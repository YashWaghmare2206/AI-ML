# Decision Tree — Classification

## 1. Simple Explanation

A **Decision Tree Classifier** is a supervised machine-learning algorithm that predicts a **categorical class** by repeatedly splitting the dataset based on feature values.

The tree has three important components:

```text
                Root
                 │
          ┌──────┴──────┐
       Condition      Condition
          │               │
       Branch           Branch
        /                   \
      ...                   ...
        │
       Leaf
        │
    Predicted Class
```

For example, suppose we have:

| Age | Income | Buys |
|---:|---:|---|
| 22 | 20K | No |
| 25 | 25K | No |
| 35 | 60K | Yes |
| 42 | 80K | Yes |

A tree might learn:

```text
             Income < 40K?
              /        \
            Yes         No
             │           │
            No          Yes
```

For a new person:

```text
Age = 30
Income = 70K
```

The tree checks:

```text
70K < 40K ? → No
                 ↓
             Predict Yes
```

The important point is that **the tree learns these decision conditions from the training data**.

---

# 2. What Does "Classification" Mean Here?

Classification means the target variable contains **discrete classes**.

Examples:

```text
Spam / Not Spam
Disease / No Disease
Cat / Dog
Pass / Fail
0 / 1
```

Mathematically, if:

\[
X = \text{input features}
\]

and

\[
y = \text{class label}
\]

then the model learns a function:

\[
f(X) \rightarrow y
\]

For example:

\[
f(\text{Age},\text{Income}) \rightarrow \{\text{Yes},\text{No}\}
\]

---

# 3. Structure of a Classification Tree

A Decision Tree consists mainly of:

### Root Node

The first decision in the tree.

```text
             Income < 40K?
```

The root is selected by the tree's splitting procedure.

---

### Internal Node

A subsequent decision.

```text
             Income < 40K?
              /        \
            Yes         No
             │           │
        Age < 30?
```

---

### Branch

The result of a decision.

```text
Income < 40K?

      Yes ──────────→
      No  ──────────→
```

---

### Leaf Node

The final prediction.

```text
              Income < 40K?
               /        \
             Yes         No
              │           │
             No          Yes
           (Leaf)       (Leaf)
```

The leaf contains the predicted class.

---

# 4. How Classification Happens

Suppose the trained tree is:

```text
              Age < 30?
              /       \
            Yes        No
             │          │
       Income < 40K?   Yes
          /    \
        No      Yes
```

For:

```text
Age = 25
Income = 50K
```

Prediction proceeds from the root:

### Step 1

Check:

\[
25 < 30
\]

True.

Move left.

### Step 2

Check:

\[
50K < 40K
\]

False.

Move right.

### Step 3

Reach the leaf:

```text
No
```

Therefore:

\[
\boxed{\text{Prediction = No}}
\]

So prediction in a decision tree is essentially **following a path from the root to a leaf**.

---

# 5. But How Does the Tree Decide the Questions?

This is the key idea we will eventually study mathematically.

Suppose the dataset has:

```text
Age
Income
Credit Score
```

The tree has many possible questions:

```text
Age < 20?
Age < 25?
Age < 30?

Income < 30K?
Income < 40K?
Income < 50K?

Credit Score < 600?
Credit Score < 700?
...
```

The tree needs to determine:

> **Which feature and which threshold should I split on?**

For classification, the split should make the resulting groups **more class-pure**.

For example, before splitting:

```text
Group:

Yes
Yes
No
Yes
No
No
Yes
```

Mixed classes.

After a good split:

```text
Left:
Yes
Yes
Yes
Yes

Right:
No
No
No
```

The resulting groups are much more pure.

The mathematical measures used to determine this are:

- **Gini**
- **Entropy**
- **Information Gain**

We will study those **later, in exactly the order you provided**.

---

# 6. Important Point

A Decision Tree Classifier does **not** directly learn a mathematical equation like linear regression:

\[
y = w_1x_1+w_2x_2+b
\]

Instead, it learns a sequence of **decision rules**:

```text
if feature_1 < threshold:
    go left
else:
    go right
```

Then eventually:

```text
return class
```

So its learned model can be thought of as:

\[
\boxed{\text{Features} \rightarrow \text{Decisions} \rightarrow \text{Leaf} \rightarrow \text{Class}}
\]

# Decision Tree — Regression

## 1. What is Decision Tree Regression?

A **Decision Tree Regressor** is a supervised learning algorithm used when the target variable is **continuous/numerical**.

Examples:

- Predict house price
- Predict temperature
- Predict salary
- Predict sales
- Predict delivery time

The basic idea is the same as classification:

> **Split the dataset into smaller groups using feature-based decisions.**

But there is one major difference:

- **Classification:** leaf → class
- **Regression:** leaf → numerical value

For example:

```text
                 Area < 1500?
                 /          \
               Yes           No
                │             │
          Bedrooms < 3?      Leaf
             /    \            │
           ...    ...        ₹85L
```

The leaves contain numerical predictions.

---

## 2. How Does a Regression Tree Predict?

Suppose we have:

| Area | Price |
|---:|---:|
| 800 | 40 |
| 1000 | 50 |
| 1200 | 60 |
| 1500 | 75 |
| 1800 | 90 |
| 2000 | 100 |

Suppose the tree creates this split:

```text
Area < 1400
```

Then the data becomes:

### Left node

| Area | Price |
|---:|---:|
| 800 | 40 |
| 1000 | 50 |
| 1200 | 60 |

### Right node

| Area | Price |
|---:|---:|
| 1500 | 75 |
| 1800 | 90 |
| 2000 | 100 |

Now the tree needs to assign a prediction to each leaf.

For a standard Decision Tree Regressor, the prediction of a leaf is the **mean target value of the training samples that reached that leaf**.

For the left leaf:

$$
\hat{y}_{left}
=
\frac{40+50+60}{3}
$$

Therefore:

$$
\hat{y}_{left}=50
$$

For the right leaf:

$$
\hat{y}_{right}
=
\frac{75+90+100}{3}
$$

Therefore:

$$
\hat{y}_{right}
=
88.33
$$

So the tree becomes approximately:

```text
                 Area < 1400?
                 /           \
               Yes            No
                │              │
                50            88.33
```

---

## 3. Prediction for a New Sample

Suppose a new house has:

```text
Area = 1100
```

The tree checks:

$$
1100 < 1400
$$

This is true, so the sample goes to the left leaf.

Therefore:

$$
\boxed{\hat{y}=50}
$$

The prediction is **50** in whatever units the target uses.

The important concept is:

> A regression tree predicts the **average target value of the samples in the final leaf**.

---

## 4. How Is the Best Split Different From Classification?

This is an important distinction.

In classification, we want a split that produces groups containing mostly the same **class**.

In regression, there are no classes.

Instead, we want a split that produces groups whose target values are **close together**.

Consider:

```text
Node A:
10, 11, 10, 12
```

The values are close together.

Compare:

```text
Node B:
5, 30, 80, 100
```

The values are highly spread out.

Node A is much more suitable as a regression leaf because its target values are concentrated around its mean.

Therefore, a regression tree tries to find splits that **reduce the variation/error within the resulting groups**.

We will mathematically study the splitting criterion later when we reach the relevant mathematical subtopics.

---

## 5. Why Does the Mean Become the Prediction?

Suppose a leaf contains target values:

$$
y_1,y_2,\dots,y_n
$$

The tree wants one constant prediction for all samples in that leaf.

Let that prediction be $c$.

The squared-error objective is:

$$
\sum_{i=1}^{n}(y_i-c)^2
$$

The value of $c$ that minimizes this quantity is the **mean**:

$$
c=\frac{1}{n}\sum_{i=1}^{n}y_i
$$

where:

- $n$ = number of samples in the leaf
- $y_i$ = target value of sample $i$
- $c$ = constant prediction assigned to the leaf

Therefore:

$$
\boxed{
\hat{y}_{leaf}
=
\frac{1}{n}\sum_{i=1}^{n}y_i
}
$$

This is why the standard Decision Tree Regressor uses the **mean target value** at a leaf.

---

## 6. Classification vs Regression

The core difference is:

| | Classification | Regression |
|---|---|---|
| Target | Categorical | Numerical |
| Leaf prediction | Class | Mean numerical value |
| Example | Spam / Not Spam | ₹50L |
| Goal of split | Improve class purity | Reduce target variation/error |
| Output | Discrete | Continuous |

So the overall structure remains:

```text
             Decision Tree
                  │
          ┌───────┴───────┐
          ↓               ↓
   Classification      Regression
          │               │
      Class leaf       Mean-value leaf
```

---

## 7. Important Point

A Decision Tree Regressor does **not necessarily produce a smooth prediction function**.

For example:

```text
Area → Prediction

800  → 50
900  → 50
1000 → 50
1100 → 50
1200 → 50
1300 → 50

1500 → 88.33
1600 → 88.33
1700 → 88.33
```

The prediction remains constant within a leaf and changes when the sample crosses a tree split.

Conceptually, this creates a **piecewise-constant prediction function**.

# The Most Important Concept

With two features, the tree is **not necessarily using both features at every decision**.

Instead, at **each node**, it asks:

> Which feature and threshold should I use to create the best split?

It might choose:

```text
Root
 ↓
Area < 1400?
```

Then:

```text
Left child
 ↓
Bedrooms < 3?
```

Then potentially another child could use:

```text
Area < 1000?
```

So the same feature can even be used **multiple times at different depths**.

The structure is:

$$
(X_1,X_2)
\rightarrow
\text{Split}
\rightarrow
\text{Split}
\rightarrow
\text{Leaf mean}
\rightarrow
\hat{y}
$$

That is how a Decision Tree Regressor handles **multiple features**.

### In our example

For the leaf ₹45:

```text
Area < 1400
      ↓
Bedrooms < 3
      ↓
   Leaf
```

$$
\boxed{

$$

So **we don't take the mean of values at the nodes along the path**. We take the mean of the **target values of the training samples that reach the final leaf**.

# Decision Tree — Gini Impurity

Now we move to the next subtopic: **Gini**.

Gini is used to measure **how impure a node is in classification**.

---

## 1. What is Gini Impurity?

Suppose a node contains:

```text
Yes  Yes  Yes  No  No
```

There are two classes:

- Yes → 3 samples
- No → 2 samples

This node contains a mixture of classes, so it is **impure**.

Compare that with:

```text
Yes  Yes  Yes  Yes  Yes
```

This node contains only one class.

It is **pure**.

Gini Impurity gives us a numerical measure of this impurity.

---

## 2. The Formula

For a node containing $K$ classes:

$$
Gini = 1-\sum_{k=1}^{K}p_k^2
$$

where:

- $K$ = number of classes
- $p_k$ = proportion of samples belonging to class $k$

So the procedure is:

```text
Count each class
      ↓
Calculate class proportion
      ↓
Square each proportion
      ↓
Add them
      ↓
Subtract from 1
```

---

## 3. Simple Numerical Example

Suppose a node contains:

```text
Yes = 3
No  = 2
```

Total samples:

$$
n=3+2=5
$$

### Step 1 — Calculate proportions

For Yes:

$$
p_{Yes}=\frac{3}{5}=0.6
$$

For No:

$$
p_{No}=\frac{2}{5}=0.4
$$

### Step 2 — Apply the formula

$$
Gini=1-(0.6^2+0.4^2)
$$

Calculate each square:

$$
0.6^2=0.36
$$

$$
0.4^2=0.16
$$

Therefore:

$$
Gini=1-(0.36+0.16)
$$

$$
Gini=1-0.52
$$

$$
\boxed{Gini=0.48}
$$

So this node has a Gini impurity of **0.48**.

---

## 4. What Does the Value Mean?

For binary classification:

### Completely pure node

Suppose:

```text
Yes = 5
No = 0
```

Then:

$$
p_{Yes}=1
$$

$$
p_{No}=0
$$

Therefore:

$$
Gini=1-(1^2+0^2)
$$

$$
\boxed{Gini=0}
$$

So:

> **Gini = 0 → completely pure node**

---

### Mixed node

Suppose:

```text
Yes = 3
No = 2
```

We already calculated:

$$
Gini=0.48
$$

There is more mixing.

---

### Maximum impurity for binary classification

Suppose:

```text
Yes = 5
No = 5
```

Then:

$$
p_{Yes}=0.5
$$

$$
p_{No}=0.5
$$

Therefore:

$$
Gini=1-(0.5^2+0.5^2)
$$

$$
=1-(0.25+0.25)
$$

$$
\boxed{Gini=0.5}
$$

For a binary classification problem, the maximum Gini impurity is:

$$
\boxed{0.5}
$$

when both classes are equally represented.

---

## 5. Mathematical Intuition

Look at the formula again:

$$
Gini=1-\sum_{k=1}^{K}p_k^2
$$

The term:

$$
\sum_{k=1}^{K}p_k^2
$$

becomes large when one class dominates.

For a pure node:

$$
1^2+0^2=1
$$

so:

$$
Gini=1-1=0
$$

For a balanced binary node:

$$
0.5^2+0.5^2=0.5
$$

so:

$$
Gini=1-0.5=0.5
$$

Therefore:

$$
\boxed{\text{More class mixing} \Rightarrow \text{higher Gini}}
$$

and:

$$
\boxed{\text{More class purity} \Rightarrow \text{lower Gini}}
$$

---

## 6. How the Tree Uses Gini

Suppose the tree is deciding between two possible splits.

### Split A

```text
Left node  → mostly Yes
Right node → mostly No
```

This will generally produce **low Gini values** in the child nodes.

### Split B

```text
Left node  → mixture of Yes and No
Right node → mixture of Yes and No
```

This produces **higher Gini values**.

The tree therefore prefers splits that create **purer child nodes**.

Later, when we reach **Information Gain**, we'll formally calculate how a parent node's impurity changes after a split.

For now, remember:

$$
\boxed{\text{Good classification split} \rightarrow \text{lower child-node impurity}}
$$

---

## 7. One Important Distinction

Gini is **not calculated using the feature values themselves**.

For example, if we have:

| Age | Class |
|---:|---|
| 20 | Yes |
| 25 | No |
| 30 | Yes |
| 35 | Yes |

Gini uses the **class distribution**:

```text
Yes = 3
No = 1
```

It does **not** directly calculate Gini from:

```text
20, 25, 30, 35
```

The feature is used to create a split first. Then Gini measures the class composition of the resulting node.

---

### Core idea

$$
\boxed{
Gini=1-\sum_{k=1}^{K}p_k^2
}
$$

- $Gini=0$ → pure node
- Higher Gini → more class mixing
- For binary classification → maximum is $0.5$
- The tree prefers splits that produce lower impurity



# Decision Tree — Entropy

Entropy is another measure of **impurity/uncertainty** used in Decision Tree Classification.

---

## 1. What is Entropy?

Suppose a node contains:

```text
Yes  Yes  Yes  No  No
```

There are two classes:

- Yes → 3
- No → 2

The node is mixed, so there is uncertainty about the class of a randomly selected sample.

**Entropy measures this uncertainty.**

- Low entropy → node is more pure
- High entropy → node is more mixed

---

## 2. Formula

For a node containing $K$ classes:

$$
H=-\sum_{k=1}^{K}p_k\log_2(p_k)
$$

where:

- $H$ = entropy of the node
- $K$ = number of classes
- $p_k$ = proportion of samples belonging to class $k$
- $\log_2$ = logarithm with base 2

---

## 3. Numerical Example

Take:

```text
Yes = 3
No  = 2
```

Total:

$$
n=5
$$

Therefore:

$$
p_{Yes}=\frac{3}{5}=0.6
$$

and:

$$
p_{No}=\frac{2}{5}=0.4
$$

Substitute into the formula:

$$
H=-(0.6\log_2(0.6)+0.4\log_2(0.4))
$$

We have approximately:

$$
\log_2(0.6)\approx-0.737
$$

and:

$$
\log_2(0.4)\approx-1.322
$$

Therefore:

$$
H=-[(0.6)(-0.737)+(0.4)(-1.322)]
$$

$$
H=-[-0.4422-0.5288]
$$

$$
\boxed{H\approx0.971}
$$

So this node has entropy approximately **0.971**.

---

# 4. Pure Node

Suppose:

```text
Yes = 5
No  = 0
```

Then:

$$
p_{Yes}=1
$$

$$
p_{No}=0
$$

Entropy:

$$
H=-(1\log_2(1)+0\log_2(0))
$$

Since:

$$
\log_2(1)=0
$$

and the second term is treated as $0$ because:

$$
\lim_{p\rightarrow0^+}p\log_2(p)=0
$$

we get:

$$
\boxed{H=0}
$$

So a completely pure node has:

$$
\boxed{\text{Entropy}=0}
$$

---

# 5. Maximum Entropy

For binary classification, maximum uncertainty occurs when both classes have equal probability:

```text
Yes = 5
No  = 5
```

Therefore:

$$
p_{Yes}=0.5
$$

$$
p_{No}=0.5
$$

Entropy:

$$
H=-(0.5\log_2(0.5)+0.5\log_2(0.5))
$$

Since:

$$
\log_2(0.5)=-1
$$

we get:

$$
H=-[(0.5)(-1)+(0.5)(-1)]
$$

$$
H=1
$$

Therefore:

$$
\boxed{H_{max}=1}
$$

for a binary classification problem.

---

# 6. Compare Different Nodes

For binary classification:

| Distribution | Entropy |
|---|---:|
| 5 Yes, 0 No | 0 |
| 4 Yes, 1 No | 0.722 |
| 3 Yes, 2 No | 0.971 |
| 2 Yes, 3 No | 0.971 |
| 1 Yes, 4 No | 0.722 |
| 0 Yes, 5 No | 0 |

Notice the pattern:

```text
Pure
  ↓
Entropy = 0

More mixed
  ↓
Entropy increases

50/50
  ↓
Maximum entropy = 1
```

---

# 7. Why Does Entropy Measure Uncertainty?

Consider these two nodes.

### Node A

```text
Yes Yes Yes Yes Yes
```

If I randomly select a sample, I know exactly what class it will be.

Probability:

$$
P(Yes)=1
$$

There is **no uncertainty**.

Therefore:

$$
H=0
$$

---

### Node B

```text
Yes No Yes No No Yes No Yes
```

The classes are much more mixed.

It is harder to predict the class of a randomly selected sample.

Therefore, entropy is higher.

---

# 8. Gini vs Entropy

You just learned Gini, so compare them:

| | Gini | Entropy |
|---|---|---|
| Purpose | Measure impurity | Measure uncertainty/impurity |
| Pure node | 0 | 0 |
| Binary maximum | 0.5 | 1 |
| Formula | $1-\sum p_k^2$ | $-\sum p_k\log_2(p_k)$ |
| Used for | Classification | Classification |

Both generally behave similarly:

$$
\boxed{\text{More pure} \Rightarrow \text{lower impurity}}
$$

The important difference is their mathematical definition.

---

## 9. Important Point

Entropy itself **does not choose the split**.

It tells us:

> "How impure is this node?"

To determine whether a particular split is good, we compare the entropy **before and after the split**.

That leads directly to the next subtopic:

**Information Gain**.

---

### Core idea

$$
\boxed{
H=-\sum_{k=1}^{K}p_k\log_2(p_k)
}
$$

For binary classification:

$$
0\leq H\leq1
$$

- $H=0$ → completely pure
- $H=1$ → maximum uncertainty
- Higher entropy → more class mixing



## Range of Gini and Entropy

## Gini Impurity

Formula:

$$
Gini = 1-\sum_{k=1}^{K}p_k^2
$$

For **$K$ classes**, the range is:

$$
\boxed{0 \leq Gini \leq 1-\frac{1}{K}}
$$

where $K$ is the number of classes.

### Binary classification: $K=2$

$$
0 \leq Gini \leq 1-\frac{1}{2}
$$

$$
\boxed{0 \leq Gini \leq 0.5}
$$

So:

- $0$ → completely pure
- $0.5$ → maximum impurity

---

## Entropy

Formula:

$$
H=-\sum_{k=1}^{K}p_k\log_2(p_k)
$$

Its range is:

$$
\boxed{0 \leq H \leq \log_2(K)}
$$

### Binary classification: $K=2$

$$
0 \leq H \leq \log_2(2)
$$

Therefore:

$$
\boxed{0 \leq H \leq 1}
$$

So:

- $0$ → completely pure
- $1$ → maximum uncertainty

---

## For 3 classes

If $K=3$:

### Gini

$$
0 \leq Gini \leq 1-\frac{1}{3}
$$

$$
\boxed{0 \leq Gini \leq 0.667}
$$

### Entropy

$$
0 \leq H \leq \log_2(3)
$$

$$
\boxed{0 \leq H \leq 1.585}
$$

---

### General rule

| Number of classes | Gini range | Entropy range |
|---:|---:|---:|
| 2 | $0$ to $0.5$ | $0$ to $1$ |
| 3 | $0$ to $0.667$ | $0$ to $1.585$ |
| 4 | $0$ to $0.75$ | $0$ to $2$ |
| $K$ | $0$ to $1-\frac{1}{K}$ | $0$ to $\log_2 K$ |

So **both have fixed ranges**, but the maximum depends on the number of classes.

# Decision Tree — Information Gain

Now we connect **Entropy** to the actual process of choosing a split.

## 1. What is Information Gain?

Entropy tells us:

> **How impure is the current node?**

But a Decision Tree needs to answer:

> **Which split should I choose?**

**Information Gain** measures how much the impurity decreases after making a split.

So:

$$
\boxed{
\text{Information Gain}
=
\text{Parent Entropy}
-
\text{Weighted Child Entropy}
}
$$

Higher Information Gain means the split produced a greater reduction in entropy.

---

# 2. Simple Example

Suppose our parent node contains:

```text
Yes = 4
No  = 4
```

So there are 8 samples.

The class probabilities are:

$$
p_{Yes}=\frac{4}{8}=0.5
$$

$$
p_{No}=\frac{4}{8}=0.5
$$

Therefore, parent entropy is:

$$
H_{parent}
=
-(0.5\log_2 0.5+0.5\log_2 0.5)
$$

Since:

$$
\log_2(0.5)=-1
$$

we get:

$$
H_{parent}=1
$$

So initially:

$$
\boxed{H_{parent}=1}
$$

---

# 3. Make a Split

Suppose the tree considers a particular feature and creates:

### Left child

```text
Yes = 4
No  = 0
```

### Right child

```text
Yes = 0
No  = 4
```

So:

```text
              Parent
           Yes = 4
            No = 4
              │
          ─── Split ───
           /           \
       Left            Right
      4 Yes            4 No
      0 No             0 Yes
```

Both child nodes are completely pure.

Therefore:

$$
H_{left}=0
$$

and:

$$
H_{right}=0
$$

---

# 4. Calculate Weighted Child Entropy

We cannot simply calculate:

$$
H_{left}+H_{right}
$$

because the children may contain different numbers of samples.

Instead, we calculate a **weighted average**.

The formula is:

$$
H_{children}
=
\frac{n_{left}}{n}H_{left}
+
\frac{n_{right}}{n}H_{right}
$$

where:

- $n$ = total number of samples in the parent
- $n_{left}$ = samples in the left child
- $n_{right}$ = samples in the right child
- $H_{left}$ = entropy of the left child
- $H_{right}$ = entropy of the right child

Here:

$$
n=8
$$

$$
n_{left}=4
$$

$$
n_{right}=4
$$

Therefore:

$$
H_{children}
=
\frac{4}{8}(0)
+
\frac{4}{8}(0)
$$

$$
H_{children}=0
$$

---

# 5. Calculate Information Gain

Now:

$$
IG
=
H_{parent}-H_{children}
$$

Substitute:

$$
IG=1-0
$$

Therefore:

$$
\boxed{IG=1}
$$

This is a **perfect split** for this binary node because the split completely separated the two classes.

---

# 6. Why Do We Need the Weighted Average?

Consider a different split.

Suppose the parent has 10 samples:


```text
Left:
Yes = 4
No  = 4       → 8 samples

Right:
Yes = 1
No = 1        → 2 samples
```

The right child is pure? No—it has one Yes and one No, so it is actually mixed.

The important point is that the children have **different sizes**.

Therefore, we calculate:

$$
H_{children}
=
\frac{8}{10}H_{left}
+
\frac{2}{10}H_{right}
$$

The larger child gets a larger weight because it contains more training samples.

This prevents a tiny child node from having the same influence as a large child node.

---

# 7. General Information Gain Formula

For a split producing $m$ child nodes:

$$
\boxed{
IG
=
H(parent)
-
\sum_{j=1}^{m}
\frac{n_j}{n}
H(child_j)
}
$$

where:

- $IG$ = information gain
- $H(parent)$ = entropy before splitting
- $m$ = number of child nodes
- $n$ = number of samples in the parent
- $n_j$ = number of samples in child $j$
- $H(child_j)$ = entropy of child $j$

For a normal binary tree split, $m=2$.

---

# 8. How Does the Tree Choose the Split?

Suppose we have:

```text
Feature 1
    ├── Split A
    ├── Split B
    └── Split C

Feature 2
    ├── Split A
    ├── Split B
    └── Split C
```

The tree calculates Information Gain for each candidate split.

For example:

| Split | Information Gain |
|---|---:|
| Feature 1 < 20 | 0.10 |
| Feature 1 < 30 | 0.25 |
| Feature 1 < 40 | 0.41 |
| Feature 2 < 5 | 0.18 |
| Feature 2 < 7 | 0.72 |
| Feature 2 < 9 | 0.30 |

The tree would select the split with the **highest Information Gain**:

$$
\boxed{IG=0.72}
$$

because it produces the largest reduction in entropy.

---

# 9. Information Gain Intuition

Think about what happens to entropy:

```text
Before split
     ↓
High entropy
     ↓
       Split
      /     \
     ↓       ↓
Low entropy Low entropy
```

Therefore:

$$
\boxed{
\text{Good split}
\rightarrow
\text{large entropy reduction}
\rightarrow
\text{high Information Gain}
}
$$

And:

$$
\boxed{
\text{Poor split}
\rightarrow
\text{small entropy reduction}
\rightarrow
\text{low Information Gain}
}
$$

---

## Important Connection

Now the relationship between the three concepts is clear:

```text
Entropy
   ↓
Measures impurity of a node
   ↓
Information Gain
   ↓
Measures reduction in entropy after a split
   ↓
Decision Tree
   ↓
Chooses a high-information-gain split
```

So **Entropy measures the state of a node**, while **Information Gain evaluates the quality of a split**.

# Decision Tree — Pruning

A Decision Tree can keep splitting the data until it becomes extremely detailed. If we allow this to happen, the tree can start **memorizing the training data** instead of learning general patterns.

**Pruning** is the process of controlling/removing unnecessary tree branches to reduce this problem.

---

## 1. Why is Pruning Needed?

Suppose we have a dataset:

```text
                    Root
                      │
              ┌───────┴───────┐
              │               │
             ...             ...
                              │
                         many splits
                              │
                         many more splits
                              │
                            Leaf
```

If the tree keeps splitting, eventually we can get very small leaves.

For example:

```text
Leaf 1 → 1 sample
Leaf 2 → 2 samples
Leaf 3 → 1 sample
Leaf 4 → 1 sample
```

The tree can become extremely specific to the training data.

This leads to:

```text
Training error ↓
Test error   ↑
```

This is **overfitting**.

---

# 2. What Does Pruning Do?

Pruning removes parts of the tree that are not sufficiently useful.

Before pruning:

```text
                    A
                   / \
                  B   C
                 / \
                D   E
```

Suppose the subtree under `B` doesn't provide enough improvement.

Pruning can replace:

```text
                    A
                   / \
                  B   C
                 / \
                D   E
```

with:

```text
                    A
                   / \
                  B   C
```

where `B` becomes a leaf.

So instead of having many detailed rules, the tree becomes simpler.

---

# 3. Two Types of Pruning

There are two main approaches:

### Pre-Pruning

Stop the tree **before it becomes too large**.

Examples:

```text
max_depth
min_samples_split
min_samples_leaf
max_leaf_nodes
```

The tree is prevented from growing beyond certain conditions.

---

### Post-Pruning

First grow a relatively large tree, then remove branches that aren't sufficiently useful.

In Scikit-Learn, an important post-pruning method is:

```python
ccp_alpha
```

which uses **Cost-Complexity Pruning**.

We'll study this mathematically and practically when we reach hyperparameters.

---

# 4. Pre-Pruning Example

Suppose the tree is allowed to grow without restriction:

```text
                    Root
                  /      \
                ...      ...
                        /   \
                      ...   ...
                           / \
                          ... ...
```

It might reach:

```text
depth = 15
```

Now suppose we impose:

```python
max_depth=3
```

The tree cannot grow beyond depth 3:

```text
              Root
             /    \
           ...    ...
          /  \    / \
        ... ... ... ...
```

This is pre-pruning.

The tree is stopped before reaching unnecessary complexity.

---

# 5. Post-Pruning Intuition

Suppose a fully grown tree looks like:

```text
                 Root
                /    \
               A      B
              / \    / \
             C   D  E   F
                / \
               G   H
```

Suppose the subtree:

```text
              D
             / \
            G   H
```

only provides a very small improvement.

We can remove it:

```text
                 Root
                /    \
               A      B
              / \    / \
             C   D  E   F
```

Now `D` becomes a leaf.

The tree becomes simpler.

---

# 6. Why Doesn't the Tree Just Become as Small as Possible?

Because an extremely small tree can **underfit**.

Consider:

```text
Very small tree
      ↓
Too simple
      ↓
Underfitting
```

And:

```text
Very large tree
      ↓
Too complex
      ↓
Overfitting
```

So we want an appropriate level of complexity.

Conceptually:

$$
\boxed{
\text{Model complexity}
\quad\leftrightarrow\quad
\text{Fit to data}
}
$$

Pruning tries to remove complexity that isn't justified by enough improvement in the model.

---

# 7. Cost-Complexity Pruning

The main mathematical idea behind post-pruning is to balance:

1. **Prediction error**
2. **Tree complexity**

The cost-complexity objective is:

$$
R_\alpha(T)=R(T)+\alpha|T|
$$

where:

- $T$ = tree
- $R(T)$ = error/impurity measure of the tree
- $\alpha$ = complexity parameter
- $|T|$ = number of terminal leaves
- $R_\alpha(T)$ = total pruning objective

The first term:

$$
R(T)
$$

measures how well the tree fits the data.

The second term:

$$
\alpha|T|
$$

penalizes having too many leaves.

Therefore:

$$
\boxed{
\text{Objective}
=
\text{Fit error}
+
\text{Complexity penalty}
}
$$

---

# 8. What Does $\alpha$ Control?

The parameter $\alpha$ controls how strongly we penalize tree complexity.

### Small $\alpha$

The complexity penalty is small:

$$
\alpha \rightarrow 0
$$

So the tree can keep more branches.

Result:

```text
More complex tree
↓
Can fit training data more closely
↓
Higher risk of overfitting
```

### Large $\alpha$

The complexity penalty becomes stronger:

$$
\alpha \uparrow
$$

More branches become undesirable.

Result:

```text
Smaller tree
↓
Less complexity
↓
Potentially less overfitting
```

But if $\alpha$ becomes too large, the tree can become too simple and **underfit**.

---

# 9. The Core Idea of Pruning

Think of the progression:

```text
No pruning
     ↓
Very complex tree
     ↓
Potential overfitting

        ↓ pruning

Simpler tree
     ↓
Less unnecessary complexity
     ↓
Better generalization potential
```

The important point is that **pruning does not mean "make the tree as small as possible."**

It means:

> Remove branches whose additional complexity is not sufficiently justified by their improvement in fitting the data.

---

## Pre-Pruning vs Post-Pruning

| | Pre-Pruning | Post-Pruning |
|---|---|---|
| When? | During tree construction | After/while considering a grown tree |
| Main idea | Stop growth | Remove branches |
| Examples | `max_depth`, `min_samples_leaf` | `ccp_alpha` |
| Goal | Prevent excessive growth | Simplify an existing tree |

# Decision Tree — Hyperparameters

Hyperparameters are settings that you choose **before training** the Decision Tree.

They control things such as:

- How deep the tree can grow
- How many samples are required for a split
- How large a leaf can be
- How splits are evaluated
- Whether the tree is pruned

We'll go one by one.

---

# 1. `criterion`

`criterion` determines **how the tree evaluates the quality of a split**.

For classification, common choices include:

```python
criterion="gini"
```

or:

```python
criterion="entropy"
```

or:

```python
criterion="log_loss"
```

### Gini

Uses:

$$
Gini=1-\sum_{k=1}^{K}p_k^2
$$

Lower impurity is better.

### Entropy

Uses:

$$
H=-\sum_{k=1}^{K}p_k\log_2(p_k)
$$

Higher information gain means a better split.

For regression, common criteria include:

```python
criterion="squared_error"
```

and others such as:

```python
criterion="absolute_error"
```

The important idea is:

> `criterion` determines the mathematical measure used to evaluate candidate splits.

---

# 2. `max_depth`

`max_depth` controls the **maximum depth of the tree**.

Example:

```python
DecisionTreeClassifier(max_depth=3)
```

means the tree can grow to a maximum depth of 3.

### Smaller `max_depth`

```text
max_depth ↓
     ↓
Simpler tree
     ↓
Less model complexity
     ↓
Higher chance of underfitting
```

### Larger `max_depth`

```text
max_depth ↑
     ↓
More complex tree
     ↓
Can capture more patterns
     ↓
Higher risk of overfitting
```

For example:

```python
max_depth=2
```

produces a shallow tree.

While:

```python
max_depth=20
```

allows much deeper splitting.

---

# 3. `min_samples_split`

This controls the **minimum number of samples required in a node before that node can be split**.

Suppose:

```python
min_samples_split=5
```

A node must contain at least 5 samples before the algorithm can consider splitting it.

### Small value

```python
min_samples_split=2
```

The tree can split very small nodes.

Result:

```text
More splitting
↓
More complex tree
↓
Higher overfitting risk
```

### Large value

```python
min_samples_split=20
```

Small nodes cannot be split.

Result:

```text
Less splitting
↓
Simpler tree
↓
Higher underfitting risk if too large
```

---

# 4. `min_samples_leaf`

This controls the **minimum number of training samples that must remain in a leaf**.

Suppose:

```python
min_samples_leaf=5
```

Then every final leaf must contain at least 5 training samples.

This is different from `min_samples_split`.

### `min_samples_split`

Controls:

> Can this node be split?

### `min_samples_leaf`

Controls:

> How small can the resulting leaf be?

For example:

```text
Node has 10 samples

         Split
        /     \
       2       8
```

If:

```python
min_samples_leaf=3
```

this split is not allowed because the left leaf would contain only 2 samples.

---

# 5. `max_leaf_nodes`

Controls the **maximum number of leaf nodes**.

Example:

```python
DecisionTreeClassifier(max_leaf_nodes=10)
```

The tree cannot have more than 10 leaves.

Smaller value:

```text
Fewer leaves
↓
Simpler tree
```

Larger value:

```text
More leaves
↓
More complex tree
```

---

# 6. `max_features`

This controls the **maximum number of features considered when looking for the best split**.

Suppose your dataset has:

```text
100 features
```

If:

```python
max_features=10
```

the tree considers only 10 features when searching for the split at a node.

This can reduce the amount of feature search and can affect the resulting tree structure.

For a normal single Decision Tree, you might commonly leave this at its default rather than restricting it unnecessarily.

It becomes particularly important in **Random Forest**, where random feature selection is a fundamental part of the algorithm.

---

# 7. `ccp_alpha`

This is related to the **Cost-Complexity Pruning** we just discussed.

The objective is:

$$
R_\alpha(T)=R(T)+\alpha|T|
$$

where:

- $R(T)$ = tree error
- $\alpha$ = complexity penalty
- $|T|$ = number of leaves

### Small `ccp_alpha`

```text
α ↓
↓
Weak complexity penalty
↓
Larger tree
```

### Large `ccp_alpha`

```text
α ↑
↓
Strong complexity penalty
↓
Smaller tree
```

Example:

```python
DecisionTreeClassifier(ccp_alpha=0.01)
```

The exact useful value depends on the dataset.

---

# 8. `random_state`

`random_state` controls randomness where the implementation uses randomized behavior.

Example:

```python
DecisionTreeClassifier(random_state=42)
```

Using the same value makes the result reproducible when randomness is involved.

It does **not** mean:

> "The model becomes better."

It means:

> "Use the same random process so I can reproduce the result."

---
