## Feature Engineering

```python
delinquency_cols = [
    "NumberOfTime30-59DaysPastDueNotWorse",
    "NumberOfTimes90DaysLate",
    "NumberOfTime60-89DaysPastDueNotWorse"
]
```

```python
df["TotalDelinquencies"] = (
        df["NumberOfTime30-59DaysPastDueNotWorse"]
        + df["NumberOfTime60-89DaysPastDueNotWorse"]
        + df["NumberOfTimes90DaysLate"]
)
```

```python
df["TotalDelinquencies"].describe()
```

```python
df[
    delinquency_cols + ["TotalDelinquencies"]
    ].head()
```

```python
age_bins = [20 , 30 , 40 , 50 , 60 , 70 , 80 , float("inf")]
age_labels = [
    "21-30",
    "31-40",
    "41-50",
    "51-60",
    "61-70",
    "71-80",
    "81+"
]

df["AgeGroup"] = pd.cut(
    df["age"],
    bins= age_bins,
    labels=age_labels,
    right= True
)
```

```python
df["AgeGroup"].value_counts().sort_index()
```

```python
df[["age", "AgeGroup"]].head(10)
```

```python
df["HighUtilization"] = (
        df["RevolvingUtilizationOfUnsecuredLines"] > 1
).astype(int)
```

```python
df["HighUtilization"].value_counts()
```

```python
df.groupby("HighUtilization")["SeriousDlqin2yrs"].mean()
```

```python
df["HighDebtRatio"] = (
        df["DebtRatio"] > 1
).astype(int)
```

```python
df["HighDebtRatio"].value_counts()
```

```python
df.groupby("HighDebtRatio")["SeriousDlqin2yrs"].mean()
```

```python
### Not much info form this feature so dropping it
df = df.drop(columns=["HighDebtRatio"])
```

```python
df["IncomeMissing"] = df["MonthlyIncome"].isna().astype(int)
```

```python
df["MonthlyIncome"] = df["MonthlyIncome"].fillna(
    df["MonthlyIncome"].median()
)
```

```python
df["TotalCreditLines"] = (
        df["NumberOfOpenCreditLinesAndLoans"]
        + df["NumberRealEstateLoansOrLines"]
)
```

```python
df[
    [
        "NumberOfOpenCreditLinesAndLoans",
        "NumberRealEstateLoansOrLines",
        "TotalCreditLines"
    ]
].head(10)
```

```python
df.groupby("TotalCreditLines")["SeriousDlqin2yrs"].mean()
```

```python
df["TotalCreditLines"].describe()
```

```python
df["Has90DaysLate"] = (
        df["NumberOfTimes90DaysLate"] > 0
).astype(int)
```

```python
df.groupby("Has90DaysLate")["SeriousDlqin2yrs"].mean()
```

```python
df["HasAnyDelinquency"] = (
        df["TotalDelinquencies"] > 0
).astype(int)
```

```python
df.groupby("HasAnyDelinquency")["SeriousDlqin2yrs"].mean()
```

```python
df["IncomePerCreditLine"] = (
        df["MonthlyIncome"] /
        df["TotalCreditLines"].replace(0, 1) # temporarily changes 0 to 1 only for the denominator, preventing division by zero.
)
```

```python
df[
    [
        "MonthlyIncome",
        "TotalCreditLines",
        "IncomePerCreditLine"
    ]
].head(10)
```

```python
df.groupby(
    pd.qcut(df["IncomePerCreditLine"], 10, duplicates="drop")
)["SeriousDlqin2yrs"].mean()
```
