### **1.1 What is Univariate Analysis? (Short Definition)**

> **Univariate Analysis is the process of analyzing a single variable (one column) independently to understand its characteristics and distribution.**

### Example

Suppose in the House Prices dataset, you study only the `SalePrice` column:

```python
df["SalePrice"]
```

and ask questions like:

* What is the average SalePrice?
* What is the minimum and maximum SalePrice?
* Is SalePrice skewed?
* Are there outliers?

Since you are analyzing **only one variable (`SalePrice`)**, this is **Univariate Analysis**.

**Easy way to remember:**

> **Uni = One → One Variable → Univariate Analysis.**


# 1.2 Why do we perform Univariate Analysis?

### Short Definition

> We perform **Univariate Analysis** to understand a single variable before studying relationships with other variables.

It helps us identify:

* How the values are distributed.
* Whether there are outliers.
* Whether the data is skewed.
* Whether there are data quality issues.
* What preprocessing may be needed.

---

### Example

Suppose we analyze only the `SalePrice` column.

```python
df["SalePrice"]
```

We discover that:

* Most houses cost between **₹100,000–₹250,000**.
* A few houses cost **₹700,000–₹800,000**.

From this, we conclude:

> **SalePrice is right-skewed and may need a log transformation.**

This is why we perform univariate analysis: **to understand what story a single variable tells us.**

### Easy way to remember

> **What is it?** → Studying one variable.
> **Why do it?** → To understand that variable before moving further.

## Short Definition

> A **Histogram** is a graph that shows how the values of a numerical variable are distributed by grouping them into intervals called **bins**.

---

## Why do we use it?

A histogram helps us answer questions like:

* Where do most values lie?
* Is the data spread out or concentrated?
* Is the distribution normal or skewed?
* Are there unusual values?

---

## Example

Suppose `SalePrice` contains:

```text
100
120
130
140
150
160
170
180
400
450
```

Instead of listing all values, a histogram groups them:

| Bin     | Count |
| ------- | ----: |
| 100–199 |     8 |
| 200–299 |     0 |
| 300–399 |     0 |
| 400–499 |     2 |

This immediately tells us:

> Most houses are cheap, while only a few are expensive.

---

## Intuition

Imagine sorting exam marks into ranges:

```text
0–10
11–20
21–30
31–40
...
```

and counting how many students fall into each range.

A histogram does exactly this for numerical data.

---

## Code Example

For House Prices:

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df["SalePrice"])
plt.show()
```

# Topic 1 → Univariate Analysis

## Subtopic 1.3.2 → KDE Plot (Kernel Density Estimation)

---

# Short Definition

> A **KDE Plot** is a smooth curve that estimates the distribution (probability density) of a numerical variable.

It is like a **smoothed version of a histogram**.

---

# Why do we use it?

A histogram can sometimes look "blocky" because of bins.

KDE helps us see the **overall shape** more clearly:

* Is the data normal?
* Is it right-skewed?
* Is it left-skewed?
* Does it have multiple peaks?

---

# Intuition

Imagine you have these heights:

```text
160 162 163 165 166 170 171 172 180
```

Histogram says:

```text
160–165 : ████
166–170 : ██
171–175 : ███
176–180 : █
```

KDE instead draws a **smooth mountain-like curve** through these values.

Think:

> Histogram = Staircase

> KDE = Smooth hill

---

# Example

Suppose `SalePrice`:

```text
100 120 130 140 150 160 170 400 450
```

Histogram:

```text
█ █ ███ █
```

KDE:

```text
      /\
     /  \
 ___/    \____
            \__
```

The KDE shows:

> Most houses are cheaper, with a long tail toward expensive houses.

---

# Code

## Basic

```python
sns.kdeplot(df["SalePrice"])
plt.show()
```

---

## With fill

```python
sns.kdeplot(df["SalePrice"], fill=True)
plt.show()
```

# Histogram vs KDE

| Histogram                   | KDE                         |
| --------------------------- | --------------------------- |
| Uses bins                   | Uses smooth curve           |
| Exact counts visible        | Overall shape visible       |
| Depends heavily on bin size | Depends on bandwidth        |
| Good for frequencies        | Good for distribution shape |

# Short Definition

> A **Box Plot** is a graph that summarizes the distribution of a numerical variable using the **five-number summary** and helps detect **outliers**.

---

# Why do we use it?

A box plot quickly tells us:

* Where the center of the data is.
* How spread out the data is.
* Whether the data is skewed.
* Whether outliers exist.

---

# Intuition

Imagine 100 students took an exam.

Instead of listing all 100 marks, you summarize them using:

* Lowest mark
* 25% mark
* Middle mark
* 75% mark
* Highest mark

A box plot visualizes exactly this.

---

# The Five-Number Summary

Suppose these values:

```text
2 4 5 7 8 10 12 15 18
```

The box plot uses:

| Statistic   | Meaning                     |
| ----------- | --------------------------- |
| Minimum     | Smallest value              |
| Q1 (25%)    | 25% of data lies below this |
| Median (Q2) | Middle value (50%)          |
| Q3 (75%)    | 75% of data lies below this |
| Maximum     | Largest non-outlier value   |

---

# Structure of a Box Plot

```text
      Outlier
         •

|-----|=======|----|
Min   Q1 Med  Q3  Max
```

### Box

```text
|=======|
```

Represents:

> Middle 50% of the data (IQR).

---

### Line inside box

```text
   |
```

Represents:

> Median (50th percentile).

---

### Whiskers

```text
|----      ----|
```

Represent:

> Range of normal values.

---

### Dots

```text
•
```

Represent:

> Outliers.

---

# Example

Suppose `SalePrice`:

```text
100
120
130
140
150
160
170
180
700
```

Most houses:

```text
100–180
```

One expensive house:

```text
700
```

Box plot:

```text
|-----|======|----|      •
                        700
```

This tells us:

> 700 is an outlier.

# Box Plot and IQR

The box represents:

> **IQR (Interquartile Range)**

Formula:

```text
IQR = Q3 − Q1
```

This is the spread of the middle 50% of data.

We'll study IQR deeply in the Outlier Analysis topic.

For now, just remember:

> Bigger IQR → More spread.

# Topic 1 → Univariate Analysis

## Subtopic 1.3.4 → Distribution Shape

When we use Histograms or KDE plots, one of the first things we observe is:

> **What shape does the distribution have?**

The three most important shapes are:

1. Normal Distribution
2. Right-Skewed Distribution
3. Left-Skewed Distribution

---

# 1. Normal Distribution

## Short Definition

> A distribution where values are symmetrically distributed around the center.

---

## Why do we use it?

Many statistical methods and ML assumptions work better when data is approximately normal.

Examples:

* Linear Regression residuals
* Hypothesis testing
* Confidence intervals

---

## Intuition

Think of human heights.

Most people have average heights.

Very short and very tall people are rare.

---

## Shape

```text
          /\
         /  \
        /    \
_______/      \_______
```

This is often called the **Bell Curve**.

---

## Example

Exam scores:

```text
60 65 68 70 72 74 75 78 80
```

Most values are near the center.

---

## How to Interpret

If the histogram/KDE looks balanced:

* Mean ≈ Median ≈ Mode
* No long tails
* Data is symmetric

---

## Common Mistakes

❌ Thinking data must be perfectly normal.

Reality:

> Most real-world data is only approximately normal.

---

## 1-Line Summary

> **Normal Distribution = Balanced data centered around the average.**

---

# 2. Right-Skewed Distribution (Positive Skew)

## Short Definition

> Most values are small, with a few very large values creating a long tail on the right.

---

## Why is it important?

Right-skewed variables often benefit from transformations such as:

```python
np.log1p()
```

to reduce skewness.

---

## Intuition

Think of salaries.

Most people earn average salaries.

A few people earn extremely high salaries.

---

## Shape

```text
        /\
       /  \______
______/           \_____
```

Long tail →

```text
Right
```

---

## Example

House Prices:

```text
100k
120k
150k
180k
200k
800k
900k
```

Most houses are affordable.

A few luxury houses pull the tail right.

---

## How to Interpret

Usually:

```text
Mean > Median
```

because large values pull the mean upward.

---

## Common Mistakes

❌ Thinking skew means bad data.

Reality:

> Skewness is normal in many datasets.

Example:

* Income
* House prices
* Medical costs

---

## 1-Line Summary

> **Right Skew = Few very large values stretch the distribution to the right.**

---

# 3. Left-Skewed Distribution (Negative Skew)

## Short Definition

> Most values are large, with a few very small values creating a long tail on the left.

---

## Why is it important?

It tells us the data behaves differently from a normal distribution and may require investigation.

---

## Intuition

Think of an easy exam.

Most students score high.

A few students score very low.

---

## Shape

```text
 ______
/      \__
          \\
           \\
```

Long tail →

```text
Left
```

---

## Example

Exam Marks:

```text
10
85
88
90
92
94
95
98
```

Most students scored high.

A few scored poorly.

---

## How to Interpret

Usually:

```text
Mean < Median
```

because small values pull the mean downward.

---

## Common Mistakes

❌ Confusing the direction of the tail.

Remember:

> **Skewness is named after the tail, NOT where most data lies.**

Right tail → Right skew.

Left tail → Left skew.

---

## Quick Comparison

| Distribution | Shape         | Mean vs Median | Example          |
| ------------ | ------------- | -------------- | ---------------- |
| Normal       | Symmetric     | Mean ≈ Median  | Heights          |
| Right Skew   | Tail on right | Mean > Median  | House Prices     |
| Left Skew    | Tail on left  | Mean < Median  | Easy Exam Scores |
