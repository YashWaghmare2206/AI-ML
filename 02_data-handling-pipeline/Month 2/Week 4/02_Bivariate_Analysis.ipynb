# Topic 2 → Bivariate Analysis

## Subtopic 2.1 → Numerical vs Numerical → Scatter Plot

---

# Short Definition

> A **Scatter Plot** is a graph used to study the relationship between two numerical variables.

Each point represents **one observation (one row)**.

---

# Why do we use it?

It helps answer questions like:

* Do the two variables move together?
* Is there a positive relationship?
* Is there a negative relationship?
* Is there no relationship?
* Are there clusters or outliers?

---

# Intuition

Suppose you want to know:

> "Do larger houses sell for higher prices?"

Instead of looking at numbers, you plot:

```text
House Size  → X-axis
Sale Price  → Y-axis
```

Each house becomes one dot.

---

# Example

House Prices Dataset:

| GrLivArea | SalePrice |
| --------: | --------: |
|      1710 |    208500 |
|      1262 |    181500 |
|      1786 |    223500 |
|      2198 |    250000 |

Scatter plot:

```text
SalePrice ↑
250k      •
220k    •
180k •
        ----------------→ GrLivArea
```

This suggests:

> Larger living area tends to have higher sale prices.

---

# Code

```python
sns.scatterplot(
    data=df,
    x="GrLivArea",
    y="SalePrice"
)

plt.show()
```

---

```text
5 → Red
7 → Blue
9 → Green
```

The plot becomes:

```text
SalePrice ↑

300k              🟢
250k        🔵
200k   🔴

      ----------------→ GrLivArea
```

Now you can see:

> "Do higher-quality houses have higher prices?"

without making another graph.

---

## If hue is categorical

Example:

```python
hue="CentralAir"
```

Suppose:

| CentralAir |
| ---------- |
| Y          |
| N          |

Then:

```text
Y → Blue
N → Orange
```

Plot:

```text
🔵 🔵 🔵 = Houses with AC
🟠 🟠 = Houses without AC
```

---

## If hue is numerical

Example:

```python
hue="OverallQual"
```

Suppose values:

```text
3 4 5 6 7 8 9
```

Then Seaborn gives a color gradient:

```text
Low Quality   High Quality
Light ─────────► Dark
```

So dots with larger values have different shades.

```text

```

Now you can see:

> "Do houses with more garage spaces tend to be larger and more expensive?"

---

# Short Definition

> **Correlation measures the strength and direction of the relationship between two numerical variables.**

It tells us:

> **When one variable changes, how does the other tend to change?**

---

# Why do we use it?

It helps answer questions like:

* Do larger houses usually have higher prices?
* Do two variables move together?
* How strong is their relationship?
* Which features are important for the target?

---

# Intuition

Suppose you observe:

### Study Hours vs Marks

| Study Hours | Marks |
| ----------: | ----: |
|           1 |    40 |
|           2 |    50 |
|           3 |    60 |
|           4 |    70 |

As study hours increase,

marks increase.

> They are positively correlated.

---

# Correlation Value

Correlation is represented by:

```text
r
```

Its value ranges from:

```text
-1  to  +1
```

---

## +1 → Perfect Positive Correlation

```text
•
  •
    •
      •
        •
```

Means:

> As X increases, Y always increases.

Example:

```text
X ↑ → Y ↑
```

---

## 0 → No Correlation

```text
 •    •
    •
•     •
   •
```

Means:

> X and Y are unrelated.

Example:

Shoe size vs exam marks.

---

## -1 → Perfect Negative Correlation

```text
        •
      •
    •
  •
•
```

Means:

> As X increases, Y always decreases.

Example:

```text
X ↑ → Y ↓
```

---

# Example (House Prices)

Suppose:

| GrLivArea | SalePrice |
| --------: | --------: |
|      1500 |    200000 |
|      1800 |    250000 |
|      2200 |    300000 |

Correlation:

```text
0.80
```

This means:

> Larger living areas generally have higher sale prices.

---

# Code

## Between Two Variables

```python
df["GrLivArea"].corr(df["SalePrice"])
```

Output:

```python
0.80
```

---

## Correlation Matrix

```python
df.corr(numeric_only=True)
```

This calculates correlations between **all numerical columns**.

---

# Important Parameters

## `other`

Used in:

```python
Series.corr(other)
```

Example:

```python
df["A"].corr(df["B"])
```

Possible values:

* Another numerical Series.

---

## `method`

Type of correlation.

Possible values:

```python
"pearson"
"spearman"
"kendall"
```

---

### Pearson (Default)

```python
method="pearson"
```

Measures:

> Linear relationships.

Most commonly used.

---

### Spearman

```python
method="spearman"
```

Measures:

> Monotonic relationships using ranks.

Useful when data is skewed or not linear.

---

### Kendall

```python
method="kendall"
```

Measures:

> Rank agreement.

Less commonly used.

---

# How to Interpret

| Correlation    | Interpretation         |
| -------------- | ---------------------- |
| 0.90 to 1.00   | Very Strong Positive   |
| 0.70 to 0.89   | Strong Positive        |
| 0.40 to 0.69   | Moderate Positive      |
| 0.10 to 0.39   | Weak Positive          |
| -0.09 to 0.09  | Little/No Relationship |
| -0.10 to -0.39 | Weak Negative          |
| -0.40 to -0.69 | Moderate Negative      |
| -0.70 to -0.89 | Strong Negative        |
| -0.90 to -1.00 | Very Strong Negative   |

---

# Common Mistakes

### Mistake 1

❌ Thinking correlation means causation.

Example:

Ice cream sales and drowning incidents may both increase in summer.

> Correlation ≠ Causation.

---

### Mistake 2

❌ Using correlation for categorical variables.

Correlation is mainly for:

> Numerical vs Numerical.

---

### Mistake 3

❌ Assuming correlation detects all relationships.

Pearson correlation mainly detects:

> Linear relationships.

It may miss curved relationships.

---

# 1-Line Summary

> **Correlation tells us how strongly and in which direction two numerical variables move together.**


### **Full Syntax**

````python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(
    method="pearson",
    min_periods=1,
    numeric_only=True
)

plt.figure(figsize=(12, 8))

sns.heatmap(
    data=corr,
    vmin=-1,
    vmax=1,
    center=0,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    linecolor="white",
    square=False,
    cbar=True,
    xticklabels=True,
    yticklabels=True
)

plt.title("Correlation Heatmap")
plt.xticks(rotation=45)
plt.yticks(rotation=0)

plt.show()
````

# 2.3.1 Scatter Plot (Feature vs Target)

## Short Definition

> A scatter plot between a numerical feature and the numerical target helps us understand how the feature affects the target.

---

## Why do we use it?

To answer questions like:

* Does this feature influence the target?
* Is the relationship positive or negative?
* Is it linear?
* Are there outliers?

---

## Example

House Prices:

| GrLivArea | SalePrice |
| --------: | --------: |
|      1500 |    200000 |

|      2200 |    300000 |

Scatter plot:

```text
SalePrice ↑

300k            •

      ----------------→ GrLivArea
```

# Entire Syntax


# Topic 2 → Bivariate Analysis

## Subtopic 2.4 → Categorical vs Numerical

### Bar Plot (Mean Target Values)

---

# Short Definition

> A **Bar Plot** compares a numerical variable across different categories by showing a summary statistic (usually the mean).

In EDA, it is often used to answer:

> **"How does the average target value differ across categories?"**

---

# Why do we use it?

It helps answer questions like:

* Which category has the highest average target?
* Which category has the lowest average target?
* Do categories differ significantly?
* Is this categorical feature useful?

---

# Intuition

Suppose you want to know:

> Which neighborhood has the highest average house price?

Instead of looking at thousands of houses individually, you calculate:

```text
Average SalePrice for each Neighborhood
```

and compare them using bars.

---

# Example (House Prices)

| Neighborhood | Mean SalePrice |
| ------------ | -------------: |
| NAmes        |         145000 |
| CollgCr      |         200000 |
| NridgHt      |         315000 |

Bar Plot:

```text
SalePrice ↑

315k     ████████████
200k     ███████
145k     █████

         ----------------→ Neighborhood
```

Interpretation:

> Houses in NridgHt are, on average, more expensive.

---

# Entire Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

sns.barplot(
    data=df,
    x="Neighborhood",
    y="SalePrice"
)

plt.xticks(rotation=45)

plt.title("Average SalePrice by Neighborhood")

plt.show()
```

---

# Important Parameters

## `data`

Dataset.

Example:

```python
data=df
```

---

## `x`

Categorical variable.

Example:

```python

```

Possible values:

* Any categorical column

Examples:

```python

x="Street"
x="Neighborhood"
```

---

## `y`

Numerical variable.

Example:

```python

```

Possible values:

Any numerical column.

---

## `estimator`

Function used to summarize the numerical variable.

Possible values:


---

# 2.5.1 Crosstab

## Short Definition

> A **Crosstab (Cross Tabulation)** is a table that shows how frequently combinations of two categorical variables occur.

---

# Why do we use it?

It helps answer questions like:

* How are two categories related?
* Which combinations occur most often?
* Are some category combinations rare?
* Is there an association between the variables?

---

# Intuition

Suppose you want to know:

> Do males and females choose different streams?

Instead of looking at individual students, you summarize:

```text
Gender × Stream
```

---

# Example

| Gender | Stream |
| ------ | ------ |
| Male   | AI     |
| Female | DS     |
| Male   | AI     |
| Female | AI     |

Crosstab:

|        | AI | DS |
| ------ | -: | -: |
| Male   |  2 |  0 |
| Female |  1 |  1 |

Interpretation:

> Most males chose AI.

---

# Entire Syntax

## Counts

```python
pd.crosstab(
    df["Gender"],
    df["Stream"]
)
```

---

## Row Percentages

```python
pd.crosstab(
    df["Gender"],
    df["Stream"],
    normalize="index"
) * 100
```

---

## Column Percentages

```python
pd.crosstab(
    df["Gender"],
    df["Stream"],
    normalize="columns"
) * 100
```

---

## Overall Percentages

```python
pd.crosstab(
    df["Gender"],
    df["Stream"],
    normalize=True
) * 100
```

---

# Important Parameters

| Parameter      | Meaning                | Values                                  | Default  |
| -------------- | ---------------------- | --------------------------------------- | -------- |
| `index`        | Row variable           | Categorical column                      | Required |
| `columns`      | Column variable        | Categorical column                      | Required |
| `normalize`    | Percentage calculation | `False`, `"index"`, `"columns"`, `True` | `False`  |
| `margins`      | Add totals             | `True`, `False`                         | `False`  |
| `margins_name` | Total row name         | String                                  | `"All"`  |

---

# How to Interpret

## High Counts

|      | AI | DS |
| ---- | -: | -: |
| Male | 50 |  5 |

Means:

> Most males prefer AI.

---

## Balanced Counts

|      | AI | DS |
| ---- | -: | -: |
| Male | 25 | 25 |

Means:

> No obvious preference.

---

# Common Mistakes

❌ Using numerical variables.

Crosstab is mainly for:

> **Categorical vs Categorical**

# 2.5.2 Heatmap of Percentages

## Short Definition

> A **Heatmap** converts the crosstab into colors so patterns are easier to see.

---

# Why do we use it?

Tables become difficult to read when there are many categories.

Heatmaps help you quickly identify:

* High-frequency combinations
* Low-frequency combinations
* Patterns and associations

---

# Example

Suppose:

|        |  AI |  DS |
| ------ | --: | --: |
| Male   | 80% | 20% |
| Female | 40% | 60% |

Heatmap:

```text
        AI      DS
Male    ████    █
Female  ██      ███
```

Darker color = Larger percentage.

---

# Entire Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

ct = pd.crosstab(
    df["Gender"],
    df["Stream"],
    normalize="index"
) * 100

plt.figure(figsize=(8,6))

sns.heatmap(
    ct,
    annot=True,
    fmt=".1f",
    cmap="Blues"
)

plt.show()
```

---

# Important Parameters

## `sns.heatmap()`

| Parameter    | Meaning         | Values                               | Default   |
| ------------ | --------------- | ------------------------------------ | --------- |
| `data`       | Crosstab matrix | DataFrame                            | Required  |
| `annot`      | Show values     | `True`, `False`                      | `False`   |
| `fmt`        | Number format   | `".1f"`, `".2f"`                     | `".2g"`   |
| `cmap`       | Color palette   | `"Blues"`, `"coolwarm"`, `"viridis"` | Default   |
| `linewidths` | Cell borders    | Number                               | `0`       |
| `linecolor`  | Border color    | Color name                           | `"white"` |
| `cbar`       | Show color bar  | `True`, `False`                      | `True`    |

---

# How to Interpret

## Dark Cells

```text
████
```

Means:

> Strong association / high percentage.

---

## Light Cells

```text
█
```

Means:

> Weak association / low percentage.

---

## Similar Shades

Means:

> Categories behave similarly.

---

# Common Mistakes

❌ Plotting raw counts when percentages are more meaningful.

Usually prefer:

```python
normalize="index"
```

for comparisons.

---

❌ Forgetting that color intensity depends on the chosen color scale.

---

# 1-Line Summary

> **Heatmaps visualize crosstab values using colors to reveal categorical relationships quickly.**
