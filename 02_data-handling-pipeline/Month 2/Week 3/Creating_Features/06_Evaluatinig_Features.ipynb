# Topic 6 — Evaluating New Features

A feature is not good because:

```text id="6v63dw"
It sounds smart
```

A feature is good because:

```text id="lqxj9j"
It improves prediction
```

---

# Complete Subtopics

## 6.1 Why Evaluate Features?

Example:

You create:

```text id="s31mkv"
Revenue = Price × Quantity
```

Looks useful.

But:

```text id="w2e4kj"
Did it actually help the model?
```

Need evaluation.

---

## 6.2 Correlation ⭐⭐⭐⭐

Measure linear relationship.

Questions answered:

```text id="9m7k9i"
Does feature move with target?
```

Examples:

```text id="p0igzl"
Hours Studied ↔ Marks

Area ↔ House Price
```

Learn:

* Positive Correlation
* Negative Correlation
* Near Zero Correlation

---

## 6.3 Mutual Information (MI) ⭐⭐⭐⭐⭐

One of the most important concepts.

Questions answered:

```text id="20frz1"
How much information
does this feature give
about the target?
```

Works even when:

```text id="v6uln2"
Relationship is not linear
```

Examples:

```text id="0r7b1k"
XGBoost
Feature Engineering
Kaggle
```

You'll use this a lot.

---

## 6.4 Feature Importance ⭐⭐⭐⭐⭐

After training a model:

```text id="4ob9v3"
Which features mattered most?
```

Example output:

| Feature        | Importance |
| -------------- | ---------- |
| Revenue        | 0.40       |
| Customer Count | 0.30       |
| Age            | 0.10       |

Very common in:

```text id="rjlwmz"
Random Forest
XGBoost
LightGBM
```

---

## 6.5 Permutation Importance ⭐⭐⭐⭐

Idea:

```text id="q6ih4z"
Break feature
↓
See performance drop
```

Large drop:

```text id="e7j6gu"
Important Feature
```

Small drop:

```text id="7oh09j"
Not Important
```

---

## 6.6 Model Comparison ⭐⭐⭐⭐⭐

Most practical evaluation.

Train:

```text id="lwtgcf"
Model A
Without New Feature
```

and

```text id="t0m5tr"
Model B
With New Feature
```

Compare:

```text id="5xqh52"
Accuracy

RMSE

MAE

F1
```

If performance improves:

```text id="z4s8ny"
Keep Feature
```

---

## 6.7 Removing Bad Features

Not every feature helps.

Remove:

```text id="9oov4o"
Noise Features

Redundant Features

Duplicated Features
```

---

## 6.8 Common Mistakes

### Mistake 1

Keeping features because they sound clever.

---

### Mistake 2

Using only correlation.

Sometimes:

```text id="vsz0w4"
Correlation = 0

Mutual Information > 0
```

Feature still useful.

---

### Mistake 3

Evaluating on training data only.

Must evaluate on:

```text id="a7imk7"
Validation Set
Test Set
```
