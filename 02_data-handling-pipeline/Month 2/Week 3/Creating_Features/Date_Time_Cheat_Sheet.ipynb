This is the **DateTime Feature Engineering Cheat Sheet** I would keep for ML projects.

Assume:

```python
df["Date"] = pd.to_datetime(df["Date"])
```

Then:

```python
df["Date"].dt
```

gives access to datetime attributes.

---

# Basic Date Components

| Feature | Code        | Example |
| ------- | ----------- | ------- |
| Year    | `dt.year`   | 2026    |
| Month   | `dt.month`  | 8       |
| Day     | `dt.day`    | 15      |
| Hour    | `dt.hour`   | 14      |
| Minute  | `dt.minute` | 30      |
| Second  | `dt.second` | 45      |

Example:

```python
df["Year"] = df["Date"].dt.year
```

---

# Week Features

| Feature      | Code                    | Example  |
| ------------ | ----------------------- | -------- |
| Day of Week  | `dt.dayofweek`          | Monday=0 |
| Day Name     | `dt.day_name()`         | Monday   |
| Week of Year | `dt.isocalendar().week` | 34       |
| Day of Year  | `dt.dayofyear`          | 227      |

Example:

```python
df["DayOfWeek"] = df["Date"].dt.dayofweek
```

---

# Month / Quarter Features

| Feature    | Code              |
| ---------- | ----------------- |
| Quarter    | `dt.quarter`      |
| Month      | `dt.month`        |
| Month Name | `dt.month_name()` |

Example:

```python
df["Quarter"] = df["Date"].dt.quarter
```

Result:

```text
Jan-Mar → 1
Apr-Jun → 2
Jul-Sep → 3
Oct-Dec → 4
```

---

# Boolean Features

Very common.

### Weekend

```python
df["IsWeekend"] = (
    df["Date"].dt.dayofweek >= 5
).astype(int)
```

Result:

```text
Saturday → 1
Sunday → 1
Monday → 0
```

# Elapsed Time Features ⭐⭐⭐⭐⭐

Most useful in ML.

### Days Since Event

```python
df["DaysSince"] = (
    pd.Timestamp.today()
    - df["Date"]
).dt.days
```

---

### Difference Between Two Dates

```python
df["Duration"] = (
    df["EndDate"]
    - df["StartDate"]
).dt.days
```

---

### Hours Difference

```python
df["Hours"] = (
    df["EndTime"]
    - df["StartTime"]
).dt.total_seconds()/3600
```

# Cyclical Encoding ⭐⭐⭐

For:

```text
Month
DayOfWeek
Hour
```

Use:

```python
np.sin(2*np.pi*x/period)

np.cos(2*np.pi*x/period)
```

Examples:

### Month

```python
df["MonthSin"] = np.sin(
    2*np.pi*df["Month"]/12
)

df["MonthCos"] = np.cos(
    2*np.pi*df["Month"]/12
)
```

---

### Hour

```python
df["HourSin"] = np.sin(
    2*np.pi*df["Hour"]/24
)

df["HourCos"] = np.cos(
    2*np.pi*df["Hour"]/24
)
```
