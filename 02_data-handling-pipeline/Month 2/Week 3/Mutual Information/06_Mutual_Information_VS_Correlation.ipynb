# **Mutual Information vs Correlation**

## **Subtopic 6.1:Where Correlation Fails But Mutual Information Succeeds**

### ML Lesson

**Suppose a feature has**:
* Correlation = 0

You might be tempted to remove it.

**But if:**
* MI = High

the feature may actually be very useful.

**This is why MI is often preferred for feature selectio**

**Key Takeaway**

A feature can have:
* Low Correlation
* High Mutual Information

when the relationship is non-linear.

This is the biggest advantage of Mutual Information over Correlation.

## **Subtopic 6.2: When Should You Use Correlation and When Should You Use Mutual Information?**

### **Use Correlation When**

You want to check:
* `"Is there a linear relationship?"`

**Examples:**
* Linear Regression
* Multicollinearity checking
* Quick EDA

| Study Hours | Marks |
| ----------- | ----- |
| 1           | 40    |
| 2           | 50    |
| 3           | 60    |

* Straight-line relationship.
* Correlation works well.

### **Use Mutual Information When**

You want to check:
* `"Does this feature contain useful information about the target?"`

**Examples:**

* Feature Selection
* Classification
* Regression
* Unknown relationships

MI can detect:
* Linear relationships
* Non-linear relationships

**Feature Engineering Rule**

If your goal is:
* `Which features should I keep?`

Use:
* `Mutual Information
* because that's exactly what it was designed for.`

### Simple Comparison

| Task                         | Better Choice      |
| ---------------------------- | ------------------ |
| Linear relationship analysis | Correlation        |
| Feature selection            | Mutual Information |
| Detect non-linear patterns   | Mutual Information |
| Quick EDA                    | Correlation        |
