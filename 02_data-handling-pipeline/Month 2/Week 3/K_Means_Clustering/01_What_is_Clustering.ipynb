# **Topic 1: What is Clustering?**

## **Subtopic 1: What is Unsupervised Learning?**

**Before understanding clustering, we first need to understand the type of learning K-Means belongs to.**

### **What is Unsupervised Learning?**

Unsupervised Learning is a type of Machine Learning in which the algorithm learns patterns from data without being given the correct answers (labels).

The algorithm receives only the input data and tries to discover hidden structures, relationships, or groups by itself.

### **Supervised vs Unsupervised (Basic Idea)**

**Supervised Learning**

The model is trained using:

* `Inputs (X) + Correct Answers (Y)`

Example:

| Hours Studied | Passed Exam |
| ------------- | ----------- |
| 2             | No          |
| 4             | Yes         |
| 6             | Yes         |


The model already knows the answers.

Its job is:

* `Learn the relationship between X and Y.`

**Unsupervised Learning**

The model is trained using:

* `Inputs (X) only`

Example:

| Person | Age | Income   |
| ------ | --- | -------- |
| A      | 25  | 30,000   |
| B      | 27  | 35,000   |
| C      | 55  | 2,00,000 |
| D      | 58  | 2,20,000 |


There is no column saying:

* `Young Group`
* `Rich Group`

The algorithm has to discover patterns itself.

Its job is:

* `Find hidden structures in the data.`

### **What does "No Labels" Mean?**

**Labeled Data**

| Animal | Legs | Can Fly | Label  |
| ------ | ---- | ------- | ------ |
| Dog    | 4    | No      | Mammal |
| Eagle  | 2    | Yes     | Bird   |

The answer is already provided.

**Unlabeled Data**

| Animal | Legs | Can Fly |
| ------ | ---- | ------- |
| ?      | 4    | No      |
| ?      | 2    | Yes     |

There is no label.

The algorithm tries to find patterns such as:

* `These animals seem similar.`
* `Those animals seem different.`

It creates groups based on similarity.

### **What Does Unsupervised Learning Try To Do?**

#### **1. Find Groups**

Example:

* `These customers behave similarly.`
* `Those customers behave differently.`

This is called:

* Clustering.

(K-Means belongs here.)

### **2. Find Hidden Relationships**

Example:

* `People who buy bread often buy butter.`

This is called:

* `Association.`

**(Not part of K-Means.)**

#### **3. Reduce Complexity**

Example:

* `Turning 100 features into 2 features while preserving information.`

This is called:

* `Dimensionality Reduction.`

**(Not part of K-Means.)**

## **Example Related to K-Means**

Suppose we have the following customer data:

| Customer | Income  | Spending Score |
| -------- | ------- | -------------- |
| A        | 30,000  | 20             |
| B        | 32,000  | 25             |
| C        | 150,000 | 80             |
| D        | 155,000 | 85             |


No one tells us:

* `A and B are Group 1.`
* `C and D are Group 2.`

K-Means will examine the similarities and may automatically discover:

**Cluster 1:**
* `A, B`

**Cluster 2:**
* `C, D`

Notice:
* `The algorithm created the groups by itself.`

## **Subtopic 2: Difference Between Classification and Clustering**

### First, the One-Line Difference

* `Classification predicts known labels.`
* `Clustering discovers unknown groups.`

### **Classification**

Classification is a supervised learning task where:

* `The model learns from data that already has labels and predicts those labels for new data.`

The algorithm knows beforehand:

* `What categories exist.`

Its job is:

* `Assign new observations to one of those known categories.`


**Example**

| Student | Hours Studied | Passed? |
| ------- | ------------- | ------- |
| A       | 2             | No      |
| B       | 4             | Yes     |
| C       | 6             | Yes     |
| D       | 1             | No      |

The labels are:

* `Yes`
* `No`

The algorithm learns:

* `Hours Studied → Passed?`

Now if a new student studies for 5 hours:

The model predicts:

* `Yes`

because it already knows the possible classes.

### **Clustering**

Clustering is an unsupervised learning task where:

* `The algorithm receives unlabeled data and tries to group similar observations together.`

The algorithm does NOT know:

* `How many groups exist,`
* `What they are called,`
* `Who belongs where.`

It must discover them itself.


#### **Example**

Suppose we have:

| Customer | Income  | Spending Score |
| -------- | ------- | -------------- |
| A        | 30,000  | 20             |
| B        | 32,000  | 25             |
| C        | 150,000 | 80             |
| D        | 155,000 | 85             |


No labels such as:
* `Low-value customer`
* `Premium customer`

exist.

**K-Means may discover:**

Cluster 1:
* `A, B`

Cluster 2:
* `C, D`

without anyone telling it.

### **Clustering Algorithms**

Examples include:

* K-Means ⭐
* Hierarchical Clustering
* DBSCAN
* Gaussian Mixture Models (GMM)

| Aspect                       | Classification                     | Clustering            |
| ---------------------------- | ---------------------------------- | --------------------- |
| Learning Type                | Supervised                         | Unsupervised          |
| Labels Available?            | Yes                                | No                    |
| Goal                         | Predict labels                     | Discover groups       |
| Output                       | Known classes                      | Clusters              |
| Categories Known Beforehand? | Yes                                | No                    |
| Example                      | Spam detection                     | Customer segmentation |
| Example Algorithms           | Logistic Regression, Random Forest | K-Means, DBSCAN       |


## **Subtopic 3: What is Clustering?**

### **What is Clustering?**

Definition

* `Clustering is the process of grouping similar data points together so that points within the same group are more similar to each other than to points in other groups.`

In simple words:

* `Clustering means finding natural groups in data.`

### **The Main Idea**

Imagine you have many data points.

You don't know:

* How many groups exist.
* What the groups should be called.
* Which point belongs to which group.

You only know:

* Some points look similar.
* Some points look different.

The clustering algorithm's job is:

* **`Automatically discover these groups.`**

### **What Does "Similar" Mean?**

This is the most important concept in clustering.

Clustering works based on the idea of similarity.

Usually:

* `Points that are closer together are considered more similar.`
* `Points that are farther apart are considered less similar.`

For K-Means specifically:

* `Similarity is measured mainly using Euclidean distance (which we'll study later).`

### Visual Intuition

Suppose we plot points on a graph.

We observe:

● ● ●

               ● ● ●

                           ● ● ●

Even without labels, our brain naturally thinks:

* These three belong together.
* Those three belong together.
* Those three belong together.

So we see:
* `3 groups.`

**This natural grouping process is exactly what clustering algorithms try to imitate.**

### **Example: Customer Data**

| Customer | Income | Spending Score |
| -------- | ------ | -------------- |
| A        | 30K    | 20             |
| B        | 35K    | 25             |
| C        | 150K   | 85             |
| D        | 155K   | 90             |
| E        | 80K    | 50             |
| F        | 85K    | 55             |

A clustering algorithm may discover:

Cluster 1:
* `Low income, low spending`

Cluster 2:
* `Middle income, medium spending`

Cluster 3:
* `High income, high spending`

Notice:

**The algorithm creates the groups first.**

Humans interpret them afterward.

### **Clustering Does NOT Predict**

Many beginners confuse clustering with prediction.

Clustering is NOT asking:

* Will this customer buy?
* Will this patient survive?
* Will this email be spam?
*
Those are prediction problems.

Instead, clustering asks:

* `What groups naturally exist?`

It is about discovery, not prediction.

### **Why Is Clustering Useful?**

Clustering helps when:

* We have lots of data,
* but no labels.

Instead of manually inspecting thousands or millions of records:

**The algorithm helps us organize the data.**

### **K-Means and Clustering**

K-Means is simply:

* `One method of performing clustering.`

It tries to create clusters by grouping points around centers called centroids.

We'll study centroids in detail soon.

For now, remember:

* **Clustering = The task**
* **K-Means = One algorithm that performs that task**

Just like:

* **Classification = Task**
*
* **Logistic Regression = One algorithm**
* **Decision Tree = Another algorithm**

The answer is:

**Different clustering algorithms use different approaches to form groups.**

Think of clustering as a problem:

* `Find natural groups in data.`

There are multiple ways to solve this problem.

### **Major Types of Clustering**

#### **1. Partition-Based Clustering**


**The algorithm divides the data into a fixed number of groups.**

Example:

Suppose:

* `K = 3`

The algorithm tries to partition the dataset into:

* Cluster 1
* Cluster 2
* Cluster 3

Characteristics

1. You specify the number of clusters beforehand.
2. Each point belongs to exactly one cluster.
3. Tries to make clusters compact.

# 1. Hierarchical Clustering

### Short Definition

> Builds clusters step-by-step by either merging smaller clusters or splitting larger ones.

Think of it as creating a **family tree of clusters**.

### Example

Students' marks:

```text
A = 90
B = 91
C = 45
D = 47
```

Starts as:

```text
A   B   C   D
```

Then merges:

```text
(A,B)   (C,D)
```

Finally:

```text
((A,B),(C,D))
```

---

# 2. DBSCAN (Density-Based Clustering)

### Short Definition

> Forms clusters based on regions where points are densely packed together and treats isolated points as noise.

### Example

```text
● ● ● ●

● ● ●

          x

                 ● ● ●
```

DBSCAN says:

```text
Cluster 1 → Left group
Cluster 2 → Right group
x → Noise (Outlier)
```

Unlike K-Means, it doesn't force `x` into a cluster.

---

# 3. Gaussian Mixture Models (GMM)

### Short Definition

> A probabilistic clustering method that assigns probabilities of belonging to each cluster instead of making a hard assignment.

### Example

Instead of saying:

```text
Customer A → Cluster 1
```

GMM says:

```text
Customer A:
80% Cluster 1
20% Cluster 2
```

The customer may partially belong to multiple groups.

---

# One-Line Summary

```text
K-Means      → Divide data into K fixed groups.

Hierarchical → Build a tree of clusters by merging/splitting.

DBSCAN       → Find dense regions and detect noise.

GMM          → Assign probabilities to clusters.
```

---

# Easy Way to Remember

```text
K-Means      → "Split into K groups."

Hierarchical → "Build a family tree."

DBSCAN       → "Find crowds, ignore loners."

GMM          → "How likely is this point to belong here?"
```
