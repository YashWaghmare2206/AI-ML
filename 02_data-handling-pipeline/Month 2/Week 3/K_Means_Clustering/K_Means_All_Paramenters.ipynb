# KMeans Parameters (scikit-learn)

Suppose you write:

```python
KMeans(...)
```

These are the important parameters you should know.

---

# 1. `n_clusters`

## Meaning

```text
Number of clusters (K).
```

Example:

```python
KMeans(n_clusters=3)
```

means:

```text
Create 3 clusters.
```

---

## Possible values

```text
Any positive integer ≥ 1
```

Examples:

```python
n_clusters=2
n_clusters=5
n_clusters=10
```

---

## Default

```python
n_clusters=8
```

---

# 2. `init`

## Meaning

```text
How initial centroids are chosen.
```

---

## Possible values

### a) `'k-means++'` ⭐ Recommended

```python
init='k-means++'
```

Meaning:

```text
Choose first centroid randomly.
Choose remaining centroids far apart.
```

Default.

---

### b) `'random'`

```python
init='random'
```

Meaning:

```text
Choose all centroids randomly.
```

Higher chance of bad initialization.

---

### c) Array-like

Example:

```python
init=[[1,2],[3,4]]
```

Meaning:

```text
You provide initial centroid coordinates.
```

Rarely used.

---

# 3. `n_init`

## Meaning

```text
How many times K-Means runs
with different initializations.
```

Example:

```python
n_init=10
```

means:

```text
Run K-Means 10 times.
Keep the solution with lowest WCSS.
```

---

## Possible values

```text
Positive integers
```

Examples:

```python
n_init=1
n_init=10
n_init=20
```

---

## Default

Recent sklearn versions:

```python
n_init='auto'
```

which internally chooses a suitable value.

---

# 4. `max_iter`

## Meaning

```text
Maximum iterations allowed
for one K-Means run.
```

Example:

```python
max_iter=300
```

means:

```text
Stop after 300 iterations
even if convergence hasn't happened.
```

---

## Possible values

```text
Positive integers
```

Examples:

```python
max_iter=100
max_iter=300
max_iter=1000
```

---

## Default

```python
300
```

---

# 5. `tol`

## Meaning

```text
Tolerance for convergence.
```

It asks:

```text
Have centroids moved so little
that we can stop?
```

---

## Smaller tol

Example:

```python
tol=0.000001
```

Means:

```text
Be very strict.
```

May take more iterations.

---

## Larger tol

Example:

```python
tol=0.1
```

Means:

```text
Stop sooner.
```

Less precise.

---

## Possible values

```text
Positive floats
```

---

## Default

```python
0.0001
```

---

# 6. `random_state`

## Meaning

```text
Controls randomness.
```

Example:

```python
random_state=42
```

Means:

```text
Same initialization
↓
Same results
```

---

## Possible values

```text
None
Any integer
```

Examples:

```python
random_state=None
random_state=42
random_state=0
random_state=100
```

---

## Default

```python
None
```

---

# 7. `algorithm`

## Meaning

```text
Which implementation of K-Means to use.
```

---

## Possible values

### a) `'lloyd'`

```python
algorithm='lloyd'
```

Standard K-Means algorithm.

Default.

---

### b) `'elkan'`

```python
algorithm='elkan'
```

Uses optimization tricks.

Often faster on dense datasets.

Uses more memory.

---

## Default

```python
'lloyd'
```

---

# Complete Example

```python
kmeans = KMeans(
    n_clusters=3,
    init='k-means++',
    n_init='auto',
    max_iter=300,
    tol=0.0001,
    random_state=42,
    algorithm='lloyd'
)
```

---

# Which Ones Matter Most?

For beginners:

```text
⭐⭐⭐⭐⭐ n_clusters
⭐⭐⭐⭐⭐ init
⭐⭐⭐⭐⭐ n_init
⭐⭐⭐⭐  random_state
⭐⭐⭐   max_iter
⭐⭐⭐   tol
⭐⭐    algorithm
```

---

# Mini Summary

| Parameter      | Meaning                 | Common Values             |
| -------------- | ----------------------- | ------------------------- |
| `n_clusters`   | Number of clusters      | 2, 3, 5, 10               |
| `init`         | Centroid initialization | `'k-means++'`, `'random'` |
| `n_init`       | Number of runs          | `'auto'`, 10, 20          |
| `max_iter`     | Max iterations          | 100, 300, 1000            |
| `tol`          | Convergence tolerance   | 0.0001                    |
| `random_state` | Reproducibility         | None, 42, 0               |
| `algorithm`    | Implementation          | `'lloyd'`, `'elkan'`      |

---

## One-line intuition

```text
n_clusters decides WHAT to find,
init decides HOW to start,
n_init decides HOW MANY chances K-Means gets,
and the remaining parameters control WHEN and HOW it stops.
```
