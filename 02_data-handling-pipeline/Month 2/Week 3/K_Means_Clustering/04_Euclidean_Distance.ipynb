
# Topic 4: Distance Metrics

## Subtopic 1: Euclidean Distance

## What is Euclidean Distance?

> **Euclidean Distance is the straight-line distance between two points.**

K-Means uses Euclidean distance to answer:

```text id="8t2e6n"
"Which centroid is closest to this point?"
```

The point is assigned to the cluster of the nearest centroid.

---

## 1-D Example

Points:

```text id="n1v7q3"
2 and 7
```

Distance:

```text id="r4k8m1"
|7 − 2| = 5
```

So, the Euclidean distance is:

```text id="x9p2j6"
5
```

---

## 2-D Example

Points:

```text id="y3m8q1"
A = (2, 3)
B = (5, 7)
```

Formula:



Calculation:

```text id="t6n4v8"
= √[(5−2)² + (7−3)²]
= √(3² + 4²)
= √25
= 5
```

Distance:

```text id="p8k5m2"
5
```

---

## 3-D Example

Points:

```text id="q7v3n1"
A = (1, 2, 3)
B = (4, 6, 3)
```

Formula:

d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2+(z_2-z_1)^2}

Calculation:

```text id="m2r8k4"
= √[(4−1)² + (6−2)² + (3−3)²]
= √(3² + 4² + 0²)
= √25
= 5
```

Distance:

```text id="j5p9v6"
5
```

---

## Mini Summary

```text id="v4n7q2"
Euclidean Distance:
• Straight-line distance between points.
• Used by K-Means to find the nearest centroid.

1-D:
|x₂ − x₁|

2-D:
√[(x₂−x₁)² + (y₂−y₁)²]

3-D:
√[(x₂−x₁)² + (y₂−y₁)² + (z₂−z₁)²]
```

# Topic 4: Distance Metrics

## Subtopic 2: Manhattan Distance (Awareness Only)

---

## What is Manhattan Distance?

> **Manhattan Distance is the distance between two points calculated by moving only horizontally and vertically.**

Unlike Euclidean distance (straight-line distance),

> Manhattan distance measures the **total movement along each axis**.

K-Means mainly uses **Euclidean distance**, but it's useful to know Manhattan distance exists.

---

## 1-D Example

Points:

```text id="m1n4k7"
2 and 7
```

Distance:

```text id="v8q2p5"
|7 − 2| = 5
```

Manhattan Distance:

```text id="x3r9m1"
5
```

In 1-D, Manhattan and Euclidean distances are the same.

---

## 2-D Example

Points:

```text id="j6k2v8"
A = (2, 3)
B = (5, 7)
```

Formula:

```text id="p4m7q2"
|x₂ − x₁| + |y₂ − y₁|
```

Calculation:

```text id="t9n3v5"
= |5−2| + |7−3|
= 3 + 4
= 7
```

Distance:

```text id="r5k8m4"
7
```

---

## 3-D Example

Points:

```text id="q2v6n9"
A = (1, 2, 3)
B = (4, 6, 3)
```

Formula:

```text id="m7p1k3"
|x₂−x₁| + |y₂−y₁| + |z₂−z₁|
```

Calculation:

```text id="v4r8q2"
= |4−1| + |6−2| + |3−3|
= 3 + 4 + 0
= 7
```

Distance:

```text id="k9m5p2"
7
```

---

## Euclidean vs Manhattan (Same Examples)

For the points used above:

| Dimensions | Euclidean | Manhattan |
| ---------- | --------: | --------: |
| 1-D        |         5 |         5 |
| 2-D        |         5 |         7 |
| 3-D        |         5 |         7 |

---

## Important Point

For K-Means:

> **Euclidean Distance is the standard choice.**

Manhattan distance is mainly discussed for awareness and because other algorithms may use it.

---

## Mini Summary

```text id="w6n3p8"
Manhattan Distance:
• Movement along axes only.
• Add absolute differences.

1-D:
|x₂ − x₁|

2-D:
|x₂−x₁| + |y₂−y₁|

3-D:
|x₂−x₁| + |y₂−y₁| + |z₂−z₁|

K-Means:
Primarily uses Euclidean Distance, not Manhattan Distance.
```

### Key Difference

```text id="y2m8q4"
Euclidean  → Straight-line distance.

Manhattan → Horizontal + Vertical movement.
```

## Subtopic 3: Effect of Feature Scaling

This is one of the **most important practical concepts** in K-Means.

---

# What is Feature Scaling?

> **Feature Scaling means bringing features to a similar range before applying K-Means.**

In simple words:

> Make sure one feature doesn't dominate the distance calculation just because its values are numerically larger.

---

# Why Is Feature Scaling Important?

Remember:

K-Means uses:

```text id="a1b2c3"
Distance
↓
Nearest Centroid
↓
Cluster Assignment
```

If one feature has much larger values than others:

> That feature will dominate the distance calculation.

As a result:

> K-Means will mostly cluster based on that feature and almost ignore the others.

---

# Example

Suppose we have two features:

| Person | Age |  Salary |
| ------ | --: | ------: |
| A      |  25 |  30,000 |
| B      |  35 | 500,000 |

Differences:

```text id="d4e5f6"
Age Difference     = 10
Salary Difference  = 470,000
```

Notice:

```text id="g7h8i9"
470,000 >> 10
```

So when Euclidean distance is calculated:

> The salary difference dominates.

K-Means will behave almost as if:

```text id="j1k2l3"
Only Salary Matters
```

even though Age might also be important.

---

# What Does Scaling Do?

Scaling transforms features so that they have comparable ranges.

Instead of:

```text id="m4n5o6"
Age:     20–60
Salary:  20,000–5,000,000
```

we might get:

```text id="p7q8r9"
Age:     0–1
Salary:  0–1
```

Now:

> Both features contribute fairly to distance calculations.

---

# Common Scaling Methods

## 1. Standardization ⭐⭐⭐⭐⭐

Formula:

```text id="s1t2u3"
(New Value) =
(Value − Mean) / Standard Deviation
```

Result:

```text id="v4w5x6"
Mean ≈ 0
Standard Deviation ≈ 1
```

Commonly used with K-Means.

---

## 2. Min-Max Scaling ⭐⭐⭐⭐

Formula:

```text id="y7z8a9"
(New Value) =
(Value − Minimum) / (Maximum − Minimum)
```

Result:

```text id="b1c2d3"
Values lie between 0 and 1
```

Also commonly used.

---

# Which One Should You Use?

For K-Means:

> **Standardization is the most common choice.**

However:

> Min-Max Scaling also works well.

The important thing is:

```text id="e4f5g6"
Scale the features.
```

Not scaling is often worse than the choice of scaler itself.

---

# What Happens If You Don't Scale?

```text id="h7i8j9"
Large-scale features dominate
↓
Distances become distorted
↓
Wrong nearest centroids
↓
Poor clusters
```

---

# Most Important Takeaway

> **Because K-Means depends on distance, and distance depends on feature values, K-Means is highly sensitive to feature scales.**

Therefore:

```text id="k1l2m3"
Scale first
↓
Then apply K-Means
```

---

# Mini Summary

```text id="n4o5p6"
Feature Scaling:

• Makes features comparable.
• Prevents large-valued features from dominating.
• Very important for K-Means.

Without Scaling:
Wrong distances → Poor clusters

Common Methods:
• Standardization
• Min-Max Scaling

Practical Rule:
Scale your data before running K-Means.
```
