# Topic 13: K-Means++ Initialization

You already know the motivation:

```text
Random initialization
↓
Bad starting centroids
↓
Poor local minima
```

K-Means++ was invented to reduce this problem.

---

# Why was K-Means++ needed?

Suppose the true clusters are:

```text
● ● ● ●        ○ ○ ○ ○
● ● ● ●        ○ ○ ○ ○
```

---

## Original K-Means

For K = 2:

Possible initialization:

```text
× ×
● ● ● ●        ○ ○ ○ ○
```

Both centroids start in the same region.

This can lead to:

```text
Poor clustering
↓
Higher WCSS
↓
Poor local minimum
```

---

# K-Means++ Idea

Instead of choosing all centroids randomly:

```text
Choose them intelligently.
```

The idea is:

```text
Initial centroids should be spread out.
```

---

# Step 1: Choose First Centroid

Choose one data point randomly.

Example:

```text
● ● ● ●        ○ ○ ○ ○

×
```

Nothing special yet.

---

# Step 2: Measure Distance

For every remaining point:

```text
Compute distance to the nearest chosen centroid.
```

Points far away get:

```text
Large distance.
```

Points nearby get:

```text
Small distance.
```

Example:

```text
Chosen centroid:

×

Nearby points:
Small distances

Far cluster:
Large distances
```

---

# Step 3: Choose Next Centroid

Choose the next centroid with probability proportional to distance².

Meaning:

```text
Farther points
↓
Higher chance of being selected.

Closer points
↓
Lower chance of being selected.
```

---

## Important Clarification

K-Means++ does NOT say:

```text
Always choose the farthest point.
```

It says:

```text
Prefer farther points.
```

So it is still somewhat random.

---

Each new centroid prefers regions far from existing centroids.

# Step 5: Run Normal K-Means

After initialization:

```text
Assignment
↓
Update
↓
Assignment
↓
Update
↓
Convergence
```

Exactly the same algorithm you already learned.

---

# Does K-Means++ Guarantee the Global Minimum?

```text
No.
```

It only makes it:

```text
More likely to find a better solution.
```

It reduces the chance of:

```text
Poor local minima.
```

but cannot eliminate them completely.

---

# K-Means vs K-Means++

| Step                         | K-Means | K-Means++               |
| ---------------------------- | ------- | ----------------------- |
| First centroid               | Random  | Random                  |
| Remaining centroids          | Random  | Prefer far-apart points |
| Main algorithm               | Same    | Same                    |
| Chance of bad initialization | Higher  | Lower                   |
| Guarantees global minimum?   | No      | No                      |

---

# Why is K-Means++ Better?

Because it avoids situations like:

```text
× ×
● ● ● ●        ○ ○ ○ ○
```

and encourages:

```text
×               ×
● ● ● ●        ○ ○ ○ ○
```

which gives K-Means a much better starting point.

---

# Mini Summary

```text
K-Means++ improves only initialization.

Process:

Choose first centroid randomly
↓
Compute distances to existing centroids
↓
Prefer farther points as new centroids
↓
Repeat until K centroids
↓
Run normal K-Means

Benefits:
• Better starting centroids
• Lower chance of poor local minima
• Often lower WCSS

Limitations:
• Does not guarantee the global minimum.
```

---

## One-line intuition

```text
K-Means++ says:
"Before clustering, let's spread out the starting centroids so K-Means doesn't begin with a bad guess."
```
