---

# What is Silhouette Score?

It measures:

```text id="g7b9rl"
How well each point fits into its own cluster
compared to other clusters.
```

In simple words, it asks:

```text id="b6wv5x"
"Is this point in the right cluster?"
```

---

# The Main Idea

For each point, we compare two distances.

## a = Cohesion

```text id="shm7rr"
Average distance to points
inside its own cluster.
```

Meaning:

```text id="c7j22e"
How close am I to my cluster friends?
```

Small is good.

---

## b = Separation

```text id="gtztg0"
Average distance to points
in the nearest other cluster.
```

Meaning:

```text id="jlwm45"
How far am I from the neighboring cluster?
```

Large is good.

---

# The Ideal Situation

You want:

```text id="itnmsr"
a → Small
b → Large
```

Meaning:

```text id="2mh0wq"
Close to own cluster
Far from other clusters
```

---

# Formula

s=\frac{b-a}{\max(a,b)}

Where:

```text id="7vwdgf"
a = Average distance to own cluster
b = Average distance to nearest other cluster
```

Don't memorize it.

Focus on the intuition.

---

# What values can it take?

Silhouette Score ranges from:

```text id="ibw4yi"
−1 to +1
```

---

# Case 1: Close to +1 ⭐⭐⭐

Suppose:

```text id="axu6mt"
a = 2
b = 10
```

Meaning:

```text id="n94x0t"
Very close to own cluster
Very far from others
```

Score:

```text id="7eclqw"
Near +1
```

Interpretation:

```text id="tmd07z"
Excellent clustering.
```

---

# Case 2: Around 0

Suppose:

```text id="pjxjlwm"
a ≈ b
```

Meaning:

```text id="vsjlwm"
The point is almost equally close
to both clusters.
```

Interpretation:

```text id="hyjlwm"
Point lies near the boundary.
```

Clusters overlap.

---

# Case 3: Negative Score

Suppose:

```text id="8djlwm"
a > b
```

Meaning:

```text id="3kjlwm"
The point is closer to another cluster
than its own cluster.
```

Interpretation:

```text id="5rjlwm"
Probably assigned to the wrong cluster.
```

---

# Visual Intuition

## Good Point

```text id="4mjlwm"
● ● ●
● X ●

           ○ ○ ○
           ○ ○ ○
```

X is close to its own cluster.

Silhouette:

```text id="jlwm10"
Near +1
```

---

## Boundary Point

```text id="1wjlwm"
● ● X ○ ○
```

X is between clusters.

Silhouette:

```text id="jlwm11"
Around 0
```

---

## Bad Point

```text id="jlwm12"
● ●

         X ○ ○ ○
```

X belongs more naturally with the ○ cluster.

Silhouette:

```text id="jlwm13"
Negative
```

---

# Overall Silhouette Score

We calculate:

```text id="jlwm14"
Silhouette score of every point
↓
Take the average
↓
Get overall silhouette score
```

---

# Choosing K using Silhouette

Try different K values.

Example:

| K | Silhouette Score |
| - | ---------------: |
| 2 |             0.45 |
| 3 |             0.62 |
| 4 |             0.55 |
| 5 |             0.41 |

Choose:

```text id="jlwm15"
K = 3
```

because:

```text id="jlwm16"
It has the highest score.
```

---

# Elbow vs Silhouette

| Elbow Method      | Silhouette Score         |
| ----------------- | ------------------------ |
| Uses WCSS         | Uses point relationships |
| Looks for a bend  | Looks for highest score  |
| Can be subjective | More quantitative        |
| Faster            | Slightly more expensive  |

---

# Mini Summary

```text id="jlwm17"
Silhouette Score measures:

How well a point fits in its own cluster
compared to other clusters.

a = Distance to own cluster
b = Distance to nearest other cluster

Score range:
−1 to +1

Near +1 → Good clustering
Around 0 → Boundary point
Negative → Probably wrong cluster

Choose K with the highest average silhouette score.
```
