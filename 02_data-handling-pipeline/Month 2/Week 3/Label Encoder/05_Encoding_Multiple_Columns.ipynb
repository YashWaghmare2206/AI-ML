# Topic 5: Encoding Multiple Columns

## Subtopic 5.1 – Applying Label Encoding to a Single Column

---

## Example

Suppose we have:

```python
import pandas as pd

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"],
    "Age": [25, 30, 22, 28]
})

print(df)
```

### Output

```text
   Gender  Age
0    Male   25
1  Female   30
2    Male   22
3  Female   28
```

---

## Step 1: Import and create the encoder

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
```

---

## Step 2: Encode only the `Gender` column

```python
df["Gender"] = encoder.fit_transform(df["Gender"])

print(df)
```

### Output

```text
   Gender  Age
0        1   25
1        0   30
2        1   22
3        0   28
```

# Topic 5: Encoding Multiple Columns

## Subtopic 5.2 – Applying Label Encoding to Multiple Categorical Columns

```
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])
df["City"] = encoder.fit_transform(df["City"])

print(df)
```

### Output (continued)

```text
   Gender  City  Age
0        1     1   25
1        0     0   30
2        1     2   22
3        0     0   28
```

---

## What happened?

### `Gender` Column

Original:

```text
Male
Female
Male
Female
```

Alphabetical order:

```text
Female
Male
```

Mapping:

```text
Female → 0
Male   → 1
```

Encoded Gender:

```text
[1, 0, 1, 0]
```

---

### `City` Column

Original:

```text
Mumbai
Delhi
Pune
Delhi
```

Alphabetical order:

```text
Delhi
Mumbai
Pune
```

Mapping:

```text
Delhi  → 0
Mumbai → 1
Pune   → 2
```

Encoded City:

```text
[1, 0, 2, 0]
```

---

# Important Point

Notice this code:

```python
encoder.fit_transform(df["Gender"])
encoder.fit_transform(df["City"])
```

We're using the **same variable name** (`encoder`), but each `fit_transform()` creates a **new mapping** because `fit()` relearns from that column.

So after encoding `City`, the encoder forgets the `Gender` mapping.

This is fine if you only want the encoded DataFrame.

However, if you later need to decode values using `inverse_transform()`, you should use **separate encoders**.

Example:

```python
gender_encoder = LabelEncoder()
city_encoder = LabelEncoder()

df["Gender"] = gender_encoder.fit_transform(df["Gender"])
df["City"] = city_encoder.fit_transform(df["City"])
```

Now:

```python
gender_encoder.inverse_transform([1])
```

gives:

```text
['Male']
```

and

```python
city_encoder.inverse_transform([1])
```

gives:

```text
['Mumbai']
```

# Subtopic 5.3 – Using Loops for Multiple Columns

Suppose you have many categorical columns:

| Gender | City | Department | Age |
| ------ | ---- | ---------- | --: |

Writing this repeatedly:

```python
df["Gender"] = ...
df["City"] = ...
df["Department"] = ...
```

becomes tedious.

Instead, use a loop.

---

## Example

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male"],
    "City": ["Mumbai", "Delhi", "Pune"],
    "Department": ["IT", "HR", "Sales"]
})

categorical_cols = ["Gender", "City", "Department"]

encoder = LabelEncoder()

for col in categorical_cols:
    df[col] = encoder.fit_transform(df[col])

print(df)
```

### Output

```text
   Gender  City  Department
0        1     1           0
1        0     0           1
2        1     2           2
```

---
---

# A Better Practice

If you need to preserve encoders:

```python
encoders = {}

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le
```

Now:

```python
encoders["Gender"]
encoders["City"]
encoders["Department"]
```

store separate mappings.

---

# Key Takeaways

### Multiple Columns

* Encode each categorical column separately.
* Numerical columns should not be encoded.

### Loops

* Useful when there are many categorical columns.
* Reduce repetitive code.

### Separate Encoders

* Use them if you need `inverse_transform()` later.
