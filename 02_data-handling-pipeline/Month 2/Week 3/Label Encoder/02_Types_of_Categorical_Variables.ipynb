# Topic 2: Types of Categorical Variables

## Subtopic 2.1 – Ordinal Variables

---

## 1. What Are Ordinal Variables?

**Ordinal variables are categorical variables whose categories have a natural order or ranking.**

In simple words:

> **Ordinal data = Categories + Meaningful Order**

The categories are labels, but one category is higher, lower, better, or worse than another.

---

## 2. Understanding "Natural Order"

Natural order means the categories can be arranged in a logical sequence.

For example, consider exam grades:

| Grade     |
| --------- |
| Poor      |
| Average   |
| Good      |
| Excellent |

We know that:

```text
Poor < Average < Good < Excellent
```

The categories have an order.

---

## 3. Examples of Ordinal Variables

### Example 1: Education Level

| Education Level |
| --------------- |
| High School     |
| Bachelor's      |
| Master's        |
| PhD             |

Natural order:

```text
High School
↓
Bachelor's
↓
Master's
↓
PhD
```

---

## 4. What Makes Ordinal Variables Special?

Two important characteristics:

### They are categorical.

For example:

```text
Low
Medium
High
```

These are labels, not actual measurements.

---

### They have order.

We know:

```text
Low < Medium < High
```

Even though we don't know the exact difference between them.

---

## 5. Can We Measure the Difference?

This is very important.

Suppose:

```text
Low    → 0
Medium → 1
High   → 2
```

We know:

```text
High > Medium > Low
```

But we **do not know** whether:

```text
Medium - Low
```

is equal to

```text
High - Medium
```

The distances between categories are unknown.

---

### Example

Movie ratings:

| Rating  |
| ------- |
| Bad     |
| Average |
| Good    |

We know:

```text
Bad < Average < Good
```

But we cannot say:

> The improvement from Bad to Average is exactly the same as Average to Good.

Only the order matters.

---

## 6. Real-World Analogy

Imagine climbing stairs.

Each step represents a category:

```text
Step 1 → Beginner
Step 2 → Intermediate
Step 3 → Advanced
```

You know that:

```text
Advanced > Intermediate > Beginner
```

However, the effort required between steps may not always be equal.

The important thing is the sequence.

---

## 7. Why Is Label Encoding Suitable for Ordinal Variables?

Because Label Encoding preserves the order.

Example:

### Original Categories

| Experience Level |
| ---------------- |
| Beginner         |
| Intermediate     |
| Advanced         |

### Encoded Values

| Experience Level | Encoded |
| ---------------- | ------: |
| Beginner         |       0 |
| Intermediate     |       1 |
| Advanced         |       2 |

This encoding reflects the actual ranking.

Since the categories naturally progress, the assigned numbers make sense.

---

## 8. Python Example

Suppose we have customer satisfaction levels.

### Original Data

```python
import pandas as pd

data = pd.DataFrame({
    "Satisfaction": [
        "Low",
        "Medium",
        "High",
        "Medium",
        "Low"
    ]
})

print(data)
```

### Output

```text
  Satisfaction
0          Low
1       Medium
2         High
3       Medium
4          Low
```

---

### Manual Ordinal Encoding

```python
mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

data["Satisfaction"] = data["Satisfaction"].map(mapping)

print(data)
```

### Output

```text
   Satisfaction
0             0
1             1
2             2
3             1
4             0
```

---

## Output Explanation

The categories were replaced using:

```text
Low    → 0
Medium → 1
High   → 2
```

The order remains preserved:

```text
0 < 1 < 2
```

which matches:

```text
Low < Medium < High
```

---

## 9. Common Mistake

A common mistake is assuming that ordinal variables behave like numerical variables.

For example:

```text
Low    → 0
Medium → 1
High   → 2
```

This does **not** mean:

```text
High is twice as much as Medium.
```

The numbers only preserve order.

They do not represent actual quantities.

---

## Important Points to Remember

* Ordinal variables are categorical variables.
* They have a meaningful order or ranking.
* The exact difference between categories is unknown.
* Label Encoding is often suitable for ordinal variables because it preserves the order.
* Examples include education level, satisfaction ratings, and clothing sizes.

---

## Summary

> **Ordinal variables are categorical variables with a natural order.**

Examples:

```text
Low < Medium < High
```

# Topic 2: Types of Categorical Variables

## Subtopic 2.2 – Nominal Variables

---

## 1. What Are Nominal Variables?

**Nominal variables are categorical variables whose categories do NOT have a natural order or ranking.**

In simple words:

> **Nominal data = Categories without any meaningful order**

The categories are just names or labels.

---

## 2. Understanding "No Natural Order"

Let's look at this example:

| Color |
| ----- |
| Red   |
| Blue  |
| Green |

Can we say:

```text
Red > Blue
```

or

```text
Green < Red
```

❌ No.

There is no correct ranking among colors.

They are simply different categories.

---

## 3. Examples of Nominal Variables

### Example 1: Cities

| City   |
| ------ |
| Mumbai |
| Delhi  |
| Pune   |

Can we say:

```text
Mumbai > Delhi > Pune
```

❌ No.

Cities do not have an inherent order.

---

### Example 2: Blood Group

| Blood Group |
| ----------- |
| A           |
| B           |
| AB          |
| O           |

Can we rank them?

❌ No.

They are just different types.

---

## 4. Why Is This Important?

Machine learning algorithms work with numbers.

Suppose we use Label Encoding on colors.

### Original Data

| Color |
| ----- |
| Red   |
| Blue  |
| Green |

After Label Encoding:

| Color | Encoded |
| ----- | ------: |
| Blue  |       0 |
| Green |       1 |
| Red   |       2 |

---

## 5. The Problem with Label Encoding Nominal Variables

Look at this:

```text
Blue  → 0
Green → 1
Red   → 2
```

A machine learning algorithm might interpret:

```text
2 > 1 > 0
```

which may look like:

```text
Red > Green > Blue
```

But this is FALSE.

The colors have no order.

The numbers were assigned only for identification.

---

## 6. Real-World Analogy

Imagine assigning locker numbers.

| Student | Locker Number |
| ------- | ------------: |
| Aditi   |           101 |
| Rahul   |           102 |
| Priya   |           103 |

Does this mean:

```text
Locker 103 student is better than Locker 101 student?
```

❌ No.

The numbers are just identifiers.

Similarly, Label Encoding gives numbers, but nominal categories do not become ordered.

---

## 7. Why Can This Be Dangerous?

Some machine learning models treat larger numbers as more important.

For example:

```text
Blue  → 0
Green → 1
Red   → 2
```

The model may think:

* Red is greater than Green.
* Green is greater than Blue.

This artificial ordering can negatively affect model performance.

---

## 8. Which Encoding Is Usually Better?

For nominal variables, **One-Hot Encoding** is usually preferred.

Instead of:

```text
Red   → 2
Blue  → 0
Green → 1
```

One-Hot Encoding creates separate columns.

Example:

| Color | Color_Blue | Color_Green | Color_Red |
| ----- | ---------: | ----------: | --------: |
| Red   |          0 |           0 |         1 |
| Blue  |          1 |           0 |         0 |
| Green |          0 |           1 |         0 |

No ordering is introduced.

We will study One-Hot Encoding in detail later.

---

## 9. Python Example

### Original Data

```python
import pandas as pd

data = pd.DataFrame({
    "City": [
        "Mumbai",
        "Delhi",
        "Pune",
        "Delhi"
    ]
})

print(data)
```

### Output

```text
     City
0   Mumbai
1    Delhi
2     Pune
3    Delhi
```

---

### Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

data["City_Encoded"] = encoder.fit_transform(data["City"])

print(data)
```

### Output

```text
     City  City_Encoded
0   Mumbai             1
1    Delhi             0
2     Pune             2
3    Delhi             0
```

---

## Output Explanation

Scikit-Learn first sorts the categories alphabetically:

```text
Delhi
Mumbai
Pune
```

Then assigns:

```text
Delhi  → 0
Mumbai → 1
Pune   → 2
```

The numbers are assigned automatically.

However:

```text
Pune > Mumbai > Delhi
```

does NOT make sense.

They are simply different cities.

---

## 10. Ordinal vs Nominal (Quick Comparison)

| Feature                  | Ordinal           | Nominal          |
| ------------------------ | ----------------- | ---------------- |
| Categories?              | Yes               | Yes              |
| Natural Order?           | Yes               | No               |
| Example                  | Low, Medium, High | Red, Blue, Green |
| Label Encoding Suitable? | Usually Yes       | Usually No       |

---

## Important Points to Remember

* Nominal variables are categorical variables without any order.
* Their categories are just labels.
* Label Encoding can introduce a false sense of ranking.
* This artificial ordering may confuse some machine learning models.
* One-Hot Encoding is often a better choice for nominal data.

---

## Summary

> **Nominal variables are categorical variables that have no natural order.**

Examples:

```text
Cities
Colors
Blood Groups
Departments
```
