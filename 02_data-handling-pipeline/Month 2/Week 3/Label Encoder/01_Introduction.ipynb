# Topic 1: Introduction

## Subtopic 1.1 – What is Categorical Data?

---

## 1. What is Data?

Data is information collected about people, objects, events, or situations.

For example:

| Name | Age | Gender |
| ---- | --: | ------ |
| John |  25 | Male   |
| Emma |  30 | Female |

In this table:

* **Name** is text data.
* **Age** is numerical data.
* **Gender** is categorical data.

---

## 2. What is Categorical Data?

Categorical data is data that represents **categories, groups, or labels** instead of numerical quantities.

It tells us **what type something is**, not **how much of something there is**.

### Examples of categorical data

| Feature         | Categories                        |
| --------------- | --------------------------------- |
| Gender          | Male, Female                      |
| Color           | Red, Blue, Green                  |
| City            | Mumbai, Delhi, Pune               |
| Department      | HR, Sales, IT                     |
| Education Level | High School, Bachelor's, Master's |

These values represent categories.

---

## 3. Why is it Called "Categorical"?

Because each value belongs to a category.

For example, suppose an online shopping website sells T-shirts.

The available sizes are:

```
Small
Medium
Large
```

These values classify the product into categories. They do not represent exact measurements.

---

## 4. Numerical Data vs Categorical Data

### Numerical Data

Numerical data consists of numbers that represent quantities.

Example:

| Age |
| --: |
|  18 |
|  25 |
|  32 |

You can perform mathematical operations:

```
32 - 18 = 14
```

This result has meaning.

---

### Categorical Data

Example:

| City   |
| ------ |
| Mumbai |
| Delhi  |
| Pune   |

Can we do:

```
Mumbai + Delhi
```

No.

These are labels, not quantities.

---

## 5. Real-World Analogy

Imagine you are filling out a college admission form.

You are asked to select your department:

* Computer Science
* Mechanical Engineering
* Civil Engineering
* Electronics Engineering

You are choosing one category from several options.

The selected department is categorical data.

---

## 6. Why is Categorical Data Important in Machine Learning?

Real-world datasets often contain categorical information.

For example:

A customer dataset may include:

| Age | Gender | City   |
| --: | ------ | ------ |
|  25 | Male   | Mumbai |
|  30 | Female | Delhi  |
|  22 | Male   | Pune   |

Here:

* Age → Numerical
* Gender → Categorical
* City → Categorical

Machine learning models frequently need this information to make predictions.

However, most algorithms cannot directly understand text categories. Later, we convert these categories into numbers using **encoding techniques**.

---

## Python Example

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [25, 30, 22],
    "Gender": ["Male", "Female", "Male"]
})

print(data)
```

### Output

```python
   Age  Gender
0   25    Male
1   30  Female
2   22    Male
```

### Explanation

* The `Age` column contains numerical data.
* The `Gender` column contains categorical data because it represents categories.

---

## Important Points to Remember

* Categorical data represents labels or groups.
* It does not represent measurable quantities.
* Examples include gender, city, color, and department.
* Many machine learning datasets contain categorical variables.
* These categories often need to be converted into numbers before training models.

# Topic 1: Introduction

## Subtopic 1.2 – Why Categorical Data Needs Encoding

---

## 1. What is Encoding?

**Encoding** means converting categorical values (text labels) into numerical values so that machine learning algorithms can work with them.

In simple words:

> **Encoding = Converting categories into numbers.**

---

## 2. Why Do We Need Encoding?

Most machine learning algorithms are built using mathematical operations.

They can understand numbers like:

```text
10
25
3.14
0
1
```

But they cannot directly understand text values like:

```text
Male
Female
Red
Blue
Mumbai
Delhi
```

Therefore, before training a machine learning model, we convert categorical data into numbers.

---

## 3. Understanding the Problem

Consider this dataset:

| Age | Gender |
| --: | ------ |
|  25 | Male   |
|  30 | Female |
|  22 | Male   |

Suppose we want to train a model.

The algorithm sees:

```text
25, Male
30, Female
22, Male
```

It understands:

```text
25
30
22
```

But it does not know what:

```text
Male
Female
```

mean.

As a result, the model cannot process the data properly.

---

## 4. What Happens Without Encoding?

If we directly pass text categories to many algorithms, we may get errors.

Example:

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X, y)
```

If `X` contains text values such as `"Male"` or `"Female"`, many algorithms will produce an error because they expect numeric input.

---

## 5. Real-World Analogy

Imagine you speak English, but someone gives you instructions in a language you don't understand.

For example:

```text
Go left.
```

You understand it.

But if someone says:

```text
左に行ってください
```

and you don't know Japanese, you cannot follow the instruction.

Similarly:

* Machine learning algorithms understand numbers.
* Categorical values are like another language.
* Encoding acts as a translator.

---

## 6. Example of Encoding

Suppose we have:

| Gender |
| ------ |
| Male   |
| Female |
| Male   |

After encoding:

| Gender |
| -----: |
|      1 |
|      0 |
|      1 |

Now the algorithm can process the values.

---

Another example:

| Color |
| ----- |
| Red   |
| Blue  |
| Green |

After encoding:

| Color |
| ----: |
|     2 |
|     0 |
|     1 |

The text categories become numerical values.

---

## 7. Does Encoding Change the Meaning?

Encoding changes the **representation**, not the actual information.

Example:

Before encoding:

```text
Male
Female
Male
```

After encoding:

```text
1
0
1
```

The category remains the same; only the format changes.

---

## 8. Python Example

### Original Data

```python
import pandas as pd

data = pd.DataFrame({
    "Gender": ["Male", "Female", "Male"]
})

print(data)
```

### Output

```text
   Gender
0     Male
1   Female
2     Male
```

---

### Encoded Data

```python
data["Gender"] = data["Gender"].map({
    "Female": 0,
    "Male": 1
})

print(data)
```

### Output

```text
   Gender
0        1
1        0
2        1
```

---

## Output Explanation

The `map()` function replaced:

```text
Female → 0
Male   → 1
```

The machine learning algorithm can now use this column because it contains numbers.

---

## Important Points to Remember

* Most machine learning algorithms require numerical input.
* They usually cannot work directly with text categories.
* Encoding converts categories into numbers.
* Encoding changes the representation, not the meaning of the data.
* Different encoding techniques exist, and each has its own use cases.

---

## Summary

> **Categorical data needs encoding because machine learning algorithms understand numbers, not text labels. Encoding converts those categories into a numerical form that algorithms can process.**

Example:

```text
Male   → 1
Female → 0
```

# Topic 1: Introduction

## Subtopic 1.3 – What is Label Encoding?

---

## 1. Definition of Label Encoding

**Label Encoding** is a technique used to convert categorical values into numerical values by assigning a unique integer (whole number) to each category.

In simple words:

> **Label Encoding replaces categories with numbers.**

---

## 2. Why Do We Use Label Encoding?

As we learned in the previous subtopic:

* Machine learning algorithms usually work with numbers.
* Many datasets contain text categories.
* Therefore, we convert those categories into numerical form.

Label Encoding is one of the simplest ways to do this.

---

## 3. How Does Label Encoding Work?

Suppose we have a column:

| Size   |
| ------ |
| Small  |
| Medium |
| Large  |

After Label Encoding, it may become:

| Size   | Encoded Value |
| ------ | ------------: |
| Small  |             2 |
| Medium |             1 |
| Large  |             0 |

Each category gets a unique number.

---

## 4. Important Note

The numbers assigned are usually **just labels**.

For example:

```text
Red   → 2
Blue  → 0
Green → 1
```

This does **not** mean:

* Red is greater than Blue.
* Green is smaller than Red.

They are simply identifiers assigned by the encoder.

---

## 5. Real-World Analogy

Imagine a school where every student is given a roll number.

| Student | Roll Number |
| ------- | ----------: |
| Aditi   |           1 |
| Rahul   |           2 |
| Priya   |           3 |

The roll number helps identify the student.

It does **not** mean:

* Rahul is better than Aditi.
* Priya is more important than Rahul.

Similarly, Label Encoding assigns numbers to categories for identification.

---

## 6. Example 1: Gender

### Original Data

| Gender |
| ------ |
| Male   |
| Female |
| Male   |
| Female |

### After Label Encoding

| Gender | Encoded |
| ------ | ------: |
| Male   |       1 |
| Female |       0 |
| Male   |       1 |
| Female |       0 |

---

## 7. Example 2: Cities

### Original Data

| City   |
| ------ |
| Mumbai |
| Delhi  |
| Pune   |

### After Label Encoding

| City   | Encoded |
| ------ | ------: |
| Delhi  |       0 |
| Mumbai |       1 |
| Pune   |       2 |

Again, these numbers are simply labels.

---

## 8. How Does Scikit-Learn Assign Numbers?

Scikit-Learn's `LabelEncoder` sorts categories alphabetically and then assigns numbers starting from 0.

Example:

Categories:

```text
Delhi
Mumbai
Pune
```

Alphabetical order:

```text
Delhi
Mumbai
Pune
```

Assigned values:

```text
Delhi  → 0
Mumbai → 1
Pune   → 2
```

Another example:

Original categories:

```text
Banana
Apple
Orange
```

Alphabetical order:

```text
Apple
Banana
Orange
```

Encoded values:

```text
Apple   → 0
Banana  → 1
Orange  → 2
```

---

## 9. Python Example Using LabelEncoder

```python
from sklearn.preprocessing import LabelEncoder

gender = ["Male", "Female", "Male", "Female"]

encoder = LabelEncoder()

encoded_gender = encoder.fit_transform(gender)

print(encoded_gender)
```

### Output

```python
[1 0 1 0]
```

---

## 10. Output Explanation

Scikit-Learn first found the unique categories:

```text
Female
Male
```

Since it sorts alphabetically:

```text
Female → 0
Male   → 1
```

Therefore:

Original:

```text
Male, Female, Male, Female
```

became:

```text
1, 0, 1, 0
```

---

## 11. Does Label Encoding Always Mean Order?

No.

This is a very important concept.

Consider:

### Colors

```text
Red
Blue
Green
```

Encoded as:

```text
Blue  → 0
Green → 1
Red   → 2
```

The encoder assigned numbers, but colors do not have a natural ranking.

So:

```text
2 > 1 > 0
```

does **not** mean:

```text
Red > Green > Blue
```

The numbers are only labels.

---

## 12. When Is Label Encoding Most Useful?

Label Encoding works best when categories already have a meaningful order.

For example:

| Satisfaction Level |
| ------------------ |
| Low                |
| Medium             |
| High               |

Encoding:

```text
Low    → 0
Medium → 1
High   → 2
```

Here, the order makes sense.

We will study this in detail in the next topic when we discuss **Ordinal vs Nominal variables**.

---

## Important Points to Remember

* Label Encoding converts categories into integers.
* Each category receives a unique number.
* Scikit-Learn assigns numbers after sorting categories alphabetically.
* The assigned numbers are often just identifiers.
* Label Encoding is especially useful when categories have a natural order.

---

## Summary

> **Label Encoding is a technique that converts categorical values into numerical labels by assigning a unique integer to each category.**

Example:

```text
Female → 0
Male   → 1
```

or

```text
Low    → 0
Medium → 1
High   → 2
```
