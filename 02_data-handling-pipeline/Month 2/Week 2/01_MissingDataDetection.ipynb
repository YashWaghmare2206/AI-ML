## 1.1.1 — `df.isnull()`

`isnull()` checks whether each value in the dataset is missing or not.

It returns a **Boolean table**:
- `True` → value is missing
- `False` → value is present

### Why this exists
In ML pipelines the first step is always data inspection. We must answer:
- Which columns have missing values?
- Which rows are incomplete?
- How bad is the dataset?

Before we drop or fill, we must **detect**. That detection starts with `df.isnull()`.

### What Pandas treats as missing
- `NaN`
- `None`
- `NaT` (for datetime)

````python
import pandas as pd
import numpy as np

data = {
    "Name": ["A", "B", "C", "D"],
    "Age":  [25, None, 30, None],
    "Salary": [40000, 50000, None, 60000]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
````

````python
# Applying isnull() — creates a True/False mask
# True  → missing value
# False → value exists
print(df.isnull())
````

````python
# isnull() does NOT count — it only detects.
# np.nan != np.nan  (IEEE 754 float rule — this is WHY we can't use == NaN)
# None == None  (Python object equality works normally)
# This is exactly why isnull() exists — to handle both cases uniformly.
import pandas as pd
import numpy as np
df2 = pd.DataFrame({
    "A": [1, 2, np.nan],
    "B": [None, 5, 6]
})
print(df2.isnull())
````

### Practice Questions

````python
# Q1: Print rows that contain at least one missing value
d1 = {
    'Name': ['A', 'B', 'C', 'D'],
    'Age':  [25, None, 30, 28],
    'Salary': [40000, 50000, None, 60000]
}
df = pd.DataFrame(d1)
print(df[df.isna().any(axis=1)])
````

## 1.1.2 — `df.isnull().sum()`

Counts missing values in each column.

### How it works internally
1. `df.isnull()` → creates True/False mask (`True=1`, `False=0`)
2. `.sum()` → adds up Trues column-wise

### Pro tricks
```python
df.isnull().sum().sort_values(ascending=False)  # worst columns first
(df.isnull().sum() / len(df)) * 100             # missing percentage
```

### Real ML workflow
- missing < 5% → fill
- missing > 50% → drop column

````python
data = {
    "Name":   ["A","B","C","D"],
    "Age":    [25, None, 30, None],
    "Salary": [40000, 50000, None, 60000]
}
df = pd.DataFrame(data)

print("Missing per column:")
print(df.isnull().sum())

print("\nMissing per row:")
print(df.isnull().sum(axis=1))
````

````python
# Sort by most missing
print(df.isnull().sum().sort_values(ascending=False))

# Missing percentage
print((df.isnull().sum() / len(df)) * 100)
````

### Practice Questions

````python
# Q1: Count missing values per column
df = pd.DataFrame({
    'Name':   ['A','B','C','D'],
    'Age':    [25, None, 30, None],
    'Salary': [40000, 50000, None, 60000]
})
print(df.isnull().sum())
````

````python
# Q2: Show only columns that contain missing values
df = pd.DataFrame({
    'A': [1, 2, None],
    'B': [None, 4, 7],
    'C': [5, 6, 8]
})
print(df.loc[:, df.isnull().any()])
````

````python
# Q3: Missing percentage per column
df = pd.DataFrame({
    'A': [1, None, 3],
    'B': [None, 4, 6],
    'C': [3, None, 7]
})
print((df.isnull().sum() / len(df)) * 100)
````

````python
# Q4: Columns sorted by missing count (highest first)
df = pd.DataFrame({
    'A': [1, 2, None],
    'B': [None, 4, 7],
    'C': [5, 6, 8]
})
print(df.isnull().sum().sort_values(ascending=False))
````

````python
# Q5: Columns where missing count > 1
df = pd.DataFrame({
    'A': [1, None, 2, None],
    'B': [None, None, None, 7],
    'C': [5, 6, 7, 8]
})
missing_count = df.isnull().sum()   # It returns an Series
print(missing_count[missing_count > 1])
````

````python
# Q6: Column with the maximum missing values  (.idxmax() = Index of Maximum Value)
df = pd.DataFrame({
    'A': [1, None, 3, None],
    'B': [None, None, None, 7],
    'C': [5, 6, 7, 8],
    'D': [7, 8, None, 10]
})
print(df.isnull().sum().idxmax())
````

````python
# Q7: Check if dataset has ANY missing values at all
df = pd.DataFrame({
    'A': [1, 4, 7],
    'B': [2, None, 8],
    'C': [3, 6, 9]
})
print(df.isnull().values.any())    # It returns True or False
````

````python
# Q8: Total missing values in entire dataset
df = pd.DataFrame({
    'A': [1, None, 5],
    'B': [None, 4, 6],
    'C': [3, None, 9]
})
print(f"Dataset has {df.isnull().sum().sum()} missing values")
````

````python
# Q9: Missing percentage for entire dataset
df = pd.DataFrame({
    'A': [1, None, 3],
    'B': [None, 4, 6],
    'C': [None, 8, 9]
})
missing_pct = (df.isnull().sum().sum() / df.size) * 100
print(f"{missing_pct:.1f}% of dataset is missing")
````

````python
# Q10: Print "Dataset is clean" or "Dataset contains missing values"
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, None, 9]
})
if df.isnull().sum().sum() == 0:
    print("Dataset is clean")
else:
    print("Dataset contains missing values")
````

## 1.1.3 — `df.isna().any(axis=1)`

Returns a Boolean Series for each **row**:
- `True` → row contains at least one missing value
- `False` → row is complete

### Why `axis=1`?
| Axis | Meaning |
|---|---|
| `axis=0` | operate column-wise |
| `axis=1` | operate row-wise |

Most common use: `df[df.isna().any(axis=1)]` — select incomplete rows.

````python
df = pd.DataFrame({
    "Name":   ["A","B","C","D"],
    "Age":    [25, None, 30, 28],
    "Salary": [40000, 50000, None, 60000]
})

# Rows WITH missing values
print("Incomplete rows:")
print(df[df.isna().any(axis=1)])   # df.isna().any(axis=1) returns an series and when passed to Datframe then mathcing of index take place like index 0  of dataframe and 0 of series if series as false at that index then the dataframe row is not shown

# Rows WITHOUT missing values  (~ = NOT operator)
print("\nComplete rows:")
print(df[~df.isna().any(axis=1)])
````

### Practice Questions

````python
# Q1: Print rows that contain missing values
df = pd.DataFrame({"A":[1,2,3], "B":[4,5,None], "C":[7,8,9]})
print(df[df.isna().any(axis=1)])
````

````python
# Q2: Print rows that do NOT contain missing values
df = pd.DataFrame({"A":[1,None,3,4], "B":[5,6,None,8], "C":[9,10,11,12]})
print(df[~df.isna().any(axis=1)])
````

````python
# Q3: Count rows with missing values
# Note: .sum() treats True=1, False=0
# .count() would be wrong — it counts all rows
df = pd.DataFrame({"A":[1,None,3,4], "B":[5,6,None,8], "C":[9,10,11,12]})
print(df.isna().any(axis=1).sum())
````

````python
# Q4: Return indices of rows that contain missing values
# .index is an attribute, not a method — no parentheses
df = pd.DataFrame({"A":[1,None,3,None], "B":[5,6,7,8], "C":[9,None,11,12]})
print(df[df.isna().any(axis=1)].index)
````

````python
# Q5: Check if ANY row contains missing values
df = pd.DataFrame({"A":[1,2,3], "B":[4,5,None], "C":[7,8,9]})

# WRONG: df.isna().values.any(axis=1) → TypeError
# CORRECT:
print(df.isna().any().any())
# df.isna()      → boolean table
# .any()         → column-wise: does each column have any True?
# .any() again   → overall: does any column have True?
````

## 1.1.4 — `df.isnull().sum().sum()`

Returns one number: **total missing cells** in the entire dataset.

**3-step breakdown:**
1. `df.isnull()` → True/False mask
2. `.sum()` → Series of missing count per column
3. `.sum()` again → adds all column counts together

**Quick check:** `df.isnull().sum().sum() > 0` → True means dataset needs cleaning.

````python
df = pd.DataFrame({
    "A": [1, None, 3],
    "B": [4, 5, None],
    "C": [7, 8, 9]
})

# Step by step
print("Step 1 - boolean mask:")
print(df.isnull())
print("\nStep 2 - count per column:")
print(df.isnull().sum())
print("\nStep 3 - total:")
print(df.isnull().sum().sum())
````

### Quick Reference Table

| Command | Meaning |
|---|---|
| `df.isnull()` | Detect missing values (boolean mask) |
| `df.isnull().sum()` | Count missing per column |
| `df.isnull().sum().sort_values()` | Sort columns by missing count |
| `(df.isnull().sum()/len(df))*100` | Missing percentage per column |
| `df.isnull().sum().sum()` | Total missing cells |
| `df.isna().any(axis=1)` | Boolean mask per row |
| `df[df.isna().any(axis=1)]` | Select incomplete rows |
