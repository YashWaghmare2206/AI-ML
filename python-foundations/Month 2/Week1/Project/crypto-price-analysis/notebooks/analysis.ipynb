```python
import pandas as pd

df = pd.read_csv("../data/raw_data.csv")
df.head()
```

```python
df.info()
```

```python
avg_price = df.groupby("coin")["price"].mean()
print(avg_price)
```

```python
import matplotlib.pyplot as plt

for coin in df["coin"].unique():

    coin_df = df[df["coin"] == coin]

    plt.plot(coin_df["date"], coin_df["price"], label=coin)

plt.legend()
plt.title("Crypto Price Trend (30 Days)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

```python
volatility = df.groupby("coin")["price"].std()

print(volatility)
```
