# Likelihood

Now let's learn the second component of Bayes.

---

## What is Likelihood?

Likelihood answers:

> If the hypothesis is true, how likely is the evidence?

Mathematically:

[
P(Evidence|Hypothesis)
]

or

[
P(E|H)
]

---

## Disease Example

Suppose a person actually has the disease.

A medical test is positive 95% of the time.

[
P(Positive|Disease)=0.95
]

This is a likelihood.

It answers:

> If disease exists, how likely is a positive test?

---

## Spam Example

Suppose:

80% of spam emails contain the word "free".

[
P(free|Spam)=0.8
]

This is the likelihood.

It answers:

> If the email is spam, how likely is the word "free"?

---

## Important Direction

Likelihood is:

[
P(Evidence|Hypothesis)
]

NOT

[
P(Hypothesis|Evidence)
]

Example:

[
P(free|Spam)
]

≠

[
P(Spam|free)
]

The second one is the posterior.

# Posterior Probability

This is the final quantity Bayes is trying to calculate.

---

## What is Posterior?

Posterior is your belief **after seeing evidence**.

Mathematically:

**`P(H|E)`**

Read as:

> Probability of the hypothesis given the evidence.

---

## Disease Example

```text
Before Evidence  → Prior

After Evidence   → Posterior
```

---

## Bayes Formula

[
P(H|E)

\frac{P(E|H)P(H)}
{P(E)}
]

The left side is the posterior.

Everything on the right side is used to calculate it.

# Bayesian Updating

This is the real idea behind Bayesian thinking.

---

## What is Bayesian Updating?

Bayesian Updating means:

> Use new evidence to update your belief.

---

## Process

```text
Prior
  +
Evidence
  ↓
Posterior
```

The posterior becomes your new belief.

---

## Example

Suppose:

**`P(Disease)=0.01`**

Before any test:

```text
1% chance of disease
```

This is the prior.

---

Now a positive test arrives.

Using Bayes:

**`P(Disease|Positive)=0.088`**

Now your belief changes:

```text
Old Belief = 1%
New Belief = 8.8%
```

The new belief is the posterior.

---

## Updating Again

Suppose a second independent test is also positive.

Now:

```text
Previous Posterior
        ↓
      New Prior
        ↓
 Second Positive Test
        ↓
 New Posterior
```

Bayesian learning keeps updating beliefs whenever new evidence arrives.

---

## General Idea

```text
Prior₁
   ↓ Evidence₁
Posterior₁

Posterior₁ becomes Prior₂
   ↓ Evidence₂
Posterior₂

Posterior₂ becomes Prior₃
   ↓ Evidence₃
Posterior₃
```

This process can continue indefinitely.

---

## Spam Example

Initial belief:

**`P(Spam)=0.30`**

---

First evidence:

```text
free
```

Posterior increases.

---

Second evidence:

```text
win
```

Update again.

---

Third evidence:

```text
money
```

Update again.

Each new piece of information changes the probability.

---

## Why This Matters in ML

Machine Learning is essentially:

```text
Start with belief
      ↓
Observe data
      ↓
Update belief
      ↓
Observe more data
      ↓
Update again
```

Bayesian models do this explicitly using probabilities.

---

## Intuition

Think:

```text
I believe X.

I see evidence.

I adjust my belief.
```

That adjustment is Bayesian Updating.

---

## What To Remember



After new evidence arrives:

```text
Prior → Evidence → Posterior
```

and the posterior can become the next prior.

This continuous belief-updating process is called **Bayesian Updating**.

# Python Examples

As you noticed earlier, Bayes itself is just multiplication, addition, and division. So the code is simple.

---

## Example 1: Disease Testing

Given:

[
P(Disease)=0.01
]

[
P(Positive|Disease)=0.95
]

[
P(Positive|NoDisease)=0.10
]

```python
p_disease = 0.01
p_no_disease = 0.99

p_positive_given_disease = 0.95
p_positive_given_no_disease = 0.10

# Evidence
p_positive = (
    p_positive_given_disease * p_disease
    + p_positive_given_no_disease * p_no_disease
)

# Posterior
p_disease_given_positive = (
    p_positive_given_disease * p_disease
) / p_positive

print(p_disease_given_positive)
```

Output:

```python
0.0876
```

---

## Example 2: Spam Detection

```python
p_spam = 0.30
p_not_spam = 0.70

p_free_given_spam = 0.80
p_free_given_not_spam = 0.20

# Evidence
p_free = (
    p_free_given_spam * p_spam
    + p_free_given_not_spam * p_not_spam
)

# Posterior
p_spam_given_free = (
    p_free_given_spam * p_spam
) / p_free

print(p_spam_given_free)
```

Output:

```python
0.6316
```

---

## Generic Bayes Function

```python
def bayes(prior, likelihood, evidence):
    return (prior * likelihood) / evidence
```

Usage:

```python
posterior = bayes(
    prior=0.30,
    likelihood=0.80,
    evidence=0.38
)

print(posterior)
```

Output:

```python
0.6316
```

---

## Using a Dictionary (Closer to ML)

```python
priors = {
    "Spam": 0.30,
    "NotSpam": 0.70
}

likelihoods = {
    "Spam": 0.80,
    "NotSpam": 0.20
}

evidence = (
    priors["Spam"] * likelihoods["Spam"]
    + priors["NotSpam"] * likelihoods["NotSpam"]
)

posterior_spam = (
    priors["Spam"] * likelihoods["Spam"]
) / evidence

print(posterior_spam)
```

---

## Key Observation

Everything in Bayes code is:

```python
evidence = sum(
    likelihood * prior
)

posterior = (
    likelihood * prior
) / evidence
```

The interesting part is **where the priors and likelihoods come from**.

In real ML:

```text
Training Data
     ↓
Estimate Priors
Estimate Likelihoods
     ↓
New Data
     ↓
Bayes Formula
     ↓
Prediction
```
