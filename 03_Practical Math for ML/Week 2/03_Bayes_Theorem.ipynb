Bayes starts with a belief.

```text
Old Belief
     +
New Evidence
     ↓
Updated Belief
```

# Reverse Probability

---

## Normal Probability

---

## Reverse Probability

But in real life we usually observe the effect first.

Example:

A test result is positive.

Now we want:



Meaning:

> Given a positive test, what is the probability the person actually has the disease?

Notice the direction changed.

---

## Forward vs Reverse

Forward:

```text
Disease
   ↓
Positive Test
```


---

Reverse:

```text
Positive Test
      ↓
Disease ?
```



---

## Important Point

These probabilities are usually very different.



The second probability also depends on:

* How common the disease is
* False positives
* Population statistics

---

## Example

Suppose:



Bayes exists to calculate the correct reverse probability.

---

## Spam Example

We may know:

[
P(free|Spam)
]

Probability that the word "free" appears in spam emails.

---

But classification needs:

[
P(Spam|free)
]

Probability that an email is spam given it contains the word "free".

Again, the probability has been reversed.

---

## Why Reverse Probability Is Hard

Forward probability:

```text
Cause → Effect
```

can often be measured directly.

Reverse probability:

```text
Effect → Cause
```

cannot usually be measured directly.

Bayes converts one into the other.

---

## Bayes as a Reverse Machine

Think:

```text
Known:
P(Evidence | Cause)

Want:
P(Cause | Evidence)
```

Bayes performs this conversion.
