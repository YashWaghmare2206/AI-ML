# Probability Trees

A Probability Tree is a visual way to represent sequential events and conditional probabilities.

---

## Why Use Trees?

They help when:

```text
One event happens
      ↓
Then another event happens
```

Examples:

* Coin tosses
* Disease testing
* Bayes Theorem
* Total Probability

---

## Example

Suppose:

* 60% students study Notes
* 40% study Videos

Pass probabilities:

* (P(Pass|Notes)=0.8)
* (P(Pass|Videos)=0.6)

Tree:

```text
                 Start
                /     \
            Notes     Videos
             0.6        0.4
            /  \       /   \
        Pass Fail  Pass  Fail
         0.8 0.2   0.6  0.4
```

---

## Rule 1: Multiply Along a Path

Example:

```text
Notes → Pass
```



---

## Rule 2: Add Multiple Paths

Pass can happen through:

```text
Notes → Pass

Videos → Pass
```



---

## Connection to Total Probability

Tree diagrams are often the easiest way to apply:



---

## Connection to Bayes

Disease testing:

```text
                Person
               /      \
        Disease     No Disease
           |             |
       Positive      Positive
```

The branches are used to calculate:

**`P(Positive)`**

and later:

**`P(Disease|Positive)`**

using Bayes.

---

## Optional Python Visualization

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

G.add_edges_from([
    ("Start", "Notes"),
    ("Start", "Videos"),
    ("Notes", "Pass"),
    ("Notes", "Fail"),
    ("Videos", "Pass"),
    ("Videos", "Fail")
])

nx.draw(G, with_labels=True)
plt.show()
```

---

## What To Remember

```text
Tree Diagram

Multiply along branches
        +
Add relevant paths
```

Used heavily in:

* Conditional Probability

# Bayes Visualization

This topic visualizes what Bayes is doing.

---

## Bayes Process

```text id="ld3g8n"
Prior
  +
Evidence
  ↓
Posterior
```

---

## Example

Suppose before seeing any email:

[
P(Spam)=0.30
]

```text id="hmg4cl"
Spam = 30%
Not Spam = 70%
```

This is the Prior.

---

## Evidence Arrives

Email contains:

```text id="k5tn1p"
free
```

This evidence supports Spam.

---

## After Bayes Update

[
P(Spam|free)=0.632
]

```text id="7td2s3"
Spam = 63.2%
Not Spam = 36.8%
```

This is the Posterior.

---

## Visualization

Before:

```text id="n8ghy7"
Spam     ███
NotSpam  ███████
```

After:

```text id="fhmftj"
Spam     ██████
NotSpam  ████
```

The evidence shifts belief toward Spam.

---

## Prior vs Posterior Graph

```python
import matplotlib.pyplot as plt

labels = ["Spam", "Not Spam"]

prior = [0.3, 0.7]

posterior = [0.632, 0.368]

plt.bar(labels, prior)

plt.title("Prior")
plt.show()

plt.bar(labels, posterior)

plt.title("Posterior")
plt.show()
```

---

## What To Remember

```text id="j2e9ew"
Prior
   ↓
Evidence
   ↓
Posterior
```

Bayes Visualization simply shows how probability moves after observing data.
