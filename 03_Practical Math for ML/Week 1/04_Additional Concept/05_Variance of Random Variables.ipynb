# Variance of Random Variables

Now that you know **Expected Value**, we can answer a new question:

> Expected value tells us the center.
> But how spread out are the values around that center?

That spread is called **Variance**.

---

## Intuition

Consider two datasets:

### Dataset A

**`4,;5,;6`**

Mean:

**`E[X]=5`**

---

### Dataset B

**`0,;5,;10`**

Mean:

**`E[X]=5`**

---

Both have the same mean.

But Dataset B is much more spread out.

Variance measures that spread.

## Understanding the Formula

### Step 1

Find the mean:

**`E[X]`**

---

### Step 2

Subtract the mean:

**`X-E[X]`**

This gives deviation from the center.

---

### Step 3

Square it:



Why square?

Because positive and negative deviations would otherwise cancel.

Example:

**`(-3)+(+3)=0`**

Not useful.

Squaring makes both positive:



---

### Step 4

Take the expected value:



Average all squared deviations.

## Standard Deviation

Variance is measured in squared units.

Example:

* Height in cm
* Variance in cm²

Not intuitive.

So we take square root:



This is called **Standard Deviation**.

---

## Interpretation

### Small Variance

Values stay close to the mean.

---

### Large Variance

Values are spread out.

---

## Relationship with Covariance

Remember:

**`Cov(X,Y)`**

measures how two variables move together.

Variance is just covariance with itself:

Var(X)=Cov(X,X)

Very important fact.

---

## Why Important in ML?

Variance appears everywhere.

### PCA

PCA searches for directions with maximum variance.

---

### Feature Analysis

High variance features often contain more information.

## Key Takeaways

* Variance measures spread around the mean.



* Small variance = tightly clustered values.
* Large variance = spread out values.
* Variance is covariance with itself.

### One-line Summary

> Variance measures the average squared distance of a random variable from its expected value.
