# Expected Value (Mean of a Random Variable)

Expected Value is one of the most important concepts in probability and ML.

It answers:

> **What value should we expect on average if an experiment is repeated many times?**

---

## Intuition

Suppose you roll a fair 6-sided die.

Possible outcomes:

**`1,2,3,4,5,6`**

Each outcome has probability:

**`1/6`**

What number should you expect on average over many rolls?

That is the **expected value**.

---

## 1. Discrete Random Variables

The variable takes specific countable values.

Examples:

* Die roll: ({1,2,3,4,5,6})
* Number of heads
* Number of customers

Formula:

[
E[X]
====

\sum_i x_i p_i
]

We use a **sum** because there are separate outcomes.

---

## 2. Continuous Random Variables

The variable can take infinitely many values in an interval.

Examples:

* Height
* Weight
* Temperature
* Time

You can't list all possible values.

Instead of probabilities, we use a **probability density function** (f(x)).

For your AI/ML roadmap, whenever we first learn a concept like Expected Value, Variance, or Covariance, I usually start with the **discrete version** because:

1. Easier to understand.
2. No calculus required.
3. The intuition is exactly the same.

Later, when you reach advanced probability or statistics, you'll see the continuous versions using integrals.

So for now, remember:

* **Discrete:** (\sum x_i p_i)
* **Continuous:** (\int x f(x),dx)

The idea is identical: **probability-weighted average**.
