# Topic 1: Mean (Arithmetic Mean)

## 1. What is Mean?

Mean is simply the **average value** of a set of numbers.

It answers the question:

> "If all the values were equally distributed, what would each value be?"

For example, if 5 students scored different marks, the mean tells us the "typical" score of the class.

---

## 2. Why do we need Mean?

Imagine these ages:

```
18, 19, 20, 19, 24
```

Instead of remembering all 5 numbers, we can summarize them with one number.

That summary is the **mean**.

Statistics is often about reducing a lot of data into a few meaningful numbers.

---

## 3. Why is Mean important in AI/ML?

Mean appears everywhere in AI/ML.

### a) Understanding data

When doing EDA, you might ask:

> "What is the average age of customers?"
>
> "What is the average house price?"

Mean helps answer these.

---

### b) Feature Scaling

Later, you'll learn **Standardization**:

```
(value − mean) / standard deviation
```

This makes features centered around zero.

Many algorithms benefit from this.

Examples:

* Linear Regression
* Logistic Regression
* SVM
* Neural Networks

---

### c) Loss Functions

Algorithms often minimize the **average error**.

Example:

**Mean Squared Error (MSE)**

Average of squared prediction errors.

You'll study this in Month 4.

---

## 4. Formula

Mean is:

> Sum of all values ÷ Number of values

Mathematically:

[
\text{Mean} = \frac{\sum x}{n}
]

where:

* (x) = values
* (\sum x) = sum of values
* (n) = total number of values

---

## 5. Example 1

Marks:

```
70, 80, 90
```

Mean

```
= (70 + 80 + 90) / 3
= 240 / 3
= 80
```

Average mark = **80**

---

## 6. Example 2

Daily temperatures:

```
30, 31, 32, 33
```

Mean

```
= (30 + 31 + 32 + 33) / 4
= 126 / 4
= 31.5°C
```

Average temperature = **31.5°C**

---

## 7. Limitation of Mean

Mean is sensitive to **outliers**.

Outliers are unusually large or small values.

Example:

Salaries:

```
25,000
27,000
28,000
30,000
5,00,000
```

Mean becomes very high because of one extremely large salary.

So sometimes the mean does **not represent the typical value well**.

This is why we'll learn **Median** next.

## 8. Python Implementation

### NumPy

```python
import numpy as np

data = [70, 80, 90]

np.mean(data)
```

Output:

```python
80.0
```

---

### Pandas

```python
import pandas as pd

s = pd.Series([70, 80, 90])

s.mean()
```

Output:

```python
80.0
```
