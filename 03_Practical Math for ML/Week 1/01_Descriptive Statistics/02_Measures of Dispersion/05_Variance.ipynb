# Topic 6: Variance

Range had a problem:

> It only looked at the smallest and largest values.

Variance solves this by asking:

> **"On average, how far are all the values from the mean?"**

---

# 1. What is Variance?

Variance measures the **spread of data around the mean**.

It tells us:

> Are the data points close to the average, or are they scattered far away?

---

## Intuition

Imagine two archers shooting arrows.

### Archer A

All arrows land close together.

🎯🎯🎯🎯🎯

Low spread.

Low variance.

---

### Archer B

Arrows land all over the board.

🎯       🎯

```
🎯

     🎯
```

🎯

High spread.

High variance.

---

# 2. Why Do We Need Variance?

Suppose two datasets have the same mean.

### Dataset A

```text
9, 10, 10, 11, 10
```

Mean = 10

Values are very close to the mean.

---

### Dataset B

```text
1, 5, 10, 15, 19
```

Mean = 10

Values are spread far from the mean.

---

Both have the same average.

But Dataset B is much more variable.

Variance captures this difference.

---

# 3. Why is Variance Important in AI/ML?

Variance appears everywhere.

Standard deviation is simply:

> Square root of variance.

## c) Model Evaluation

You'll later hear:

> Bias-Variance Tradeoff

One of the most important ML concepts.

This variance is related to how sensitive a model is to training data.

# 4. How Do We Calculate Variance?

### Step 1

Find the mean.

### Step 2

Find how far each value is from the mean.

These are called deviations.

### Step 3

Square each deviation.

Why square?

Because positive and negative deviations cancel each other.

Example:

```text
+2 and -2

Sum = 0
```

Even though both are far from the mean.

Squaring fixes this.

---

### Step 4

Take the average of those squared deviations.

That average is the variance.

---

# Formula

For a population:

[
\text{Variance}=\frac{\sum (x-\mu)^2}{N}
]

where:

* (x) = each value
* (\mu) = population mean
* (N) = number of observations

---

# Population vs Sample Variance

You may see two formulas.

## Population Variance

Used when you have the entire population.

Divide by:

[
N
]

---

## Sample Variance

Used when working with a sample.

Divide by:

[
n-1
]

Most real-world datasets are samples.

For now:

> Just remember that **Pandas and NumPy usually use sample variance by default (n−1)** unless specified otherwise.

We'll revisit this later if needed.

---

# Example 1

Dataset:

```text
2, 4, 6
```

Mean:

```text
(2+4+6)/3 = 4
```

Deviations:

```text
2−4 = -2
4−4 = 0
6−4 = 2
```

Squares:

```text
4, 0, 4
```

Average:

```text
(4+0+4)/3
= 8/3
≈ 2.67
```

Variance ≈ **2.67**

---

# Example 2

Dataset:

```text
9, 10, 11
```

Mean:

```text
10
```

Deviations:

```text
-1, 0, 1
```

Squares:

```text
1, 0, 1
```

Variance:

```text
(1+0+1)/3
= 2/3
≈ 0.67
```

Variance ≈ **0.67**

Notice:

> The values are tightly clustered around the mean.

So variance is smaller.

---

# One Problem with Variance

Variance uses **squared units**.

Example:

If the data is in kilograms,

Variance is in:

```text
kilograms²
```

which is harder to interpret.

That's why we take the square root.

This gives us:

> **Standard Deviation**

which is in the original units.

And that's our next topic.

---

# Python Implementation

## NumPy

Population Variance:

```python
import numpy as np

data = [2, 4, 6]

np.var(data)
```

Output:

```python
2.6666666666666665
```

---

Sample Variance:

```python
np.var(data, ddof=1)
```

Output:

```python
4.0
```

`ddof=1` means divide by (n-1).

---

## Pandas

```python
import pandas as pd

s = pd.Series([2, 4, 6])

s.var()
```

Output:

```python
4.0
```

Pandas uses **sample variance** by default.

---

# Quick Recap

* Variance measures **how spread out data is around the mean**.
* Low variance → values are close to the mean.
* High variance → values are widely spread.
* It uses **all observations**, unlike range.
* Squaring deviations prevents positives and negatives from cancelling.
* Variance is used in:

  * Standard deviation
  * Feature scaling
  * Bias-Variance Tradeoff
  * Many ML algorithms.
* Limitation:

  * Units become squared.
