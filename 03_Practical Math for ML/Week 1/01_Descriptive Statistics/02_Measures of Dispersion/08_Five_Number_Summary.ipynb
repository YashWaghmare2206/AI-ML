# 1. What is the Five-Number Summary?

It is a way to summarize an entire dataset using just **5 numbers**.

These 5 numbers describe:

> The center, spread, and possible outliers in the data.

The five numbers are:

| Statistic   | Meaning         |
| ----------- | --------------- |
| Minimum     | Smallest value  |
| Q1          | 25th percentile |
| Median (Q2) | Middle value    |
| Q3          | 75th percentile |
| Maximum     | Largest value   |

---

# 2. Why Do We Need It?

Imagine this dataset:

```text
5, 7, 8, 9, 10, 12, 15
```

Instead of looking at all 7 values, we can summarize it using 5 numbers.

This gives us a quick understanding of:

* Where the data starts
* Where most of the data lies
* The middle value
* Where the data ends

---

## Five-Number Summary

| Statistic | Value |
| --------- | ----: |
| Minimum   |     2 |
| Q1        |     3 |
| Median    |     6 |
| Q3        |     9 |
| Maximum   |    10 |

---
# 3. Why is it Important in AI/ML?

## a) Boxplots

A boxplot is built entirely from these 5 numbers.

```
Min    Q1   Median   Q3    Max
|------|====|====|----|
```

The box represents the middle 50% of the data.

---

## b) Detecting Outliers

Using Q1 and Q3, we calculate:

[
IQR = Q3 - Q1
]

Then:

* Lower Bound = Q1 − 1.5 × IQR
* Upper Bound = Q3 + 1.5 × IQR

Values outside these bounds are potential outliers.

---

## c) Quick Data Understanding

Instead of reading thousands of values, you can immediately know:

* Typical values
* Spread of the middle data
* Whether extreme values exist

---

# 4. What Does Each Number Tell You?

Suppose:

| Statistic |    Salary |
| --------- | --------: |
| Minimum   |   ₹20,000 |
| Q1        |   ₹35,000 |
| Median    |   ₹50,000 |
| Q3        |   ₹70,000 |
| Maximum   | ₹5,00,000 |

Interpretation:

* Lowest salary = ₹20,000
* 25% earn ≤ ₹35,000
* Half earn ≤ ₹50,000
* 75% earn ≤ ₹70,000
* Highest salary = ₹5,00,000

Notice:

> The maximum salary is much larger than Q3.

This suggests a possible outlier.

The Five-Number Summary consists of:

1. Minimum

3. Median (Q2)
4. Q3 (75th percentile)
5. Maximum

It helps you:

* Understand a dataset quickly
* Build and interpret boxplots
* Detect outliers using IQR
* Summarize large datasets efficiently
