# Topic 7: Standard Deviation

This is probably the **most used measure of spread** in statistics, data science, and AI/ML.

If someone says:

> "The average score is 80 with a standard deviation of 5."

They're telling you not only the average, but also **how much the scores typically vary from that average**.

---

# 1. What is Standard Deviation?

Standard deviation tells us:

> **On average, how far are the values from the mean?**

It measures the spread of the data, just like variance.

The difference is:

> Standard deviation is expressed in the **same units as the original data**, making it much easier to interpret.

---

# 2. Why Do We Need Standard Deviation?

Remember variance?

Variance had a problem:

If the data was in kilograms,

Variance was in:

```text
kilograms²
```

which isn't intuitive.

Standard deviation fixes this by taking the square root of variance.

---

# Formula

> Standard Deviation = √Variance

Mathematically:

[
\text{Standard Deviation} = \sqrt{\text{Variance}}
]

Population standard deviation:

[
\sigma = \sqrt{\frac{\sum (x-\mu)^2}{N}}
]

You don't need to memorize this formula.

Just remember:

> **Standard Deviation = Square Root of Variance**

---

# 3. Intuition

Suppose the average height in a class is:

```text
170 cm
```

### Low Standard Deviation

Most students are around:

```text
168–172 cm
```

People are similar.

Small spread.

---

### High Standard Deviation

Students range from:

```text
145–195 cm
```

Huge differences.

Large spread.

---

# 4. Why is Standard Deviation Important in AI/ML?

It is used everywhere.

---

## a) Feature Standardization

One of the most important preprocessing techniques:

[
z = \frac{x-\text{mean}}{\text{standard deviation}}
]

This transforms data so that:

* Mean = 0
* Standard deviation = 1

Used in:

* Linear Regression
* Logistic Regression
* SVM
* PCA
* Neural Networks

---

## b) Detecting Outliers

Values that are very far from the mean (for example, 3 standard deviations away) may be outliers.

You'll use this during EDA.

---

## c) Understanding Variability

Two features can have the same mean.

Standard deviation tells you which one is more consistent.

---

---

# How to Interpret Standard Deviation

Suppose:

```text
Average Salary = ₹50,000
Standard Deviation = ₹2,000
```

This means:

> Most salaries are fairly close to ₹50,000.

---

Now suppose:

```text
Average Salary = ₹50,000
Standard Deviation = ₹50,000
```

This means:

> Salaries vary enormously.

The average alone doesn't tell the full story.

---

# Variance vs Standard Deviation

| Feature                | Variance  | Standard Deviation |
| ---------------------- | --------- | ------------------ |
| Measures spread?       | ✅         | ✅                  |
| Uses all observations? | ✅         | ✅                  |
| Same units as data?    | ❌         | ✅                  |
| Easier to interpret?   | ❌         | ✅                  |
| Commonly reported?     | Sometimes | Very often         |

---

# Python Implementation

## NumPy

Population Standard Deviation:

```python
import numpy as np

data = [2, 4, 6]

np.std(data)
```

Output:

```python
1.632993161855452
```

---

Sample Standard Deviation:

```python
np.std(data, ddof=1)
```

Output:

```python
2.0
```

---

## Pandas

```python
import pandas as pd

s = pd.Series([2, 4, 6])

s.std()
```

Output:

```python
2.0
```

Pandas uses **sample standard deviation** by default.

---

# One Important Thing

Many beginners think:

> Large standard deviation = bad.

That's not true.

It simply means:

> **The data is more spread out.**

Whether that's good or bad depends on the problem.

For example:

* Investment returns → High SD means high risk.
* Manufacturing → High SD means inconsistent quality.
* Human heights → Moderate SD is normal.

---

# Quick Recap

* Standard deviation measures how far values typically lie from the mean.
* It is:

[
\text{Standard Deviation} = \sqrt{\text{Variance}}
]

* Same units as the original data.
* Easier to interpret than variance.
* Used heavily in:

  * Feature scaling
  * Outlier detection
  * EDA
  * Many ML algorithms.
* Small SD → values clustered together.
* Large SD → values spread out.

One-line intuition to remember for Scaling

StandardScaler →

"How unusual is this value compared to the average?"

MinMaxScaler →

"Where does this value lie between the smallest and largest values?"

That's the fundamental difference between the two.
