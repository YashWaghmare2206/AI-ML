# Poisson Distribution

## 1. What is it?

A **Poisson Distribution** models the probability of a certain number of events occurring in a fixed interval of:

* Time
* Area
* Distance
* Volume

when those events occur randomly.

### Simple Definition

```text id="9g3e2h"
Poisson Distribution
=
Probability of getting k events
in a fixed interval
```

---

## 2. Intuition

Binomial asks:

```text id="dkvqzn"
Out of n trials,
how many successes occur?
```

Poisson asks:

```text id="k25yqt"
In a fixed period,
how many events occur?
```

Examples:

* Number of calls received in 1 hour.
* Number of accidents at a junction in a month.
* Number of customers entering a shop in 10 minutes.

Notice:

We are not counting successes and failures anymore.

We are counting **occurrences of events**.

---

## 3. When do we use Poisson Distribution?

Use it when:

### 1. Events are counted

Examples:

* Calls
* Accidents
* Customers
* Emails

---

### 2. Events occur randomly

You cannot predict the exact moment.

---

### 3. Fixed interval

Example:

```text id="x3n4h7"
1 hour
1 day
1 kilometer
1 webpage visit period
```

---

### 4. Average rate is known

Example:

```text id="hczj7n"
Average 4 calls per hour
```

---

## 4. Mathematical Definition

Let:

(lambda) = average number of events in the interval.

Then,

Probability of exactly (k) events:



---

### Meaning of Symbols

| Symbol    | Meaning                |
| --------- | ---------------------- |
| (k)       | Events we want         |
| (\lambda) | Average events         |
| (e)       | Euler's number ≈ 2.718 |
| (k!)      | Factorial              |

---

## 5. Why do we need λ?

λ tells us the average behavior.

Example:

```text id="d7s0pr"
Average calls per hour = 4
```

Example:

* 1 million website users.
* Probability of clicking ad = 0.00001.

Using Binomial becomes difficult.

Poisson becomes easier.

Poisson:

```text id="uxc4ll"
Count events
```

---

### ❌ λ is not a probability

Many students think:

```text id="brcnx7"
λ = probability
```

Wrong.

```text id="mh3vgf"
λ = average number of events
```

Example:

[
\lambda=10
]

means average 10 events.

Not 10%.

---

### ❌ Mean = Variance

This is true only for Poisson, not for all distributions.

### **When Binomial is not used and we opt for Poisson**

This is the exact situation where Poisson was invented.

# Example: Binomial → Poisson Approximation

## Problem

A website has:

```text
n = 1000 visitors
```

Each visitor clicks an ad with probability:

```text
p = 0.002
```

Find the probability that exactly 3 visitors click the ad.

---

# Method 1: Binomial

Since:

* Fixed trials = 1000 visitors
* Click / No Click
* Independent

This is Binomial.

Formula:

[
P(X=3)=\binom{1000}{3}(0.002)^3(0.998)^{997}
]

This is extremely annoying to calculate manually.

---

# Why Poisson Can Be Used?

Check conditions:

```text
n is very large?      Yes (1000)
p is very small?      Yes (0.002)
```

Therefore:

```text
Use Poisson Approximation
```

---

# Step 1: Calculate λ

For Binomial:

```text
λ = np
```

So:

[
\lambda = 1000 \times 0.002
]

[
\lambda = 2
]

Interpretation:

```text
Average clicks = 2
```

---

# Step 2: Use Poisson Formula

We want exactly:

[
k=3
]

Poisson formula:

[
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
]

Substitute:

[
P(X=3)
======

\frac{e^{-2}(2)^3}{3!}
]

---

# Step 3: Solve

[
3!=6
]

[
2^3=8
]

[
e^{-2}\approx0.1353
]

Therefore:

[
P(X=3)
======

\frac{0.1353\times8}{6}
]

# [

0.1804
]

---

# Final Answer

[
\boxed{P(X=3)\approx0.1804}
]

---

# What Happened Here?

Originally:

[
P(X=3)=\binom{1000}{3}(0.002)^3(0.998)^{997}
]

Huge ugly calculation.

Instead we converted:

```text
Binomial(n=1000,p=0.002)
↓
Poisson(λ=np=2)
```

and solved using:

[
P(X=3)
======

\frac{e^{-2}2^3}{3!}
]

much more easily.

---

# Rule to Remember

When:

```text
n ≥ 20 (usually much larger)
p ≤ 0.05 (very small)
```

and

[
\lambda=np
]

is moderate,

then:

```text
Binomial ≈ Poisson
```

---

### One More Quick Example

Suppose:

```text
10000 emails
Probability of spam = 0.0005
```

Then:

[
n=10000
]

[
p=0.0005
]

[
\lambda=np=5
]

Instead of Binomial:

[
\binom{10000}{k}(0.0005)^k(0.9995)^{10000-k}
]

use:

[
P(X=k)=\frac{e^{-5}5^k}{k!}
]

Much simpler.

---

The intuition is:

```text
Binomial:
Count successes in n trials.

When n becomes huge and p becomes tiny,
we stop caring about individual trials.

We only care about
"How many events happened?"

That becomes Poisson.
```
