# 68–95–99.7 Rule (Empirical Rule)

## 1. What is it?

The **68–95–99.7 Rule** tells us how much data lies within:

* 1 Standard Deviation
* 2 Standard Deviations
* 3 Standard Deviations

of the mean in a Normal Distribution.

---

## 2. Why do we need it?

Suppose:

```text id="m5trsi"
Mean = 70
SD = 10
```

You ask:

```text id="hb8f59"
Where do most students' marks lie?
```

The 68–95–99.7 Rule answers this.

---

## 3. The Rule




contains approximately:

```text id="ucf2bl"
68%
```

of the data.

---

### Within 2 SD



contains approximately:

```text id="6ec8pw"
95%
```

---

### Within 3 SD



contains approximately:

```text id="jv5s7e"
99.7%
```

Yes.

are often considered unusual.

Example:

```text id="e0snm5"
Mean salary = 50,000
SD = 10,000
```

Someone earning:

```text id="34w9nc"
200,000
```

is far from the mean and may be an outlier.

---

### Feature Analysis

When exploring datasets:

```python
df.describe()
```

you often use mean and SD to understand whether values are unusually large or small.

## 8. Key Points to Remember

```text id="9kvf2l"
Within 1 SD → 68%

Within 2 SD → 95%

Within 3 SD → 99.7%
```

For:

```text id="epyxoa"
Mean = 70
SD = 10
```

```text id="bj70lx"
60–80 → 68%

50–90 → 95%

40–100 → 99.7%
```

---

## 9. Common Confusion

### ❌ Exactly 68%?

Not necessarily.

```text id="qlzd90"
Approximately 68%
```

for a Normal Distribution.

---

### ❌ Works for Every Distribution?

No.

This rule is specifically for:

```text id="cb6q1o"
Normal Distributions
```

---

**Topic Completed:** 68–95–99.7 Rule ✅

Next remaining topics:

```text id="fh0ysq"
1. Mean and Spread of Distributions
2. Real-life Examples & Comparison
```
