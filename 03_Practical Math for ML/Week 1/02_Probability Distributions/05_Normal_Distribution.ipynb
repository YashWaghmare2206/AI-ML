# Normal Distribution

## 1. What is it?

A **Normal Distribution** is a probability distribution where:

* Most values are near the center.
* Fewer values occur as we move away from the center.
* The graph looks like a **bell-shaped curve**.

### Simple Definition

```text id="y9srmn"
Most observations occur near the average.
Extreme observations are rare.
```

---

## 2. Intuition

Imagine measuring heights of 10,000 people.

You'll find:

```text id="k8mjyr"
Very short people → Few
Average height people → Many
Very tall people → Few
```

So the graph looks like:

```text id="e4n3bp"
        /\
      /    \
    /        \
  /            \
------------------
```

Highest point:

```text id="olzk5x"
Mean (Average)
```

---

## 3. Why do we need it?

Many real-world things naturally follow a Normal Distribution.

Examples:

* Heights
* Weight
* Exam scores
* Measurement errors
* IQ scores

Because of this, Normal Distribution becomes one of the most important distributions in statistics and ML.

Example:

Height depends on:

* Genetics
* Nutrition
* Sleep
* Environment

Many factors combine, producing a Normal Distribution.

## 13. Common Confusions

### ❌ Mean is not the spread

Mean tells:

```text id="oj3z9u"
Where the center is
```

Standard deviation tells:

```text id="8f3l11"
How spread out the data is
```

---

### ❌ Normal Distribution is Continuous

Unlike Bernoulli, Binomial and Poisson.

Normal can take values like:

```text id="g6ix2e"
65.2
65.21
65.213
...
```

Infinite possibilities.

---

### ❌ Peak does NOT mean maximum value

The peak simply indicates:

```text id="u6jg1g"
Most likely region
```

not the largest possible value.
