# Pandas Statistics

---

## 1. `.describe()`

### What it does

Returns summary statistics.

### Syntax

```python
df.describe()
```

### Returns

```text
count
mean
std
min
25%
50%
75%
max
```

### Example

```python
df["Age"].describe()
```

---

## 2. `.mean()`

### What it does

Returns average value.

### Syntax

```python
df["Age"].mean()
```

### Important Parameters

```python
df.mean(numeric_only=True)
```

| Parameter    | Use                        |
| ------------ | -------------------------- |
| numeric_only | Ignore non-numeric columns |

---

## 3. `.median()`

### What it does

Returns middle value after sorting.

### Syntax

```python
df["Age"].median()
```

### Example

```python
df["Salary"].median()
```

---

## 4. `.mode()`

### What it does

Returns most frequent value.

### Syntax

```python
df["City"].mode()
```

### Important Point

Returns a Series.

```python
df["City"].mode()[0]
```

to get first mode.

### Example

```python
A
B
B
C
```

Output:

```python
B
```

---

## 5. `.quantile()`

### What it does

Returns specified quantile.

### Syntax

```python
df["Age"].quantile(q)
```

### Common Values

| q    | Meaning |
| ---- | ------- |
| 0.25 | Q1      |
| 0.50 | Median  |
| 0.75 | Q3      |

### Examples

```python
df["Age"].quantile(0.25)
```

Q1

```python
df["Age"].quantile(0.75)
```

Q3

---

## 6. `.value_counts()`

### What it does

Counts occurrences of unique values.

### Syntax

```python
df["Gender"].value_counts()
```

### Example

```python
Male
Male
Female
Male
```

Output:

```python
Male      3
Female    1
```

### Important Parameters

```python
df["Gender"].value_counts(normalize=True)
```

Returns percentages.

---

```python
df["Gender"].value_counts(dropna=False)
```

Includes NaN values.

---

# Important Interview Differences

### `.describe()`

```python
df.describe()
```

Numerical columns only.

---

For categorical columns:

```python
df.describe(include="object")
```

or

```python
df.describe(include="all")
```

---

### `.std()`

Pandas default:

```python
df["Age"].std()
```

uses:

```python
ddof=1
```

Sample SD.

---

### `.var()`

Pandas default:

```python
df["Age"].var()
```

uses:

```python
ddof=1
```

Sample Variance.
