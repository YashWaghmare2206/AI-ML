### Important Interview Point

```python
np.var()
np.std()
```

By default use:

```python
ddof=0
```

which means:

```text id="8f1v5q"
Population Variance
Population SD
```

For sample statistics:

```python
ddof=1
```

### **Important**

# Step 1: Why does `ddof` exist at all?

Suppose population is:

```text
10 20 30 40 50
```

Population mean:

```text
30
```

Population variance can be calculated exactly because:

```text
We know every value.
```

No estimation is needed.

---

Now imagine you don't have all values.

You only sample:

```text
20 30 40
```

Now you want to estimate:

```text
Population variance
```

from only 3 observations.

This is where the problem begins.

---

# Step 2: What would we naturally do?

Compute sample mean:

```text
(20+30+40)/3 = 30
```

Then deviations:

```text
-10
0
10
```

Squares:

```text
100
0
100
```

Sum:

```text
200
```

Now divide by:

```text
n = 3
```

Variance:

```text
200/3 = 66.67
```

Looks reasonable.

---

# Step 3: The hidden problem

Notice something weird.

The sample mean was computed from the same data:

```text
20 30 40
```

Because of that, the sample mean gets pulled toward the sample points.

That makes deviations slightly smaller on average.

Meaning:

```text
Sample variance tends to be too small.
```

This happens systematically across millions of random samples.

---

# Step 4: Statisticians tested it

Imagine taking:

```text
Thousands of samples
```

from the same population.

For each sample:

```text
Compute variance using n
```

Then average those variances.

Result:

```text
Average sample variance
<
True population variance
```

So it is biased downward.

---

# Step 5: How much correction is needed?

Mathematicians proved:

If you divide by:

```text
n-1
```

instead of

```text
n
```

then:

```text
Average sample variance
=
True population variance
```

This removes the bias.

# Step 6: Why exactly n-1?

This is the part most courses skip.

Suppose sample:

```text
20
30
?
```

and sample mean must be:

```text
30
```

If first two values are fixed:

```text
20
30
```

the third value is forced:

```text
40
```

It cannot be anything else.

Why?

Because the mean must stay 30.

---

So after calculating the sample mean:

```text
Only n-1 values are truly free.
```

The last one becomes determined automatically.

This is called:

```text
Degrees of Freedom
```

Hence:

```text
Degrees of Freedom = n - 1
```

and NumPy names it:

```python
ddof
```

which means:

```text
Delta Degrees Of Freedom
```

---

# Final intuition

```text
Population:
Know everything
→ divide by N
```

```text
Sample:
Estimating from incomplete data
+
Mean estimated from sample itself
→ variance becomes too small
→ divide by n-1
```

---

# For AI/ML

You do NOT need to derive this.

Just remember:

```python
np.var(arr)           # ddof=0
np.std(arr)           # ddof=0
```

Population formula.

```python
np.var(arr, ddof=1)
np.std(arr, ddof=1)
```

Sample formula.

And the reason:

```text
n-1 compensates for the bias introduced
when the mean is estimated from the sample itself.
```

That's the actual reason, not "because one row is missing."
