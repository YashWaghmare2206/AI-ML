# 01_Gradient_Vector

In the previous topic, we introduced the gradient.

A **gradient vector** is simply the vector formed by all partial derivatives of a function.

---

## Definition



---

## Example



---

## What Does Each Value Mean?

For:



* 4 → how fast the function changes along the x-direction
* 6 → how fast the function changes along the y-direction

Since 6 is larger, the function is changing faster along y at that point.

---

## Why It's a Vector

A derivative for one variable is a single number.

For multiple variables, we need one derivative per variable.

So we store them together as a vector:

**`[4,6]`**

This lets us work with many parameters efficiently.

---

## Why ML Cares



---

## Python

```python
import numpy as np

x, y = 2, 3

gradient = np.array([2*x, 2*y])

print(gradient)
```

Output:

```python
[4 6]
```

---

### Key Takeaway

A gradient vector is:

> A vector containing all partial derivatives of a multivariable function.



In ML, the gradient vector tells us how the loss changes with respect to every model parameter.

# Example to understand

Let's take the simplest possible example.

## Loss Function

[
L(w)=w^2
]

This is our error function.

Goal:

> Find the value of (w) that makes the loss as small as possible.

---

## Step 1: Compute Derivative

[
L(w)=w^2
]

[
\frac{dL}{dw}=2w
]

This derivative is the gradient (since there is only one variable).

---

## Step 2: Start Somewhere

Let's start with:

[
w=3
]

Current loss:

[
L(3)=3^2=9
]

---

## Step 3: Find Gradient

[
\frac{dL}{dw}=2w
]

At (w=3):

[
\frac{dL}{dw}=2(3)=6
]

Gradient = 6

Meaning:

> If we increase (w), loss will increase.

---

## Step 4: Apply Gradient Descent

Learning rate:

[
\alpha=0.1
]

Update rule:

[
w_{new}=w_{old}-\alpha \times gradient
]

Substitute values:

[
w_{new}=3-0.1(6)
]

[
w_{new}=3-0.6
]

[
w_{new}=2.4
]

---

## Step 5: Check New Loss

Old loss:

[
L(3)=9
]

New loss:

[
L(2.4)=2.4^2
]

[
=5.76
]

Loss decreased:

[
9 \rightarrow 5.76
]

✅ Good

---

## One More Iteration

Current:

[
w=2.4
]

Gradient:

[
2(2.4)=4.8
]

Update:

[
w=2.4-0.1(4.8)
]

[
w=1.92
]

Loss:

[
L(1.92)=3.6864
]

Again:

[
5.76 \rightarrow 3.6864
]

✅ Loss decreased.

---

## What Eventually Happens?

Iterations:

[
3
\rightarrow
2.4
\rightarrow
1.92
\rightarrow
1.536
\rightarrow
...
\rightarrow
0
]

The weight keeps moving toward:

[
w=0
]

because:

[
L(0)=0
]

which is the minimum possible loss.

---

### What the Gradient Was Telling Us

At (w=3):

Gradient = 6

This means:

> "Move right (positive direction) and loss increases."

So we move left:

[
-6
]

At (w=-3):

Gradient:

[
2(-3)=-6
]

Now the gradient points left.

So we move opposite:

[
-(-6)=+6
]

which moves us right toward 0.

Notice something important:

The gradient always points **away from the minimum**, and the negative gradient always points **toward the minimum**.

That is why Gradient Descent works.
