## Python

```python
import sympy as sp

x = sp.Symbol('x')

f = 3*x**2 + 2*x + 5

print(sp.diff(f, x))
```

Output:

```python
6*x + 2
```

`SymPy` stands for **Symbolic Python**.

It is a Python library for **symbolic mathematics**, meaning it works with mathematical symbols and expressions instead of only numerical values.

For example:

```python
import sympy as sp

x = sp.Symbol('x')
```

Here:

```python
x = sp.Symbol('x')
```

does **not** assign a number to `x`.

Instead, it tells SymPy:

> "Treat `x` as a mathematical variable."

Now you can build expressions:

```python
f = 3*x**2 + 2*x + 5
```

SymPy stores this as the algebraic expression:

[
3x^2 + 2x + 5
]

Then:

```python
sp.diff(f, x)
```

computes the symbolic derivative:

[
6x + 2
]

without needing a value for (x).

---

### Difference from NumPy

**NumPy**

```python
x = 3
f = 3*x**2 + 2*x + 5
print(f)
```

Output:

```python
38
```

Everything becomes a number.

---

**SymPy**

```python
x = sp.Symbol('x')
f = 3*x**2 + 2*x + 5
print(f)
```

Output:

```python
3*x**2 + 2*x + 5
```

It keeps the mathematical expression intact.

---

### Do We Use SymPy in ML?

Usually **no**.

Most ML libraries like:

* NumPy
* PyTorch
* TensorFlow

compute derivatives numerically or automatically.

We use SymPy mainly for:

* Learning calculus
* Verifying derivatives
* Mathematical exploration

Not for training real ML models.

The line

```python
x = sp.Symbol('x')
```

simply means:

> "Create a symbolic mathematical variable named x."
