## Why ML Cares

In real ML systems:

* Sometimes derivatives are difficult to derive manually.
* Numerical derivatives can be used to check whether gradients are correct.
* Gradient checking in deep learning uses this idea.

Modern frameworks use **Automatic Differentiation**, which is faster and more accurate than numerical differentiation.

---

## Python

```python
def f(x):
    return x**2

x = 3
h = 0.0001

derivative = (f(x + h) - f(x)) / h

print(derivative)
```

Output:

```python
6.0001
```
