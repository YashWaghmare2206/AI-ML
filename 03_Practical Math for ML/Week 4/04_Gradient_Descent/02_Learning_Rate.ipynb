## If Learning Rate Is Too Large



The updates overshoot and may never reach the minimum.

Loss can increase instead of decrease.

---

## Goal

Choose a learning rate that:

* is not too small
* is not too large
* reaches the minimum efficiently

---

## Why ML Cares

Gradient tells us:

> Which direction to move.

Learning Rate tells us:

> How far to move.

Both are needed.

---

### Key Takeaway

In:



* Gradient decides the direction.
* Learning Rate decides the size of the step.

A good learning rate helps Gradient Descent converge quickly and stably.
