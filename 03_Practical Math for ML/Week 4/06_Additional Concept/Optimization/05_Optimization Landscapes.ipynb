# What Is an Optimization Landscape?

An optimization landscape is simply:

> The graph of the loss function over all possible parameter values.

---

Suppose a model has one parameter:

**`w`**

Loss:

**`L(w)`**

The graph of (L) against (w) is the optimization landscape.

---
