# What Is a Convex Function?

A function is convex if:

> Any straight line connecting two points on the graph lies above or on the graph.

Formal definition is not important yet.

The shape is what matters.

---

## Convex Shape

Think of a bowl:

```text
      •
    /   \
   /     \
  /       \
 /         \
-------------
```

A convex function has one bowl-shaped minimum.

## Why Gradient Descent Loves Convex Functions

Imagine standing on a smooth bowl.

The gradient always points toward the bottom.

There are no traps.

Gradient Descent eventually finds the minimum.

---

## Convex Optimization

Optimization problems involving convex functions are called:

> Convex Optimization Problems

These are easier because:

* No bad local minima
* One global minimum
* Gradient Descent works reliably

---

## Example in ML

Linear Regression uses:

[
MSE
===

\frac1n\sum(y-\hat y)^2
]

The loss surface of linear regression is convex.

Therefore Gradient Descent can reliably find the best parameters.
