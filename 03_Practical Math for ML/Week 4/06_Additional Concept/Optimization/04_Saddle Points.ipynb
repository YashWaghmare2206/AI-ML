# Saddle Points

After local minima and global minima, the next important concept is **saddle points**.

In modern deep learning, saddle points are often a bigger problem than local minima.

---

# What Is a Saddle Point?

A saddle point is a point where:



(the gradient is zero)

but it is **not**:

* A local minimum
* A local maximum

---

## Intuition

At a saddle point:

* One direction goes upward.
* Another direction goes downward.

So it is neither a valley nor a peak.

---

# Why Saddle Points Matter in Deep Learning

Neural networks may have:

* Millions of parameters
* Extremely high-dimensional loss surfaces

In such spaces:

* Saddle points are very common.
* Local minima are often less of a problem.

---

Historically people worried about:

> Local minima.

Modern research found:

> Saddle points are often the bigger challenge.

---

# How Optimizers Escape Saddle Points

Optimizers like:

* SGD
* Adam
* RMSProp

contain randomness and momentum.

This often helps them move away from saddle points instead of staying stuck.

---

# Why This Topic Exists

When optimizing:



a zero gradient does **not** always mean success.

It could mean:

* Global minimum ✅
* Local minimum ✅
* Saddle point ❌

We need additional information to distinguish them.

---

# Key Takeaways

* Saddle point has zero gradient.
* It is not a minimum.
* It is not a maximum.
* One direction goes up, another goes down.
* Common in neural network optimization.
* Hessian matrix helps identify saddle points.

### One-line Summary

> A saddle point is a stationary point where the gradient is zero but the function is neither a minimum nor a maximum.
