# 04_System_of_Linear_Equations

## 6. No Solution

---

## Formula

There is **no special formula** for no solution.

Generally:

* The equations represent **parallel lines** (for 2 variables).
* The coefficient matrix is **singular**.



---

## Explanation

A system has **no solution** when there is **no single set of values** that satisfies all the equations.

For two variables, this happens when the lines are **parallel**.

Parallel lines never intersect, so there is no common point.

---

## Example



---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 1],
    [1, 1]
])

B = np.array([2, 5])

np.linalg.solve(A, B)
```

**Output**

```text
LinAlgError: Singular matrix
```

Again, `np.linalg.solve()` raises an error because the system **does not have a unique solution**.

> **Note:** The same error (`Singular matrix`) is raised for both **infinite solutions** and **no solution**. `np.linalg.solve()` only solves systems with a unique solution.

---

## ML Connection

A "no solution" situation is less common in ML because most ML algorithms are designed to find the **best approximate solution** rather than requiring an exact solution.

However, inconsistent systems can appear when:

* Data contains conflicting constraints.
* Mathematical models are over-constrained.
* During theoretical analysis of linear systems.
