## Explanation

The inverse matrix "undoes" the effect of the original matrix.

Think:

```text id="f52v1h"
Number     Inverse

5          1/5

10         1/10
```

because



Similarly:

```text id="vxk2bq"
Matrix     Inverse Matrix

A          A⁻¹
```

because

## Python / NumPy

```python id="5dd06l"
import numpy as np

A = np.array([
    [2, 0],
    [0, 3]
])

print(np.linalg.inv(A))
```

Output:

```text id="s81y4e"
[[0.5        0.        ]
 [0.         0.33333333]]
```

---

## Verification

```python id="vm0l9n"
A_inv = np.linalg.inv(A)

print(A @ A_inv)
```

Output:

```text id="xtm6gu"
[[1. 0.]
 [0. 1.]]
```

Identity matrix obtained.

## Important Note

Not every matrix has an inverse.



Since determinant is zero:

```text id="5b1yyn"
No inverse exists
```

This matrix is called **singular**.

---

### Memory Chain

```text id="ijw69m"
det ≠ 0
↓
Non-Singular
↓
Inverse Exists
↓
Can solve AX = B

det = 0
↓
Singular
↓
No Inverse
↓
Cannot uniquely solve AX = B
```
