## Explanation

In practice, you will almost never compute inverse matrices by hand.

NumPy provides built-in functions.

The most common ones are:

```python
np.linalg.inv()
np.linalg.solve()
```

---

## 1. Finding the Inverse

Matrix:



Code:

```python
import numpy as np

A = np.array([
    [2, 3],
    [4, 5]
])

A_inv = np.linalg.inv(A)

print(A_inv)
```

Output:

```text
[[-2.5  1.5]
 [ 2.  -1. ]]
```

---

## 2. Verify the Inverse

Check:



Code:

```python
print(A @ A_inv)
```

Output:

```text
[[1. 0.]
 [0. 1.]]
```

This is the Identity Matrix.

---

## 3. Determinant

Before finding an inverse, check:

```python
print(np.linalg.det(A))
```

Output:

```text
-2.0
```

Since:

```text
det ≠ 0
```

the inverse exists.

---

## 4. Solving AX = B

Suppose:



```python
A = np.array([
    [2, 3],
    [4, 5]
])

B = np.array([8, 10])
```

### Recommended

```python
X = np.linalg.solve(A, B)

print(X)
```

Output:

```text
[2.2 1.2]
```

---

### Less Preferred

```python
X = np.linalg.inv(A) @ B
```

This gives the same answer but is usually slower and less numerically stable.

---

## 5. Singular Matrix Example

```python
A = np.array([
    [1, 2],
    [2, 4]
])

np.linalg.inv(A)
```

Output:

```text
LinAlgError: Singular matrix
```

Why?

Because:



No inverse exists.

## ML Connection

These are the NumPy functions you'll actually use in AI/ML:

| Task        | Function                  |
| ----------- | ------------------------- |
| Determinant | `np.linalg.det()`         |
| Inverse     | `np.linalg.inv()`         |
| Solve AX=B  | `np.linalg.solve()`       |
| Eigenvalues | `np.linalg.eig()`         |
| Matrix Rank | `np.linalg.matrix_rank()` |

You'll encounter these repeatedly in:

* Linear Regression
* PCA
* Eigenvectors
* Matrix Decomposition
* Optimization

---

## Important Note

For interviews and practical ML:

```text
np.linalg.solve()
```

is generally preferred over:

```text
np.linalg.inv(A) @ B
```

because it directly solves the system without explicitly computing the inverse.
