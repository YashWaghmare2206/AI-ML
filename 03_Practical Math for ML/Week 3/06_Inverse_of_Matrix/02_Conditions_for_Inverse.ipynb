## Condition

A matrix has an inverse **only if**:



This is the most important condition.

---

## Explanation

Not every matrix can be inverted.

To check whether an inverse exists:

1. Calculate the determinant.
2. If determinant ≠ 0 → inverse exists.
3. If determinant = 0 → inverse does not exist.

## Another Important Condition

The matrix must be **square**.

Valid:

```text id="6h9lr9"
2×2
3×3
4×4
```

Invalid:

```text id="qwy0kn"
2×3
3×5
4×2
```

Rectangular matrices do not have a standard inverse.

---

## Python / NumPy

```python id="uz5k9u"
import numpy as np

A = np.array([
    [2, 3],
    [4, 5]
])

print(np.linalg.det(A))
```

Output:

```text id="9x7o9i"
-2.0
```

Since determinant is not zero:

```python id="9it0cm"
np.linalg.inv(A)
```

works successfully.

---

Example with determinant zero:

```python id="9kzhyr"
A = np.array([
    [1, 2],
    [2, 4]
])

np.linalg.inv(A)
```

Output:

```text id="xy6hgm"
LinAlgError: Singular matrix
```

---

## ML Connection

Many ML formulas require matrix inversion.

If the matrix is singular:

```text id="6xj4hn"
No inverse
↓
Formula fails
```

This is why duplicate or highly dependent features can cause problems.
