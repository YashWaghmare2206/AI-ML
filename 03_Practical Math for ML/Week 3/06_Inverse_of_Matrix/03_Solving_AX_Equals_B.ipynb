Output:

```text id="x8n2wq"
[[2.]
 [2.]]
```

---

## Better NumPy Method

Instead of:

```python id="1w7b5m"
np.linalg.inv(A) @ B
```

Use:

```python id="n6c2k8"
np.linalg.solve(A, B)
```

because it is:

* Faster
* More accurate
* Preferred in ML

---

## ML Connection

This idea appears everywhere:

### Linear Regression



### Least Squares

### Optimization

### Numerical Linear Algebra

Many ML formulas are ultimately solving some version of:



---

## Important Note

The formula



works only if:

```text id="p5r8v3"
A is square
and
det(A) ≠ 0
```

Otherwise:

```text id="m3n6w1"
No inverse exists
```

and the formula cannot be used.

---

## Memory Chain

```text id="t7k4q9"
AX = B
↓
Need X
↓
Multiply by A⁻¹
↓
X = A⁻¹B
```
