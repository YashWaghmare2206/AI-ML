# 03_Matrices

## 4. Scalar Multiplication

---

## Formula



Every element of the matrix is multiplied by the scalar.

---

## Explanation

Scalar multiplication for matrices is exactly the same as for vectors.

A scalar multiplies **every element** in the matrix.

* Positive scalar → Values increase or decrease, but signs stay the same.
* Negative scalar → Values change sign and are multiplied by the scalar.
* Zero → Every element becomes 0.

---

## Example



---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

print(3 * A)
```

**Output**

```text
[[ 3  6]
 [ 9 12]]
```

Another example:

```python
print(-2 * A)
```

**Output**

```text
[[ -2  -4]
 [ -6  -8]]
```

---

## ML Connection

Scalar multiplication is used throughout ML.

Examples:

* Scaling feature values.
* Updating weights during **Gradient Descent**:

[
W_{\text{new}}
==============

## W_{\text{old}}

\alpha \times \text{Gradient}
]

where:

* (W) = Weight matrix
* (\alpha) = Learning rate (scalar)

The learning rate multiplies every value in the gradient matrix.

---

## Important Note

The **shape of the matrix never changes**.

Example:

If

[
A
]

has shape

[
(2,3)
]

then

[
5A
]

also has shape

[
(2,3)
