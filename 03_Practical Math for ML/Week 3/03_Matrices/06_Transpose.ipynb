## Explanation

A **transpose** is obtained by **swapping rows and columns**.

That means:

* Row 1 becomes Column 1.
* Row 2 becomes Column 2.
* Row 3 becomes Column 3.
* And so on.

You can think of it as **flipping the matrix across its main diagonal** (top-left to bottom-right).

---

## Example



---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(A.T)
```

**Output**

```text
[[1 4]
 [2 5]
 [3 6]]
```

You can also use:

```python
np.transpose(A)
```

Both produce the same result.

---

## ML Connection

Transpose is used throughout Machine Learning:

* Matrix multiplication
* Linear Regression
* PCA
* Covariance matrices
* Neural Networks

For example, in Linear Regression, you'll later see formulas like:

[
X^TX
]

where:

* (X) = Dataset matrix
* (X^T) = Transpose of the dataset matrix

### 3. 1D NumPy vectors

Be careful with NumPy:

```python
v = np.array([1, 2, 3])

print(v.shape)
```

Output:

```text
(3,)
```

Now transpose it:

```python
print(v.T)
```

Output:

```text
[1 2 3]
```

The output **does not change**.

Why?

Because a 1D NumPy array has no distinction between rows and columns.

If you want a true column vector:

```python
v = np.array([[1],
              [2],
              [3]])

print(v.T)
```

Output:

```text
[[1 2 3]]
```

Here the transpose works as expected because `v` is a **2D matrix**.
