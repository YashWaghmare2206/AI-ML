## Explanation

Matrix addition works exactly like vector addition.

To add two matrices:

1. They must have the **same shape**.
2. Add each corresponding element.

Every element in one matrix is added to the element at the **same row and column** in the other matrix.

---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A + B)
```

**Output**

```text
[[ 6  8]
 [10 12]]
```

---

## ML Connection

Matrix addition is used frequently in ML.

Examples:

* Adding a **bias matrix** to predictions.
* Combining feature matrices.
* Updating parameters during optimization.
* Image processing (adding pixel values).

Although you may not write `A + B` directly very often, ML libraries perform millions of matrix additions internally.

---

## Important Note

### Rule: Shapes must be the same

✅ Valid:

[
(2,3)+(2,3)
]

❌ Invalid:

[
(2,3)+(3,2)
]



This is **not allowed** because the shapes are different.

In NumPy:

```python
A.shape  # (2,2)
B.shape  # (2,3)

A + B
```
