## 8. Diagonal Matrix

---

## Formula

A **Diagonal Matrix** is a **square matrix** in which:

* Elements on the **main diagonal** can be any value.
* All **non-diagonal elements are 0**.

Example:



---

## Explanation

Only the **main diagonal** (top-left to bottom-right) can contain non-zero values.

Everything else must be **0**.

---

## Example

Valid diagonal matrix:

[
\begin{bmatrix}
4 & 0 & 0\
0 & 9 & 0\
0 & 0 & 6
\end{bmatrix}
]

Not a diagonal matrix:



because **1** is outside the main diagonal.

---

## Python / NumPy

```python
import numpy as np

D = np.diag([2, 5, 7])

print(D)
```

**Output**

```text
[[2 0 0]
 [0 5 0]
 [0 0 7]]
```

---

## ML Connection

Diagonal matrices are used in:

* Covariance matrices
* Feature scaling
* Eigenvalue decomposition
* PCA
* Matrix decompositions

---

## Important Note

Every **Identity Matrix** is a **Diagonal Matrix**, but not every **Diagonal Matrix** is an Identity Matrix.

## 9. Symmetric Matrix

---

## Formula

A matrix is **symmetric** if:



This means the matrix is equal to its transpose.

---

## Explanation

A symmetric matrix looks the same even after transposing it.

This happens because:

[
a_{ij}=a_{ji}
]

In other words,

* Row 1, Column 2 = Row 2, Column 1
* Row 1, Column 3 = Row 3, Column 1
* Row 2, Column 3 = Row 3, Column 2

---

## Example



it is a **Symmetric Matrix**.

---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [2, 5, 6],
    [3, 6, 8]
])

print(np.array_equal(A, A.T))
```

**Output**

```text
True
```

---

## ML Connection

Symmetric matrices appear in:

* Covariance matrices
* Correlation matrices
* PCA
* Eigenvalue decomposition
* Optimization algorithms

Many important ML matrices are symmetric because they have useful mathematical properties.

---

## Important Note

A symmetric matrix **must be square**.

Example:

```text
(3×3) ✅ Can be symmetric

(2×2) ✅ Can be symmetric

(2×3) ❌ Cannot be symmetric
```

because:

[
A=(m,n)
]

and

[
A^T=(n,m)
]

If (m \ne n), then (A) and (A^T) don't even have the same shape, so they cannot be equal.
