# 03_Matrices

## 1. Matrix Representation

---

## Formula

A **matrix** is a rectangular arrangement of numbers in rows and columns.



---

## Explanation

A **matrix** is simply a collection of vectors (or numbers) arranged in a table.

Think of it as a spreadsheet:

```text
Rows
↓
[ 1   2   3 ]
[ 4   5   6 ]
[ 7   8   9 ]
        →
      Columns
```

Each value has a unique position identified by:

* Row number
* Column number

For example, in the above matrix:

* `5` is at **Row 2, Column 2**
* `3` is at **Row 1, Column 3**
* `7` is at **Row 3, Column 1**

---

## Example

Suppose we have data for 3 students:

| Height | Weight | Age |
| ------ | ------ | --- |
| 170    | 65     | 22  |
| 180    | 75     | 25  |
| 160    | 55     | 20  |

This can be represented as the matrix:



Here:

* Each **row** = One student (one data sample).
* Each **column** = One feature.

This is exactly how datasets are stored in Machine Learning.

---

## Python / NumPy

```python
import numpy as np

X = np.array([
    [170, 65, 22],
    [180, 75, 25],
    [160, 55, 20]
])

print(X)
```

**Output**

```text
[[170  65  22]
 [180  75  25]
 [160  55  20]]
```

To access an element:

```python
print(X[1, 2])
```

**Output**

```text
25
```

Why?

* Indexing starts from **0** in NumPy.
* `X[1,2]` means:

  * Row index `1` → 2nd row
  * Column index `2` → 3rd column

So the value is **25**.

---

## ML Connection

In Machine Learning, the entire dataset is usually stored as a matrix.

For example:

```text
          Features
        Height Weight Age
Student1   170     65   22
Student2   180     75   25
Student3   160     55   20
```

In notation:

* **Rows** → Samples (observations).
* **Columns** → Features (variables).

Most ML algorithms expect the input data as a matrix, often denoted by **X**.

---

## Important Note

Don't confuse a **vector** with a **matrix**.

| Vector          | Matrix              |
| --------------- | ------------------- |
| One-dimensional | Two-dimensional     |
| `[3, 4, 5]`     | `[[3,4,5],[6,7,8]]` |
| Shape: `(3,)`   | Shape: `(2,3)`      |

A matrix is essentially a collection of vectors arranged in rows (or equivalently, columns).
