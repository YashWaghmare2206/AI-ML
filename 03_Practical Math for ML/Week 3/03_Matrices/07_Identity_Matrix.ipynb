## 7. Identity Matrix

---

## Formula



---

## Explanation

The Identity Matrix plays the same role in matrix multiplication that **1** plays in normal multiplication.



---

## Example



---

## Python / NumPy

```python
import numpy as np

I = np.eye(3)

print(I)
```

**Output**

```text
[[1. 0. 0.]
 [0. 1. 0.]
 [0. 0. 1.]]
```

Example:

```python
A = np.array([
    [2, 3],
    [4, 5]
])

I = np.eye(2)

print(A @ I)
```

**Output**

```text
[[2. 3.]
 [4. 5.]]
```

---

## ML Connection

Identity matrices appear in many ML algorithms:

* Matrix inverse

[
A \times A^{-1}=I
]

* Linear Regression (Regularization)
* PCA
* Covariance matrices
* Matrix decompositions

You'll encounter them frequently in advanced linear algebra and optimization.

---

## Important Notes

### 1. Identity matrix must be square

✅ Valid:

```text
(2×2)

(3×3)

(5×5)
```

❌ Invalid:

```text
(2×3)

(3×4)
```

An identity matrix **cannot** be rectangular.

---

### 2. Don't confuse Identity Matrix with Diagonal Matrix



Every **identity matrix is a diagonal matrix**, but **not every diagonal matrix is an identity matrix**.
