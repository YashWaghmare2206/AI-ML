# 07_Eigenvalues_and_Eigenvectors

## 2. Eigenvectors

---

## Definition

An **eigenvector** is a non-zero vector whose direction remains unchanged when multiplied by a matrix.



---

## Explanation

Most vectors change direction when multiplied by a matrix.

Example:

```text id="52e42f"
Matrix × Vector
↓
New direction
```

But a few special vectors don't.

They only get:

* Stretched
* Shrunk
* Flipped

These special vectors are called **eigenvectors**.

---

## Example



Direction stayed the same.

Only the length changed.



---

## Visual Intuition

```text id="f6vxv4"
Before:

      ↑
      |
      |
      |

After:

          ↑
          |
          |
          |
          |
          |
```

Same direction.

Different length.

That is an eigenvector.

---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [2,0],
    [0,3]
])

values, vectors = np.linalg.eig(A)

print(values)
print(vectors)
```

Output:

```text
[2. 3.]

[[1. 0.]
 [0. 1.]]
```

Meaning:

```text
Eigenvalues:
2
3

Eigenvectors:
[1,0]
[0,1]
```

---

## ML Connection

Eigenvectors tell us:

```text id="8mr2lu"
Important directions
inside data
```

In PCA:

* Eigenvectors → Important directions
* Eigenvalues → Importance of those directions

Together they help reduce dimensions while keeping maximum information.

---

## Important Note

### Eigenvector ≠ Eigenvalue

```text id="lzdys8"
Eigenvector
→ Direction

Eigenvalue
→ Scaling amount
```
