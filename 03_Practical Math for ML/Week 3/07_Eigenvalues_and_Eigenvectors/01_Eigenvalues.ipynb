## Before the Definition

Normally when a matrix multiplies a vector:



the vector's:

* Length changes
* Direction changes

Both can change.

---

## Special Case

Sometimes a vector keeps the **same direction** after multiplication.

Only its length changes.



---

## Definition

An **Eigenvalue** is the amount by which an eigenvector is scaled after multiplication by a matrix.

It is represented by:



(Greek letter Lambda)

## Python / NumPy

We usually compute eigenvalues with:

```python id="31wthf"
np.linalg.eig()
```

Example:

```python id="4dqudx"
import numpy as np

A = np.array([
    [2, 0],
    [0, 2]
])

values, vectors = np.linalg.eig(A)

print(values)
```

Output:

```text id="2w0zpb"
[2. 2.]
```

These are the eigenvalues.

---

## ML Connection

Eigenvalues tell us:

* How important a direction is.
* How much variance exists in a direction.
* Which directions contain useful information.

This becomes extremely important in:

* PCA
* Dimensionality Reduction
* Covariance Matrices
* Computer Vision
* Recommendation Systems

---

## Important Note

An eigenvalue is **not a vector**.

It is just a number.

```text id="d7wsu8"
Eigenvalue  → Number

Eigenvector → Vector
```
