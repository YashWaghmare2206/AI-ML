## Why Are Eigenvalues and Eigenvectors Important?

Without PCA:

```text id="uq76ix"
100 Features
↓
Model
```

This can be:

* Slow
* Expensive
* Noisy

PCA helps reduce dimensions while keeping most of the information.

---

## PCA Goal

Convert:

```text id="ez79jz"
100 Features
```

into

```text id="kwwz5s"
10 Important Features
```

while losing as little information as possible.

---

## Where Do Eigenvalues and Eigenvectors Come In?

PCA computes the covariance matrix:



---

## Meaning of Eigenvectors in PCA

Eigenvectors represent:

```text id="7tjl4h"
Important directions
in the data
```

Think:

```text id="9zcltb"
Direction A
Direction B
Direction C
```

Some directions contain more information than others.

PCA tries to find those directions.

---

## Meaning of Eigenvalues in PCA

Eigenvalues tell us:

```text id="bd7vjy"
How much information
(or variance)
exists in that direction
```

Large eigenvalue:

```text id="4wj3if"
Important direction
```

Small eigenvalue:

```text id="h7phze"
Less important direction
```

---

## Example

Suppose PCA produces:

| Eigenvector | Eigenvalue |
| ----------- | ---------- |
| v₁          | 120        |
| v₂          | 35         |
| v₃          | 2          |

Interpretation:

```text id="lgb8lf"
v₁ → Most information

v₂ → Some information

v₃ → Almost no information
```

So PCA may keep:

```text id="lchijc"
v₁
v₂
```

and discard:

```text id="e95ksz"
v₃
```

---

## Visual Intuition

Imagine data spread like this:

```text id="s9j4jh"
************
***********
**********
*********
```

The longest spread direction contains the most variance.

PCA finds that direction.

That direction is given by an eigenvector.

The amount of variance is given by the corresponding eigenvalue.

Eigenvalues and eigenvectors are the mathematical foundation behind PCA.
