Output:

```text id="2gc12v"
[2. 3.]
```

NumPy solves the characteristic equation internally.

---

## ML Connection

You rarely solve:



by hand in ML.

But every time you use:

```python id="c5nsmg"
np.linalg.eig()
```

### What does the characteristic equation give?

```text id="sqwsl9"
Characteristic Equation
↓
Eigenvalues
```

After finding eigenvalues:

```text id="7xvjlwm"
Eigenvalues
↓
Eigenvectors
```

So the order is:

```text id="n3l47z"
Characteristic Equation
↓
Eigenvalues
↓
Eigenvectors
```

---

### Memory Chain

```text id="lnjtz0"
Av = λv
↓
(A - λI)v = 0
↓
det(A - λI) = 0
↓
Solve for λ
↓
Get Eigenvalues
```
