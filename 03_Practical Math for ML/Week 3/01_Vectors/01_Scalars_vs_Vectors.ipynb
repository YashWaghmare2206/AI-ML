# 01_Vectors

## 1. Scalars vs Vectors

---

## Formula

There is **no mathematical formula** for this topic because it is a basic concept.

---

## Explanation

### Scalar

A **scalar** is a quantity that has **only magnitude (size/value)**.

It does **not** have any direction.

Examples:

* Temperature = **30°C**
* Mass = **5 kg**
* Time = **10 s**
* Age = **20 years**

Only the value matters.

---

### Vector

A **vector** is a quantity that has:

* **Magnitude**
* **Direction**

Both are required to completely describe a vector.

Examples:

* Velocity = **20 m/s East**
* Force = **10 N upward**
* Displacement = **5 m North**
* Wind = **15 km/h South**

If the direction is missing, the vector information is incomplete.

---

## Example

Suppose a person walks:

* **5 meters** → Scalar (only distance/value)
* **5 meters East** → Vector (distance + direction)

Another example:

* Temperature = **25°C** → Scalar
* Wind = **20 km/h North** → Vector

---

## Python / NumPy

```python
import numpy as np

# Scalar
a = 10

# Vector
v = np.array([3, 4])

print("Scalar:", a)
print("Vector:", v)
```

**Output**

```text
Scalar: 10
Vector: [3 4]
```

---

## ML Connection

In Machine Learning:

* A **scalar** usually represents a single value.

  * Example:

    * Loss = **0.25**
    * Learning rate = **0.01**

* A **vector** represents multiple values together.

  * Example:

    * One data sample with features:

      ```
      [Height, Weight, Age]
      =
      [170, 65, 22]
      ```
    * Word embeddings in NLP.
    * Image pixels stored as vectors.

Almost every ML dataset is ultimately represented using **vectors** (and collections of vectors form matrices, which you'll learn next).
