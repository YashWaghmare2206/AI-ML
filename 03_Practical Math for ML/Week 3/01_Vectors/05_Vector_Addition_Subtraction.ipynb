## Explanation

To add or subtract vectors:

* Both vectors **must have the same number of dimensions**.
* Perform the operation on each corresponding component.

For example:

```text
First components  → Add/Subtract
Second components → Add/Subtract
Third components  → Add/Subtract
...
```

If the vectors have different dimensions, the operation is **not defined**.

For example:

```text
[1, 2] + [3, 4]      ✅ Valid

[1, 2] + [3, 4, 5]   ❌ Invalid
```

## Python / NumPy

```python
import numpy as np

A = np.array([2, 3])
B = np.array([4, 1])

print("Addition:", A + B)
print("Subtraction:", A - B)
```

**Output**

```text
Addition: [6 4]
Subtraction: [-2  2]
```

## ML Connection

Vector addition and subtraction are used in many ML tasks:

* Updating model parameters during **Gradient Descent**.
* Finding the **difference (error)** between predicted and actual values.
* Computing distances between data points.
* Working with embeddings (e.g., word vectors).

Example:

If

* Predicted vector = `[4, 7]`
* Actual vector = `[5, 9]`

Then the error vector is:

[
[5,9]-[4,7]=[1,2]
]

Many optimization algorithms use these vector operations internally.

## Important Note

Addition and subtraction are only possible when both vectors have the **same dimensions**.

Examples:

| Vector A  | Vector B  | Valid? |
| --------- | --------- | ------ |
| `[1,2]`   | `[3,4]`   | ✅      |
| `[1,2,3]` | `[4,5,6]` | ✅      |
| `[1,2]`   | `[3,4,5]` | ❌      |
