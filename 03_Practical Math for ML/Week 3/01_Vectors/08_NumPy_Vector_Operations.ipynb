## 8. NumPy Vector Operations

---

## Explanation

NumPy represents vectors as **1-dimensional arrays** (`ndarray`).

You can perform operations such as:

* Creating vectors
* Addition
* Subtraction
* Scalar multiplication
* Magnitude (Norm)
* Unit vector
* Distance between vectors

Most operations are **element-wise**, and NumPy performs them efficiently.

---

## Example

```python
import numpy as np

# Create vectors
A = np.array([2, 3])
B = np.array([5, 7])

# Addition
print("Addition:", A + B)

# Subtraction
print("Subtraction:", A - B)

# Scalar Multiplication
print("Scalar Multiplication:", 3 * A)

# Magnitude (Norm)
print("Magnitude of A:", np.linalg.norm(A))

# Unit Vector
print("Unit Vector of A:", A / np.linalg.norm(A))

# Distance Between Vectors
print("Distance:", np.linalg.norm(A - B))
```

**Output**

```text
Addition: [ 7 10]
Subtraction: [-3 -4]
Scalar Multiplication: [6 9]
Magnitude of A: 3.605551275463989
Unit Vector of A: [0.5547002  0.83205029]
Distance: 5.0
```

## ML Connection

NumPy is the foundation of almost every ML library:

* **scikit-learn**
* **TensorFlow**
* **PyTorch** (internally similar tensor operations)
* **SciPy**

Almost every ML algorithm performs vector operations using optimized numerical libraries like NumPy.

For example, a dataset with 10,000 samples and 20 features is stored as a NumPy array (or a similar tensor structure), and all computations are done using vectorized operations instead of Python loops.

---

## Important Note

One of NumPy's biggest advantages is **vectorization**.

Instead of writing loops:

```python
result = []

for i in range(len(A)):
    result.append(A[i] + B[i])
```

You simply write:

```python
result = A + B
```

This code is:

* Shorter
* Easier to read
* Much faster

Vectorization is one of the reasons NumPy is widely used in AI and Machine Learning.
