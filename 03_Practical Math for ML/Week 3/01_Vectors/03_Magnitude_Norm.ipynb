## Explanation

The **magnitude (or norm)** tells us the **length (size)** of a vector.

It ignores the direction and only measures **how long the vector is**.

Think of it as the distance from the **origin (0, 0)** to the point represented by the vector.

## Python / NumPy

```python
import numpy as np

v = np.array([3, 4])

magnitude = np.linalg.norm(v)

print(magnitude)
```

**Output**

```text
5.0
```

Another example:

```python
v = np.array([1, 2, 2])

print(np.linalg.norm(v))
```

**Output**

```text
3.0
```

---

## ML Connection

Magnitude is used throughout Machine Learning, for example:

* Calculating the **distance** between data points.
* **Feature normalization** (creating unit vectors).
* Measuring similarity between vectors.
* Algorithms like **KNN**, **K-Means**, and **Cosine Similarity** rely on vector norms.

---

### Important Note

You'll often hear the words:

* **Magnitude**
* **Length**
* **Norm**

For now, you can treat them as the **same thing**.

Technically, there are different types of norms (L1, L2, L∞), but unless specified, **"norm" usually refers to the L2 norm**, which is the magnitude you've just learned.
