## Explanation

The **distance between two vectors** tells us **how far apart they are**.

The smaller the distance:

* More similar the vectors.

The larger the distance:

* Less similar the vectors.

This is called the **Euclidean Distance**.

## Python / NumPy

```python
import numpy as np

A = np.array([2, 3])
B = np.array([5, 7])

distance = np.linalg.norm(A - B)

print(distance)
```

**Output**

```text
5.0
```

---

## ML Connection

Distance is one of the most important concepts in Machine Learning.

It is used in:

* **K-Nearest Neighbors (KNN)** → Find the closest data points.
* **K-Means Clustering** → Assign points to the nearest cluster center.
* **Anomaly Detection** → Detect points that are far from others.
* **Recommendation Systems** → Measure similarity between users/items.

Many ML algorithms compare vectors by computing their distances.

---

## Important Note



Distance is **never negative** because it is the square root of squared values.
