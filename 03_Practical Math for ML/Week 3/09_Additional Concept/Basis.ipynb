# Basis

Basis is one of the most important concepts in Linear Algebra.

It combines the two topics we already learned:

* Span
* Linear Independence

---

## Definition

A **basis** is a set of vectors that:

1. Span the space.
2. Are linearly independent.

Both conditions must be true.

## Connection with Span

Span asks:

> What space can these vectors generate?

Basis asks:

> What is the minimum independent set of vectors that generates that space?

---

## Connection with Linear Independence

Independence removes unnecessary vectors.

Span ensures the whole space is covered.

Basis requires both.

---

## Why Basis Matters in ML

Many ML techniques find a better basis for data.

Examples:

* PCA finds new basis directions that capture maximum variance.
* SVD decomposes data using special basis vectors.
* Embeddings represent data using basis-like directions in high-dimensional spaces.

---

## Key Takeaways

* Basis = Span + Linear Independence.
* A basis spans the entire space.
* A basis contains no redundant vectors.
* A space can have multiple different bases.


**One-line summary:**

> A basis is the smallest set of linearly independent vectors that can generate an entire space.
