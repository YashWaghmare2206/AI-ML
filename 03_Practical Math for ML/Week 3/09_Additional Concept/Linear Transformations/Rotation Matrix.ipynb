# Rotation Matrix

A rotation matrix rotates a vector by some angle while:

* Preserving length
* Preserving angles
* Preserving shape

This is why rotation matrices are orthogonal matrices.

## Effect on Basis Vectors

Standard basis:



A rotation matrix rotates both basis vectors.

Since basis vectors rotate, the whole coordinate system rotates.

---

## Properties

### Preserves Length

**`||Rv||=||v||`**

### Preserves Angles

Angle between vectors remains unchanged.

### Preserves Shape

Squares remain squares.

Circles remain circles.

Only orientation changes.

### One-line Summary
