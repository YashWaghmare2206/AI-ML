# Shear Matrix

A shear transformation **slants** or **skews** the space.

Unlike:

* Rotation → turns
* Scaling → stretches
* Reflection → flips

A shear transformation **pushes points sideways** while keeping one direction fixed.

---

## Intuition

Imagine a rectangle.

After a shear, it becomes a parallelogram.

The shape is tilted.

## What Happens to Length?

Lengths are generally **not preserved**.

---

## What Happens to Angles?

Angles are generally **not preserved**.

A right angle may no longer be 90°.

---

## What Happens to Area?

Interesting fact:

For a shear matrix

**`det(S)=1`**

Therefore:

> Area is preserved.

The shape changes, but the area remains the same.

---

## Is Shear Orthogonal?

No.

Because:

* Length changes.
* Angles change.

Therefore



So shear matrices are not orthogonal.

---

## Why Important in ML?

Shear itself is less common than rotation and scaling, but the concept is important because:

* It is a valid linear transformation.
* Helps understand matrix transformations.
* Appears in graphics and image processing.
* Useful for understanding determinants and geometric effects of matrices.

---

## Summary of All Transformations

| Transformation | Length Preserved | Angle Preserved | Orthogonal |
| -------------- | ---------------- | --------------- | ---------- |
| Rotation       | Yes              | Yes             | Yes        |
| Reflection     | Yes              | Yes             | Yes        |
| Scaling        | No               | Usually No      | No         |
| Shear          | No               | No              | No         |

---

## Key Takeaways

* Shear slants the space.



* Length changes.
* Angles change.
* Area stays the same.
* Not an orthogonal transformation.

### One-line Summary

> A shear matrix tilts vectors in one direction while keeping the other direction fixed, changing shape but preserving area.
