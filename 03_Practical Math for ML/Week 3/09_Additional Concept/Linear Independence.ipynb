# Linear Independence

This topic is directly connected to Span.

---

## Definition

A set of vectors is **linearly independent** if none of the vectors can be written as a linear combination of the others.

A set of vectors is **linearly dependent** if at least one vector can be written using the others.

---

## Intuition

Think in terms of **directions**.

* Independent vectors → each adds a new direction.
* Dependent vectors → some vector adds no new direction.

Since span depends on directions, independence tells us whether a vector actually increases the span.

## Connection with Span

Span answers:

> "What space can these vectors generate?"

Linear Independence answers:

> "Does each vector contribute a new direction to that space?"

So:

* Span = generated space.
* Independence = whether vectors are actually useful in generating that space.

---

* Independent vectors add new directions.
* Dependent vectors are redundant.
* If one vector can be created from others, the set is dependent.
