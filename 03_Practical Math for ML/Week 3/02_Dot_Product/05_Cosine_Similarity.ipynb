## Explanation

Cosine Similarity measures **how similar the directions of two vectors are**.

It **does not care about the lengths (magnitudes)** of the vectors.

Instead, it compares the **angle** between them.

The result is always between **-1 and 1**.

| Cosine Similarity | Meaning                                   |
| ----------------: | ----------------------------------------- |
|             **1** | Same direction (most similar)             |
|             **0** | Perpendicular (no directional similarity) |
|            **-1** | Opposite directions                       |

## Python / NumPy

```python
import numpy as np

A = np.array([1, 2])
B = np.array([2, 4])

cos_sim = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

print(cos_sim)
```

**Output**

```text
1.0
```

## ML Connection

Cosine Similarity is widely used when the **direction** of vectors matters more than their size.

Common applications:

* **Text Similarity**
* **Sentence Embeddings**
* **Word Embeddings (Word2Vec, GloVe)**
* **Recommendation Systems**
* **Semantic Search**
* **Large Language Models (LLMs)**

Example:

If two sentence embeddings have:

```text
Cosine Similarity = 0.98
```

their meanings are very similar.

If:

```text
Cosine Similarity = 0.10
```

their meanings are quite different.

## Important Note

### Euclidean Distance vs Cosine Similarity

| Euclidean Distance                     | Cosine Similarity                             |
| -------------------------------------- | --------------------------------------------- |
| Measures how **far apart** vectors are | Measures how **similar their directions** are |
| Depends on magnitude                   | Independent of magnitude                      |
| Smaller distance = more similar        | Larger cosine similarity = more similar       |

### Example

```text
A = [1, 2]
B = [2, 4]
```

* Euclidean Distance ≠ 0 (because the vectors have different lengths)
* Cosine Similarity = 1 (because they point in the same direction)

This is why Cosine Similarity is often preferred for comparing text embeddings.
