## Explanation

The **dot product** is calculated by:

1. Multiply the corresponding components of the two vectors.
2. Add all the products.

The result of a **dot product is always a scalar (single number)**, **not a vector**.

**Requirements:**

* Both vectors must have the **same number of dimensions**.

## Python / NumPy

```python
import numpy as np

A = np.array([2, 3])
B = np.array([4, 5])

dot = np.dot(A, B)

print(dot)
```

**Output**

```text
23
```

You can also use the `@` operator:

```python
dot = A @ B

print(dot)
```

**Output**

```text
23
```

## Important Note

A dot product:

* ✅ Takes **two vectors** as input.
* ✅ Returns **one scalar value**.
* ✅ Requires both vectors to have the **same dimensions**.

Example:

| Expression          | Result                           |
| ------------------- | -------------------------------- |
| `[2,3] · [4,5]`     | `23`                             |
| `[1,2,3] · [4,5,6]` | `32`                             |
| `[1,2] · [3,4,5]`   | ❌ Invalid (different dimensions) |
