# 05_Determinant

## 3. Singular Matrix

---

## Formula

A matrix is called **Singular** if:

[
\boxed{\det(A)=0}
]

---

## Explanation

A singular matrix is a matrix whose determinant is zero.

When the determinant becomes zero:

* ❌ Inverse does not exist.
* ❌ Unique solution does not exist.
* ❌ Matrix cannot be inverted.

This is the "bad" case in linear algebra.

---

## Example



---

## Why is it singular?

Notice:



One row contains no new information.

The matrix has redundant information.

This causes the determinant to become zero.

---

## Python / NumPy

```python id="gvg4zn"
import numpy as np

A = np.array([
    [1, 2],
    [2, 4]
])

print(np.linalg.det(A))
```

**Output**

```text id="axk58s"
0.0
```

Trying to find the inverse:

```python id="ycqlf5"
np.linalg.inv(A)
```

Output:

```text id="2r2ggt"
LinAlgError: Singular matrix
```

---

## ML Connection

Singular matrices often appear when features are duplicates or highly dependent.

Example dataset:

| Height(cm) | Height(m) |
| ---------- | --------- |
| 170        | 1.70      |
| 180        | 1.80      |
| 160        | 1.60      |

The second feature is just the first feature scaled.

So the columns are dependent.

This can create singular matrices in ML.
