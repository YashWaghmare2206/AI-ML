# Theory

**Vectorization** means performing operations on entire vectors/matrices at once instead of using explicit Python loops.

Instead of:

```python
result = []

for i in range(len(x)):
    result.append(x[i] * w[i])
```

we do:

```python
result = x * w
```

using NumPy.

---
