# 07. Underfitting

Underfitting is the opposite of overfitting.

---

# Theory

**Underfitting** occurs when the model is too simple to learn the actual pattern in the data.

As a result:

* Training performance is poor.
* Validation performance is also poor.

The model cannot even fit the training data properly.

---

# Intuition

Imagine a student learning multiplication.

Training:

```text
2 × 3 = 6
4 × 5 = 20
7 × 8 = 56
```

The student only learns:

> "The answer is always around 10."

He performs badly on training questions and badly on exam questions.

That is underfitting.

---

# Example



But you fit:

**`y = wx+b`**

a straight line.

The line cannot capture the curve.

---

# Visual Idea

Data follows a curve.

Model:

```text
Straight line
```

Result:

* Misses the pattern
* Large errors everywhere

---

# Mathematical View



No matter how gradient descent adjusts (w) and (b), a line cannot become a parabola.

The model's capacity is too low.

---

# How to Detect Underfitting

Training Error:

```text
High
```

Validation Error:

```text
High
```

Unlike overfitting, there is usually no large gap.

---

### Example

Training Accuracy:

```text
65%
```

Validation Accuracy:

```text
63%
```

Both poor.

Likely underfitting.

---

### Regression Example

Training MSE:

```text
80
```

Validation MSE:

```text
85
```

Both large.

Again, underfitting.

---

# Train vs Validation Error

Typical pattern:

```text
Training Error    High
Validation Error  High
```

Compare with overfitting:

```text
Training Error    Low
Validation Error  High
```

---

# Why It Happens

### 1. Model Too Simple



---

### 2. Too Few Features

Important information missing.

---

### 3. Excessive Regularization

Weights forced too close to zero.

Model loses learning ability.

---

### 4. Insufficient Training

Gradient descent stopped too early.

# Bias Perspective

Underfitting usually means:

```text
High Bias
Low Variance
```

We'll discuss this formally in the next topic.

---

# Important Insight

An underfit model is not memorizing.

It is failing to learn even the main pattern.

That's why both training and validation performance are poor.

---

# Comparison

| Property         | Underfitting | Overfitting |
| ---------------- | ------------ | ----------- |
| Training Error   | High         | Low         |
| Validation Error | High         | High        |
| Model Complexity | Too Low      | Too High    |
| Learns Pattern?  | Not enough   | Too much    |
| Generalization   | Poor         | Poor        |

---

# What You Should Remember

1. Underfitting = model too simple.
2. Training and validation performance are both poor.
3. High training error is the key signal.
4. Common causes:

   * Simple model
   * Missing features
   * Too much regularization
   * Too little training
5. Increase complexity or improve features to reduce underfitting.
6. Underfitting corresponds to **high bias**.
