## Python Example

Original:

```python
import numpy as np

X = np.array([[1],
              [2],
              [3],
              [4]])
```

Create polynomial feature manually:

```python
X_poly = np.hstack([X, X**2])
```

Result:

```text
[
 [1  1]
 [2  4]
 [3  9]
 [4 16]
]
```

---

## Scikit-Learn

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)
```

Output:

```text
[
 [1,1,1]
 [1,2,4]
 [1,3,9]
 [1,4,16]
]
```

The first column is the bias term.

## Important Observation

Degree increases feature count rapidly.

For one feature:

Degree 2:

```text
x
x²
```

Degree 3:

```text
x
x²
x³
```

Not bad.

---

For 10 features:

Degree 2 generates:

```text
x₁²
x₂²
...
x₁x₂
x₁x₃
...
```

Feature count grows very quickly.

This can increase overfitting.

---

## Choosing Degree

Too low:

```text
Underfit
```

Too high:

```text
Overfit
```

The goal is finding the right complexity.

### **So what changed when x raised to 2 is added to the eqn?**

Before:

```text
Feature: x
Model: y = wx + b
```



Suppose the true relationship is:

```text
x : 1  2  3  4
y : 1  4  9 16
```

which follows (y=x^2).

A straight line cannot fit those points well.

---

After adding (x^2):

```text
Feature 1: x
Feature 2: x²

Model:
y = w₁x + w₂x² + b
```

Now the model can create curves because the prediction contains an (x^2) term.


For example:

```text
x    x²
1     1
2     4
3     9
4    16
```

The algorithm is still just finding the best (w_1), (w_2), and (b).

What changed is that you've given it a new feature that lets the prediction bend.

A good way to think about it:

```text
y = wx + b
        ↓
Only straight lines possible

y = w₁x + w₂x² + b
        ↓
Parabolas become possible

y = w₁x + w₂x² + w₃x³ + b
        ↓
Even more complex curves become possible
```

So adding the (x^2) feature increases the set of shapes the model can represent. That's why polynomial features help when the relationship between (x) and (y) is curved rather than linear.
