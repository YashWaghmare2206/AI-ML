# What Problem Does It Solve?

Naive Bayes is a **supervised machine learning algorithm** mainly used for **classification**.

It predicts **which class an input belongs to** by calculating the probability of each class and selecting the one with the **highest probability**.

In simple terms:

> **Given some features (evidence), what is the most probable class?**

For example:

| Features                               | Predict                      |
| -------------------------------------- | ---------------------------- |
| Email contains "Win", "Prize", "Money" | Spam                         |
| Review: "Amazing quality, loved it"    | Positive                     |
| Symptoms: Fever, Cough                 | Flu                          |
| News article words                     | Sports / Politics / Business |

Unlike Logistic Regression, which learns a decision boundary, Naive Bayes works using **probability**.

---

## Example

Suppose we have two classes:

* Spam
* Not Spam

Email:

> "You won free money"

Naive Bayes computes

* Probability(email is Spam)
* Probability(email is Not Spam)

Whichever probability is larger becomes the prediction.

---

# Why Was It Developed?

Earlier algorithms like Linear Regression were designed for predicting **continuous values**.

Later, Logistic Regression solved binary classification by learning weights through optimization.

However, there were problems:

* Optimization can be slow on huge datasets.
* High-dimensional data (thousands of features) becomes computationally expensive.
* Text classification contains tens of thousands of words.

Researchers wanted an algorithm that:

* Is mathematically simple.
* Trains extremely fast.
* Works well with many features.
* Needs little data preprocessing.
* Gives probabilistic predictions.

Naive Bayes was developed based on **Bayes' Theorem** from probability theory.

Instead of learning parameters through gradient descent, it directly estimates probabilities from the training data.

### 2. Fast Prediction

---

Unlike many algorithms that require large amounts of data.

### 4. Handles High-Dimensional Data

# Limitations

### 1. Strong Independence Assumption

This is the biggest limitation.

Naive Bayes assumes

```
Every feature is independent
```

which is rarely true in real-world data.

Example:

```
Word = "New"

Word = "York"
```

These words are clearly related.

Naive Bayes treats them as independent.

---

### 2. Zero Probability Problem

If a feature never appeared in training,

its probability becomes

```
0
```

which can make the entire class probability zero.

(This is later solved using **Laplace Smoothing**.)

---

### 3. Lower Accuracy on Complex Relationships

If features interact heavily,

algorithms like Random Forest, SVM, or Neural Networks usually perform better.

---

### 4. Decision Boundary is Simple

Cannot model highly complicated boundaries.

---

### 5. Probability Estimates May Be Poor

Although classification accuracy can be good,

the predicted probabilities are often not perfectly calibrated.

# Comparison with Previous Algorithms

| Algorithm           | Type           | Main Idea                             | Strength                           | Weakness                     |
| ------------------- | -------------- | ------------------------------------- | ---------------------------------- | ---------------------------- |
| Linear Regression   | Regression     | Predict continuous values             | Simple                             | Cannot classify              |
| Logistic Regression | Classification | Learns weights using gradient descent | Good decision boundary             | Training can be slower       |
| **Naive Bayes**     | Classification | Uses Bayes' Theorem and probabilities | Extremely fast, excellent for text | Assumes independent features |

### Logistic Regression vs Naive Bayes

| Logistic Regression                   | Naive Bayes                           |
| ------------------------------------- | ------------------------------------- |
| Learns parameters using optimization  | Computes probabilities directly       |
| No independence assumption            | Assumes feature independence          |
| Often better with correlated features | Excellent with text data              |
| Usually needs more training time      | Very fast training                    |
| Produces linear decision boundary     | Produces probabilistic classification |

---

# Summary

* Naive Bayes is a **probabilistic classification algorithm**.
* It is based on **Bayes' Theorem**.
* It predicts the class with the **highest posterior probability**.
* It is especially effective for **text classification**, **spam detection**, **sentiment analysis**, and **document classification**.
* Its biggest assumption is that **all features are conditionally independent given the class**, which is why it is called **"Naive"**.
* Despite this strong assumption, it often performs remarkably well in practice, particularly on high-dimensional datasets.
