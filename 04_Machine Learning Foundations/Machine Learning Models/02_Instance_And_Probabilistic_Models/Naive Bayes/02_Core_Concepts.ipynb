# Working Intuition

Before learning the mathematics, let's build the intuition behind how Naive Bayes makes predictions.

Suppose you receive an email:

```text
Congratulations!
You won a free lottery prize.
```

You want to classify it as:

* Spam
* Not Spam

Instead of learning a decision boundary (like Logistic Regression), Naive Bayes asks:

> **"Which class is more probable given these words?"**

It computes:

* Probability that this email is Spam.
* Probability that this email is Not Spam.

Whichever probability is larger becomes the prediction.

---

## General Idea

Assume we have features:

```text
X = (x₁, x₂, x₃, ..., xₙ)
```

We want to predict one class from:

```text
Y ∈ {C₁, C₂, C₃, ...}
```

Naive Bayes calculates:

```text
P(Class | Features)
```

for every class.

Example:

```text
P(Spam | Email)

P(Not Spam | Email)
```

Then chooses the larger one.

---

## Why is it Called "Naive"?

The algorithm assumes:

> Every feature is independent of every other feature **after knowing the class**.

Example:

Email:

```text
Free
Money
Winner
Prize
```

Naive Bayes assumes

```text
Free   ⟂   Money

Money  ⟂   Prize

Winner ⟂   Free
```

although in reality these words are often related.

This unrealistic assumption makes the algorithm **"Naive."**

---

## Intuition with a Medical Example

Suppose we want to predict:

```text
Disease
```

Features:

```text
Fever

Cough

Headache
```

Patient:

```text
Fever = Yes

Cough = Yes

Headache = No
```

Naive Bayes calculates:

```text
Probability(Flu)

Probability(COVID)

Probability(Cold)
```

and predicts the disease with the highest probability.

---

## What Happens During Training?

Naive Bayes **does not optimize weights**.

Instead, it simply counts things.

Example:

Training Data

| Email        | Class    |
| ------------ | -------- |
| Free money   | Spam     |
| Hello friend | Not Spam |
| Win lottery  | Spam     |

During training it stores information like:

```text
How many spam emails?

How many non-spam emails?

How often does "free" appear in spam?

How often does "friend" appear in non-spam?
```

That's essentially the entire training process.

---

## What Happens During Prediction?

Given a new sample:

```text
Free

Winner

Prize
```

Naive Bayes:

1. Computes probability of Spam.
2. Computes probability of Not Spam.
3. Compares them.
4. Returns the larger one.

---

# Important Terminology

Understanding these terms is essential before studying Bayes' Theorem.

---

## 1. Feature (Predictor)

An input variable.

Example:

| Feature       |
| ------------- |
| Age           |
| Salary        |
| Fever         |
| Word = "Free" |

Notation:

```text
X
```

or

```text
x₁, x₂, ...
```

---

## 2. Target (Class)

The output we want to predict.

Examples:

```text
Spam

Not Spam
```

```text
Positive

Negative
```

Notation:

```text
Y
```

or

```text
C
```

---

## 3. Prior Probability

Probability of a class **before looking at any features**.

Example:

Suppose:

100 emails

80 spam

20 not spam

Then

```text
P(Spam) = 80/100 = 0.8

P(Not Spam) = 20/100 = 0.2
```

This is called the **Prior Probability**.

---

## 4. Likelihood

Probability of observing a feature given a class.

Example:

Among Spam emails:

```text
Word "Free" appears in 60 of 80 emails
```

Then

```text
P(Free | Spam)
=
60/80
=
0.75
```

This is the **Likelihood**.

---

## 5. Evidence

Overall probability of observing the feature.

Example:

```text
P(Free)
```

It is used for normalization in Bayes' Theorem.

For classification, since this value is the same for every class, we usually ignore it when comparing classes.

---

## 6. Posterior Probability

This is what Naive Bayes wants to compute.

It answers:

> **After observing the features, what is the probability of each class?**

Example:

```text
P(Spam | Email)
```

The class with the highest posterior probability is chosen.

---

## 7. Conditional Probability

Probability of one event occurring **given** another event has already occurred.

Example:

```text
P(Fever | Flu)
```

means:

> Probability that a patient has a fever **if** they already have the flu.

This concept is the foundation of Bayes' Theorem.

---

# Assumptions

Naive Bayes relies on several assumptions.

---

## 1. Conditional Independence (Most Important)

After knowing the class,

all features are independent.

Mathematically:

```text
P(x₁,x₂,...,xₙ | C)
=
P(x₁|C)
×
P(x₂|C)
×
...
×
P(xₙ|C)
```

This assumption makes the computation much simpler.

---

## 2. Training and Test Data Follow Similar Distributions

The data used during prediction should come from a distribution similar to the training data.

Otherwise, probabilities estimated during training may no longer be accurate.

---

## 3. Features are Relevant

Irrelevant or noisy features can reduce performance.

Although Naive Bayes is relatively robust, it still benefits from meaningful features.

---

# Types / Variants

Naive Bayes has several variants depending on the type of data.

---

## 1. Gaussian Naive Bayes

Used for **continuous numerical features**.

Example:

```text
Height

Weight

Age

Temperature
```

Assumes each feature follows a **Normal (Gaussian) Distribution** within each class.

Common datasets:

* Iris
* Breast Cancer
* Wine

Scikit-learn:

```python
from sklearn.naive_bayes import GaussianNB
```

---

## 2. Multinomial Naive Bayes

Used for **count data**.

Examples:

```text
Word frequencies

Document word counts

Number of purchases
```

Common applications:

* Spam Detection
* Text Classification
* Sentiment Analysis

Scikit-learn:

```python
from sklearn.naive_bayes import MultinomialNB
```

---

## 3. Bernoulli Naive Bayes

Used for **binary features**.

Features contain only:

```text
0

1
```

Examples:

```text
Word exists?

Yes / No

Clicked Ad?

Yes / No
```

Scikit-learn:

```python
from sklearn.naive_bayes import BernoulliNB
```

---

## 4. Complement Naive Bayes

Designed for **imbalanced text datasets**.

Often performs better than Multinomial NB when classes are highly imbalanced.

Scikit-learn:

```python
from sklearn.naive_bayes import ComplementNB
```

---

## 5. Categorical Naive Bayes

Used for **categorical features**.

Example:

```text
Color

Red

Blue

Green
```

Scikit-learn:

```python
from sklearn.naive_bayes import CategoricalNB
```

---

## Choosing the Right Variant

| Data Type          | Variant       |
| ------------------ | ------------- |
| Continuous values  | GaussianNB    |
| Word counts        | MultinomialNB |
| Binary features    | BernoulliNB   |
| Categorical values | CategoricalNB |
| Imbalanced text    | ComplementNB  |

# Decision Boundary / Model Behavior

Naive Bayes creates decision boundaries based on **posterior probabilities**, not by directly optimizing a separating line.

### Binary Classification

```text
                Class A
XXXXXXXXXXXXXXXX|
XXXXXXXXXXXXXXXX|
XXXXXXXXXXXXXXXX|
----------------|----------------
                |
                |
                |
                |
                Class B
```

The boundary is where:

```text
P(Class A | X)
=
P(Class B | X)
```

---

### Behavior

* Predicts the class with the highest probability.
* Very fast even with thousands of features.
* Works exceptionally well for text classification.
* Performance decreases when features are highly dependent.

---

# Advantages & Disadvantages

## Advantages

* Extremely fast training.
* Fast prediction.
* Simple implementation.
* Works well on high-dimensional data.
* Performs well on text classification.
* Requires relatively little training data.
* Outputs class probabilities.
* Less prone to overfitting than many complex models.

---

## Disadvantages

* Assumes feature independence.
* Zero-frequency problem (solved later using Laplace Smoothing).
* Poor probability calibration.
* Struggles with complex feature interactions.
* Usually outperformed by advanced models on complex datasets.

---

# When To Use / Avoid

## Use Naive Bayes When

✅ Text Classification

✅ Spam Detection

✅ Sentiment Analysis

✅ Language Detection

✅ Small datasets

✅ High-dimensional datasets

✅ Need very fast training

---

## Avoid Naive Bayes When

❌ Features are strongly correlated.

❌ Decision boundary is highly complex.

❌ Maximum predictive accuracy is required.

❌ Large nonlinear relationships exist among features.

❌ Feature independence assumption is clearly violated.
