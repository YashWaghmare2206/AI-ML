# Step-by-Step Workflow

Unlike Logistic Regression, Naive Bayes **does not learn weights using gradient descent**. Instead, it estimates probabilities directly from the training data.

The overall workflow consists of **two phases**:

1. **Training**
2. **Prediction**

---

## Overall Workflow

```text
                Training Dataset
                       │
                       ▼
            Calculate Prior Probabilities
                       │
                       ▼
         Calculate Likelihood Probabilities
                       │
                       ▼
          Store All Probability Values
                       │
──────────────────────────────────────────────
                 New Sample Arrives
                       │
                       ▼
     Compute Posterior Probability for Every Class
                       │
                       ▼
        Choose Class with Highest Probability
                       │
                       ▼
                 Final Prediction
```

---

# Training Process

The goal during training is **not** to learn coefficients.

Instead, Naive Bayes learns **probabilities**.

Suppose our training dataset is:

| Email         | Class    |
| ------------- | -------- |
| Free Prize    | Spam     |
| Free Money    | Spam     |
| Hello Friend  | Not Spam |
| Meeting Today | Not Spam |
| Win Lottery   | Spam     |

---

## Step 1: Find All Classes

The algorithm first identifies every unique class.

```text
Spam

Not Spam
```

---

## Step 2: Calculate Prior Probability

Prior Probability answers:

> **How common is each class before seeing any features?**

Formula:

```text
P(Class)
=
(Number of samples in that class)
/
(Total number of samples)
```

Example:

Total emails = 5

Spam = 3

Not Spam = 2

Therefore,

```text
P(Spam)
=
3/5
=
0.6

P(Not Spam)
=
2/5
=
0.4
```

These values are stored.

---

## Step 3: Calculate Likelihood

Now compute how often every feature appears inside each class.

Example:

### Spam emails

```
Free Prize
Free Money
Win Lottery
```

Word counts:

| Word    | Count |
| ------- | ----: |
| Free    |     2 |
| Prize   |     1 |
| Money   |     1 |
| Win     |     1 |
| Lottery |     1 |

---

### Not Spam emails

```
Hello Friend
Meeting Today
```

| Word    | Count |
| ------- | ----: |
| Hello   |     1 |
| Friend  |     1 |
| Meeting |     1 |
| Today   |     1 |

These probabilities are stored.

---

## Step 4: Save Everything

After training, Naive Bayes stores:

* Prior probabilities
* Feature probabilities (Likelihoods)

Training is complete.

No optimization.

No iterations.

No gradient descent.

---

# Prediction Process

Suppose a new email arrives:

```text
Free Lottery
```

We want to decide:

```text
Spam ?

Not Spam ?
```

---

## Step 1: Compute Posterior for Spam

The algorithm combines:

* Prior probability of Spam
* Probability of "Free"
* Probability of "Lottery"

Conceptually:

```text
Posterior(Spam)

=

Prior

×

Likelihood(Free)

×

Likelihood(Lottery)
```

---

## Step 2: Compute Posterior for Not Spam

Similarly,

```text
Posterior(Not Spam)

=

Prior

×

Likelihood(Free)

×

Likelihood(Lottery)
```

---

## Step 3: Compare Both

Example:

```text
Spam
=
0.032

Not Spam
=
0.001
```

Since

```text
0.032 > 0.001
```

Prediction:

```text
Spam
```

---

# Small Numerical Example

Let's work through a complete example.

---

## Training Dataset

| Weather | Play Cricket |
| ------- | ------------ |
| Sunny   | Yes          |
| Sunny   | Yes          |
| Rainy   | No           |
| Sunny   | Yes          |
| Rainy   | No           |

Suppose we want to predict:

```text
Weather = Sunny
```

---

## Step 1: Prior

Total samples = 5

Yes = 3

No = 2

Therefore,

```text
P(Yes)

=

3/5

=

0.6

P(No)

=

2/5

=

0.4
```

---

## Step 2: Likelihood

Among Yes:

Sunny appears 3 times.

```text
P(Sunny | Yes)

=

3/3

=

1
```

Among No:

Sunny appears 0 times.

```text
P(Sunny | No)

=

0/2

=

0
```

---

## Step 3: Posterior Scores

For **Yes**:

```text
Score

=

0.6

×

1

=

0.6
```

For **No**:

```text
Score

=

0.4

×

0

=

0
```

Prediction:

```text
Play Cricket = Yes
```

---

## Problem Observed

Notice:

```text
P(Sunny | No)

=

0
```

Because one probability became zero,

the entire posterior became zero.

This is called the **Zero Frequency Problem**.

It is solved using **Laplace Smoothing**, which we'll cover later.

# How Naive Bayes Thinks

Instead of asking:

> "Which line separates the classes?"

(as Logistic Regression or SVM does)

Naive Bayes asks:

> **"Given these observed features, which class is most probable?"**

This difference is very important.

* **Logistic Regression** → learns a decision boundary.
* **Decision Trees** → learn splitting rules.
* **KNN** → looks at neighboring samples.
* **Naive Bayes** → computes probabilities.

---

# Algorithm Summary

## Training

1. Read training dataset.
2. Find all classes.
3. Compute prior probabilities.
4. Compute likelihood probabilities for every feature.
5. Store these probabilities.

---

## Prediction

1. Receive a new sample.
2. For every class:

   * Start with the prior probability.
   * Multiply by the likelihood of each feature (under the independence assumption).
3. Compute a score for every class.
4. Predict the class with the highest score.

---

## Time Complexity

Let:

* **N** = Number of training samples
* **d** = Number of features
* **C** = Number of classes

### Training

Computing probabilities requires scanning all samples and features:

```text
O(N × d)
```

### Prediction (per sample)

For each class, compute the contribution of all features:

```text
O(C × d)
```

Naive Bayes is therefore one of the **fastest classification algorithms**, especially for high-dimensional datasets such as text.
