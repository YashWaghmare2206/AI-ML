# K-Nearest Neighbors (KNN)

---

# What Problem Does It Solve?

Many machine learning problems require predicting the output for **new, unseen data**.

Examples:

* Is this email spam?
* Will this customer buy a product?
* Which species does this flower belong to?
* What house price is expected for this location?

To answer these questions, we need a model that can make predictions using previously observed data.

KNN solves this by following a very simple idea:

> **Objects that are similar are likely to have similar outputs.**

Instead of trying to learn a mathematical equation during training, KNN simply stores the training data. When a new sample arrives, it looks for the most similar training samples (its nearest neighbors) and predicts based on them.

---

## Classification Problem

Suppose we have data like:

| Height | Weight | Sport     |
| ------ | ------ | --------- |
| 170    | 65     | Football  |
| 172    | 68     | Football  |
| 160    | 52     | Badminton |
| 158    | 50     | Badminton |

A new person:

```
Height = 171
Weight = 66
```

KNN asks:

> Which existing people are most similar?

If most nearby people play Football,

Prediction:

```
Football
```

---

## Regression Problem

Suppose:

| House Size | Price |
| ---------- | ----- |
| 900        | 45L   |
| 1000       | 50L   |
| 1100       | 55L   |

New house:

```
1050 sq.ft
```

Nearby houses:

```
1000 → 50L
1100 → 55L
```

Average

```
(50 + 55)/2 = 52.5L
```

Prediction:

```
52.5L
```

# Why Was It Developed?

Earlier algorithms (like Linear Regression or Logistic Regression) assume there is some mathematical relationship between input and output.

Examples:

Linear Regression assumes

[
y = w_1x_1+w_2x_2+b
]

Logistic Regression assumes

[
P(y)=\sigma(wx+b)
]

These assumptions work well when the relationship is approximately linear.

However, many real-world datasets are much more complex.

Example:

```
Dog
      ●

             Cat
                 ●

Dog ●

            Cat ●
```

There may not exist one straight line that separates the classes.

Researchers wanted an algorithm that:

* makes almost no assumptions
* works on irregular datasets
* is simple to understand
* can handle both classification and regression

This led to **K-Nearest Neighbors**.

Instead of learning a formula,

it simply says:

> "Look at nearby examples and copy their behavior."

# Advantages

## 1. Very Simple

One of the easiest ML algorithms to understand and implement.

---

## 2. No Training Phase

Unlike many algorithms, KNN does not estimate model parameters.

Training mainly consists of storing the dataset.

This makes training extremely fast.

---

## 3. Works for Classification and Regression

Classification:

Majority vote

Regression:

Average of neighbors

---

## 4. No Assumption About Data

Unlike Linear Regression,

KNN does not assume:

* linearity
* normal distribution
* independence

It adapts to the data.

---

## 5. Naturally Handles Multi-Class Problems

Example:

Classes:

```
Dog
Cat
Horse
Cow
```

No modification is required.

---

## 6. Learns Complex Decision Boundaries

Decision boundaries can become highly irregular because predictions depend on local neighborhoods rather than a single global equation.

---

## 7. Easy to Update

Adding new training data is straightforward.

No need to retrain a mathematical model—just include the new samples in the stored dataset.

---

# Limitations

## 1. Slow Prediction

Training:

Very fast.

Prediction:

Slow.

Why?

Every prediction requires computing the distance from the new sample to (potentially) every training sample.

---

## 2. High Memory Usage

The entire training dataset must be stored.

Large datasets consume more memory.

---

## 3. Sensitive to Feature Scaling

Example:

Features:

```
Age = 20

Salary = 900000
```

Salary has much larger numerical values and can dominate the distance calculation.

Therefore, scaling (e.g., StandardScaler or MinMaxScaler) is usually essential before using KNN.

---

## 4. Sensitive to Irrelevant Features

If you include features that have little or no predictive value, the distance calculation becomes less meaningful.

Feature selection or dimensionality reduction can improve performance.

---

## 5. Suffers from the Curse of Dimensionality

As the number of features increases:

* distances between points become more similar
* identifying truly "nearest" neighbors becomes harder
* prediction quality may decrease

---

## 6. Choosing K is Important

Small K:

```
High variance
Sensitive to noise
```

Large K:

```
High bias
May oversmooth decision boundaries
```

Selecting an appropriate value of **K** is crucial.

---

## 7. Sensitive to Noise and Outliers

If noisy or mislabeled samples are close to the query point, they can strongly influence the prediction, especially when **K** is small.

### Key Takeaways

* KNN is an **instance-based (lazy learning)** algorithm: it stores training examples instead of learning model parameters.
* It can solve **both classification and regression** problems.
* Predictions are made by examining the **K nearest neighbors** according to a distance metric.
* It is simple and effective for many small- to medium-sized datasets.
* Proper **feature scaling** and **choice of K** are critical for good performance.
* Prediction becomes slower as the dataset grows because distances must be computed to many training samples.
