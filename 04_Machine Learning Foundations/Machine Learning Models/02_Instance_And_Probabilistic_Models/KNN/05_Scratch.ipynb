# **KNN Classifier Scratch Implementation**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class KNNClassifier:

    def __init__(self , k = 3):
        self.k = k
        self.X_train = None
        self.y_train = None


    # Here it stores the dataset
    def fit(self , X_train , y_train):
        self.X_train = X_train
        self.y_train = y_train

    # Calculates the distance between points
    def cal_eucledian(self , x1 , x2):
        return np.sqrt(np.sum(x1 - x2) ** 2)

    # It predicts the category
    def predict_single(self , x1):

        distances = []   # stores distances of each point from the input value

        for x2 in self.X_train:
            dist = self.cal_eucledian(x1 , x2)
            distances.append(dist)


        k_indices = np.argsort(distances)[:self.k] # It " np.argsort() " sorts values in ascending value and returns the list with index of value in ascending value

        k_labels = [self.y_train[i] for i in k_indices]   # We save the category of the samples close to our input

        unique_val , count = np.unique(k_labels , return_counts=True)  # it counts each category

        return unique_val[np.argmax(count)] # return the maximum count value of the category

    def predict(self, X):

        predictions = []

        for x in X:
            predictions.append(self.predict_single(x))

        return np.array(predictions)
```

```python
X_train = np.array([
    [170,65],
    [172,68],
    [160,52],
    [158,50],
    [175,70]
])

y_train = np.array([
    "Football",
    "Football",
    "Badminton",
    "Badminton",
    "Football"
])

X_test = np.array([
    [171,66],
    [159,51]
])


model = KNNClassifier(k=3)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print(predictions)
```
