# 2. Training Process

During training, SVM performs the following steps:

1. Receive training dataset.
2. Scale the features (recommended).
3. Choose a kernel.
4. Choose hyperparameters (`C`, `gamma`, etc.).
5. Find the hyperplane that maximizes the margin.
6. Identify support vectors.
7. Save the learned model.

```text
Training Data
      │
      ▼
Feature Scaling
      │
      ▼
Choose Kernel
      │
      ▼
Optimization
      │
      ▼
Maximum Margin Hyperplane
      │
      ▼
Support Vectors Stored
```

---

# 3. Prediction Process

Suppose a new sample arrives.

```text
?

```

The prediction steps are:

1. Scale the new sample using the same scaler.
2. Compute its position relative to the learned hyperplane.
3. Determine on which side of the hyperplane it lies.
4. Assign the corresponding class.

```text
New Sample
      │
      ▼
Scale Features
      │
      ▼
Decision Function
      │
      ▼
Left Side ? → Class A

Right Side ? → Class B
```

## Key Takeaways

* SVM searches for the **best separating hyperplane**, not just any hyperplane.
* It maximizes the **margin** between classes.
* **Support vectors** are the only training points that directly influence the decision boundary.
* For non-linear data, SVM uses **kernel functions** to create flexible decision boundaries.
* Feature scaling is highly recommended before training SVM.


## **The different hyperplane is calculated dby t hyperparameter values right and how multiple and best is selected.?**

Good question. The answer is **not exactly**.

The different hyperplanes are **not generated by trying different hyperparameter values**. Instead, they arise from the optimization problem itself.

Here's how it works.

---

## Step 1: Imagine many possible separating hyperplanes

For a linearly separable dataset:

```text
● ● ●        ▲ ▲ ▲
```

There are infinitely many lines that can separate them.

```text
Line 1

● ● | ▲ ▲

----------------

Line 2

● ●   | ▲ ▲

----------------

Line 3

● ●     | ▲ ▲
```

These are **mathematically possible hyperplanes**, not hyperplanes created by changing `C` or another hyperparameter.

---

## Step 2: How does SVM choose the best one?

For every possible hyperplane, SVM computes its **margin**.

```text
Margin = Distance from hyperplane to nearest point
```

Example:

```text
Line A

● ● | ▲ ▲

Margin = 1

----------------

Line B

● ●   | ▲ ▲

Margin = 2

----------------

Line C

● ●     | ▲ ▲

Margin = 0.5
```

SVM chooses the one with the **largest margin**.

---

## Step 3: Does SVM actually try every line?

**No.**

There are infinitely many possible hyperplanes, so checking each one is impossible.

Instead, SVM formulates an **optimization problem**:

> Find the hyperplane that maximizes the margin.

Mathematically, it solves:

[
\min \frac{1}{2}|w|^2
]

subject to

[
y_i(w^Tx_i+b)\ge1
]

The optimization algorithm directly computes the optimal (w) and (b). It does **not** enumerate all possible hyperplanes.

---

## Then what does the hyperparameter `C` do?

`C` changes the optimization objective.

### Small `C`

Allows some mistakes.

```text
Large Margin

● ●     |      ▲ ▲

One point may be misclassified.
```

---

### Large `C`

Tries hard to classify every point correctly.

```text
Small Margin

● ● | ▲ ▲
```

So changing `C` changes **which hyperplane becomes optimal**, but SVM still solves an optimization problem—it doesn't generate a list of hyperplanes and pick one.

---

## Complete Flow

```text
Training Data
      │
      ▼
Choose Hyperparameters (C, kernel, gamma)
      │
      ▼
Form Optimization Problem
      │
      ▼
Optimization Solver computes w and b
      │
      ▼
Optimal Hyperplane Found
      │
      ▼
Support Vectors Identified
```

So the sequence is:

* Hyperparameters (like `C`) define the optimization problem.
* The optimization solver computes **one optimal hyperplane**.
* It does **not** explicitly create or test every possible hyperplane.
