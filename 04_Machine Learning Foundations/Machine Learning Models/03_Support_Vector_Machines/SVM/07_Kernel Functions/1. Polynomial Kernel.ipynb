# 1. Why Isn't the Homogeneous Polynomial Kernel Enough?

We previously derived the homogeneous polynomial kernel:

$$\boldsymbol{K(x_i,x_j)=(x_i^Tx_j)^d}$$

where:

* $x_i$ = first training sample
* $x_j$ = second training sample
* $d$ = polynomial degree

This kernel works correctly, but it has a limitation.

---

## Problem

Suppose

$$\boldsymbol{x_i=(1,0)}$$

and

$$\boldsymbol{x_j=(0,1)}$$

Their dot product is

$$\boldsymbol{x_i^Tx_j=1\times0+0\times1=0}$$

Now suppose the degree is 3.

The kernel becomes

$$\boldsymbol{K(x_i,x_j)=0^3=0}$$

So the kernel says the similarity between these two samples is **zero**.

For some datasets, this is too restrictive because even though the original dot product is zero, we may still want the transformed feature space to capture useful relationships.

---

# 2. Adding a Constant

To solve this, we modify the kernel.

Instead of

$$\boldsymbol{K(x_i,x_j)=(x_i^Tx_j)^d}$$

we use

$$\boldsymbol{K(x_i,x_j)=(x_i^Tx_j+c)^d}$$

where $c$ is a constant.

---

## Example

Suppose

* $d=3$
* $c=1$

and

$$\boldsymbol{x_i^Tx_j=0}$$

Then

$$\boldsymbol{K(x_i,x_j)=(0+1)^3=1}$$

Instead of returning **0**, the kernel now returns **1**.

So adding the constant prevents the similarity from becoming zero simply because the original dot product is zero.

---

# 3. Why Does Adding $c$ Help?

This is the most important part.

Consider

$$\boldsymbol{(x_i^Tx_j+1)^2}$$

Expand it using

$$\boldsymbol{(a+b)^2=a^2+2ab+b^2}$$

We get

$$\boldsymbol{(x_i^Tx_j)^2+2(x_i^Tx_j)+1}$$

Notice what happened.

Without the constant:

$$\boldsymbol{(x_i^Tx_j)^2}$$

Only the **degree-2** term exists.

With the constant:

$$\boldsymbol{(x_i^Tx_j)^2+2(x_i^Tx_j)+1}$$

Now we have:

* Degree 2 term
* Degree 1 term
* Constant term (degree 0)

So the kernel is no longer using only degree-2 information.

It also includes lower-degree information, making the feature space richer and more flexible.

---

# 4. The Practical Polynomial Kernel

Most machine learning libraries (including scikit-learn) use

$$\boldsymbol{K(x_i,x_j)=(\gamma x_i^Tx_j+coef0)^d}$$

This introduces three parameters.

---

## Degree ($d$)

Controls the highest polynomial degree.

Examples:

Degree 2:

$$\boldsymbol{(\cdot)^2}$$

Degree 3:

$$\boldsymbol{(\cdot)^3}$$

A larger degree allows the SVM to learn more complex decision boundaries.

---

## Gamma ($\gamma$)

Gamma scales the dot product **before** raising it to the power.

Suppose

$$\boldsymbol{x_i^Tx_j=17}$$

If

$$\boldsymbol{\gamma=1}$$

then

$$\boldsymbol{\gamma x_i^Tx_j=17}$$

If

$$\boldsymbol{\gamma=0.5}$$

then

$$\boldsymbol{\gamma x_i^Tx_j=8.5}$$

If

$$\boldsymbol{\gamma=2}$$

then

$$\boldsymbol{\gamma x_i^Tx_j=34}$$

Gamma controls how strongly the original similarity influences the kernel value.

---

## coef0

`coef0` is the constant that we discussed earlier.

If

$$\boldsymbol{coef0=0}$$

the kernel becomes

$$\boldsymbol{(\gamma x_i^Tx_j)^d}$$

which is essentially the homogeneous polynomial kernel (up to the scaling by $\gamma$).

If

$$\boldsymbol{coef0>0}$$

the expansion includes lower-degree terms, making the kernel more expressive.

---

# 5. Numerical Example

Suppose

$$\boldsymbol{x_i=(2,3)}$$

and

$$\boldsymbol{x_j=(1,5)}$$

### Compute the dot product

$$\boldsymbol{x_i^Tx_j=2\times1+3\times5=17}$$

Now choose

* Degree = 3
* $\gamma=2$
* `coef0` = 1

The kernel becomes

$$\boldsymbol{K(x_i,x_j)=(2\times17+1)^3}$$

$$\boldsymbol{=35^3}$$

$$\boldsymbol{=42875}$$

Notice the sequence:

1. Compute the original dot product.
2. Multiply by $\gamma$.
3. Add `coef0`.
4. Raise the result to the chosen degree.

---

# 6. Effect of the Parameters

| Parameter        | If Increased | Effect                                  |
| ---------------- | ------------ | --------------------------------------- |
| Degree ($d$)     | Higher       | More complex decision boundaries        |
| Gamma ($\gamma$) | Higher       | Similarity changes more rapidly         |
| coef0            | Higher       | Gives more weight to lower-degree terms |

---

# 7. Summary

The polynomial kernel used in practice is

$$\boldsymbol{K(x_i,x_j)=(\gamma x_i^Tx_j+coef0)^d}$$

where:

* $d$ controls the polynomial degree.
* $\gamma$ scales the original dot product.
* `coef0` introduces lower-degree terms through the polynomial expansion.
* This kernel still computes the transformed-space dot product **without explicitly constructing the transformed feature vectors**.
