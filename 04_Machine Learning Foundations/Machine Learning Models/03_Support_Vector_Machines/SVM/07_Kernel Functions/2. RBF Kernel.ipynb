# 1. RBF (Gaussian) Kernel

So far, we have studied the **Polynomial Kernel**, which measures similarity using the **dot product** between two samples.

The **RBF (Radial Basis Function) Kernel** takes a completely different approach.

Instead of asking:

> "How aligned are these two vectors?"

it asks:

> **"How far apart are these two samples?"**

So, the RBF kernel is based on **distance**, not the dot product.

---

# 2. Measuring the Distance Between Two Samples

Suppose we have two training samples:

$$
\boldsymbol{x_i=(x_{i1},x_{i2})}
$$

and

$$
\boldsymbol{x_j=(x_{j1},x_{j2})}
$$

The Euclidean distance between them is

$$
\boldsymbol{|x_i-x_j|=\sqrt{(x_{i1}-x_{j1})^2+(x_{i2}-x_{j2})^2}}
$$

Since the square root is computationally unnecessary for the kernel, we use the **squared Euclidean distance**:

$$
\boldsymbol{|x_i-x_j|^2=(x_{i1}-x_{j1})^2+(x_{i2}-x_{j2})^2}
$$

This value tells us how far apart the two samples are.

---

# 3. Why Use the Squared Distance?

Suppose

$$
\boldsymbol{x_i=(2,3)}
$$

and

$$
\boldsymbol{x_j=(5,7)}
$$

Compute the squared distance.

$$
\boldsymbol{|x_i-x_j|^2=(2-5)^2+(3-7)^2}
$$

$$
\boldsymbol{=(-3)^2+(-4)^2}
$$

$$
\boldsymbol{=9+16=25}
$$

The ordinary distance would be

$$
\boldsymbol{\sqrt{25}=5}
$$

For the RBF kernel, we only need the squared distance:

$$
\boldsymbol{25}
$$

Using the squared distance avoids the square-root operation while preserving the relative ordering of distances.

---

# 4. Converting Distance into Similarity

Distance and similarity are opposite ideas.

* Small distance → High similarity
* Large distance → Low similarity

If we used only the squared distance:

$$
\boldsymbol{|x_i-x_j|^2}
$$

then

* Similar samples would produce values near **0**.
* Dissimilar samples would produce **large** values.

However, a kernel should return **larger values for more similar samples**.

So we need a function that:

* Produces a value close to **1** when the distance is small.
* Produces a value close to **0** when the distance is large.

The exponential function provides exactly this behavior.

---

# 5. The RBF Kernel Formula

The RBF kernel is

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

where:

* $x_i$ = first training sample
* $x_j$ = second training sample
* $|x_i-x_j|^2$ = squared Euclidean distance
* $\gamma$ = parameter controlling how quickly similarity decreases with distance
* $\exp(\cdot)$ is the exponential function.

---

# 6. What Happens When the Distance Changes?

Suppose

$$
\boldsymbol{\gamma=1}
$$

### Case 1: Same Sample

If

$$
\boldsymbol{x_i=x_j}
$$

then

$$
\boldsymbol{|x_i-x_j|^2=0}
$$

The kernel becomes

$$
\boldsymbol{K(x_i,x_j)=e^{-1\times0}=e^0=1}
$$

A sample is perfectly similar to itself.

---

### Case 2: Small Distance

Suppose

$$
\boldsymbol{|x_i-x_j|^2=1}
$$

Then

$$
\boldsymbol{K(x_i,x_j)=e^{-1}}
$$

which is approximately

$$
\boldsymbol{0.368}
$$

The samples are still reasonably similar.

---

### Case 3: Large Distance

Suppose

$$
\boldsymbol{|x_i-x_j|^2=10}
$$

Then

$$
\boldsymbol{K(x_i,x_j)=e^{-10}}
$$

which is approximately

$$
\boldsymbol{0.000045}
$$

The similarity is now almost zero.

---

# 7. Summary So Far

The RBF kernel measures similarity based on **distance**, not the dot product.

Its formula is

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

It has these important properties:

* If two samples are identical, the kernel value is **1**.
* As the distance increases, the kernel value decreases smoothly toward **0**.
* The parameter $\gamma$ controls how quickly this decrease happens.

---

# 1. Understanding the Parameter $\boldsymbol{\gamma}$

In the RBF kernel, the formula is

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

Everything in this formula is fixed except **$\gamma$**.

So, to understand the behavior of the RBF kernel, we only need to understand what changing $\gamma$ does.

---

# 2. Where Does $\boldsymbol{\gamma}$ Appear?

Notice that $\gamma$ multiplies the squared distance.

$$
\boldsymbol{-\gamma|x_i-x_j|^2}
$$

This means that before taking the exponential, we first compute

1. The squared distance.
2. Multiply it by $\gamma$.
3. Add the negative sign.
4. Apply the exponential.

So $\gamma$ determines **how large or small the exponent becomes**.

---

# 3. Example with a Fixed Distance

Suppose the squared distance between two samples is

$$
\boldsymbol{|x_i-x_j|^2=4}
$$

Now let's see what happens for different values of $\gamma$.

---

### Case 1: Small $\gamma$

Suppose

$$
\boldsymbol{\gamma=0.1}
$$

The exponent becomes

$$
\boldsymbol{-0.1\times4=-0.4}
$$

Therefore,

$$
\boldsymbol{K(x_i,x_j)=e^{-0.4}}
$$

which is approximately

$$
\boldsymbol{0.670}
$$

Even though the samples are some distance apart, the similarity is still fairly high.

---

### Case 2: Medium $\gamma$

Suppose

$$
\boldsymbol{\gamma=1}
$$

The exponent becomes

$$
\boldsymbol{-1\times4=-4}
$$

Therefore,

$$
\boldsymbol{K(x_i,x_j)=e^{-4}}
$$

which is approximately

$$
\boldsymbol{0.0183}
$$

The similarity has dropped significantly.

---

### Case 3: Large $\gamma$

Suppose

$$
\boldsymbol{\gamma=10}
$$

The exponent becomes

$$
\boldsymbol{-10\times4=-40}
$$

Therefore,

$$
\boldsymbol{K(x_i,x_j)=e^{-40}}
$$

which is approximately

$$
\boldsymbol{4.25\times10^{-18}}
$$

This value is extremely close to zero.

So, even a moderate distance now makes the samples appear almost completely unrelated.

---

# 4. What Does This Mean?

Suppose the distance between two samples stays exactly the same.

Only $\gamma$ changes.

| $\gamma$ | Kernel Value | Interpretation                   |
| -------: | -----------: | -------------------------------- |
|    Small |        Large | Samples are considered similar   |
|   Medium |     Moderate | Similarity decreases             |
|    Large |   Very Small | Samples are considered different |

So:

* **Small $\gamma$** → Similarity decreases **slowly** with distance.
* **Large $\gamma$** → Similarity decreases **very quickly** with distance.

---

# 5. Intuition

Think of $\gamma$ as controlling the **range of influence** of each training sample.

* If $\gamma$ is **small**, a training sample influences points that are **far away**.
* If $\gamma$ is **large**, a training sample mainly influences points that are **very close** to it.

This directly affects the shape of the decision boundary learned by the SVM.

---

# 6. Summary

The RBF kernel is

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

The parameter $\gamma$ controls **how quickly similarity decreases as distance increases**.

* Small $\gamma$ → Slow decay of similarity.
* Large $\gamma$ → Fast decay of similarity.

The distance between the samples may stay the same, but changing $\gamma$ changes the kernel value, which changes how the SVM views the relationship between those samples.

---

# 1. How Does $\boldsymbol{\gamma}$ Affect the Decision Boundary?

Until now, we have learned that:

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

and that:

* Small $\gamma$ → Similarity decreases slowly.
* Large $\gamma$ → Similarity decreases quickly.

Now the question is:

> **How does this change the decision boundary learned by the SVM?**

---

# 2. Small $\boldsymbol{\gamma}$

Suppose

$$
\boldsymbol{\gamma=0.01}
$$

Then the exponent

$$
\boldsymbol{-\gamma|x_i-x_j|^2}
$$

changes very slowly because $\gamma$ is very small.

Even if two samples are fairly far apart,

$$
\boldsymbol{K(x_i,x_j)}
$$

is still reasonably large.

This means that **many training samples are considered similar to one another.**

---

## What Does the SVM Learn?

Since many samples influence each other, the SVM cannot create very small local bends in the decision boundary.

Instead, it learns a **smooth and simple** boundary.

For example:

```text
Class A        Decision Boundary        Class B

● ● ● ● ●
             /
            /
           /
          /
                     ▲ ▲ ▲ ▲ ▲
```

The boundary changes gradually.

---

# 3. Consequence of a Small $\boldsymbol{\gamma}$

A very small $\gamma$ may produce a boundary that is **too simple**.

Some training samples may be misclassified because the boundary does not bend enough to separate them correctly.

This is called

**Underfitting.**

Characteristics:

* Smooth boundary
* High bias
* Low variance
* May miss important patterns in the data

---

# 4. Large $\boldsymbol{\gamma}$

Now suppose

$$
\boldsymbol{\gamma=100}
$$

The exponent becomes much more negative for even small distances.

Therefore,

$$
\boldsymbol{K(x_i,x_j)}
$$

drops toward zero very quickly.

Only samples that are extremely close to each other are considered similar.

---

## What Does the SVM Learn?

Since each training sample mainly influences its immediate neighborhood, the SVM can create many small bends in the decision boundary.

Example:

```text
● ● ●      ▲

   \__
      \___
          \__
             \___

         ▲     ▲
```

The boundary becomes highly irregular.

---

# 5. Consequence of a Large $\boldsymbol{\gamma}$

The model tries to classify almost every training sample correctly.

As a result, it may also learn the random noise present in the training data.

This is called

**Overfitting.**

Characteristics:

* Highly complex boundary
* Low bias
* High variance
* Excellent training accuracy
* Poor performance on unseen test data

---

# 6. Comparison

| Small $\gamma$                        | Large $\gamma$                            |
| ------------------------------------- | ----------------------------------------- |
| Similarity decreases slowly           | Similarity decreases quickly              |
| Each sample influences a large region | Each sample influences only nearby points |
| Smooth decision boundary              | Highly complex decision boundary          |
| May underfit                          | May overfit                               |

---

# 7. Visual Intuition

### Small $\boldsymbol{\gamma}$

```text
Influence Region

          ●
    *************
  *****************
 *******************
  *****************
    *************
```

One training sample affects a **large area**.

---

### Large $\boldsymbol{\gamma}$

```text
Influence Region

      ●
    ***
   *****
    ***
```

One training sample affects only a **small neighborhood**.

---

# 8. Important Observation

Notice that **$\gamma$ does not directly draw the decision boundary.**

Instead:

1. $\gamma$ changes the kernel values.

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

2. These kernel values determine how similar the training samples are.

3. The SVM optimization algorithm uses these similarities to compute the optimal separating boundary.

So, **$\gamma$ influences the decision boundary indirectly through the kernel values.**

---

# 9. Summary

* **Small $\gamma$**

  * Similarity decreases slowly.
  * Large influence region.
  * Smooth boundary.
  * Can underfit.

* **Large $\gamma$**

  * Similarity decreases rapidly.
  * Small influence region.
  * Complex boundary.
  * Can overfit.

---

# 1. Why Is the RBF Kernel Called an Infinite-Dimensional Kernel?

When we studied the Polynomial Kernel, we saw that it implicitly maps the original features into a **higher-dimensional feature space**.

For example:

Original features:

$$
\boldsymbol{x=(x_1,x_2)}
$$

Degree 2 polynomial mapping:

$$
\boldsymbol{\phi(x)=\left(x_1^2,\sqrt{2}x_1x_2,x_2^2\right)}
$$

Here:

* Original space: **2 dimensions**
* Mapped space: **3 dimensions**

If we increase the degree to 3:

$$
\boldsymbol{\phi(x)=\left(x_1^3,\sqrt{3}x_1^2x_2,\sqrt{3}x_1x_2^2,x_2^3\right)}
$$

Now:

* Original space: **2 dimensions**
* Mapped space: **4 dimensions**

As the degree increases, the number of features also increases.

However, **for every polynomial degree, the number of features is still finite.**

---

# 2. What About the RBF Kernel?

The RBF kernel is

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

Notice something important.

There is **no explicit feature mapping** like

$$
\boldsymbol{\phi(x)=\left(\cdots\right)}
$$

This naturally raises the question:

> **If there is no mapping, what feature space is the RBF kernel using?**

The surprising answer is:

> **It is equivalent to mapping the data into an infinite-dimensional feature space.**

---

# 3. How Do We Know That?

To understand this, we need to look at the exponential function.

The exponential function has the Taylor series expansion

$$
\boldsymbol{e^z=1+z+\frac{z^2}{2!}+\frac{z^3}{3!}+\frac{z^4}{4!}+\cdots}
$$

Notice that the series **never ends**.

There is always another term.

---

# 4. Apply This to the RBF Kernel

The RBF kernel is

$$
\boldsymbol{K(x_i,x_j)=e^{-\gamma|x_i-x_j|^2}}
$$


Using the Taylor expansion,

$$
\boldsymbol{e^{-\gamma|x_i-x_j|^2}
==================================
$$

$$
1
+
\frac{\left(-\gamma|x_i-x_j|^2\right)}{1!}
+
\frac{\left(-\gamma|x_i-x_j|^2\right)^2}{2!}
+
\frac{\left(-\gamma|x_i-x_j|^2\right)^3}{3!}
+\cdots}
$$

Look carefully.

The expansion contains

* Constant term
* First-order term
* Second-order term
* Third-order term
* Fourth-order term
* ...

There is **no highest degree**.

The series continues forever.

---

# 5. Why Does This Mean Infinite Features?

Remember the Polynomial Kernel.

Degree 2 produced features involving degree-2 terms.

Degree 3 produced features involving degree-3 terms.

Degree 4 produced features involving degree-4 terms.

Now imagine a polynomial that contains

* Degree 1
* Degree 2
* Degree 3
* Degree 4
* Degree 5
* Degree 6

...

without ever stopping.

Each new degree introduces additional feature interactions.

Since there is **no largest degree**, there is **no finite number of mapped features**.

Therefore, the implicit feature vector

$$
\boldsymbol{\phi(x)}
$$

must contain **infinitely many components**.

---

# 6. Does the SVM Actually Create Infinite Features?

No.

This is the beauty of the kernel trick.

The SVM **never computes**

$$
\boldsymbol{\phi(x)}
$$

Instead, it computes only

$$
\boldsymbol{K(x_i,x_j)=e^{-\gamma|x_i-x_j|^2}}
$$

The kernel value is exactly the same as the dot product in that infinite-dimensional space,

but the infinite-dimensional vector itself is **never constructed**.

This is why kernel methods are computationally practical.

---

# 7. Why Is This Powerful?

Suppose we actually tried to build

$$
\boldsymbol{\phi(x)}
$$

for the RBF kernel.

It would have infinitely many features.

Such a vector:

* Could never be fully stored in memory.
* Could never be completely computed.
* Could never be used directly in matrix multiplication.

Yet, because of the kernel trick, we only compute

$$
\boldsymbol{K(x_i,x_j)}
$$

which is just a single number.

This lets the SVM benefit from an infinite-dimensional feature space without the impossible cost of explicitly representing it.

---

# 8. Summary

* The Polynomial Kernel maps data into a **finite-dimensional** feature space.
* The RBF kernel is mathematically equivalent to mapping data into an **infinite-dimensional** feature space.
* This comes from the infinite Taylor series of the exponential function.
* The SVM never constructs the infinite feature vector.
* It only computes the kernel value

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

which behaves exactly like the dot product in that infinite-dimensional space.


Yes, exactly.

The Taylor series expansion is **not** what is computed during training.

The expansion

$$
\boldsymbol{e^z=1+z+\frac{z^2}{2!}+\frac{z^3}{3!}+\cdots}
$$

is used only to **prove mathematically** that the RBF kernel is equivalent to an infinite-dimensional feature mapping.

It is **not** what SVM implementations compute.

---

In practice, the backend simply computes

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

This is the actual computation performed by libraries like **scikit-learn**, **LIBSVM**, and others.

The steps are:

1. Compute the difference

$$
\boldsymbol{x_i-x_j}
$$

2. Compute the squared Euclidean distance

$$
\boldsymbol{|x_i-x_j|^2}
$$

3. Multiply by $-\gamma$

$$
\boldsymbol{-\gamma|x_i-x_j|^2}
$$

4. Compute the exponential

$$
\boldsymbol{\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

That's it.

---

The Taylor series is only a **theoretical proof**.

It tells us:

> "Although we are computing only one exponential expression, mathematically it is equivalent to taking the dot product of two vectors that contain infinitely many features."

So you can think of it like this:

```text
Theory
Taylor Expansion
        │
        ▼
Proves RBF ≡ Infinite-dimensional feature mapping

────────────────────────────────────────────

Actual Implementation
        │
        ▼
K(xi, xj) = exp(-γ||xi - xj||²)
```

You **never** compute the infinite feature vector, and you **never** evaluate the Taylor series term by term during training. The backend directly computes the exponential formula, and the mathematics guarantees that this is equivalent to working in an infinite-dimensional feature space.


**`Why their is need to multiple the gamma can we just skip it?`**

In short:

Yes, we **could** define the RBF kernel without $\gamma$:

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-|x_i-x_j|^2\right)}
$$

But then the rate at which similarity decreases would be **fixed**.

Adding $\gamma$

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

lets us **control how sensitive the kernel is to distance**:

* Small $\gamma$ → Similarity decreases slowly.
* Large $\gamma$ → Similarity decreases quickly.

So, **$\gamma$ is included to make the RBF kernel tunable for different datasets.**


# 1. Advantages and Disadvantages of the RBF Kernel

The RBF kernel is the **most commonly used kernel** in SVM because it can model very complex decision boundaries without explicitly constructing high-dimensional features.

However, it also has some limitations.

| **Advantages**                                                                                                 | **Disadvantages**                                                                                         |
| -------------------------------------------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------------------- |
| Can learn **non-linear** decision boundaries.                                                                  | Requires tuning of **$\boldsymbol{\gamma}$** and **$\boldsymbol{C}$**.                                    |
| Equivalent to working in an **infinite-dimensional feature space** without explicitly creating those features. | Sensitive to the choice of hyperparameters. Poor choices can lead to underfitting or overfitting.         |
| Works well on many real-world datasets where the relationship is not linear.                                   | Training becomes slow on very large datasets because the kernel matrix grows with the number of samples.  |
| Usually performs well even when the number of features is small.                                               | Requires **feature scaling** (such as StandardScaler), otherwise distance calculations become misleading. |
| Often gives better accuracy than the Linear Kernel when the classes are not linearly separable.                | Harder to interpret than a Linear SVM because there is no simple linear decision boundary.                |

---

# 2. Why Feature Scaling Is Necessary

The RBF kernel depends on the squared Euclidean distance.

$$
\boldsymbol{K(x_i,x_j)=\exp\left(-\gamma|x_i-x_j|^2\right)}
$$

Suppose one feature ranges from

* 0 to 1

while another ranges from

* 0 to 1,000,000

The second feature will dominate the distance calculation.

As a result, the kernel will mainly consider differences in the large-scale feature and almost ignore the smaller-scale feature.

Therefore, before using the RBF kernel, we usually standardize or normalize the features.

---

# 3. Why Is the RBF Kernel the Default?

Many datasets are **not linearly separable**, and we usually do not know the correct relationship beforehand.

The RBF kernel is flexible enough to model:

* Simple decision boundaries.
* Moderately complex boundaries.
* Highly complex boundaries.

By adjusting only **$\boldsymbol{\gamma}$** and **$\boldsymbol{C}$**, it can adapt to many different datasets.

Because of this flexibility, libraries such as **scikit-learn** use **`kernel="rbf"`** as the default choice for `SVC`.

---

# 4. Summary

The RBF kernel is generally the **best default choice** when:

* The relationship between features and classes is unknown.
* The data is not linearly separable.
* The dataset size is moderate.
* The features have been properly scaled.

Its main drawbacks are the need to tune **$\boldsymbol{\gamma}$** and **$\boldsymbol{C}$**, and the higher computational cost compared to a linear kernel.
