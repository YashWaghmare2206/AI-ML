# Support Vector Machine (SVM)

## What Problem Does It Solve?

Support Vector Machine (SVM) is a **supervised machine learning algorithm** used for:

* **Classification** (most common)
* **Regression** (Support Vector Regression - SVR)
* **Outlier Detection** (One-Class SVM)

Its main goal is to find the **best decision boundary (hyperplane)** that separates different classes while maximizing the distance (margin) from the nearest data points.

---

## Why Was It Developed?

Earlier algorithms like Logistic Regression work well when classes are **linearly separable**.

However, many real-world datasets:

* Are **not perfectly separable**
* Have **complex boundaries**
* Need better generalization on unseen data

SVM was developed to:

* Maximize the margin between classes.
* Reduce overfitting by choosing the safest decision boundary.
* Handle non-linear data using the **Kernel Trick**.

---

## Real World Applications

### 1. Face Recognition

SVM can classify whether a detected face belongs to a particular person based on extracted facial features.

Example:

* Person A
* Person B
* Person C

After feature extraction, SVM predicts whose face it is.

---

### 2. Spam Email Detection

Email features such as:

* Number of suspicious words
* Number of links
* Sender reputation
* Attachment information

are used by SVM to classify an email as:

* Spam
* Not Spam

---

## Advantages

* Works well on high-dimensional datasets.
* Effective even when the number of features is greater than the number of samples.
* Can model non-linear relationships using kernels.
* Maximizes margin, leading to good generalization.
* Memory efficient because only **support vectors** are used.

---

## Limitations

* Training becomes slow for very large datasets.
* Choosing the correct kernel and hyperparameters can be difficult.
* Less interpretable than Logistic Regression.
* Sensitive to feature scaling.
* Training complexity increases rapidly with dataset size.

---

## Comparison with Previous Algorithms

| Algorithm             | Main Idea                                                | Difference from SVM                                                                                          |
| --------------------- | -------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------ |
| Linear Regression     | Predict continuous values                                | SVM is mainly for classification and margin maximization.                                                    |
| Ridge Regression      | Linear regression with L2 regularization                 | Solves regression, whereas SVM focuses on finding the optimal separating boundary.                           |
| Lasso Regression      | Linear regression with L1 regularization                 | Performs feature selection; SVM does not inherently do feature selection.                                    |
| Elastic Net           | Combination of L1 and L2                                 | Regularized regression, not a margin-based classifier.                                                       |
| Polynomial Regression | Fits polynomial curves                                   | SVM can model complex boundaries using kernels without explicitly creating polynomial features.              |
| KNN                   | Classifies using nearest neighbors                       | KNN stores all training data and predicts based on neighbors, while SVM learns an optimal decision boundary. |
| Naive Bayes           | Uses Bayes' theorem with feature independence assumption | Naive Bayes is probabilistic; SVM is geometric and margin-based.                                             |
