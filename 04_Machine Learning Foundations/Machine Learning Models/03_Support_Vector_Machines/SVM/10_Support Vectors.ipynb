# 1. What are Support Vectors?

A **Support Vector** is a training sample that directly determines the position of the optimal decision boundary.

In other words,

> If you move a support vector, the optimal hyperplane will usually change.

---

# 2. Why are they called "Support" Vectors?

Recall the equation obtained from the KKT Stationarity condition:

$$\boldsymbol{w=\sum_{i=1}^{n}\alpha_i y_i x_i}$$

Now recall Complementary Slackness:

$$\boldsymbol{\alpha_i\left(y_i(w^Tx_i+b)-1\right)=0}$$

From this equation, we concluded:

* If $$\boldsymbol{\alpha_i=0}$$

  then the point contributes nothing to $$\boldsymbol{w}$$.

* If $$\boldsymbol{\alpha_i>0}$$

  then the point contributes to $$\boldsymbol{w}$$.

Therefore,

$$\boldsymbol{w=\sum_{\text{Support Vectors}}\alpha_i y_i x_i}$$

Notice something remarkable.

The summation effectively ignores every point whose

$$\boldsymbol{\alpha_i=0}$$

Therefore, the decision boundary is constructed **only from the support vectors**.

This is why they are called **Support Vectors**:

> They **support** or **define** the optimal separating hyperplane.

---

# 3. Which Points Become Support Vectors?

For a **Hard Margin SVM**, Complementary Slackness gives

$$\boldsymbol{\alpha_i\left(y_i(w^Tx_i+b)-1\right)=0}$$

Suppose

$$\boldsymbol{\alpha_i>0}$$

Then

$$\boldsymbol{y_i(w^Tx_i+b)-1=0}$$

which implies

$$\boldsymbol{y_i(w^Tx_i+b)=1}$$

But this equation is exactly the **margin boundary**.

Therefore,

> Every support vector lies exactly on one of the two margin lines.

---

# 4. What About the Other Training Points?

Suppose a point is far from the margin.

For example,

$$\boldsymbol{y_i(w^Tx_i+b)=5}$$

Then

$$\boldsymbol{y_i(w^Tx_i+b)-1=4}$$

Complementary Slackness requires

$$\boldsymbol{\alpha_i\times4=0}$$

Hence,

$$\boldsymbol{\alpha_i=0}$$

This means

* the point contributes nothing to $$\boldsymbol{w}$$,
* it is **not** a support vector,
* moving it slightly usually does **not** affect the decision boundary.

---

# 5. Geometric Interpretation

Imagine the optimal separator.

```text
Positive Class

        ●
          ●
------------- Margin
       ★
===================== Decision Boundary
       ★
------------- Margin
          ●
        ●

Negative Class
```

Where

* **★** = Support Vectors
* **●** = Ordinary training samples

The support vectors touch the margins.

The ordinary points lie farther away.

---

# 6. Why Doesn't SVM Use Every Training Point?

Many machine learning algorithms depend on **all** training samples.

For example:

* k-NN stores every point.
* Linear Regression uses every point when estimating the parameters.

SVM is different.

It only needs the points that are **closest to the decision boundary**.

Why?

Because those points determine the **maximum possible margin**.

Points far away from the boundary already satisfy the constraints comfortably.

Moving them slightly does not change where the maximum-margin hyperplane should be.

---

# 7. Mathematical Proof

Recall

$$\boldsymbol{w=\sum_{i=1}^{n}\alpha_i y_i x_i}$$

Suppose there are five training points.

Assume

$$\boldsymbol{\alpha=[0,;0,;2,;1.5,;0]}$$

Then

$$\boldsymbol{w=0\cdot y_1x_1+0\cdot y_2x_2+2y_3x_3+1.5y_4x_4+0\cdot y_5x_5}$$

Simplifying,

$$\boldsymbol{w=2y_3x_3+1.5y_4x_4}$$

Notice that

* Points 1, 2, and 5 disappear completely.
* Only points 3 and 4 determine the classifier.

Therefore,

Points 3 and 4 are the **support vectors**.

---

# 8. Key Insight

The beauty of SVM is its **sparsity**.

Instead of storing information from every training point, it stores information only from the training points with

$$\boldsymbol{\alpha_i>0}$$

These points are called **support vectors**, and they alone determine the optimal decision boundary.

---

# 2. Types of Support Vectors (Soft Margin SVM)

In a **Hard Margin SVM**, every support vector lies exactly on the margin.

However, in a **Soft Margin SVM**, support vectors are **not all the same**.

There are **three types**.

---

# 2.1 Why Do We Have Different Types?

Recall the Soft Margin constraint:

$$\boldsymbol{y_i(w^Tx_i+b)\ge1-\xi_i}$$

where

$$\boldsymbol{\xi_i\ge0}$$

The slack variable allows some points to

* touch the margin,
* enter the margin,
* or even be misclassified.

Therefore, support vectors can belong to different categories.

---

# 2.2 Type 1: Support Vectors on the Margin

These are the ideal support vectors.

They satisfy

$$\boldsymbol{\xi_i=0}$$

and

$$\boldsymbol{y_i(w^Tx_i+b)=1}$$

They lie exactly on the margin.

Graphically,

```text
Positive

      ●
------------- Margin
      ★
==================== Decision Boundary
      ★
------------- Margin
      ●

Negative
```

The ★ points are support vectors touching the margin.

These are exactly the same type of support vectors that exist in Hard Margin SVM.

---

# 2.3 Type 2: Support Vectors Inside the Margin

Sometimes a point is correctly classified but lies inside the margin.

Then

$$\boldsymbol{0<\xi_i<1}$$

and

$$\boldsymbol{0<y_i(w^Tx_i+b)<1}$$

Notice two things.

The point is still

* on the correct side of the decision boundary,
* but inside the margin.

Example

Suppose

$$\boldsymbol{y_i(w^Tx_i+b)=0.7}$$

Then

$$\boldsymbol{\xi_i=0.3}$$

because

$$\boldsymbol{0.7=1-0.3}$$

These points are still support vectors because they influence the optimal hyperplane.

---

# 2.4 Type 3: Misclassified Support Vectors

Sometimes a point lies on the wrong side of the decision boundary.

Then

$$\boldsymbol{\xi_i>1}$$

and

$$\boldsymbol{y_i(w^Tx_i+b)<0}$$

For example,

$$\boldsymbol{y_i(w^Tx_i+b)=-0.6}$$

This means

* the point is misclassified,
* but Soft Margin SVM allows this,
* because forcing perfect classification might reduce generalization.

These points are also support vectors.

---

# 2.5 Complete Classification

| Type           | Condition                          | Slack Variable             | Support Vector? |
| -------------- | ---------------------------------- | -------------------------- | --------------- |
| Outside Margin | $$\boldsymbol{y_i(w^Tx_i+b)>1}$$   | $$\boldsymbol{\xi_i=0}$$   | ❌ No            |
| On Margin      | $$\boldsymbol{y_i(w^Tx_i+b)=1}$$   | $$\boldsymbol{\xi_i=0}$$   | ✅ Yes           |
| Inside Margin  | $$\boldsymbol{0<y_i(w^Tx_i+b)<1}$$ | $$\boldsymbol{0<\xi_i<1}$$ | ✅ Yes           |
| Misclassified  | $$\boldsymbol{y_i(w^Tx_i+b)<0}$$   | $$\boldsymbol{\xi_i>1}$$   | ✅ Yes           |

This table is one of the most important summaries in Soft Margin SVM.

---

# 2.6 Why Are Points Outside the Margin Not Support Vectors?

Suppose

$$\boldsymbol{y_i(w^Tx_i+b)=4}$$

Then the point already satisfies the constraint comfortably.

Moving this point slightly usually does not change the optimal hyperplane.

Therefore,

$$\boldsymbol{\alpha_i=0}$$

and the point contributes nothing to

$$\boldsymbol{w=\sum_i\alpha_i y_i x_i}$$

Hence it is **not** a support vector.

---

# 2.7 Why Are All Other Points Support Vectors?

Every point that

* touches the margin,
* lies inside the margin,
* or is misclassified,

places a restriction on where the hyperplane can move.

Therefore,

$$\boldsymbol{\alpha_i>0}$$

and these points contribute to

$$\boldsymbol{w}$$

They directly influence the learned classifier.

---

# 2.8 Visual Summary

```text
Positive Class

          ●        ← Outside Margin (Not SV)

--------------------------- Margin

      ★          ← On Margin

    ★            ← Inside Margin

=========================== Decision Boundary

       ★         ← Misclassified

--------------------------- Margin

          ●       ← Outside Margin (Not SV)

Negative Class
```

Where:

* **●** = Ordinary training points
* **★** = Support vectors

Notice that **every point that is on the margin, inside the margin, or misclassified is a support vector**, while points well outside the margin are not.

---

# 2.9 Key Takeaway

For **Hard Margin SVM**:

* Every support vector lies exactly on the margin.

For **Soft Margin SVM**:

Support vectors can be:

1. On the margin
2. Inside the margin
3. Misclassified

This flexibility is one of the main reasons **Soft Margin SVM** works well on real-world datasets that are not perfectly separable.


# 3. What Happens If We Remove a Support Vector?

This is one of the most important properties of SVM.

Let's understand it mathematically and intuitively.

---

# 3.1 Recall the Weight Vector

From the Stationarity condition,

$$\boldsymbol{w=\sum_{i=1}^{n}\alpha_i y_i x_i}$$

Notice that only points with

$$\boldsymbol{\alpha_i>0}$$

appear in this equation.

These are exactly the support vectors.

---

# 3.2 Suppose We Remove a Non-Support Vector

Assume we have five training samples.

Their multipliers are

$$\boldsymbol{\alpha=[0,;2,;1,;0,;3]}$$

Then

$$\boldsymbol{w=2y_2x_2+y_3x_3+3y_5x_5}$$

Notice that

* Point 1 has

  $$\boldsymbol{\alpha_1=0}$$

* Point 4 has

  $$\boldsymbol{\alpha_4=0}$$

Now remove Point 1.

The equation becomes

$$\boldsymbol{w=2y_2x_2+y_3x_3+3y_5x_5}$$

Nothing changes.

The decision boundary remains exactly the same.

---

# 3.3 Why?

Because that point never contributed to

$$\boldsymbol{w}$$

Its coefficient was already

$$\boldsymbol{0}$$

So removing it has no effect.

This is why SVM ignores most training samples after training.

---

# 3.4 Now Remove a Support Vector

Suppose Point 3 is removed.

Previously,

$$\boldsymbol{w=2y_2x_2+y_3x_3+3y_5x_5}$$

Now

$$\boldsymbol{w=2y_2x_2+3y_5x_5}$$

The weight vector changes.

Since

$$\boldsymbol{w}$$

changes,

the hyperplane

$$\boldsymbol{w^Tx+b=0}$$

also changes.

Therefore,

the margin changes,

and sometimes even the support vectors themselves change after retraining.

---

# 3.5 Geometric Interpretation

Imagine the optimal margin.

```text
Positive

          ●

---------------------- Margin

      ★

====================== Decision Boundary

      ★

---------------------- Margin

          ●

Negative
```

Suppose we remove the top **●**.

Nothing changes because it is far from the margin.

Now remove one of the **★** points.

The SVM must find a new maximum-margin separator.

The hyperplane usually shifts.

New support vectors may also appear.

---

# 3.6 Why Does This Happen?

Remember the optimization objective.

$$\boldsymbol{\min\frac{1}{2}|w|^2}$$

subject to

$$\boldsymbol{y_i(w^Tx_i+b)\ge1}$$

The support vectors are the points that make these constraints **tight**.

They determine exactly where the maximum-margin hyperplane can be placed.

If one of these critical constraints disappears, the optimization problem changes.

Therefore,

the optimal solution also changes.

---

# 3.7 Real-World Analogy (Minimal)

Think of stretching a rope between two walls.

Only the points touching the rope determine its position.

Objects far away from the rope don't affect it.

Similarly,

* Support vectors "touch" the margin.
* Other points are too far away to influence the separator.

---

# 3.8 Summary

| Remove...                                        | Effect on Hyperplane       |
| ------------------------------------------------ | -------------------------- |
| Non-support vector ($$\boldsymbol{\alpha_i=0}$$) | Usually no change          |
| Support vector ($$\boldsymbol{\alpha_i>0}$$)     | Hyperplane usually changes |

---

# 3.9 Key Insight

This explains why SVM is called a **sparse model**.

Although it is trained on all samples, the final classifier depends only on the relatively small subset of training points with

$$\boldsymbol{\alpha_i>0}$$

These are the **support vectors**, and they alone determine the decision boundary and the maximum margin.


# 4. Number of Support Vectors and Model Complexity

Now that we know **what support vectors are**, let's answer another important question:

> **Is having more support vectors good or bad?**

The answer is: **it depends**, but in general, **fewer support vectors lead to a simpler model**.

---

# 4.1 Why Does the Number of Support Vectors Matter?

Recall the prediction equation for a kernel SVM:

$$\boldsymbol{f(x)=\sum_{i=1}^{n}\alpha_i y_iK(x_i,x)+b}$$

Notice something important.

Only training points with

$$\boldsymbol{\alpha_i>0}$$

appear in the summation.

These are exactly the **support vectors**.

Therefore, if there are

* 20 support vectors → prediction uses 20 kernel computations.
* 500 support vectors → prediction uses 500 kernel computations.

So the prediction cost depends on the **number of support vectors**, not on the total number of training samples.

---

# 4.2 Example

Suppose you trained two SVM models on the same dataset.

### Model A

Training samples:

$$\boldsymbol{1000}$$

Support vectors:

$$\boldsymbol{25}$$

Prediction:

$$\boldsymbol{f(x)=\sum_{i=1}^{25}\alpha_i y_iK(x_i,x)+b}$$

---

### Model B

Training samples:

$$\boldsymbol{1000}$$

Support vectors:

$$\boldsymbol{600}$$

Prediction:

$$\boldsymbol{f(x)=\sum_{i=1}^{600}\alpha_i y_iK(x_i,x)+b}$$

Model B must compute far more kernel values for every prediction.

Therefore, it is slower.

---

# 4.3 Why Can the Number of Support Vectors Increase?

Several factors influence this.

### Small Margin

If the classes overlap heavily,

many points lie close to the boundary.

Therefore,

many become support vectors.

---

### Large Value of $$\boldsymbol{C}$$

A large

$$\boldsymbol{C}$$

tries to classify every point correctly.

As a result,

more points become important,

which often increases the number of support vectors.

---

### Large Value of $$\boldsymbol{\gamma}$$ (RBF)

A large

$$\boldsymbol{\gamma}$$

creates a very flexible boundary.

Many nearby points influence that boundary,

so the number of support vectors often increases.

---

# 4.4 What Happens When There Are Fewer Support Vectors?

A model with fewer support vectors usually has:

* Faster prediction.
* Lower memory usage.
* Simpler decision boundary.
* Better generalization (in many cases).

However,

too few support vectors can also indicate **underfitting**.

---

# 4.5 Relationship with Generalization

Generally,

* Very few support vectors → model may be too simple.
* Moderate number → often a good balance.
* Very large number (close to the size of the training set) → model may be overfitting.

There is no fixed "ideal" number.

It depends on the dataset.

---

# 4.7 Summary

| Number of Support Vectors | Effect                                                      |
| ------------------------- | ----------------------------------------------------------- |
| Small                     | Faster prediction, simpler model                            |
| Moderate                  | Usually a good balance                                      |
| Very large                | Slower prediction, more complex model, possible overfitting |

---

# 4.8 Key Insight

Unlike many machine learning algorithms, **SVM does not rely on all training samples during prediction**.

Instead, it stores and uses only the **support vectors**.

This is why support vectors are central to SVM: **they determine the decision boundary, the prediction cost, and much of the model's complexity.**
