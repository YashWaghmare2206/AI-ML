# 1. SVM Hyperparameters

So far, we have learned about the kernel functions.

Now let's look at the **hyperparameters** that control how an SVM learns.

The two most important hyperparameters are:

* $\boldsymbol{C}$ (Regularization Parameter)
* $\boldsymbol{\gamma}$ (Kernel Scaling Parameter)

We already studied $\boldsymbol{\gamma}$.

Now we'll study **$\boldsymbol{C}$**.

---

# 2. What is $\boldsymbol{C}$?

Remember the Soft Margin SVM objective:

$$
\boldsymbol{\min \frac{1}{2}|w|^2 + C\sum_{i=1}^{n}\xi_i}
$$

Here,

* $$\boldsymbol{\frac{1}{2}|w|^2}$$ tries to **maximize the margin**.
* $$\boldsymbol{\sum\xi_i}$$ measures the total margin violations (classification errors and points inside the margin).
* $$\boldsymbol{C}$$ decides **how much importance** we give to those violations.

So, **$\boldsymbol{C}$ is a weighting factor (regularization parameter)**.

It balances two competing goals:

1. **Large margin**
2. **Few training errors**

---

# 3. Why Do We Need $\boldsymbol{C}$?

Suppose we remove $C$.

The objective becomes

$$
\boldsymbol{\min \frac{1}{2}|w|^2 + \sum_{i=1}^{n}\xi_i}
$$

Now both terms have equal importance.

But every dataset is different.

Some datasets contain:

* Noise
* Outliers
* Incorrect labels

Sometimes we want to **ignore a few mistakes** to obtain a larger margin.

Other times we want to **avoid mistakes as much as possible**.

Without $C$, we cannot control this trade-off.

---

# 4. What Does a Small $\boldsymbol{C}$ Mean?

Suppose

$$
\boldsymbol{C=0.01}
$$

The second term

$$
\boldsymbol{C\sum\xi_i}
$$

becomes very small.

Therefore, the optimization mainly focuses on

$$
\boldsymbol{\frac{1}{2}|w|^2}
$$

The SVM prefers a **larger margin**, even if it allows more training mistakes.

Characteristics:

* Larger margin
* More misclassified training samples
* Better resistance to noise
* Lower risk of overfitting

---

# 5. What Does a Large $\boldsymbol{C}$ Mean?

Suppose

$$
\boldsymbol{C=1000}
$$

Now

$$
\boldsymbol{C\sum\xi_i}
$$

becomes much larger.

The optimizer strongly penalizes every margin violation.

The SVM will try very hard to classify the training data correctly.

Characteristics:

* Smaller margin
* Fewer training errors
* May fit noise
* Higher risk of overfitting

---

# 6. Summary

| Small $\boldsymbol{C}$              | Large $\boldsymbol{C}$          |
| ----------------------------------- | ------------------------------- |
| Large margin                        | Small margin                    |
| Allows more mistakes                | Penalizes mistakes heavily      |
| Better generalization in noisy data | Fits training data more closely |
| Lower chance of overfitting         | Higher chance of overfitting    |

---

# 7. Important Difference Between $\boldsymbol{C}$ and $\boldsymbol{\gamma}$

Students often confuse these two parameters.

| Parameter             | Controls                                                             |
| --------------------- | -------------------------------------------------------------------- |
| $\boldsymbol{C}$      | **Penalty for training errors** (regularization)                     |
| $\boldsymbol{\gamma}$ | **How quickly similarity decreases with distance** (kernel behavior) |

A simple way to remember them is:

* **$\boldsymbol{C}$** asks: *"How much should I punish mistakes?"*
* **$\boldsymbol{\gamma}$** asks: *"How local should each training point's influence be?"*

These two parameters work together. In practice, choosing good values for both has a major impact on the performance of an RBF SVM.
