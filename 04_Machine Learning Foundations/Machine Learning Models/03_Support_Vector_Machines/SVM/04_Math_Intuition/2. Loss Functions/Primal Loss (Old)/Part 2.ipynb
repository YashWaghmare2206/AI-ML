# 16. Problem with Hard Margin SVM

Hard Margin works only when

* Data is perfectly separable.
* No noisy samples exist.
* No outliers exist.

Real-world datasets rarely satisfy these conditions.

Example

```text
Positive

● ● ●

      ○

----------------

○ ○ ○
```

Here one positive point lies inside the negative class.

No line can classify every sample correctly.

Therefore

Hard Margin SVM fails.

---

# 17. Soft Margin SVM

Soft Margin SVM allows some mistakes.

Instead of saying

> Every sample **must** satisfy the margin,

it says

> Samples are allowed to violate the margin,
> but they receive a penalty.

The optimization now becomes a trade-off between

* Larger margin
* Fewer classification errors

---

Observe carefully.

There are now **two objectives**.

Objective 1



Make the margin as large as possible.

---

Objective 2



Reduce violations of the margin.

The optimizer tries to satisfy both objectives simultaneously.

---

# 21. What is C?

C is called the

**Regularization Parameter**

or

**Penalty Parameter**

It controls the trade-off between

* Large Margin
* Small Classification Error

---

## Large C

Suppose

**`C = 1000`**

Then every mistake becomes extremely expensive.

The optimizer tries to classify every point correctly.

Margin becomes smaller.

Chance of overfitting increases.

---

## Small C

Suppose

**`C=0.01`**

Mistakes become inexpensive.

The optimizer allows more violations.

Margin becomes larger.

Generalization usually improves.
