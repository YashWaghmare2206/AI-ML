# 1. What is Support Vector Machine (SVM)?

Support Vector Machine (SVM) is a **supervised machine learning algorithm** used for:

* Classification
* Regression (Support Vector Regression - SVR)

However, SVM is mainly used for **classification**.

The objective of SVM is **not just to separate two classes**, but to find the **best possible decision boundary** that separates the classes with the **maximum margin**.

Unlike Logistic Regression, which focuses on minimizing prediction error using probabilities, SVM focuses on **maximizing the distance between the decision boundary and the nearest training samples**.

This usually gives better generalization on unseen data.

---

# 2. Types of SVM

There are two major optimization formulations.

## 1. Primal SVM

Uses the original optimization problem.

It directly learns

* Weight vector ((w))
* Bias ((b))

The optimization is performed using the primal objective function.

---

## 2. Dual SVM

Instead of optimizing (w) and (b) directly, the optimization problem is converted into its **dual form** using Lagrange Multipliers.

Advantages:

* Kernel Trick becomes possible.
* Efficient for high-dimensional datasets.
* Only support vectors participate in the optimization.

Most practical libraries such as **scikit-learn** internally solve the **dual optimization problem**.

# 8. Margin

A margin is the **minimum distance between the decision boundary and the nearest training samples**.

Consider the following.

```
Positive Class

●      ●

------------------------  Positive Margin

------------------------  Hyperplane

------------------------  Negative Margin

○      ○
```

The center line is the decision boundary.

The upper line passes through the nearest positive samples.

The lower line passes through the nearest negative samples.

The distance between these two margin lines is called the **margin width**.

# 10. Support Vectors

The training samples lying exactly on the margin boundaries are called **Support Vectors**.

Example

```
Positive Class

      ●
           ●   ← Support Vector

------------------  +1

------------------   0

------------------  -1

○  ← Support Vector
      ○
```

Only these nearest samples determine the final hyperplane.

Removing points that are far from the margin usually does not change the hyperplane.

Removing a support vector often changes the hyperplane.

Hence the name

**Support Vector Machine**

because the hyperplane is "supported" by these critical vectors.

---

# 11. Hard Margin SVM

Hard Margin SVM assumes that the data is **perfectly linearly separable**.

That means:

* No overlap between classes.
* No noisy data.
* No misclassified samples.
* Every sample must satisfy the margin.

---

## Objective

Find the hyperplane with the **maximum margin** while correctly classifying **every** training sample.
