# 3. KKT Condition 2: Primal Feasibility

We have already completed **Stationarity**.

Now let's study the second KKT condition.

---

# 3.1 What Does "Primal Feasibility" Mean?

The word **feasible** means **allowed** or **valid**.

The primal optimization problem has certain constraints.

A solution is called **primal feasible** only if it satisfies **all the original constraints**.

In other words,

> **The solution obtained by the optimizer must lie inside the feasible region.**

---

# 3.2 Hard Margin SVM

Recall the optimization problem.

Objective:

$$\boldsymbol{\min_{w,b};\frac{1}{2}|w|^2}$$

Subject to

$$\boldsymbol{y_i(w^Tx_i+b)\ge1,\qquad i=1,2,\dots,n}$$

These constraints define the feasible region.

Therefore, the **Primal Feasibility** condition is simply

$$\boldsymbol{y_i(w^Tx_i+b)\ge1,\qquad\forall i}$$

Notice that this is **not a new equation**.

It is exactly the original constraint of the optimization problem.

---

# 3.3 Soft Margin SVM

For soft margin SVM, we introduced slack variables.

The optimization problem becomes

$$\boldsymbol{\min_{w,b,\xi};\frac{1}{2}|w|^2+C\sum_{i=1}^{n}\xi_i}$$

Subject to

$$\boldsymbol{y_i(w^Tx_i+b)\ge1-\xi_i}$$

and

$$\boldsymbol{\xi_i\ge0}$$

Therefore, the Primal Feasibility conditions are

$$\boldsymbol{y_i(w^Tx_i+b)\ge1-\xi_i}$$

and

$$\boldsymbol{\xi_i\ge0}$$

Again, these are simply the original constraints.

---

# 3.4 Why Is This a KKT Condition?

You might wonder:

> "Why is this even called a KKT condition if it's just the original constraint?"

The answer is:

KKT describes **what an optimal solution must satisfy**.

One of those requirements is:

> **The optimal solution must satisfy every original constraint.**

If even one constraint is violated, then the solution is **not feasible**, so it cannot be the optimal solution.

---

# 3.5 Example

Suppose after optimization we obtain

$$\boldsymbol{w=(2,1),\qquad b=-3}$$

Now consider a training sample

$$\boldsymbol{x=(1,2),\qquad y=+1}$$

Compute

$$\boldsymbol{y(w^Tx+b)}$$

First,

$$\boldsymbol{w^Tx=(2)(1)+(1)(2)=4}$$

Then,

$$\boldsymbol{y(w^Tx+b)=1(4-3)=1}$$

Since

$$\boldsymbol{1\ge1}$$

this sample satisfies the primal feasibility condition.

---

Now suppose another point gives

$$\boldsymbol{y(w^Tx+b)=0.7}$$

Since

$$\boldsymbol{0.7<1}$$

the constraint is violated.

Therefore,

* This solution is **not primal feasible** for a **Hard Margin SVM**.
* It cannot be the optimal solution.

---

# 3.6 Geometric Interpretation

Imagine the feasible region as the shaded area.

```text
Feasible Region

██████████████████████
██████████████████████
██████●███████████████   ← Valid Solution
██████████████████████

      ○                  ← Invalid (Outside)
```

* **●** satisfies every constraint.
* **○** violates at least one constraint.

Primal Feasibility says:

> The optimizer is allowed to choose only points inside the feasible region.

---

# 3.7 Summary

For **Hard Margin SVM**, Primal Feasibility is

$$\boldsymbol{y_i(w^Tx_i+b)\ge1}$$

For **Soft Margin SVM**, Primal Feasibility is

$$\boldsymbol{y_i(w^Tx_i+b)\ge1-\xi_i}$$

and

$$\boldsymbol{\xi_i\ge0}$$

**Intuition:** Primal Feasibility simply ensures that the solution found by the optimizer obeys the original SVM constraints. It does not create new equations; it guarantees that the chosen solution is valid before we can consider it optimal.
