# 4. KKT Condition 3: Dual Feasibility

---

# 4.1 What is Dual Feasibility?

Recall that while constructing the Lagrangian, we introduced one Lagrange multiplier for every constraint.

For Hard Margin SVM,

$$\boldsymbol{L(w,b,\alpha)=\frac{1}{2}|w|^2-\sum_{i=1}^{n}\alpha_i\left[y_i(w^Tx_i+b)-1\right]}$$

Here,

$$\boldsymbol{\alpha_1,\alpha_2,\ldots,\alpha_n}$$

are the **dual variables** (also called **Lagrange multipliers**).

These variables cannot take arbitrary values.

They must satisfy a condition.

This condition is called **Dual Feasibility**.

---

# 4.2 The Condition

For every training sample,

$$\boldsymbol{\alpha_i\ge0,\qquad\forall i}$$

This means every Lagrange multiplier must be **non-negative**.

Negative values are **not allowed**.

---

# 4.3 Why Must They Be Non-Negative?

Recall our constraint

$$\boldsymbol{y_i(w^Tx_i+b)\ge1}$$

Rewrite it as

$$\boldsymbol{1-y_i(w^Tx_i+b)\le0}$$

Suppose we define

$$\boldsymbol{g_i(w,b)=1-y_i(w^Tx_i+b)}$$

Then the constraint becomes

$$\boldsymbol{g_i(w,b)\le0}$$

For minimization problems, every constraint of the form

$$\boldsymbol{g(x)\le0}$$

is added to the Lagrangian as

$$\boldsymbol{L=f(x)+\sum_i\alpha_i g_i(x)}$$

where

$$\boldsymbol{\alpha_i\ge0}$$

This is a fundamental result from constrained optimization.

If negative multipliers were allowed, the optimizer could reward violating a constraint instead of penalizing it.

---

# 4.4 Intuition

Think of each constraint as having a **penalty weight**.

The multiplier

$$\boldsymbol{\alpha_i}$$

tells us how important that constraint is.

Suppose

$$\boldsymbol{\alpha_i=0}$$

Then that constraint contributes nothing to the Lagrangian.

Now suppose

$$\boldsymbol{\alpha_i=15}$$

Violating that constraint would greatly increase the objective, so the optimizer tries much harder to satisfy it.

Therefore,

* Small $$\boldsymbol{\alpha_i}$$ → constraint is less influential.
* Large $$\boldsymbol{\alpha_i}$$ → constraint is more influential.

Since these are **penalty weights**, they cannot be negative.

A negative penalty would effectively become a **reward** for violating the constraint, which would contradict the purpose of constrained optimization.

---

# 4.5 Why Is This Important in SVM?

Remember the equation from **Stationarity**:

$$\boldsymbol{w=\sum_{i=1}^{n}\alpha_i y_i x_i}$$

Notice something interesting.

If

$$\boldsymbol{\alpha_i=0}$$

then

$$\boldsymbol{\alpha_i y_i x_i=0}$$

That training sample contributes nothing to the weight vector.

If

$$\boldsymbol{\alpha_i>0}$$

then the sample contributes to

$$\boldsymbol{w}$$

This observation leads directly to the idea of **support vectors**, which we'll fully explain after completing all four KKT conditions.

---

# 4.6 Summary

The **Dual Feasibility** condition is

$$\boldsymbol{\alpha_i\ge0,\qquad\forall i}$$

Its purpose is to ensure that the Lagrange multipliers behave as **penalty weights** rather than rewards.

So far, we have:

| KKT Condition      | Mathematical Form                                                                 | Meaning                                          |
| ------------------ | --------------------------------------------------------------------------------- | ------------------------------------------------ |
| Stationarity       | $$\boldsymbol{\frac{\partial L}{\partial w}=0,;\frac{\partial L}{\partial b}=0}$$ | Connects primal and dual variables.              |
| Primal Feasibility | $$\boldsymbol{y_i(w^Tx_i+b)\ge1}$$ (Hard Margin)                                  | The solution satisfies the original constraints. |
| Dual Feasibility   | $$\boldsymbol{\alpha_i\ge0}$$                                                     | Every Lagrange multiplier must be non-negative.  |

The **fourth KKT condition—Complementary Slackness—is the most important one** because it explains **why only support vectors have non-zero multipliers** and mathematically identifies which training points determine the decision boundary.
