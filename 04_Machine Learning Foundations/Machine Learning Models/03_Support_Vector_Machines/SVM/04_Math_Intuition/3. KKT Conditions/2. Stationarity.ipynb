# 2. KKT Condition 1: Stationarity

This is the first and most important KKT condition because it is the one we actually used to derive the dual formulation.

---

# 2.1 Recall the Primal Problem

For simplicity, consider the **Hard Margin SVM**.

We want to minimize

$$\boldsymbol{\min_{w,b}\ \frac{1}{2}\|w\|^2}$$

subject to

$$\boldsymbol{y_i(w^Tx_i+b)\ge1}$$

Since there is a constraint, we cannot directly differentiate the objective function.

Instead, we construct the **Lagrangian**.

---

# 2.2 Lagrangian

The Lagrangian is

$$\boldsymbol{L(w,b,\alpha) = \frac{1}{2}\|w\|^2 - \sum_{i=1}^{n} \alpha_i \left[y_i(w^Tx_i+b)-1\right]}$$

where

$$\boldsymbol{\alpha_i\ge0}$$

are the Lagrange multipliers.

---

# 2.3 What Does Stationarity Mean?

The word **stationary** means:

> **The Lagrangian has reached a point where moving slightly in any primal variable does not decrease it further.**

Mathematically, the gradient of the Lagrangian with respect to every **primal variable** must be zero.

In SVM, the primal variables are $\boldsymbol{w}$ and $\boldsymbol{b}$ (and for soft margin, also $\boldsymbol{\xi}$).

Therefore, we require

$$\boldsymbol{\frac{\partial L}{\partial w}=0}$$

and

$$\boldsymbol{\frac{\partial L}{\partial b}=0}$$

For soft margin, we additionally require

$$\boldsymbol{\frac{\partial L}{\partial \xi_i}=0}$$

---

# 2.4 Differentiate with Respect to $\boldsymbol{w}$

Start with

$$\boldsymbol{L = \frac{1}{2}\|w\|^2 - \sum_i \alpha_i \left[y_i(w^Tx_i+b)-1\right]}$$

Take the derivative with respect to $\boldsymbol{w}$.

### First Term

We know

$$\boldsymbol{\frac{\partial}{\partial w}\left(\frac{1}{2}\|w\|^2\right) = w}$$

---

### Second Term

The derivative of

$$\boldsymbol{\sum_i \alpha_i y_i (w^Tx_i+b)}$$

with respect to $\boldsymbol{w}$ is

$$\boldsymbol{\sum_i \alpha_i y_i x_i}$$

because

$$\boldsymbol{\frac{\partial}{\partial w}(w^Tx_i) = x_i}$$

The constant $\boldsymbol{\sum_i \alpha_i}$ has derivative zero.

---

# 2.5 Set the Gradient Equal to Zero

Combining both derivatives,

$$\boldsymbol{\frac{\partial L}{\partial w} = w - \sum_i \alpha_i y_i x_i}$$

Stationarity requires

$$\boldsymbol{\frac{\partial L}{\partial w}=0}$$

Therefore,

$$\boldsymbol{w - \sum_i \alpha_i y_i x_i = 0}$$

Rearranging,

$$\boldsymbol{w = \sum_i \alpha_i y_i x_i}$$

---

# 2.6 Why Is This Equation So Important?

This equation tells us something remarkable:

$$\boldsymbol{w = \sum_i \alpha_i y_i x_i}$$

It means the weight vector is **not learned independently**.

Instead, it is completely determined by the training samples and their Lagrange multipliers.

This equation is the foundation of the **dual formulation** and later allows us to apply the **kernel trick** by replacing dot products with kernel functions.

---

# 2.7 Differentiate with Respect to $\boldsymbol{b}$

Now compute

$$\boldsymbol{\frac{\partial L}{\partial b}}$$

Only the constraint term contains $\boldsymbol{b}$, so

$$\boldsymbol{\frac{\partial L}{\partial b} = -\sum_i \alpha_i y_i}$$

Stationarity requires

$$\boldsymbol{\frac{\partial L}{\partial b}=0}$$

Therefore,

$$\boldsymbol{\sum_i \alpha_i y_i = 0}$$

This is the second equation we used when deriving the dual problem.

---

# 2.8 Summary

From the **Stationarity** condition, we obtain two fundamental equations:

$$\boldsymbol{w = \sum_i \alpha_i y_i x_i}$$

and

$$\boldsymbol{\sum_i \alpha_i y_i = 0}$$

These are exactly the equations you substituted into the primal objective to derive the SVM dual optimization problem. Stationarity is therefore the KKT condition that mathematically connects the primal variables to the dual variables.

A short intuition is:

**The stationarity condition means we adjust the primal variables $w$ and $b$ until the Lagrangian reaches its minimum with respect to them. At this point, changing $w$ or $b$ slightly no longer changes (decreases) the Lagrangian, so their gradients become zero. Solving these equations gives**

$$\boldsymbol{w=\sum_i \alpha_i y_i x_i}$$

and

$$\boldsymbol{\sum_i \alpha_i y_i=0}$$

These equations allow us to eliminate $w$ and $b$ and express the optimization entirely in terms of the dual variables $\alpha_i$.
