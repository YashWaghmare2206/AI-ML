# 5. KKT Condition 4: Complementary Slackness

This is the **most important KKT condition** in SVM.

It is the condition that mathematically explains:

* Why only support vectors matter.
* Why most training points have $$\boldsymbol{\alpha_i=0}$$.
* Why only a few points determine the decision boundary.

---

# 5.1 The Condition

For a **Hard Margin SVM**, Complementary Slackness is

$$\boldsymbol{\alpha_i\left(y_i(w^Tx_i+b)-1\right)=0,\qquad\forall i}$$

This equation looks simple, but it has a very deep meaning.

---

# 5.2 Understanding the Equation

The equation is

$$\boldsymbol{\alpha_i\left(y_i(w^Tx_i+b)-1\right)=0}$$

Notice that it is a **product of two quantities**.

The first quantity is

$$\boldsymbol{\alpha_i}$$

The second quantity is

$$\boldsymbol{y_i(w^Tx_i+b)-1}$$

Since their product must be zero,

**at least one of them must be zero**.

This gives us two possible cases.

---

# 5.3 Case 1

Suppose

$$\boldsymbol{\alpha_i=0}$$

Then

$$\boldsymbol{\alpha_i\left(y_i(w^Tx_i+b)-1\right)=0}$$

is automatically satisfied regardless of the value of

$$\boldsymbol{y_i(w^Tx_i+b)-1}$$

This means the point **does not contribute** to

$$\boldsymbol{w=\sum_i\alpha_i y_i x_i}$$

Such a point is **not a support vector**.

---

# 5.4 Case 2

Suppose

$$\boldsymbol{\alpha_i>0}$$

Then the only way for the product to be zero is

$$\boldsymbol{y_i(w^Tx_i+b)-1=0}$$

or

$$\boldsymbol{y_i(w^Tx_i+b)=1}$$

But what does this equation represent?

It is exactly the equation of the **margin**.

Therefore,

every point with

$$\boldsymbol{\alpha_i>0}$$

lies **on the margin**.

These are the **support vectors**.

---

# 5.5 Geometric Interpretation

Imagine the separating hyperplane.

```text
Class +1

        ●
----------------------   Margin
      ★
======================   Decision Boundary
      ★
----------------------   Margin
        ●

Class -1
```

* **★** = Support vectors
* **●** = Ordinary training samples

The support vectors satisfy

$$\boldsymbol{y_i(w^Tx_i+b)=1}$$

The ordinary points satisfy

$$\boldsymbol{y_i(w^Tx_i+b)>1}$$

---

# 5.6 Why Do Ordinary Points Have $$\boldsymbol{\alpha_i=0}$$?

Suppose a point is far away from the margin.

Then

$$\boldsymbol{y_i(w^Tx_i+b)=4}$$

Therefore

$$\boldsymbol{y_i(w^Tx_i+b)-1=3}$$

This is **not zero**.

For Complementary Slackness to hold,

$$\boldsymbol{\alpha_i\times3=0}$$

The only possibility is

$$\boldsymbol{\alpha_i=0}$$

So every point that is **strictly outside the margin** must have

$$\boldsymbol{\alpha_i=0}$$

Such points have **no influence** on the learned classifier.

---

# 5.7 Why Are They Called Support Vectors?

Recall the equation from Stationarity:

$$\boldsymbol{w=\sum_{i=1}^{n}\alpha_i y_i x_i}$$

Now divide the training points into two groups.

### Ordinary points

$$\boldsymbol{\alpha_i=0}$$

Their contribution is

$$\boldsymbol{0\times y_i x_i=0}$$

They disappear completely from the equation.

---

### Support vectors

$$\boldsymbol{\alpha_i>0}$$

Their contribution remains in

$$\boldsymbol{w}$$

Therefore,

$$\boldsymbol{w}$$

is constructed **only from the support vectors**.

This is why they are called **support vectors**:

> They **support** or **determine** the optimal separating hyperplane.

---

# 5.8 Summary of All Four KKT Conditions

| Condition                   | Mathematical Form                                                                 | Interpretation                                        |
| --------------------------- | --------------------------------------------------------------------------------- | ----------------------------------------------------- |
| **Stationarity**            | $$\boldsymbol{\frac{\partial L}{\partial w}=0,;\frac{\partial L}{\partial b}=0}$$ | Connects the primal and dual variables.               |
| **Primal Feasibility**      | $$\boldsymbol{y_i(w^Tx_i+b)\ge1}$$                                                | Every solution must satisfy the original constraints. |
| **Dual Feasibility**        | $$\boldsymbol{\alpha_i\ge0}$$                                                     | Lagrange multipliers must be non-negative.            |
| **Complementary Slackness** | $$\boldsymbol{\alpha_i\left(y_i(w^Tx_i+b)-1\right)=0}$$                           | Determines which points are support vectors.          |

---

# 5.9 The Most Important Insight

Complementary Slackness tells us:

* If

  $$\boldsymbol{\alpha_i=0}$$

  then the point is **not a support vector**.

* If

  $$\boldsymbol{\alpha_i>0}$$

  then

  $$\boldsymbol{y_i(w^Tx_i+b)=1}$$

  so the point lies exactly on the margin and **is a support vector**.

This single condition explains why SVM models are typically sparse: **only the support vectors have non-zero coefficients and influence the final decision boundary**, while the remaining training points do not contribute directly to the model.
