# 1. Why Do We Need KKT Conditions?

When solving an optimization problem, there are two cases:

### Case 1: No Constraints

Example:

$$
\boldsymbol{
f(x)=x^2
}
$$

To find the minimum, we simply compute

$$
\boldsymbol{
\frac{df}{dx}=0
}
$$

This gives the optimal solution.

---

### Case 2: Constraints Exist

SVM is **not** unconstrained.

For a hard-margin SVM, we must satisfy

$$
\boldsymbol{
y_i(w^Tx_i+b)\ge1
}
$$

For a soft-margin SVM,

$$
\boldsymbol{
y_i(w^Tx_i+b)\ge1-\xi_i
}
$$

and

$$
\boldsymbol{
\xi_i\ge0
}
$$

Now, we **cannot** simply differentiate the objective function because the solution must also satisfy these constraints.

---

# 2. The Problem

Suppose the unconstrained minimum occurs here:

```
          ●  ← Minimum of the function
```

But imagine this point violates the constraints.

The optimizer is **not allowed** to choose it.

Instead, it must find the **best feasible point**.

```
Feasible Region

██████████████████
██████████████████
██████●███████████  ← Best feasible solution
██████████████████
```

So optimization now has **two goals**:

1. Minimize the objective.
2. Satisfy every constraint.

---

# 3. This is Where KKT Comes In

The **Karush–Kuhn–Tucker (KKT) Conditions** provide the mathematical rules that an optimal solution must satisfy **when constraints are present**.

Think of them as an extension of

$$
\boldsymbol{
\nabla f(x)=0
}
$$

for constrained optimization.

Instead of only checking the gradient, KKT also checks whether the constraints are respected.

---

# 4. KKT in SVM

For SVM, KKT tells us:

* Is this solution optimal?
* Are all constraints satisfied?
* Which training points affect the decision boundary?
* Which points become support vectors?

Without KKT:

* We cannot derive the dual formulation.
* We cannot identify support vectors mathematically.
* We cannot solve SVM efficiently.

---

# 5. The Four KKT Conditions

For SVM, there are **four** conditions:

1. **Stationarity**
2. **Primal Feasibility**
3. **Dual Feasibility**
4. **Complementary Slackness**

Each one has a specific role:

| Condition               | Purpose                                                                                                                              |
| ----------------------- | ------------------------------------------------------------------------------------------------------------------------------------ |
| Stationarity            | Links the primal variables ($\boldsymbol{w}$, $\boldsymbol{b}$, $\boldsymbol{\xi}$) with the dual variables ($\boldsymbol{\alpha}$). |
| Primal Feasibility      | Ensures the original SVM constraints are satisfied.                                                                                  |
| Dual Feasibility        | Ensures the Lagrange multipliers are valid (non-negative).                                                                           |
| Complementary Slackness | Determines which points are support vectors and whether a constraint is active.                                                      |

---

# 6. The Big Picture

During the dual derivation, you already used these conditions without studying them individually.

The flow is:

```
Primal Optimization
        │
        ▼
Construct Lagrangian
        │
        ▼
Apply KKT Conditions
        │
        ▼
Obtain Dual Problem
        │
        ▼
Train SVM
        │
        ▼
Identify Support Vectors
```

KKT is therefore the **bridge** between the primal and dual formulations.

---

At this point, we have **not yet derived** the four conditions. We only understand **why KKT is needed**. The next topic is the **first KKT condition: Stationarity**, where you'll see how we obtain equations like

$$
\boldsymbol{
w=\sum_i\alpha_i y_i x_i
}
$$

directly from the KKT conditions.


```
Dataset
   │
   ▼
Preprocessing
   │
   ▼
Feature Scaling
   │
   ▼
Train-Test Split
   │
   ▼
Choose Kernel
   │
   ▼
Formulate the SVM Optimization Problem
   │
   ▼
Construct the Lagrangian
   │
   ▼
Apply KKT Conditions
   │
   ▼
Derive the Dual Optimization Problem
   │
   ▼
Solve for α (Lagrange Multipliers)
   │
   ▼
Support Vectors = Points with α > 0
   │
   ▼
Compute Decision Function
   │
   ├── Linear:      f(x) = wᵀx + b
   │
   └── Kernel SVM:  f(x) = Σ αᵢyᵢK(xᵢ, x) + b
   │
   ▼
Predict New Data
   │
   ▼
Evaluate Performance
```
