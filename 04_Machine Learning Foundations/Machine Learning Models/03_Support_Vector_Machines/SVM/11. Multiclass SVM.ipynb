# 1. Why Do We Need Multiclass SVM?

So far, everything we've studied assumes that there are only **two classes**.

For example,

* Spam vs Not Spam
* Cat vs Dog
* Positive vs Negative

This is called **binary classification**.

SVM was originally designed only for binary classification.

---

# 2. The Binary Nature of SVM

Recall the decision function.

$$\boldsymbol{f(x)=w^Tx+b}$$

The prediction rule is

$$\boldsymbol{\hat{y}=\operatorname{sign}(w^Tx+b)}$$

The sign function produces only two outputs.

$$\boldsymbol{\operatorname{sign}(z)=\begin{cases}+1, & z>0 \\ -1, & z<0\end{cases}}$$

Therefore, SVM can naturally predict only

* $$\boldsymbol{+1}$$
* $$\boldsymbol{-1}$$

Hence, the original SVM is a **binary classifier**.

---

# 3. What Happens If There Are More Than Two Classes?

Suppose we have three classes.

* Class A
* Class B
* Class C

A single hyperplane

$$\boldsymbol{w^Tx+b=0}$$

can only divide the feature space into **two regions**.

One side is

$$\boldsymbol{+1}$$

The other side is

$$\boldsymbol{-1}$$

There is no third side.

Therefore, a single SVM cannot directly classify three or more classes.

---

# 4. The Solution

Instead of modifying the mathematics of SVM, we build **multiple binary SVM classifiers**.

These classifiers work together to solve a multiclass problem.

The two standard approaches are:

1. **One-vs-Rest (OvR)**

2. **One-vs-One (OvO)**

These strategies convert a multiclass problem into several binary classification problems.

---

# 5. Example

Suppose we have

* Cat
* Dog
* Horse

Instead of training one classifier,

we train multiple binary SVMs.

Each classifier solves an easier binary problem.

Exactly how these classifiers are created depends on whether we use **One-vs-Rest** or **One-vs-One**.

---

# 6. Summary

The original SVM is naturally a **binary classifier** because its decision function predicts only

$$\boldsymbol{+1}$$

or

$$\boldsymbol{-1}$$

A single hyperplane cannot directly separate more than two classes.

To solve multiclass classification, we train **multiple binary SVMs** and combine their predictions.

The two most common strategies are:

* **One-vs-Rest (OvR)**
* **One-vs-One (OvO)**

# 2. One-vs-Rest (OvR)

One-vs-Rest (OvR) is the simplest and one of the most widely used techniques to extend SVM for multiclass classification.

The idea is very simple:

> **For every class, train one binary SVM that separates that class from all the remaining classes.**

---

# 2.1 Suppose We Have Three Classes

Assume our dataset contains

* A
* B
* C

Instead of training one classifier, we train **three** classifiers.

---

### Classifier 1

Treat

* A as the positive class

and

* B and C together as the negative class.

```text
Positive : A

Negative : B + C
```

---

### Classifier 2

Treat

* B as the positive class

and

* A and C together as the negative class.

```text
Positive : B

Negative : A + C
```

---

### Classifier 3

Treat

* C as the positive class

and

* A and B together as the negative class.

```text
Positive : C

Negative : A + B
```

So, for three classes, we train **three binary SVMs**.

---

# 2.2 General Formula

If there are

$$\boldsymbol{k}$$

classes,

then OvR trains

$$\boldsymbol{k}$$

binary classifiers.

Therefore,

$$\boldsymbol{\text{Number of Classifiers}=k}$$

Examples:

| Number of Classes | Number of OvR Classifiers |
| ----------------: | ------------------------: |
|                 3 |                         3 |
|                 5 |                         5 |
|                10 |                        10 |
|               100 |                       100 |

---

# 2.3 How Does Prediction Work?

Suppose a new sample arrives.

Instead of predicting immediately,

we pass it through **every classifier**.

Each classifier computes its decision function.

Classifier A

$$\boldsymbol{f_A(x)=w_A^Tx+b_A}$$

Classifier B

$$\boldsymbol{f_B(x)=w_B^Tx+b_B}$$

Classifier C

$$\boldsymbol{f_C(x)=w_C^Tx+b_C}$$

Each classifier returns a **decision score**.

Example:

| Classifier | Score |
| ---------- | ----: |
| A vs Rest  |   2.8 |
| B vs Rest  |  -0.5 |
| C vs Rest  |   1.3 |

The prediction is simply the classifier with the **largest decision score**.

Since

$$\boldsymbol{2.8>1.3>-0.5}$$

the sample is classified as

**Class A**.

---

# 2.4 Why Use the Largest Score?

Each classifier answers the question:

> "How confident am I that this sample belongs to my class?"

A larger decision function value means

* the sample lies farther inside the positive region,
* therefore the classifier is more confident.

Hence,

OvR predicts

$$\boldsymbol{\hat{y}=\arg\max_i f_i(x)}$$

where

* $$\boldsymbol{f_i(x)}$$ is the decision score of the $$\boldsymbol{i^{th}}$$ classifier.

---

# 2.5 Example

Suppose we want to classify flowers into

* Rose
* Lily
* Sunflower

Three SVMs are trained.

| Classifier | Positive Class | Negative Class   |
| ---------- | -------------- | ---------------- |
| SVM 1      | Rose           | Lily + Sunflower |
| SVM 2      | Lily           | Rose + Sunflower |
| SVM 3      | Sunflower      | Rose + Lily      |

A new flower produces

| Classifier        | Score |
| ----------------- | ----: |
| Rose vs Rest      |   4.2 |
| Lily vs Rest      |   1.1 |
| Sunflower vs Rest |  -0.8 |

The largest score is

$$\boldsymbol{4.2}$$

Therefore,

the prediction is **Rose**.

---

# 2.6 Advantages

* Very simple to understand.
* Only

  $$\boldsymbol{k}$$

  classifiers are needed.
* Fast to train when the number of classes is small.
* Works well for many practical problems.

---

# 2.7 Disadvantages

Each classifier is trained against **all remaining classes combined**.

For example,

```text
A  vs  (B + C + D + E)
```

The negative class becomes very large and heterogeneous.

As the number of classes increases,

each binary problem becomes more difficult,

which can reduce classification accuracy.

---

# 2.8 Summary

In **One-vs-Rest (OvR)**:

* Train one binary SVM for each class.
* Each classifier separates **one class** from **all other classes**.
* For

  $$\boldsymbol{k}$$

  classes,

  train

  $$\boldsymbol{k}$$

  classifiers.
* During prediction, compute the decision score from every classifier.
* Choose the class with the **highest decision score**:

$$\boldsymbol{\hat{y}=\arg\max_i f_i(x)}$$

OvR is simple, efficient, and widely used, but each classifier must distinguish one class from a potentially large mixture of all other classes.

# 3. One-vs-One (OvO)

Now let's study the second multiclass strategy.

Unlike **One-vs-Rest**, One-vs-One compares **every pair of classes separately**.

---

# 3.1 Main Idea

Instead of training one classifier against all remaining classes,

OvO trains one classifier for **every pair of classes**.

Suppose we have

* A
* B
* C

The classifiers are

```text
A vs B

A vs C

B vs C
```

Notice that each classifier only learns to distinguish **two classes**.

---

# 3.2 Example

Suppose our classes are

* Cat
* Dog
* Horse

We train

### Classifier 1

```text
Cat vs Dog
```

Ignore Horse completely.

---

### Classifier 2

```text
Cat vs Horse
```

Ignore Dog completely.

---

### Classifier 3

```text
Dog vs Horse
```

Ignore Cat completely.

So every classifier sees only **two classes**.

---

# 3.3 How Many Classifiers?

Suppose there are

$$\boldsymbol{k}$$

classes.

Every pair of classes requires one classifier.

The number of pairs is

$$\boldsymbol{\frac{k(k-1)}{2}}$$

Therefore,

$$\boldsymbol{\text{Number of OvO Classifiers}=\frac{k(k-1)}{2}}$$

---

### Examples

| Number of Classes | OvO Classifiers |
| ----------------: | --------------: |
|                 3 |               3 |
|                 4 |               6 |
|                 5 |              10 |
|                10 |              45 |
|               100 |            4950 |

Notice how quickly the number grows.

---

# 3.4 How Does Prediction Work?

Suppose we have

* A
* B
* C

A new sample arrives.

Each classifier makes one prediction.

Example

| Classifier | Prediction |
| ---------- | ---------- |
| A vs B     | A          |
| A vs C     | C          |
| B vs C     | C          |

Now count the votes.

| Class | Votes |
| ----- | ----: |
| A     |     1 |
| B     |     0 |
| C     |     2 |

The class with the **most votes** wins.

Therefore,

Prediction = **C**

---

# 3.5 Voting Rule

Each classifier casts **one vote**.

The predicted class is

> The class receiving the highest number of votes.

This is called **Majority Voting**.

Unlike OvR,

OvO **does not compare decision scores from different classifiers**.

It simply counts votes.

---

# 3.6 Why Does OvO Often Work Better?

Consider OvR.

One classifier learns

```text
Cat

vs

Dog + Horse + Lion + Elephant + Tiger
```

The negative class contains many different patterns.

Now consider OvO.

```text
Cat vs Dog
```

Only two classes are involved.

This is a much simpler classification problem.

Each SVM learns a cleaner boundary.

This often leads to higher accuracy.

---

# 3.7 Disadvantages

The main drawback is the number of classifiers.

Suppose

$$\boldsymbol{k=100}$$

Then

$$\boldsymbol{\frac{100\times99}{2}=4950}$$

binary SVMs must be trained.

Training and managing thousands of classifiers can become expensive.

---

# 3.8 Advantages

* Each classifier solves a simple binary problem.
* Often achieves better accuracy than OvR.
* Works particularly well with SVM because each classifier is trained on only two classes, resulting in smaller training subsets.

---

# 3.9 Summary

In **One-vs-One (OvO)**:

* Train one classifier for **every pair of classes**.
* Number of classifiers:

$$\boldsymbol{\frac{k(k-1)}{2}}$$

* Each classifier votes for one class.
* The class with the **maximum votes** is selected.

Compared with OvR:

* OvR trains fewer classifiers.
* OvO trains more classifiers but each one solves a simpler problem.

# 4. One-vs-Rest (OvR) vs One-vs-One (OvO)

Now that we have studied both strategies individually, let's compare them.

---

# 4.1 Number of Classifiers

Suppose there are

$$\boldsymbol{k}$$

classes.

### One-vs-Rest

Train one classifier for each class.

$$\boldsymbol{\text{Number of classifiers}=k}$$

---

### One-vs-One

Train one classifier for every pair of classes.

$$\boldsymbol{\text{Number of classifiers}=\frac{k(k-1)}{2}}$$

---

### Example

Suppose

$$\boldsymbol{k=5}$$

Then

| Strategy | Number of Classifiers |
| -------- | --------------------: |
| OvR      |                     5 |
| OvO      |                    10 |

Now suppose

$$\boldsymbol{k=10}$$

| Strategy | Number of Classifiers |
| -------- | --------------------: |
| OvR      |                    10 |
| OvO      |                    45 |

As the number of classes increases, OvO requires significantly more classifiers.

---

# 4.2 Training Time

### OvR

Each classifier is trained against **all remaining classes**.

Example

```text
Cat

vs

Dog + Horse + Cow + Lion
```

Each classifier sees almost the **entire dataset**.

There are fewer classifiers,

but each classifier trains on more data.

---

### OvO

Each classifier only trains on **two classes**.

Example

```text
Cat

vs

Dog
```

The Horse, Cow, and Lion samples are ignored.

Each classifier trains on a much smaller dataset,

but there are many more classifiers.

---

# 4.3 Prediction Time

### OvR

Every classifier evaluates the new sample once.

Total predictions

$$\boldsymbol{k}$$

---

### OvO

Every classifier votes once.

Total predictions

$$\boldsymbol{\frac{k(k-1)}{2}}$$

Therefore,

OvO generally requires more classifier evaluations during prediction.

---

# 4.4 Accuracy

### OvR

Each classifier must separate

```text
One Class

vs

Everything Else
```

The negative class can be very diverse.

This makes the learning problem harder.

---

### OvO

Each classifier distinguishes only between two classes.

```text
Cat

vs

Dog
```

This is a much simpler task.

Therefore,

OvO often achieves slightly better accuracy.

---

# 4.5 Memory Requirement

### OvR

Stores

$$\boldsymbol{k}$$

models.

---

### OvO

Stores

$$\boldsymbol{\frac{k(k-1)}{2}}$$

models.

Hence,

OvO usually requires more memory.

---

# 4.6 Complexity Comparison

| Feature                | OvR                | OvO                               |
| ---------------------- | ------------------ | --------------------------------- |
| Number of classifiers  | $$\boldsymbol{k}$$ | $$\boldsymbol{\frac{k(k-1)}{2}}$$ |
| Dataset per classifier | Large              | Small                             |
| Total models stored    | Fewer              | More                              |
| Prediction             | Faster             | Slower                            |
| Memory                 | Lower              | Higher                            |
| Accuracy               | Good               | Often Better                      |
| Implementation         | Simple             | More Complex                      |

---

# 4.7 Which Strategy is Better?

There is no universal winner.

### Use OvR when

* Number of classes is large.
* Training speed is important.
* Simplicity is preferred.

---

### Use OvO when

* Accuracy is the primary goal.
* Number of classes is moderate.
* Using SVM with nonlinear kernels.

---

# 4.8 Why Does SVM Commonly Use OvO?

SVM training complexity increases rapidly with the number of training samples.

In OvO,

each classifier trains on only **two classes**.

Although there are more classifiers,

each optimization problem is much smaller.

This is one reason why **OvO often works well for kernel SVMs**.

---

# 4.9 Summary

| Aspect                | One-vs-Rest             | One-vs-One                        |
| --------------------- | ----------------------- | --------------------------------- |
| Main Idea             | One class vs all others | Every pair of classes             |
| Number of Classifiers | $$\boldsymbol{k}$$      | $$\boldsymbol{\frac{k(k-1)}{2}}$$ |
| Prediction            | Highest decision score  | Majority voting                   |
| Training Dataset      | Almost all samples      | Only two classes                  |
| Training Simplicity   | Simpler                 | More classifiers                  |
| Accuracy              | Good                    | Often slightly better             |
| Memory                | Lower                   | Higher                            |

# 5. Which Strategy Does scikit-learn Use?

Now that we understand both **OvR** and **OvO**, let's see what **scikit-learn** actually uses.

The answer depends on **which SVM implementation you choose**.

---

# 5.1 `SVC`

The class

```python
from sklearn.svm import SVC
```

uses

**One-vs-One (OvO)**

by default for multiclass classification.

Suppose there are

$$\boldsymbol{k}$$

classes.

Then `SVC` automatically trains

$$\boldsymbol{\frac{k(k-1)}{2}}$$

binary SVMs.

You do **not** have to implement OvO yourself.

Example

```python
from sklearn.svm import SVC

model = SVC(kernel="rbf")

model.fit(X_train, y_train)
```

If your dataset has

* Class 0
* Class 1
* Class 2

then internally, `SVC` trains

```text
0 vs 1

0 vs 2

1 vs 2
```

and combines their predictions using majority voting.

---

# 5.2 Why Does `SVC` Use OvO?

Kernel SVM training becomes computationally expensive as the number of training samples increases.

In OvO,

each classifier uses only the samples belonging to two classes.

For example, if there are four classes:

```text
A

B

C

D
```

then the classifier

```text
A vs B
```

is trained using only samples from **A** and **B**.

Samples from **C** and **D** are ignored.

Each optimization problem is therefore much smaller.

Although there are more classifiers,

each one is easier to train.

For this reason, **OvO is generally a good default choice for kernel-based SVC**.

---

# 5.3 `LinearSVC`

The class

```python
from sklearn.svm import LinearSVC
```

uses

**One-vs-Rest (OvR)**

by default.

Instead of training classifiers for every pair of classes,

it trains

$$\boldsymbol{k}$$

classifiers.

Example

```python
from sklearn.svm import LinearSVC

model = LinearSVC()

model.fit(X_train, y_train)
```

If there are three classes,

internally it trains

```text
0 vs Rest

1 vs Rest

2 vs Rest
```

---

# 5.4 Why Does `LinearSVC` Use OvR?

`LinearSVC` is optimized for **linear decision boundaries** and large datasets.

Linear SVMs train much faster than kernel SVMs.

Using OvR means:

* Fewer classifiers.
* Lower memory usage.
* Faster training.
* Faster prediction.

For large datasets with many classes, this is often more efficient.

---

# 5.5 Can We Change the Strategy?

### For `SVC`

The multiclass strategy is **internally based on OvO**.

You generally **cannot switch `SVC` itself to use OvR** through a parameter.

If you specifically want OvR behavior with `SVC`, you can wrap it using scikit-learn's multiclass utilities.

Example:

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.svm import SVC

model = OneVsRestClassifier(SVC(kernel="rbf"))
```

Similarly, you can explicitly create an OvO wrapper:

```python
from sklearn.multiclass import OneVsOneClassifier
from sklearn.svm import SVC

model = OneVsOneClassifier(SVC(kernel="rbf"))
```

---

### For `LinearSVC`

The default strategy is OvR.

If you want OvO instead, you can also use the wrapper:

```python
from sklearn.multiclass import OneVsOneClassifier
from sklearn.svm import LinearSVC

model = OneVsOneClassifier(LinearSVC())
```

---

# 5.6 Quick Comparison

| Class       | Default Multiclass Strategy |
| ----------- | --------------------------- |
| `SVC`       | One-vs-One (OvO)            |
| `LinearSVC` | One-vs-Rest (OvR)           |

---

# 5.7 Summary

* `SVC` uses **One-vs-One (OvO)** by default.
* `LinearSVC` uses **One-vs-Rest (OvR)** by default.
* OvO is well suited for kernel SVMs because each classifier is trained on only two classes.
* OvR is efficient for linear SVMs and large datasets.
* If needed, scikit-learn provides `OneVsRestClassifier` and `OneVsOneClassifier` wrappers to explicitly choose the multiclass strategy regardless of the base classifier.
