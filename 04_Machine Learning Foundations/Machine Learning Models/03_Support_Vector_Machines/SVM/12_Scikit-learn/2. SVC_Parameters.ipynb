# 11.9 `SVC` Parameters

When we create an SVM model, we can control its behavior using **parameters**.

Example:

```python
from sklearn.svm import SVC

model = SVC(
    C=1.0,
    kernel="rbf",
    gamma="scale"
)
```

Here,

* `C`
* `kernel`
* `gamma`

are parameters.

---

The complete constructor is

```python
SVC(
    C=1.0,
    kernel="rbf",
    degree=3,
    gamma="scale",
    coef0=0.0,
    shrinking=True,
    probability=False,
    tol=1e-3,
    cache_size=200,
    class_weight=None,
    verbose=False,
    max_iter=-1,
    decision_function_shape="ovr",
    break_ties=False,
    random_state=None
)
```

# 11.11 `C`

### Definition

`C` is the **regularization parameter**.

It controls the trade-off between

* maximizing the margin
* minimizing classification errors

---

### Default

```python
C = 1.0
```

---

### Small `C`

Example

```python
SVC(C=0.1)
```

Small `C`

* allows more misclassification
* creates a wider margin
* stronger regularization
* may underfit

---

### Large `C`

Example

```python
SVC(C=100)
```

Large `C`

* penalizes mistakes heavily
* creates a narrower margin
* weaker regularization
* may overfit

---

### Summary

| Small `C`              | Large `C`               |
| ---------------------- | ----------------------- |
| Wider Margin           | Narrower Margin         |
| More Errors Allowed    | Fewer Errors Allowed    |
| Strong Regularization  | Weak Regularization     |
| Lower Overfitting Risk | Higher Overfitting Risk |

---

# 11.14 `gamma`

`gamma` determines how much influence a single training sample has.

It is used only with

* RBF
* Polynomial
* Sigmoid kernels

It is ignored for the linear kernel.

---

### Default

```python
gamma="scale"
```

---

### Options

```python
gamma="scale"
```

Automatically computed as

$$\boldsymbol{\frac{1}{n_{\text{features}}\times\operatorname{Var}(X)}}$$

where:

* $$\boldsymbol{n_{\text{features}}}$$ is the number of features.
* $$\boldsymbol{\operatorname{Var}(X)}$$ is the variance of the training data.

This is the recommended default because it adapts to the scale of the dataset.

---

```python
gamma="auto"
```

Automatically computed as

$$\boldsymbol{\frac{1}{n_{\text{features}}}}$$

This does **not** consider the data variance.

---

```python
gamma=0.1
```

or any positive float.

---

### Small `gamma`

* Smooth decision boundary
* Larger influence region
* May underfit

---

### Large `gamma`

* Complex decision boundary
* Small influence region
* May overfit

---

### Summary

| Small `gamma`         | Large `gamma`        |
| --------------------- | -------------------- |
| Smooth Boundary       | Complex Boundary     |
| Lower Variance        | Higher Variance      |
| Underfitting Possible | Overfitting Possible |

# 11.16 `coef0`

`coef0` is an independent constant term used in the **Polynomial** and **Sigmoid** kernels.

It is **ignored** when using:

* `kernel="linear"`
* `kernel="rbf"`

---

## Default

```python
coef0 = 0.0
```

---

## Used With

```python
SVC(kernel="poly", coef0=1.0)
```

or

```python
SVC(kernel="sigmoid", coef0=0.5)
```

---

## Polynomial Kernel

The polynomial kernel is

$$\boldsymbol{K(x_i,x_j)=\left(\gamma x_i^Tx_j+\text{coef0}\right)^{\text{degree}}}$$

Notice that `coef0` is **added before raising to the power**.

---

## Sigmoid Kernel

The sigmoid kernel is

$$\boldsymbol{K(x_i,x_j)=\tanh(\gamma x_i^Tx_j+\text{coef0})}$$

Again, `coef0` shifts the input to the sigmoid function.

---

## Effect

Increasing `coef0`

* increases the influence of lower-order terms in the polynomial kernel.
* shifts the sigmoid function horizontally.

In practice, most users leave it at its default value unless tuning a polynomial or sigmoid kernel.

---

## Summary

| Parameter | Used With           | Default |
| --------- | ------------------- | ------- |
| `coef0`   | Polynomial, Sigmoid | `0.0`   |

---

# 11.17 `shrinking`

`shrinking` enables or disables the **shrinking heuristic** used by the optimization algorithm.

---

## Default

```python
shrinking = True
```

---

## Purpose

During optimization, some training samples quickly become unlikely to be support vectors.

The shrinking heuristic temporarily removes such samples from optimization, reducing the amount of computation.

This usually speeds up training.

---

## Example

```python
SVC(shrinking=True)
```

or

```python
SVC(shrinking=False)
```

---

## Should You Change It?

Usually **No**.

Leave it as

```python
True
```

unless you have a specific reason or are benchmarking performance.

---

## Summary

| Value   | Meaning                               |
| ------- | ------------------------------------- |
| `True`  | Use shrinking heuristic (recommended) |
| `False` | Disable shrinking heuristic           |

---

# 11.20 `tol`

`tol` stands for **tolerance**.

It controls **when the optimization algorithm should stop training**.

Instead of searching forever for the perfect solution, the optimizer stops when the improvement becomes smaller than the specified tolerance.

---

## Default

```python
tol = 1e-3
```

which means

$$\boldsymbol{tol=0.001}$$

---

## Example

```python
SVC(tol=1e-4)
```

or

```python
SVC(tol=1e-2)
```

---

## Small `tol`

Example

```python
SVC(tol=1e-5)
```

Effects:

* More optimization iterations
* Better convergence
* Longer training time

---

## Large `tol`

Example

```python
SVC(tol=1e-1)
```

Effects:

* Fewer optimization iterations
* Faster training
* May stop before reaching the best solution

---

## Summary

| Small `tol`                | Large `tol`                |
| -------------------------- | -------------------------- |
| More Accurate Optimization | Faster Training            |
| Longer Training            | Less Accurate Optimization |

---

# 11.21 `cache_size`

Kernel SVM repeatedly computes kernel values.

Instead of recomputing them every time,

SVC stores many of these computations in memory called the **kernel cache**.

`cache_size` specifies the size of this cache.

---

## Default

```python
cache_size = 200
```

The value is measured in **megabytes (MB)**.

---

## Example

```python
SVC(cache_size=500)
```

---

## Effect

Larger cache

* More memory usage
* Fewer repeated kernel computations
* Can speed up training for large datasets

Smaller cache

* Less memory
* More kernel recomputation
* Possibly slower training

---

## Should You Change It?

Usually only when

* dataset is large
* sufficient RAM is available

Otherwise, the default value is suitable.

---

# 11.22 `class_weight`

Sometimes the dataset is **imbalanced**.

Example

| Class   | Samples |
| ------- | ------: |
| Class A |     950 |
| Class B |      50 |

Without adjustment,

the classifier may focus more on the majority class.

`class_weight` allows us to assign different importance to each class.

---

## Default

```python
class_weight = None
```

All classes receive equal weight.

---

## Automatic Balancing

```python
SVC(class_weight="balanced")
```

Scikit-learn automatically computes class weights based on class frequencies.

Minority classes receive higher weights.

---

## Manual Weights

```python
SVC(class_weight={0:1, 1:5})
```

Meaning

* Class 0 weight = 1
* Class 1 weight = 5

Misclassifying class 1 is penalized five times more heavily.

---

## When to Use

Use `class_weight` when

* classes are imbalanced,
* minority classes are important,
* you want to improve recall for rare classes.

---

# 11.26 `decision_function_shape`

This parameter controls **how the decision scores are returned for multiclass classification**.

> **Important:** It **does not change how `SVC` is trained.**

`SVC` always trains using **One-vs-One (OvO)** internally.

`decision_function_shape` only changes the **format of the output** produced by `decision_function()`.

---

## Default

```python
decision_function_shape = "ovr"
```

---

## Available Values

```python
decision_function_shape="ovr"
```

or

```python
decision_function_shape="ovo"
```

---

# What is `decision_function()`?

Instead of predicting only the class label,

```python
model.predict(X_test)
```

we can obtain the **decision scores**.

```python
model.decision_function(X_test)
```

These scores indicate how confident the classifier is.

---

# Option 1 : `"ovr"` (Default)

```python
SVC(decision_function_shape="ovr")
```

Suppose there are

$$\boldsymbol{3}$$

classes.

Then

```python
model.decision_function(X)
```

returns

```text
[
 [2.3, -1.2, 0.5],
 [0.1, 3.8, -2.1],
 ...
]
```

Each sample has

* one score for Class 0
* one score for Class 1
* one score for Class 2

The output shape is

$$\boldsymbol{(n_{\text{samples}},;n_{\text{classes}})}$$

This format is usually easier to interpret.

---

# Option 2 : `"ovo"`

```python
SVC(decision_function_shape="ovo")
```

Now the output contains one score for **every pair of classes**.

Suppose there are

$$\boldsymbol{3}$$

classes.

The classifiers are

```text
0 vs 1

0 vs 2

1 vs 2
```

The output shape becomes

$$\boldsymbol{\left(n_{\text{samples}},;\frac{k(k-1)}{2}\right)}$$

where

$$\boldsymbol{k}$$

is the number of classes.

---

## Summary

| Value   | Output Shape                                                       |
| ------- | ------------------------------------------------------------------ |
| `"ovr"` | $$\boldsymbol{(n_{\text{samples}},;n_{\text{classes}})}$$          |
| `"ovo"` | $$\boldsymbol{\left(n_{\text{samples}},;\frac{k(k-1)}{2}\right)}$$ |

---

## Important Note

Even if you write

```python
SVC(decision_function_shape="ovr")
```

the model is **still trained using One-vs-One**.

Only the format of `decision_function()` changes.

---

# 11.27 `break_ties`

Sometimes two or more classes receive exactly the same highest decision score.

This is called a **tie**.

The `break_ties` parameter specifies how such ties should be handled.

---

## Default

```python
break_ties = False
```

---

## Example

```python
SVC(break_ties=True)
```

---

## When Does It Work?

`break_ties` is only used when

```python
decision_function_shape="ovr"
```

If

```python
decision_function_shape="ovo"
```

then `break_ties` has no effect.

---

## Why Use It?

If several classes have equal highest scores,

setting

```python
break_ties=True
```

uses additional decision values to choose the most appropriate class instead of relying on an arbitrary tie-breaking rule.

This may improve predictions in rare tie situations.

---

## Should You Change It?

Usually,

```python
break_ties=False
```

is sufficient.

Enable it only if you specifically want improved handling of ties and are using

```python
decision_function_shape="ovr"
```

---

# 11.28 `random_state`

Some parts of SVM training involve randomness.

For example,

* probability estimation (`probability=True`)
* certain internal operations that may use randomized procedures

The `random_state` parameter controls this randomness.

---

## Default

```python
random_state = None
```

---

## Example

```python
SVC(random_state=42)
```

---

## Why Use It?

Suppose you train the same model multiple times.

Without a fixed random state,

small random differences may occur in operations that depend on randomness.

Using

```python
random_state=42
```

helps make those results reproducible.

---

## Common Values

```python
random_state=0
```

```python
random_state=42
```

```python
random_state=123
```

The actual number is arbitrary; it simply initializes the random number generator.
