# 11.1 SVM Classes in scikit-learn

The `sklearn.svm` module provides three main SVM classifiers.

## 1. `SVC`

```python
from sklearn.svm import SVC
```

**Purpose**

General Support Vector Classifier.

**Supports**

* Linear Kernel
* Polynomial Kernel
* RBF Kernel
* Sigmoid Kernel
* Custom Kernels

Uses **LIBSVM** internally.

Supports both

* Binary Classification
* Multiclass Classification

---

## 2. `LinearSVC`

```python
from sklearn.svm import LinearSVC
```

Specialized for **Linear SVM**.

Only supports

```text
Linear Kernel
```

Uses **LIBLINEAR** internally.

Designed for

* Large datasets
* High-dimensional data
* Faster training

---

## 3. `NuSVC`

```python
from sklearn.svm import NuSVC
```

Similar to `SVC`.

Instead of parameter

$$\boldsymbol{C}$$

it uses

$$\boldsymbol{\nu}$$

which controls

* Margin errors
* Number of support vectors

It is less commonly used than `SVC`.

# 11.3 Which One Should You Use?

### Use `SVC`

When

* dataset is small or medium-sized,
* nonlinear decision boundaries are expected,
* kernel trick is needed.

---

### Use `LinearSVC`

When

* dataset is very large,
* features are high-dimensional (e.g., text classification),
* a linear boundary is sufficient.

---

### Use `NuSVC`

When

* you want to control the model using the parameter $$\boldsymbol{\nu}$$ instead of $$\boldsymbol{C}$$.
* This is a specialized use case and is less common in practice.

---

# 11.4 Basic Workflow

The general workflow for using any SVM model in scikit-learn is:

```text
Import Model
        ↓
Create Model
        ↓
Fit on Training Data
        ↓
Predict on Test Data
        ↓
Evaluate Performance
```
