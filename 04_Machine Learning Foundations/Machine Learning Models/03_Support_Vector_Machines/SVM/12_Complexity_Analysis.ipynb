# 12_Complexity_Analysis.ipynb

Complexity analysis helps us understand **how much time and memory an algorithm requires** as the dataset grows.

For SVM, the complexity depends heavily on **whether we use a Linear SVM or a Kernel SVM**.

---

# 1. Why Study Complexity?

When choosing an algorithm, accuracy is not the only consideration.

We also need to know:

* How long does training take?
* How fast is prediction?
* How much memory is required?
* Will it scale to millions of samples?

Complexity analysis answers these questions.

---

# 2. Factors Affecting SVM Complexity

The complexity mainly depends on:

* Number of training samples

  $$\boldsymbol{n}$$

* Number of features

  $$\boldsymbol{d}$$

* Number of support vectors

  $$\boldsymbol{s}$$

where

* $$\boldsymbol{n}$$ = training samples
* $$\boldsymbol{d}$$ = features
* $$\boldsymbol{s}$$ = support vectors

---

# 3. Training Complexity

Training is the most computationally expensive part of SVM.

---

## Linear SVM

Linear SVM uses optimization methods designed specifically for linear problems.

Approximate training complexity:

$$\boldsymbol{O(nd)}$$

where

* $$\boldsymbol{n}$$ = number of samples
* $$\boldsymbol{d}$$ = number of features

This scales well even for very large datasets.

---

## Kernel SVM

Kernel SVM solves a quadratic optimization problem and computes a kernel matrix.

Approximate training complexity:

$$\boldsymbol{O(n^2d)\text{ to }O(n^3)}$$

The exact complexity depends on

* optimization algorithm,
* kernel,
* number of support vectors,
* convergence.

As the number of samples grows, training becomes much slower.

---

# 4. Why Is Kernel SVM Slow?

Suppose

$$\boldsymbol{n=1000}$$

The kernel matrix has

$$\boldsymbol{1000\times1000=1,!000,!000}$$

entries.

Now suppose

$$\boldsymbol{n=10000}$$

The kernel matrix becomes

$$\boldsymbol{10000\times10000=100,!000,!000}$$

entries.

The kernel matrix grows approximately as

$$\boldsymbol{n^2}$$

This is one of the main reasons kernel SVMs become expensive on large datasets.

---

# 5. Prediction Complexity

Prediction depends on the type of SVM.

---

## Linear SVM

Prediction computes

$$\boldsymbol{w^Tx+b}$$

Complexity:

$$\boldsymbol{O(d)}$$

Only one dot product is required.

Prediction is very fast.

---

## Kernel SVM

Prediction computes

$$\boldsymbol{f(x)=\sum_i\alpha_i y_iK(x_i,x)+b}$$

The summation includes only the support vectors.

If there are

$$\boldsymbol{s}$$

support vectors,

prediction complexity is approximately

$$\boldsymbol{O(sd)}$$

More support vectors mean slower prediction.

---

# 6. Memory Complexity

---

## Linear SVM

Stores only

* weight vector

  $$\boldsymbol{w}$$

* bias

  $$\boldsymbol{b}$$

Memory complexity:

$$\boldsymbol{O(d)}$$

---

## Kernel SVM

Stores

* all support vectors,
* dual coefficients,
* intercept.

Approximate memory complexity:

$$\boldsymbol{O(sd)}$$

If many support vectors exist,

memory usage increases significantly.

---

# 7. Complexity Summary

| Operation  | Linear SVM             | Kernel SVM                                |
| ---------- | ---------------------- | ----------------------------------------- |
| Training   | $$\boldsymbol{O(nd)}$$ | $$\boldsymbol{O(n^2d)\text{ to }O(n^3)}$$ |
| Prediction | $$\boldsymbol{O(d)}$$  | $$\boldsymbol{O(sd)}$$                    |
| Memory     | $$\boldsymbol{O(d)}$$  | $$\boldsymbol{O(sd)}$$                    |

---

# 8. Why Is `LinearSVC` Faster Than `SVC`?

`LinearSVC`

* learns only

  $$\boldsymbol{w}$$

  and

  $$\boldsymbol{b}$$

* does not compute kernel matrices.

* uses the **LIBLINEAR** optimizer.

---

`SVC`

* computes kernel values,
* solves the dual optimization problem,
* stores support vectors,
* uses the **LIBSVM** optimizer.

Therefore,

`LinearSVC` is significantly faster on large datasets.

---

# 9. When Should You Use `LinearSVC`?

Use `LinearSVC` when:

* the data is approximately linearly separable,
* the dataset contains many samples,
* training speed is important,
* memory is limited.

---

# 10. When Should You Use `SVC`?

Use `SVC` when:

* the relationship is nonlinear,
* kernel functions are required,
* the dataset is small or medium-sized,
* higher predictive accuracy is desired.

---

# 11. Practical Guidelines

| Dataset Size                                                | Recommendation                                       |
| ----------------------------------------------------------- | ---------------------------------------------------- |
| Less than 10,000 samples                                    | `SVC` is generally practical.                        |
| Tens of thousands or more samples                           | Prefer `LinearSVC` if a linear boundary is suitable. |
| Unknown nonlinear relationship with manageable dataset size | Start with `SVC(kernel="rbf")`.                      |
| Very high-dimensional sparse data (e.g., text)              | `LinearSVC` is usually preferred.                    |

*These are practical guidelines rather than strict rules; the best choice depends on your data and computational resources.*

---

# 12. Key Takeaways

* **Linear SVM** scales well to large datasets because it avoids kernel computations.
* **Kernel SVM** is more flexible but significantly more expensive to train.
* Prediction time for Kernel SVM depends on the **number of support vectors**, not just the number of training samples.
* Choosing between `LinearSVC` and `SVC` is often a trade-off between **speed** and **model flexibility**.
