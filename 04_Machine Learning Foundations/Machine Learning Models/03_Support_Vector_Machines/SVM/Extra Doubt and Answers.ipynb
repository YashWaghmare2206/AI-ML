## **So how the margin concept mapped in polynomial and RBF kernel?**

This is one of the most important concepts in SVM. The answer is that **the margin concept does not change**. Only the **space in which the margin is computed** changes.

Let's understand it step by step.

---

# 1. Linear SVM

In Linear SVM, the decision boundary is

$$
\boldsymbol{w^Tx+b=0}
$$

The margin is

$$
\boldsymbol{\frac{2}{|w|}}
$$

This margin is measured **in the original feature space**.

For example, if your data has two features, the margin is computed in that **2D space**.

---

# 2. Polynomial Kernel

Suppose we use a polynomial kernel.

The original data

$$
\boldsymbol{x}
$$

is implicitly mapped to

$$
\boldsymbol{\phi(x)}
$$

Now the SVM does **not** find a margin in the original space.

Instead, it finds the maximum margin in the transformed feature space.

The decision boundary becomes

$$
\boldsymbol{w^T\phi(x)+b=0}
$$

Notice that only

$$
\boldsymbol{x}
$$

has changed to

$$
\boldsymbol{\phi(x)}
$$

The optimization problem is **exactly the same**.

The margin is still

$$
\boldsymbol{\frac{2}{|w|}}
$$

but now

$$
\boldsymbol{w}
$$

belongs to the transformed feature space.

---

# 3. RBF Kernel

Exactly the same idea applies.

Instead of explicitly constructing

$$
\boldsymbol{\phi(x)}
$$

the kernel trick computes

$$
\boldsymbol{K(x_i,x_j)}
$$

The SVM behaves **as if** it had mapped the data into an infinite-dimensional feature space.

The decision boundary is still

$$
\boldsymbol{w^T\phi(x)+b=0}
$$

and the margin is still

$$
\boldsymbol{\frac{2}{|w|}}
$$

The only difference is that

$$
\boldsymbol{\phi(x)}
$$

is never explicitly computed.

---

# 4. Then Why Does the Boundary Look Curved?

This often confuses students.

In the transformed feature space, the separating boundary is **always a hyperplane**.

When we visualize the classifier back in the original feature space, that hyperplane appears as a curved boundary.

So:

* **Feature space:** Straight hyperplane with maximum margin.
* **Original space:** Curved decision boundary.

The margin is maximized in the **feature space**, not in the original space.

---

# 5. Summary

| Linear SVM                                    | Kernel SVM                                                                              |
| --------------------------------------------- | --------------------------------------------------------------------------------------- |
| Margin computed in the original feature space | Margin computed in the transformed feature space                                        |
| Boundary: $$\boldsymbol{w^Tx+b=0}$$           | Boundary: $$\boldsymbol{w^T\phi(x)+b=0}$$                                               |
| Hyperplane appears straight                   | Hyperplane is still straight in feature space, but appears curved in the original space |

So the **maximum-margin principle never changes**. Whether you use a **Linear**, **Polynomial**, or **RBF** kernel, SVM is always finding the **maximum-margin hyperplane**. The only difference is **which feature space that hyperplane lives in**.
