# 1. How Does SVM Compute the Margin Without Knowing $\boldsymbol{\phi(x)}$?

This is the natural next question.

You might think:

> "For the margin, we need $\boldsymbol{w}$. But $\boldsymbol{w}$ depends on $\boldsymbol{\phi(x)}$. If we never compute $\boldsymbol{\phi(x)}$, how can we compute the margin?"

The answer lies in the **dual formulation** that we derived earlier.

---

# 2. Recall the Dual Solution

From the dual formulation, we proved that

$$
\boldsymbol{w=\sum_{i=1}^{n}\alpha_i y_i\phi(x_i)}
$$

Notice something important.

The weight vector is **not stored directly**.

Instead, it is represented as a weighted sum of the transformed training samples.

So we never need to explicitly compute every coordinate of $\boldsymbol{w}$.

---

# 3. What Happens During Prediction?

For a new sample $\boldsymbol{x}$, the decision function is

$$
\boldsymbol{f(x)=w^T\phi(x)+b}
$$

Now substitute

$$
\boldsymbol{w=\sum_{i=1}^{n}\alpha_i y_i\phi(x_i)}
$$

We get

$$
\boldsymbol{f(x)=\left(\sum_{i=1}^{n}\alpha_i y_i\phi(x_i)\right)^T\phi(x)+b}
$$

Using the distributive property,

$$
\boldsymbol{f(x)=\sum_{i=1}^{n}\alpha_i y_i\left(\phi(x_i)^T\phi(x)\right)+b}
$$

Now apply the kernel trick.

Since

$$
\boldsymbol{\phi(x_i)^T\phi(x)=K(x_i,x)}
$$

the decision function becomes

$$
\boldsymbol{f(x)=\sum_{i=1}^{n}\alpha_i y_iK(x_i,x)+b}
$$

This is the formula actually used by kernel SVMs.

Notice that **$\boldsymbol{\phi(x)}$ has completely disappeared.**

---

# 4. Where Is the Margin Then?

The optimization algorithm still learns the coefficients

$$
\boldsymbol{\alpha_1,\alpha_2,\ldots,\alpha_n}
$$

Using these coefficients and the kernel values, it finds the hyperplane that maximizes the margin in the feature space.

So the algorithm **never constructs**

$$
\boldsymbol{\phi(x)}
$$

or

$$
\boldsymbol{w}
$$

explicitly.

Instead, everything is expressed using

$$
\boldsymbol{K(x_i,x_j)}
$$

---

# 5. Does the Formula for the Margin Change?

No.

The geometric margin is still

$$
\boldsymbol{\frac{2}{|w|}}
$$

However, in kernel SVM, we usually **do not compute this value explicitly**, because computing

$$
\boldsymbol{|w|}
$$

would require the feature-space representation.

Instead, the optimization process automatically maximizes the margin while working entirely with kernel values.

---

# 6. What Does the Backend Actually Store?

For a kernel SVM, the trained model stores:

* Support vectors.
* Their corresponding coefficients

$$
\boldsymbol{\alpha_i}
$$

* Their class labels

$$
\boldsymbol{y_i}
$$

* The bias

$$
\boldsymbol{b}
$$

* The chosen kernel (Linear, Polynomial, RBF, etc.).

It **does not store an explicit weight vector** as a linear SVM does.

---

# 7. Summary

For a **Linear SVM**:

* The model learns and stores

$$
\boldsymbol{w}
$$

* Prediction is

$$
\boldsymbol{f(x)=w^Tx+b}
$$

For a **Kernel SVM**:

* The model never explicitly computes

$$
\boldsymbol{w}
$$

or

$$
\boldsymbol{\phi(x)}
$$

* Prediction is made using

$$
\boldsymbol{f(x)=\sum_{i=1}^{n}\alpha_i y_iK(x_i,x)+b}
$$

This is the key advantage of the kernel trick: it allows SVM to work in extremely high-dimensional—even infinite-dimensional—feature spaces while only computing kernel values.
