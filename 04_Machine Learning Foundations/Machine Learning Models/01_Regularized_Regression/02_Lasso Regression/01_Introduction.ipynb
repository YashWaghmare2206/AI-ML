# 02 — Lasso Regression

## 1. What is Lasso Regression?

**Lasso (Least Absolute Shrinkage and Selection Operator)** is a regularized version of **Linear Regression**.

It tries to achieve two goals at the same time:

1. Fit the training data well.
2. Keep the model simple by shrinking coefficients.

Unlike Ridge Regression, **Lasso can make some coefficients exactly zero**, effectively removing those features from the model.

---

## 2. Why was Lasso introduced?

Before Lasso, we had **Linear Regression**.

Linear Regression only minimizes prediction error.



It doesn't care whether:

* coefficients are huge,
* irrelevant features are included,
* the model overfits.

| Age            |                       12.4 |           10.8 |

Notice:

The useless feature **Favorite Color** still exists.

Its coefficient is **very small**, but **not zero**.

---

## The Problem with Ridge

Imagine you have **10,000 features**.

Only **120** are actually useful.

Ridge may produce something like

| Feature       |  Weight |
| ------------- | ------: |
| Feature 1     |     2.4 |
| Feature 2     |    -1.7 |
| Feature 3     |   0.003 |
| Feature 4     |  -0.001 |
| Feature 5     |  0.0002 |
| ...           |     ... |
| Feature 10000 | 0.00005 |

Almost every feature remains in the model.

Although many coefficients are tiny, the model still has to:

* store all of them,
* multiply every feature during prediction,
* keep irrelevant variables.

---

## What if we could completely remove useless features?

Instead of

```text
Feature A → 2.4
Feature B → 0.002
Feature C → -0.001
Feature D → 0.0005
```

it would be better if we had

```text
Feature A → 2.4
Feature B → 0
Feature C → 0
Feature D → 0
```

Now the model automatically ignores those features.

This idea led to **Lasso Regression**.

---

## What does Lasso do?

Lasso minimizes prediction error **and** forces some coefficients to become exactly zero.

For example,

Before Lasso

```text
Age             5.4
Income          3.1
Height          0.07
Weight          0.01
Favorite Color  0.003
```

After Lasso

```text
Age             5.4
Income          3.1
Height          0
Weight          0
Favorite Color  0
```

Only the important features remain.

---

## Why is this useful?

Lasso provides **automatic feature selection**.

Benefits include:

* Simpler models
* Faster predictions
* Less memory usage
* Easier interpretation
* Reduced overfitting in many cases

---

## Summary

* **Linear Regression** → minimizes only prediction error.
* **Ridge Regression** → shrinks coefficients but rarely makes them zero.
* **Lasso Regression** → shrinks coefficients and can make some exactly zero, automatically selecting important features.
