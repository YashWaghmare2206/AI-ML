# 4. Lasso Cost Function

Now that you understand **why the derivative becomes `sign(w)`**, we can write the complete cost function.

---

## Linear Regression

Linear Regression minimizes only the prediction error.



Goal:

> Fit the training data as accurately as possible.

---

## Ridge Regression

Ridge adds an **L2 penalty**.



The second term discourages large weights.

---

## Lasso Regression

Lasso replaces the square with the **absolute value**.



This is the only mathematical difference from Ridge.

---

# Understanding every term

### First term



This is the **Mean Squared Error (MSE)**.

It measures prediction error.

---

### Second term



This is called the **L1 Regularization** or **L1 Penalty**.

It encourages coefficients to become zero.

---

### α (Alpha)

Exactly like Ridge,

(\alpha) controls the strength of regularization.

---

If

```text
α = 0
```

No regularization.

Lasso becomes ordinary Linear Regression.

---

If

```text
α = 0.01
```

Very weak regularization.

---

If

```text
α = 100
```

Very strong regularization.

Many coefficients become zero.

---

# Example

Suppose

```text
Weights

[4, -3, 1]
```

Prediction Error

```text
25
```

Alpha

```text
2
```

---

### Compute L1 penalty

[
|4|+|-3|+|1|
============

# 4+3+1

8
]

Penalty

[
2\times8
========

16
]

Total loss

```text
25 + 16 = 41
```

---

Now suppose Gradient Descent updates the weights.

New weights

```text
[3, -2, 0]
```

Prediction error increases slightly.

Suppose

```text
Prediction Error = 28
```

Penalty

[
|3|+|-2|+|0|
============

5
]

Penalty

[
2\times5
========

10
]

Total loss

```text
28 + 10 = 38
```

Notice:

Prediction error became worse

```text
25 → 28
```

But the total loss became

```text
41 → 38
```

Lasso prefers the second model because it is simpler.

---

# Why isn't the bias included?

Notice



There is **no bias term**.

Exactly like Ridge,

we regularize only the weights.

Reason:

The bias shifts the regression line up or down.

Penalizing it generally doesn't help reduce model complexity.
