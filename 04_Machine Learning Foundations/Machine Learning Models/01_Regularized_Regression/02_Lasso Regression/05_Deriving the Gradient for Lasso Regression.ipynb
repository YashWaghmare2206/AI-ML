# Vectorized NumPy Implementation

The gradient becomes

```python
error = y_pred - y

dw = (
    (2 / n_samples) * (X_train.T @ error)
    + (alpha / n_samples) * np.sign(self.w)
)

db = (2 / n_samples) * np.sum(error)
```

Weight update

```python
self.w -= self.learning_rate * dw
```

Bias update

```python
self.b -= self.learning_rate * db
```

# Comparison of the Three Algorithms

| Algorithm             | (dw)                                                                            |
| --------------------- | ------------------------------------------------------------------------------- |
| **Linear Regression** | (\displaystyle \frac{2}{n}X^T(\hat y-y))                                        |
| **Ridge Regression**  | (\displaystyle \frac{2}{n}X^T(\hat y-y)+\frac{2\alpha}{n}w)                     |
| **Lasso Regression**  | (\displaystyle \frac{2}{n}X^T(\hat y-y)+\frac{\alpha}{n}\operatorname{sign}(w)) |

Notice how only the **regularization term changes**:

* **Linear Regression:** No regularization.
* **Ridge:** Adds a term proportional to the current weight.
* **Lasso:** Adds a constant push determined only by the sign of each weight.
