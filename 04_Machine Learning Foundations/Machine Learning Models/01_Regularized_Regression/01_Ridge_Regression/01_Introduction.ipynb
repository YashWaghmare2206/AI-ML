# What Problem Does It Solve?

Linear Regression works well when the relationship between features and the target is approximately linear. However, in real datasets it often faces two major problems:

### 1. Overfitting

When a model learns the training data too closely, it also learns the noise instead of the true relationship.

Example:

Suppose we predict house prices using:

* Area
* Bedrooms
* Bathrooms
* Age
* Distance from city
* School rating
* 200 additional unnecessary features

Ordinary Linear Regression tries to fit all features, even the useless ones.

Result:

* Excellent training accuracy
* Poor testing accuracy

This is called **overfitting**.

---

### 2. Multicollinearity

Sometimes two or more features contain almost identical information.

Example:

```
Height (cm)
Height (inch)
```

or

```
Salary
Annual Income
```

These variables are highly correlated.

Linear Regression struggles because there are many equally good coefficient values.

Example:

Instead of

```
Weight = 3
Height = 5
```

it might learn

```
Weight = 100
Height = -92
```

Both produce similar predictions, but the coefficients become unstable.

---

### Ridge Regression solves these problems.

Instead of only minimizing prediction error, it also discourages very large coefficient values.

Instead of saying

> Fit the data as perfectly as possible

Ridge says

> Fit the data well, but keep coefficients small.

# Why Was It Developed?

Linear Regression has existed for over two centuries and works well under ideal conditions.

However, modern datasets introduced new challenges:

* Hundreds of features
* Thousands of correlated variables
* Noisy measurements
* Limited training samples

Researchers observed that minimizing only prediction error often produced unstable models.

Ridge Regression was introduced to improve model stability by adding a penalty on coefficient size.

This technique is called **regularization**.

Its goal is not to increase training accuracy.

Its goal is to improve **test accuracy**.

# Advantages

### 1. Reduces Overfitting

Keeps coefficients smaller.

Generalizes better on unseen data.

---

### 2. Handles Multicollinearity

Produces much more stable coefficients.

---

### 3. Uses All Features

Unlike Lasso, Ridge does **not** remove features.

Every feature contributes a little.

---

### 4. More Stable Than Linear Regression

Small changes in data produce smaller coefficient changes.

---

### 5. Easy to Train

Optimization remains convex.

There is still one global optimum.

---

### 6. Works Well with Many Features

Especially useful when

```
Number of Features >> Number of Samples
```

---

# Limitations

### 1. Does Not Perform Feature Selection

Every coefficient remains in the model.

Some become very small but rarely exactly zero.

---

### 2. Harder to Interpret

Since all variables remain, interpretation becomes more difficult than Lasso.

---

### 3. Requires Feature Scaling

Because the penalty depends on coefficient size, features should usually be standardized before training.

Without scaling:

```
Income: 100000
Age: 30
```

Income dominates the penalty.

---

### 4. Requires Choosing α (Regularization Strength)

Choosing too small:

Almost Linear Regression.

Choosing too large:

Underfitting.

---

### 5. Still Assumes Linear Relationships

If the true relationship is highly nonlinear, Ridge alone is insufficient.

## Key Takeaways

* Ridge Regression is an extension of **Linear Regression**.
* It introduces **L2 Regularization**, which penalizes large coefficients.
* Its primary purpose is to **reduce overfitting** and **handle multicollinearity**.
* Unlike **Lasso Regression**, Ridge **does not eliminate features**; it only shrinks their coefficients.
* It generally improves **generalization performance** on unseen data while keeping the optimization problem convex.
