# 4. Python Implementation

Suppose:

[
z=2x-6
]

---

```python
import numpy as np

def predict(x):
    z = 2*x - 6

    if z >= 0:
        return 1
    else:
        return 0
```

---

Predictions:

```python
for x in [1,2,3,4,5]:
    print(x, predict(x))
```

Output:

```python
1 0
2 0
3 1
4 1
5 1
```

---

Using Sigmoid:

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))

x = 4

z = 2*x - 6

p = sigmoid(z)

print(p)
```

Output:

```python
0.8808
```

---

Since:

[
p > 0.5
]

Predict:

[
1
]
