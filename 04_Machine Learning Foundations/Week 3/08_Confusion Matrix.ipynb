# 1. Theory

Suppose our classifier predicts whether an email is Spam (1) or Not Spam (0).

We compare

* Actual Label
* Predicted Label

There are only **four possible outcomes**.

---

## Case 1

Actual = 1

Prediction = 1

Correct.

This is called

# True Positive (TP)

The model correctly predicted a positive class.

Example

Spam email

↓

Predicted Spam

✅ Correct

---

## Case 2

Actual = 0

Prediction = 0

Correct.

Called

# True Negative (TN)

Example

Normal Email

↓

Predicted Normal

✅ Correct

---

## Case 3

Actual = 0

Prediction = 1

Wrong.

Called

# False Positive (FP)

The model predicted Positive

but actually it was Negative.

Example

Normal Email

↓

Predicted Spam

❌ Wrong

---

## Case 4

Actual = 1

Prediction = 0

Wrong.

Called

# False Negative (FN)

The model predicted Negative

but actually it was Positive.

Example

Spam Email

↓

Predicted Normal

❌ Wrong

---

# Why is it called "Confusion" Matrix?

Because it shows **where the model got confused.**

Example

The model thought

Healthy

↓

Disease

False Positive.

The model got confused.

---

# 4. Python Implementation

Using sklearn

```python
from sklearn.metrics import confusion_matrix

actual = [1,1,0,0,1,0]

predicted = [1,0,0,1,1,0]

cm = confusion_matrix(actual,predicted)

print(cm)
```

Output

```text
[[2 1]
 [1 2]]
```

---

## How to read this?

This confuses many beginners.

Sklearn prints

```text
           Pred 0   Pred 1

Actual 0     TN      FP

Actual 1     FN      TP
```

So

```text
[[TN FP]
 [FN TP]]
```

For our example

```text
[[2 1]
 [1 2]]
```

means

TN = 2

FP = 1

FN = 1

TP = 2

# Why is this useful?

Suppose Accuracy is

95%.

Looks great.

But Confusion Matrix may show

```text
TP = 0

FN = 100

FP = 0

TN = 900
```

Meaning

The model never detects disease.

Accuracy alone hides this.

Confusion Matrix exposes it.

# Key Takeaways

Always remember these four:

| Actual | Predicted | Name           |
| ------ | --------- | -------------- |
| 1      | 1         | True Positive  |
| 1      | 0         | False Negative |
| 0      | 1         | False Positive |
| 0      | 0         | True Negative  |

---

### Memory Trick

**True / False** → Was the prediction correct?

**Positive / Negative** → What did the model predict?

This simple rule lets you identify TP, TN, FP, and FN in seconds.
