# Week 1 → Topic 7: Batch Gradient Descent

In Topic 6, we learned **Gradient Descent**.

Now we'll learn a specific version called **Batch Gradient Descent**, which is what is usually taught first in Linear Regression.

---

## Recall the Goal

We want to find:

[
w,b
]

such that:

[
J(w,b)
]

is minimized.

Gradient Descent updates:

[
w := w-\alpha\frac{\partial J}{\partial w}
]

[
b := b-\alpha\frac{\partial J}{\partial b}
]

The question is:

> How do we calculate these gradients?

This depends on how much data we use in each update.

---

# What Does "Batch" Mean?

A **batch** is simply a group of training examples.

Suppose we have:

| x | y  |
| - | -- |
| 1 | 5  |
| 2 | 7  |
| 3 | 9  |
| 4 | 11 |

Total examples:

[
m=4
]

In **Batch Gradient Descent**, every update uses **all 4 training examples**.

---

## Process

### Step 1

Use current (w,b).

---

### Step 2

Predict for **every training example**.

| x | Actual y | Predicted y |
| - | -------- | ----------- |
| 1 | 5        | ?           |
| 2 | 7        | ?           |
| 3 | 9        | ?           |
| 4 | 11       | ?           |

---

### Step 3

Calculate error for all examples.

---

### Step 4

Calculate gradients using all examples.

---

### Step 5

Update (w,b).

---

### Step 6

Repeat.

---

# Why Is It Called Batch Gradient Descent?

Because:

```text
Entire Dataset
      ↓
One Gradient Calculation
      ↓
One Parameter Update
```

The whole dataset is treated as a single batch.

---

# Visualization

Dataset:

```text
1000 examples
```

Batch GD:

```text
1000 examples
      ↓
Compute Gradient
      ↓
Update Once
```

One update per full dataset pass.

---

# Mathematical Form

For Linear Regression:

Cost:

[
J(w,b)=\frac{1}{2m}
\sum_{i=1}^{m}
(\hat y_i-y_i)^2
]

Notice:

[
m
]

appears in the formula.

This means every training example contributes to the gradient.

The update is based on the entire dataset.

---

# Characteristics

### Stable

Because all data is used.

Gradient estimates are accurate.

Cost decreases smoothly.

```text
Smooth Path
 ↓
 ↓
 ↓
Minimum
```

---

### Deterministic

Same dataset.

Same initialization.

Same learning rate.

Produces identical updates every run.

---

### Accurate Gradient

Uses complete information.

The gradient truly represents the direction of maximum increase of the full cost function.

---

# Disadvantage

Suppose:

```text
100 examples
```

No issue.

---

Suppose:

```text
10,000,000 examples
```

Now every update requires processing:

```text
10,000,000 rows
```

before updating once.

Training becomes slow.

---

# Comparison with Other Gradient Descent Variants

### Batch Gradient Descent

Uses:

```text
All Training Examples
```

before updating.

---

### Stochastic Gradient Descent (SGD)

Uses:

```text
1 Example
```

before updating.

---

### Mini-Batch Gradient Descent

Uses:

```text
Small Group
```

before updating.

Example:

```text
32 examples
64 examples
128 examples
```

Most modern deep learning uses mini-batches.

---

# Why Do We Learn Batch GD First?

Because:

* Easier mathematically
* Easier to derive
* Smooth cost curve
* Introduces optimization concepts clearly

Andrew Ng's Linear Regression derivation is based on Batch Gradient Descent.

---

# Connection to Everything So Far

```text
Dataset
   ↓
Linear Model
   ↓
Predictions
   ↓
Cost Function
   ↓
Batch Gradient Descent
   ↓
Update w,b
   ↓
Lower Cost
```

Repeat until convergence.

---

# Key Takeaway

Batch Gradient Descent is simply Gradient Descent where:

> Every parameter update is computed using the entire training dataset.

```text
All Data
   ↓
Compute Gradient
   ↓
Update Parameters
   ↓
Repeat
```

It is:

* Stable
* Accurate
* Easy to understand
