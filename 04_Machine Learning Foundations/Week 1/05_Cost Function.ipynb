## What Is a Cost Function?

A Cost Function is a mathematical function that measures how bad the model's predictions are.

Input:

* Model predictions
* Actual values

Output:

* A single number representing total error

```text
Small Cost  → Good Model

Large Cost  → Bad Model
```

---

## What Does MSE Mean?

> On average, how large are the squared prediction errors?

## Connection to Linear Regression

Remember:

The model is trying to find:

**`w`**

and

**`b`**

The cost function converts every possible pair of ((w,b)) into a number:

**`J(w,b)`**

```text
(w,b)
   ↓
Predictions
   ↓
Cost Function
   ↓
Error Value
```

---

## Training Objective

The objective of Linear Regression is:



Meaning:

> Find the values of (w) and (b) that produce the smallest possible cost.

Everything that comes next (Gradient Descent) is simply the method used to minimize this cost.

---

## The Big Picture So Far

```text
Dataset
   ↓
Assume ŷ = wx + b
   ↓
Make Predictions
   ↓
Compute Cost (MSE)
   ↓
Adjust w and b
   ↓
Lower Cost
```

The cost function does **not** improve the model.

It only **measures how wrong the model currently is**.

The algorithm that improves the model is **Gradient Descent**, which is the next topic.

### Key Takeaway

Cost Function = Error Measurement

**`J(w,b)`**

takes the model's predictions and returns a single number.

* High Cost → Poor fit
* Low Cost → Good fit

The entire training process is ultimately trying to make this number as small as possible.
