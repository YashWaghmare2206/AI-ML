# Week 1 → Topic 3: Regression

Regression is a type of **supervised learning** where the target variable is a **continuous numerical value**.

---

## Formal Definition



Examples:

| Input               | Output          |
| ------------------- | --------------- |
| House Features      | ₹75,00,000      |
| Temperature Factors | 32.5°C          |
| Engine Data         | 18.4 km/L       |
| Advertising Spend   | ₹2,50,000 Sales |

Notice that the output is numeric and continuous.

---

## Why Is It Called Regression?

Historically, the term comes from statistics, specifically the work of Francis Galton on "regression toward the mean."

In ML, regression simply means:

> Predicting a continuous value.

---

# Regression vs Classification

### Regression

Output:



Examples:

```text
23.5
100000
78.9
```

---

### Classification

Output:

```text
Spam
Not Spam
```

or

```text
0
1
```

or

```text
Cat
Dog
Bird
```

Regression predicts numbers.

Classification predicts categories.

# Example: House Price Prediction

Dataset:

| Area | Price |
| ---- | ----- |
| 1000 | 50L   |
| 1200 | 60L   |
| 1500 | 75L   |

Input:

**`X = Area`**

Output:

**`Y = Price`**

The model learns a relationship:

**`Price = f(Area)`**

---

# What Does the Model Actually Learn?

The model learns a mathematical relationship between inputs and outputs.

Suppose data follows:

| Area | Price |
| ---- | ----- |
| 1    | 5     |
| 2    | 7     |
| 3    | 9     |
| 4    | 11    |

A model may learn:



This equation can then predict prices for unseen houses.

### Target (Y)

Mathematically:

**`X → Y`**

---

# Simple vs Multiple Regression

## Simple Regression

One feature.

**`Y = wX + b`**

Example:

```text
Area → Price
```

Only one input variable.

---

## Multiple Regression

Multiple features.



Example:

```text
Area
Bedrooms
Age
Location Score

↓
Price
```

You'll study this in Week 2.

# Regression Line

Suppose we plot:

| X | Y  |
| - | -- |
| 1 | 5  |
| 2 | 7  |
| 3 | 9  |
| 4 | 11 |

A straight line can describe the pattern.

That line is called the **regression line**.

Equation:

[
y = wx + b
]

Where:

* (w) = slope (weight)
* (b) = intercept (bias)

# Assumption Behind Linear Regression

Linear Regression assumes:

**`Y`**

changes approximately linearly with

**`X`**

Meaning:

```text
Increase X
↓
Y changes roughly proportionally
```

Not every dataset follows a linear relationship.

When relationships become more complex, we use:

* Polynomial Regression
* Decision Trees
* Random Forests
* Neural Networks
