### ROC in simple terms

1. The model predicts **probabilities** for each sample.
2. ROC changes the **classification threshold** from **1 → 0** (or equivalently evaluates all thresholds between 0 and 1).
3. At **each threshold**:

   * Convert probabilities into class labels (0 or 1).
   * Compute the **Confusion Matrix**.
   * Calculate:

     * **TPR (Recall)** = TP / (TP + FN)
     * **FPR** = FP / (FP + TN)
4. Plot the point **(FPR, TPR)**.
5. Connecting all these points gives the **ROC Curve**.

---

### What makes a good ROC curve?

A good classifier has points that are:

* **High** → High TPR (many positives detected)
* **Left** → Low FPR (few false alarms)

So the ideal curve hugs the **top-left corner**.

```
TPR
1.0 | ┌──────────
    | │
    | │
0.5 | │
    | │
0.0 +-------------------
    0      FPR      1
```

The diagonal line from **(0,0)** to **(1,1)** represents **random guessing**.

### One small correction to your statement

You wrote:

> "if all points left it means we have more correct predictions"

A more accurate statement is:

> **If the ROC curve stays close to the top-left corner, the model achieves a high True Positive Rate while keeping the False Positive Rate low, meaning it distinguishes the two classes well.**
