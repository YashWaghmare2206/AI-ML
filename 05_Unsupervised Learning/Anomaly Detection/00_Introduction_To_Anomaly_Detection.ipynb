## 1. What is an Anomaly?

### Intuition

An **anomaly** is an observation that behaves very differently from the pattern we normally see in the data.

Imagine a bank looking at one customer's transactions:

```text id="8v8h1c"
₹500
₹700
₹450
₹800
₹650
₹550
₹5,00,000  ← unusual
```

Most transactions are around a few hundred rupees.

The ₹5,00,000 transaction is dramatically different.

We would call it an **anomaly**.

So the basic idea is:

$$
\boxed{\text{Anomaly} = \text{An observation that is unusually different from the expected pattern}}
$$

---

### Why do we need anomaly detection?

In many datasets, unusual observations are exactly what we want to find.

For example:

**Banking**

```text
Normal transactions
        ↓
₹500, ₹700, ₹400, ₹800...
        ↓
Unusual ₹10,00,000 transaction
        ↓
Investigate
```

**Machine monitoring**

```text
Normal vibration
        ↓
20, 21, 19, 20, 21...
        ↓
Sudden vibration = 80
        ↓
Possible machine problem
```

**Cybersecurity**

```text
Normal network behavior
        ↓
Regular traffic pattern
        ↓
Sudden unusual traffic
        ↓
Possible attack
```

Therefore, anomaly detection helps us answer:

> **"Which observations do not look like the normal pattern?"**

---

### Important point: anomaly does NOT always mean error

Suppose a machine normally produces:

$$
20,\ 21,\ 19,\ 20,\ 21
$$

and suddenly produces:

$$
40
$$

The value 40 is unusual, but we don't immediately know why.

It could be:

* a faulty sensor,
* a machine failure,
* a legitimate unusual operating condition.

Therefore:

$$
\boxed{\text{Anomaly} \neq \text{Error}}
$$

Anomaly detection tells us:

> **"This observation is unusual and deserves attention."**

It does not automatically tell us the cause.

---

## Normal vs Anomalous Data

Now let's make the idea more precise.

Suppose we have:

$$
10,\ 11,\ 12,\ 10,\ 13,\ 11,\ 12,\ 10,\ 50
$$

Most observations lie between approximately:

$$
10 \rightarrow 13
$$

while:

$$
50
$$

is far away.

We can visualize the idea as:

```text
10  11  12  13
●   ●   ●   ●
●   ●   ●   ●

                                50
                                 ●
                                 ↑
                              Anomaly
```

The first eight observations form what we can call the **normal pattern**.

The observation 50 is an **anomalous observation**.

So conceptually:

$$
\text{Data}
\rightarrow
\text{Identify normal pattern}
\rightarrow
\text{Find observations that don't fit}
$$

---

### Normal Data

**Normal data** represents observations that are consistent with the dominant or expected behavior of the dataset.

For example:

$$
10,\ 11,\ 12,\ 10,\ 13,\ 11
$$

These observations are relatively close to one another.

---

### Anomalous Data

**Anomalous data** represents observations that significantly deviate from the expected pattern.

For example:

$$
50
$$

is very different from:

$$
10,\ 11,\ 12,\ 10,\ 13,\ 11
$$

---

### But "normal" is not always simply "close together"

This is very important.

Consider two-dimensional data:

```text
          y
          ↑
          |
      ● ● ●
     ● ● ● ●
      ● ● ●
          |
          |                    ●
          |                    ↑
          |                 unusual
          |
          +--------------------------→ x
```

The algorithm isn't necessarily asking:

> "Is this point far away from the entire dataset?"

Different algorithms use different definitions of **unusual**.

For example:

* Statistical methods may ask whether a value is statistically extreme.
* Isolation Forest asks whether a point can be isolated unusually quickly.
* One-Class SVM learns a boundary around normal data.
* LOF asks whether a point has much lower local density than its neighbors.

So the deeper idea is:

$$
\boxed{\text{Different algorithms define "unusual" in different ways}}
$$

This will become especially important when we study the individual algorithms.

---

## A useful mental model

For now, think of anomaly detection as:

```text id="6r8j3m"
                    DATA
                      │
                      ↓
              Find normal pattern
                      │
                      ↓
             Examine each observation
                      │
                      ↓
             Is it unusual?
                  /       \
                NO         YES
                ↓           ↓
             NORMAL      ANOMALY
```

At this stage, don't worry about **how** the normal pattern is found.


## 2. Normal vs Anomalous Data

### Intuition

Before detecting an anomaly, we need to understand one basic question:

> **What does "normal" mean?**

Suppose we are monitoring the temperature of a machine.

For several hours, we observe:

$$
50,\ 51,\ 49,\ 52,\ 50,\ 51,\ 49
$$

These values are all around the same region.

We can consider this behavior **normal**.

Now suppose we observe:

$$
95
$$

That value is very different from the previous observations.

It may be an anomaly.

So conceptually:

```text
Normal observations
      ↓
50, 51, 49, 52, 50, 51, 49
      ↓
Follow the expected pattern


Anomalous observation
      ↓
95
      ↓
Does not follow the expected pattern
```

The important idea is:

$$
\boxed{\text{Normal} = \text{consistent with the expected pattern}}
$$

$$
\boxed{\text{Anomalous} = \text{significantly inconsistent with the expected pattern}}
$$

---

### Normal data is usually a pattern, not a single value

Consider:

$$
10,\ 11,\ 12,\ 13,\ 11,\ 12
$$

There isn't one exact "normal" value.

Instead, there is a **region of normal behavior**:

```text
        Normal region

     10  11  12  13
      ●   ●   ●   ●
       ●   ●   ●


                         50
                          ●
                       Anomaly
```

This distinction becomes very important in machine learning.

We generally don't want to memorize:

> "Normal = 11"

Instead, we want to understand:

> "Values around this region are typical."

---

## Mathematics

Suppose our dataset contains \(n\) observations:

$$
x_1,x_2,\ldots,x_n
$$

where:

* \(x_i\) = the \(i\)-th observation
* \(n\) = total number of observations

An anomaly detector can conceptually assign each observation an **anomaly score**:

$$
s(x_i)
$$

where:

* \(s(x_i)\) represents how unusual \(x_i\) is.

The exact meaning of this score depends on the algorithm.

For example, we might have:

$$
s(x_i)=0.1
$$

for one observation and:

$$
s(x_j)=0.9
$$

for another.

If a larger score means "more unusual," then:

$$
0.1 \rightarrow \text{more normal}
$$

$$
0.9 \rightarrow \text{more anomalous}
$$

We can then use a threshold \(T\):

$$
s(x_i) > T
\rightarrow
\text{Anomaly}
$$

$$
s(x_i) \leq T
\rightarrow
\text{Normal}
$$

Here:

* \(s(x_i)\) = anomaly score of observation \(x_i\)
* \(T\) = decision threshold

**Important:** not every algorithm uses a score where larger means more anomalous. The direction depends on the method and implementation.

---

## Tiny Example

Suppose an algorithm produces these scores:

| Observation | Score |
| ----------: | ----: |
|          10 |  0.10 |
|          11 |  0.08 |
|          12 |  0.12 |
|          11 |  0.09 |
|          50 |  0.95 |

Suppose:

$$
T=0.50
$$

For \(x=10\):

$$
0.10 \leq 0.50
$$

so it is **Normal**.

For \(x=50\):

$$
0.95 > 0.50
$$

so it is an **Anomaly**.

The workflow is therefore:

$$
\text{Observation}
\rightarrow
\text{Anomaly Score}
\rightarrow
\text{Compare with Threshold}
\rightarrow
\text{Normal / Anomaly}
$$

The important thing is that **the algorithm must first determine what makes an observation unusual**. The score is simply a way of representing that unusualness.

---

### A more important example: anomalies can be close to other points

Consider this dataset:

```text
● ● ● ● ● ●
        ●
```

The bottom point might be an anomaly even though it isn't extremely far away.

This is why anomaly detection becomes more interesting in multiple dimensions and in datasets with different local patterns.

For example, later **LOF** will detect anomalies by asking:

> "Is this point in a region that is much less dense than the region around its neighbors?"

So we should not think:

$$
\boxed{\text{Anomaly} = \text{Always the farthest point}}
$$

Instead:

$$
\boxed{\text{Anomaly} = \text{Point that violates the relevant normal pattern}}
$$

The definition of that **normal pattern** is what differs between anomaly-detection algorithms.


## 3. Types of Anomalies

Not every anomaly looks the same.

There are **three important types**:

1. **Point Anomaly**
2. **Contextual Anomaly**
3. **Collective Anomaly**

The key difference is **what makes the observation unusual**.

---

### 3.1 Point Anomaly

#### Intuition

A **point anomaly** is a single observation that is significantly different from the rest of the data.

Example: daily ATM withdrawal amounts:

$$
500,\ 700,\ 600,\ 800,\ 550,\ 650,\ 5000
$$

Most values are around a few hundred rupees.

But:

$$
5000
$$

is unusually large.

So the individual point itself is suspicious.

```text id="w7u0d3"
500   550   600   650   700   800
 ●     ●     ●     ●     ●     ●

                                      5000
                                        ●
                                        ↑
                                    Anomaly
```

#### Key idea

The anomaly can be identified by looking at the **individual observation** relative to the rest of the data.

$$
\boxed{\text{Point Anomaly} = \text{One observation is significantly different}}
$$

---

### Tiny Example

Suppose:

$$
X = \{10,11,12,10,13,11,50\}
$$

The value:

$$
50
$$

is far away from the other observations.

Therefore:

$$
50 \rightarrow \text{Point Anomaly}
$$

The important thing is that **we don't need a sequence or special context** to recognize it as unusual.

---

## 3.2 Contextual Anomaly

#### Intuition

A **contextual anomaly** is an observation that is unusual **only when we consider its context**.

This is more subtle.

Consider temperature.

Suppose the temperature is:

$$
35^\circ C
$$

Is that an anomaly?

Not necessarily.

In Mumbai during a hot afternoon, it might be completely normal.

But:

$$
35^\circ C
$$

at **3 AM in winter** could be highly unusual.

The value itself hasn't changed.

The **context** has changed.

```text id="5g1n6j"
Temperature = 35°C

Afternoon
   ↓
Normal


3 AM in winter
   ↓
Potential anomaly
```

So:

$$
\boxed{\text{Contextual Anomaly} =
\text{Normal/abnormal depends on context}}
$$

---

### What is "context"?

Context is additional information that tells us **under what circumstances the observation occurred**.

For example:

* Time
* Location
* Season
* User
* Operating condition
* Day of week

Suppose we monitor website traffic.

```text id="8i0jyc"
10,000 visitors

During daytime
      ↓
Normal

At 3 AM
      ↓
Potential anomaly
```

The number 10,000 isn't inherently anomalous.

Its interpretation depends on **when** it occurred.

---

### Point vs Contextual Anomaly

| Type               | What makes it unusual?                 |
| ------------------ | -------------------------------------- |
| Point anomaly      | The point itself is unusual            |
| Contextual anomaly | The point is unusual given its context |

A useful mental distinction:

```text id="wz2p4m"
Point anomaly
    ↓
"Is this value unusual?"


Contextual anomaly
    ↓
"Is this value unusual HERE / NOW / UNDER THESE CONDITIONS?"
```

---

## 3.3 Collective Anomaly

#### Intuition

A **collective anomaly** occurs when a **group or sequence of observations** behaves unusually, even though the individual observations may look normal by themselves.

Consider network traffic:

```text id="k8l7d4"
Normal:
10, 11, 9, 12, 10, 11

Suspicious sequence:
10, 11, 10, 12, 11, 10
10, 11, 12, 11, 10...
```

Each individual value might look normal.

But suppose these values represent a system sending requests, and suddenly there is a particular repeated pattern over a short period.

The **sequence as a whole** may indicate suspicious behavior.

That is a collective anomaly.

---

### Simple example

Imagine a heartbeat sensor.

Normal heartbeat pattern:

```text id="8jq4h1"
72 → 75 → 73 → 74 → 72 → 76
```

Now imagine:

```text id="7j3n4d"
72 → 74 → 73 → 72 → 74 → 73
```

Each value individually looks perfectly reasonable.

But if the **pattern of the sequence** is abnormal for the system, the sequence can be considered anomalous.

Therefore:

$$
\boxed{\text{Collective Anomaly} =
\text{A group of observations is anomalous as a whole}}
$$

---

## The Three Types Together

```text id="0n5y4k"
                    ANOMALIES
                        │
          ┌─────────────┼─────────────┐
          ↓             ↓             ↓
       POINT       CONTEXTUAL      COLLECTIVE
          │             │             │
     One point       Context        Group/sequence
      unusual        matters          unusual
          │             │             │
        ₹5000       35°C at 3AM    Suspicious
                                   sequence
```

### The easiest way to remember them

**Point:**

> "This point is strange."

**Contextual:**

> "This point is strange **in this situation**."

**Collective:**

> "These points are strange **together**."

These distinctions matter because different anomaly-detection methods are better suited to different kinds of unusual behavior.


## 4. Anomaly Detection vs Classification

These two problems can look similar because both eventually produce labels such as **Normal** or **Anomaly**, but they are fundamentally different.

### Intuition

Suppose a bank wants to identify fraudulent transactions.

There are two possible situations.

### Situation 1 — Classification

The bank already has historical transactions labeled:

```text
Transaction     Label
₹500            Normal
₹700            Normal
₹50,000         Fraud
₹1,20,000       Fraud
₹800            Normal
```

The model can learn:

$$
\text{Transaction features} \rightarrow \text{Known class}
$$

This is **classification**.

The important point is:

> **We already know the correct labels during training.**

---

### Situation 2 — Anomaly Detection

Now imagine the bank has millions of transactions but **doesn't have reliable fraud labels**.

It wants to discover transactions that behave differently from the normal pattern.

```text
Millions of transactions
          ↓
Mostly normal behavior
          ↓
Learn normal pattern
          ↓
Find unusual transactions
          ↓
Potential anomalies
```

This is **anomaly detection**.

The model isn't necessarily learning:

> "Here are examples of fraud."

Instead, it is trying to determine:

> "What does normal behavior look like, and which observations don't fit it?"

---

## The Fundamental Difference

### Classification

Classification learns from **known classes**.

$$
\boxed{
\text{Features} + \text{Known Labels}
\rightarrow
\text{Learn Classes}
}
$$

For example:

$$
X \rightarrow \{\text{Normal},\text{Fraud}\}
$$

where the training dataset explicitly tells the model which observations are normal and which are fraudulent.

---

### Anomaly Detection

Anomaly detection generally starts without explicit labels for every observation.

$$
\boxed{
\text{Mostly Unlabeled Data}
\rightarrow
\text{Identify Unusual Observations}
}
$$

The model may produce something like:

$$
\text{Normal} \quad \text{or} \quad \text{Anomaly}
$$

but these labels are **decisions produced by the detection method**, rather than pre-existing class labels.

---

## Small Example

Suppose we have transaction amounts:

$$
500,\ 600,\ 550,\ 700,\ 650,\ 5000
$$

### Classification

Imagine we already know:

$$
500 \rightarrow \text{Normal}
$$

$$
600 \rightarrow \text{Normal}
$$

$$
550 \rightarrow \text{Normal}
$$

$$
700 \rightarrow \text{Normal}
$$

$$
650 \rightarrow \text{Normal}
$$

$$
5000 \rightarrow \text{Fraud}
$$

The algorithm learns from these **known labels**.

---

### Anomaly Detection

Instead, suppose the labels are unavailable:

$$
500,\ 600,\ 550,\ 700,\ 650,\ 5000
$$

The algorithm observes that most values are around:

$$
500 \rightarrow 700
$$

while:

$$
5000
$$

is very different.

It can therefore flag:

$$
5000 \rightarrow \text{Anomaly}
$$

without having been explicitly told:

> "5000 is fraud."

---

## Another Important Difference

Classification usually assumes that the classes we care about are represented in the training data.

For example:

```text id="7x9p3q"
Training data

Normal
Normal
Fraud
Normal
Fraud
Normal
```

The model learns the distinction between the known classes.

But anomalies are often:

* rare,
* previously unseen,
* constantly changing,
* difficult to label.

For example, a cybersecurity system may encounter a completely new attack.

There may be **no labeled example of that attack** in the training data.

Anomaly detection can still potentially identify it if its behavior differs sufficiently from normal behavior.

---

## Side-by-Side Comparison

| Property         | Classification          | Anomaly Detection                              |
| ---------------- | ----------------------- | ---------------------------------------------- |
| Training labels  | Usually required        | Often unavailable                              |
| Main goal        | Predict known classes   | Find unusual observations                      |
| Typical question | "Which class is this?"  | "Is this unusual?"                             |
| Rare events      | Need labeled examples   | Can potentially detect unseen unusual behavior |
| Output           | Class label/probability | Anomaly score/decision                         |
| Example          | Spam vs Not Spam        | Unusual network activity                       |

---

### Important conceptual distinction

Classification asks:

$$
\boxed{\text{"Which known category does this belong to?"}}
$$

Anomaly detection asks:

$$
\boxed{\text{"Does this observation behave unusually?"}}
$$

These questions are related, but they are **not the same problem**.

And importantly, an anomaly detector saying:

$$
\text{Anomaly}
$$

does **not automatically mean**:

$$
\text{Fraud}
$$

It means the observation is unusual according to the detector.

A human or another system may still need to investigate what that unusual behavior actually represents.


## 5. Unsupervised vs Semi-supervised Anomaly Detection

The key difference here is **what information about normal/anomalous data we have during training**.

---

### 5.1 Unsupervised Anomaly Detection

#### Intuition

Suppose you receive a dataset like this:

```text id="q2k8sx"
Transaction Amount

500
650
700
550
600
520
5000
```

But you have **no labels**.

You don't know:

* which transactions are normal,
* which are fraudulent,
* or even how many anomalies exist.

The algorithm must discover unusual observations **from the data itself**.

That is **unsupervised anomaly detection**.

Conceptually:

```text id="nq8v3k"
Unlabeled Data
      ↓
Find structure / pattern
      ↓
Determine unusual observations
      ↓
Anomaly / Normal
```

This is useful because in many real-world datasets, obtaining anomaly labels is difficult.

For example, a company may have millions of network requests but only a tiny number of confirmed attacks.

---

### 5.2 Semi-supervised Anomaly Detection

Now consider a different situation.

Suppose we have a large dataset of **known normal behavior**:

```text id="q8gk2h"
Normal data

● ● ● ●
 ● ● ● ●
● ● ● ●
```

We don't have many examples of anomalies.

Instead, we train the model primarily using **normal data**.

The model learns:

> "This is what normal behavior looks like."

Then, when a new observation arrives:

```text id="7k3p2r"
Training:

Normal data
     ↓
Learn normal pattern
     ↓
New observation
     ↓
Does it fit?
   /       \
 YES       NO
 ↓          ↓
Normal    Anomaly
```

This is commonly called **semi-supervised anomaly detection**.

---

### Why "semi-supervised"?

Because we have **some information about the desired behavior**, usually:

$$
\boxed{\text{Known Normal Data}}
$$

but we don't have a complete set of labels such as:

```text
Normal
Normal
Anomaly
Normal
Anomaly
...
```

Instead:

```text
Normal examples
      ↓
Learn what normal looks like
      ↓
Anything sufficiently different
      ↓
Potential anomaly
```

---

## Small Example

Suppose a machine normally operates at temperatures:

$$
50,\ 51,\ 49,\ 52,\ 50,\ 51
$$

We can train a model using these normal observations.

Now a new temperature arrives:

$$
90
$$

The model asks:

> "Does 90 resemble the normal operating behavior?"

If not:

$$
90 \rightarrow \text{Anomaly}
$$

Notice something important:

We did **not** need examples such as:

$$
90 \rightarrow \text{Anomaly}
$$

during training.

We only provided examples of normal behavior.

---

## Unsupervised vs Semi-supervised

|               | Unsupervised               | Semi-supervised               |
| ------------- | -------------------------- | ----------------------------- |
| Labels        | No labels                  | Usually normal examples       |
| Training data | Mixed/unknown data         | Primarily normal data         |
| Main idea     | Discover unusual structure | Learn normal behavior         |
| Question      | "What looks unusual?"      | "Does this look like normal?" |
| Example       | Isolation Forest           | One-Class SVM                 |

There is an important terminology point:

In practical machine learning literature, **One-Class SVM is often described as a one-class or novelty-detection method**, and depending on the training setup it can be used in a semi-supervised/novelty-detection setting. Isolation Forest is commonly used as an unsupervised anomaly detector.

---

### The conceptual difference

Think of two people trying to identify suspicious people at an airport.

**Unsupervised approach:**

> "I don't know who is suspicious. I'll examine everyone's behavior and look for unusual patterns."

**Semi-supervised approach:**

> "Here are many examples of normal passenger behavior. Learn what normal looks like, then flag people who don't fit."

So:

$$
\boxed{
\text{Unsupervised}
\rightarrow
\text{Discover unusual patterns from unlabeled data}
}
$$

$$
\boxed{
\text{Semi-supervised}
\rightarrow
\text{Learn normal behavior and detect deviations}
}
$$


## 6. General Anomaly Detection Workflow

Now let's put the ideas together into one general process.

The exact internal mechanism changes between algorithms, but the overall workflow is similar.

### Step 1 — Input Data

We start with observations.

For example, suppose we are monitoring machines using two measurements:

* Temperature
* Vibration

Our data might look like:

$$
\begin{bmatrix}
50 & 20 \\
52 & 21 \\
49 & 19 \\
51 & 20 \\
90 & 80
\end{bmatrix}
$$

Here:

* Each **row** represents one observation.
* The first column represents temperature.
* The second column represents vibration.

The last observation is potentially unusual because both measurements are much higher.

---

### Step 2 — Understand the Normal Pattern

The algorithm tries to determine what behavior is typical.

Different algorithms do this differently.

For example:

```text id="f4c0qh"
Statistical Method
      ↓
How far from expected statistics?


Isolation Forest
      ↓
How easily can the point be isolated?


One-Class SVM
      ↓
Does the point lie inside the normal boundary?


LOF
      ↓
Is the point's local density unusually low?
```

This is one of the most important ideas in the entire topic:

$$
\boxed{\text{Different algorithms learn "normal" differently}}
$$

---

### Step 3 — Calculate an Anomaly Score

The algorithm evaluates an observation and produces some measure of how unusual it is.

Conceptually:

$$
x_i
\rightarrow
s(x_i)
$$

where:

* \(x_i\) = observation
* \(s(x_i)\) = anomaly-related score

The meaning and direction of this score depend on the algorithm.

For some methods:

$$
\text{larger score} \rightarrow \text{more anomalous}
$$

For others, the interpretation may be reversed.

So we should **never memorize "high score = anomaly" without checking the specific algorithm**.

---

### Step 4 — Apply a Decision Rule

The score is converted into a final decision.

Conceptually:

$$
\text{Score}
\rightarrow
\text{Decision Rule}
\rightarrow
\begin{cases}
\text{Normal}\\
\text{Anomaly}
\end{cases}
$$

For example, if larger scores mean more anomalous:

$$
s(x_i)>T
\rightarrow
\text{Anomaly}
$$

where \(T\) is the threshold.

---

### Step 5 — Investigate the Anomaly

An anomaly detector usually **flags** an observation.

It doesn't necessarily explain the cause.

For example:

```text id="8qk3jm"
Machine Data
     ↓
Anomaly Detector
     ↓
Machine #17 = Anomaly
     ↓
Investigation
     ↓
Possible overheating
```

This distinction is important:

$$
\boxed{\text{Detection} \neq \text{Diagnosis}}
$$

The detector tells us:

> "Something unusual happened."

Further analysis determines:

> "Why did it happen?"

---

# Complete Workflow

Putting everything together:

```text id="8i6y8q"
                    INPUT DATA
                        ↓
                Data Preparation
                        ↓
              Understand Normal
                   Pattern
                        ↓
              Apply Detection
                  Algorithm
                        ↓
              Anomaly Score
                        ↓
               Decision Rule
                        ↓
                 ┌──────┴──────┐
                 ↓             ↓
              NORMAL       ANOMALY
                               ↓
                         Investigation
```

### Example

Suppose our machine produces:

```text id="d8f1ak"
Temperature    Vibration

50             20
52             21
49             19
51             20
90             80
```

The workflow is:

$$
\text{Machine Data}
\rightarrow
\text{Learn/identify normal behavior}
\rightarrow
\text{Evaluate each observation}
\rightarrow
\text{Anomaly Score}
\rightarrow
\text{Decision}
$$

The final observation might be flagged:

$$
(90,80)
\rightarrow
\boxed{\text{Anomaly}}
$$

But **why** it is flagged depends entirely on the algorithm.

---

## One Final Concept for This Notebook

You have now seen the basic idea behind anomaly detection:

$$
\boxed{
\text{Data}
\rightarrow
\text{Normal Pattern}
\rightarrow
\text{Unusualness}
\rightarrow
\text{Decision}
}
$$

The four methods we study next will implement this idea in very different ways:

| Method              | Core question                                                 |
| ------------------- | ------------------------------------------------------------- |
| Statistical Methods | **How statistically unusual is this point?**                  |
| Isolation Forest    | **How easily can this point be isolated?**                    |
| One-Class SVM       | **Does this point fall outside the learned normal boundary?** |
| LOF                 | **Is this point much less dense than its neighbors?**         |
