## 1. Statistical Intuition for Anomaly Detection

Since you already know **mean, variance, standard deviation, Z-score, and IQR**, we'll focus on how these ideas become **anomaly-detection methods** rather than relearning their basic definitions.

### The core idea

Suppose we have:

$$
10,\ 11,\ 12,\ 10,\ 13,\ 11,\ 12,\ 10,\ 50
$$

We can visually see that \(50\) is unusual.

But a computer needs a **measurable definition of unusual**.

Statistics gives us a way to describe the normal data and then measure how much an observation deviates from it.

The basic reasoning is:

```text id="4kj8m2"
Data
 ↓
Describe typical behavior
 ↓
Measure deviation of each observation
 ↓
Is the deviation unusually large?
 ↓
Anomaly
```

---

## Why deviation matters

Suppose the typical value is around:

$$
\mu = 11
$$

Consider two observations:

$$
x_1=12
$$

and

$$
x_2=50
$$

Their deviations from the mean are:

$$
|12-11|=1
$$

and

$$
|50-11|=39
$$

Clearly, \(50\) is much more unusual.

But there's a problem.

### Is a deviation of 10 large?

We cannot answer that just from the deviation itself.

Imagine two datasets.

**Dataset A:**

$$
9,\ 10,\ 11,\ 12,\ 13
$$

A deviation of \(10\) from the center would be enormous.

**Dataset B:**

$$
0,\ 5,\ 10,\ 15,\ 20
$$

A deviation of \(10\) might be completely normal.

So anomaly detection needs to consider not just:

$$
\text{How far is the point from the center?}
$$

but also:

$$
\text{How spread out is normal data?}
$$

This is why **standard deviation** becomes useful.

---

## From distance to standardized distance

Suppose:

$$
\mu = 50
$$

and:

$$
\sigma = 5
$$

Now consider:

$$
x=60
$$

The raw deviation is:

$$
x-\mu=60-50=10
$$

But relative to the standard deviation:

$$
\frac{x-\mu}{\sigma}
=
\frac{10}{5}
=
2
$$

So \(x=60\) is **2 standard deviations above the mean**.

This is what the Z-score gives us.

$$
\boxed{
z=\frac{x-\mu}{\sigma}
}
$$

You already know the formula. The important anomaly-detection interpretation is:

> **Z-score converts raw distance into distance relative to the typical spread of the data.**

That makes values much easier to judge.

---

## Why this is useful

Compare:

### Dataset A

$$
\mu=50,\quad \sigma=5
$$

and:

$$
x=60
$$

Then:

$$
z=2
$$

### Dataset B

$$
\mu=50,\quad \sigma=20
$$

and:

$$
x=60
$$

Then:

$$
z=0.5
$$

The observation is the same distance from the mean:

$$
|60-50|=10
$$

But its **relative unusualness** is different.

```text id="j3p5q8"
Dataset A
spread = small
      ↓
10 is relatively large
      ↓
z = 2


Dataset B
spread = large
      ↓
10 is relatively small
      ↓
z = 0.5
```

This is the key statistical intuition.

---

## How do we turn this into an anomaly detector?

We need a **decision threshold**.

For example, suppose we decide:

$$
|z|>3
\rightarrow
\text{Anomaly}
$$

and:

$$
|z|\leq3
\rightarrow
\text{Normal}
$$

Why absolute value?

Because both directions can be unusual.

For example:

$$
z=4
$$

means the observation is far **above** the mean.

While:

$$
z=-4
$$

means it is far **below** the mean.

Both are potentially anomalous.

Therefore:

$$
\boxed{|z| \text{ measures how many standard deviations away the point is}}
$$

---

## Tiny Numerical Example

Suppose:

$$
\mu=100
$$

$$
\sigma=10
$$

and we observe:

$$
x=135
$$

Calculate:

$$
z=\frac{x-\mu}{\sigma}
$$

Substitute:

$$
z=\frac{135-100}{10}
$$

$$
z=\frac{35}{10}
$$

$$
z=3.5
$$

So:

$$
|z|=3.5
$$

If our chosen threshold is:

$$
T=3
$$

then:

$$
3.5>3
$$

Therefore:

$$
\boxed{x=135\rightarrow\text{Anomaly}}
$$

### Interpretation

The important interpretation isn't simply:

> "The Z-score is 3.5."

It is:

> **The observation is 3.5 standard deviations above the mean, so according to our chosen threshold it is unusually far from the typical data.**

---

## But there is a major limitation

This method assumes that **mean and standard deviation provide a useful description of normal behavior**.

What if the data contains a huge outlier?

Consider:

$$
10,\ 11,\ 10,\ 12,\ 11,\ 10,\ 100
$$

The \(100\) is clearly unusual.

But \(100\) can influence:

* the mean
* the standard deviation

This can make the detector less effective.

That's one reason we also have the **IQR method**, which relies on percentiles and is generally more robust to extreme values.

So we now have the basic statistical detection philosophy:

$$
\boxed{
\text{Typical location}
+
\text{Typical spread}
\rightarrow
\text{Measure unusualness}
}
$$

And the next subsection will apply this specifically to **Z-score anomaly detection**, including how thresholds such as \(2\) and \(3\) should actually be interpreted.


## 2. Z-Score Method

You already know how to calculate a Z-score, so we'll focus on **using it as an anomaly detector**.

### 2.1 Core idea

For every observation \(x\), calculate:

$$
z=\frac{x-\mu}{\sigma}
$$

where:

* \(x\) = observation
* \(\mu\) = mean of the data
* \(\sigma\) = standard deviation

The Z-score tells us how far the observation is from the mean **in units of standard deviation**.

Then we use:

$$
|z| > T
$$

to determine whether the observation is anomalous.

Here \(T\) is our chosen threshold.

---

### 2.2 Choosing the threshold

Common choices are:

$$
T=2
$$

or

$$
T=3
$$

For example:

$$
|z|>3 \rightarrow \text{Anomaly}
$$

means:

> We consider an observation anomalous if it is more than 3 standard deviations away from the mean.

A larger threshold means we are **less sensitive**.

```text id="4b1yq6"
Threshold = 2
     ↓
More points can be flagged
     ↓
More sensitive


Threshold = 3
     ↓
Only more extreme points flagged
     ↓
Less sensitive
```

So:

$$
T\uparrow
\rightarrow
\text{fewer anomalies detected}
$$

and:

$$
T\downarrow
\rightarrow
\text{more anomalies detected}
$$

---

### 2.3 Tiny numerical example

Suppose:

$$
\mu=50
$$

$$
\sigma=5
$$

and we have three observations:

$$
x_1=55,\qquad x_2=65,\qquad x_3=35
$$

Let's use:

$$
T=3
$$

#### Observation 1

$$
z_1=\frac{55-50}{5}
$$

$$
z_1=1
$$

Therefore:

$$
|z_1|=1<3
$$

$$
\boxed{x_1=55\rightarrow\text{Normal}}
$$

---

#### Observation 2

$$
z_2=\frac{65-50}{5}
$$

$$
z_2=3
$$

Therefore:

$$
|z_2|=3
$$

If our rule is strictly:

$$
|z|>3
$$

then:

$$
\boxed{x_2=65\rightarrow\text{Normal}}
$$

It is **right at the threshold**, not beyond it.

If we instead define the rule as:

$$
|z|\geq3
$$

then it would be flagged.

This shows that the exact threshold rule matters.

---

#### Observation 3

$$
z_3=\frac{35-50}{5}
$$

$$
z_3=-3
$$

Again:

$$
|z_3|=3
$$

So with the strict \(>3\) rule, it is not flagged.

---

### 2.4 A clearer example

Let's instead use:

$$
x=70
$$

Then:

$$
z=\frac{70-50}{5}
$$

$$
z=4
$$

Therefore:

$$
|z|=4>3
$$

So:

$$
\boxed{70\rightarrow\text{Anomaly}}
$$

Interpretation:

> \(70\) is **4 standard deviations above the mean**, which exceeds our chosen threshold of 3.

---

## 2.5 Complete workflow

```text id="j4t7k2"
Input observations
       ↓
Calculate mean μ
       ↓
Calculate standard deviation σ
       ↓
Calculate Z-score for each x
       ↓
Take |z|
       ↓
Compare with threshold T
       ↓
 ┌─────────────────┐
 │                 │
|z| ≤ T          |z| > T
 │                 │
 ↓                 ↓
Normal          Anomaly
```

So the Z-score itself is a **measurement of unusualness**.

The threshold converts that measurement into a **decision**.

$$
\boxed{
\text{Z-score}
\rightarrow
\text{Measure unusualness}
}
$$

$$
\boxed{
\text{Threshold}
\rightarrow
\text{Make the anomaly decision}
}
$$


## 3. IQR Method

You already know how to calculate \(Q_1\), \(Q_3\), and IQR, so we'll focus on **why IQR works for anomaly detection and why it can be preferable to Z-score**.

### Intuition

Suppose we have:

$$
10,\ 11,\ 12,\ 12,\ 13,\ 14,\ 15,\ 16,\ 50
$$

We want to identify the unusual value.

The Z-score approach uses:

$$
\text{Mean + Standard Deviation}
$$

But mean and standard deviation can be strongly affected by extreme values.

The IQR method takes a different approach:

> **Instead of describing the data using its mean and standard deviation, describe the central 50% of the data.**

The central 50% is the region between:

$$
Q_1
$$

and:

$$
Q_3
$$

So visually:

```text id="y3t9kx"
        Lower 25%       Middle 50%       Upper 25%

───────────────┬──────────────────────┬───────────────
               Q1                     Q3
               │<------ IQR ------>│
```

The middle 50% is much less affected by extreme observations.

---

## Why is this useful?

Imagine:

$$
10,\ 11,\ 12,\ 13,\ 14
$$

Now add an extreme value:

$$
1000
$$

The mean moves significantly toward 1000, and the standard deviation becomes much larger.

But the **middle portion of the data** is still around:

$$
10 \rightarrow 14
$$

That's why IQR is called a **robust** measure.

> Robust means that extreme observations have relatively little influence on the statistic.

---

## Mathematics

You already know:

$$
IQR=Q_3-Q_1
$$

where:

* \(Q_1\) = 25th percentile
* \(Q_3\) = 75th percentile
* \(IQR\) = range containing the middle 50% of observations

We then construct two boundaries.

### Lower bound

$$
L=Q_1-1.5(IQR)
$$

### Upper bound

$$
U=Q_3+1.5(IQR)
$$

Here:

* \(L\) = lower boundary
* \(U\) = upper boundary

The anomaly rule becomes:

$$
\boxed{x<L \rightarrow \text{Anomaly}}
$$

or:

$$
\boxed{x>U \rightarrow \text{Anomaly}}
$$

Otherwise:

$$
L\leq x\leq U
\rightarrow
\text{Normal}
$$

---

## Tiny Numerical Example

Consider:

$$
10,\ 11,\ 12,\ 13,\ 14,\ 15,\ 16,\ 17
$$

Suppose we obtain:

$$
Q_1=11.5
$$

and:

$$
Q_3=15.5
$$

Therefore:

$$
IQR=15.5-11.5=4
$$

Now calculate the boundaries.

### Lower boundary

$$
L=11.5-1.5(4)
$$

$$
L=11.5-6
$$

$$
L=5.5
$$

### Upper boundary

$$
U=15.5+1.5(4)
$$

$$
U=15.5+6
$$

$$
U=21.5
$$

Therefore our normal region according to the IQR rule is:

$$
5.5\leq x\leq21.5
$$

Now suppose we receive:

$$
x=30
$$

Since:

$$
30>21.5
$$

we classify:

$$
\boxed{30\rightarrow\text{Anomaly}}
$$

But:

$$
x=15
$$

satisfies:

$$
5.5\leq15\leq21.5
$$

so:

$$
\boxed{15\rightarrow\text{Normal}}
$$

---

## Why \(1.5\times IQR\)?

This is an important question.

The \(1.5\) is **not derived from a universal law saying that every value beyond \(1.5IQR\) is an anomaly**.

It is a widely used **rule of thumb**, originating from the traditional boxplot convention.

It creates a boundary sufficiently outside the central 50% that observations beyond it are treated as potential outliers.

So:

$$
\boxed{1.5\times IQR=\text{chosen practical cutoff}}
$$

not:

$$
\boxed{1.5\times IQR=\text{mathematical definition of an anomaly}}
$$

This distinction is important.

---

## Z-score vs IQR

Now you can see the fundamental difference:

### Z-score

```text id="q8x5r2"
Mean
 ↓
Standard Deviation
 ↓
How many SDs away?
 ↓
Threshold
 ↓
Anomaly
```

### IQR

```text id="7f3m2a"
Q1 ───────────── Q3
       Middle 50%
             ↓
        Calculate IQR
             ↓
     Create boundaries
             ↓
   Outside boundaries?
        /          \
      No            Yes
      ↓              ↓
   Normal         Anomaly
```

The biggest practical difference is:

$$
\boxed{\text{Z-score uses mean + SD}}
$$

while:

$$
\boxed{\text{IQR uses percentiles}}
$$

Because percentiles are less affected by extreme values, **IQR is generally more robust when the data contains strong outliers or is skewed**.

One caveat: the exact numerical values of quartiles can depend on the percentile calculation convention/software, so different implementations can occasionally give slightly different boundaries.


## 4. Percentile-Based Detection

### Intuition

The IQR method already uses **percentiles**. But we can use percentiles directly for anomaly detection without calculating IQR.

The basic idea is very simple:

> **Define the normal region using extreme percentiles, and flag observations outside that region.**

Suppose we have a dataset of transaction amounts.

We could say:

> The lowest 1% and highest 1% of transactions are unusual.

Then:

```text id="7r5m2k"
        Bottom 1%       Normal 98%       Top 1%
             │<---------------------->│
             ↓                         ↓
          Anomaly                   Anomaly
```

So instead of asking:

> "How many standard deviations away is this?"

we ask:

> **"Is this observation in an extremely rare percentile of the data?"**

---

## Mathematics

Let:

* \(P_{\alpha}\) = lower percentile
* \(P_{1-\alpha}\) = upper percentile
* \(\alpha\) = chosen tail proportion

For example, if:

$$
\alpha=0.01
$$

then:

$$
P_{\alpha}=P_{1\%}
$$

and:

$$
P_{1-\alpha}=P_{99\%}
$$

We can define:

$$
L=P_1
$$

$$
U=P_{99}
$$

Then:

$$
x<L
\rightarrow
\text{Anomaly}
$$

or:

$$
x>U
\rightarrow
\text{Anomaly}
$$

So the middle:

$$
1\%\rightarrow99\%
$$

is treated as the normal region.

---

## Tiny Numerical Example

Suppose we have many transaction values and calculate:

$$
P_1=100
$$

and:

$$
P_{99}=10,000
$$

Our rule becomes:

$$
100\leq x\leq10,000
\rightarrow
\text{Normal}
$$

Now consider:

$$
x=50
$$

Since:

$$
50<100
$$

we flag it:

$$
\boxed{50\rightarrow\text{Anomaly}}
$$

Similarly:

$$
x=15,000
$$

Since:

$$
15,000>10,000
$$

we get:

$$
\boxed{15,000\rightarrow\text{Anomaly}}
$$

---

## Choosing the Percentile

This is the most important practical decision.

You could choose:

$$
P_1,\ P_{99}
$$

or:

$$
P_{0.5},\ P_{99.5}
$$

or:

$$
P_5,\ P_{95}
$$

depending on how aggressively you want to flag observations.

### Lower percentile thresholds

For example:

$$
P_5,\ P_{95}
$$

means you're flagging approximately the lowest 5% and highest 5%.

That's a **much more sensitive** detector.

Whereas:

$$
P_1,\ P_{99}
$$

flags approximately the extreme 1% on each side.

So:

$$
\boxed{\text{More extreme percentile}
\rightarrow
\text{fewer points flagged}}
$$

and:

$$
\boxed{\text{Less extreme percentile}
\rightarrow
\text{more points flagged}}
$$

---

## Why use percentile detection?

Consider data such as income, transaction amount, website traffic, etc.

These can be **highly skewed**.

For example:

$$
20,\ 25,\ 30,\ 35,\ 40,\ 50,\ 100,\ 500,\ 5000
$$

The data doesn't necessarily resemble a symmetric normal distribution.

In such situations, a rule based on:

$$
\mu\pm3\sigma
$$

may not represent "rare" behavior very well.

Percentiles don't require us to assume that the data follows a normal distribution.

That's their main advantage.

---

## IQR vs Percentile Detection

They are related but different.

### IQR

Uses:

$$
Q_1,\ Q_3
$$

and then expands beyond the middle 50%:

$$
Q_1-1.5IQR
$$

$$
Q_3+1.5IQR
$$

### Percentile-based detection

Directly chooses the boundaries:

$$
P_1,\ P_{99}
$$

for example.

So:

```text id="e6z1t4"
IQR
 ↓
Q1, Q3
 ↓
IQR
 ↓
1.5 × IQR
 ↓
Boundaries


Percentile
 ↓
Choose percentiles directly
 ↓
Boundaries
```

---

## Important Limitation

Percentile-based detection will **always flag a certain proportion of observations** if applied directly to the same dataset used to calculate the percentiles.

For example, using:

$$
P_1,\ P_{99}
$$

essentially defines the extreme tails as unusual.

But that doesn't necessarily mean those observations are **truly anomalous**.

They may simply be legitimate extreme values.

For example, if the richest 1% of customers naturally spend much more, those customers aren't necessarily fraudulent.

So:

$$
\boxed{\text{Rare} \neq \text{Automatically anomalous}}
$$

This is a recurring principle in anomaly detection.

**Statistical methods identify unusual observations; domain knowledge determines whether the unusualness is actually meaningful.**


## 5. Multivariate Considerations

So far, our statistical methods have mostly looked at **one feature at a time**.

For example:

$$
\text{Temperature}=90
$$

and we ask whether 90 is unusual.

But real datasets usually contain **multiple features**.

---

### Intuition

Suppose we monitor a machine using:

* Temperature
* Vibration

Consider these observations:

```text id="5w4h8q"
Temperature    Vibration

50             20
51             21
49             19
52             20
90             80
```

The last observation:

$$
(90,80)
$$

looks clearly unusual.

But here's where things become interesting.

Imagine instead:

```text id="q2f7km"
Temperature    Vibration

50             20
51             21
49             19
52             20
60             80
```

Looking at each feature separately:

* Temperature = 60 might not be extremely unusual.
* Vibration = 80 might be unusual.

But imagine a different example where **both individual values are normal**, yet their combination is unusual.

For example:

```text id="9h4m2k"
Temperature    Pressure

50             100
51             102
49             98
52             101
60             100
```

Maybe temperature 60 is not extreme enough by itself, and pressure 100 is completely normal.

But if normal machines follow a relationship like:

$$
\text{higher temperature}
\rightarrow
\text{higher pressure}
$$

then:

$$
(60,100)
$$

might violate the normal relationship.

That is the key multivariate idea:

$$
\boxed{\text{A point can look normal in each feature individually but anomalous when features are considered together.}}
$$

---

## Univariate vs Multivariate

### Univariate

We look at one feature:

$$
x=\text{Temperature}
$$

and calculate something such as:

$$
z=\frac{x-\mu}{\sigma}
$$

or use IQR.

```text id="x9q3pd"
Temperature
    ↓
Statistical analysis
    ↓
Normal / Anomaly
```

---

### Multivariate

Now each observation has multiple features:

$$
\mathbf{x}=
\begin{bmatrix}
x_1\\
x_2\\
\vdots\\
x_d
\end{bmatrix}
$$

where:

* \(d\) = number of features
* \(x_1,x_2,\ldots,x_d\) = feature values

For example, with two features:

$$
\mathbf{x}=
\begin{bmatrix}
\text{Temperature}\\
\text{Pressure}
\end{bmatrix}
$$

Now we care about the **joint behavior** of the features.

---

## A visual way to understand it

Imagine normal observations forming a diagonal pattern:

```text id="u3v8q1"
Pressure
   ↑
   |
   |             ●
   |          ●
   |       ●
   |    ●
   | ●
   |
   +--------------------→ Temperature
```

The two variables have a relationship.

Now place:

```text id="a8m2xz"
Pressure
   ↑
   |
   |             ●
   |          ●
   |       ●
   |    ●
   | ●          X
   |            ↑
   |         unusual combination
   |
   +--------------------→ Temperature
```

The \(X\) might not be extreme in either feature individually.

But its **combination** is unusual.

This is something simple Z-score or IQR applied independently to each column can miss.

---

## Why this matters

Suppose we independently apply Z-score:

```text id="q7v5kd"
Temperature → Z-score → Normal
Pressure    → Z-score → Normal
```

We might conclude:

$$
\boxed{\text{Normal}}
$$

But if the relationship between temperature and pressure is abnormal, the point may actually be anomalous.

So:

$$
\boxed{
\text{Univariate detection}
\neq
\text{Complete multivariate anomaly detection}
}
$$

---

## What can we do?

For genuinely multivariate anomaly detection, we need methods that consider the features **together**.

There are statistical approaches such as **Mahalanobis distance**, which considers:

* the distance from the center,
* the scale of each feature,
* and the relationships/correlation between features.

And later, the algorithms we're studying provide other ways to handle multidimensional data:

```text id="5t7k1p"
Multiple Features
       ↓
 ┌─────┼──────────┐
 ↓     ↓          ↓
IF    OC-SVM     LOF
 ↓     ↓          ↓
Different definitions
of unusualness
```

We don't need to introduce Mahalanobis distance in this notebook because it would become another major statistical method that isn't part of your planned structure.

The important takeaway is simply:

> **Applying Z-score or IQR independently to every feature does not necessarily detect anomalies caused by unusual combinations of features.**

---

## One more practical issue: feature scale

Suppose our features are:

* Age: \(20\)–\(80\)
* Income: \(20,000\)–\(2,00,000\)

Their numerical scales are very different.

This matters for methods that use distances.

For example:

$$
\text{Age difference}=10
$$

and:

$$
\text{Income difference}=50,000
$$

cannot simply be treated as comparable raw distances.

This is one reason **scaling/standardization** becomes important for many multivariate algorithms.

However, whether scaling is required depends on the particular anomaly-detection algorithm.

---

### Core idea

The progression is:

$$
\boxed{
\text{One feature}
\rightarrow
\text{Is this value unusual?}
}
$$

versus:

$$
\boxed{
\text{Multiple features}
\rightarrow
\text{Is this combination of values unusual?}
}
$$

That distinction becomes very important once we leave simple statistical rules and move into **Isolation Forest, One-Class SVM, and LOF**.


## 6. Python Implementation

Now we'll implement the three statistical approaches we've learned:

1. **Z-score**
2. **IQR**
3. **Percentile-based detection**

The goal isn't just to run code. We want to connect:

$$
\text{Theory}
\rightarrow
\text{Code}
\rightarrow
\text{Anomaly Decision}
$$

---

# 6.1 Z-Score Implementation

Let's start with a tiny dataset.

```python
import numpy as np
from scipy.stats import zscore

data = np.array([10, 11, 12, 10, 13, 11, 12, 10, 50])

z_scores = zscore(data)

print(z_scores)
```

### What happens here?

`zscore(data)` calculates:

$$
z_i=\frac{x_i-\mu}{\sigma}
$$

for every observation.

So instead of calculating each Z-score manually, Python calculates all of them.

---

### Set an anomaly threshold

Suppose we choose:

$$
T=3
$$

```python
threshold = 3

anomalies = np.abs(z_scores) > threshold

print(anomalies)
```

`np.abs()` is important because we care about:

$$
|z|
$$

rather than whether \(z\) is positive or negative.

The result is a Boolean array:

```text
False
False
False
False
False
False
False
False
True
```

So the final value, 50, has been flagged.

We can retrieve the actual anomalous values:

```python
print(data[anomalies])
```

Expected output:

```text
[50]
```

---

# 6.2 IQR Implementation

For IQR, we can calculate the quartiles using NumPy.

```python
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

The code directly represents what you already know mathematically:

$$
IQR=Q_3-Q_1
$$

then:

$$
L=Q_1-1.5(IQR)
$$

$$
U=Q_3+1.5(IQR)
$$

Now detect the anomalies:

```python
anomalies = (data < lower_bound) | (data > upper_bound)

print(data[anomalies])
```

Expected output:

```text
[50]
```

The `|` means **OR**.

So we're asking:

```text
Is data below lower_bound?
          OR
Is data above upper_bound?
```

---

# 6.3 Percentile-Based Detection

Now suppose we want to flag observations below the 10th percentile or above the 90th percentile.

```python
lower_percentile = np.percentile(data, 10)
upper_percentile = np.percentile(data, 90)

anomalies = (
    (data < lower_percentile) |
    (data > upper_percentile)
)

print(data[anomalies])
```

The important part is that the thresholds come **directly from the data's percentiles**.

We aren't using:

$$
\mu
$$

or:

$$
\sigma
$$

---

# 6.4 Putting the Methods Together

The same dataset can be analyzed using different definitions of unusualness:

```text
                    DATA
                      ↓
          ┌───────────┼───────────┐
          ↓           ↓           ↓
       Z-score       IQR      Percentile
          ↓           ↓           ↓
      Z-score      Quartiles   Percentiles
          ↓           ↓           ↓
      Threshold    Boundaries   Boundaries
          ↓           ↓           ↓
       Anomaly     Anomaly      Anomaly
```

The important lesson is:

$$
\boxed{
\text{Different statistical methods can produce different anomaly decisions}
}
$$

because they define "unusual" differently.

---

# 6.5 When Should You Use Which?

### Z-score

Useful when:

* Data is approximately normally distributed.
* Mean and standard deviation represent the data well.
* You want a standardized measure of deviation.

```text
Data
 ↓
Mean + SD
 ↓
Z-score
 ↓
Threshold
```

### IQR

Useful when:

* Data is skewed.
* Extreme values may exist.
* You want a more robust method.

```text
Data
 ↓
Q1 + Q3
 ↓
IQR
 ↓
Bounds
```

### Percentile

Useful when:

* You want a direct control over how much data is considered extreme.
* Distribution assumptions are undesirable.
* Domain requirements naturally suggest a percentile cutoff.

```text
Data
 ↓
Choose percentile
 ↓
Boundary
 ↓
Anomaly
```

---

# 6.6 One Important Warning

Don't blindly do:

```python
threshold = 3
```

for every dataset.

Similarly, don't blindly use:

```python
1.5 * IQR
```

or:

```python
1% / 99%
```

without understanding the data.

The choice depends on:

* Distribution
* Amount of noise
* Cost of false positives
* Cost of missed anomalies
* Domain knowledge

The code is easy.

**Choosing an appropriate detection strategy is the important part.**

# Statistical Methods — Notebook Recap

### Core intuition

$$
\boxed{
\text{Describe normal behavior}
\rightarrow
\text{Measure deviation}
\rightarrow
\text{Flag unusually extreme observations}
}
$$

### Z-score

$$
z=\frac{x-\mu}{\sigma}
$$

Decision:

$$
|z|>T
\rightarrow
\text{Anomaly}
$$

### IQR

$$
IQR=Q_3-Q_1
$$

$$
L=Q_1-1.5(IQR)
$$

$$
U=Q_3+1.5(IQR)
$$

Outside the bounds:

$$
\rightarrow \text{Anomaly}
$$

### Percentile

Choose lower and upper percentiles and flag observations outside them.

### Main limitation

These simple statistical methods are strongest when the statistical assumptions match the data. They can struggle with:

* Highly complex distributions
* Strong multivariate relationships
* Changing patterns
* High-dimensional data

And importantly:

$$
\boxed{
\text{Unusual} \neq \text{necessarily wrong}
}
$$

The statistical method **detects** the unusual observation; interpretation and domain knowledge help determine **why it is unusual**.
