## The easiest way to remember all 4

Think of the **same suspicious point** and ask four different questions:

### 1. Statistical Method

```text
"What is the normal range?"

        normal
    |-----------|
                  X
                  ↑
              far away
```

It checks whether the point is **statistically extreme**.

Example:

```text
Salaries:
40k 42k 43k 45k 44k 46k 500k
                       ↑
                    anomaly
```

---

### 2. Isolation Forest

```text
"How quickly can I isolate this point?"
```

Imagine randomly cutting the data:

```text
Normal points
● ● ● ● ● ●
     │
     │ random cuts
     ↓
● ● ●
● ● ●

              X
              │
              ↓
         isolated quickly
```

**Shorter path → more anomalous.**

---

### 3. One-Class SVM

```text
"What region represents normal data?"
```

Imagine drawing a boundary around normal observations:

```text
        ┌───────────────┐
        │ ● ● ● ●       │
        │ ● ● ● ● ●     │
        │   ● ● ●       │
        └───────────────┘
                       X
                       ↑
                    anomaly
```

**Inside boundary → normal**

**Outside boundary → anomaly**

---

### 4. LOF

```text
"Is this point much less dense than its neighbors?"
```

Imagine:

```text
Dense normal region

● ● ●
● ● ●
● ● ●

                    X
                   ●
```

The important thing is not simply that `X` is far away.

LOF asks:

> "Are the points around X also sparse?"

If:

```text
Normal neighborhood:
● ● ● ● ●
similar density → normal
```

but:

```text
X
        ●

neighbors:
● ● ● ● ●
```

then X has **much lower local density** → anomaly.

# The biggest difference

The four methods can detect the **same anomaly for completely different reasons**:

```text
                    ANOMALY
                       │
       ┌───────────────┼────────────────┐
       │               │                │
 Statistical      Isolation         One-Class
       │             Forest              SVM
       │               │                 │
 "Far from        "Easy to          "Outside
  statistics"      isolate"          boundary"
                       │
                       │
                      LOF
                       │
                "Less dense than
                 its neighbors"
```

### One-line memory trick

| Method               | Remember it as                          |
| -------------------- | --------------------------------------- |
| **Statistical**      | 📏 **Far from normal**                  |
| **Isolation Forest** | ✂️ **Easy to isolate**                  |
| **One-Class SVM**    | 🛑 **Outside normal boundary**          |
| **LOF**              | 👥 **Different density from neighbors** |

And the most important conceptual distinction is:

**Statistical → deviation**
**Isolation Forest → isolation**
**One-Class SVM → boundary**
**LOF → local density**


## When to Use Which?

| Method               | Use when                                  | Main idea                    |
| -------------------- | ----------------------------------------- | ---------------------------- |
| **Statistical**      | Simple data, clear statistical range      | 📏 Far from normal           |
| **Isolation Forest** | Large/multivariate datasets               | ✂️ Easy to isolate           |
| **One-Class SVM**    | Mostly normal data + complex boundary     | 🛑 Outside normal boundary   |
| **LOF**              | Local density / different-density regions | 👥 Less dense than neighbors |
