# 1. Association Rules — Core Concept

## 1.1 What is Association Rule Learning?

**Association Rule Learning** is an unsupervised learning technique used to discover **relationships between items/events in a dataset**.

The basic idea is:

> **If certain items occur together frequently, can we discover a meaningful relationship between them?**

The classic example is **market basket analysis**.

Suppose a supermarket has transactions like:

| Transaction | Items               |
| ----------- | ------------------- |
| T1          | Milk, Bread         |
| T2          | Milk, Bread, Butter |
| T3          | Bread, Butter       |
| T4          | Milk, Bread, Butter |
| T5          | Milk, Eggs          |

We might discover a rule such as:

$$
\text{Milk} \rightarrow \text{Bread}
$$

This means:

> Customers who buy **Milk** tend to also buy **Bread**.

Notice that this is **not saying Milk causes Bread**.

It is only saying there is an **association** between them.

---

# 1.2 What exactly are we trying to discover?

There are two main things:

### 1. Frequent Itemsets

First, we find groups of items that occur together frequently.

For example:

$$
\{\text{Milk, Bread}\}
$$

or

$$
\{\text{Milk, Bread, Butter}\}
$$

These are called **itemsets**.

---

### 2. Association Rules

Then we turn useful itemsets into rules.

For example:

$$
\{\text{Milk, Bread}\}
\rightarrow
\text{Butter}
$$

or

$$
\text{Milk}
\rightarrow
\text{Bread}
$$

The general form is:

$$
X \rightarrow Y
$$

where:

* $X$ = **antecedent** (left-hand side)
* $Y$ = **consequent** (right-hand side)

For example:

$$
\{\text{Milk, Bread}\}
\rightarrow
\{\text{Butter}\}
$$

Here:

* Antecedent = Milk + Bread
* Consequent = Butter

---

# 1.3 Why is this useful?

Imagine a supermarket has **1 million transactions**.

You cannot manually inspect every possible combination and notice relationships.

Association Rule Learning automatically searches for patterns such as:

$$
\text{Diapers} \rightarrow \text{Beer}
$$

$$
\text{Phone} \rightarrow \text{Phone Case}
$$

$$
\text{Laptop} \rightarrow \text{Mouse}
$$

$$
\text{Bread, Butter} \rightarrow \text{Milk}
$$

The same idea can be applied outside shopping:

* **E-commerce:** products purchased together
* **Recommendation systems:** items likely to be selected together
* **Web usage:** pages visited together
* **Healthcare:** symptoms occurring together
* **Banking:** transaction patterns
* **Cybersecurity:** events occurring together

---

# 1.4 The Important Idea

Association Rule Learning can be viewed as a **two-stage problem**:

$$
\boxed{
\text{Transactions}
\rightarrow
\text{Frequent Itemsets}
\rightarrow
\text{Association Rules}
}
$$

For example:

### Step 1 — Find frequent combinations

From:

$$
\{\text{Milk, Bread, Butter}\}
$$

we might determine that:

$$
\{\text{Milk, Bread}\}
$$

occurs frequently.

### Step 2 — Generate rules

From that frequent itemset, we can generate:

$$
\text{Milk} \rightarrow \text{Bread}
$$

and

$$
\text{Bread} \rightarrow \text{Milk}
$$

But **both directions do not necessarily have the same strength**.

That's where measures such as **support, confidence, and lift** become important.

---

# 1.5 One Important Distinction

Association Rule Learning is **not prediction in the normal supervised-learning sense**.

There is no target variable like:

```text
Features → Target
```

Instead, we have:

```text
Transactions
     ↓
Find patterns
     ↓
Find relationships
```

For example, we're not given:

> "Predict whether a customer will buy Bread."

Instead, we're asking:

> "What items tend to occur together in these transactions?"

That's why it belongs to **unsupervised learning**.

---

## The complete picture we'll build

Eventually, you'll understand this entire pipeline:

$$
\text{Raw Transactions}
\rightarrow
\text{Itemsets}
\rightarrow
\text{Frequent Itemsets}
\rightarrow
\text{Candidate Rules}
\rightarrow
\text{Support}
\rightarrow
\text{Confidence}
\rightarrow
\text{Lift}
\rightarrow
\text{Useful Rules}
$$

For now, the key idea is simply:

> **Association Rule Learning discovers useful relationships between things that occur together.**


# 2. Important Terminology

Before calculating **support, confidence, and lift**, we need to understand the basic objects used in Association Rule Learning.

---

## 2.1 Transaction

A **transaction** is one individual record containing the items/events that occurred together.

Example supermarket dataset:

| Transaction ID | Items               |
| -------------- | ------------------- |
| T1             | Milk, Bread         |
| T2             | Milk, Bread, Butter |
| T3             | Bread, Butter       |
| T4             | Milk, Bread, Butter |
| T5             | Milk, Eggs          |

Each row is one **transaction**.

So:

$$
T_1=\{\text{Milk, Bread}\}
$$

and

$$
T_2=\{\text{Milk, Bread, Butter}\}
$$

A transaction doesn't have to be a shopping basket. It could represent:

* one customer's purchase
* one patient's symptoms
* one user's website activity
* one day's events

The important idea is that **items occurring within the same transaction are considered together**.

---

## 2.2 Item

An **item** is a single element inside a transaction.

From:

$$
T_2=\{\text{Milk, Bread, Butter}\}
$$

the individual items are:

* Milk
* Bread
* Butter

So:

> **Transaction = collection of items**

---

## 2.3 Itemset

An **itemset** is simply a collection of one or more items.

For example:

$$
\{\text{Milk}\}
$$

is a **1-itemset**.

$$
\{\text{Milk, Bread}\}
$$

is a **2-itemset**.

$$
\{\text{Milk, Bread, Butter}\}
$$

is a **3-itemset**.

The number of items determines the size of the itemset.

| Itemset               | Size |
| --------------------- | ---: |
| {Milk}                |    1 |
| {Milk, Bread}         |    2 |
| {Milk, Bread, Butter} |    3 |

---

## 2.4 Candidate Itemset

A **candidate itemset** is an itemset that the algorithm is currently considering as a possible frequent itemset.

For example, suppose we have:

$$
\{\text{Milk, Bread}\}
$$

The algorithm checks:

> Does this combination occur frequently enough?

If yes → it can become a **frequent itemset**.

If no → it is discarded.

This distinction becomes especially important in **Apriori**, because Apriori generates many candidate itemsets and then eliminates the ones that cannot be useful.

---

## 2.5 Frequent Itemset

A **frequent itemset** is an itemset that occurs often enough according to a chosen **minimum support threshold**.

Suppose:

$$
\text{minimum support}=40\%
$$

and:

$$
\{\text{Milk, Bread}\}
$$

appears in 3 out of 5 transactions.

Its support is:

$$
\frac{3}{5}=60\%
$$

Since:

$$
60\% \geq 40\%
$$

the itemset is **frequent**.

So:

$$
\boxed{
\text{Candidate Itemset}
\rightarrow
\text{Check Support}
\rightarrow
\text{Frequent or Not Frequent}
}
$$

---

# 2.6 Antecedent

The **antecedent** is the **left-hand side** of an association rule.

For:

$$
\{\text{Milk}\}
\rightarrow
\{\text{Bread}\}
$$

the antecedent is:

$$
\boxed{\{\text{Milk}\}}
$$

Think:

> **If this happens...**

---

# 2.7 Consequent

The **consequent** is the **right-hand side** of an association rule.

For:

$$
\{\text{Milk}\}
\rightarrow
\{\text{Bread}\}
$$

the consequent is:

$$
\boxed{\{\text{Bread}\}}
$$

Think:

> **...this tends to happen too.**

Therefore:

$$
\boxed{
\text{Antecedent}
\rightarrow
\text{Consequent}
}
$$

means:

> When the antecedent occurs, the consequent tends to occur as well.

---

# 2.8 A Rule Can Have Multiple Items on Either Side

It isn't necessary to have only one item on each side.

For example:

$$
\{\text{Milk, Bread}\}
\rightarrow
\{\text{Butter}\}
$$

Here:

* Antecedent = {Milk, Bread}
* Consequent = {Butter}

Or:

$$
\{\text{Milk}\}
\rightarrow
\{\text{Bread, Butter}\}
$$

Here:

* Antecedent = {Milk}
* Consequent = {Bread, Butter}

---

# 2.9 Transaction vs Itemset

This distinction is very important.

Consider:

$$
T_2=\{\text{Milk, Bread, Butter}\}
$$

This is a **transaction** because it represents an actual customer's purchase.

But:

$$
\{\text{Milk, Bread}\}
$$

is an **itemset** because it is simply a combination of items we're analyzing.

An itemset **may appear inside many transactions**.

For example:

| Transaction | Items               |
| ----------- | ------------------- |
| T1          | Milk, Bread         |
| T2          | Milk, Bread, Butter |
| T3          | Bread, Butter       |
| T4          | Milk, Bread, Butter |

The itemset:

$$
\{\text{Milk, Bread}\}
$$

appears in:

$$
T_1,T_2,T_4
$$

So we can count how often that itemset occurs.

That count is what eventually allows us to calculate **support**.

---

# 2.10 Putting Everything Together

Using our dataset:

| Transaction | Items               |
| ----------- | ------------------- |
| T1          | Milk, Bread         |
| T2          | Milk, Bread, Butter |
| T3          | Bread, Butter       |
| T4          | Milk, Bread, Butter |
| T5          | Milk, Eggs          |

We have:

### Items

$$
\text{Milk},\text{Bread},\text{Butter},\text{Eggs}
$$

### Example itemsets

$$
\{\text{Milk, Bread}\}
$$

$$
\{\text{Bread, Butter}\}
$$

$$
\{\text{Milk, Bread, Butter}\}
$$

### Example frequent itemset

If:

$$
\{\text{Milk, Bread}\}
$$

occurs frequently enough, it becomes a **frequent itemset**.

From it, we can potentially generate rules such as:

$$
\text{Milk}\rightarrow\text{Bread}
$$

or:

$$
\text{Bread}\rightarrow\text{Milk}
$$

And **support, confidence, and lift** will tell us whether those rules are actually useful.


# 3. Support, Confidence, and Lift

These are the **three most important measures** used to determine whether an association rule is useful.

Consider this rule:

$$
\text{Milk}\rightarrow\text{Bread}
$$

We want to answer three different questions:

1. **Support:** How frequently do Milk and Bread occur together?
2. **Confidence:** When Milk occurs, how often does Bread also occur?
3. **Lift:** Is Bread actually more likely to occur when Milk occurs, compared with Bread occurring normally?

---

## 3.1 Support

### Intuition

**Support tells us how common an itemset is in the entire dataset.**

For an itemset $X$:

$$
\text{Support}(X)
=
\frac{\text{Number of transactions containing }X}
{\text{Total number of transactions}}
$$

For a rule:

$$
X\rightarrow Y
$$

support is based on the **combined itemset**:

$$
\text{Support}(X\rightarrow Y)
=
\text{Support}(X\cup Y)
$$

### Example

Consider:

| Transaction | Items               |
| ----------- | ------------------- |
| T1          | Milk, Bread         |
| T2          | Milk, Bread, Butter |
| T3          | Bread, Butter       |
| T4          | Milk, Bread, Butter |
| T5          | Milk, Eggs          |

Total transactions:

$$
5
$$

The combination:

$$
\{\text{Milk, Bread}\}
$$

appears in:

$$
T_1,T_2,T_4
$$

So:

$$
\text{Support}(\text{Milk, Bread})
=
\frac{3}{5}
=
0.6
=
60\%
$$

So Milk and Bread occur together in **60% of all transactions**.

### Why support matters

Suppose you discover:

$$
\text{Rare Product A}\rightarrow\text{Rare Product B}
$$

with very high confidence, but this combination appears in only **0.01%** of transactions.

It may not be practically useful.

Support helps us identify patterns that occur often enough to matter.

---

# 3.2 Confidence

### Intuition

Confidence answers:

> **When the antecedent occurs, how often does the consequent also occur?**

For:

$$
X\rightarrow Y
$$

the formula is:

$$
\text{Confidence}(X\rightarrow Y)
=
\frac{\text{Support}(X\cup Y)}
{\text{Support}(X)}
$$

Equivalently:

$$
\text{Confidence}(X\rightarrow Y)
=
\frac{\text{Transactions containing both }X\text{ and }Y}
{\text{Transactions containing }X}
$$

---

### Example

For:

$$
\text{Milk}\rightarrow\text{Bread}
$$

Milk appears in:

$$
T_1,T_2,T_4,T_5
$$

So Milk appears in:

$$
4
$$

transactions.

Milk and Bread together appear in:

$$
T_1,T_2,T_4
$$

So:

$$
3
$$

transactions contain both.

Therefore:

$$
\text{Confidence}
=
\frac{3}{4}
=
0.75
=
75\%
$$

Meaning:

> **75% of transactions containing Milk also contain Bread.**

---

# 3.3 Direction Matters

This is extremely important.

Consider:

$$
\text{Milk}\rightarrow\text{Bread}
$$

and:

$$
\text{Bread}\rightarrow\text{Milk}
$$

They have the **same support**, because both are based on:

$$
\{\text{Milk, Bread}\}
$$

But their **confidence can be different**.

Bread appears in:

$$
T_1,T_2,T_3,T_4
$$

So Bread occurs in:

$$
4
$$

transactions.

Therefore:

$$
\text{Confidence}(\text{Bread}\rightarrow\text{Milk})
=
\frac{3}{4}
=
75\%
$$

In this particular dataset, both happen to be 75%.

But imagine Bread appeared in 5 transactions while Milk appeared in only 4. Then:

$$
\text{Confidence}(\text{Milk}\rightarrow\text{Bread})
=
\frac{3}{4}
=
75\%
$$

while:

$$
\text{Confidence}(\text{Bread}\rightarrow\text{Milk})
=
\frac{3}{5}
=
60\%
$$

So:

> **Support is not directional, but confidence is directional.**

---

# 3.4 Lift

Confidence alone can be misleading.

Suppose:

$$
\text{Confidence}(\text{Milk}\rightarrow\text{Bread})=80\%
$$

That sounds strong.

But what if **Bread is already purchased by 80% of everyone**?

Then Milk may not have any meaningful relationship with Bread.

This is where **Lift** becomes important.

---

## Intuition

Lift asks:

> **How much more likely is the consequent when the antecedent occurs compared with the consequent occurring normally?**

The formula is:

$$
\text{Lift}(X\rightarrow Y)
=
\frac{\text{Confidence}(X\rightarrow Y)}
{\text{Support}(Y)}
$$

Another equivalent form is:

$$
\text{Lift}(X\rightarrow Y)
=
\frac{\text{Support}(X\cup Y)}
{\text{Support}(X)\times\text{Support}(Y)}
$$

---

## 3.5 Example of Lift

Using our dataset:

Bread appears in:

$$
T_1,T_2,T_3,T_4
$$

Therefore:

$$
\text{Support}(\text{Bread})
=
\frac{4}{5}
=
80\%
$$

We already calculated:

$$
\text{Confidence}(\text{Milk}\rightarrow\text{Bread})
=
75\%
$$

Therefore:

$$
\text{Lift}
=
\frac{75\%}{80\%}
=
0.9375
$$

So:

$$
\boxed{\text{Lift}=0.9375}
$$

This is actually interesting.

Despite the 75% confidence, the lift is **less than 1**.

That means Milk does **not** increase the likelihood of Bread occurring; Bread is actually slightly less likely among Milk transactions than in the overall dataset.

---

# 3.6 How to Interpret Lift

The most important interpretation is:

### Lift > 1

Positive association.

$$
X\rightarrow Y
$$

means $Y$ occurs **more often than expected** when $X$ occurs.

---

### Lift = 1

No meaningful association.

$$
P(Y|X)=P(Y)
$$

The occurrence of $X$ doesn't change the likelihood of $Y$.

---

### Lift < 1

Negative association.

When $X$ occurs, $Y$ is **less likely** than normal.

---

### Quick interpretation

$$
\boxed{
\begin{array}{c|c}
\text{Lift} & \text{Interpretation}\\
\hline
>1 & \text{Positive association}\\
=1 & \text{No association}\\
<1 & \text{Negative association}
\end{array}
}
$$

---

# 3.7 Support vs Confidence vs Lift

| Measure        | Main Question                                 |
| -------------- | --------------------------------------------- |
| **Support**    | How common is the combination?                |
| **Confidence** | Given X, how often does Y occur?              |
| **Lift**       | Is Y actually more likely because X occurred? |

Think of them as three different filters:

$$
\text{Support}
\rightarrow
\text{Is the pattern common?}
$$

$$
\text{Confidence}
\rightarrow
\text{Is the rule reliable?}
$$

$$
\text{Lift}
\rightarrow
\text{Is the relationship actually meaningful?}
$$

---

# 3.8 Why We Need All Three

A rule can have:

* high support but weak confidence
* high confidence but low support
* high confidence but lift close to 1
* low support but very high lift

Therefore, looking at only one metric can give a misleading picture.

A useful association rule generally needs to be considered from **all three perspectives**:

$$
\boxed{
\text{Commonness}
+
\text{Reliability}
+
\text{Actual Association}
}
$$

which correspond to:

$$
\boxed{
\text{Support}
+
\text{Confidence}
+
\text{Lift}
}
$$


# Which Values are best for each parameters?

### 1. Support — higher is generally better

Support tells you **how common the pattern is**.

| Support  | Interpretation          |
| -------- | ----------------------- |
| Very low | Rare pattern            |
| Moderate | Occurs reasonably often |
| High     | Very common pattern     |

For example:

* Support = **1%** → combination is rare
* Support = **20%** → fairly common
* Support = **60%** → very common

But **high support isn't automatically better**. A very common combination may be obvious and not useful.

---

### 2. Confidence — higher is generally better

Confidence tells you:

> If I have X, how often do I get Y?

For example:

$$
\text{Confidence}(X\rightarrow Y)=90\%
$$

means 90% of transactions containing X also contain Y.

| Confidence | Interpretation                     |
| ---------- | ---------------------------------- |
| 50%        | Weak/moderate depending on context |
| 70%        | Stronger                           |
| 90%        | Very strong                        |

But again, **high confidence alone can be misleading**.

If Y is purchased by 95% of everyone, then a 95% confidence rule:

$$
X\rightarrow Y
$$

isn't particularly impressive.

That's why we need **lift**.

---

### 3. Lift — this one has a special interpretation

Here, the key value is **1**, not simply "higher is always better."

|         Lift | Meaning                       |
| -----------: | ----------------------------- |
|      **< 1** | Negative association          |
|      **= 1** | No association                |
|      **> 1** | Positive association          |
| **Much > 1** | Stronger positive association |

For example:

$$
\text{Lift}=1.0
$$

→ X doesn't really affect Y.

$$
\text{Lift}=1.5
$$

→ Y is **1.5× as likely** when X occurs compared with its normal occurrence.

$$
\text{Lift}=3
$$

→ Y is **3× as likely** when X occurs.

So, for positive associations:

> **Higher lift generally means a stronger association.**

---

## The important part

Don't think:

> "I need support > 50%, confidence > 80%, lift > 2."

There are **no universal cutoffs**.

Instead, think:

$$
\boxed{
\text{High enough Support}
+
\text{High enough Confidence}
+
\text{Lift meaningfully > 1}
}
$$

The actual thresholds depend heavily on the dataset.

For example, in a supermarket with thousands of products, a support of **1%** might already represent thousands of transactions and therefore be useful.

### Simple mental model

**Support → Is it common enough?**

**Confidence → Is the rule reliable enough?**

**Lift → Is the relationship actually interesting?**

And one more important point: **"best rule" usually isn't the rule with the highest value of one metric.** You normally filter by minimum support/confidence and then examine lift and business usefulness.

# 4. Other Important Measures

Support, confidence, and lift are the core measures, but a few additional measures are useful for understanding association rules properly.

We don't need to memorize all of them immediately. The most important ones to know are **Leverage** and **Conviction**.

---

## 4.1 Leverage

### Intuition

Lift tells us **how many times more likely** two items are to occur together than expected.

Leverage instead tells us the **actual difference** between:

> How often the two items occur together

and

> How often we would expect them to occur together if they were independent.

The formula is:

$$
\text{Leverage}(X\rightarrow Y)
=
\text{Support}(X\cup Y)
-
\text{Support}(X)\times\text{Support}(Y)
$$

---

### Example

Suppose:

$$
\text{Support}(X)=0.4
$$

$$
\text{Support}(Y)=0.5
$$

If X and Y were independent, we'd expect:

$$
0.4\times0.5=0.2
$$

Suppose their actual combined support is:

$$
\text{Support}(X\cup Y)=0.3
$$

Then:

$$
\text{Leverage}
=
0.3-0.2
=
0.1
$$

So the actual co-occurrence is **10 percentage points higher** than what we'd expect under independence.

---

### Interpretation

$$
\boxed{
\text{Leverage}>0
}
$$

→ Positive association.

$$
\boxed{
\text{Leverage}=0
}
$$

→ Independence.

$$
\boxed{
\text{Leverage}<0
}
$$

→ Negative association.

Unlike lift, leverage is expressed as an **absolute difference**, not a ratio.

---

# 4.2 Conviction

Conviction is another measure of how strongly a rule suggests that the consequent follows the antecedent.

The formula is:

$$
\text{Conviction}(X\rightarrow Y)
=
\frac{1-\text{Support}(Y)}
{1-\text{Confidence}(X\rightarrow Y)}
$$

It focuses particularly on how often the rule is **wrong**.

---

### Intuition

Consider:

$$
X\rightarrow Y
$$

If confidence is high, then whenever X occurs, Y usually occurs.

So there are fewer cases where:

$$
X\text{ occurs but }Y\text{ does not}
$$

Conviction measures this idea.

---

### Example

Suppose:

$$
\text{Support}(Y)=0.5
$$

and:

$$
\text{Confidence}(X\rightarrow Y)=0.8
$$

Then:

$$
\text{Conviction}
=
\frac{1-0.5}{1-0.8}
$$

$$
=
\frac{0.5}{0.2}
=
2.5
$$

So:

$$
\boxed{\text{Conviction}=2.5}
$$

A larger conviction generally indicates a stronger directional implication.

---

# 4.3 What Value Represents "Good"?

For the additional measures:

### Leverage

$$
\text{Leverage}>0
$$

is desirable for a positive association.

The larger it is, the greater the **absolute difference from independence**.

### Conviction

$$
\text{Conviction}>1
$$

indicates positive directional dependence.

Higher values generally indicate stronger rules.

A conviction of:

$$
1
$$

corresponds to independence.

---

# 4.4 Don't Confuse the Measures

Let's put everything together:

| Measure        | What it tells you                                        | Generally desirable           |
| -------------- | -------------------------------------------------------- | ----------------------------- |
| **Support**    | How common is the combination?                           | Higher, subject to usefulness |
| **Confidence** | How often does Y occur when X occurs?                    | Higher                        |
| **Lift**       | How much stronger is the relationship than independence? | > 1                           |
| **Leverage**   | How much actual extra co-occurrence exists?              | > 0                           |
| **Conviction** | How strongly does X imply Y?                             | > 1                           |

The important distinction is:

$$
\boxed{
\text{Support = frequency}
}
$$

$$
\boxed{
\text{Confidence = reliability}
}
$$

$$
\boxed{
\text{Lift = relative association}
}
$$

$$
\boxed{
\text{Leverage = absolute association}
}
$$

$$
\boxed{
\text{Conviction = directional implication}
}
$$

In practice, **support, confidence, and lift are the three you should understand most deeply**. Leverage and conviction are useful additional measures when evaluating rules more carefully.


# Lift V/S Leverage

The main difference is **relative vs absolute association**.

Suppose:

$$
X \rightarrow Y
$$

### Lift

Lift asks:

> **How many times more often do X and Y occur together than we would expect if they were independent?**

$$
\text{Lift}
=
\frac{\text{Support}(X\cup Y)}
{\text{Support}(X)\times\text{Support}(Y)}
$$

Example:

$$
\text{Actual support}=0.30
$$

$$
\text{Expected support}=0.20
$$

Then:

$$
\text{Lift}=\frac{0.30}{0.20}=1.5
$$

**Interpretation:** They occur together **1.5× as much as expected**.

---

### Leverage

Leverage asks:

> **How much more frequently do X and Y occur together than expected?**

$$
\text{Leverage}
=
\text{Actual support}
-
\text{Expected support}
$$

Using the same example:

$$
\text{Leverage}=0.30-0.20=0.10
$$

**Interpretation:** They occur together **10 percentage points more than expected**.

---

### Direct comparison

|                      | **Lift**                   | **Leverage**        |
| -------------------- | -------------------------- | ------------------- |
| Measures             | Relative difference        | Absolute difference |
| Formula              | Actual ÷ Expected          | Actual − Expected   |
| Example              | 1.5× expected              | +0.10               |
| Baseline             | 1                          | 0                   |
| Positive association | > 1                        | > 0                 |
| Main question        | "How many times stronger?" | "How much extra?"   |

### Easy way to remember

**Lift = "How many times?"**

**Leverage = "How much more?"**

So if:

$$
\text{Expected}=20\%
$$

and:

$$
\text{Actual}=30\%
$$

then:

$$
\boxed{\text{Lift}=1.5}
$$

means **1.5 times expected**, while

$$
\boxed{\text{Leverage}=0.10}
$$

means **10 percentage points above expected**.

# 5. Association Rule Structure

Now that we know the measures, let's understand **how actual rules are formed**.

An association rule has the general form:

$$
X\rightarrow Y
$$

where:

* $X$ = **Antecedent**
* $Y$ = **Consequent**
* $X\cap Y=\emptyset$

The items on the left and right must be different.

---

## 5.1 From an Itemset to Rules

Suppose we have the frequent itemset:

$$
\{\text{Milk, Bread, Butter}\}
$$

We can split this itemset into different antecedent and consequent combinations.

For example:

$$
\{\text{Milk, Bread}\}\rightarrow\{\text{Butter}\}
$$

$$
\{\text{Milk, Butter}\}\rightarrow\{\text{Bread}\}
$$

$$
\{\text{Bread, Butter}\}\rightarrow\{\text{Milk}\}
$$

We can also have one item on the left:

$$
\{\text{Milk}\}\rightarrow\{\text{Bread, Butter}\}
$$

$$
\{\text{Bread}\}\rightarrow\{\text{Milk, Butter}\}
$$

$$
\{\text{Butter}\}\rightarrow\{\text{Milk, Bread}\}
$$

So one frequent itemset can produce **multiple rules**.

---

## 5.2 How Many Rules Can Be Generated?

Suppose an itemset contains $n$ items.

Every non-empty subset can potentially become the antecedent.

The remaining items become the consequent.

The number of possible rules is:

$$
2^n-2
$$

Why?

There are:

$$
2^n
$$

possible subsets.

We exclude:

* the empty set
* the complete itemset

Therefore:

$$
2^n-2
$$

### Example: 3-itemset

For:

$$
\{\text{A,B,C}\}
$$

we have:

$$
2^3-2=6
$$

possible rules.

They are:

$$
A\rightarrow BC
$$

$$
B\rightarrow AC
$$

$$
C\rightarrow AB
$$

$$
AB\rightarrow C
$$

$$
AC\rightarrow B
$$

$$
BC\rightarrow A
$$

---

# 5.3 Important: Rules Are Directional

Consider:

$$
A\rightarrow B
$$

and:

$$
B\rightarrow A
$$

These are **two different rules**.

Their support is the same because both involve:

$$
\{A,B\}
$$

But their confidence can be different:

$$
\text{Confidence}(A\rightarrow B)
\neq
\text{Confidence}(B\rightarrow A)
$$

because their antecedents are different.

This is one reason we must be careful when interpreting association rules.

---

# 5.4 Frequent Itemset vs Association Rule

These are **not the same thing**.

### Frequent itemset

$$
\{A,B,C\}
$$

simply tells us:

> A, B, and C occur together frequently.

### Association rule

$$
\{A,B\}\rightarrow\{C\}
$$

makes a directional statement:

> When A and B occur, C tends to occur as well.

So the overall process is:

$$
\text{Transactions}
\rightarrow
\text{Frequent Itemsets}
\rightarrow
\text{Association Rules}
$$

---

# 5.5 Rule Generation Is a Filtering Process

Not every possible rule is useful.

Suppose we generate:

$$
A\rightarrow B
$$

We then calculate:

$$
\text{Support}
$$

$$
\text{Confidence}
$$

$$
\text{Lift}
$$

and possibly other measures.

Then we can filter the rules.

For example:

$$
\text{All possible rules}
\rightarrow
\text{Minimum Support}
\rightarrow
\text{Minimum Confidence}
\rightarrow
\text{Check Lift}
\rightarrow
\text{Useful Rules}
$$

This becomes especially important when the dataset contains thousands or millions of transactions.

The number of possible combinations can become enormous.

That is the problem **Apriori and FP-Growth** are designed to help solve efficiently.

# 6. Association Rule Learning — Complete Workflow

Before studying **Apriori**, it's important to understand the overall process without worrying about how the algorithm efficiently finds the patterns.

The complete workflow is:

$$
\text{Transactions}
\rightarrow
\text{Itemsets}
\rightarrow
\text{Frequent Itemsets}
\rightarrow
\text{Rules}
\rightarrow
\text{Evaluate Rules}
$$

---

## 6.1 Step 1 — Collect Transactions

Suppose we have:

| Transaction | Items               |
| ----------- | ------------------- |
| T1          | Milk, Bread         |
| T2          | Milk, Bread, Butter |
| T3          | Bread, Butter       |
| T4          | Milk, Bread, Butter |
| T5          | Milk, Eggs          |

This is our input data.

---

## 6.2 Step 2 — Find Itemsets

We can form combinations of items.

### 1-itemsets

$$
\{\text{Milk}\}
$$

$$
\{\text{Bread}\}
$$

$$
\{\text{Butter}\}
$$

$$
\{\text{Eggs}\}
$$

### 2-itemsets

$$
\{\text{Milk, Bread}\}
$$

$$
\{\text{Milk, Butter}\}
$$

$$
\{\text{Milk, Eggs}\}
$$

$$
\{\text{Bread, Butter}\}
$$

and so on.

### 3-itemsets

$$
\{\text{Milk, Bread, Butter}\}
$$

etc.

The problem is that the number of possible itemsets grows very quickly as the number of items increases.

---

# 6.3 Step 3 — Calculate Support

For every candidate itemset, we determine how frequently it occurs.

For example:

$$
\{\text{Milk, Bread}\}
$$

occurs in:

$$
T_1,T_2,T_4
$$

Therefore:

$$
\text{Support}=
\frac{3}{5}
=
60\%
$$

Suppose we choose:

$$
\text{Minimum Support}=40\%
$$

Then:

$$
60\%\geq40\%
$$

so:

$$
\{\text{Milk, Bread}\}
$$

is a **frequent itemset**.

---

# 6.4 Step 4 — Remove Infrequent Itemsets

Suppose:

$$
\{\text{Milk, Eggs}\}
$$

appears only once.

Then:

$$
\text{Support}
=
\frac{1}{5}
=
20\%
$$

Since:

$$
20\%<40\%
$$

we discard it.

Therefore:

$$
\boxed{
\text{Candidate Itemset}
\rightarrow
\text{Support Check}
\rightarrow
\text{Frequent / Discard}
}
$$

---

# 6.5 Step 5 — Generate Rules

Once we have frequent itemsets, we can create rules.

Suppose:

$$
\{\text{Milk, Bread}\}
$$

is frequent.

Possible rules include:

$$
\text{Milk}\rightarrow\text{Bread}
$$

and:

$$
\text{Bread}\rightarrow\text{Milk}
$$

---

# 6.6 Step 6 — Evaluate the Rules

Now calculate:

### Support

How common is the complete combination?

### Confidence

How often does the consequent occur when the antecedent occurs?

### Lift

Is the association actually stronger than what we'd expect by chance/independence?

So:

$$
\text{Rule}
\rightarrow
\text{Support}
+
\text{Confidence}
+
\text{Lift}
\rightarrow
\text{Interpretation}
$$

---

# 6.7 The Problem With the Basic Approach

Imagine we have:

$$
1000
$$

different items.

The number of possible itemsets is:

$$
2^{1000}-1
$$

which is an enormous number.

We obviously cannot blindly generate every possible combination and calculate its support.

So we need **efficient algorithms**.

That's where:

$$
\boxed{\text{Apriori}}
$$

and

$$
\boxed{\text{FP-Growth}}
$$

come in.

Their goal is not to change what an association rule means.

Instead, they provide **efficient ways to find frequent itemsets**.

---

# 6.8 The Big Picture

The distinction is important:

### Association Rule Learning

Tells us **what we're trying to discover**:

$$
\text{Frequent Itemsets}
\rightarrow
\text{Association Rules}
$$

### Apriori

Provides one algorithm for efficiently finding those frequent itemsets.

### FP-Growth

Provides another, generally more efficient approach for finding frequent itemsets.

So:

$$
\boxed{
\text{Association Rules}
=
\text{Problem / Technique}
}
$$

while:

$$
\boxed{
\text{Apriori and FP-Growth}
=
\text{Algorithms used to solve it}
}
$$

This is the foundation we'll use when we move into the **Apriori algorithm**.


## Association Rule Learning — Rules Recap

Suppose we have:

$$
X\rightarrow Y
$$

There are **two types of rules/concepts** you should distinguish.

### 1. Frequent Itemset

A **frequent itemset** is simply a group of items that appears together frequently.

Example:

$$
\{\text{Milk, Bread, Butter}\}
$$

It does **not** have a direction.

It only says:

> Milk, Bread, and Butter commonly occur together.

---

### 2. Association Rule

An association rule gives that itemset a **direction**:

$$
X\rightarrow Y
$$

Example:

$$
\text{Milk}\rightarrow\text{Bread}
$$

It means:

> When Milk occurs, Bread tends to occur as well.

It does **not** mean Milk causes someone to buy Bread.

---

## The Measures Used to Evaluate a Rule

For:

$$
\text{Milk}\rightarrow\text{Bread}
$$

### Support

**Question:** How common are Milk and Bread together?

$$
\text{Support}
=
\frac{\text{Milk + Bread transactions}}
{\text{Total transactions}}
$$

Higher generally means the pattern is more common.

---

### Confidence

**Question:** When Milk occurs, how often does Bread occur?

$$
\text{Confidence}
=
\frac{\text{Milk + Bread transactions}}
{\text{Milk transactions}}
$$

Higher generally means the rule is more reliable.

---

### Lift

**Question:** Is Bread actually more likely when Milk occurs than it normally is?

$$
\text{Lift}
=
\frac{\text{Confidence}(\text{Milk}\rightarrow\text{Bread})}
{\text{Support}(\text{Bread})}
$$

Interpretation:

* **Lift > 1** → positive association
* **Lift = 1** → no association
* **Lift < 1** → negative association

---

### Leverage

**Question:** How much more often do Milk and Bread occur together than we'd expect if they were independent?

$$
\text{Leverage}
=
\text{Support}(\text{Milk,Bread})
-
\text{Support}(\text{Milk})\times\text{Support}(\text{Bread})
$$

Interpretation:

* **> 0** → positive association
* **= 0** → independence
* **< 0** → negative association

---

### Conviction

**Question:** How strongly does Milk imply Bread, particularly by looking at cases where the rule fails?

$$
\text{Conviction}
=
\frac{1-\text{Support}(\text{Bread})}
{1-\text{Confidence}(\text{Milk}\rightarrow\text{Bread})}
$$

Interpretation:

* **= 1** → independence
* **> 1** → positive directional relationship
* Higher → generally stronger implication

---

### The easiest way to remember

| Measure        | Ask yourself                               |
| -------------- | ------------------------------------------ |
| **Support**    | "How common is the combination?"           |
| **Confidence** | "When X happens, how often does Y happen?" |
| **Lift**       | "How many times more likely is Y with X?"  |
| **Leverage**   | "How much extra co-occurrence is there?"   |
| **Conviction** | "How strongly does X imply Y?"             |

The **three core measures** to focus on most are:

$$
\boxed{\text{Support + Confidence + Lift}}
$$

Leverage and conviction are additional measures for deeper rule evaluation.
