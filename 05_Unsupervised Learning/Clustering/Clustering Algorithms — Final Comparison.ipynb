# How to Choose — The Main Idea

### 1. Data looks like round/compact groups

Use **K-Means**.

```text
● ●        ● ●
 ● ●      ● ●
```

Think:

$$
\boxed{\text{K-Means}\rightarrow\text{compact spherical clusters}}
$$

---

### 2. You don't know the hierarchy/group structure

Use **Hierarchical Clustering**.

It is useful when you want to see:

```text
all points
   ↓
small groups
   ↓
larger groups
   ↓
larger groups
```

The **dendrogram** is the important visualization.

---

### 3. Data has irregular shapes + noise

Use **DBSCAN**.

For example:

```text
   ● ● ●
 ●       ●
 ●       ●
   ● ● ●

             ● ●
           ●     ●
             ●
```

DBSCAN can discover these shapes because it focuses on **density**, not simply distance from a center.

And it can mark isolated points as:

```text
noise = -1
```

---

### 4. Data has complicated/non-linear structure

Use **Spectral Clustering**.

Its main idea is:

$$
X
\rightarrow
W
\rightarrow
D
\rightarrow
L
\rightarrow
\text{Eigenvectors}
\rightarrow
\text{K-Means}
\rightarrow
\text{Clusters}
$$

It converts the data into a **graph-based representation** so that relationships that aren't obvious using ordinary distance can become easier to separate.

---

### 5. Data looks like overlapping elliptical distributions

Use **GMM**.

GMM learns:

$$
\mu_k,\quad\Sigma_k,\quad\pi_k
$$

and gives probabilities:

$$
P(C_k\mid x)
$$

So instead of simply saying:

```text
Point → Cluster 1
```

it can say:

$$
P(C_1\mid x)=0.7
$$

$$
P(C_2\mid x)=0.3
$$

That's why GMM is particularly useful when you care about **uncertainty and soft membership**.

---

# The Simplest Decision Rule

Remember these five words:

$$
\boxed{
\text{K-Means}\rightarrow\text{Centers}
}
$$

$$
\boxed{
\text{Hierarchical}\rightarrow\text{Hierarchy}
}
$$

$$
\boxed{
\text{DBSCAN}\rightarrow\text{Density}
}
$$

$$
\boxed{
\text{Spectral}\rightarrow\text{Graph}
}
$$

$$
\boxed{
\text{GMM}\rightarrow\text{Probability}
}
$$

And for **shape**:

```text
Spherical / compact     → K-Means
Hierarchy needed        → Hierarchical
Irregular + noise       → DBSCAN
Non-linear / graph      → Spectral
Elliptical + overlap    → GMM
```

One important point: **there is no universally "best" clustering algorithm**. The data shape, noise, dimensionality, and what you want from the clusters determine which one is appropriate.
