# Topic 1 — Recommender Systems

## 1. Concept

A **Recommender System** is a system that predicts or suggests items that a user may be interested in.

Examples:

* Netflix → movies/shows
* YouTube → videos
* Amazon → products
* Spotify → songs
* Instagram → posts/reels

The goal is essentially:

$$
\boxed{\text{User} \rightarrow \text{predict items the user may like}}
$$

The lecture introduces recommender systems through the problem of **making recommendations**.

---

## 2. Basic Recommendation Problem

Suppose we have:

* **Users**
* **Items**
* Some information about how users interacted with those items.

For example, movie ratings:

| Movie                | Alice | Bob | Carol | Dave |
| -------------------- | ----: | --: | ----: | ---: |
| Love at Last         |     5 |   5 |     0 |    0 |
| Romance Forever      |     5 |   ? |     ? |    0 |
| Cute Puppies of Love |     ? |   4 |     0 |    ? |
| Nonstop Car Chases   |     0 |   0 |     5 |    4 |
| Swords vs Karate     |     0 |   0 |     5 |    ? |

The lecture uses exactly this movie-rating setup to introduce recommender systems.

The system's job is to predict the `?` values.

For example:

$$
\boxed{\text{What rating would Alice give "Cute Puppies of Love"?}}
$$

---

## 3. Important Notation

The lecture defines:

### \(n_u\)

Number of users.

$$
n_u = \text{number of users}
$$

In the example:

$$
n_u=4
$$

### \(n_m\)

Number of movies/items.

$$
n_m = \text{number of movies}
$$

In the example:

$$
n_m=5
$$

### \(r(i,j)\)

Indicates whether user \(j\) has rated movie \(i\).

$$
r(i,j)=
\begin{cases}
1 & \text{if user }j\text{ rated movie }i\\
0 & \text{otherwise}
\end{cases}
$$

### \(y(i,j)\)

The actual rating given by user \(j\) to movie \(i\).

It is defined only when:

$$
r(i,j)=1
$$

These definitions are given in the lecture's notation slide.

---

## 4. Example

Suppose:

```text
Alice rated "Love at Last" = 5
```

Then:

$$
r(1,1)=1
$$

and

$$
y(1,1)=5
$$

But suppose Alice hasn't rated "Cute Puppies of Love":

$$
r(3,1)=0
$$

Therefore:

$$
y(3,1)
$$

is **not defined yet**.

---

## 5. Core Problem

A recommender system tries to learn from the information we already have:

```text
Existing user-item interactions
              ↓
        Learn patterns
              ↓
      Predict unknown values
              ↓
       Make recommendations
```

For this lecture, the first major problem is:

$$
\boxed{\text{Predict missing user-item ratings/interactions}}
$$

---

## 6. Important distinction

Don't confuse:

**`0`**

with

**`?`**

In the lecture's rating example:

* `5` → user gave 5 stars
* `0` → user gave 0 / did not rate positively
* `?` → rating is unknown/not available

Later, the lecture changes from numerical ratings to **binary feedback** such as likes, clicks, and purchases.

---

### What you need to remember

```text
Recommender System
        ↓
User + Item
        ↓
Use available interactions
        ↓
Predict unknown preference
        ↓
Recommend items
```

And the basic notation:

$$
\boxed{n_u,\ n_m,\ r(i,j),\ y(i,j)}
$$

# Topic 2 — Predicting Movie Ratings

---

## 1. Problem

We have a table of users and movies:

| Movie                | Alice | Bob | Carol | Dave |
| -------------------- | ----: | --: | ----: | ---: |
| Love at Last         |     5 |   5 |     0 |    0 |
| Romance Forever      |     5 |   ? |     ? |    0 |
| Cute Puppies of Love |     ? |   4 |     0 |    ? |
| Nonstop Car Chases   |     0 |   0 |     5 |    4 |
| Swords vs. Karate    |     0 |   0 |     5 |    ? |

The `?` means the user's rating is unknown.

So the recommender system needs to predict values such as:

$$
\boxed{y(3,1)}
$$

which means:

> What rating will **Alice (user 1)** give **Cute Puppies of Love (movie 3)**?

The PDF uses this exact movie-rating example.

---

# 2. What information do we have?

The lecture introduces **features of movies**.

For example, suppose every movie has two features:

$$
x_1 = \text{romance}
$$

$$
x_2 = \text{action}
$$

So each movie can be represented by a feature vector:

$$
x^{(i)} =
\begin{bmatrix}
x_1^{(i)}\\
x_2^{(i)}
\end{bmatrix}
$$

For example, the PDF gives:

$$
x^{(1)}=
\begin{bmatrix}
0.9\\
0
\end{bmatrix}
$$

for **Love at Last**.

And:

$$
x^{(3)}=
\begin{bmatrix}
0.99\\
0
\end{bmatrix}
$$

for **Cute Puppies of Love**.

So:

```text
Movie
  ↓
Features
  ↓
[romance, action]
```

---

# 3. User's preferences

Now we need to represent what a particular user likes.

For Alice, the lecture uses:

$$
w^{(1)}
=
\begin{bmatrix}
5\\
0
\end{bmatrix}
$$

and

$$
b^{(1)}=0
$$

This means Alice's parameters indicate a strong preference toward the first feature in this example.

The important point is:

> **Movie → has features \(x^{(i)}\)**
> **User → has parameters \(w^{(j)}, b^{(j)}\)**

This is the setup shown on the PDF's collaborative-filtering slide.

---

# 4. Prediction Formula

Now comes the main formula of this topic.

For **user \(j\)** and **movie \(i\)**:

$$
\boxed{
\hat y(i,j)=w^{(j)}\cdot x^{(i)}+b^{(j)}
}
$$

Where:

* \(x^{(i)}\) = features of movie \(i\)
* \(w^{(j)}\) = preferences/parameters of user \(j\)
* \(b^{(j)}\) = bias for user \(j\)
* \(\hat y(i,j)\) = predicted rating

The PDF explicitly describes this as essentially **linear regression**.

---

# 5. Example from the PDF

For Alice:

$$
w^{(1)}
=
\begin{bmatrix}
5\\
0
\end{bmatrix}
$$

For movie 3:

$$
x^{(3)}
=
\begin{bmatrix}
0.99\\
0
\end{bmatrix}
$$

and:

$$
b^{(1)}=0
$$

Therefore:

$$
\hat y(3,1)
=
w^{(1)}\cdot x^{(3)}+b^{(1)}
$$

Calculate the dot product:

$$
=
(5)(0.99)+(0)(0)+0
$$

$$
=4.95
$$

So the predicted rating is:

$$
\boxed{4.95}
$$

This exact calculation appears in the PDF.

---

# 6. Why is this called Collaborative Filtering?

The important idea is that we **don't manually specify Alice's preferences**.

We want the algorithm to **learn**:

$$
w^{(1)},b^{(1)}
$$

from the ratings Alice has already given.

Similarly, for every other user:

$$
w^{(2)},b^{(2)}
$$

$$
w^{(3)},b^{(3)}
$$

$$
\cdots
$$

The PDF then introduces the cost function to learn these parameters.

---

## 7. The key structure

Remember this:

$$
\boxed{
\text{User parameters}
+
\text{Movie features}
\rightarrow
\text{Predicted rating}
}
$$

or:

$$
\boxed{
w^{(j)}\cdot x^{(i)}+b^{(j)}
}
$$

This is the **core prediction equation** for the collaborative-filtering approach introduced here.

---

### What you should know from this topic

| Symbol                           | Meaning                              |
| -------------------------------- | ------------------------------------ |
| \(x^{(i)}\)                      | Features of movie \(i\)              |
| \(w^{(j)}\)                      | Parameters/preferences of user \(j\) |
| \(b^{(j)}\)                      | Bias of user \(j\)                   |
| \(\hat y(i,j)\)                  | Predicted rating                     |
| \(w^{(j)}\cdot x^{(i)}+b^{(j)}\) | Prediction                           |

**Main formula:**

$$
\boxed{\hat y(i,j)=w^{(j)}\cdot x^{(i)}+b^{(j)}}
$$



# Topic 3 — Collaborative Filtering Cost Function

## 1. Concept

We have the prediction:

$$
\hat y^{(i,j)}
=
w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

We need to **learn** \(w^{(j)}\) and \(b^{(j)}\) so that our predicted ratings are close to the ratings the user actually gave.

For this, we define a **cost function**.

---

## 2. Cost Function for One User

For user \(j\), the cost function is:

$$
\boxed{
J(w^{(j)},b^{(j)})
=
\frac{1}{2}
\sum_{i:r(i,j)=1}
\left(
w^{(j)}\cdot x^{(i)}+b^{(j)}-y^{(i,j)}
\right)^2
+
\frac{\lambda}{2}
\sum_{k=1}^{n}
(w_k^{(j)})^2
}
$$

There are two important parts.

### Part 1 — Prediction error

$$
\frac{1}{2}
\sum_{i:r(i,j)=1}
\left(
w^{(j)}\cdot x^{(i)}+b^{(j)}-y^{(i,j)}
\right)^2
$$

This measures:

$$
\boxed{\text{Predicted rating} - \text{Actual rating}}
$$

for movies that the user has actually rated.

The condition

$$
r(i,j)=1
$$

means **only use movies that user \(j\) has rated**.

---

## 3. Why do we subtract the actual rating?

Suppose Alice actually rated a movie:

$$
y^{(i,1)}=5
$$

and our model predicts:

$$
\hat y^{(i,1)}=4.5
$$

Then the error is:

$$
4.5-5=-0.5
$$

Squaring it:

$$
(-0.5)^2=0.25
$$

So the cost gets a contribution of \(0.25/2\).

If the prediction is very wrong, the squared error becomes larger.

---

## 4. Regularization Term

The second part is:

$$
\boxed{
\frac{\lambda}{2}
\sum_{k=1}^{n}(w_k^{(j)})^2
}
$$

This is the **regularization term**.

It penalizes large values of the weights \(w\).

So the cost has:

$$
\boxed{
\text{Cost}
=
\text{Prediction error}
+
\text{Regularization}
}
$$

The PDF includes this regularization term in the collaborative-filtering cost function.

---

# 5. Cost Function for All Users

We don't only have Alice.

We have:

$$
j=1,2,\ldots,n_u
$$

So we need to learn:

$$
w^{(1)},b^{(1)}
$$

$$
w^{(2)},b^{(2)}
$$

$$
\cdots
$$

$$
w^{(n_u)},b^{(n_u)}
$$

The complete cost function becomes:

$$
\boxed{
J=
\frac{1}{2}
\sum_{j=1}^{n_u}
\sum_{i:r(i,j)=1}
\left(
w^{(j)}\cdot x^{(i)}
+b^{(j)}
-y^{(i,j)}
\right)^2
+
\frac{\lambda}{2}
\sum_{j=1}^{n_u}
\sum_{k=1}^{n}
(w_k^{(j)})^2
}
$$

This combines the cost across **all users**.

---

## 6. What are we minimizing?

We want to find parameters that make the cost as small as possible:

$$
\boxed{
\min_{w^{(j)},b^{(j)}} J
}
$$

In simple words:

```text
Choose w and b
      ↓
Make predictions
      ↓
Compare with actual ratings
      ↓
Calculate error
      ↓
Minimize total error
```

---

## 7. Important Understanding

Suppose Alice has rated only 4 movies.

We **do not calculate the error using an unknown `?` rating**.

We calculate:

$$
\sum_{i:r(i,1)=1}
$$

So only Alice's known ratings participate in her cost.

For example:

```text
Movie 1 → 5  ✓ use
Movie 2 → 5  ✓ use
Movie 3 → ?  ✗ don't use
Movie 4 → 0  ✓ use
Movie 5 → 0  ✓ use
```

That's what the condition

$$
\boxed{r(i,j)=1}
$$

is doing.

---

## 8. One-line Summary

$$
\boxed{
\text{Cost Function}
=
\text{How wrong our predictions are}
+
\text{Penalty for large weights}
}
$$

And we minimize this cost to learn the users' \(w\) and \(b\).


# Topic 4 — Collaborative Filtering Algorithm

## 1. Concept

We have learned how to define the cost function. Now we need to **find the values of the parameters** that minimize this cost.

The parameters we want to learn are:

$$
w^{(1)}, b^{(1)}, w^{(2)}, b^{(2)}, \ldots
$$

for all users.

We use **gradient descent** to minimize the cost function.

---

## 2. What are we learning?

For each user \(j\):

$$
w^{(j)}
$$

and

$$
b^{(j)}
$$

For each movie \(i\), we also learn its feature vector:

$$
x^{(i)}
$$

So collaborative filtering learns **both**:

$$
\boxed{\text{User parameters } w,b}
$$

and

$$
\boxed{\text{Movie features }x}
$$

---

## 3. Cost Function

The combined objective is:

$$
\boxed{
\min_{w,b,x} J(w,b,x)
}
$$

where:

$$
J(w,b,x)
=
\frac{1}{2}
\sum_{j=1}^{n_u}
\sum_{i:r(i,j)=1}
\left(
w^{(j)}\cdot x^{(i)}
+b^{(j)}
-y^{(i,j)}
\right)^2
$$

plus regularization terms for \(w\) and \(x\).

---

## 4. Gradient Descent

We repeatedly update the parameters in the direction that **reduces the cost**.

General gradient descent:

$$
w := w-\alpha\frac{\partial J}{\partial w}
$$

$$
b := b-\alpha\frac{\partial J}{\partial b}
$$

where:

* \(\alpha\) = learning rate
* \(\frac{\partial J}{\partial w}\) = derivative of cost with respect to \(w\)
* \(\frac{\partial J}{\partial b}\) = derivative of cost with respect to \(b\)

The PDF then applies this idea to the collaborative-filtering parameters.

---

## 5. What gets updated?

For collaborative filtering, gradient descent updates:

$$
\boxed{w^{(j)}}
$$

$$
\boxed{b^{(j)}}
$$

and

$$
\boxed{x^{(i)}}
$$

So the algorithm is simultaneously learning:

```text
Users
  ↓
w, b

Movies
  ↓
x
```

while trying to minimize the rating prediction error.

---

## 6. Important Idea

Earlier, we had:

$$
\hat y^{(i,j)}
=
w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

Initially, we don't know the correct \(w\), \(b\), or \(x\).

Gradient descent repeatedly adjusts them:

$$
\boxed{
w,b,x
\rightarrow
\text{make prediction}
\rightarrow
\text{calculate cost}
\rightarrow
\text{update}
\rightarrow
\text{repeat}
}
$$

until the cost is minimized.

---

## 7. Key Point

Collaborative filtering **learns the movie features instead of requiring us to manually specify them**.

So rather than saying beforehand:

```text
Movie 1:
romance = 0.9
action  = 0
```

the algorithm can learn useful feature values \(x^{(i)}\) from the users' ratings.

This is the central idea behind the collaborative-filtering algorithm presented in the lecture.


# Topic 5 — Binary Labels: Likes, Favourites and Clicks

## 1. Concept

Until now, we used ratings such as:

$$
0,1,2,3,4,5
$$

But many recommender systems don't have star ratings.

Instead, they observe whether the user **interacted with an item or not**.

So the target becomes a **binary label**:

$$
\boxed{y^{(i,j)} \in \{0,1\}}
$$

The PDF gives examples such as purchases, favourites/likes, time spent, and clicks.

---

## 2. Meaning of the Labels

The PDF defines:

$$
y^{(i,j)} =
\begin{cases}
1 & \text{user engaged with the item}\\
0 & \text{user did not engage with the item}\\
? & \text{item has not yet been shown}
\end{cases}
$$

For example:

| Movie                | Alice | Bob | Carol | Dave |
| -------------------- | ----: | --: | ----: | ---: |
| Love at Last         |     1 |   1 |     0 |    0 |
| Romance Forever      |     1 |   ? |     ? |    0 |
| Cute Puppies of Love |     ? |   1 |     0 |    ? |
| Nonstop Car Chases   |     0 |   0 |     1 |    1 |
| Swords vs. Karate    |     0 |   0 |     1 |    ? |

This binary version of the movie matrix is shown in the PDF.

---

## 3. What Can `1` Represent?

Depending on the recommender system, `1` could mean:

* User purchased the item after seeing it.
* User liked/favourited the item.
* User spent at least 30 seconds on the item.
* User clicked the item.

For example, in a video recommender:

```text
Video shown → User clicked      → y = 1
Video shown → User didn't click → y = 0
Video not shown                 → ?
```

---

## 4. Important: `0` vs `?`

These are **not the same**.

### `0`

The item **was shown**, but the user didn't engage.

$$
\boxed{0=\text{shown but no engagement}}
$$

### `?`

The item **has not been shown**, so we don't know what the user would do.

$$
\boxed{?=\text{unknown}}
$$

This distinction is explicitly made in the lecture.

---

## 5. Prediction Changes

Previously, we predicted a numerical rating:

$$
\hat y=w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

Now we want to predict:

$$
\boxed{P(y^{(i,j)}=1)}
$$

That is:

> **What is the probability that user \(j\) will engage with item \(i\)?**

For binary labels, the PDF applies the **sigmoid function**:

$$
\boxed{
f(x)=g(w^{(j)}\cdot x^{(i)}+b^{(j)})
}
$$

where:

$$
\boxed{
g(z)=\frac{1}{1+e^{-z}}
}
$$

This converts the output into a value between:

$$
0 \leq f(x) \leq 1
$$

so it can represent a probability.

---

## 6. Example

Suppose:

$$
w^{(j)}\cdot x^{(i)}+b^{(j)}=2
$$

Then:

$$
g(2)=\frac{1}{1+e^{-2}}
$$

$$
g(2)\approx0.881
$$

So:

$$
\boxed{P(y^{(i,j)}=1)\approx88.1\%}
$$

The model predicts a high probability that the user will engage with that item.

---

## 7. Key Difference

### Rating-based

$$
\boxed{\hat y=w^{(j)}\cdot x^{(i)}+b^{(j)}}
$$

Predict:

> **How much will the user rate the item?**

### Binary

$$
\boxed{
\hat y=g(w^{(j)}\cdot x^{(i)}+b^{(j)})
}
$$

Predict:

> **What is the probability that the user will engage with the item?**


# Topic 6 — From Regression to Binary Classification

## 1. Concept

For numerical ratings, we used:

$$
\hat y^{(i,j)}
=
w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

For binary labels, we want the output to represent a **probability**.

So we apply the **sigmoid function** to the linear prediction.

---

## 2. Prediction

The linear part is:

$$
z=w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

Then apply sigmoid:

$$
\boxed{
f_{w,b,x}(x^{(i)})
=
g\left(w^{(j)}\cdot x^{(i)}+b^{(j)}\right)
}
$$

where:

$$
\boxed{
g(z)=\frac{1}{1+e^{-z}}
}
$$

---

## 3. Why Sigmoid?

The linear expression:

$$
w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

can produce any value:

$$
-\infty < z < \infty
$$

But a probability must be between:

$$
0 \leq P \leq 1
$$

The sigmoid converts \(z\) into this range:

$$
\boxed{
-\infty < z < \infty
\quad\rightarrow\quad
0 < g(z) < 1
}
$$

---

## 4. Interpretation

The output represents:

$$
\boxed{
P(y^{(i,j)}=1\mid x^{(i)},j)
}
$$

In words:

> Probability that user \(j\) will engage with item \(i\).

---

## 5. Example

Suppose:

$$
z=w^{(j)}\cdot x^{(i)}+b^{(j)}=2
$$

Then:

$$
g(2)=\frac{1}{1+e^{-2}}
$$

$$
g(2)\approx0.88
$$

Therefore:

$$
\boxed{P(y=1)\approx0.88}
$$

So the model predicts about an **88% probability of engagement**.

---

## 6. Comparison

### Regression

$$
\boxed{
\hat y=w\cdot x+b
}
$$

Output can be any real number.

### Binary classification

$$
\boxed{
\hat y=g(w\cdot x+b)
}
$$

Output is between 0 and 1 and can represent a probability.

---

## Key Formula

$$
\boxed{
P(y^{(i,j)}=1)
=
g(w^{(j)}\cdot x^{(i)}+b^{(j)})
}
$$

with

$$
\boxed{
g(z)=\frac{1}{1+e^{-z}}
}
$$


# Topic 7 — Cost Function for Binary Applications

## 1. Concept

For binary recommendations, the output is:

$$
f_{w,b,x}(x^{(i)})=
g(w^{(j)}\cdot x^{(i)}+b^{(j)})
$$

where the output represents:

$$
P(y^{(i,j)}=1)
$$

Now we need a **cost function** that measures how well this probability prediction matches the actual binary label.

---

## 2. Loss for One Example

For one training example, the loss is:

$$
\boxed{
L(f(x),y)
=
-y\log(f(x))
-(1-y)\log(1-f(x))
}
$$

where:

* \(y\) = actual label (`0` or `1`)
* \(f(x)\) = predicted probability

This is the **logistic loss / binary cross-entropy loss** shown in the PDF.

---

## 3. When \(y=1\)

If the actual label is:

$$
y=1
$$

then:

$$
L(f(x),1)
=
-\log(f(x))
$$

Suppose the model predicts:

$$
f(x)=0.9
$$

Then:

$$
L=-\log(0.9)
$$

which is small.

So:

$$
\boxed{\text{Correct high probability → small loss}}
$$

---

## 4. When \(y=0\)

If:

$$
y=0
$$

then:

$$
L(f(x),0)
=
-\log(1-f(x))
$$

Suppose:

$$
f(x)=0.1
$$

Then:

$$
L=-\log(0.9)
$$

Again, the loss is small.

So:

$$
\boxed{\text{Correct low probability → small loss}}
$$

---

## 5. What if the prediction is wrong?

Suppose:

$$
y=1
$$

but the model predicts:

$$
f(x)=0.01
$$

Then:

$$
L=-\log(0.01)
$$

which is large.

Therefore:

$$
\boxed{\text{Very confident wrong prediction → large loss}}
$$

---

## 6. Cost for All Examples

The individual losses are combined to form the cost:

$$
\boxed{
J(w,b,x)
=
\sum_{(i,j):r(i,j)=1}
L\left(f_{w,b,x}(x^{(i)}),y^{(i,j)}\right)
}
$$

with regularization terms also included in the complete objective.

The goal remains:

$$
\boxed{\min J(w,b,x)}
$$

---

## 7. Key Idea

For binary recommender systems:

$$
\boxed{
\text{Prediction}
\rightarrow
\text{Probability}
\rightarrow
\text{Log Loss}
\rightarrow
\text{Minimize Cost}
}
$$

### Remember

$$
\boxed{
L(f(x),y)
=
-y\log(f(x))
-(1-y)\log(1-f(x))
}
$$

* Actual `1` → want predicted probability close to **1**
* Actual `0` → want predicted probability close to **0**
* Confident wrong prediction → **large loss**


# Topic 8 — Mean Normalization

## 1. Concept

Mean normalization is used in the recommender-system implementation to make the rating data easier for the model to learn from.

The lecture demonstrates this using a user who has **not rated any movies**.

---

## 2. Problem

Suppose we have:

| Movie                | Alice | Bob | Carol | Dave | Eve |
| -------------------- | ----: | --: | ----: | ---: | --: |
| Love at Last         |     5 |   5 |     0 |    0 |   ? |
| Romance Forever      |     5 |   ? |     ? |    0 |   ? |
| Cute Puppies of Love |     ? |   4 |     0 |    ? |   ? |
| Nonstop Car Chases   |     0 |   0 |     5 |    4 |   ? |
| Swords vs. Karate    |     0 |   0 |     5 |    ? |   ? |

Eve has **not rated any movie**.

If we directly use the original rating matrix, there is no information from Eve's ratings to help determine her preferences.

---

## 3. Calculate Mean Rating

For each movie, calculate the average of the known ratings.

The lecture gives:

$$
\mu =
\begin{bmatrix}
2.5\\
2.5\\
2\\
2.25\\
1.25
\end{bmatrix}
$$

So, for example, the mean for Movie 1 is:

$$
\mu_1=\frac{5+5+0+0}{4}=2.5
$$

---

## 4. Normalize the Ratings

For each known rating:

$$
\boxed{
y_{\text{norm}}^{(i,j)}
=
y^{(i,j)}-\mu_i
}
$$

So we subtract the movie's mean rating from each known rating.

For Movie 1:

$$
5-2.5=2.5
$$

and:

$$
0-2.5=-2.5
$$

Therefore the normalized values for Movie 1 become:

$$
[2.5,\;2.5,\;-2.5,\;-2.5]
$$

The lecture's normalized matrix shows these deviations from the movie means.

---

## 5. Why Normalize?

After normalization:

* Positive value → rating is **above the movie's average**
* Negative value → rating is **below the movie's average**
* Zero → rating is **equal to the movie's average**

Example:

$$
2.5
$$

means the user's rating was 2.5 points above that movie's average.

---

## 6. Prediction After Normalization

The model predicts the **normalized rating**:

$$
w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

Then we add the movie mean back:

$$
\boxed{
\hat y^{(i,j)}
=
w^{(j)}\cdot x^{(i)}
+b^{(j)}
+\mu_i
}
$$

This allows the model to produce the final predicted rating.

---

## 7. Important Point

Mean normalization changes the problem from:

$$
\text{predict the raw rating}
$$

to:

$$
\boxed{\text{predict how much the rating differs from the movie's average}}
$$

Then:

$$
\boxed{
\text{Final prediction}
=
\text{predicted difference}
+
\text{movie mean}
}
$$

The lecture uses this normalization specifically in its recommender-system implementation.
