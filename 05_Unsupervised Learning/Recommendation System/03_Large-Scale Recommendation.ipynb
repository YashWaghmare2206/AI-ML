## 18. Large-Scale Recommendation

Suppose a platform has:

$$
1,000,000
$$

movies/items.

We want to recommend maybe **10 items** to one user.

Doing a complicated ranking calculation on all 1,000,000 items would be expensive.

So the system uses:

$$
\boxed{
    1,000,000
\rightarrow
\text{Retrieval}
\rightarrow
1,000
\rightarrow
\text{Ranking}
\rightarrow
10
}
$$

The PDF describes this two-stage structure: retrieval first produces a smaller set of candidates, then ranking selects the final recommendations.

---

# Stage 1 — Retrieval

### Goal

**Quickly find a relatively small set of items that could be relevant.**

Suppose the user is interested in:

```text
Action + Sci-Fi
```

There may be 1,000,000 total items, but perhaps only a few thousand are potentially relevant.

The retrieval system tries to get those candidates.

For example:

$$
1,000,000
\rightarrow
1,000
$$

These 1,000 items are called **candidate items**.

### Important

Retrieval does **not** need to perfectly order the 1,000 items.

Its main job is:

> **Don't miss potentially good items while reducing the enormous catalogue to a manageable candidate set.**

---

# Stage 2 — Ranking

Now we only have:

$$
1,000
$$

candidate items instead of:

$$
1,000,000
$$

The ranking model can now perform a **more detailed calculation** for each candidate.

For example, it could consider:

* how well the item matches the user
* previous user interactions
* item characteristics
* other features used by the recommendation system

It assigns a score to each candidate:

$$
s_1,s_2,\ldots,s_{1000}
$$

Then sort them:

$$
s_{(1)} \geq s_{(2)} \geq \cdots
$$

and select the items at the top.

For example:

$$
1,000
\rightarrow
10
$$

These 10 are the items ultimately shown to the user.

---

# Why not just rank all 1,000,000?

This is the key reason for the architecture.

Imagine ranking one item requires some computational work.

Doing that for:

$$
1,000,000
$$

items is much more expensive than doing it for:

$$
1,000
$$

items.

So:

### Without retrieval

$$
1,000,000
\rightarrow
\text{expensive ranking}
\rightarrow
10
$$

### With retrieval

$$
1,000,000
\xrightarrow{\text{fast retrieval}}
1,000
\xrightarrow{\text{detailed ranking}}
10
$$

The expensive model only processes the **small candidate set**.

---

# A concrete example

Imagine YouTube has millions of videos.

You open YouTube.

### Retrieval

The system might quickly identify candidates such as:

```text
Video A
Video B
Video C
...
Video Z
```

from the huge catalogue based on signals such as the user's interests and previously watched content.

Suppose it retrieves:

$$
1,000 \text{ candidate videos}
$$

### Ranking

Now a more detailed model evaluates those 1,000 candidates and produces scores:

| Video | Score |
| ----- | ----: |
| A     |  0.97 |
| B     |  0.32 |
| C     |  0.91 |
| D     |  0.84 |
| ...   |   ... |

After sorting:

$$
A,\ C,\ D,\ldots
$$

the system can select the highest-scoring items to show.

---

# The most important distinction

### Retrieval asks:

> **"Which items are worth considering?"**

### Ranking asks:

> **"Among those candidates, which ones should be placed at the top?"**

So:

$$
\boxed{\text{Retrieval = candidate generation}}
$$

$$
\boxed{\text{Ranking = detailed ordering of candidates}}
$$

The PDF specifically describes retrieval as selecting a relatively small set of candidates from a large catalogue, followed by ranking those candidates.

And importantly, **retrieval doesn't mean the system knows the final 10 items immediately**. It first creates a pool that contains potentially useful items; the ranking stage makes the more detailed selection.


## 19. Retrieval

Now let's look specifically at **how the retrieval stage works**.

The purpose of retrieval is to take a huge catalogue and produce a **small set of candidate items** that are potentially relevant to the user.

### 19.1 Input

We start with a user and a large catalogue.

For example:

$$
\text{Catalogue} = 1,000,000\text{ items}
$$

The system has information about the user, including their learned representation:

$$
v_{\text{user}}
$$

and item representations:

$$
v_1,v_2,\ldots,v_{1,000,000}
$$

---

### 19.2 Candidate Generation

The retrieval system searches for items that could be relevant to the user.

Conceptually:

$$
\text{User}
\rightarrow
\text{Retrieval system}
\rightarrow
\text{Candidate items}
$$

For example:

$$
1,000,000
\rightarrow
1,000
$$

So instead of passing one million items to the expensive ranking stage, we only pass 1,000 candidates.

---

### 19.3 Multiple Sources of Candidates

An important point in the PDF is that the candidate set can be generated from **multiple retrieval sources**.

For example, candidates could come from different recommendation methods:

$$
\text{Candidates}
=
\text{source 1}
+
\text{source 2}
+
\text{source 3}
+\cdots
$$

The resulting candidates are then passed to the ranking system.

This allows the system to retrieve potentially useful items using different signals.

---

### 19.4 Retrieval vs. Final Recommendation

Retrieval is **not the final decision**.

It produces:

$$
\boxed{\text{Candidate set}}
$$

The ranking stage then evaluates these candidates in more detail.

So the complete flow is:

$$
\boxed{
\text{Large catalogue}
\rightarrow
\underbrace{\text{Retrieval}}_{\text{find candidates}}
\rightarrow
\text{Candidate set}
\rightarrow
\underbrace{\text{Ranking}}_{\text{order candidates}}
\rightarrow
\text{Recommendations}
}
$$

### Key idea

The retrieval stage prioritizes **speed and recall**: it should quickly bring potentially relevant items into the candidate set so that the ranking model can examine them more carefully.


## 20. Ranking

After **retrieval** has reduced the huge catalogue to a manageable set of candidates, the **ranking stage** decides which candidates should appear first.

The PDF describes ranking as taking the retrieved candidate items and assigning them scores based on how relevant they are to the user.

### 20.1 Input to Ranking

Suppose retrieval produced:

$$
1000 \text{ candidates}
$$

For each candidate, the ranking model can use information about:

* the user
* the candidate item
* the user's previous interactions
* other relevant features

Conceptually:

$$
(\text{user},\text{item})
\rightarrow
\text{ranking model}
\rightarrow
\text{score}
$$

---

### 20.2 Calculate a Score

For every candidate item \(i\), the model produces a score:

$$
s_i = f(\text{user},\text{item}_i)
$$

For example:

| Candidate | Score |
| --------- | ----: |
| Movie A   |  0.91 |
| Movie B   |  0.43 |
| Movie C   |  0.87 |
| Movie D   |  0.72 |

The score represents the model's estimate of how relevant that item is to the user.

---

### 20.3 Sort the Candidates

After calculating the scores, the candidates are ordered:

$$
0.91 > 0.87 > 0.72 > 0.43
$$

So:

$$
\text{Movie A}
\rightarrow
\text{Movie C}
\rightarrow
\text{Movie D}
\rightarrow
\text{Movie B}
$$

The highest-ranked items can then be shown to the user.

---

### 20.4 Why use a separate Ranking Stage?

Because now we're working with a **much smaller set**.

For example:

$$
1,000,000
\xrightarrow{\text{Retrieval}}
1,000
\xrightarrow{\text{Ranking}}
10
$$

The ranking model can therefore afford to perform a **more detailed computation** on each candidate.

The important distinction is:

$$
\boxed{
\text{Retrieval: Which items should we consider?}
}
$$

$$
\boxed{
\text{Ranking: In what order should we present those items?}
}
$$

This retrieval → ranking architecture is the large-scale recommendation approach described in the PDF.


## 21. Retrieval vs. Ranking Trade-off

The key issue is that **retrieval and ranking have different goals**.

### Retrieval

Retrieval needs to be:

* **Fast**
* Able to handle a **very large catalogue**
* Good at finding a broad set of potentially relevant items

Its goal is to avoid missing good candidates while reducing the catalogue size.

$$
\boxed{
\text{Large catalogue}
\rightarrow
\text{Fast retrieval}
\rightarrow
\text{Candidate set}
}
$$

### Ranking

Ranking can be more computationally expensive because it only works on the smaller candidate set.

Its goal is to determine the **best ordering** of those candidates.

$$
\boxed{
\text{Candidate set}
\rightarrow
\text{Detailed ranking}
\rightarrow
\text{Final recommendations}
}
$$

### The important trade-off

Suppose:

$$
1,000,000 \rightarrow 100
$$

The retrieval stage is extremely aggressive.

That's computationally efficient, but there is a risk:

> A potentially good item might never enter the 100 candidates.

If that happens, the ranking model **cannot recommend it**, because ranking only sees the retrieved candidates.

So retrieval needs to balance:

$$
\boxed{\text{Efficiency} \quad \text{vs.} \quad \text{not missing relevant items}}
$$

The PDF describes using **offline experiments** to evaluate different retrieval approaches and this trade-off.

### Simple example

Imagine 1,000,000 movies.

**Option A:**

$$
1,000,000 \rightarrow 10
$$

Very fast, but many potentially relevant movies may be excluded.

**Option B:**

$$
1,000,000 \rightarrow 10,000
$$

More candidates reach ranking, reducing the chance of missing relevant items, but ranking now has much more work.

So the system has to choose a suitable candidate-set size.

### Overall picture

$$
\boxed{
\underbrace{1,000,000}_{\text{catalogue}}
\xrightarrow[\text{fast}]{\text{Retrieval}}
\underbrace{1,000\text{ candidates}}_{\text{manageable}}
\xrightarrow[\text{detailed}]{\text{Ranking}}
\underbrace{10}_{\text{recommendations}}
}
$$

The central idea is that **retrieval determines what is eligible to be ranked, while ranking determines the order of the eligible items**.
