# SECTION 2 — Content-Based Filtering

## 10. Finding Related Items

In collaborative filtering, we learned **item feature vectors** \(x^{(i)}\). We can now use these vectors to find items that are similar to each other.

### Core idea

Suppose we have two movies:

$$
x^{(i)} =
\begin{bmatrix}
0.9\\
0.1
\end{bmatrix}
$$

and

$$
x^{(k)} =
\begin{bmatrix}
0.8\\
0.2
\end{bmatrix}
$$

To determine how similar they are, calculate the squared distance:

$$
d(i,k)=\|x^{(i)}-x^{(k)}\|^2
$$

For these two movies:

$$
d=(0.9-0.8)^2+(0.1-0.2)^2
$$

$$
=0.01+0.01=0.02
$$

A **smaller distance means the two items are more similar**.

### General formula

For item \(i\), find another item \(k\) such that:

$$
\boxed{
k=\arg\min_k\|x^{(i)}-x^{(k)}\|^2
}
$$

This gives the item whose learned feature vector is closest to item \(i\).

### Important point

The \(x^{(i)}\) vectors were **learned during collaborative filtering**.

So we can use those learned vectors to answer:

> **"Which other items have similar learned features to this item?"**

However, the individual learned features may not have an obvious human interpretation.

### Key takeaway

$$
\boxed{\text{Similar item vectors} \Rightarrow \text{Similar items}}
$$

This is the first building block for recommending items based on their characteristics.


## 11. Cold Start Problem

The **cold start problem** occurs when a recommender system has **new users or new items** for which there is little or no interaction data.

The PDF focuses on two cases:

### 1. New item

Suppose a new movie is added.

It has **no ratings yet**:

$$
\text{Movie X} \rightarrow \text{no user ratings}
$$

A collaborative filtering system has little information to learn the movie's \(x^{(i)}\) vector from.

Therefore, it becomes difficult to recommend that movie.

---

### 2. New user

Suppose a new user creates an account.

They haven't rated or interacted with anything yet:

$$
\text{New User} \rightarrow \text{no interaction history}
$$

The system doesn't yet know what this user likes.

Therefore, personalized recommendations are difficult.

---

### How can this be addressed?

The PDF introduces **side information** as a way to help with the cold-start problem.

For example, information about:

* the item
* the user
* other characteristics available outside the interaction history

can provide additional information for making recommendations.

### Key idea

Collaborative filtering mainly relies on:

$$
\boxed{\text{User-item interactions}}
$$

But for a **new user/item**, those interactions don't exist yet.

That's why **cold start** is a challenge for recommender systems.


## 12. Collaborative Filtering vs Content-Based Filtering

The two approaches both try to recommend items to users, but they use **different sources of information**.

---

### 12.1 Collaborative Filtering

Collaborative filtering uses **user-item interactions** such as ratings, likes, clicks, or purchases.

For example:

| User    | Movie A | Movie B | Movie C |
| ------- | ------: | ------: | ------: |
| Alice   |       5 |       4 |       ? |
| Bob     |       5 |       4 |       5 |
| Charlie |       1 |       2 |       1 |

The system can learn relationships from these interactions.

In the approach we've studied:

$$
\hat y^{(i,j)}
=
w^{(j)}\cdot x^{(i)}+b^{(j)}
$$

The system learns both:

* \(w^{(j)},b^{(j)}\) → user parameters
* \(x^{(i)}\) → item features

So the item features themselves can be learned from the interaction data.

---

### 12.2 Content-Based Filtering

Content-based filtering uses **information/features describing the user and item**.

For example, a movie could have features such as:

$$
x^{(i)}=
[\text{action},\text{romance},\text{comedy},\text{drama}]
$$

A user can also be represented by their preferences for these features.

The system then tries to determine:

> **Does this item's content/features match this user's preferences?**

The PDF describes this approach as using information about the **user and item**, rather than relying only on the interaction matrix.

---

## 12.3 Main Difference

The simplest way to remember it:

|                     | Collaborative Filtering          | Content-Based Filtering                         |
| ------------------- | -------------------------------- | ----------------------------------------------- |
| Main information    | User-item interactions           | User/item features                              |
| Learns from         | Ratings, likes, clicks, etc.     | Characteristics of users/items                  |
| Main question       | "What did similar users like?"   | "Does this item match this user's preferences?" |
| Item representation | Can be learned from interactions | Based on item features                          |
| New item problem    | Can struggle if no interactions  | Can use item features                           |

---

### 12.4 Example

Suppose **Alice likes**:

* Action Movie A
* Action Movie B
* Action Movie C

#### Collaborative filtering

The system may observe:

> Users who liked A and B also liked D.

Therefore:

$$
\text{Recommend D to Alice}
$$

It doesn't necessarily need to understand what makes A, B, and D "action movies."

---

#### Content-based filtering

The system knows that Alice prefers:

$$
[\text{action}=high,\text{romance}=low]
$$

and Movie D has:

$$
[\text{action}=high,\text{romance}=low]
$$

So D matches Alice's preferences.

$$
\text{User preference vector}
\approx
\text{Item feature vector}
$$

Therefore, D can be recommended.

---

### 12.5 Important Connection

The distinction is **not simply**:

> Collaborative filtering = no features
> Content-based = features

Collaborative filtering can **learn item features from interaction data**, as we saw earlier.

The important distinction is **where the information comes from**:

$$
\boxed{
\text{Collaborative Filtering}
\rightarrow
\text{interaction data}
}
$$

$$
\boxed{
\text{Content-Based Filtering}
\rightarrow
\text{user/item descriptive information}
}
$$

The PDF then moves toward using **user and item features to learn how well a user and item match**.


## 13. User & Item Features

Content-based filtering needs information that describes **the user** and **the item**.

The PDF gives examples of features that can be used for both.

### 13.1 User Features

A user can be represented using features describing their preferences or characteristics.

For example:

$$
x_{\text{user}}=
\begin{bmatrix}
\text{preference for romance}\\
\text{preference for action}\\
\text{age}
\end{bmatrix}
$$

The exact features depend on the application.

---

### 13.2 Item Features

An item can also be represented by features describing it.

For a movie, for example:

$$
x_{\text{movie}}=
\begin{bmatrix}
\text{romance}\\
\text{action}\\
\text{comedy}\\
\text{drama}
\end{bmatrix}
$$

So instead of representing a movie only by its ID, we represent it using information about the movie.

---

### 13.3 User–Item Matching

Now the goal is to determine whether the user's preferences match the item's features.

Conceptually:

$$
\boxed{
\text{User features}
\rightarrow
\text{matching model}
\rightarrow
\text{item preference score}
}
$$

A high score means the item is a good match for that user.

---

### Important distinction from Collaborative Filtering

In collaborative filtering, we learned item features from the **interaction data**.

In content-based filtering, we can use **descriptive information about users and items** to construct their representations.


## 14. Learning to Match

The main idea here is to **learn how well a particular user matches a particular item**.

Instead of manually deciding:

> “This user likes action movies, so recommend action movies.”

we build a model that **learns the matching relationship from data**.

### 14.1 User and Item Vectors

Represent a user with a feature vector:

$$
x_{\text{user}}
$$

and an item with another feature vector:

$$
x_{\text{item}}
$$

For example:

$$
x_{\text{user}}=
\begin{bmatrix}
0.9\\
0.2\\
0.1
\end{bmatrix}
$$

$$
x_{\text{item}}=
\begin{bmatrix}
0.8\\
0.3\\
0.1
\end{bmatrix}
$$

The model needs to learn a function:

$$
f(x_{\text{user}},x_{\text{item}})
$$

that produces a **compatibility/relevance score**.

---

### 14.2 What does the score mean?

Conceptually:

$$
\boxed{
\text{User}+\text{Item}
\rightarrow
\text{Matching model}
\rightarrow
\text{Score}
}
$$

A higher score means the model predicts that the user is more likely to be interested in that item.

For example:

| User  | Movie   | Matching score |
| ----- | ------- | -------------: |
| Alice | Movie A |           0.92 |
| Alice | Movie B |           0.31 |
| Alice | Movie C |           0.78 |

The recommender can use these scores to identify items that are relevant to Alice.

---

### 14.3 Why "Learning to Match"?

Because we are **not manually specifying the rules** for what makes a user and item compatible.

The model learns the relationship from training data:

$$
\boxed{
\text{Training interactions}
\rightarrow
\text{learn matching function}
}
$$

This idea leads directly into using **neural networks** to learn the user and item representations and their matching.

### Key takeaway

> **Learning to match = learn a function that measures how well a user and an item fit together.**


## 15. Deep Learning for Content-Based Filtering

The next step is to use a **neural network** to learn the user and item representations used for matching.

The PDF's key idea is:

$$
\boxed{
\text{User features}
\rightarrow
\text{Neural Network}
\rightarrow
\text{User vector}
}
$$

and separately:

$$
\boxed{
\text{Item features}
\rightarrow
\text{Neural Network}
\rightarrow
\text{Item vector}
}
$$

These two learned vectors are then used to determine how well the user and item match.

### 15.1 Two Neural Networks

There are **two networks**:

**User network**

$$
x_{\text{user}}
\rightarrow
NN_{\text{user}}
\rightarrow
v_{\text{user}}
$$

**Item network**

$$
x_{\text{item}}
\rightarrow
NN_{\text{item}}
\rightarrow
v_{\text{item}}
$$

where:

* \(x_{\text{user}}\) = input features describing the user
* \(x_{\text{item}}\) = input features describing the item
* \(v_{\text{user}}\) = learned user vector
* \(v_{\text{item}}\) = learned item vector

---

### 15.2 Why Two Networks?

Because users and items have **different types of information**.

For example:

```text
User information
      ↓
 User Network
      ↓
User vector

Movie information
      ↓
Movie Network
      ↓
Movie vector
```

The two networks transform their respective inputs into vectors in the **same representation space**.

---

### 15.3 Matching the Vectors

Once we have:

$$
v_{\text{user}}
$$

and

$$
v_{\text{item}}
$$

we can measure how well they match.

The PDF uses the idea of comparing these learned vectors to determine the recommendation score.

Conceptually:

$$
\boxed{
v_{\text{user}} \quad \text{vs.} \quad v_{\text{item}}
\rightarrow
\text{matching score}
}
$$

A higher matching score means the model predicts that the user is more likely to be interested in the item.

### The overall architecture

$$
\boxed{
\text{User features}
\rightarrow NN_{\text{user}}
\rightarrow v_{\text{user}}
}
$$

$$
\boxed{
\text{Item features}
\rightarrow NN_{\text{item}}
\rightarrow v_{\text{item}}
}
$$

$$
\boxed{
v_{\text{user}},v_{\text{item}}
\rightarrow
\text{matching score}
}
$$

This is the **deep-learning version of content-based filtering** described in the PDF.


## 17. Learned User & Item Vectors

The neural networks from the previous topic transform the original user and item features into **learned vectors**.

### 17.1 User Vector

The user network produces:

$$
v_{\text{user}}
$$

For example, the network might produce a **32-dimensional vector**:

$$
v_{\text{user}}\in\mathbb{R}^{32}
$$

### 17.2 Item Vector

The item/movie network produces:

$$
v_{\text{item}}
$$

with the same dimension:

$$
v_{\text{item}}\in\mathbb{R}^{32}
$$

The PDF specifically gives the example of **32-dimensional learned user and movie vectors**.

---

### 17.3 How are they used?

The vectors can be compared using their dot product:

$$
\boxed{
\text{score}
=
v_{\text{user}}\cdot v_{\text{item}}
}
$$

For example:

$$
v_{\text{user}}=[0.8,0.2,0.6]
$$

$$
v_{\text{item}}=[0.7,0.4,0.5]
$$

Then:

$$
v_{\text{user}}\cdot v_{\text{item}}
=
(0.8)(0.7)+(0.2)(0.4)+(0.6)(0.5)
$$

$$
=0.56+0.08+0.30
$$

$$
=0.94
$$

A higher score indicates a stronger match.

---

### 17.4 What makes these vectors "learned"?

We don't manually decide:

> Dimension 1 = action
> Dimension 2 = romance
> Dimension 3 = comedy

Instead, the neural networks **learn the vector representations from the training data**.

The training objective teaches the networks to produce representations that make users and items with positive interactions have high matching scores.

Therefore:

$$
\boxed{
\text{Raw features}
\rightarrow
\text{Neural Network}
\rightarrow
\text{Learned vector}
}
$$

and then:

$$
\boxed{
v_{\text{user}}\cdot v_{\text{item}}
\rightarrow
\text{matching score}
}
$$

### 17.5 Finding Recommendations

Once the user vector has been learned, we can compare it with the vectors of many items:

$$
v_{\text{user}}\cdot v_1
$$

$$
v_{\text{user}}\cdot v_2
$$

$$
\vdots
$$

$$
v_{\text{user}}\cdot v_n
$$

Items with higher matching scores can be considered more relevant to that user.
