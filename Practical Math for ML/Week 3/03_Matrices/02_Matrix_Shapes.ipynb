## 2. Matrix Shapes

---

## Formula

The **shape** of a matrix is written as:



where:

* (m) = Number of **rows**
* (n) = Number of **columns**

---

## Explanation

The shape tells us the **size** of the matrix.

To find the shape:

1. Count the rows.
2. Count the columns.

Always write it as:

```text
(rows, columns)
```

**Not**

```text
(columns, rows) ❌
```

---

## Example

### Example 1


---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(A.shape)
```

**Output**

```text
(2, 3)
```

Another example:

```python
B = np.array([
    [7],
    [8],
    [9]
])

print(B.shape)
```

**Output**

```text
(3, 1)
```

---

## Important Note

### NumPy Vector vs Matrix

```python
v = np.array([1, 2, 3])
print(v.shape)
```

Output:

```text
(3,)
```

This is a **1D vector**, **not** a matrix.

---

```python
v = np.array([[1],
              [2],
              [3]])
print(v.shape)
```

Output:

```text
(3, 1)
```

This **is** a matrix (specifically, a column matrix).

Similarly,

```python
v = np.array([[1, 2, 3]])
print(v.shape)
```

Output:

```text
(1, 3)
```

This is a **row matrix**.

> **Remember:** In NumPy,
>
> * `[1,2,3]` → Shape `(3,)` → 1D vector
> * `[[1],[2],[3]]` → Shape `(3,1)` → Column matrix
> * `[[1,2,3]]` → Shape `(1,3)` → Row matrix

This distinction becomes important when performing **matrix multiplication**, which is the next topic.
