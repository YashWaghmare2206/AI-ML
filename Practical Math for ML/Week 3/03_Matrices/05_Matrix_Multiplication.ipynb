# Explanation

Unlike matrix addition, **matrix multiplication is NOT element-wise**.

Instead:

* Take **one row** from the first matrix.
* Take **one column** from the second matrix.
* Compute their **dot product**.
* That becomes one element in the result matrix.

So matrix multiplication is basically **repeated dot products**.

---

# Rule for Matrix Multiplication

---

### Resulting Shape



A quick trick:

```text
(m × n) × (n × p)

      n = n ✅

Result:

(m × p)
```

# Example

Let

[
A=
\begin{bmatrix}
1 & 2\
3 & 4
\end{bmatrix}
]

Shape:

```text
(2,2)
```

and

[
B=
\begin{bmatrix}
5 & 6\
7 & 8
\end{bmatrix}
]

Shape:

```text
(2,2)
```

Since

```text
Columns of A = 2
Rows of B    = 2
```

Multiplication is possible.

---

### Step 1 → First Row × First Column

Take

```text
Row:
[1 2]
```

and

```text
Column:
[5
 7]
```

Dot product:

[
(1\times5)+(2\times7)
=====================

# 5+14

19
]

This becomes the first element.

---

### Step 2 → First Row × Second Column

Row:

```text
[1 2]
```

Column:

```text
[6
 8]
```

Dot product:

[
(1\times6)+(2\times8)
=====================

# 6+16

22
]

---

### Step 3 → Second Row × First Column

Row:

```text
[3 4]
```

Column:

```text
[5
 7]
```

Dot product:

[
15+28
=====

43
]

---

### Step 4 → Second Row × Second Column

Row:

```text
[3 4]
```

Column:

```text
[6
 8]
```

Dot product:

[
18+32
=====

50
]

---

Final answer:

[
A\times B=
\begin{bmatrix}
19 & 22\
43 & 50
\end{bmatrix}
]

# Python / NumPy

Use either `@` or `np.matmul()`.

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A @ B)
```

**Output**

```text
[[19 22]
 [43 50]]
```

Equivalent:

```python
np.matmul(A, B)
```
