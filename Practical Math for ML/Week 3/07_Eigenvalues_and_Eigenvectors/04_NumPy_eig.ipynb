# 07_Eigenvalues_and_Eigenvectors

## 4. NumPy `eig()`

---

## Purpose

Instead of manually solving:

[
\det(A-\lambda I)=0
]

NumPy can directly compute:

* Eigenvalues
* Eigenvectors

using:

```python
np.linalg.eig()
```

---

## Syntax

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
```

Returns:

```text
eigenvalues  → λ values

eigenvectors → corresponding vectors
```

---

## Example

[
A=
\begin{bmatrix}
2 & 0\
0 & 3
\end{bmatrix}
]

Code:

```python
import numpy as np

A = np.array([
    [2, 0],
    [0, 3]
])

values, vectors = np.linalg.eig(A)

print(values)
print(vectors)
```

Output:

```text
[2. 3.]

[[1. 0.]
 [0. 1.]]
```

---

## Understanding the Output

Eigenvalues:

```text
[2. 3.]
```

means:

```text
λ₁ = 2
λ₂ = 3
```

---

Eigenvectors:

```text
[[1. 0.]
 [0. 1.]]
```

Read **column-wise**, not row-wise.

First column:

[
\begin{bmatrix}
1\
0
\end{bmatrix}
]

is the eigenvector for:

[
\lambda=2
]

---

Second column:

[
\begin{bmatrix}
0\
1
\end{bmatrix}
]

is the eigenvector for:

[
\lambda=3
]

---

## Verification

Take:

[
v=
\begin{bmatrix}
1\
0
\end{bmatrix}
]

Multiply:

[
Av
==

\begin{bmatrix}
2\
0
\end{bmatrix}
]

which equals:

[
2v
]

Therefore:

```text
Eigenvector ✓
Eigenvalue = 2 ✓
```

---

## Another Example

```python
A = np.array([
    [4, 2],
    [1, 3]
])

values, vectors = np.linalg.eig(A)

print(values)
print(vectors)
```

NumPy computes everything automatically.

In real ML work, this is what you'll use.

---

## ML Connection

Many ML algorithms internally compute eigenvalues/eigenvectors:

* PCA
* Spectral Clustering
* Covariance Analysis
* Recommendation Systems
* Image Compression

Most of the time:

```python
np.linalg.eig()
```

is all you'll need.

---

## Important Note

### Eigenvectors are stored as columns

Given:

```text
[[a b]
 [c d]]
```

Read as:

```text
Eigenvector 1 = [a,c]

Eigenvector 2 = [b,d]
```

Not row-wise.

---

### Memory Trick

```text
np.linalg.eig(A)

↓

Returns

(values, vectors)

↓

Eigenvalues first

Eigenvectors second
```
