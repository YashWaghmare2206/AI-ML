In practice, however, **we do not compute the inverse manually**. NumPy provides a direct solver.

## Explanation

To solve a system of linear equations in NumPy, use:

```python
np.linalg.solve(A, B)
```

where:

* `A` = Coefficient matrix
* `B` = Constant matrix (or vector)

NumPy computes the unknown values in **X**.

---

## Example



### Step 3: Solve for X

```python
import numpy as np

A = np.array([
    [2, 3],
    [4, 1]
])

B = np.array([8, 10])

X = np.linalg.solve(A, B)

print(X)
```

**Output**

```text
[2.2 1.2]
```

So,

---

## Important Note

### Use `np.linalg.solve()` instead of computing the inverse.

✅ Recommended:

```python
X = np.linalg.solve(A, B)
```

❌ Less preferred:

```python
X = np.linalg.inv(A) @ B
```

Why?

* `np.linalg.solve()` is **faster**.
* It is **more numerically stable**.
* It avoids explicitly computing the inverse, which can introduce more floating-point error.
