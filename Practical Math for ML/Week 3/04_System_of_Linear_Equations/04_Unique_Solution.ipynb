## Explanation

A **unique solution** means there is **exactly one set of values** for the unknown variables that satisfies **all the equations**.

For example:



and **no other values** satisfy the system.

Graphically (for two variables), this happens when the two lines **intersect at exactly one point**.

---

## Example



---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 1],
    [1, -1]
])

B = np.array([5, 1])

X = np.linalg.solve(A, B)

print(X)
```

**Output**

```text
[3. 2.]
```

---

## ML Connection

Many ML algorithms assume a **unique solution** exists.

Examples:

* Linear Regression (under certain conditions)
* Solving normal equations
* Optimization problems

If the solution is unique, the algorithm can find one definite set of model parameters.

## Important Note

For **2 variables**, think of the equations as **two lines**:
