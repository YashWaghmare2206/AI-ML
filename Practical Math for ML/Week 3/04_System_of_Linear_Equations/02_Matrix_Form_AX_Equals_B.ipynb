## Python / NumPy

```python
import numpy as np

A = np.array([
    [2, 3],
    [4, 1]
])

X = np.array([
    [1],
    [2]
])

print(A @ X)
```

**Output**

```text
[[8]
 [6]]
```

Notice that this computes the left-hand side of the equations for the chosen values (x=1) and (y=2). It is **not solving** the system yet.

---

## ML Connection

Many ML problems are written in matrix form.

Example:

[
\boxed{Y = XW}
]

where:

* (X) = Feature matrix
* (W) = Weight matrix
* (Y) = Predictions

This has the same structure as:

[
AX=B
]

Learning ML becomes much easier once you're comfortable reading matrix equations.
