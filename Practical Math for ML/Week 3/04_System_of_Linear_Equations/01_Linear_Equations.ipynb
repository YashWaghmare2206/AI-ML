## Formula

A **linear equation** in one variable is:

**`ax+b=0`**

where:

* (a) and (b) are constants.
* (x) is the unknown variable.
* (a \neq 0).

---

For two variables:

**`ax+by=c`**

For three variables:

**`ax+by+cz=d`**

---

## Explanation

A **linear equation** is an equation where:

* Every variable has **power 1**.
* Variables are **not multiplied together**.
* There are **no square roots, exponents, or trigonometric functions** involving the variables.

## ML Connection

Many ML problems are based on solving systems of linear equations.

Examples:

* Linear Regression
* Least Squares
* Optimization
* Matrix Inversion
* PCA

Most ML models eventually reduce to solving equations involving unknown parameters.

---

## Important Note

A **single linear equation** can have:

* One unknown (`x`)
* Two unknowns (`x, y`)
* Three unknowns (`x, y, z`)
* Or even thousands of unknowns in ML.

When we have **multiple linear equations together**, it is called a **System of Linear Equations**, which is the next topic.
