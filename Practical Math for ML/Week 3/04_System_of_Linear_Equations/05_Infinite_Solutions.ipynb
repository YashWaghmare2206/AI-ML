## 5. Infinite Solutions

---

## Formula

There is **no special formula** for infinite solutions.

In general:

* The equations represent the **same line** (for 2 variables).
* One equation is a **multiple of the other**.

Also,



so the coefficient matrix is **singular**.

---

## Explanation

A system has **infinite solutions** when **both equations describe the same relationship**.

Instead of intersecting at one point, they lie exactly on top of each other.

That means **every point on that line satisfies both equations**.

---

## Example


---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [1, 1],
    [2, 2]
])

B = np.array([2, 4])

np.linalg.solve(A, B)
```

Output:

```text
LinAlgError: Singular matrix
```

Why?

Because there is **no unique solution**.

`np.linalg.solve()` is designed to solve systems with **exactly one solution**, so it raises an error for singular matrices.

---

## ML Connection

Infinite solutions can occur in ML when:

* Features are **duplicates**.
* One feature is a multiple of another.
* There is **multicollinearity**.

Example:

```text
Height (cm)

Height (m)
```

These two features contain the same information.

This can make the coefficient matrix singular, leading to infinitely many valid parameter values.
