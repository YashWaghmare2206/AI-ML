---

## Explanation

This formula tells us the **angle between two vectors**.

To find the angle:

1. Compute the **dot product**.
2. Compute the **magnitude** of both vectors.
3. Divide the dot product by the product of the magnitudes.
4. Apply the inverse cosine ((\cos^{-1})).

The result is the angle between the vectors.

---

## Example





---

## Python / NumPy

```python
import numpy as np

A = np.array([1, 0])
B = np.array([0, 1])

dot = np.dot(A, B)

angle = np.arccos(
    dot / (np.linalg.norm(A) * np.linalg.norm(B))
)

print(np.degrees(angle))
```

**Output**

```text
90.0
```

> `np.arccos()` returns the angle in **radians**.
> `np.degrees()` converts it to **degrees**.

---

## ML Connection

The angle between vectors is useful because it measures **directional similarity**.

Examples:

* Two text embeddings with a **small angle** represent similar meanings.
* Recommendation systems compare user/item vectors based on their angle.
* This concept is the foundation of **Cosine Similarity**, which you'll learn next.

---

## Important Note

Different angles indicate different relationships:

| Angle                              | Meaning                    |
| ---------------------------------- | -------------------------- |
| (0^\circ)                          | Same direction             |
| Between (0^\circ) and (90^\circ)   | Similar direction          |
| (90^\circ)                         | Perpendicular (orthogonal) |
| Between (90^\circ) and (180^\circ) | Opposite tendency          |
| (180^\circ)                        | Exactly opposite direction |
