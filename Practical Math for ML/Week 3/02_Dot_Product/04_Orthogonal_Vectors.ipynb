## Python / NumPy

```python
import numpy as np

A = np.array([2, 3])
B = np.array([3, -2])

dot = np.dot(A, B)

print(dot)
```

**Output**

```text
0
```

Since the dot product is **0**, the vectors are orthogonal.

---

## ML Connection

Orthogonal vectors appear in many ML and mathematics topics:

* **Principal Component Analysis (PCA)** → Principal components are orthogonal.
* **QR Decomposition**
* **Orthogonal basis vectors**
* Reducing redundancy between features.

Orthogonality is useful because orthogonal vectors are **independent in direction**, meaning they don't overlap.

---

## Important Note

A dot product of **0** means the vectors are **perpendicular**, **not** that one of the vectors is the zero vector.

For example:

```text
A = [1, 0]
B = [0, 1]
```

Both are non-zero vectors, yet:



Also remember:

* **Positive dot product** → Angle is **less than 90°**
* **Zero dot product** → Angle is **90°**
* **Negative dot product** → Angle is **greater than 90°**
