## Explanation

The dot product is one of the most frequently used operations in ML.

Whenever a model combines **features** with **weights**, it is usually computing a **dot product**.

---

## Example 1: Linear Regression

Suppose a house has these features:

```text
Area = 1200
Bedrooms = 3
```

Feature vector:

[
\mathbf{x}=[1200,;3]
]

Model weights:

[
\mathbf{w}=[200,;50000]
]

Prediction:

[
y=\mathbf{w}\cdot\mathbf{x}+b
]

Dot product:

[
(200\times1200)+(50000\times3)
]

[
=240000+150000
]

[
=390000
]

If

[
b=10000
]

then

[
y=400000
]

---

## Example 2: Neural Networks

Suppose

Features:

[
\mathbf{x}=[2,;4]
]

Weights:

[
\mathbf{w}=[0.5,;0.8]
]

The neuron first computes:

[
\mathbf{w}\cdot\mathbf{x}
]

[
=(0.5\times2)+(0.8\times4)
]

[
=1+3.2
]

[
=4.2
]

Then it adds the bias and passes the result through an activation function.

---

## Python / NumPy

```python
import numpy as np

x = np.array([2, 4])
w = np.array([0.5, 0.8])
b = 1

prediction = np.dot(w, x) + b

print(prediction)
```

**Output**

```text
5.2
```

---

## ML Connection

The dot product is used in almost every major ML algorithm:

| Algorithm                     | Use of Dot Product                      |
| ----------------------------- | --------------------------------------- |
| Linear Regression             | Compute predictions                     |
| Logistic Regression           | Compute prediction score before sigmoid |
| Neural Networks               | Compute neuron outputs                  |
| Support Vector Machines (SVM) | Decision boundary calculations          |
| PCA                           | Matrix/vector computations              |
| Recommendation Systems        | User-item similarity                    |
| Transformers                  | Attention scores (Query · Key)          |
| Embeddings                    | Similarity calculations                 |

If you study deep learning later, you'll see **millions or billions of dot products** being computed during training and inference.

---

## Important Note

From now on, whenever you see an equation like:

[
\mathbf{w}\cdot\mathbf{x}
]

you should immediately recognize it as:

* Multiply corresponding elements.
* Add the products.
* Get **one scalar value**.

This scalar is then used for predictions, classifications, similarity calculations, or as input to the next computation.
