## Explanation

Previously, we learned the **algebraic formula**:



This computes the dot product using the vector components.

The **geometric formula** shows what the dot product means geometrically.

It depends on three things:

1. The length of vector **A**.
2. The length of vector **B**.
3. The angle between them.

The value of the dot product changes as the angle changes.

---

## Example





---

## Python / NumPy

```python
import numpy as np

A = np.array([3, 4])
B = np.array([4, 0])

dot = np.dot(A, B)

print(dot)
```

**Output**

```text
12
```

Here, NumPy uses the **algebraic formula** internally. Both formulas always produce the **same result**.

## ML Connection

The geometric interpretation is important because it tells us **how aligned two vectors are**.

Many ML applications use this idea:

* **Cosine Similarity** compares the angle between vectors.
* **Text embeddings**: Similar sentences have vectors pointing in similar directions.
* **Recommendation systems**: Similar users/items have vectors with small angles between them.

## Important Note

The value of the dot product depends on the angle:

| Angle ((\theta)) | (\cos\theta) |    Dot Product   |
| ---------------: | :----------: | :--------------: |
|        (0^\circ) |       1      | Maximum positive |
|       (90^\circ) |       0      |         0        |
|      (180^\circ) |      -1      | Maximum negative |

This leads to an important observation:

* Vectors pointing in the **same direction** → large positive dot product.
* Vectors that are **perpendicular** → dot product is **0**.
* Vectors pointing in **opposite directions** → negative dot product.
