## **User Input**

```python
import numpy as np

def matrix_creation():
    rows = int(input("Rows: "))
    cols = int(input("Columns: "))

    matrix = []

    for i in range(rows):
        while True:
            row = list(map(float, input(f"Row {i+1}: ").split()))

            if len(row) == cols:
                matrix.append(row)
                break

            print(f"Please enter exactly {cols} values.")

    return np.array(matrix)


n = int(input("Enter the number of matrices needed: "))

arr_list = []

for i in range(n):
    print(f"\nEnter Matrix {i+1}")
    arr_list.append(matrix_creation())


print("\nStored Matrices:")
for i, matrix in enumerate(arr_list, start=1):
    print(f"\nMatrix {i}")
    print(matrix)
    print("Shape:", matrix.shape)
```

## **Main Logic**

```python
operation = {
    1: "Matrix Addition",
    2: "Matrix Multiplication",
    3: "Transpose",
    4: "Determinant",
    5: "Inverse",
    6: "Eigenvalues",
    7: "Eigenvectors"
}

c = int(input(f"Enter the choice : {operation}\n"))

if c not in operation:
    print("Invalid Choice")
else:
    print(f"\n{operation[c]}")

    if c in [3, 4, 5, 6, 7]:
        m = int(input("Enter Matrix Number: ")) - 1
    else:
        m = int(input("Enter 1st Matrix Number: ")) - 1
        n = int(input("Enter 2nd Matrix Number: ")) - 1

    match c:

        case 1:
            if arr_list[m].shape == arr_list[n].shape:
                print(arr_list[m] + arr_list[n])
            else:
                print("Matrices must have same shape")

        case 2:
            if arr_list[m].shape[1] == arr_list[n].shape[0]:
                print(arr_list[m] @ arr_list[n])
            else:
                print("Matrix multiplication not possible")

        case 3:
            print(arr_list[m].T)

        case 4:
            if arr_list[m].shape[0] == arr_list[m].shape[1]:
                print(np.linalg.det(arr_list[m]))
            else:
                print("Matrix must be square")

        case 5:
            if arr_list[m].shape[0] == arr_list[m].shape[1]:
                if np.linalg.det(arr_list[m]) != 0:
                    print(np.linalg.inv(arr_list[m]))
                else:
                    print("Inverse does not exist")
            else:
                print("Matrix must be square")

        case 6:
            if arr_list[m].shape[0] == arr_list[m].shape[1]:
                eig_values, eig_vectors = np.linalg.eig(arr_list[m])
                print(eig_values)
            else:
                print("Matrix must be square")

        case 7:
            if arr_list[m].shape[0] == arr_list[m].shape[1]:
                eig_values, eig_vectors = np.linalg.eig(arr_list[m])
                print(eig_vectors)
            else:
                print("Matrix must be square")
```
