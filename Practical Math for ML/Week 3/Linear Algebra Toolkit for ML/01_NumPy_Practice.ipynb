```python
import numpy as np
```

## **Vectors**

### **1. Vector Addition**

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

r = a + b
print(r)
```

### **2. Vector Subtraction**

```python
r = a - b
print(r)
```

### **3. Dot Product**

```python
d1 = np.dot(a , b) # Both are same method
d2 = a @ b

print(d1)
print(d2)
```

### **4. Magnitude (Norm)**

```python
magnitude = np.linalg.norm(a)
print(magnitude)
```

### **5. Distance Between Vectors**

```python
distance = np.linalg.norm((a-b))
print(distance)
```

## **Matrices**

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
```

### **1. Matrix Addition**

```python
r = A + B
print(r)
```

### **2. Matrix Multiplication**

```python
r1 = np.matmul(A , B)
r2 = A @ B
print(r1)
print(r2)
```

### **3. Transpose**

```python
t1 = A.T
t2 = np.transpose(A)
print(t1)
print(t2)
```

### **4. Determinant**

```python
det = np.linalg.det(A)
print(det)
```

### **5. Inverse**

```python
inverse = np.linalg.inv(A)
print(inverse)
```

### **6. Eigenvalues & Eigenvectors**

```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

| Operation                | Function                 |
| ------------------------ | ------------------------ |
| Vector Addition          | `a + b`                  |
| Vector Subtraction       | `a - b`                  |
| Dot Product              | `np.dot(a,b)` or `a @ b` |
| Magnitude                | `np.linalg.norm(a)`      |
| Distance                 | `np.linalg.norm(a-b)`    |
| Matrix Addition          | `A + B`                  |
| Matrix Multiplication    | `A @ B`                  |
| Transpose                | `A.T`                    |
| Determinant              | `np.linalg.det(A)`       |
| Inverse                  | `np.linalg.inv(A)`       |
| Eigenvalues/Eigenvectors | `np.linalg.eig(A)`       |
| Solve Ax=b               | `np.linalg.solve(A,b)`   |
