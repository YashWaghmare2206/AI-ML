## What is Q?

(Q) contains special vectors that are:

* Unit vectors (length = 1)
* Perpendicular to each other

Such vectors are called **orthonormal**.

## Python / NumPy

```python id="hlgkgj"
import numpy as np

A = np.array([
    [1, 1],
    [1, -1]
])

Q, R = np.linalg.qr(A)

print(Q)
print(R)
```

---

The function returns:

```text id="4gbl7u"
Q → Orthogonal Matrix

R → Upper Triangular Matrix
```

## ML Connection

QR decomposition is used in:
