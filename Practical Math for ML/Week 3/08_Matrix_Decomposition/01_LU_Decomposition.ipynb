## What is Matrix Decomposition?

Matrix decomposition means:

```text id="93upfy"
One Matrix
↓
Break into smaller matrices
```

Just like:

```text id="hrsb4g"
12 = 3 × 4
```

we break a matrix into simpler matrices.

---

## LU Decomposition

A matrix (A) is decomposed into:

[
\boxed{A=LU}
]

where:

* (L) = Lower Triangular Matrix
* (U) = Upper Triangular Matrix

## Python

NumPy does not directly provide LU decomposition.

Usually SciPy is used:

```python id="gbvc4v"
from scipy.linalg import lu

P, L, U = lu(A)
```

For your roadmap, just know the concept.

---

## ML Connection

LU decomposition is used in:

* Solving large systems of equations
* Numerical optimization
* Scientific computing
* Internally by many linear algebra libraries

You usually won't call LU directly in beginner ML, but libraries use it behind the scenes.

---

## Important Note

Don't memorize how to manually compute LU decomposition.

For AI/ML, remember:

```text id="gpp0gd"
A
↓
LU Decomposition
↓
L × U
↓
Faster solving of AX = B
```
