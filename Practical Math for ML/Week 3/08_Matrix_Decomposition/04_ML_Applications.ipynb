## Why Did We Learn LU, QR, and SVD?

The goal is not to manually decompose matrices.

The goal is to understand how ML libraries solve problems efficiently.

---

## 1. Solving Linear Systems

You learned:

[
AX=B
]

For small matrices:

```python
np.linalg.solve(A, B)
```

works fine.

For very large matrices, libraries internally use decompositions such as:

```text
LU
QR
```

to solve faster.

---

## 2. Linear Regression

Linear Regression often requires solving large equation systems.

Instead of directly computing inverses:

```text
Inverse ❌
```

libraries often use:

```text
QR ✅
```

because it is more numerically stable.

---

## 3. PCA

Remember:

```text
Eigenvectors → Directions

Eigenvalues → Importance
```

PCA can be implemented using:

```text
Eigen Decomposition

or

SVD
```

In practice, many libraries use SVD.

---

## 4. Dimensionality Reduction

Suppose a dataset has:

```text
100 Features
```

Many features contain little information.

SVD helps identify:

```text
Most important directions
```

and remove less useful ones.

---

## 5. Recommendation Systems

Movie recommendations, product recommendations, etc.

Example:

* User × Movie matrix
* User × Product matrix

SVD helps discover hidden patterns.

Used historically in systems like Netflix recommendation engines.

---

## 6. Image Compression

An image can be represented as a matrix.

SVD allows us to keep:

```text
Large singular values
```

and discard:

```text
Tiny singular values
```

Result:

```text
Smaller file size
+
Similar image quality
```

---

## What Should You Remember?

### LU

```text
A = LU

Used for solving equations efficiently.
```

---

### QR

```text
A = QR

Used for least squares
and linear regression.
```

---

### SVD

```text
A = UΣVᵀ

Used for:

PCA
Compression
Recommendations
Dimensionality Reduction
```
