# Dimension

Dimension is directly connected to **Basis**.

In fact:

> **Dimension = Number of vectors in a basis**

---

## Definition

The dimension of a vector space is the number of vectors required in its basis.

If a basis contains:

* 1 vector → dimension = 1
* 2 vectors → dimension = 2
* 3 vectors → dimension = 3

---

Only one basis vector is needed.

Therefore:

---

## Intuition

Dimension tells us:

> How many independent directions exist in the space?

### Line

Only one direction.

Dimension = 1.

### Plane

Two independent directions.

Dimension = 2.

### 3D Space

Three independent directions.

Dimension = 3.

---

## Relationship with Span

Recall:

* One independent vector → spans a line.
* Two independent vectors → span a plane.
* Three independent vectors → span a 3D space.

The number of independent vectors needed is exactly the dimension.

---

## Key Takeaways

* Dimension = number of vectors in a basis.
* Dimension = number of independent directions.
* Line → dimension 1.
* Plane → dimension 2.
* 3D space → dimension 3.
* Dependent vectors do not increase dimension.
* Dimension counts independent vectors, not total vectors.

### One-line Summary

> Dimension tells us how many independent directions are needed to represent a space.
