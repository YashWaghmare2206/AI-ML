# Span

Span is one of the most important concepts in Linear Algebra because **Basis**, **Dimension**, **Linear Independence**, **PCA**, and **SVD** all depend on it.

---

## Definition

The **span** of a set of vectors is the set of **all possible linear combinations** of those vectors.

## Why Span Matters in ML

Suppose your dataset has features:

* Height
* Weight

Every data point is a vector:



The span tells us:

**What region of the feature space can be represented using those vectors.**

Later:

* Basis = minimum vectors needed to create the span
* Dimension = size of the span
* PCA = find directions that span most variance
* SVD = decomposition using important spanning directions

So span is the foundation for many ML algorithms.

---

## Key Takeaways

* Span = all possible linear combinations of vectors.
* One non-zero vector spans a line.
* Two independent vectors span a plane.
* Three independent vectors span 3D space.
* Dependent vectors do not increase span.
* Every span contains the zero vector.
* Span describes the space that can be generated from given vectors.
