## Why Orthogonality Matters in ML

Orthogonal features contain different information.

Examples:

* PCA produces orthogonal principal components.
* SVD uses orthogonal vectors.
* Orthogonal directions help avoid redundant information.

## Key Takeaways

* Orthogonal = perpendicular.
* Two vectors are orthogonal if their dot product is zero.
* Dot product zero means angle = 90°.
* Orthogonal vectors are independent.
* Orthogonal vectors represent different directions.
* PCA and SVD heavily rely on orthogonality.

### One-line Summary

> Orthogonality means vectors are perpendicular, which mathematically is equivalent to having a dot product of zero.
