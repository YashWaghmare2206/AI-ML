## Geometric Meaning

An orthogonal matrix can:

* Rotate vectors
* Reflect vectors

But it cannot:

* Stretch them
* Shrink them

The length remains unchanged.

## Angle Preservation

Orthogonal matrices also preserve angles.

If two vectors form an angle before transformation, they form the same angle afterward.

Therefore orthogonal transformations preserve geometry.

---

## Why Does This Happen?

The columns of an orthogonal matrix are:

1. Unit vectors (length = 1)
2. Mutually orthogonal



---

## Orthonormal

A set of vectors is orthonormal if:

* Orthogonal to each other
* Each vector has magnitude 1

Orthogonal Matrix = Matrix whose columns are orthonormal vectors.
