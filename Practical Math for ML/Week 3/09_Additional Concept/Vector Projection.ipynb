# Vector Projection

Vector Projection answers the question:

> "How much of one vector lies in the direction of another vector?"

This is one of the most important concepts behind:

* Linear Regression
* PCA
* Least Squares
* SVD

---

## Intuition



You want to know:

> How much of (a) points in the direction of (b)?

The answer is the **projection of (a) onto (b)**.

---

## Geometric Idea

Imagine dropping a perpendicular from vector (a) onto vector (b).

The shadow formed on (b) is the projection.

That shadow tells us how much of (a) is aligned with (b).

## Why Projection Matters in ML

### Linear Regression

Predictions are projections of data onto a feature space.

### PCA

Projects data onto principal components.

### SVD

Uses projections repeatedly.

### Similarity Analysis

Projection measures how strongly one vector aligns with another.

---

## Relationship with Previous Topics

### Orthogonality

Projection becomes zero if vectors are orthogonal.

### Basis

Coordinates of a vector in a basis are obtained through projections.

### Dimension

Projection allows movement between spaces of different dimensions.
