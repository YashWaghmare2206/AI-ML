# Reflection Matrix

A reflection matrix **flips** vectors across a line.

Think of it as creating a mirror image.

Examples:

* Reflection across x-axis
* Reflection across y-axis
* Reflection across the line (y=x)

## What Happens to Length?

Reflection preserves length.

If

**`||v||=5`**

then after reflection:

**`||Rv||=5`**

---

## What Happens to Angles?

Angles are preserved.

Only orientation changes.

## Connection to Previous Topics

### Orthogonality

Reflection matrices are orthogonal matrices.

### Linear Transformations

Reflection is a linear transformation.

### Basis

Reflection changes basis vectors.

### Rotation

Both rotation and reflection preserve length and angles.

Difference:

* Rotation turns.
* Reflection flips.

---

## Key Takeaways



* Preserves lengths.
* Preserves angles.
* Is an orthogonal transformation.

### One-line Summary

> A reflection matrix flips vectors across a line while preserving distances and angles.
