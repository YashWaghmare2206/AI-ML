# ML Perspective

For AI/ML, you rarely manually use these 3D formulas.

The important takeaway is:

* Rotation → Orthogonal matrix
* Scaling → Diagonal matrix
* Reflection → Orthogonal matrix with determinant (-1)
* Shear → General linear transformation

Later in **PCA**, **SVD**, and **Eigenvectors**, you'll see these transformations again, often in much higher dimensions than 3D. A matrix with 100 features is essentially performing these same ideas in a 100-dimensional space.
