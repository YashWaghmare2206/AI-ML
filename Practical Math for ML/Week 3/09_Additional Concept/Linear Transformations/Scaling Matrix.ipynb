# Scaling Matrix

A scaling matrix changes the **size (magnitude)** of vectors.

Unlike rotation:

* Direction may stay the same.
* Length changes.

---

## Definition

A scaling transformation stretches or shrinks vectors.

## Geometric Interpretation

### Uniform Scaling



Everything grows equally.

A square remains a square.

---

### Non-Uniform Scaling



The x-direction stretches more.

A square may become a rectangle.

---

## Properties

### Changes Length

Unlike rotation:



in general.

---

### Does Not Preserve Distance

Distances between points change.

---

### Does Not Preserve Angles

If scaling factors differ, angles can change.

---

### Linear Transformation

Scaling still satisfies:

**`T(u+v)=T(u)+T(v)`**

and

**`T(cu)=cT(u)`**

Therefore it is linear.

---

## Why Important in ML?

Scaling appears everywhere.

### Feature Scaling

Example:

* Age: 0–100
* Salary: 0–1,000,000

We scale features so one does not dominate the other.

### StandardScaler

Transforms data to:

* Mean = 0
* Variance = 1

### MinMaxScaler

Scales data into:

**`[0,1]`**

range.

---

## Connection to Previous Topics

### Linear Transformations

Scaling is a linear transformation.

### Basis

Scaling changes basis vectors.

### Dimension

Scaling does not change dimension.

### Orthogonal Matrices

Scaling matrices are generally **not orthogonal** because they change lengths.

---

## Key Takeaways

* Scaling changes vector size.
* General matrix:



* Uniform scaling stretches equally.
* Non-uniform scaling stretches different directions differently.
* Length changes.
* Angles may change.
* Widely used in feature preprocessing.

### One-line Summary

> A scaling matrix stretches or shrinks vectors by multiplying their coordinates by scaling factors.
