# Linear Transformations

This is the last major topic before the specific transformations (Rotation, Scaling, Reflection, Shear).

---

## Definition

A linear transformation is a function that takes a vector as input and produces another vector as output while preserving linearity.

Usually written as:

**`T(x)`**

or

**`y = Ax`**

where:

* (x) = input vector
* (A) = transformation matrix
* (y) = transformed vector

---

## Intuition

A linear transformation changes vectors by:

* Rotating them
* Scaling them
* Reflecting them
* Shearing them

without breaking the structure of the vector space.

## Two Conditions of Linearity

A transformation is linear if it satisfies both:

### 1. Additivity



Transforming the sum is the same as transforming individually and then adding.

---

### 2. Homogeneity



Scaling before transforming is the same as transforming before scaling.

---

## Why These Conditions Matter

Linear transformations preserve:

* Straight lines
* Vector addition
* Scalar multiplication

They do not create curves or distort the basic structure of the space.

---

## Matrix View

Every matrix represents a linear transformation.

**`y = Ax`**

Think of the matrix as a machine.

Input:

**`x`**

Output:

**`Ax`**

Different matrices perform different transformations.

## Geometric Interpretation

A transformation can:

### Scale

Stretch or shrink.

### Rotate

Change direction.

### Reflect

Flip across a line.

### Shear

Slant the space.

All of these are linear transformations.

## Why Linear Transformations Matter in ML

Almost everything in ML can be viewed as matrix transformations.

Examples:

### Neural Networks

**`y = Wx + b`**

The (Wx) part is a linear transformation.

### PCA

Transforms data into a new basis.

### SVD

Decomposes transformations into simpler ones.

### Computer Vision

Image rotations, scaling, and projections use transformation matrices.

## Key Takeaways

* A linear transformation maps vectors to vectors.
* Usually represented as:

**`y = Ax`**

* Must satisfy:

  * Additivity
  * Homogeneity
* Matrices represent linear transformations.
* Rotations, scaling, reflections, and shears are all linear transformations.
* Basis vectors determine the entire transformation.
