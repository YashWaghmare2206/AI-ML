## Explanation

Scalar multiplication means multiplying **each element of a vector** by a single number (scalar).

Effects:

* If (c > 1): the vector becomes **longer**.
* If (0 < c < 1): the vector becomes **shorter**.
* If (c < 0): the vector changes **direction** and its length is multiplied by (|c|).

The **direction** stays the same for positive scalars and reverses for negative scalars.

---

## Python / NumPy

```python
import numpy as np

v = np.array([2, 3])

print("Multiply by 4 :", 4 * v)
print("Multiply by -2:", -2 * v)
```

**Output**

```text
Multiply by 4 : [ 8 12]
Multiply by -2: [-4 -6]
```
