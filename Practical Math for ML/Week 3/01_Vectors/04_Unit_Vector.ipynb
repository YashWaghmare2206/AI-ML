## Explanation

A **unit vector** is a vector whose **magnitude is exactly 1**.

It keeps the **same direction** as the original vector but changes its length to **1**.

So, when we convert a vector into a unit vector:

* ✅ Direction stays the same.
* ✅ Magnitude becomes 1.

This process is called **normalization**.

---

## Example



---

## Python / NumPy

```python
import numpy as np

v = np.array([3, 4])

unit_vector = v / np.linalg.norm(v)

print(unit_vector)
print(np.linalg.norm(unit_vector))
```

**Output**

```text
[0.6 0.8]
1.0
```

---

## ML Connection

Unit vectors are useful because many ML algorithms care about the **direction** of a vector rather than its length.

Examples:

* **Cosine Similarity**
* **Text Embeddings**
* **Recommendation Systems**
* **Semantic Search**
* **Feature Normalization**

Normalizing vectors makes comparisons fair, even when their magnitudes are very different.
