## Explanation

A vector is simply an **ordered collection of numbers**.

Each number represents one value (called a **component**) in a particular dimension.

For example:

```
[3, 4]
```

has:

* 2 components
* 2 dimensions (2D vector)

Similarly,

```
[2, 5, 7]
```

has:

* 3 components
* 3 dimensions (3D vector)

The **order matters**.

For example,

```
[2, 5]
```

is **not** the same as

```
[5, 2]
```

because each position represents a different dimension.

---

## Example

Suppose we represent a student using three features:

* Height = 170 cm
* Weight = 65 kg
* Age = 22 years

The vector is:



This is a **3-dimensional vector** because it contains 3 components.

---

## Python / NumPy

```python
import numpy as np

v1 = np.array([3, 4])

v2 = np.array([170, 65, 22])

print(v1)
print(v2)

print(v1.shape)
print(v2.shape)
```

**Output**

```text
[3 4]
[170  65  22]

(2,)
(3,)
```

* `(2,)` means a vector with **2 components**.
* `(3,)` means a vector with **3 components**.

---

## ML Connection

In Machine Learning, **every data sample is represented as a vector of features**.

Example dataset:

| Height | Weight | Age |
| ------ | ------ | --- |
| 170    | 65     | 22  |
| 180    | 75     | 25  |
| 160    | 55     | 20  |

Internally, each row is stored as a vector:

```
[170, 65, 22]

[180, 75, 25]

[160, 55, 20]
```

Almost every ML algorithm (Linear Regression, Logistic Regression, SVM, Neural Networks, etc.) works with vectors.
