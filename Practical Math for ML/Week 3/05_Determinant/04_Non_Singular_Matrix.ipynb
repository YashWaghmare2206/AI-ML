## Explanation

A non-singular matrix is simply a matrix whose determinant is **not zero**.

When the determinant is non-zero:

* ✅ Inverse exists
* ✅ Unique solution exists
* ✅ Matrix can be inverted

This is the "good" case in linear algebra.

---

## Example



---

## Why is it non-singular?

Notice:

```text
Row 1 = [2,3]
Row 2 = [4,5]
```

Row 2 is **not** a multiple of Row 1.

The rows contain independent information.

Therefore the determinant is non-zero.

---

## Python / NumPy

```python
import numpy as np

A = np.array([
    [2, 3],
    [4, 5]
])

print(np.linalg.det(A))
```

Output:

```text
-2.0
```

Finding the inverse:

```python
np.linalg.inv(A)
```

Output:

```text
[[-2.5  1.5]
 [ 2.  -1. ]]
```

Notice that NumPy successfully computes the inverse because the matrix is **non-singular**.

---

## ML Connection

Most ML algorithms work best when matrices are non-singular.

Examples:

* Linear Regression
* Matrix inversion
* PCA
* Eigenvalue computations
* Optimization methods

If a matrix becomes singular, many algorithms either fail or need special techniques to handle it.

---

## Important Note

### Determinant ≠ 0

Always remember:

```text
det ≠ 0
↓
Non-Singular Matrix
↓
Inverse Exists
↓
Unique Solution Exists
```

---

### Singular vs Non-Singular

| Property        | Singular          | Non-Singular       |
| --------------- | ----------------- | ------------------ |
| Determinant     | 0                 | ≠ 0                |
| Inverse Exists  | ❌                 | ✅                  |
| Unique Solution | ❌                 | ✅                  |
| Matrix Type     | Bad for inversion | Good for inversion |

---

### Memory Trick

```text
Non-Singular
=
det ≠ 0
=
Invertible
=
Unique Solution
```

This is exactly the opposite of a singular matrix.
