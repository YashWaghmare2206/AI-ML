## Explanation

The determinant is a **single number** computed from a square matrix.

It tells us important things about the matrix:

* Can the matrix be inverted?
* Does a system of equations have a unique solution?
* Is the matrix singular or non-singular?

For now, think of it as:

> A special value calculated from a matrix.

## Python / NumPy

```python id="3t6k7r"
import numpy as np

A = np.array([
    [2, 3],
    [4, 5]
])

print(np.linalg.det(A))
```

**Output**

```text id="y9jnh0"
-2.0
```

## Important Note

### What does determinant tell us?

For now, remember:

| Determinant    | Meaning                  |
| -------------- | ------------------------ |
| (\det(A)\neq0) | Matrix is invertible     |
| (\det(A)=0)    | Matrix is not invertible |

And from the previous topic:

| Determinant    | System of Equations     |
| -------------- | ----------------------- |
| (\det(A)\neq0) | Unique solution         |
| (\det(A)=0)    | Infinite or No solution |
