## 3. Detecting Redundant Features

Suppose a dataset contains:

| Height(cm) | Height(m) |
| ---------- | --------- |
| 170        | 1.70      |
| 180        | 1.80      |
| 160        | 1.60      |

The second feature is just the first feature scaled.

This creates dependency between columns.

Result:

```text
Determinant → 0
Matrix → Singular
```

This tells us there is redundant information.

## What should you actually remember?

For AI/ML interviews, coding, and practical work:

```text
det ≠ 0
→ Matrix is invertible
→ Unique solution exists

det = 0
→ Matrix is singular
→ No inverse
→ Infinite or No solution
```

That's about 90% of what you'll use determinants for in early ML.
