# Conditional Probability

---

## 1. Why Conditional Probability Exists

Normal probability:

```text
P(A)
```

Probability of A happening.

Conditional probability:

```text
P(A|B)
```

Probability of A happening **given that B has already happened**.

---

## 2. Meaning of P(A|B)

Read as:

```text
Probability of A given B
```

Example:

```text
Probability student passes
given they studied.
```

Not:

```text
Probability student passes
```

but

```text
Probability student passes
after knowing they studied.
```

---

## 3. Intuition Behind Conditioning

When new information arrives:

```text
Sample space shrinks
```

Example:

From all students → only students who studied.

Probability changes because our universe changed.

This is the most important intuition.

---

## 4. Conditional Probability Formula

```text
P(A|B) = P(A ∩ B) / P(B)
```

Understand:

```text
A ∩ B
```

means:

```text
A and B both occur
```

---

## 5. Why This Formula Works

Derivation from reduced sample space.

Visual explanation using:

```text
Venn Diagram
```

This makes the formula easy to remember instead of memorizing it.

---

## 6. P(B|A) vs P(A|B)

Very important.

Many students think:

```text
P(A|B) = P(B|A)
```

Wrong.

Example:

```text
P(Disease | Positive Test)

≠

P(Positive Test | Disease)
```

This confusion is exactly why Bayes Theorem exists.

---

## 7. Independent Events & Conditional Probability

If A and B are independent:

```text
P(A|B)=P(A)
```

Knowing B changes nothing.

Connection with previous topic.

---

## 8. Dependent Events

If A affects B:

```text
P(A|B) ≠ P(A)
```

Examples:

```text
Cards without replacement
Rain and umbrella
```

---

## 9. Multiplication Rule

Rearranging formula:

From

```text
P(A|B)=P(A∩B)/P(B)
```

we get:

```text
P(A∩B)=P(A|B)P(B)
```

This is called:

```text
Multiplication Rule
```

Used everywhere in Bayes.

---

## 10. Special Case: Independent Events

Since:

```text
P(A|B)=P(A)
```

Multiplication Rule becomes:

```text
P(A∩B)=P(A)P(B)
```

which you already learned.

## 15. Common Mistakes

### Mistake 1

```text
P(A|B)=P(B|A)
```

False.

---

### Mistake 2

Using:

```text
P(A)
```

instead of:

```text
P(A|B)
```

after information is given.

---

### Mistake 3

Not updating sample space.

Conditional probability is all about:

```text
New Information
→ New Sample Space
→ New Probability
```
