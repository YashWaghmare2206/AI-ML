# Posterior_Distributions

## 1. What is a Distribution?

Before learning **Prior Distribution** and **Posterior Distribution**, we must understand what a **distribution** is.

---

## What is a Distribution?

A distribution tells us:

> How probability is spread across possible values.

---

## Example 1: Dice

Possible outcomes:


**`1,2,3,4,5,6`**


Each outcome has probability:

**`1 / 6`**

Distribution:

| Value | Probability |
| ----- | ----------- |
| 1     | 1/6         |
| 2     | 1/6         |
| 3     | 1/6         |
| 4     | 1/6         |
| 5     | 1/6         |
| 6     | 1/6         |

This table is a probability distribution.

---

## Example 2: Coin

Possible outcomes:

**`H,T`**


Probabilities:

**`P(H)=0.5`**

**`P(T)=0.5`**


Distribution:

| Outcome | Probability |
| ------- | ----------- |
| H       | 0.5         |
| T       | 0.5         |

---

## Why Do We Need Distributions?

Sometimes we don't want the probability of one event.

Instead we want probabilities for **all possible values**.

Example:

Age of people:

```text
18
19
20
21
22
...
```

A distribution tells us how likely each value is.

---

A distribution shows where probability is concentrated.

# Prior Distribution

Earlier, a **Prior** was a single probability:

**`P(Spam)=0.3`**

But sometimes we don't know exactly which value is correct.

Instead, we assign probabilities to many possible values.

That collection of probabilities is called a **Prior Distribution**.

---

## Example

Suppose we are estimating:

**`p = Probability of getting Head`**


We don't know the true value.

Possible values:

| p   | Probability |
| --- | ----------- |
| 0.2 | 0.1         |
| 0.5 | 0.7         |
| 0.8 | 0.2         |

This table is a Prior Distribution.

---

## Difference

### Prior Probability

One value:

**`P(Spam)=0.3`**

---

### Prior Distribution

Many possible values:



---

## Intuition

Before seeing any data:

```text
What do I believe?
```

The answer may not be a single number.

Instead:

```text
I think 0.5 is most likely
0.8 is possible
0.2 is less likely
```

That entire belief is the Prior Distribution.

---

## Coin Example

Before tossing a coin:

```text
Maybe fair
Maybe slightly biased
Maybe heavily biased
```

You assign probabilities to all possibilities.

That collection becomes the Prior Distribution.

---

## Visual Intuition

```text
Probability
 ^
 |
 |       *
 |     * *
 |   * * *
 | * * * *
 +---------------->
      p values
```

The graph shows which values you believe are more likely before observing data.

---

## Why We Need It

In Bayesian statistics we often estimate unknown quantities:

```text
Probability of rain
Probability of disease
Model parameter
Coin bias
```

Instead of one value:

```text
p = 0.5
```

we keep uncertainty:

```text
p could be many values
```

This uncertainty is represented by a Prior Distribution.

---

## ML Connection

In Bayesian ML:

```text
Parameter
      ↓
Prior Distribution
      ↓
Observe Data
      ↓
Posterior Distribution
```

The model starts with a Prior Distribution and updates it after seeing data.

---

## What To Remember



Think:

```text
Before Data
     ↓
Distribution of Beliefs
```

rather than a single probability.

### **But the probabilty of an parameter is calculated so how can it have muiltiple value of which we make distribution of?**

Short answer:

The parameter has only **one true value**, but **we don't know what it is**.

So the distribution is not over reality, it's over **our belief** about the possible values.

Example:


**P(Head)=?**

True value might be:


**0.63**

but we don't know that.

So we say:

```text
0.4 → possible
0.5 → more likely
0.6 → very likely
0.7 → possible
```

This collection of beliefs is the **Prior Distribution**.

**One true value exists.**
**Many possible values represent our uncertainty about it.**


# Posterior Distribution

Now we update our Prior Distribution after seeing data.

The result is called the **Posterior Distribution**.

---

## Prior Distribution

Before data:

```text
Possible value of P(Head)

0.2 → possible
0.5 → most likely
0.8 → possible
```

This is our belief before observing anything.

---

## Observe Data

Suppose we toss a coin:

```text
H H H H H H H H T T
```

8 Heads out of 10 tosses.

---

## Update Belief

Now values near:

**`0.8`**

become more likely.

Values near:

**`0.2`**

become less likely.

---

## Posterior Distribution

After seeing data:

```text
0.2 → very unlikely
0.5 → somewhat likely
0.8 → very likely
```

This new distribution is the Posterior Distribution.

---

## Intuition

```text
Prior Distribution
        +
Observed Data
        ↓
Posterior Distribution
```

---

## Difference

### Prior Distribution

Belief before data.

### Posterior Distribution

Belief after data.

---

## ML Connection

Bayesian ML models do:

```text
Prior Distribution
       ↓
Observe Training Data
       ↓
Posterior Distribution
```

and use the posterior for prediction.

---

## What To Remember



or simply:

```text
Prior + Data = Posterior
```

# Bayesian Thinking

Bayesian Thinking is a way of reasoning.

The idea is simple:

```text id="vjlwm5"
Start with a belief
       ↓
Observe evidence
       ↓
Update the belief
```

---

## Traditional Thinking

Many people think:

```text id="pyn6ns"
Evidence arrived
      ↓
Make conclusion
```

---

## Bayesian Thinking

Bayesian thinking says:

```text id="q6a5so"
Initial Belief
       +
Evidence
       ↓
Updated Belief
```

The old belief is never ignored.

---

## Example: Weather

Before looking outside:

```text id="bfjlwm"
Chance of rain = 30%
```

This is your prior belief.

---

Then you see:

```text id="epx4hh"
Dark clouds
```

Now your belief increases.

Maybe:

```text id="xj8nt4"
Chance of rain = 70%
```

This is your updated belief.

---

## Example: Coin

Before tossing:

```text id="50jp2d"
Coin is probably fair
```

After seeing:

```text id="fl00oq"
8 Heads out of 10 tosses
```

You update your belief.

Now you think:

```text id="q3v4mc"
Coin may be biased toward Heads
```

---

## Core Principle

Bayesian Thinking is:


---

## ML Connection

Machine Learning often does exactly this:

```text id="x3txha"
Initial Model
      ↓
See Data
      ↓
Update Belief
      ↓
Better Prediction
```

Bayesian ML makes this process explicit using probabilities.

# Updating Beliefs

This is the process that Bayesian Thinking follows.

---

## What is Updating Beliefs?

When new evidence arrives, we modify our current belief.

```text
Old Belief
     +
New Evidence
     ↓
New Belief
```

---

## Example

Before tossing a coin:

```text
P(Head) = 0.5
```

You believe it is probably fair.

---

After observing:

```text
10 tosses
8 Heads
2 Tails
```

Your belief changes.

```text
P(Head) is probably greater than 0.5
```

You updated your belief using data.

---

## Continuous Updating

```text
Prior
  ↓
Data 1
  ↓
Posterior 1
  ↓
Data 2
  ↓
Posterior 2
  ↓
Data 3
  ↓
Posterior 3
```

Each posterior becomes the next prior.

---

## Why It Matters

The more data you observe:

```text
More Evidence
      ↓
More Accurate Belief
```

Your uncertainty gradually decreases.

---

## Disease Example

Initially:

```text
Disease Probability = 1%
```

Positive test:

```text
Disease Probability = 8.8%
```

Second positive test:

```text
Disease Probability increases further
```

Every new piece of evidence updates the belief.

---

## ML Connection

In Bayesian ML:

```text
Initial Belief
      ↓
Training Data
      ↓
Updated Belief
      ↓
More Data
      ↓
Updated Again
```

The model continuously learns by updating probabilities.

---

## What To Remember

[

Using\ new\ evidence\ to\ modify\ current\ beliefs
}
]
