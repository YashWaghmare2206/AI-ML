# Non-Convex Functions

Now let's study the opposite of convex functions.

These are much more important in modern AI because most deep learning models optimize **non-convex** loss functions.

---

# What Is a Non-Convex Function?

A function is non-convex if it does **not** have the bowl-shaped structure of a convex function.

It can have:

* Multiple valleys
* Multiple peaks
* Flat regions
* Saddle points

---

## Intuition

Convex:

```text id="g26cmy"
      .
    /   \
   /     \
  /       \
-----------
```

One bowl.

One minimum.

---

Non-Convex:

```text id="7u2iqg"
     /\__
    /    \_
 __/       \__
```

Many ups and downs.

Many possible minima.

## Why Is This a Problem?

Suppose we want to minimize:

**`f(x)`**

using Gradient Descent.

In a convex function:

* One minimum exists.
* Every path leads to it.

In a non-convex function:

* Multiple minima may exist.
* Gradient Descent may stop at a suboptimal point.

---

## Local Minimum

A point that is lower than nearby points.

But not necessarily the lowest point overall.

---

## Global Minimum

The lowest point in the entire function.

---

Non-convex functions may have many local minima and one global minimum.

---

## Example

Imagine:

```text id="hpnj4o"
     _
   _/ \_
 _/     \__
```

Three valleys.

If Gradient Descent starts near the left valley, it may stop there even if a deeper valley exists on the right.

---

## Why Deep Learning Is Non-Convex

Neural networks contain:

* Many layers
* Many weights
* Nonlinear activations

The resulting loss surface is highly non-convex.

---

Instead of a simple bowl:

```text id="k6qupd"
      .
    /   \
   /     \
```

you get a landscape with millions of dimensions and many valleys.

---

## Does Non-Convex Mean Impossible?

No.

Modern optimization algorithms often find very good solutions.

Even without finding the true global minimum.

---

In practice:

* SGD
* Adam
* RMSProp

usually find solutions that work extremely well.

---

## Convex vs Non-Convex

| Property          | Convex         | Non-Convex            |
| ----------------- | -------------- | --------------------- |
| Global Minimum    | One            | May be many minima    |
| Local Minima      | Same as global | Different from global |
| Easy to Optimize  | Yes            | Harder                |
| Linear Regression | Yes            | No                    |
| Neural Networks   | No             | Yes                   |

## Key Takeaways

* Non-convex functions are not bowl-shaped.
* They can have multiple valleys and peaks.
* Gradient Descent may get trapped in local minima.
* Neural network training is generally non-convex.
* Modern ML still works well despite non-convexity.

### One-line Summary

> A non-convex function has multiple possible minima and a complex optimization landscape, making optimization more difficult than convex optimization.
