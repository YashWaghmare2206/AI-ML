# Local vs Global Minima

Now that you understand convex and non-convex functions, this topic becomes very easy.

---

# What Is a Minimum?

A minimum is a point where the function value is low.

There are two important types:

1. Local Minimum
2. Global Minimum

---

# Local Minimum

A point is a **local minimum** if:

> It is lower than all nearby points.

Notice:

**Nearby points**, not the entire function.

---

Example:

```text id="vh6y6m"
      _
   __/ \__
 _/       \_
```

The left valley is a local minimum.

If you look nearby, it is the lowest point.

But there may be a deeper valley elsewhere.

---

# Global Minimum

A point is a **global minimum** if:

> It is the lowest point in the entire function.

Example:

```text id="uj39yk"
      _
   __/ \__
 _/       \____
```

The right valley is lower than every other point.

That is the global minimum.

Therefore:

* Local Minimum = Global Minimum

Always.

This is why convex optimization is easy.

---

# Non-Convex Function Case

Consider:



Now there are multiple valleys.

```text id="vuzhsq"
      _
   __/ \__
 _/       \__
```

One valley may be deeper.

So:

* Multiple local minima possible
* One global minimum

---

# Why Does This Matter?

Suppose we use Gradient Descent.

Start here:

```text id="n7s10l"
      _
   __/ \__
 _/       \__
 ↑
 Start
```

Gradient Descent moves downhill.

It may stop at the nearest valley.

---

Problem:

The nearest valley might be only a local minimum.

A better minimum may exist elsewhere.

---

# Optimization Goal

What do we really want?



We want the **global minimum**.

Because it gives the smallest loss.

---

But in non-convex optimization:

Finding the global minimum can be difficult.

---

# Deep Learning Reality

Neural network losses are highly non-convex.

In practice:

* We usually don't know the global minimum.
* We often settle for a very good local minimum.

Surprisingly:

That is often enough for excellent performance.

---

# Relationship with Gradient Descent

Gradient Descent only sees:

* Current position
* Local slope

It cannot see the entire landscape.

Therefore it may:

* Reach global minimum
* Reach local minimum

depending on where it starts.

---

---

# ML Examples

### Linear Regression

Convex loss.

Global minimum is easy to find.

---

### Neural Networks

Non-convex loss.

Many local minima.

Optimization is harder.

---

# Key Takeaways

* Local minimum = lowest point in a neighborhood.
* Global minimum = lowest point in entire function.
* Convex functions:

  * Local = Global
* Non-convex functions:

  * Multiple local minima possible
* Gradient Descent may stop at a local minimum.
* Neural networks are usually non-convex.

### One-line Summary

> A local minimum is the lowest nearby point, while a global minimum is the lowest point in the entire optimization landscape.
