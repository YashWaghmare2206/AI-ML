## Why ML Cares

Suppose a loss function depends on two weights:

**`L(w_1,w_2)`**

To reduce the loss, we need to know:

* How loss changes when (w_1) changes
* How loss changes when (w_2) changes

These are partial derivatives:



Gradient Descent uses these values to update weights.

---

## Python

Using SymPy:

```python
import sympy as sp

x, y = sp.symbols('x y')

f = x**2 + y**2

print(sp.diff(f, x))
print(sp.diff(f, y))
```

Output:

```python
2*x
2*y
```

## Key Takeaway

For a multivariable function:

**`f(x,y)`**

a partial derivative measures:

> "How much does the output change if only one variable changes while the others stay fixed?"
