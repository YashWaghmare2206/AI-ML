## What Does the Gradient Tell Us?

The gradient points in the direction of:

> **Maximum increase** of the function.

For minimizing a loss function, we move in the **opposite direction**:



This is the core idea behind Gradient Descent.

---

## Why ML Cares



This tells us:

* how sensitive loss is to (w_1)
* how sensitive loss is to (w_2)

Using this information, we update the weights to reduce loss.
