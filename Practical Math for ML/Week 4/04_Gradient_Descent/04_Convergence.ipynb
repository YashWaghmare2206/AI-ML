# 04_Convergence

**Convergence** means:

> Gradient Descent has reached (or is very close to) the minimum cost.

In other words:

**`J(w)`**

is no longer decreasing significantly.

---

## Example

Suppose the cost values are:

```text
100
↓
50
↓
20
↓
8
↓
3
↓
1.2
↓
0.5
↓
0.49
↓
0.489
↓
0.4889
```

Initially the cost drops a lot.

Later the changes become tiny.

This is convergence.

## What Happens If Learning Rate Is Bad?

### Too Small

Converges very slowly.

```text
10
9.999
9.998
9.997
...
```

---

### Too Large

May never converge.

```text
10
-20
40
-80
160
...
```

The updates keep overshooting.

---

## ML Interpretation

When training finishes:

* Weights have stabilized.
* Cost is near minimum.
* Further updates produce almost no improvement.

That state is called **convergence**.

---

## Important Note

Convergence does **not** guarantee:

> Cost = 0

It only means:

> Gradient Descent cannot find a significantly better solution anymore.
