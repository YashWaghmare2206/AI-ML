## Python

```python
import numpy as np

y = np.array([10, 7, 5])
y_pred = np.array([8, 6, 4])

cost = np.mean((y - y_pred) ** 2)

print(cost)
```

Output:

```python
2.0
```

---

### Key Takeaway

* **Loss Function** → error for one sample.
* **Cost Function** → average error over all samples.
* Training a model means finding weights that minimize the cost function.
