3. Compute gradient.
4. Move opposite to gradient.

## ML Training Loop

```text
Initialize weights

Repeat:
    Predict
    Compute Cost
    Compute Gradient
    Update Weights

Until Cost Stops Decreasing
```

This is essentially how linear regression, logistic regression, and neural networks learn.

Gradient Descent is:

> A repeated process that updates weights in the opposite direction of the gradient to minimize the cost function.
