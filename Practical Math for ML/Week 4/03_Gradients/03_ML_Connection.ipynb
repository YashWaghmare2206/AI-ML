## Entire ML Training in One Loop

```text
Initialize weights

Repeat:
    Make prediction
    Compute loss
    Compute gradient
    Update weights

Until loss stops decreasing
```

---

## Where Everything Fits

| Topic              | ML Purpose                       |
| ------------------ | -------------------------------- |
| Derivative         | Change of function               |
| Partial Derivative | Change wrt one weight            |
| Gradient           | All partial derivatives together |
| Loss Function      | Measures prediction error        |
| Negative Gradient  | Direction of decreasing loss     |
| Gradient Descent   | Uses gradient to update weights  |

---

### Key Takeaway

The entire reason we learned:

* Derivatives
* Partial Derivatives
* Gradients

is because ML training is simply:

> Compute loss → compute gradient → move weights opposite to the gradient → reduce error.
