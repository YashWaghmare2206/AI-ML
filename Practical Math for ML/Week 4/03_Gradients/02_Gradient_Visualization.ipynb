# 02_Gradient_Visualization

Now let's visualize what the gradient is doing.

Consider the loss function:



---

## The Graph

The graph is a bowl-shaped curve.

* Minimum point is at:

**`w=0`**

* Loss is:

**`L(0)=0`**

This is where we want to reach.

---

## Point 1: Right Side

Suppose:

**`w=3`**

Gradient:



Positive gradient.

Meaning:

* Function is increasing as we move right.
* To reduce loss, move left.

So Gradient Descent moves:



toward 0.

---

## Point 2: Left Side

Suppose:



Gradient:



Negative gradient.

Meaning:

* Function is decreasing as we move right.
* To reduce loss, move right.

Gradient Descent:



Again moving toward 0.

---

## Point 3: Minimum



No slope.

No direction to move.

We have reached the minimum.

---

## What the Sign Tells You

| Gradient | Meaning                                |
| -------- | -------------------------------------- |
| Positive | Move left to reduce loss               |
| Negative | Move right to reduce loss              |
| Zero     | At a minimum (or maximum/saddle point) |

---

## ML Interpretation



---

### Key Takeaway

For any point on the loss surface:

* Gradient = uphill direction
* Negative Gradient = downhill direction
* Gradient magnitude = how steep the hill is
* Gradient (=0) = flat region (possible optimum)

This intuition is more important than the formulas.
