# 01_What_Is_Derivative

## Definition

The derivative measures **how fast a function changes** with respect to its input.

If:

`**y = f(x)**`

then the derivative is written as:

**`f'(x)`**



It tells us:

> "If x changes a little, how much does y change?"

## Why ML Cares

Machine Learning is mostly:

1. Make prediction
2. Calculate error (loss)
3. Find how error changes when weights change
4. Adjust weights to reduce error

Step 3 uses **derivatives**.

Without derivatives:

* No Gradient Descent
* No Neural Network training
* No Optimization

---

## Python

```python
import numpy as np

def f(x):
    return x**2

x = 3
h = 0.0001

derivative = (f(x + h) - f(x)) / h

print(derivative)
```

Output:

```python
6.0001
```

Close to the actual derivative value:

```python
6
```

---

### Key Takeaway

A derivative tells us the **rate of change of a function** at a specific point.

For ML, derivatives tell us:

> "Which direction should the model parameters move to reduce error?"
