# 03_SGD_MiniBatch

So far, we assumed the gradient is computed using the **entire dataset**.

For large datasets, that can be very slow.

This leads to three approaches.

---

## 1. Batch Gradient Descent

Uses **all training samples** to compute the gradient.

Suppose dataset size:


**`100000`**

samples.

For every update:

```text
Use all 100000 samples
→ Compute Cost
→ Compute Gradient
→ Update Weights
```

### Pros

* Stable updates
* Accurate gradient

### Cons

* Slow
* Expensive for large datasets

---

## 2. Stochastic Gradient Descent (SGD)

Uses **one sample at a time**.

```text
Take 1 sample
→ Compute Gradient
→ Update Weights
```

Example:

```text
Sample 1 → update
Sample 2 → update
Sample 3 → update
...
```

### Pros

* Very fast updates
* Works on huge datasets

### Cons

* Noisy updates
* Loss may bounce up and down

---

## 3. Mini-Batch Gradient Descent

Uses a small group of samples.

Example batch size:

**`32`**

```text
Take 32 samples
→ Compute Gradient
→ Update Weights

Take next 32 samples
→ Compute Gradient
→ Update Weights
```

### Pros

* Faster than Batch GD
* More stable than SGD
* Efficient on GPUs

### Cons

* Still an approximation

---

## Comparison

| Method        | Samples per Update               |
| ------------- | -------------------------------- |
| Batch GD      | Entire Dataset                   |
| SGD           | 1 Sample                         |
| Mini-Batch GD | Small Batch (16, 32, 64, 128...) |

---

## Why Deep Learning Uses Mini-Batches

Suppose dataset:

**`1,000,000`**

images.

Computing gradients on all images before every update is too slow.

Instead:

```text
32 images
→ update

32 images
→ update

32 images
→ update
```

This is much faster.

---

## Important Note

The Gradient Descent formula remains the same:



The only difference is:

**How the gradient is computed.**

* Batch → all samples
* SGD → one sample
* Mini-Batch → small subset

---

### Key Takeaway

* **Batch GD** = accurate but slow.
* **SGD** = fastest but noisy.
* **Mini-Batch GD** = balance of speed and stability.

This is why almost all modern ML and Deep Learning training uses **Mini-Batch Gradient Descent**.
