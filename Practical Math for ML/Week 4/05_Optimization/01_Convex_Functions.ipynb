# 01_Convex_Functions

A **convex function** is a function with only one bowl-shaped minimum.

This is the best type of function for Gradient Descent.

---

## Example



This graph has:

* One minimum point
* No local minima
* No local maxima

The minimum is at:

**`x=0`**

---

## Why Is It Important?

If a function is convex:

No matter where you start,

```text
Left side  \
            \
             \__
            /
Right side /
```

Gradient Descent will eventually reach the same minimum.

```text
\__/^\__/
```

---

## ML Connection

For:

* Linear Regression + MSE

the cost function is convex.

So Gradient Descent finds the global minimum.

---

For:

* Deep Neural Networks

the loss surface is usually non-convex.

There can be many local minima and saddle points.

---

## Convex vs Non-Convex

| Property                  | Convex | Non-Convex       |
| ------------------------- | ------ | ---------------- |
| One minimum               | ✅      | ❌                |
| Multiple valleys          | ❌      | ✅                |
| Easier optimization       | ✅      | ❌                |
| Gradient Descent reliable | ✅      | Less predictable |

---

### Key Takeaway

A convex function has a single bowl-shaped minimum.

For convex functions:

> Any gradient descent path eventually leads to the same minimum.

That's why convex optimization is much easier than non-convex optimization.
