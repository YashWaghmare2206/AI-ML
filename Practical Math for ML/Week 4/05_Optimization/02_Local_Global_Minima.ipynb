# 02_Local_Global_Minima

When optimizing a function, not all minima are equal.

---

## Global Minimum

The **lowest point in the entire function**.

Example:



Minimum:

**`x=0`**

This is the lowest point anywhere on the graph.

So it is the **global minimum**.

---

## Local Minimum

A point that is lower than nearby points, but not necessarily the lowest point overall.

Imagine:

```text
      /\        /\
     /  \______/  \
         \__/
```

The small valley is a local minimum.

There may be a deeper valley elsewhere.

---

## Example

Consider a non-convex function:

```text
\__/^\____/
```

There are two valleys.

* Left valley → Local Minimum
* Right deeper valley → Global Minimum

Gradient Descent may stop at the left valley if it starts nearby.

---

## Why ML Cares

For convex functions:

* Local Minimum = Global Minimum

Only one valley exists.

Easy optimization.

---

For neural networks:

* Many local minima
* Saddle points
* Complex loss surfaces

Gradient Descent may converge to different solutions depending on the starting weights.

---

## Important

A local minimum is still a minimum.

At that point:



So Gradient Descent thinks:

> "I can't go downhill anymore."

Even though a better minimum may exist elsewhere.

---

## Global vs Local

| Property                        | Local Minimum | Global Minimum |
| ------------------------------- | ------------- | -------------- |
| Lower than nearby points        | ✅             | ✅              |
| Lowest point in entire function | ❌             | ✅              |
| Gradient ≈ 0                    | ✅             | ✅              |
| Best possible solution          | ❌             | ✅              |

---

### Key Takeaway

* **Local Minimum** → lowest point in a small neighborhood.
* **Global Minimum** → lowest point in the entire function.

Optimization tries to find the global minimum, but in non-convex functions Gradient Descent may stop at a local minimum instead.
