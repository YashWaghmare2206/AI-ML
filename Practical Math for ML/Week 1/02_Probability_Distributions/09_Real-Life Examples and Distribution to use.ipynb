---

## Examples

### Spam Detection



---

## Examples

### Coin Tosses

```text id="b9r4w8"
10 tosses

Probability of exactly
7 heads
```

---

### Student Passes

```text id="c3m7x1"
20 students

Probability exactly
15 pass
```

---

## Question Style

```text id="n8k5p2"
Out of n trials,
how many successes?
```

---

# 3. Poisson Distribution

## Pattern

```text id="y4v8m3"
Count events

In fixed time/space
```

# Master Comparison

| Distribution | Data Type  | What It Counts           |
| ------------ | ---------- | ------------------------ |
| Bernoulli    | Discrete   | One success/failure      |
| Binomial     | Discrete   | Successes in n trials    |
| Poisson      | Discrete   | Events in interval       |
| Normal       | Continuous | Measurements around mean |

# The 4 Questions Trick

When you see a problem, ask:

### Question 1

```text id="v2m8r6"
Only one trial?
```

Yes →

```text id="e7p4n1"
Bernoulli
```

---

### Question 2

```text id="u3k9w5"
Many trials?

Counting successes?
```

Yes →

```text id="r6m2q8"
Binomial
```

---

### Question 3

```text id="c1v7n4"
Counting events in time/space?
```

Yes →

```text id="b8p3m6"
Poisson
```

---

### Question 4

```text id="k4r9w2"
Continuous measurements around an average?
```

Yes →

```text id="x5n1q7"
Normal
```
