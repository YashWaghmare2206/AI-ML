# Mean and Spread of Distributions

This topic ties together everything you've learned so far.

---

## 1. What is Mean?

Mean tells us:

```text id="t1g7f2"
Where the center of the distribution is.
```

Think:

```text id="ev9m8w"
Average value
```

Examples:

| Distribution | Mean Meaning                |
| ------------ | --------------------------- |
| Bernoulli    | Average success rate        |
| Binomial     | Average number of successes |
| Poisson      | Average events              |
| Normal       | Center of the bell curve    |

---

## 2. What is Spread?

Spread tells us:

```text id="q4n8k3"
How much the data varies around the mean.
```

Small spread:

```text id="5v7r9j"
Data tightly clustered
```

Large spread:

```text id="d3w2m6"
Data widely scattered
```

---

## 3. Measuring Spread

### Variance

Measures average squared distance from the mean.

```text id="w6x4n1"
Higher variance
=
More spread
```

---

### Standard Deviation

Square root of variance.

```text id="m2c7b8"
Most commonly used measure of spread
```

Reason:

It is in the same units as the original data.

---

## 4. Example

### Dataset A

```text id="r9k3v5"
68
69
70
71
72
```

Mean:

```text id="n8w5q2"
70
```

Spread:

```text id="f1y7z4"
Very small
```

---

### Dataset B

```text id="h3c8m6"
20
40
70
100
120
```

Mean:

```text id="k6v2p9"
70
```

Spread:

```text id="g4x8r1"
Very large
```

---

Notice:

```text id="b7m1d5"
Same mean

Different spread
```

This is why mean alone is not enough.

Mean
