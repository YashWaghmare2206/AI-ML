# Standard Normal Distribution

## 1. What is it?

A **Standard Normal Distribution** is a Normal Distribution that has been standardized so that:

```text id="7g8m7a"
Mean (μ) = 0

Standard Deviation (σ) = 1
```

---

## 2. Why do we need it?

Suppose we have two students.

### Student A

```text id="sq8hsr"
Math Marks

Mean = 70
SD = 10

Student Score = 80
```

---

### Student B

```text id="6w0o1e"
Physics Marks

Mean = 50
SD = 5

Student Score = 60
```

Question:

```text id="i8yn7k"
Who performed better relative to their class?
```

We cannot compare directly because:

```text id="b3u14e"
Different means
Different spreads
```

We need a common scale.

That common scale is the **Standard Normal Distribution**.

---

## 3. The Main Idea

Convert every value into:

```text id="8uyrlj"
How many standard deviations
away from the mean?
```

This value is called the **Z-score**.

---

## 4. Z-Score Formula



where:

| Symbol | Meaning            |
| ------ | ------------------ |
| x      | Actual value       |
| μ      | Mean               |
| σ      | Standard Deviation |
| z      | Z-score            |

---

## 7. Positive and Negative Z-Scores

### Positive


z>0


Means:

```text id="z3t5f7"
Above mean
```

Example:

```text id="a2f1qq"
z = 1
z = 2
z = 3
```

---

### Negative


z<0


Means:

```text id="9sj5iz"
Below mean
```

Example:

```text id="yxsvhy"
z = -1
z = -2
z = -3
```

---

### Zero

z=0


Means:

```text id="rrh0li"
Exactly at the mean
```

---

## 8. What Does Standardization Do?

Original Distribution:

```text id="5cmh2w"
Mean = 70
SD = 10
```

After standardization:

```text id="rz6i4z"
Mean = 0
SD = 1
```

Every Normal Distribution can be converted to this common form.

---

## 9. AI/ML Relevance

This is extremely important.

### StandardScaler

In Scikit-Learn:

```python
from sklearn.preprocessing import StandardScaler
```

Z-score:

```text id="xy10rw"
2
```

The score is still 90.

2 is just its position relative to the mean.

---

### ❌ Negative Z-score does NOT mean bad

It only means:

```text id="m7m9jg"
Below the mean
```

Nothing more.
