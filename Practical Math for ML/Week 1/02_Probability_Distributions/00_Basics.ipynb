**What is it?**
Events that cannot happen together.

---

## 8. Independent Events

**What is it?**

Coin toss and die roll.

---

## 9. Dependent Events

**What is it?**

**Example:**
Drawing two cards without replacement.

The first card changes what remains.

# 10–13: Probability Rules

## 10. Basic Probability Formula

**What is it?**
Probability = Favorable outcomes ÷ Total outcomes.

```text
P(A) = Number of favorable outcomes / Total outcomes
```

**Example:**
Probability of getting an even number on a die:

```text
P(A) = 3/6 = 1/2
```

**Code:**

```python
probability = 3 / 6
```

---

## 11. Complement Rule

**What is it?**
Probability that an event does NOT occur.

```text
P(A') = 1 - P(A)
```

**Notation:**

* (A') (A complement) means "not A".

**Example:**
P(rain) = 0.3

```text
P(no rain) = 1 − 0.3 = 0.7
```

**Code:**

```python
p_no_rain = 1 - 0.3
```

---

## 12. Addition Rule

**What is it?**
Used to find the probability of A **or** B.

For mutually exclusive events:

```text
P(A ∪ B) = P(A) + P(B)
```

**Notation:**

* (∪) means "OR".

**Example:**
Probability of rolling 2 or 5:

```text
1/6 + 1/6 = 2/6
```

**Code:**

```python
p = 1/6 + 1/6
```

---

## 13. Multiplication Rule

**What is it?**
Used to find the probability of A **and** B.

For independent events:

```text
P(A ∩ B) = P(A) × P(B)
```

**Notation:**

* (∩) means "AND".

**Example:**
Getting Head and rolling a 6:

```text
1/2 × 1/6 = 1/12
```

**Code:**

```python
p = (1/2) * (1/6)
```

#### Multiplication Rule

### 1. Dependent Events

One event **affects** the probability of the other.

Formula:
```
P(A \B) = P(A) * P(B|A)
```

* (P(B|A)) means "Probability of B **given that A has already happened**."

**Example:** Drawing 2 cards from a deck **without replacement**.

* First card is an Ace:

```
  P(A)= 4 / 52
```

* Second card is an Ace given the first was an Ace:

```
 P(B|A)= 3 / 51
```

So,

P(A \B)= (4 / 52)  *  (3 / 51)

---

### 2. Independent Events

One event **does not affect** the other.

Formula:

```
P(A B) = P(A) * P(B)
```

because

```
P(B|A)=P(B)
```

**Example:** Tossing a coin and rolling a die.

* Heads:

```
  P(H)=1 / 2
```

* Rolling a 6:

  ```
  P(6)= 1/ 6
  ```

### How to know which one to use?

Ask yourself:

> **"Does the first event change the probability of the second?"**

* ✅ **Yes** → Dependent → Use (P(A) \times P(B|A))
* ❌ **No** → Independent → Use (P(A) \times P(B))

### Easy trick

* **With replacement** → Usually Independent
* **Without replacement** → Usually Dependent

So the multiplication rule is the **same concept (AND = multiply)**, but the second probability changes depending on whether the events are dependent or independent.
