## 4. Why do we need it?

Bernoulli Distribution helps us model situations involving:

* A single trial
* Two outcomes
* Known probability of success

It forms the foundation for many other distributions, especially the **Binomial Distribution** (which we'll study next).

### Variance

The variance is:

[
Var(X)=p(1-p)
]

### Derivation

Using,

[
Var(X)=E[X^2]-[E(X)]^2
]

Since,

[
X^2=X
]

for (X=0) or (1),

[
E[X^2]=p
]

Therefore,

[
Var(X)=p-p^2
]

[
=p(1-p)
]

## 9. Key Points to Remember

* Bernoulli Distribution models **one trial**.
* There are exactly **two outcomes**.
* Success is represented by **1**.
* Failure is represented by **0**.
* Probability of success = (p).
* Probability of failure = (1-p).
* Mean:

[
E[X]=p
]

* Variance:

[
Var(X)=p(1-p)
]

* Bernoulli Distribution is the building block of the **Binomial Distribution**.

---

## 10. Common Mistakes / Confusions

### ❌ Bernoulli vs Binomial

Students often think they are the same.

* Bernoulli → **1 trial**
* Binomial → **n independent Bernoulli trials**

Example:

* Toss a coin once → Bernoulli
* Toss a coin 10 times → Binomial

---

### ❌ Success does NOT mean something good

"Success" simply means the outcome of interest.

Example:

For fraud detection,

* Fraud = Success (1)
* Not Fraud = Failure (0)

Even though fraud is undesirable.

---

### ❌ Mean is not always 0.5

Mean equals (p), not necessarily 0.5.

Only fair experiments have (p=0.5).
