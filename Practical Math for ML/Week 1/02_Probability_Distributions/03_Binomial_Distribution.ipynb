# Binomial Distribution

## 1. What is it?

A **Binomial Distribution** describes the probability of getting a certain number of **successes** in a fixed number of **independent Bernoulli trials**.

### Simple Definition

> Binomial Distribution = Repeated Bernoulli Experiments

---

## 2. Intuition

Remember Bernoulli?

```text
Bernoulli:
1 experiment
↓
Success or Failure
```

Now suppose we repeat that experiment multiple times.

```text
Bernoulli + Bernoulli + Bernoulli + ...
= Binomial
```

### Ask yourself:

> "I am doing the same experiment multiple times. What is the probability of getting exactly k successes?"

If yes → Binomial Distribution.

---

## 3. Conditions for Binomial Distribution

All 4 conditions must be true.

### 1. Fixed number of trials

Example:

* Toss coin 10 times.
* Check 20 products.

Number of trials is fixed.

---

### 2. Only two outcomes

Each trial has:

* Success
* Failure

---

### 3. Same probability of success

If


p=0.5


for the first trial,

it remains 0.5 for all trials.

---

### 4. Trials are independent

One trial does not affect another.

Example:

Coin tosses.

---

## 4. Mathematical Definition

Suppose,

* (n) = number of trials
* (k) = number of successes
* (p) = probability of success
* (1-p) = probability of failure

Then,

The probability of getting exactly (k) successes is:



This is called the **Binomial Probability Mass Function (PMF).**

---

## 5. Understanding the Formula

The formula has 3 parts.

### Part 1



means:

> In how many ways can we arrange the successes?

---

### Part 2


means:

> Probability of getting k successes.

---

### Part 3



means:

> Probability of getting the remaining failures.

---

## 6. What is (\binom{n}{k})?

It is called **Combination**.

Formula:



It means:

> Number of ways to choose k successes from n trials.

---

## 7. Example 1: Coin Toss

### Problem

A fair coin is tossed 3 times.

Find the probability of getting exactly 2 Heads.

---

### Step 1

Identify values.

Number of trials:

[
n=3
]

Successes required:

[
k=2
]

Probability of Head:

[
p=0.5
]

Failure probability:

[
1-p=0.5
]

---

### Step 2

Apply formula.

[
P(X=2)
======

\binom{3}{2}(0.5)^2(0.5)^1
]

---

### Step 3

Find combinations.

[
\binom{3}{2}
============

# \frac{3!}{2!1!}

3
]

---

### Step 4

Calculate.

[
P(X=2)
======

3\times0.25\times0.5
]

# [

3\times0.125
]

# [

0.375
]

Therefore,

[
\boxed{P(X=2)=0.375}
]

---

### Why 3?

Because these arrangements exist:

```text
HHT
HTH
THH
```

Exactly 2 Heads.

Total = 3 ways.

## 12. Key Points to Remember

* Binomial = repeated Bernoulli trials.
* Fixed number of trials.
* Only two outcomes.
* Same probability of success.
* Independent trials.
* PMF:

[
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
]

* Mean:

[
E[X]=np
]

* Variance:

[
Var(X)=np(1-p)
]

---

## 13. Common Confusions

### Bernoulli vs Binomial

```text
Bernoulli
↓
One trial

Binomial
↓
Multiple Bernoulli trials
```

Example:

* Toss coin once → Bernoulli
* Toss coin 10 times → Binomial
