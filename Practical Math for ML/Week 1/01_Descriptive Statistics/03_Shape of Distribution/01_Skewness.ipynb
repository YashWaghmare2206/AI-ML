# Topic 12: Skewness

We are now starting the **Shape of Distribution** section.

This is one of the most important EDA concepts because it tells you:

> **Is the data balanced, or is it leaning more towards one side?**

---

# 1. What is Skewness?

Skewness measures the **asymmetry** of a distribution.

It answers:

> Is the data distributed evenly around the center, or does one side have a longer tail?

---

# 2. What is a Symmetric Distribution?

Imagine most people in a class have heights close to the average.

```text
        /\
       /  \
      /    \
     /      \
----/--------\----
```

Left side and right side are roughly mirror images.

This is called a **symmetric distribution**.

In such distributions:

> Mean ≈ Median

---

# 3. Positive Skew (Right Skew)

## What does it look like?

```text
      /\
     /  \
    /    \______
---/------------->
```

The **tail extends towards the right**.

---

## What causes it?

A few extremely large values pull the distribution to the right.

---

## Example 1: Salaries

Most employees earn:

```text
₹25,000 – ₹60,000
```

But a few CEOs earn:

```text
₹5,00,000+
```

Those large salaries create a long right tail.

---

## Relationship between Mean and Median

> Mean > Median

Why?

Because the large values pull the mean upward.

---

# 4. Negative Skew (Left Skew)

## What does it look like?

```text
          /\
     _____/  \
<------------\----
```

The **tail extends towards the left**.

---

## What causes it?

A few extremely small values pull the distribution to the left.

---

## Example 2: Easy Exam

Suppose most students score:

```text
80–100
```

But a few students score:

```text
10–20
```

The low scores create a long left tail.

---

## Relationship between Mean and Median

> Mean < Median

Because the small values drag the mean downward.

---

# 5. Summary

| Distribution  | Tail Direction | Mean vs Median |
| ------------- | -------------- | -------------- |
| Symmetric     | No long tail   | Mean ≈ Median  |
| Positive Skew | Right          | Mean > Median  |
| Negative Skew | Left           | Mean < Median  |

---

# 6. Why is Skewness Important in AI/ML?

## a) Feature Understanding

It helps you understand:

> Is this feature normally distributed?

or

> Does it have extreme values?

---

## b) Choosing Mean vs Median

Remember earlier?

For skewed data:

> Median is often a better measure of center than Mean.

---

## c) Data Transformations

Later in ML, if a feature is highly positively skewed, you may apply transformations like:

* Log Transformation
* Square Root Transformation

to make the distribution more balanced.

You'll see this in feature engineering.

---

# 7. Python Implementation

## Pandas

```python
import pandas as pd

s = pd.Series([1, 2, 2, 3, 4, 20])

s.skew()
```

Interpretation:

* Positive value → Positive skew
* Negative value → Negative skew
* Near 0 → Approximately symmetric

# Quick Recap

* Skewness measures the **asymmetry** of data.
* Positive skew:

  * Tail to the right
  * Mean > Median
* Negative skew:

  * Tail to the left
  * Mean < Median
* Symmetric:

  * Mean ≈ Median
* Important for:

  * EDA
  * Choosing mean vs median
  * Detecting unusual distributions
  * Future feature transformations.

---

## One intuition to remember forever

> **The tail tells the skew.**

* Tail right → Positive skew.
* Tail left → Negative skew.
