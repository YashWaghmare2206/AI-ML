# Topic 13: Identifying Skewness from Plots

Now we'll learn something very practical:

> Instead of calculating skewness, how do you **look at a graph and immediately tell whether data is positively skewed, negatively skewed, or symmetric?**

This is a skill you'll use all the time in EDA.

---

# Rule #1: Look at the Tail

> **The direction of the longer tail determines the skew.**

Not the peak.

Not where most data is.

The **tail**.

---

# 1. Symmetric Distribution

Looks like:

```text
         /\
        /  \
       /    \
      /      \
-----/--------\-----
```

### Characteristics

* Left side ≈ Right side
* Mean ≈ Median
* No obvious skew

### Examples

* Adult heights (approximately)
* Measurement errors

---

# 2. Positive Skew (Right Skew)

Looks like:

```text
      /\
     /  \
    /    \________
---/----------------->
```

### Notice:

* Most data is on the **left**
* Tail extends to the **right**

### Interpretation

A few large values pull the distribution rightward.

### Examples

* Salaries
* House prices
* Customer spending

---

# 3. Negative Skew (Left Skew)

Looks like:

```text
          /\
 ________/  \
<----------------\----
```

### Notice:

* Most data is on the **right**
* Tail extends to the **left**

### Interpretation

A few very small values pull the distribution leftward.

### Examples

* Scores on an easy exam
* Retirement ages (sometimes)

---

# How to Read a Histogram

Suppose you see:

```text
███████████▇▆▅▄▃▂
```

Most bars are on the left.

The bars gradually decrease toward the right.

The tail is on the right.

Therefore:

> ✅ Positive Skew

---

Suppose you see:

```text
▂▃▄▅▆▇███████████
```

Tail is on the left.

Therefore:

> ✅ Negative Skew

# How to Identify Skewness Using Boxplots

A boxplot can also reveal skewness.

## Symmetric

```text
----|====|====|----
```

Both whiskers have similar lengths.

---

## Positive Skew

```text
--|====|===|----------
```

Right whisker is longer.

The tail extends right.

> Positive skew.

---

## Negative Skew

```text
----------|===|====|--
```

Left whisker is longer.

The tail extends left.

> Negative skew.

---

# Quick Trick

### Histogram

> Follow the tail.

### Boxplot

> Compare the whiskers.

Longer right whisker → Positive skew.

Longer left whisker → Negative skew.

---

# Why is this useful in AI/ML?

During EDA, before even calculating anything, you can quickly identify:

* Features with extreme values
* Whether Mean or Median is better
* Whether transformations might be needed

This visual skill saves a lot of time.

# Quick Recap

* The **tail determines the skew**.
* Histogram:

  * Tail right → Positive skew
  * Tail left → Negative skew
* Boxplot:

  * Longer right whisker → Positive skew
  * Longer left whisker → Negative skew
* This is one of the fastest EDA techniques used in practice.
