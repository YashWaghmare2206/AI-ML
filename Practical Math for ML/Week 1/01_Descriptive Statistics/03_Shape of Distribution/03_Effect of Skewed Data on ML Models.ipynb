# Topic 14: Effect of Skewed Data on ML Models

This is the last conceptual topic in **Descriptive Statistics**, and it answers:

> **"Why should an ML engineer care if data is skewed?"**

Until now, skewness was just a way to describe data.

Now you'll see why it matters in machine learning.

---

# 1. Does every ML model care about skewness?

> **No.**

Some models are affected by skewed features, while others don't care much.

---

# 2. Problem 1: Mean gets distorted

Suppose house prices are:

```text
₹30L, ₹35L, ₹40L, ₹45L, ₹10Cr
```

Mean:

```text
≈ ₹2.1Cr
```

Median:

```text
₹40L
```

If you use the mean to understand a "typical house price":

> You'll get a misleading picture.

---

## Why does this matter?

During EDA:

* You may make wrong conclusions.
* You may choose poor preprocessing strategies.

---

# 3. Problem 2: Feature Scaling Can Be Affected

Remember StandardScaler?

It uses:

* Mean
* Standard Deviation

Both are influenced by outliers.

Suppose incomes are:

```text
₹25k, ₹30k, ₹35k, ₹40k, ₹50L
```

That ₹50L income:

* Increases the mean
* Increases the SD

As a result:

> The scaling of normal observations gets distorted.

---

# 4. Problem 3: Some Models Assume Approximately Normal Data

Certain models work better when features are not heavily skewed.

Examples include:

### Linear Regression

Large skew can sometimes affect assumptions about residuals.

---

### Logistic Regression

May converge better with transformed features.

---

### PCA (Principal Component Analysis)

Can be influenced by extreme values.

---

# 5. Models That Usually Don't Care Much

Tree-based models are generally robust.

Examples:

✅ Decision Trees

✅ Random Forest

✅ XGBoost

Why?

Because they split based on thresholds.

Example:

```text
Income < ₹50,000 ?
```

They don't depend heavily on means or distances.

---

# 6. What Do Data Scientists Do?

If a feature is highly positively skewed:

Example:

```text
Income
House Prices
Transaction Amounts
```

They often apply transformations.

Common ones are:

### Log Transformation

Turns:

```text
100
1000
10000
```

into something closer together:

```text
2
3
4
```

(Conceptually)

---

### Square Root Transformation

Used for moderate skew.

---

These transformations make the distribution more balanced.

---

# Example 1: House Prices

Original:

```text
₹30L, ₹35L, ₹40L, ₹45L, ₹10Cr
```

Problems:

* Mean distorted
* Scaling distorted
* Harder to model

After a log transformation:

The huge gap becomes smaller.

The feature becomes easier to work with.

---

# Example 2: Customer Spending

Most customers:

```text
₹500–₹2000
```

Few customers:

```text
₹1,00,000+
```

Positively skewed.

Transforming the feature can improve analysis.

---

# Do we always transform skewed data?

> **No.**

Ask yourself:

### Is the skew causing problems?

If yes:

* Transform it.

If no:

* Leave it.

Especially for tree-based models.

---

# Summary Table

| Situation                  | What Happens                          |
| -------------------------- | ------------------------------------- |
| Mean used on skewed data   | Can be misleading                     |
| StandardScaler             | Can be influenced                     |
| Distance-based models      | May be affected                       |
| Linear/Logistic Regression | Sometimes benefit from transformation |
| Tree-based models          | Usually less affected                 |
| Highly skewed features     | Often transformed                     |

---

# Quick Recap

* Skewness isn't just descriptive; it affects preprocessing and modeling.
* Positive skew often appears in:

  * Salaries
  * House prices
  * Spending
* Skewed features may distort:

  * Mean
  * Standard Deviation
  * StandardScaler
* Transformations like **log transforms** can help.
* Tree-based models are generally more robust to skewness.
