# Topic 8: Interquartile Range (IQR)

This is one of the **most practical statistics concepts** in AI/ML because it is heavily used for:

* Outlier Detection
* Boxplots
* EDA (Exploratory Data Analysis)

---

# 1. Why Do We Need IQR?

Range had a problem:

> It only uses the minimum and maximum values.

Example:

```text
1, 2, 3, 4, 100
```

Range = 99

But one extreme value (100) heavily affects it.

We need a measure of spread that:

> **Ignores extreme values and focuses on the middle part of the data.**

That's exactly what IQR does.

---

# 2. What is IQR?

IQR tells us:

> **How spread out the middle 50% of the data is.**

It ignores the lowest 25% and highest 25%.

---

# 3. First, What Are Quartiles?

Arrange the data in ascending order.

Quartiles divide data into **4 equal parts**.

```
Q1      Q2      Q3
|-------|-------|
25%    50%     75%
```

* **Q1 (First Quartile)** → 25% of data lies below this value.
* **Q2 (Second Quartile)** → Median (50%).
* **Q3 (Third Quartile)** → 75% of data lies below this value.

---

# 4. Formula

```
IQR = Q3 - Q1
```

It measures:

> The spread of the middle 50% of the data.

---

# Example 1

Data:

```text
1, 2, 3, 4, 5
```

Already sorted.

Median (Q2):

```text
3
```

Lower half:

```text
1, 2
```

Q1:

```text
1.5
```

Upper half:

```text
4, 5
```

Q3:

```text
4.5
```

IQR:

[
4.5 - 1.5 = 3
]

IQR = **3**

---

# Example 2

Data:

```text
2, 4, 6, 8, 10, 12, 14, 16
```

Median:

```text
(8+10)/2 = 9
```

Lower half:

```text
2, 4, 6, 8
```

Q1:

```text
(4+6)/2 = 5
```

Upper half:

```text
10, 12, 14, 16
```

Q3:

```text
(12+14)/2 = 13
```

IQR:

[
13-5=8
]

IQR = **8**

---

# 5. Why Is IQR Important in AI/ML?

## Outlier Detection

This is its biggest use.

We define:

### Lower Bound

[
Q1 - 1.5 \times IQR
]

### Upper Bound

[
Q3 + 1.5 \times IQR
]

Any value outside these bounds is considered a **potential outlier**.

---

## Salary Example

Salaries:

```text
25,000
27,000
28,000
30,000
500,000
```

Mean and range get heavily affected.

But IQR focuses on the middle salaries.

So:

> The ₹500,000 salary is easily flagged as an outlier.

---

# 6. IQR vs Standard Deviation

| Feature                 | IQR   | Standard Deviation |
| ----------------------- | ----- | ------------------ |
| Uses all observations?  | ❌ No  | ✅ Yes              |
| Robust to outliers?     | ✅ Yes | ❌ No               |
| Used in boxplots?       | ✅ Yes | ❌ No               |
| Used in StandardScaler? | ❌ No  | ✅ Yes              |
| Best for skewed data?   | ✅ Yes | ❌ Not always       |

---

# 7. Python Implementation

## NumPy

```python
import numpy as np

data = [2, 4, 6, 8, 10, 12, 14, 16]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print(iqr)
```

Output:

```python
7.0
```

(Note: NumPy may use a slightly different quartile calculation method than our manual example.)

---

## Pandas

```python
import pandas as pd

s = pd.Series([2, 4, 6, 8, 10, 12, 14, 16])

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)

iqr = q3 - q1

print(iqr)
```

---

# Quick Recap

* Quartiles divide data into 4 equal parts.
* Q1 → 25%
* Q2 → Median
* Q3 → 75%

IQR:

```

IQR = Q3 - Q1

```
It measures:

> **The spread of the middle 50% of the data.**

Most importantly:

> **IQR is one of the most common methods for detecting outliers in EDA.**

Many people memorize:

> Lower Bound = Q1 − 1.5 × IQR
> Upper Bound = Q3 + 1.5 × IQR

without knowing **why 1.5?**

---

# Short Answer

> **1.5 is not a mathematical law.**
>
> It is an empirical rule proposed by John Tukey because it works well in practice for detecting outliers.

---

# The intuition

IQR measures the spread of the **middle 50%** of the data.

Now we need to decide:

> "How far away from this middle region should a value be before we call it an outlier?"

If we use too small a multiplier:

### Example: 0.5 × IQR

Many normal values get labeled as outliers.

Too strict.

---

If we use too large a multiplier:

### Example: 5 × IQR

Even truly unusual values won't be detected.

Too lenient.

---

# Why specifically 1.5?

Tukey found that:

> **1.5 × IQR provides a good balance.**

It catches unusually distant observations while keeping most normal observations.


# Salary Example

Suppose:

* Q1 = ₹40,000
* Q3 = ₹60,000

Then:

```text
IQR = 60,000 − 40,000 = 20,000
```

Bounds:

Lower:

```text
40,000 − 1.5 × 20,000
= ₹10,000
```

Upper:

```text
60,000 + 1.5 × 20,000
= ₹90,000
```

Any salary outside:

```text
₹10,000 to ₹90,000
```

is considered a **potential outlier**.

Notice:

* ₹70,000 → not an outlier.
* ₹85,000 → not an outlier.
* ₹5,00,000 → definitely an outlier.

---

# Is 1.5 always used?

No.

Sometimes people use:

* **1.5 × IQR** → mild outliers (most common)
* **3 × IQR** → extreme outliers

But in EDA, boxplots, and most ML preprocessing:

> **1.5 × IQR is the standard convention.**

# Topic 9: Quartiles (Q1, Q2, Q3)


# 1. What are Quartiles?

Quartiles divide an **ordered dataset into 4 equal parts**.

Think of it like cutting a pizza into 4 equal slices.

```text
0%       25%       50%       75%      100%
|---------|---------|---------|---------|
         Q1        Q2        Q3
```

Each section contains about **25% of the observations**.

---

# 2. What do Q1, Q2, and Q3 mean?

## Q1 (First Quartile)

> 25% of the data lies below this value.

Also called the **25th percentile**.

---

## Q2 (Second Quartile)

> 50% of the data lies below this value.

This is simply the **Median**.

---

## Q3 (Third Quartile)

> 75% of the data lies below this value.

Also called the **75th percentile**.

---

# The intuition

Suppose you have exam scores.

If:

```text
Q1 = 60
```

then:

> 25% of students scored **60 or less**.

---

If:

```text
Q2 = 75
```

then:

> Half the students scored **75 or less**.

---

If:

```text
Q3 = 90
```

then:

> 75% of students scored **90 or less**.

Only 25% scored above 90.

---

# Example 1

Data:

```text
2, 4, 6, 8, 10
```

Already sorted.

### Q2 (Median)

Middle value:

```text
6
```

---

### Q1

Lower half:

```text
2, 4
```

Median of lower half:

```text
(2 + 4)/2 = 3
```

Q1 = **3**

---

### Q3

Upper half:

```text
8, 10
```

Median of upper half:

```text
(8 + 10)/2 = 9
```

Q3 = **9**

---

Final:

| Quartile | Value |
| -------- | ----: |
| Q1       |     3 |
| Q2       |     6 |
| Q3       |     9 |

---

# Why are Quartiles important in AI/ML?

## a) Boxplots

Boxplots are built using:

* Minimum
* Q1
* Median (Q2)
* Q3
* Maximum

These five values summarize the distribution.

---

## b) Outlier Detection

IQR uses quartiles:

[
IQR = Q3 - Q1
]

Without quartiles, IQR doesn't exist.

---

## c) Understanding Data Distribution

Quartiles help answer questions like:

> "How do the bottom 25% compare to the top 25%?"

which is often more useful than just looking at the average.

---

# Quartiles vs Mean

Suppose salaries are:

```text
25,000
27,000
28,000
30,000
500,000
```

Mean becomes:

```text
122,000
```

which is misleading.

Quartiles still describe the bulk of employees reasonably well.

That's why quartiles are very useful for skewed data.

---

```python
import numpy as np

data = [2, 4, 6, 8, 10]

np.percentile(data, 25)
np.percentile(data, 50)
np.percentile(data, 75)
```

# Topic 10: Percentiles

This topic will feel easy because you already understand quartiles.

In fact:

> **Quartiles are just special percentiles.**

---

# 1. What are Percentiles?

Percentiles divide data into **100 equal parts**.

They answer:

> **"What percentage of the data lies below this value?"**

---

# 2. Intuition

Suppose you scored in the:

> **90th percentile** in an exam.

This does **NOT** mean you scored 90%.

It means:

> You scored **better than 90% of the students**.

Only 10% scored higher than you.

---

# 3. Relationship with Quartiles

Remember:

| Quartile    | Percentile      |
| ----------- | --------------- |
| Q1          | 25th Percentile |
| Q2 (Median) | 50th Percentile |
| Q3          | 75th Percentile |

So quartiles are just three important percentiles.

---

# Example 1

Suppose the 80th percentile of salaries is:

```text
₹80,000
```

This means:

> **80% of employees earn ₹80,000 or less.**

Only 20% earn more than ₹80,000.

---

# Example 2

Suppose your height is at the 95th percentile.

It means:

> **You are taller than 95% of the people in that group.**

Only 5% are taller.

---

# 4. Why are Percentiles Important in AI/ML?

## a) Understanding Distributions

They tell you how data is distributed.

Example:

* 90th percentile house price
* 95th percentile response time
* 99th percentile latency

---

## b) Outlier Detection

Instead of using IQR, sometimes people remove:

* Top 1%
* Bottom 1%

using percentiles.

Example:

Remove values below the 1st percentile and above the 99th percentile.

---

## c) Feature Understanding

They help answer questions like:

> "What does a high value actually mean?"

Example:

A customer spending in the 98th percentile is among the top spenders.

---

# Percentiles vs Percentage

This is a common confusion.

## Percentage

> Out of 100.

Example:

You scored **90%** marks.

---

## Percentile

> Relative ranking.

Example:

You are in the **90th percentile**, meaning you performed better than 90% of people.

They are NOT the same.

---

# Python Implementation

## NumPy

```python
import numpy as np

data = [10, 20, 30, 40, 50]

np.percentile(data, 25)   # 25th percentile
np.percentile(data, 50)   # Median
np.percentile(data, 75)   # 75th percentile
```

---

## Pandas

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50])

s.quantile(0.25)
s.quantile(0.50)
s.quantile(0.75)
```

---

  > **k% of observations lie below that value.**

  * Q1 = 25th percentile

## One intuition to remember forever

> **Mean tells you the average.**

> **Median tells you the middle.**

> **Percentiles tell you where you stand relative to everyone else.**

Actually,

> **Quantiles ARE calculated on sorted data.**

This is a very important point.

---

## What happens internally?

Suppose you do:

```python
s.quantile(0.25)
```

on

```python
[10, 2, 8, 4, 6]
```

You might think Pandas uses the data as-is.

But internally, it first sorts the data:

```text
2, 4, 6, 8, 10
```

Then it finds the required quantile.

So even though **you don't have to sort the data yourself**, the algorithm effectively does.

---

## Example

Data:

```python
import pandas as pd

s = pd.Series([10, 2, 8, 4, 6])

s.quantile(0.5)
```

Output:

```python
6.0
```

Why?

Because internally it becomes:

```text
2, 4, 6, 8, 10
```

The 50th percentile (median) is:

```text
6
```

---

## Then why didn't we sort in Python examples?

Because libraries like:

* Pandas (`quantile()`)
* NumPy (`percentile()`)

do it for you.

So:

### Manually

✅ You must sort first.

### Using Pandas/NumPy

✅ They sort internally.
