# Topic 5: Range

We are now entering the next section:

> **Measures of Dispersion**

You already know how to find the "center" of data (Mean, Median, Mode).

Now we'll answer a different question:

> **How spread out is the data?**

Two datasets can have the same mean but be completely different in how their values are distributed.

Range is the simplest way to measure this spread.

---

# What is Range?

**Range** tells us:

> **How far apart the smallest and largest values are.**

It measures the total spread of the dataset.

---

# Formula


Range = Maximum Value -  Minimum Value

Very simple.

Just subtract the smallest value from the largest value.

---

# Why Do We Need Range?

Imagine two classes.

### Class A Marks

```text
78, 79, 80, 81, 82
```

Mean = 80

Range:

```
82 − 78 = 4
```

---

### Class B Marks

```text
50, 65, 80, 95, 110
```

Mean = 80

Range:

```
110 − 50 = 60
```

---

Both classes have the **same mean (80)**.

But are they similar?

No.

* Class A students performed very consistently.
* Class B students had huge differences in performance.

Range reveals this difference.

---

# Why is Range Important in AI/ML?

Range gives an initial idea about the spread of a feature.

During EDA, you might ask:

> "How much do house prices vary?"

> "What is the spread of customer ages?"

A large range often indicates:

* High variability
* Possible outliers
* Need for deeper investigation

---

### Age

Seems reasonable.

# Limitation of Range

Range only uses **two values**:

* Smallest value
* Largest value

It completely ignores everything in between.

Example:

Dataset A:

```text
1, 2, 3, 4, 100
```

Range:

```
99
```

Dataset B:

```text
1, 50, 51, 52, 100
```

Range:

```
99
```

Same range.

But the datasets are clearly very different.

---

# Why Will We Learn Variance Next?

Because range is too simplistic.

We need a measure that considers **all observations**, not just the extremes.

That measure is **Variance**.

Variance is one of the most important statistical concepts in AI/ML.

---

# Python Implementation

## NumPy

```python
import numpy as np

data = [4, 8, 10, 15, 20]

range_value = np.max(data) - np.min(data)

print(range_value)
```

Output:

```python
16
```

---

## Pandas

```python
import pandas as pd

s = pd.Series([4, 8, 10, 15, 20])

range_value = s.max() - s.min()

print(range_value)
```

Output:

```python
16
```

---

# Quick Recap

* Range measures the **total spread** of data.
* Formula:

[
\text{Range} = \text{Max} - \text{Min}
]

* Same mean does not mean same spread.
* Large range can indicate high variability or possible outliers.
* Limitation:

  * Uses only two values.
  * Ignores the rest of the dataset.
* Because of this limitation, we move to **Variance** next.
