# What is Mode?

The **Mode** is the value that appears **most frequently** in a dataset.

It answers the question:

> **"Which value occurs the most?"**

Unlike mean and median, mode doesn't care about averages or positions.

It only cares about **frequency**.

---

# Why do we need Mode?

Sometimes the most common value is more useful than the average.

For example:

* Which shoe size is sold the most?
* Which blood group is most common?
* Which product color do customers buy the most?

The answer comes from the **mode**.

---

# Why is Mode important in AI/ML?

## a) Understanding Categorical Data

Mean and median don't work for categories.

Suppose a feature is:

```text
Red, Blue, Red, Green, Red
```

You cannot calculate an average color.

But you can find the most common color.

Mode = **Red**

---

## b) Handling Missing Values

During data preprocessing, sometimes missing values in categorical columns are filled using the **mode**.

Example:

| Gender  |
| ------- |
| Male    |
| Female  |
| Male    |
| Missing |

Since **Male** occurs most frequently:

Missing value → Male

You'll use this later in data cleaning.

---

# Types of Mode

## 1. Unimodal

Only one value appears most frequently.

Example:

```text
2, 3, 3, 4, 5
```

Mode = **3**

---

## 2. Bimodal

Two values have the highest frequency.

Example:

```text
1, 2, 2, 3, 3, 4
```

Modes = **2 and 3**

---

## 3. Multimodal

More than two values share the highest frequency.

Example:

```text
1, 1, 2, 2, 3, 3
```

Modes = **1, 2, and 3**

---

## Example 1

Exam scores:

```text
80, 90, 90, 70, 60
```

Frequency:

| Score | Count |
| ----- | ----- |
| 60    | 1     |
| 70    | 1     |
| 80    | 1     |
| 90    | 2     |

Mode = **90**

---

## Example 2

Favorite fruits:

```text
Apple
Banana
Apple
Mango
Apple
Banana
```

Frequency:

| Fruit  | Count |
| ------ | ----- |
| Apple  | 3     |
| Banana | 2     |
| Mango  | 1     |

Mode = **Apple**

---

# Limitations of Mode

Mode isn't always useful.

Example:

```text
1, 2, 3, 4, 5
```

Every value occurs once.

There is **no mode**.

Also, a dataset can have multiple modes, making interpretation harder.

---

# Mean vs Median vs Mode

| Measure | Represents          | Sensitive to Outliers? | Works for Categories? |
| ------- | ------------------- | ---------------------- | --------------------- |
| Mean    | Average             | Yes                    | No                    |
| Median  | Middle Value        | No                     | No                    |
| Mode    | Most Frequent Value | No                     | Yes                   |

---

# Python Implementation

## NumPy

NumPy doesn't have a direct mode function.

Usually, Pandas is preferred.

---

## Pandas

```python
import pandas as pd

s = pd.Series([80, 90, 90, 70, 60])

s.mode()
```

Output:

```python
0    90
dtype: int64
```

If multiple modes exist:

```python
s = pd.Series([1, 2, 2, 3, 3])

s.mode()
```

Output:

```python
0    2
1    3
dtype: int64
```

# Quick Recap

* Mode = **most frequently occurring value**.
* Useful for **categorical data**.
* Can be:

  * Unimodal
  * Bimodal
  * Multimodal
* Often used for:

  * Understanding categories
  * Filling missing categorical values.
* Not affected by outliers.

# Topic 4: When to Use Mean vs Median vs Mode

This is one of the most useful intuitions in AI/ML and EDA.

Knowing **which measure to trust** is more important than just knowing how to calculate them.

---

# The Main Question

Suppose someone gives you a dataset.

How do you decide whether to use:

* Mean?
* Median?
* Mode?

The answer depends on **the type and shape of the data**.

---

# 1. Use Mean When...

### The data is numerical and has no extreme outliers.

Mean uses every value, so it works well when the dataset is fairly balanced.

### Example 1

Student marks:

```text
72, 75, 78, 80, 85
```

Average mark:

```text
78
```

This represents the class well because there are no unusual values.

---

# 2. Use Median When...

### The data has outliers or is skewed.

Median focuses on the middle position.

It ignores how large or small extreme values are.

### Example 2

Monthly salaries:

```text
25,000
27,000
28,000
30,000
500,000
```

Mean:

```text
122,000
```

Median:

```text
28,000
```

Which better represents a typical employee?

👉 Median.

---

# 3. Use Mode When...

### The data is categorical.

If values are labels instead of numbers, mean and median don't make sense.

Example:

```text
Red
Blue
Red
Green
Red
```

Mode:

```text
Red
```
---

# Quick Decision Table

| Situation                             | Best Measure |
| ------------------------------------- | ------------ |
| Numerical data without outliers       | Mean         |
| Numerical data with outliers/skewness | Median       |
| Categorical data                      | Mode         |

---

# How Data Scientists Think

When performing EDA, they often ask:

### Step 1

Is this feature numerical or categorical?

If categorical:

→ Check Mode.

---

### Step 2

If numerical:

Are there extreme values?

* No → Mean
* Yes → Median
