# Topic 2: Median

# What is Median?

The **Median** is the **middle value** of a dataset after arranging the values in order.

It answers the question:

> "What value lies exactly at the center of the data?"

Unlike the mean, the median focuses on the **position** of values rather than their total.

---

# Why do we need Median?

Sometimes, a few extremely large or small values distort the mean.

Median gives us the **typical middle value**, making it more reliable in such cases.

---

# Why is Median important in AI/ML?

## a) Robust to Outliers

Suppose customer incomes contain a few billionaires.

The mean income becomes misleading.

Median provides a better representation of a "typical" customer.

---

## b) Data Understanding (EDA)

During exploratory data analysis, comparing **mean vs median** helps detect skewness.

For example:

* Mean ≈ Median → data may be symmetric.
* Mean > Median → possible positive skew.
* Mean < Median → possible negative skew.

You'll use this later when studying distributions.

---

# How to Find the Median

### Step 1:

Arrange the data in ascending order.

### Step 2:

* If the number of observations is **odd**:
  → Median = middle value.

* If the number of observations is **even**:
  → Median = average of the two middle values.

---

# Example 1 (Odd Number of Values)

Data:

```
7, 2, 9, 4, 5
```

Arrange:

```
2, 4, 5, 7, 9
```

Middle value:

```
5
```

Median = **5**

---

# Example 2 (Even Number of Values)

Data:

```
10, 2, 8, 4
```

Arrange:

```
2, 4, 8, 10
```

Two middle values:

```
4 and 8
```

Median:

```
(4 + 8) / 2 = 6
```

Median = **6**

---

# Mean vs Median

Consider salaries:

```
25,000
27,000
28,000
30,000
500,000
```

### Mean

```
= 122,000
```

### Median

Ordered data:

```
25,000
27,000
28,000
30,000
500,000
```

Middle value:

```
28,000
```

Median = **28,000**

### Which better represents the typical salary?

👉 Median.

Because it is **not heavily affected by the outlier (500,000).**

# Python Implementation

## NumPy

```python
import numpy as np

data = [7, 2, 9, 4, 5]

np.median(data)
```

Output:

```python
5.0
```

---

## Pandas

```python
import pandas as pd

s = pd.Series([7, 2, 9, 4, 5])

s.median()
```

Output:

```python
5.0
```

---

# Quick Recap

* Median = middle value of ordered data.
* Arrange data before finding it.
* Odd observations → one middle value.
* Even observations → average of two middle values.
* Less affected by outliers.
* Useful in EDA and understanding skewed data.
* Comparing mean and median often reveals the shape of the distribution.
