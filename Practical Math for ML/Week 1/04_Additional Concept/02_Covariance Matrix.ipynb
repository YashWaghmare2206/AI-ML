# Covariance Matrix

Now that you know covariance between **two variables**, let's extend it to **many variables**.

A covariance matrix stores the covariance between **every pair of features**.

---

## Why Do We Need It?

Suppose a dataset has 3 features:

* Height
* Weight
* Age

We can compute:

* Cov(Height, Weight)
* Cov(Height, Age)
* Cov(Weight, Age)

Instead of storing them separately, we put everything into one matrix.

That matrix is called the **covariance matrix**.

## Why Covariance Matrix Matters in ML

This is one of the most important matrices in ML.

### PCA

PCA starts by computing the covariance matrix.

Then it finds:

* Eigenvalues
* Eigenvectors

of that covariance matrix.

Those eigenvectors become the principal components.

---

### Feature Relationships

Covariance matrix helps identify:

* Related features
* Redundant information
* Important directions in data

---

### Multivariate Statistics

Almost every multivariate statistical method uses covariance matrices.

---

## Relationship with Linear Algebra

Notice:

* Covariance Matrix → Matrix
* PCA → Eigenvectors of that matrix
* SVD → Closely related to covariance matrices

This is where Linear Algebra and Statistics begin to connect.

---

## Key Takeaways

* Covariance matrix stores covariance between all feature pairs.
* Diagonal entries = variances.
* Off-diagonal entries = covariances.
* Covariance matrix is always symmetric.
* Used heavily in PCA and dimensionality reduction.

### One-line Summary

> A covariance matrix is a table that stores the variance of each feature and the covariance between every pair of features in a dataset.
