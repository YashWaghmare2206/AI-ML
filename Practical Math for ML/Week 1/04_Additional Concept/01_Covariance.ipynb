---

# What Covariance Measures

Covariance tells us:

> How two variables change together.

Suppose we have:

* Height
* Weight

If taller people generally weigh more:

* Height increases
* Weight increases

Covariance is positive.

---

## Positive Covariance

When variables move in the same direction.

| Height | Weight |
| ------ | ------ |
| ↑      | ↑      |
| ↓      | ↓      |

Covariance > 0

---

## Negative Covariance

When variables move in opposite directions.

Example:

* Speed increases
* Travel time decreases

| Speed | Time |
| ----- | ---- |
| ↑     | ↓    |
| ↓     | ↑    |

Covariance < 0

---

## Zero Covariance

No clear linear relationship.

One variable changing gives no information about the other.

Covariance ≈ 0

# Understanding the Formula

### Step 1

Subtract means.



This tells us whether a value is above or below average.

---

### Step 2

Multiply deviations.



---

If both are above average:

Positive × Positive = Positive

---

If both are below average:

Negative × Negative = Positive

---

If one is above and one below:

Positive × Negative = Negative

---

### Step 3

Average everything.

The sign of the average tells the relationship.

---

# Important Limitation

Covariance tells direction.

It does **not tell strength**.

Example:

Dataset A:

**`Cov=10`**

Dataset B:

**`Cov=1000`**

You cannot conclude B is more related.

Why?

Because covariance depends on units.

---

Example:

Height in:

* meters
* centimeters

produces completely different covariance values.

This is why we later use **Correlation**.

Correlation normalizes covariance.

---

# Relationship with Variance

Variance is a special case of covariance.



A variable's covariance with itself equals its variance.

Very important fact.

---

# Why Covariance Matters in ML

Suppose features:

* Height
* Weight

Large positive covariance means:

The features contain similar information.

---

PCA uses covariance to find:

* Important directions
* Redundant features
* Maximum variance directions

The entire PCA algorithm starts from the covariance matrix.
