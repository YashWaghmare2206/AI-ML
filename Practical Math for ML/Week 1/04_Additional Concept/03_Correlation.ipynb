# Correlation

You already know that covariance measures whether two variables move together.

The problem is:

> Covariance depends on units.

That's why we use **Correlation**.

---

## What Correlation Measures

Correlation measures:

> The strength and direction of a linear relationship between two variables.

Unlike covariance:

* Always between **-1 and +1**
* Independent of units
* Easy to compare

## Correlation = +1

Perfect positive relationship.

As X increases, Y increases exactly.

Example:

| X | Y |
| - | - |
| 1 | 2 |
| 2 | 4 |
| 3 | 6 |
| 4 | 8 |

[
r = 1
]

---

## Correlation = -1

Perfect negative relationship.

As X increases, Y decreases exactly.

Example:

| X | Y |
| - | - |
| 1 | 8 |
| 2 | 6 |
| 3 | 4 |
| 4 | 2 |

[
r=-1
]

---

## Correlation = 0

No linear relationship.

Knowing X tells you almost nothing about Y linearly.

[
r=0
]

---

## Interpretation Table

| Correlation | Meaning                |
| ----------- | ---------------------- |
| +1          | Perfect positive       |
| +0.8        | Strong positive        |
| +0.5        | Moderate positive      |
| 0           | No linear relationship |
| -0.5        | Moderate negative      |
| -0.8        | Strong negative        |
| -1          | Perfect negative       |

---

## Why Correlation Is Better Than Covariance

Suppose:

### Dataset A

Height in meters.

Covariance = 10

### Dataset B

Height in centimeters.

Covariance = 1000

The relationship is identical.

Only units changed.

Covariance changes.

Correlation stays the same.

That's why correlation is usually preferred.

# Spearman Correlation

Spearman uses **ranks** instead of actual values.

Example:

| Score | Rank |
| ----- | ---- |
| 95    | 1    |
| 90    | 2    |
| 80    | 3    |
| 70    | 4    |

Instead of comparing values, it compares order.

---

## What Spearman Measures

Spearman measures:

> Whether variables move in the same monotonic direction.

As one increases, does the other generally increase?

---

## Why Useful?

Consider:



As X increases:

Y always increases.

Pearson may struggle.

Spearman remains high.

---

## Pearson vs Spearman

| Pearson                      | Spearman                        |
| ---------------------------- | ------------------------------- |
| Uses actual values           | Uses ranks                      |
| Measures linear relationship | Measures monotonic relationship |
| Sensitive to outliers        | More robust                     |
| Most common in ML            | Used when data is non-linear    |

## Why Correlation Matters in ML

### Feature Selection

Highly correlated features often carry similar information.

---

### Multicollinearity Detection

If two features have correlation close to:

**`+1`**

or

**`-1`**

they may be redundant.

---

### EDA

Correlation matrices are one of the first things data scientists check.

---

## Key Takeaways

* Correlation measures strength and direction of relationship.
* Range:

**`[-1,+1]`**

* Pearson measures linear relationships.
* Spearman measures monotonic relationships.
* Correlation removes unit effects.
* Highly correlated features may be redundant.

### One-line Summary

> Correlation is a normalized version of covariance that measures how strongly and in what direction two variables are related.
