# Short Definition

> **IQR (Interquartile Range) measures the spread of the middle 50% of the data.**

It is used by boxplots to determine potential outliers.

---

# The 5 Numbers Behind a Boxplot

```text
Minimum
↓
Q1
↓
Median (Q2)
↓
Q3
↓
Maximum
```

---

## Q1 (25th Percentile)

> 25% of the data lies below Q1.

Example:

```text
Data:
10, 12, 14, 16, 18, 20, 22, 24

Q1 ≈ 13
```

---

## Q2 (Median)

> Middle value of the data.

Example:

```text
Median = 17
```

---

## Q3 (75th Percentile)

> 75% of the data lies below Q3.

Example:

```text
Q3 ≈ 21
```

---

# IQR Formula

```python
IQR = Q3 - Q1
```

Example:

```text
Q1 = 13
Q3 = 21

IQR = 21 - 13 = 8
```

---

# Lower and Upper Bounds

These determine potential outliers.

## Lower Bound

```python
Lower = Q1 - 1.5 * IQR
```

---

## Upper Bound

```python
Upper = Q3 + 1.5 * IQR
```

---

# Example

Suppose:

```text
Q1 = 13
Q3 = 21
IQR = 8
```

Then:

```python
Lower = 13 - 1.5×8 = 1
Upper = 21 + 1.5×8 = 33
```

Any value:

```text
< 1
or
> 33
```

is considered a potential outlier.

---

# Why 1.5 × IQR?

> It is a rule proposed by **John Tukey**.

It:

* Works well for many distributions.
* Doesn't assume normality.
* Flags unusually distant observations.
* Is a guideline, not a strict law.

---

# Keep or Remove?

Ask:

### Remove if:

```text
Age = 250 years
Negative house price
Typing mistakes
Sensor errors
```

---

### Keep if:

```text
Luxury mansions
High-income customers
Rare but genuine observations
```

Especially in House Prices:

> Expensive houses are often genuine outliers.
