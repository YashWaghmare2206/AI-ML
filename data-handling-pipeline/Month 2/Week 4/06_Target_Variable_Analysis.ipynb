## 6.1 Distribution of Target

### What?

Study how the target values are distributed.

### Why?

To understand whether the target is normal, skewed, or has outliers.

### Example

```python
sns.histplot(df["SalePrice"], kde=True)
```

Question answered:

> "How are house prices distributed?"

---

## 6.2 Target Skewness

### What?

Measure whether the target leans to one side.

### Why?

Highly skewed targets can negatively affect some models.

### Example

```python
df["SalePrice"].skew()
```

Example output:

```text
1.88
```

Interpretation:

> SalePrice is heavily right-skewed.

---

## 6.3 Target Transformations

### What?

Transform the target to reduce skewness.

### Why?

Makes the distribution more normal and can improve model performance.

### Example

```python
df["SalePrice_log"] = np.log1p(df["SalePrice"])
```

Question answered:

> "Should I use log-transformed SalePrice instead?"
