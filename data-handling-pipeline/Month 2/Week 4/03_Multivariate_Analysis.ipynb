---

# Short Definition

> A **Correlation Heatmap** visualizes the correlation matrix using colors, making it easier to spot patterns among many variables.

Think of it as:

```text
Correlation Matrix
        +
Colors
=
Correlation Heatmap
```

---

# Why do we use it?

Instead of reading this:

|   |    A |    B |    C |
| - | ---: | ---: | ---: |
| A | 1.00 | 0.75 | 0.20 |
| B | 0.75 | 1.00 | 0.10 |
| C | 0.20 | 0.10 | 1.00 |

we can immediately see:

```text
Strong → Dark
Weak → Light
Negative → Different color
```

It helps answer:

* Which features are highly correlated?
* Which features may cause multicollinearity?
* Which features relate strongly to the target?

---

# Intuition

Suppose you have 50 numerical features.

Without a heatmap:

```text
50 × 50 = 2500 numbers
```

Impossible to understand quickly.

With a heatmap:

> You can spot patterns in seconds.

---

# Advanced Usage 1 → Mask Upper Triangle

Problem:

The heatmap is symmetric.

```text
A-B = B-A
```

So half the plot is duplicated.

---

## Entire Syntax

```python
import numpy as np

corr = df.corr(numeric_only=True)

mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.show()
```

---

# Important Parameters

## `mask`

Hide specific cells.

Example:

```python
mask=mask
```

Result:

Only one triangle is shown.

---

Select rows and columns.

```python
df.loc[rows, columns]
```

> Keep only the important features.

---

# Short Definition

> A **Pairplot** automatically creates scatter plots between every pair of numerical variables and histograms/KDE plots on the diagonal.

Think of it as:

> **Scatter Plot Matrix**

---

# Why do we use it?

Instead of manually doing:

```text
A vs B
A vs C
A vs D
B vs C
B vs D
C vs D
...
```

Pairplot generates them all automatically.

It helps answer:

* Which variables are related?
* Which variables are not related?
* Are there clusters?
* Are there outliers?
* What does the distribution of each variable look like?

---

# Intuition

Suppose you have 3 numerical features:

```text
SalePrice
GrLivArea
GarageArea
```

Normally, you'd make:

```text
SalePrice vs GrLivArea
SalePrice vs GarageArea
GrLivArea vs GarageArea
```

and also look at the individual distributions.

Pairplot does all of this in **one command**.

---

# Example

Suppose you select:

```python
cols = [
    "SalePrice",
    "GrLivArea",
    "GarageArea"
]
```

Pairplot creates:

```text
                SalePrice  GrLivArea  GarageArea

SalePrice       Hist       Scatter    Scatter

GrLivArea       Scatter    Hist       Scatter

GarageArea      Scatter    Scatter    Hist
```

# Topic 3 → Multivariate Analysis

## Subtopic 3.4 → Grouped Aggregations (`groupby`)

---

# Short Definition

> `groupby()` is used to **split the data into groups and calculate statistics for each group separately.**

Think of it as:

```text
Group → Calculate → Combine
```

---

# Why do we use it?

It helps answer questions like:

* What is the average SalePrice in each Neighborhood?
* Which MSZoning has the highest median SalePrice?
* How many houses are there in each Neighborhood?
* What are the min and max prices by category?

---

# Intuition

Suppose you have:

| Neighborhood | SalePrice |
| ------------ | --------: |
| NAmes        |    150000 |
| NAmes        |    170000 |
| CollgCr      |    220000 |
| CollgCr      |    240000 |

You mentally do:

```text
NAmes
↓
150000, 170000
↓
Mean = 160000

CollgCr
↓
220000, 240000
↓
Mean = 230000
```

`groupby()` automates this.

---

# General Syntax

```python
df.groupby("GroupColumn")["ValueColumn"].aggregation()
```

Example:

```python
df.groupby("Neighborhood")["SalePrice"].mean()
```

---

# 1. Mean

## Short Definition

> Calculates the average value for each group.

---

## Entire Syntax

```python
df.groupby(
    "Neighborhood"
)["SalePrice"].mean()
```

---

### Output

| Neighborhood | SalePrice |
| ------------ | --------: |
| CollgCr      |    230000 |
| NAmes        |    160000 |

---

### Interpretation

> Houses in CollgCr are more expensive on average.

---

# 2. Median

## Short Definition

> Calculates the middle value for each group.

---

## Entire Syntax

```python
df.groupby(
    "Neighborhood"
)["SalePrice"].median()
```

---

### Why use Median?

Because:

> Median is less affected by outliers.

Example:

```text
100k, 120k, 900k
```

Mean becomes large.

Median stays representative.

---

# 3. Sum

## Short Definition

> Adds values within each group.

---

## Entire Syntax

```python
df.groupby(
    "Neighborhood"
)["SalePrice"].sum()
```

---

### Interpretation

> Total sales amount contributed by each neighborhood.

---

# 4. Count

## Short Definition

> Counts non-missing observations in each group.

---

## Entire Syntax

```python
df.groupby(
    "Neighborhood"
)["SalePrice"].count()
```

---

### Interpretation

> Number of houses sold in each neighborhood.

---

# 5. Size

## Short Definition

> Counts all rows in each group.

---

## Entire Syntax

```python
df.groupby(
    "Neighborhood"
).size()
```

---

# Count vs Size

Suppose:

| Neighborhood | SalePrice |
| ------------ | --------: |
| NAmes        |    150000 |
| NAmes        |       NaN |

### Count

```python
df.groupby("Neighborhood")["SalePrice"].count()
```

Output:

```text
NAmes    1
```

---

### Size

```python
df.groupby("Neighborhood").size()
```

Output:

```text
NAmes    2
```

---

# 6. Multiple Aggregations

## Short Definition

> Calculate multiple statistics at once.

---

| Syntax                                     | Returns            |
| ------------------------------------------ | ------------------ |
| `df.groupby("A")`                          | `DataFrameGroupBy` |
| `df.groupby("A")["B"]`                     | `SeriesGroupBy`    |
| `df.groupby("A")["B"].mean()`              | `Series`           |
| `df.groupby("A")[["B","C"]].mean()`        | `DataFrame`        |
| `df.groupby("A")["B"].agg(["mean","max"])` | `DataFrame`        |
| `df.groupby("A").size()`                   | `Series`           |

## Entire Syntax

```python
df.groupby(
    "Neighborhood"
)["SalePrice"].agg(
    ["mean", "median", "min", "max"]
)
```

---

### Output

| Neighborhood |   mean | median |    min |    max |
| ------------ | -----: | -----: | -----: | -----: |
| NAmes        | 160000 | 160000 | 150000 | 170000 |
| CollgCr      | 230000 | 230000 | 220000 | 240000 |

---

# Pivot Table vs GroupBy

| GroupBy                       | Pivot Table           |
| ----------------------------- | --------------------- |
| More flexible                 | Easier summaries      |
| Can return Series/DataFrame   | Always DataFrame      |
| Requires reshaping sometimes  | Built-in table format |
| Better for complex operations | Better for reporting  |

---

# When to Use Which?

Use **GroupBy** when:

* Doing transformations
* Complex operations
* Custom logic

Use **Pivot Tables** when:

# Common Mistakes

### Mistake 1

❌ Forgetting `values`.

You may summarize unintended columns.

---

### Mistake 2

❌ Forgetting the default aggregation.

Pivot Tables use:

```python
aggfunc="mean"
```

unless specified otherwise.

---

### Mistake 3

❌ Expecting a Series.

Pivot Tables always return:

> **DataFrames**

---

# Most Useful House Prices Syntax

### Average SalePrice by Neighborhood

```python
pd.pivot_table(
    data=df,
    values="SalePrice",
    index="Neighborhood"
)
```

---

### Neighborhood × Zoning Analysis

```python
pd.pivot_table(
    data=df,
    values="SalePrice",
    index="Neighborhood",
    columns="MSZoning"
)
```

---

### Multiple Statistics

```python
pd.pivot_table(
    data=df,
    values="SalePrice",
    index="Neighborhood",
    aggfunc=["mean","median","count"]
)
```
