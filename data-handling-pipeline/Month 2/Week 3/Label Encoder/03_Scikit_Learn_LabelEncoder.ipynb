# Topic 3: Scikit-Learn's `LabelEncoder`

## Subtopic 3.1 – `fit()`

This is one of the most important concepts in Scikit-Learn.

Many beginners memorize `fit()` without understanding what it actually does.

After this lesson, you'll understand exactly what happens behind the scenes.

---

# 1. What is `fit()`?

> **`fit()` means "learn from the data."**

For `LabelEncoder`, learning means:

> **Find all unique categories and remember how they should be encoded.**

---

# 2. Syntax

```python
encoder.fit(data)
```

Example:

```python
encoder.fit(["Low", "Medium", "High"])
```

---

# 3. What Does `fit()` Actually Do?

Suppose we have:

```python
levels = ["Low", "Medium", "High", "Medium", "Low"]
```

When we run:

```python
encoder.fit(levels)
```

the encoder does two things:

### Step 1: Finds unique categories

From:

```text
Low
Medium
High
Medium
Low
```

it extracts:

```text
Low
Medium
High
```

Notice that duplicates are removed.

---

### Step 2: Sorts them alphabetically

Scikit-Learn's `LabelEncoder` sorts categories alphabetically.

Alphabetical order:

```text
High
Low
Medium
```

---

### Step 3: Assigns numbers starting from 0

```text
High   → 0
Low    → 1
Medium → 2
```

and remembers this mapping internally.

---

# 4. Very Important

`fit()` does NOT change your data.

Many beginners think:

```python
encoder.fit(levels)
```

will give:

```text
[1, 2, 0, 2, 1]
```

❌ Wrong.

It only learns the mapping.

Your original data remains unchanged.

---

# 5. Real-World Analogy

Imagine a teacher enters a classroom.

Students:

```text
Rahul
Aditi
Priya
Rahul
```

The teacher first prepares the attendance list.

They note:

```text
Aditi
Priya
Rahul
```

and assign roll numbers:

```text
Aditi → 0
Priya → 1
Rahul → 2
```

The teacher has **learned** the students.

But attendance hasn't been marked yet.

Similarly:

> `fit()` prepares the mapping.

It doesn't encode the data.

---

# 6. Python Example

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

levels = ["Low", "Medium", "High", "Medium", "Low"]

encoder.fit(levels)

print(encoder.classes_)
```

---

## Output

```python
['High' 'Low' 'Medium']
```

---

# 7. What is `classes_`?

`classes_` is an attribute that stores what the encoder learned.

After fitting:

```python
encoder.classes_
```

shows:

```text
['High' 'Low' 'Medium']
```

This means:

The encoder learned these categories in alphabetical order.

---

# 8. How Can We Know the Mapping?

Since numbering starts from 0:

| Category | Number |
| -------- | -----: |
| High     |      0 |
| Low      |      1 |
| Medium   |      2 |

The mapping is:

```text
High   → 0
Low    → 1
Medium → 2
```

---

# 9. Example with Cities

Suppose:

```python
cities = ["Mumbai", "Delhi", "Pune", "Delhi"]
```

We do:

```python
encoder.fit(cities)

print(encoder.classes_)
```

Output:

```python
['Delhi' 'Mumbai' 'Pune']
```

Mapping becomes:

| City   | Encoding |
| ------ | -------: |
| Delhi  |        0 |
| Mumbai |        1 |
| Pune   |        2 |

Again:

> `fit()` only learned this mapping.

No encoding has happened yet.

---

# 10. Why Is `fit()` Needed?

Imagine you want to transform:

```text
Low
Medium
High
```

into numbers.

The encoder first needs to know:

```text
What categories exist?
```

Without learning them, it cannot encode anything.

So:

```text
fit()
↓
learn categories
↓
transform()
↓
convert categories into numbers
```

---

# 11. What Happens If We Skip `fit()`?

Suppose:

```python
encoder = LabelEncoder()

encoder.transform(["Low", "High"])
```

Python raises an error similar to:

```python
NotFittedError:
This LabelEncoder instance is not fitted yet.
```

because the encoder doesn't know the categories.

It hasn't learned anything.

---

# Important Points to Remember

* `fit()` means "learn from the data."
* It finds unique categories.
* It removes duplicates.
* It sorts categories alphabetically.
* It assigns numbers internally starting from 0.
* It stores learned categories in `classes_`.
* It does NOT modify your data.
* It must usually be called before `transform()`.

---

# Topic 3: Scikit-Learn's `LabelEncoder`

## Subtopic 3.2 – `transform()`

In the previous subtopic, we learned that:

> **`fit()` learns the categories and their mapping.**

Now we'll learn:

> **`transform()` actually converts the categories into numbers.**

This is where the real encoding happens.

---

# 1. What is `transform()`?

`transform()` uses the mapping learned by `fit()` and applies it to the data.

In simple words:

> **`transform()` = "Use what you learned to encode the data."**

---

# 2. Syntax

```python
encoder.transform(data)
```

Example:

```python
encoder.transform(["Low", "Medium", "High"])
```

---

# 3. How Does It Work?

Suppose we have:

```python
levels = ["Low", "Medium", "High", "Medium", "Low"]
```

First, we fit the encoder:

```python
encoder.fit(levels)
```

The encoder learns:

```text
High   → 0
Low    → 1
Medium → 2
```

Now, when we do:

```python
encoder.transform(levels)
```

it replaces each category using the learned mapping.

---

# 4. Step-by-Step Example

### Original Data

```text
Low
Medium
High
Medium
Low
```

### Learned Mapping (from `fit()`)

```text
High   → 0
Low    → 1
Medium → 2
```

### After `transform()`

```text
1
2
0
2
1
```

---

# 5. Python Example

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

levels = ["Low", "Medium", "High", "Medium", "Low"]

encoder.fit(levels)

encoded = encoder.transform(levels)

print(encoded)
```

### Output

```python
[1 2 0 2 1]
```

---

# 6. Output Explanation

The encoder had learned:

| Category | Encoding |
| -------- | -------: |
| High     |        0 |
| Low      |        1 |
| Medium   |        2 |

Therefore:

| Original | Encoded |
| -------- | ------: |
| Low      |       1 |
| Medium   |       2 |
| High     |       0 |
| Medium   |       2 |
| Low      |       1 |

Result:

```python
[1 2 0 2 1]
```

---

# 7. Real-World Analogy

Imagine a teacher has already prepared roll numbers.

| Student | Roll Number |
| ------- | ----------: |
| Aditi   |           0 |
| Priya   |           1 |
| Rahul   |           2 |

Preparing this list is like:

```python
fit()
```

Now, marking attendance:

```text
Rahul
Aditi
Priya
Rahul
```

becomes:

```text
2
0
1
2
```

This is like:

```python
transform()
```

The teacher uses the already learned roll numbers.

---

# 8. Can We Transform New Data?

Yes, but only if the categories were seen during `fit()`.

Example:

```python
encoder.fit(["Low", "Medium", "High"])
```

Learned:

```text
High   → 0
Low    → 1
Medium → 2
```

Now:

```python
encoder.transform(["Medium", "Low"])
```

Output:

```python
[2 1]
```

This works because both categories were already learned.

---

# 9. What If We Transform an Unseen Category?

Suppose:

```python
encoder.fit(["Low", "Medium", "High"])
```

Then we try:

```python
encoder.transform(["Very High"])
```

Python raises an error:

```python
ValueError:
y contains previously unseen labels
```

because:

> The encoder never learned `"Very High"`.

We'll study unseen categories in detail later.

---

# 10. Does `transform()` Change the Original Data?

No.

Example:

```python
levels = ["Low", "Medium", "High"]

encoded = encoder.transform(levels)
```

Now:

```python
print(levels)
```

Output:

```python
['Low', 'Medium', 'High']
```

and:

```python
print(encoded)
```

Output:

```python
[1 2 0]
```

The original data stays the same.

The transformed values are returned separately.

---

# 11. Visual Flow

```text
Original Data
↓
Low
Medium
High
Medium
Low

fit()
↓
Learn Mapping

High   → 0
Low    → 1
Medium → 2

transform()
↓
Apply Mapping

1
2
0
2
1
```

---

## Difference Between `fit()` and `transform()`

| Method        | What It Does                              |
| ------------- | ----------------------------------------- |
| `fit()`       | Learns categories and creates the mapping |
| `transform()` | Uses the learned mapping to encode data   |

# Topic 3: Scikit-Learn's `LabelEncoder`

## Subtopic 3.5 – `fit_transform()`

So far, we have learned:

* `fit()` → Learn the categories and create the mapping.
* `transform()` → Use that mapping to encode the data.

Now let's learn:

> **`fit_transform()` = `fit()` + `transform()`**

It combines both steps into one.

---

# 1. What is `fit_transform()`?

`fit_transform()` first:

1. Learns the categories (`fit`)
2. Immediately encodes the data (`transform`)

In simple words:

> **Learn and encode in one step.**

---

# 2. Syntax

```python
encoder.fit_transform(data)
```

Example:

```python
encoder.fit_transform(["Low", "Medium", "High"])
```

---

# 3. How Does It Work Internally?

Suppose we have:

```python
levels = ["Low", "Medium", "High", "Medium", "Low"]
```

When we do:

```python
encoder.fit_transform(levels)
```

Internally, Python does this:

```python
encoder.fit(levels)

encoded = encoder.transform(levels)
```

It simply combines the two operations.

---

# 4. Step-by-Step Example

### Original Data

```text
Low
Medium
High
Medium
Low
```

---

### Step 1: `fit()`

Find unique categories:

```text
Low
Medium
High
```

Sort alphabetically:

```text
High
Low
Medium
```

Create mapping:

```text
High   → 0
Low    → 1
Medium → 2
```

---

### Step 2: `transform()`

Apply the mapping:

```text
Low    → 1
Medium → 2
High   → 0
Medium → 2
Low    → 1
```

---

### Final Output

```text
1
2
0
2
1
```

---

# 5. Python Example

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

levels = ["Low", "Medium", "High", "Medium", "Low"]

encoded = encoder.fit_transform(levels)

print(encoded)
```

---

## Output

```python
[1 2 0 2 1]
```

---

# 6. Output Explanation

The encoder learned:

| Category | Encoding |
| -------- | -------: |
| High     |        0 |
| Low      |        1 |
| Medium   |        2 |

Then immediately transformed the data.

Result:

| Original | Encoded |
| -------- | ------: |
| Low      |       1 |
| Medium   |       2 |
| High     |       0 |
| Medium   |       2 |
| Low      |       1 |

Output:

```python
[1 2 0 2 1]
```

---

# 7. Real-World Analogy

Imagine a teacher enters a new classroom.

Instead of:

### First

Preparing roll numbers:

```text
Aditi → 0
Priya → 1
Rahul → 2
```

and then

### Second

Marking attendance,

the teacher does both at once.

As each student is processed:

* Learn the students.
* Record attendance using roll numbers.

This is similar to:

```python
fit_transform()
```

---

# 8. Why Use `fit_transform()`?

It saves time and code.

Instead of writing:

```python
encoder.fit(data)

encoded = encoder.transform(data)
```

you can simply write:

```python
encoded = encoder.fit_transform(data)
```

Both give the same result.

---

# 9. Should We Always Use `fit_transform()`?

No.

It depends on the situation.

---

## During Training

When working with training data:

```python
encoded_train = encoder.fit_transform(train_data)
```

✅ Common practice.

Because the encoder needs to learn the categories.

---

## During Testing

For test data:

❌ Do NOT do this:

```python
encoder.fit_transform(test_data)
```

Why?

Because the encoder would relearn categories from the test set.

Instead:

```python
encoded_test = encoder.transform(test_data)
```

Use the mapping learned from training data.

---

# Example

### Training Data

```python
train = ["Low", "Medium", "High"]
```

Fit and transform:

```python
encoder.fit_transform(train)
```

Learned mapping:

```text
High   → 0
Low    → 1
Medium → 2
```

---

### Test Data

```python
test = ["Medium", "Low"]
```

Correct approach:

```python
encoder.transform(test)
```

Output:

```python
[2 1]
```

---

### Wrong Approach

```python
encoder.fit_transform(test)
```

Now the encoder relearns from:

```text
Medium
Low
```

New mapping:

```text
Low    → 0
Medium → 1
```

The original mapping is lost.

This creates inconsistent encodings.

---

# 10. Visual Flow

## Separate Steps

```text
Data
↓
fit()
↓
Learn mapping
↓
transform()
↓
Encoded data
```

---

## Combined Step

```text
Data
↓
fit_transform()
↓
Learn mapping + Encode
```


## **Important**

### Short Answer

Because:

> **`fit_transform()` creates a new mapping, while `transform()` uses the mapping learned from the training data.**

The model was trained using the training mapping. If you use `fit_transform()` on test data, the same categories may get different numbers.

### Example

Training data:

```text
Low, Medium, High
```

Training mapping:

```text
High   → 0
Low    → 1
Medium → 2
```

Model learns using these numbers.

---

Test data:

```text
Medium, Low
```

Using `transform()`:

```text
Medium → 2
Low    → 1
```

✅ Same mapping as training.

Using `fit_transform()`:

```text
Low    → 0
Medium → 1
```

❌ New mapping.

---

### One-line rule to remember:

> **Train: `fit_transform()` | Test: `transform()`**

because **the test data must use the same encoding rules that the model learned during training.**


# Topic 3: Scikit-Learn's `LabelEncoder`

## Subtopic 3.6 – `inverse_transform()`

So far, we have learned:

* `fit()` → Learn categories
* `transform()` → Convert categories into numbers
* `fit_transform()` → Do both together

Now we'll learn:

> **`inverse_transform()` converts the encoded numbers back to the original categories.**

---

# 1. What is `inverse_transform()`?

`inverse_transform()` does the opposite of `transform()`.

In simple words:

> **Numbers → Original Categories**

---

# 2. Syntax

```python
encoder.inverse_transform(encoded_data)
```

Example:

```python
encoder.inverse_transform([1, 2, 0])
```

---

# 3. How Does It Work?

Suppose we fitted the encoder on:

```python
levels = ["Low", "Medium", "High"]
```

The encoder learned:

| Category | Encoding |
| -------- | -------: |
| High     |        0 |
| Low      |        1 |
| Medium   |        2 |

So:

```text
High   → 0
Low    → 1
Medium → 2
```

Now `inverse_transform()` simply reverses this mapping:

```text
0 → High
1 → Low
2 → Medium
```

---

# 4. Step-by-Step Example

### Original Data

```text
Low
Medium
High
```

---

### After `transform()`

```text
1
2
0
```

---

### After `inverse_transform()`

```text
Low
Medium
High
```

We get back the original categories.

---

# 5. Python Example

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

levels = ["Low", "Medium", "High"]

encoder.fit(levels)

encoded = encoder.transform(levels)

print(encoded)
```

### Output

```python
[1 2 0]
```

---

Now decode them:

```python
decoded = encoder.inverse_transform(encoded)

print(decoded)
```

### Output

```python
['Low' 'Medium' 'High']
```

---

# 6. Output Explanation

The encoder had learned:

| Category | Encoding |
| -------- | -------: |
| High     |        0 |
| Low      |        1 |
| Medium   |        2 |

So:

```text
1 → Low
2 → Medium
0 → High
```

Therefore:

```python
[1 2 0]
```

became:

```python
['Low' 'Medium' 'High']
```

---

# 7. Real-World Analogy

Imagine a teacher's attendance register.

| Student | Roll Number |
| ------- | ----------: |
| Aditi   |           0 |
| Priya   |           1 |
| Rahul   |           2 |

Suppose someone gives you:

```text
2
0
1
```

You can look at the register and decode them:

```text
2 → Rahul
0 → Aditi
1 → Priya
```

This is exactly what `inverse_transform()` does.

---

# 8. Why Is `inverse_transform()` Useful?

Sometimes, after making predictions, you want the original labels back.

Example:

Suppose a model predicts:

```python
[2, 1, 0]
```

Instead of showing:

```text
2
1
0
```

you can convert them into:

```text
Medium
Low
High
```

which is easier to understand.

---

# 9. Can We Decode Numbers That Were Never Learned?

No.

Suppose the encoder learned:

| Category | Encoding |
| -------- | -------: |
| High     |        0 |
| Low      |        1 |
| Medium   |        2 |

Now you try:

```python
encoder.inverse_transform([3])
```

You'll get an error similar to:

```text
ValueError:
y contains previously unseen labels: [3]
```

because the encoder never learned what `3` means.

---

# 10. Visual Flow

```text
Original Categories
↓
Low
Medium
High

transform()
↓
Encoded Values

1
2
0

inverse_transform()
↓
Original Categories

Low
Medium
High
```

---

# Important Points to Remember

* `inverse_transform()` reverses Label Encoding.
* It converts numbers back to their original categories.
* It uses the mapping learned during `fit()`.
* It is the opposite of `transform()`.
* It cannot decode numbers that were never learned.

---

# Summary

> **`inverse_transform()` converts encoded numerical labels back into their original categorical values.**

Example:

Learned mapping:

```text
High   → 0
Low    → 1
Medium → 2
```

Then:

```python
encoder.inverse_transform([1, 2, 0])
```

Output:

```python
['Low', 'Medium', 'High']
```
