# Topic 6: Handling Unseen Categories

## Subtopic 6.1 – What Are Unseen Labels?

This is one of the most practical problems you'll face in real ML projects.

---

# What are unseen labels?

> **Unseen labels are categories that were NOT present during `fit()`, but appear later during `transform()`.**

In simple words:

> The encoder sees a category it has never seen before.

---

# Example

## Training Data

```python
train = ["Low", "Medium", "High"]
```

Fit the encoder:

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

encoder.fit(train)
```

The encoder learns:

| Category | Encoding |
| -------- | -------: |
| High     |        0 |
| Low      |        1 |
| Medium   |        2 |

---

## Test Data

Suppose later you get:

```python
test = ["Medium", "Very High"]
```

Notice:

```text
Very High
```

was never present in the training data.

This is an **unseen label**.

---

# Visual Understanding

### Seen Labels

Training data:

```text
Low
Medium
High
```

Encoder knows these.

---

### Unseen Label

Later:

```text
Very High
```

Encoder thinks:

> "I don't know what this is. Nobody taught me this category."

---

# Real-World Analogy

Imagine a teacher memorized these students:

```text
Aditi
Rahul
Priya
```

Now during attendance, someone says:

```text
Karan
```

The teacher says:

> "Who is Karan? He wasn't in my list."

Karan is an **unseen label**.

---

# Python Example

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

encoder.fit(["Low", "Medium", "High"])

print(encoder.classes_)
```

Output:

```python
['High' 'Low' 'Medium']
```

The encoder knows only these categories.

Now suppose:

```python
encoder.transform(["Medium"])
```

Output:

```python
[2]
```

Works perfectly.

---

But:

```python
encoder.transform(["Very High"])
```

will fail because:

```text
Very High
```

was never learned.

---

# Why does this happen?

Because:

```text
fit()
↓
learn categories
↓
transform()
↓
use only those learned categories
```

The encoder does **not learn new categories during `transform()`**.



# Topic 6: Handling Unseen Categories

## Subtopic 6.3 – Basic Ways to Handle Unseen Categories

Now we know:

* `fit()` learns categories.
* `transform()` uses those categories.
* Unseen categories cause errors.

The next question is:

> **What should we do when unseen categories appear?**

---

# Method 1: Replace with `"Unknown"` (Most Common)

Suppose training data is:

```text
Low
Medium
High
```

Before transforming test data, replace unseen values with `"Unknown"`.

## Example

Training categories:

```python
known_categories = ["Low", "Medium", "High"]
```

Test data:

```python
test = ["Medium", "Very High"]
```

Replace unseen values:

```python
processed_test = [
    value if value in known_categories else "Unknown"
    for value in test
]

print(processed_test)
```

### Output

```python
['Medium', 'Unknown']
```

Now you can decide how `"Unknown"` should be handled.

---

# Method 2: Add `"Unknown"` During Training

Training data:

```text
Low
Medium
High
Unknown
```

Fit the encoder:

```python
encoder.fit(["Low", "Medium", "High", "Unknown"])
```

Learned mapping:

| Category | Encoding |
| -------- | -------: |
| High     |        0 |
| Low      |        1 |
| Medium   |        2 |
| Unknown  |        3 |

Now if test data contains:

```text
Very High
```

replace it with:

```text
Unknown
```

and transform safely.

---

# Method 3: Use a Dictionary with `get()`

Instead of `LabelEncoder`, manual encoding can help.

Example:

```python
mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

print(mapping.get("Medium", -1))
print(mapping.get("Very High", -1))
```

### Output

```python
2
-1
```

Here:

```text
-1
```

means:

> Unknown category.

---

# Method 4: Use Encoders Designed for Unknown Categories

`LabelEncoder` is mainly intended for target labels and doesn't handle unknowns well.

For feature columns, other encoders are often better.

Examples:

* `OneHotEncoder(handle_unknown="ignore")`
* `OrdinalEncoder(handle_unknown="use_encoded_value")`

You'll study these later.

For now, just remember:

> **`LabelEncoder` has no built-in support for unseen categories.**

---

# Which Method Should You Remember?

As a beginner:

### Remember this:

```text
Training:
Fit on known categories.

Testing:
Replace unseen categories with "Unknown" (or another placeholder).
```

This is enough for most learning purposes.

---

# Example

Training:

```text
Low
Medium
High
Unknown
```

Mapping:

```text
High     → 0
Low      → 1
Medium   → 2
Unknown  → 3
```

Test:

```text
Medium
Very High
```

Replace:

```text
Medium
Unknown
```

Transform:

```text
2
3
```

No error occurs.

---

# Important Points

* Unseen categories are common in real-world data.
* `LabelEncoder` cannot handle them automatically.
* A common solution is to introduce an `"Unknown"` category.
* Manual mappings can use a default value like `-1`.
* Specialized encoders provide built-in support.
