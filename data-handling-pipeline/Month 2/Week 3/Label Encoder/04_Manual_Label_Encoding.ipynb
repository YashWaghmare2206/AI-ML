# Topic 4: Manual Label Encoding

## Subtopic 4.1 – Encoding Using Dictionaries

---

## What is it?

Instead of using Scikit-Learn's `LabelEncoder`, you manually define the mapping yourself using a Python dictionary.

Example:

```python
{
    "Low": 0,
    "Medium": 1,
    "High": 2
}
```

---

## Why use it?

Because sometimes **you want full control over the encoding**.

Remember:

`LabelEncoder` assigns values alphabetically.

For example:

```text
High   → 0
Low    → 1
Medium → 2
```

But what if you want:

```text
Low    → 0
Medium → 1
High   → 2
```

You can manually define it.

---

## Example

Suppose we have:

| Satisfaction |
| ------------ |
| Low          |
| Medium       |
| High         |
| Medium       |

---

### Step 1: Create a dictionary

```python
mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}
```

---

### Step 2: Use the dictionary

```python
data = ["Low", "Medium", "High", "Medium"]

encoded = [mapping[item] for item in data]

print(encoded)
```

### Output

```python
[0, 1, 2, 1]
```

---

## Explanation

The dictionary works like this:

```text
Low     → 0
Medium  → 1
High    → 2
Medium  → 1
```

Result:

```text
[0, 1, 2, 1]
```

---

## When should you use dictionary encoding?

### Use it when:

✅ You know the exact order.

Examples:

```text
Low → Medium → High
```

```text
Beginner → Intermediate → Advanced
```

```text
Poor → Average → Good → Excellent
```

---

### Don't use it when:

❌ The categories don't have an order.

Example:

```text
Red, Blue, Green
```

because you'll again introduce artificial ordering.

---

## Dictionary Encoding vs LabelEncoder

| Feature                 | Dictionary | LabelEncoder                 |
| ----------------------- | ---------- | ---------------------------- |
| Automatic               | ❌ No       | ✅ Yes                        |
| You control the mapping | ✅ Yes      | ❌ No                         |
| Alphabetical encoding   | ❌ No       | ✅ Yes                        |
| Good for ordinal data   | ✅ Yes      | ⚠️ Yes, but order may differ |


# Topic 4: Manual Label Encoding

## Subtopic 4.2 – Encoding Using Pandas `map()`

---

## What is `map()`?

`map()` is a Pandas method used to replace values in a Series (column) using a dictionary.

In simple words:

> **`map()` takes your manual mapping and applies it to a DataFrame column.**

---

## Syntax

```python
df["column_name"] = df["column_name"].map(mapping_dictionary)
```

---

## Example

Suppose we have this DataFrame:

```python
import pandas as pd

df = pd.DataFrame({
    "Satisfaction": ["Low", "Medium", "High", "Medium"]
})

print(df)
```

### Output

```text
  Satisfaction
0          Low
1       Medium
2         High
3       Medium
```

---

## Step 1: Create the mapping

```python
mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}
```

---

## Step 2: Apply `map()`

```python
df["Satisfaction"] = df["Satisfaction"].map(mapping)

print(df)
```

### Output

```text
   Satisfaction
0             0
1             1
2             2
3             1
```

---

## What happens if a value is not in the dictionary?

Example:

```python
df = pd.DataFrame({
    "Satisfaction": ["Low", "Medium", "Very High"]
})

mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

print(df["Satisfaction"].map(mapping))
```

### Output

```text
0    0.0
1    1.0
2    NaN
Name: Satisfaction, dtype: float64
```

---

## Why `NaN`?

Because:

```text
Very High
```

doesn't exist in the dictionary.

Pandas doesn't know what number to assign, so it returns:

```text
NaN
```

which means **missing value**.

---


## Comparison

| Feature                    | `map()`       | `LabelEncoder`       |
| -------------------------- | ------------- | -------------------- |
| Manual mapping             | ✅ Yes         | ❌ No                 |
| Automatic mapping          | ❌ No          | ✅ Yes                |
| Unseen values              | Returns `NaN` | Raises an error      |
| Works on DataFrame columns | ✅ Yes         | ⚠️ Needs extra steps |
