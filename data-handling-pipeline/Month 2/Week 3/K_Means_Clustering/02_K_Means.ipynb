# What is K-Means?

## Definition

> **K-Means is an unsupervised learning algorithm that partitions data into K clusters by grouping similar data points together and representing each cluster by its centroid.**

Let's break this down.

---

# Breaking the Definition

## "Unsupervised Learning Algorithm"

This means:

```text
Input Features (X)
Only
```

There are no labels.

Example:

| Customer | Income | Spending Score |
| -------- | ------ | -------------- |
| A        | 30K    | 20             |
| B        | 35K    | 25             |
| C        | 150K   | 85             |

No one tells us:

```text
Premium Customer
Budget Customer
```

K-Means must discover the groups itself.

---

## "Partitions Data"

Partition means:

> Divide the dataset into groups.

Example:

Suppose we have 9 customers and choose:

```text
K = 3
```

K-Means will divide them into:

```text
Cluster 1
Cluster 2
Cluster 3
```

Every data point must belong to exactly one cluster.

---

# "Groups Similar Data Points Together"

This is the heart of K-Means.

It asks:

```text
Which points are close to each other?
```

Points that are close:

```text
→ Same cluster
```

Points that are far apart:

```text
→ Different clusters
```

---

# "Represented by its Centroid"

Each cluster has a center point called a:

> **Centroid**

Example:

Suppose this cluster exists:

```text
● ● ●
● X ●
● ● ●
```

The:

```text
X
```

represents the cluster center.

That center is called:

> **Centroid**

K-Means uses these centroids to form clusters.

We'll study centroids deeply in upcoming subtopics.

---

# Simple Definition

If the formal definition feels complicated, remember this:

> **K-Means is an algorithm that automatically groups similar data points into K clusters using cluster centers called centroids.**

---

# Why Is It Called "K-Means"?

The name itself explains the algorithm.

---

## K

Means:

> The number of clusters we want.

Example:

```text
K = 2
```

means:

```text
2 clusters
```

---

## Means

Means:

> Arithmetic averages.

The centroid of a cluster is calculated using the mean of the points inside it.

Example:

Suppose a cluster contains:

```text
2, 4, 6
```

Its centroid is:

```text
(2 + 4 + 6) / 3 = 4
```

That is the "Mean" part.

---

# Example

Suppose we have these points:

| Point | Value |
| ----- | ----- |
| A     | 2     |
| B     | 3     |
| C     | 4     |
| D     | 10    |
| E     | 11    |
| F     | 12    |

Suppose:

```text
K = 2
```

K-Means may discover:

## Cluster 1

```text
2, 3, 4
```

Centroid:

```text
(2+3+4)/3 = 3
```

---

## Cluster 2

```text
10, 11, 12
```

Centroid:

```text
(10+11+12)/3 = 11
```

Notice:

> Similar points ended up together.

---

# Visual Intuition

Imagine points on a line:

```text
2  3  4          10 11 12
```

Your brain naturally says:

```text
These belong together.

Those belong together.
```

K-Means tries to do exactly that mathematically.

---

# What Is the Goal of K-Means?

The objective is:

> **Create clusters where points are close to their own centroid and far from other centroids.**

In simple words:

```text
Small distance inside clusters.

Large distance between clusters.
```

---

# Important Characteristics

K-Means:

* Is an unsupervised learning algorithm.
* Requires specifying K beforehand.
* Groups similar observations.
* Uses centroids (means).
* Assigns each point to exactly one cluster.
* Primarily relies on Euclidean distance.

---

# Interview Answer

If asked:

> "What is K-Means?"

You can answer:

> K-Means is an unsupervised clustering algorithm that partitions a dataset into K clusters by assigning data points to the nearest centroid and updating those centroids iteratively until convergence. The goal is to minimize the distance between points and their respective cluster centroids.

---

# Important Clarification

Many beginners think:

> "K-Means understands what the clusters mean."

It doesn't.

K-Means only says:

```text
These points are similar.
```

Humans later interpret:

```text
Cluster 1 = Premium Customers

Cluster 2 = Budget Customers
```

The names come from us.

---

# Mini Summary

```text
K-Means:
• Unsupervised Learning Algorithm
• Divides data into K clusters
• Groups similar points together
• Uses centroids (means) as cluster centers
• Requires K beforehand

Goal:
Keep points close to their own centroid
and far from other centroids.
```

> **K is the number of clusters that we want the K-Means algorithm to create.**

> **No.**

This is one of the biggest limitations of K-Means.

K-Means asks you:

```text
How many clusters do you want me to create?
```

You must provide:

```text
K = ?
```

before the algorithm starts.

Many beginners think:

> "K means the number of natural clusters in the dataset."

> **The number of clusters we ASK the algorithm to create.**

## Subtopic 3: What is a Cluster?

So far, you know:

```text
K-Means
↓
Creates K clusters
↓
Groups similar points together
```

> **What exactly is a cluster?**

Is it just a random group?

# Definition of a Cluster

> **A cluster is a collection of data points that are more similar to each other than to points in other groups.**

In simple words:

> A cluster is a group of "similar" observations.

---

# The Main Idea

Suppose we have these points:

```text
2   3   4       10   11   12
```

Your brain naturally thinks:

```text
2, 3, 4 belong together.

10, 11, 12 belong together.
```

So you mentally create:

```text
Cluster 1 = {2, 3, 4}

Cluster 2 = {10, 11, 12}
```

These are clusters.

---

# Important Point

A cluster is NOT:

```text
A label,
A category name,
A prediction.
```

It is simply:

> A set of observations that look similar according to some measure.

For K-Means, that measure is mainly:

> **Distance.**

Closer points:

```text
→ Same cluster
```

Farther points:

```text
→ Different clusters
```

# Visual Intuition

Imagine points on a graph.

```text
● ● ●

           ● ● ●

                     ● ● ●
```

Even without mathematics, you immediately see:

```text
3 groups.
```

Each group is a:

> **Cluster.**

---

# What Makes a Good Cluster?

A good cluster has two properties.

---

## 1. High Similarity Within the Cluster

Points inside the same cluster should be close.

Example:

```text
●●●●●
```

This is good.

The points resemble each other.

---

## 2. Low Similarity Between Clusters

Different clusters should be well separated.

Example:

```text
●●●●●          ▲▲▲▲▲
```

This is also good.

The groups are distinct.

---

# Bad Clustering

Suppose we get:

```text
● ▲ ● ▲
▲ ● ▲ ●
● ▲ ● ▲
```

Everything overlaps.

Questions arise:

```text
Which group does this point belong to?

Are these really separate groups?
```

This is poor clustering.

---

# Cluster ≠ Label

This is one of the biggest misconceptions.

Many beginners think:

```text
Cluster 0
↓
Must mean something.
```

Not necessarily.

K-Means only says:

```text
These points belong together.
```

The numbers:

```text
Cluster 0
Cluster 1
Cluster 2
```

are just identifiers.

They do NOT mean:

```text
Cluster 2 > Cluster 1

Cluster 0 = Bad Customers

Cluster 1 = Good Customers
```

Humans decide what the clusters represent.

---

# Cluster vs Class

This is another common confusion.

## Class

Known beforehand.

Example:

```text
Spam
Not Spam
```

---

## Cluster

Discovered from the data.

Example:

```text
Group A
Group B
Group C
```

whose meaning is determined later.

---
# Mini Summary

```text
Cluster:
A group of similar data points.

Good Cluster:
• High similarity within the cluster.
• Low similarity between clusters.

Cluster ≠ Label.
Cluster numbers are just names.

K-Means:
Finds which points belong together.
Humans decide what those groups mean.
```

# Topic 2: What is K-Means?

## Subtopic 4: What is a Centroid?

This is one of the **most important concepts in K-Means**.

If you understand centroids properly, the entire K-Means algorithm becomes much easier.

---

# What is a Centroid?

## Definition

> A **centroid** is the center point of a cluster.

In simple words:

> It is the "representative point" of a cluster.

Think of it as:

```text
Cluster
↓
Its center
↓
Centroid
```

---

# Intuition

Suppose you have this cluster:

```text
● ● ●
 ● X ●
● ● ●
```

The point:

```text
X
```

represents the middle of the cluster.

That middle point is called:

> **The centroid.**

---

# Why Do We Need a Centroid?

K-Means needs a way to answer:

> "Which cluster should this point belong to?"

It does this by comparing:

```text
Distance from Point
↓
to each Centroid
```

The point is assigned to:

> The cluster whose centroid is closest.

So centroids act like:

> **Cluster leaders or reference points.**

---

# How Is a Centroid Calculated?

This is where the **"Means"** in K-Means comes from.

A centroid is calculated using the:

> **Arithmetic mean (average)** of all points in the cluster.

---

# Example 1 (1-D Data)

Suppose a cluster contains:

```text
2, 4, 6
```

The centroid is:

```text
(2 + 4 + 6) / 3
= 12 / 3
= 4
```

So:

```text
Cluster:
2, 4, 6

Centroid:
4
```

Notice:

> The centroid doesn't have to be one of the actual data points.

It just represents the center.

---

# Example 2 (2-D Data)

Suppose a cluster has these points:

| Point |  x |  y |
| ----- | -: | -: |
| A     |  2 |  4 |
| B     |  4 |  6 |
| C     |  6 |  8 |

---

## Step 1: Average the x-values

```text
(2 + 4 + 6) / 3 = 4
```

---

## Step 2: Average the y-values

```text
(4 + 6 + 8) / 3 = 6
```

---

## Centroid

```text
(4, 6)
```

So:

> The centroid is found by taking the mean of each feature separately.

---

# Visual Intuition

Imagine these points:

```text
(2,4)

      (4,6)

             (6,8)
```

The center lies around:

```text
(4,6)
```

That becomes the centroid.

---

# Important Observation

Suppose the points are:

```text
2, 4, 100
```

The centroid becomes:

```text
(2 + 4 + 100) / 3
= 106 / 3
≈ 35.33
```

Notice:

> 35.33 is NOT an actual data point.

This tells us:

### Centroids can exist where no real observation exists.

They are mathematical centers.

---

# Why Are Centroids Important in K-Means?

K-Means repeatedly does two things:

## Step 1

```text
Assign points
↓
to nearest centroid
```

---

## Step 2

```text
Recalculate centroids
↓
using the mean of assigned points
```

Then repeat.

So:

> Centroids keep moving until the clusters stabilize.

We'll study this entire process in Topic 3.

---

# Real-Life Analogy

Imagine a group of friends deciding where to meet.

Their homes are located at:

```text
2 km
4 km
6 km
```

The average location is:

```text
4 km
```

So they choose:

> 4 km as the meeting point.

That meeting point is like the centroid.

---

# Common Misconception

Many beginners think:

> "The centroid must be one of the actual data points."

❌ Wrong.

Example:

```text
Points:
1, 2
```

Centroid:

```text
(1 + 2) / 2 = 1.5
```

There is no data point at 1.5.

Yet:

> 1.5 is still the centroid.

---



# Important Points to Remember

* A centroid is the center of a cluster.
* It represents the cluster.
* It is calculated using the arithmetic mean.
* It does not have to be an actual data point.
* K-Means uses centroids to assign points to clusters.
* Centroids are updated repeatedly during the algorithm.

---

# Mini Summary

```text
Centroid:
Center of a cluster.

How to calculate:
Take the mean of each feature.

Example:
Points: 2, 4, 6
Centroid: 4

Important:
• May not be a real data point.
• Used to assign points to clusters.
• Updated repeatedly in K-Means.
```

# Topic 2: What is K-Means?

## Subtopic 5: Goal of K-Means

Now you understand:

```text id="l6x8av"
K → Number of clusters

Cluster → Group of similar points

Centroid → Center of a cluster
```

The final question before learning the actual algorithm is:

> **What is K-Means trying to achieve?**

---

# Goal of K-Means

> **The goal of K-Means is to create K clusters such that points are as close as possible to their own centroid and as far as possible from other centroids.**

In simple words:

> **Make points within a cluster very similar and different clusters well separated.**

---

# Think of It This Way

Suppose we have these points:

```text id="tcrh3h"
2   3   4        10   11   12
```

A good clustering would be:

```text id="jlwmxr"
Cluster 1:
2, 3, 4

Centroid = 3

Cluster 2:
10, 11, 12

Centroid = 11
```

Why?

Because:

```text id="ttmxm6"
2, 3, 4 are close to 3

10, 11, 12 are close to 11
```

---

# What K-Means Wants to Avoid

Suppose K-Means formed:

```text id="2g6gc5"
Cluster 1:
2, 12

Centroid = 7

Cluster 2:
3, 4, 10, 11

Centroid = 7
```

Now notice:

```text id="ht2eh6"
2 is far from 7

12 is far from 7

3 and 11 are also far apart
```

These clusters don't make much sense.

So:

> K-Means tries to avoid clusters where points are far from their centroid.

---

# The Real Objective

K-Means tries to minimize something called:

> **Within-Cluster Sum of Squares (WCSS)**

You'll study this deeply later.

For now, think of it as:

> **The total distance of all points from their own centroid.**

---

# Visual Intuition

## Good Clustering

```text id="yrwzrq"
● ● ●     X


            ▲ ▲ ▲      Y
```

Where:

```text id="xgvxwe"
X = Centroid of Cluster 1
Y = Centroid of Cluster 2
```

Points are close to their own centroid.

---

## Bad Clustering

```text id="sg2yru"
●          X          ●

▲          Y          ▲
```

Points are spread far from their centroid.

This increases the total distance.

K-Means doesn't like this.

---

# The Goal in One Sentence

Imagine K-Means saying:

> **"I want every point to be as close as possible to the center of its own cluster."**

That's literally what it's optimizing.

---

---

# Why Is This Goal Important?

Because if points within a cluster are close:

```text id="6dcf1j"
The cluster becomes meaningful.
```

It becomes easier to interpret:

```text id="vr2vzf"
Premium Customers

Budget Customers

Frequent Buyers
```

instead of random mixtures.

---

# Mathematical Version (Don't Memorize Yet)

K-Means minimizes:

```text id="yqv0s7"
Total distance² between points and their centroids.
```

We'll learn the exact formula later when studying WCSS.

For now:

> **Focus only on the intuition.**

---

# Important Points to Remember

* K-Means wants compact clusters.
* Points should be close to their own centroid.
* Different clusters should be well separated.
* It minimizes the total distance (specifically squared distance) within clusters.
* This objective leads to meaningful groupings.

---

# Mini Summary

```text id="7zcaz5"
Goal of K-Means:

Create K clusters where:

✓ Points are close to their own centroid.
✓ Points in different clusters are separated.
✓ Within-cluster distance is minimized.

K-Means tries to make:
"Similar points stay together."
```
