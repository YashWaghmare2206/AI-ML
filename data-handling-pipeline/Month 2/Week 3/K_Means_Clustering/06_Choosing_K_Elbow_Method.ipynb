# Topic 6: Choosing K (Elbow Method)

So far, we've been doing:

```text id="jlwm4m"
Choose K
↓
Run K-Means
```

But an important question is:

```text id="z9nzzk"
How do we know what K should be?
```

Why K = 2?

Why not 3, 5, or 10?

---

## The Problem

K-Means does **not** automatically choose K.

You have to provide it:

```text id="r2ywpj"
K = ?
```

Choosing the wrong K can lead to poor clustering.

---

# Intuition

Suppose you have customer data.

If you choose:

```text id="j3bjk2"
K = 1
```

everything goes into one cluster.

Too simple.

---

If you choose:

```text id="hbmjlwm"
K = Number of data points
```

each point gets its own cluster.

Too complex.

---

So we want:

```text id="l58d5o"
Not too few clusters
Not too many clusters
```

---

# What happens as K increases?

Suppose:

```text id="wvx42r"
K = 1
```

One centroid serves everyone.

Points are far from the centroid.

---

```text id="0uvc5d"
K = 2
```

Two centroids.

Distances decrease.

---

```text id="bgjlwm3"
K = 3
```

Distances decrease further.

---

In general:

```text id="v4cx77"
As K increases,
clusters become smaller,
and points get closer to centroids.
```

---

# The Quantity We Measure

We measure:

```text id="v0ohrk"
WCSS (Within-Cluster Sum of Squares)
```

Also called:

```text id="c5f3j0"
Inertia
```

For now, think of it as:

```text id="c1w6qn"
Total clustering error.
```

Lower WCSS means:

```text id="sxg8a6"
Points are closer to their centroids.
```

---

# What happens to WCSS?

As K increases:

```text id="xpl4ga"
K ↑
↓
WCSS ↓
```

Because more centroids fit the data better.

Example:

| K | WCSS |
| - | ---: |
| 1 | 1000 |
| 2 |  600 |
| 3 |  350 |
| 4 |  250 |
| 5 |  220 |
| 6 |  205 |

Notice:

```text id="4kgnhj"
Initially:
Large improvements.

Later:
Very small improvements.
```

---

# The Elbow Method

We plot:

```text id="vtrh7d"
X-axis → K
Y-axis → WCSS
```

Suppose the graph looks like:

```text id="9e4e8v"
```


There is a bend.

Like an elbow.

---

# What is the Elbow?

The elbow is:

```text id="i4l1ph"
The point after which
adding more clusters
does not significantly reduce WCSS.
```

In other words:

Before elbow:

```text id="exm2vv"
Big improvement.
```

After elbow:

```text id="4bljze"
Tiny improvement.
```

---

# Choosing K

We choose:

```text id="yh2f7d"
K = Elbow point
```

because it balances:

```text id="4syt6j"
Simple model
+
Good clustering
```

---

# Example

Suppose:

| K | WCSS |
| - | ---: |
| 1 | 1000 |
| 2 |  500 |
| 3 |  250 |
| 4 |  200 |
| 5 |  180 |
| 6 |  170 |

Improvements:

```text id="70qqwq"
1→2 : 500 decrease
2→3 : 250 decrease
3→4 : 50 decrease
4→5 : 20 decrease
5→6 : 10 decrease
```

The biggest slowing down happens around:

```text id="lfjlwm"
K = 3
```

So we'd choose:

```text id="xjlwm4"
K = 3
```

---

# Is the elbow always obvious?

No.

Sometimes:

```text id="zjlwm8"
The bend is clear.
```

Sometimes:

```text id="jlwm88"
The curve is smooth,
and there is no obvious elbow.
```

In such cases:

```text id="jlwm99"
Elbow Method becomes subjective.
```

That's why another method exists:

```text id="iuyt55"
Silhouette Score
```

which we'll learn later.

---

# Mini Summary

```text id="mm34vf"
K-Means cannot choose K automatically.

As K increases:
WCSS decreases.

Elbow Method:
Plot K vs WCSS.

Choose the point where:
Large improvements stop
and small improvements begin.

That point is called the elbow.
```

---

## One-line intuition

```text id="y6cvk9"
Choose K where adding another cluster stops being "worth it."
```
