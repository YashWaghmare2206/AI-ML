# Topic 12: Limitations of K-Means ⭐⭐⭐⭐⭐

You've already learned the assumptions.

Now, limitations are simply:

```text
"What practical problems does K-Means have because of those assumptions?"
```

Some limitations come directly from the assumptions, while others are new.

---

# 1. Sensitive to Initialization ⭐⭐⭐⭐⭐

## Problem

K-Means starts with:

```text
Random initial centroids.
```

Different starting points can lead to different results.

---

### Example

Suppose the true clusters are:

```text
● ● ● ●        ○ ○ ○ ○
● ● ● ●        ○ ○ ○ ○
```

---

### Good Initialization

```text
×                  ×
```

Result:

```text
Cluster 1 → ●
Cluster 2 → ○
```

Correct.

---

### Bad Initialization

```text
× ×
```

Both centroids start here:

```text
● ● ● ●        ○ ○ ○ ○
× ×
```

K-Means may converge to a poorer solution.

---

## Why?

Because K-Means minimizes WCSS locally.

It does NOT guarantee:

```text
Lowest possible WCSS globally.
```

It can get stuck in:

```text
Local minima.
```

---

# 2. Need to Specify K ⭐⭐⭐⭐⭐

## Problem

K-Means asks:

```text
How many clusters?
```

before starting.

---

### Example

Hospital patients:

```text
How many groups exist?

2?
4?
7?
```

Nobody knows.

---

## Why?

Because K-Means assumes:

```text
K is already known.
```

It never discovers K automatically.

---

# 3. Sensitive to Outliers ⭐⭐⭐⭐

## Problem

A few extreme points can affect centroids.

---

### Example

Most salaries:

```text
30k
35k
40k
45k
```

One outlier:

```text
10,000,000
```

---

## Why?

Because centroid =

```text
Mean of points.
```

Means are sensitive to extreme values.

The centroid gets pulled toward the outlier.

Example:

```text
2, 4, 6
Centroid = 4

2, 4, 100
Centroid = 35.33
```

---

# 4. Poor Performance on Non-Spherical Clusters ⭐⭐⭐⭐⭐

## Problem

True clusters may be curved.

---

### Example

```text
((((((

)))))))
```

---

## Why?

Because K-Means assumes:

```text
One centroid
↓
One circular region.
```

Curves cannot be represented by one centroid.

---

# 5. Struggles with Different Sizes ⭐⭐⭐⭐

## Problem

Small clusters near large clusters may disappear.

---

### Example

```text
1000 Regular customers

20 VIP customers
```

---

## Why?

Because:

```text
Large clusters contribute more to WCSS.
```

K-Means tries to reduce total error.

It may:

```text
Split large cluster
instead of
preserving small cluster.
```

---

# 6. Struggles with Different Densities ⭐⭐⭐⭐

## Problem

Sparse clusters can be split.

Dense clusters can dominate.

---

### Example

```text
Dense:

●●●●●●●

Sparse:

○      ○

    ○
```

---

## Why?

Because sparse clusters have:

```text
Large distances to centroids
↓
High WCSS.
```

K-Means tries to reduce this.

---

# 7. Works Only with Numerical Data ⭐⭐⭐

## Problem

K-Means uses:

```text
Means.
```

---

### Example

Can we average?

```text
Red
Blue
Green
```

No.

Can we average?

```text
Dog
Cat
Bird
```

No.

---

## Why?

Because centroids require:

```text
Arithmetic mean.
```

which only makes sense for numerical features.

---

# Mini Summary

| Limitation                  | Why?                                   |
| --------------------------- | -------------------------------------- |
| Sensitive to initialization | Random starts can lead to local minima |
| Need K beforehand           | K-Means cannot discover K              |
| Sensitive to outliers       | Means get pulled by extreme values     |
| Non-spherical clusters      | One centroid creates circular regions  |
| Different sizes             | Large clusters dominate WCSS           |
| Different densities         | Sparse clusters have high WCSS         |
| Numerical data only         | Means require numeric values           |

---

## One-line intuition

```text
K-Means is simple and fast,
but it sees the world through three lenses:

• Centroids (means)
• Euclidean distance
• WCSS minimization

Whenever reality doesn't fit these lenses,
K-Means starts making mistakes.
```

## **Local Minimum V/S Global Minimum**

# Local Minimum vs Global Minimum (Short)

Suppose K-Means can produce these WCSS values:

| Clustering Solution | WCSS |
| ------------------- | ---: |
| A                   |  120 |
| B                   |   90 |
| C                   |   70 |
| D                   |   50 |

---

## Global Minimum

```text
The absolute smallest WCSS possible.
```

In this example:

```text
WCSS = 50
```

because:

```text
50 is the best among ALL possible solutions.
```

---

## Local Minimum

```text
A solution that looks best nearby,
but is not the best overall.
```

Suppose K-Means starts randomly and reaches:

```text
WCSS = 70
```

Around 70, small changes give:

```text
72
71
70
71
73
```

So K-Means thinks:

```text
"I can't improve anymore."
```

and stops.

But elsewhere:

```text
WCSS = 50
```

still exists.

Therefore:

```text
70 = Local Minimum
50 = Global Minimum
```

---

# Mini Summary

```text
Global Minimum:
Best WCSS among ALL possible clusterings.

Local Minimum:
Best WCSS in the nearby region,
but not the best overall.
```

## One-line intuition

```text
Local minimum = "best I found from where I started."

Global minimum = "best that actually exists."
```
