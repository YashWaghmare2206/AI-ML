# Topic 11: Assumptions of K-Means ⭐⭐⭐⭐

K-Means works well in many situations.

But it secretly assumes certain things about the data.

If these assumptions are violated:

```text
K-Means may produce poor clusters.
```

---

# Assumption 1: Clusters are roughly spherical ⭐⭐⭐⭐⭐

K-Means assumes clusters look like:

```text
   ● ● ●
 ● ● × ● ●
   ● ● ●
```

where:

```text
× = centroid
```

Meaning:

```text
Points are distributed around the centroid
in all directions.
```

---

## Why?

Because K-Means uses:

```text
Euclidean distance.
```

Euclidean distance naturally creates:

```text
Circular (2D)
Spherical (3D)
Ball-shaped (higher dimensions)
```

regions around centroids.

---

## Works Well

```text
 ●●●     ○○○
●×●●     ○○×○
 ●●●     ○○○
```

---

## Works Poorly

Suppose the true clusters are:

```text
((((((

)))))))
```

Long curved shapes.

K-Means tries to split them into circles.

Result:

```text
Wrong clusters.
```

---

# Assumption 2: Clusters have similar sizes

K-Means prefers:

```text
Cluster A: 100 points
Cluster B: 120 points
Cluster C: 90 points
```

---

## Problem

Suppose:

```text
Cluster A: 1000 points
Cluster B: 20 points
```

The large cluster can dominate.

The small cluster may get:

```text
Merged
or
Broken incorrectly.
```

---

# Assumption 3: Clusters have similar densities

K-Means assumes:

```text
Points are similarly packed.
```

Example:

Works well:

```text
●●●      ○○○
●●●      ○○○
```

---

Poor case:

```text
Dense cluster:

●●●●●●●

Sparse cluster:

○      ○

     ○
```

Different densities can confuse K-Means.

---

# Assumption 4: Distance means similarity ⭐⭐⭐⭐⭐

K-Means assumes:

```text
Close points
↓
Similar

Far points
↓
Different
```

---

## Example

For customer segmentation:

Using:

```text
Age
Income
```

this assumption often makes sense.

---

## But sometimes it doesn't

Suppose two customers:

```text
Same salary
Different hobbies
Different purchasing behavior
```

Numerically close:

```text
Yes
```

Actually similar:

```text
Maybe not.
```

So feature choice matters.

---

# Assumption 5: Features are properly scaled ⭐⭐⭐⭐⭐

Because K-Means uses:

```text
Euclidean distance.
```

It assumes:

```text
Features contribute fairly.
```

Without scaling:

```text
Salary dominates Age.
```

We already learned:

```text
Wrong distances
↓
Wrong clusters
```

---

# Assumption 6: K is known beforehand

K-Means assumes:

```text
You already know K.
```

But in reality:

```text
You often don't.
```

That's why we use:

```text
Elbow Method
Silhouette Score
```

to estimate K.

---

# Summary of Assumptions

| Assumption            | Meaning                                |
| --------------------- | -------------------------------------- |
| Spherical clusters    | Clusters are roughly ball-shaped       |
| Similar sizes         | No extremely tiny vs huge clusters     |
| Similar densities     | Clusters are similarly packed          |
| Distance = similarity | Euclidean distance reflects similarity |
| Proper scaling        | Features contribute fairly             |
| K known               | Number of clusters must be chosen      |

---

# What if assumptions fail?

K-Means can still run.

But:

```text
Good assumptions
↓
Good clusters

Broken assumptions
↓
Potentially misleading clusters
```

---

# Mini Summary

```text
K-Means works best when:

✓ Clusters are spherical.
✓ Cluster sizes are similar.
✓ Cluster densities are similar.
✓ Euclidean distance reflects similarity.
✓ Features are scaled.
✓ K is chosen appropriately.

These are not strict rules,
but violations can reduce clustering quality.
```

---

## One-line intuition

```text
K-Means is powerful, but it sees the world as groups of similarly sized balls separated by Euclidean distance.
```

## **Example of Each Assumption and Reason why would it Fail**

Perfect. This is actually the **most important part** of assumptions: not just knowing them, but understanding **why K-Means fails** when they are violated.

---

# Assumption 1: Clusters are roughly spherical

### ✓ Works

Young vs Old customers:

```text
● ● ●          ○ ○ ○
● ● ●          ○ ○ ○
```

---

### ✗ Fails

Cars moving on two curved roads:

```text
((((((

)))))))
```

---

### Why does K-Means fail?

Because K-Means thinks:

```text
Cluster
↓
One centroid
↓
Points should be close to that centroid
↓
Circle (sphere) around centroid
```

But curved data cannot be captured by a single circle.

So K-Means does:

```text
Curve
↓
Cuts it into circular pieces
↓
Wrong clusters
```

---

# Assumption 2: Similar cluster sizes

(Size = number of points)

### ✓ Works

```text
100 customers
120 customers
90 customers
```

---

### ✗ Fails

```text
1000 Regular customers
20 VIP customers
```

---

### Why does K-Means fail?

Because K-Means minimizes:

```text
Total WCSS
```

The large cluster contributes much more to WCSS.

K-Means thinks:

```text
"If I reduce error for 1000 points,
I can reduce WCSS a lot."

instead of

"If I protect these 20 VIP points."
```

So it may:

```text
Split the large cluster
↓
Ignore or absorb the small cluster
```

---

# Assumption 3: Similar densities

(Density = how closely packed points are)

### ✓ Works

```text
● ● ●        ○ ○ ○
● ● ●        ○ ○ ○
```

---

### ✗ Fails

```text
Dense:

●●●●●●●

Sparse:

○      ○

   ○
```

---

### Why does K-Means fail?

Because K-Means assumes:

```text
One centroid should represent a cluster.
```

Sparse clusters have:

```text
Large distances to the centroid.
```

Therefore:

```text
Large WCSS.
```

K-Means thinks:

```text
"This sparse cluster has too much error."

Maybe I should:
↓
Split the sparse cluster

or

Merge it with another cluster.
```

because that reduces total WCSS.

---

# Assumption 4: Distance reflects similarity

### ✓ Works

Customer segmentation using:

```text
Age
Income
```

---

### ✗ Fails

Movie recommendation using only:

```text
Age
Salary
```

---

### Why does K-Means fail?

Because K-Means only sees:

```text
Numbers.
```

It assumes:

```text
Close in feature space
↓
Actually similar.
```

But if important features are missing:

```text
Close numerically
≠
Close behaviorally.
```

Example:

```text
Age = 30
Salary = 50k
```

One likes horror movies.

One likes romantic movies.

K-Means says:

```text
Same cluster.
```

Humans say:

```text
Very different.
```

---

# Assumption 5: Features are properly scaled

### ✓ Works

```text
Age
Salary
↓
Standardize
↓
Fair contribution
```

---

### ✗ Fails

```text
Age: 20–60
Salary: 20,000–5,000,000
```

---

### Why does K-Means fail?

Because Euclidean distance becomes:

```text
Age contribution:
10² = 100

Salary contribution:
470000² = 220,900,000,000
```

K-Means effectively sees:

```text
Salary only.
```

So:

```text
Distances become distorted
↓
Wrong assignments
↓
Wrong clusters
```

---

# Assumption 6: K is known

### ✓ Works

Company wants:

```text
Budget
Regular
Premium
```

K = 3.

---

### ✗ Fails

Hospital data:

```text
How many patient groups exist?
```

Nobody knows.

---

### Why does K-Means fail?

Because K-Means never asks:

```text
"How many clusters naturally exist?"
```

It blindly trusts:

```text
K given by the user.
```

If you choose:

```text
K too small
↓
Different groups get merged.
```

If you choose:

```text
K too large
↓
True groups get split.
```

The algorithm itself won't warn you.

---

# The Complete Picture

| Assumption Violated   | Why K-Means Fails                                                          |
| --------------------- | -------------------------------------------------------------------------- |
| Non-spherical shapes  | Uses circles around centroids; curves can't be represented by one centroid |
| Different sizes       | Large clusters dominate WCSS and sacrifice small clusters                  |
| Different densities   | Sparse clusters have high WCSS and get split/merged                        |
| Distance ≠ similarity | K-Means only sees numbers, not true meaning                                |
| No scaling            | Large-valued features dominate Euclidean distance                          |
| Wrong K               | K-Means blindly follows the chosen K                                       |
