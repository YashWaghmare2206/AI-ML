# Topic 10: Interpreting Clusters

Suppose you ran K-Means and got:

```python
labels = [0, 0, 1, 2, 1, 2]
```

The biggest question is:

```text
What do these clusters actually mean?
```

---

# Important Principle

K-Means only does:

```text
Group similar points together.
```

It does NOT know:

```text
Customer types
Rich/Poor
High-value customers
Student/Senior citizens
```

It only knows:

```text
Distance.
```

---

# Example

Suppose after clustering customers, you get:

| Age | Salary | Cluster |
| --- | -----: | ------: |

# Does Cluster 0 mean "Poor"?

```text
No.
```

Does Cluster 1 mean "Rich"?

```text
No.
```

K-Means never gives meanings.

# Why is Interpretation Important?

Because clustering itself isn't the final goal.

The goal is usually:

```text
Business decisions
Customer understanding
Personalized marketing
Resource allocation
Insights
```

K-Means gives:

```text
Groups
```

Humans provide:

```text
Meaning.
```

---

# Centroids Help Interpretation

Remember:

```python
kmeans.cluster_centers_
```

After converting back to original units:

```python
scaler.inverse_transform(...)
```

Suppose you get:

| Cluster | Age | Salary |
| ------- | --: | -----: |
| 0       |  26 | 32,500 |
| 1       |  61 | 92,500 |
| 2       |  41 | 52,500 |

These centroid values summarize:

```text
The "typical" member of each cluster.
```

---

# Are Cluster Numbers Meaningful?

Suppose today:

```text
Cluster 0 → Young professionals
Cluster 1 → Seniors
Cluster 2 → Middle-aged
```

Tomorrow, K-Means might return:

```text
Cluster 0 → Seniors
Cluster 1 → Middle-aged
Cluster 2 → Young professionals
```

The numbering can change.

Therefore:

```text
Cluster IDs have NO inherent meaning.
```

Only the characteristics matter.

---

# Practical Workflow

```text
Run K-Means
↓
Get cluster labels
↓
Compute statistics for each cluster
↓
Study the centroids
↓
Identify patterns
↓
Assign meaningful descriptions
```

---

# Mini Summary

```text
K-Means only groups similar points.

It does NOT know what the groups mean.

To interpret clusters:
• Analyze cluster statistics.
• Examine centroids.
• Look for common characteristics.
• Assign human-readable meanings.

Cluster numbers themselves have no meaning.
```

---

## One-line intuition

```text
K-Means discovers the groups.
Humans explain what those groups represent.
```
