# Topic 7: WCSS (Inertia)

In the Elbow Method, we kept saying:

```text
WCSS decreases
WCSS decreases
```

But what exactly is WCSS?

---

# What is WCSS?

WCSS stands for:

```text
Within-Cluster Sum of Squares
```

It is also called:

```text
Inertia
```

It measures:

```text
How tightly packed the clusters are.
```

or

```text
Total clustering error.
```

---

# What does it calculate?

For every point:

```text
Find distance to its own centroid
↓
Square that distance
↓
Add them up
```

---

## Formula

WCSS=\sum_{i=1}^{K}\sum_{x\in C_i}|x-\mu_i|^2

Don't memorize it yet.

Just remember:

```text
For each cluster:
    For each point:
        Distance to centroid
        ↓
        Square it
        ↓
        Add it
```

---

# Example

Suppose one cluster has:

```text
Points:     2, 4, 6
Centroid:   4
```

Distances from centroid:

```text
2 → |2−4| = 2
4 → |4−4| = 0
6 → |6−4| = 2
```

Square them:

```text
2² = 4
0² = 0
2² = 4
```

Add them:

```text
WCSS for this cluster

= 4 + 0 + 4
= 8
```

---

Suppose another cluster has:

```text
Points:    10, 11
Centroid:  10.5
```

Distances:

```text
|10−10.5| = 0.5
|11−10.5| = 0.5
```

Squares:

```text
0.25
0.25
```

Add:

```text
0.5
```

---

# Total WCSS

Add WCSS of all clusters:

```text
Cluster 1 → 8
Cluster 2 → 0.5
```

Total:

```text
WCSS = 8.5
```

---

# What does WCSS mean?

### Large WCSS

```text
Points are far from centroids.
Clusters are spread out.
Poor clustering.
```

Example:

```text
●      ●

         ×

●             ●
```

Points are far from the centroid.

---

### Small WCSS

```text
Points are close to centroids.
Clusters are compact.
Better clustering.
```

Example:

```text
 ● ●
● × ●
  ●
```

Points are tightly grouped around the centroid.

---

# Why do we square?

Same reason as variance.

If we simply added distances:

```text
Negative and positive directions don't matter,
but larger distances should be penalized more.
```

Squaring:

```text
• Makes everything positive
• Penalizes large distances more heavily
```

Example:

```text
Distance = 2  → 4
Distance = 4  → 16
```

Doubling distance causes:

```text
4× the penalty.
```

---

# Why does WCSS always decrease when K increases?

This is very important.

Suppose:

```text
K = 1
```

One centroid serves everyone.

Large distances.

Large WCSS.

---

Now:

```text
K = 2
```

Two centroids.

Points become closer.

WCSS decreases.

---

If:

```text
K = Number of data points
```

Each point becomes its own centroid.

Distance:

```text
0
```

Therefore:

```text
WCSS = 0
```

---

So:

```text
K ↑
↓
WCSS ↓
```

It can never increase.

---

# Why can't WCSS increase?

Because when you add another centroid:

```text
You are giving K-Means more flexibility.
```

At worst:

```text
The new centroid is ignored.
```

At best:

```text
It reduces distances.
```

Therefore:

```text
WCSS either decreases
or stays the same.
```

Never increases.

---

# Why don't we simply choose huge K?

Because:

```text
K = Number of points
↓
WCSS = 0
↓
Every point becomes its own cluster
↓
No useful grouping
```

This is why we use:

```text
Elbow Method
```

to balance:

```text
Low WCSS
+
Meaningful clusters
```

---

# Mini Summary

```text
WCSS (Inertia):

For each point:
Distance to centroid
↓
Square it
↓
Add them all

Small WCSS:
Compact clusters.

Large WCSS:
Spread-out clusters.

As K increases:
WCSS always decreases.

K = Number of points:
WCSS = 0.
```

---

## One-line intuition

```text
WCSS tells us how much "scatter" remains inside the clusters.
Lower WCSS means tighter clusters.
```
