# Topic 3: How K-Means Works

## Subtopic 1: Choose K

We are now entering the **actual algorithm**.

The entire K-Means algorithm can be summarized as:

```text
Choose K
↓
Initialize K centroids
↓
Assign points to nearest centroid
↓
Update centroids
↓
Repeat until convergence
```

Today, we'll understand the **first step**.

---

# Step 1: Choose K

## What Does It Mean?

> Before K-Means starts, we must tell it how many clusters we want.

That number is:

```text
K
```

---

# Why Is This the First Step?

Remember:

K-Means itself does NOT know how many groups exist.

It asks us:

> **"How many clusters should I create?"**

We answer:

```text
K = ?
```

Only then can it begin.

---

# Example

Suppose we have these data points:

```text
2   3   4      10   11   12
```

---

## If We Choose:

```text
K = 2
```

K-Means tries to create:

```text
2 clusters
```

Possible result:

```text
Cluster 1:
2, 3, 4

Cluster 2:
10, 11, 12
```

---

## If We Choose:

```text
K = 3
```

K-Means is forced to create:

```text
3 clusters
```

Possible result:

```text
Cluster 1:
2, 3

Cluster 2:
4

Cluster 3:
10, 11, 12
```

Notice:

> The same data gives different answers depending on K.

---

# Does K Mean the True Number of Groups?

This is a very important concept.

Many beginners think:

> "K is the actual number of groups in the data."

Not exactly.

K actually means:

> **The number of clusters we ASK K-Means to create.**

Whether that matches the true structure depends on whether we chose a good K.

---

# Visual Intuition

Imagine these points:

```text
● ● ●

          ● ● ●

                    ● ● ●
```

Most humans naturally see:

```text
3 groups
```

---

## If You Choose K = 2

K-Means says:

```text
"Okay, I'll somehow fit these into 2 clusters."
```

Even if 3 groups actually exist.

---

## If You Choose K = 5

K-Means says:

```text
"Okay, I'll split these into 5 clusters."
```

Even if only 3 natural groups exist.

---

# Choosing K Too Small

Suppose the true groups are:

```text
Students
Professionals
Retired People
```

You choose:

```text
K = 2
```

K-Means may combine two different groups.

Example:

```text
Cluster 1:
Students + Professionals

Cluster 2:
Retired People
```

Problem:

> Different groups get merged together.

---

# Choosing K Too Large

Suppose the true groups are:

```text
Students
Professionals
```

You choose:

```text
K = 5
```

Possible result:

```text
Students Group A
Students Group B
Professionals Group A
Professionals Group B
Professionals Group C
```

Problem:

> Natural groups get unnecessarily split.

---

# Then How Do We Choose K?

At this point, you might ask:

> "How do I know the correct K?"

Good question.

We'll study methods later:

```text
Elbow Method
Silhouette Score
```

These help estimate a good value of K.

For now:

> Just understand that K must be chosen before running K-Means.

---

# What Happens After Choosing K?

Suppose we choose:

```text
K = 3
```

K-Means now knows:

```text
"I need to form exactly 3 clusters."
```

The next question becomes:

> "Where should the 3 cluster centers start?"

That leads to the next step:

```text
Initialize K centroids.
```

---

# Important Points to Remember

* K is chosen before K-Means starts.
* K determines how many clusters will be formed.
* K-Means does not automatically know the correct K.
* Different K values can produce different clusterings.
* Choosing K too small merges groups.
* Choosing K too large splits groups.

---

# Mini Summary

```text
Step 1: Choose K

K = Number of clusters to create.

K-Means asks:
"How many groups do you want?"

You answer:
K = ?

Wrong K:
• Too small → Merge groups.
• Too large → Split groups.

Good K:
• Produces meaningful clusters.
```

## Subtopic 2: Initialize K Centroids

Now K-Means knows:

```text id="cysvl8"
K = Number of clusters to create.
```

For example:

```text id="6e0v7q"
K = 2
```

means:

> "I need to create 2 clusters."

But now K-Means faces a problem:

> **Where should these 2 clusters start?**

This is where initialization comes in.

---

# What Does "Initialize K Centroids" Mean?

> **Initialization means choosing the starting positions of the K centroids before the algorithm begins clustering.**

In simple words:

> K-Means first guesses where the cluster centers might be.

These are called:

```text id="u2t2dn"
Initial Centroids
```

---

# Why Do We Need Initial Centroids?

Remember:

K-Means forms clusters around centroids.

But at the beginning:

```text id="zqz1ci"
No clusters exist.
No centroids exist.
```

So K-Means says:

> "Let me start with some temporary centroids."

Then it improves them later.

---

# Example

Suppose our data points are:

```text id="wwn74d"
2   3   4   10   11   12
```

and we choose:

```text id="3tnmgo"
K = 2
```

We need:

```text id="ebmnm2"
2 starting centroids.
```

---

# How Are Initial Centroids Chosen?

## Method 1: Random Initialization

The simplest approach.

K-Means randomly picks K points from the dataset.

Example:

From:

```text id="u8u2e4"
2   3   4   10   11   12
```

Suppose it randomly selects:

```text id="fc2j67"
3 and 11
```

These become:

```text id="wx9jsv"
Centroid 1 = 3
Centroid 2 = 11
```

These are only **starting guesses**.

They will move later.

---

# Important Point

At this stage:

```text id="v6brv3"
Centroid 1 = 3
Centroid 2 = 11
```

does NOT mean these are the final centroids.

They are simply:

> The initial positions.

---

# Visual Intuition

Suppose the points are:

```text id="ibxjkc"
2   3   4        10   11   12
```

Random initialization may give:

```text id="3mdk3x"
     C1                      C2
      ↓                       ↓
2   3   4        10   11   12
```

where:

```text id="e5k9h2"
C1 = 3
C2 = 11
```

K-Means then starts improving these choices.

---

# Another Example

Same dataset:

```text id="uj0h3d"
2   3   4        10   11   12
```

Random initialization might instead choose:

```text id="dkrp1w"
2 and 4
```

So:

```text id="vgoe0v"
Centroid 1 = 2
Centroid 2 = 4
```

Notice:

> Both centroids started in the same region.

This may lead to poor clustering initially.

---

# Why Is Initialization Important?

Because:

> Different starting centroids can produce different results.

Example:

Good initialization:

```text id="nmdlp9"
3 and 11
```

Bad initialization:

```text id="tfod4r"
2 and 4
```

The algorithm may still recover, but sometimes it may end up with worse clusters.

---

# Local Minimum Problem (Intuition Only)

Suppose K-Means starts poorly.

It may settle into a solution that is:

```text id="6gf5w8"
Good,
but not the best possible.
```

This is called getting stuck in a:

> **Local Minimum**

We'll discuss this more under limitations.

For now, just remember:

> Initialization affects the final result.

---

# Does K-Means Always Use Random Initialization?

Historically:

> Yes.

But modern implementations usually use:

> **K-Means++**

instead.

K-Means++ chooses smarter starting centroids.

Benefits:

```text id="ow4x1t"
Better starting points.

More stable results.

Faster convergence.
```

We'll study K-Means++ later.

For now:

> Assume K-Means starts with random centroids.

---

# What Happens Next?

Suppose we have:

Data:

```text id="0jdywq"
2   3   4   10   11   12
```

Initial centroids:

```text id="78byod"
C1 = 3
C2 = 11
```

The next question is:

> **Which points belong to C1 and which belong to C2?**

K-Means answers this by measuring distances.

This is called:

> **The Assignment Step.**

---

# Important Points to Remember

* After choosing K, K-Means initializes K centroids.
* Initial centroids are starting guesses.
* Random initialization often selects K data points.
* Initial centroids are not final centroids.
* Different initializations can produce different results.
* K-Means++ improves initialization.

---

# Mini Summary

```text id="wfwgkl"
Step 2: Initialize K Centroids

Why?
• K-Means needs starting cluster centers.

How?
• Usually choose K random points.

Important:
• These are temporary guesses.
• They move later.
• Initialization affects results.

Next:
Assign each point to its nearest centroid.
```

# Topic 3: How K-Means Works

## Subtopic 3: Assignment Step

This is the moment when K-Means forms clusters for the **first time**.

So far, we have done:

```text id="yuljjg"
Step 1: Choose K

Step 2: Initialize K Centroids
```

Now we ask:

> **Which data points belong to which centroid?**

K-Means answers this using the Assignment Step.

---

# What is the Assignment Step?

> **Each data point is assigned to the cluster of the nearest centroid.**

In simple words:

> Every point asks,

```text id="uxft9w"
"Which centroid is closest to me?"
```

and joins that cluster.

---

# How Does K-Means Decide "Nearest"?

It measures:

> **Distance between the point and each centroid.**

Usually, K-Means uses:

> **Euclidean Distance**

(which we'll study in detail later).

For now, think of it as:

> Normal straight-line distance.

---

# Example

Suppose we have:

Data points:

```text id="v0x84t"
2   3   4   10   11   12
```

Initial centroids:

```text id="t48k9j"
C1 = 3
C2 = 11
```

---

# Point = 2

Distance to C1:

```text id="th3uwm"
|2 − 3| = 1
```

Distance to C2:

```text id="6gk8km"
|2 − 11| = 9
```

Nearest centroid:

```text id="1zjlwm"
C1
```

Assignment:

```text id="gq3nsy"
2 → Cluster 1
```

---

# Point = 3

Distance to C1:

```text id="1w9d7r"
|3 − 3| = 0
```

Distance to C2:

```text id="pq5r6k"
|3 − 11| = 8
```

Nearest centroid:

```text id="sqn3qe"
C1
```

Assignment:

```text id="l4q1lb"
3 → Cluster 1
```

---

# Point = 4

Distance to C1:

```text id="0y6a8u"
|4 − 3| = 1
```

Distance to C2:

```text id="u5odq6"
|4 − 11| = 7
```

Nearest centroid:

```text id="30wp8s"
C1
```

Assignment:

```text id="omdlfv"
4 → Cluster 1
```

---

# Point = 10

Distance to C1:

```text id="fjlwmx"
|10 − 3| = 7
```

Distance to C2:

```text id="dsm4ml"
|10 − 11| = 1
```

# Important Point

At this stage:

> The centroids DO NOT move.

We simply ask:

```text id="s2kqj4"
"Which centroid is closest?"
```

and assign points.

The centroid update happens in the next step.

---

# What If a Point Is Exactly in the Middle?

Example:

Centroids:

```text id="yzwk8v"
C1 = 4
C2 = 8
```

Point:

```text id="9w0mdr"
6

Tie!

* Assign it to the first centroid encountered.
* Use a predefined rule.

This situation is rare.

---

# Important Points to Remember

* Every point compares distances to all centroids.
* The point joins the nearest centroid's cluster.
* Centroids do not change during this step.
* This is the first time clusters are formed.
* Euclidean distance is commonly used.

```


# Topic 3: How K-Means Works

## Subtopic 4: Update Step

So far, we've done:

```text
Step 1: Choose K
Step 2: Initialize K Centroids
Step 3: Assign Points to Nearest Centroid
```

After the assignment step, K-Means asks:

> **"Now that I know which points belong to each cluster, are my centroids still at the center of those clusters?"**

Usually, the answer is:

> **No.**

Therefore, K-Means updates the centroids.

---

# What is the Update Step?

> **The update step recalculates each centroid by taking the mean of all points assigned to that cluster.**

In simple words:

> Find the new center of each cluster.

---

# Why Do We Need This Step?

Remember:

The initial centroids were just guesses.

Example:

```text
Data:
2   3   4   10   11   12

Initial Centroids:
C1 = 3
C2 = 11
```

These starting positions may not be the true centers.

After forming clusters, we improve them.

---

# Example 1

From the Assignment Step, we obtained:

## Cluster 1

```text
2, 3, 4
```

Current centroid:

```text
C1 = 3
```

---

## Cluster 2

```text
10, 11, 12
```

Current centroid:

```text
C2 = 11
```

---

## Update Cluster 1

New centroid:

```text
(2 + 3 + 4) / 3
= 9 / 3
= 3
```

Therefore:

```text
New C1 = 3
```

---

## Update Cluster 2

New centroid:

```text
(10 + 11 + 12) / 3
= 33 / 3
= 11
```

Therefore:

```text
New C2 = 11
```

---

# Result

Old centroids:

```text
C1 = 3
C2 = 11
```

New centroids:

```text
C1 = 3
C2 = 11
```

They didn't move.

This means these centroids were already at the centers of their clusters.

---

# Example 2 (Centroids Move)

Suppose after assignment we have:

## Cluster 1

```text
2, 3, 10
```

Current centroid:

```text
C1 = 2
```

---

## Cluster 2

```text
11, 12
```

Current centroid:

```text
C2 = 12
```

---

## Update Cluster 1

New centroid:

```text
(2 + 3 + 10) / 3
= 15 / 3
= 5
```

So:

```text
New C1 = 5
```

---

## Update Cluster 2

New centroid:

```text
(11 + 12) / 2
= 23 / 2
= 11.5
```

So:

```text
New C2 = 11.5
```

---

# What Happened?

Old centroids:

```text
C1 = 2
C2 = 12
```

became

```text
C1 = 5
C2 = 11.5
```

Notice:

> The centroids moved closer to the centers of their assigned points.

---

# Visual Intuition

### Before Update

```text
2   3   10       11   12
↑                 ↑
C1=2             C2=12
```

---

### After Update

```text
2   3   10       11   12
      ↑            ↑
     C1=5       C2=11.5
```

The centroids shifted toward the middle of their clusters.

---

# Important Observation

The new centroid:

> **Does NOT have to be an actual data point.**

Example:

Cluster:

```text
11, 12
```

New centroid:

```text
11.5
```

There is no point at 11.5.

Yet:

> It is the correct centroid.

---

# What Happens Next?

After updating:

K-Means asks again:

```text
"Now that the centroids moved,
should some points change clusters?"
```

So it goes back to:

```text
Assignment Step
↓
Update Step
↓
Assignment Step
↓
Update Step
...
```

This repeats until the centroids stop changing.

---

# Important Points to Remember

* The update step recalculates centroids.
* The new centroid is the mean of all points in that cluster.
* Centroids often move after updating.
* Updated centroids may not be actual data points.
* After updating, K-Means performs the assignment step again.

---

# Mini Summary

```text
Step 4: Update Step

For each cluster:

Take all assigned points
↓
Compute their mean
↓
Replace old centroid with new centroid

Result:
Centroids move toward the centers of their clusters.

Then:
Repeat Assignment → Update until centroids stop moving.
```

# Topic 3: How K-Means Works ⭐⭐⭐⭐⭐

## Subtopic 5: Repeat Until Convergence

So far, K-Means has been doing this:

```text id="1rco43"
Choose K
↓
Initialize Centroids
↓
Assign Points
↓
Update Centroids
```

After updating the centroids, an important question arises:

> **How does K-Means know when to stop?**

The answer is:

> **It stops when it converges.**

---

# What is Convergence?

## Definition

> **Convergence means that the centroids no longer change (or change very little) between iterations.**

In simple words:

> K-Means stops when nothing meaningful changes anymore.

---

# The K-Means Loop

K-Means keeps repeating these two steps:

```text id="u5y9gx"
Assignment Step
↓
Update Step
↓
Assignment Step
↓
Update Step
↓
...
```

until convergence.

---

# Example

Suppose we have:

```text id="lsk3e8"
2   3   4   10   11   12
```

and

```text id="7rbjlwm"
K = 2
```

---

## Iteration 1

Initial Centroids:

```text id="c3xkxt"
C1 = 2
C2 = 12
```

---

### Assignment

```text id="tl7m3k"
Cluster 1:
2, 3, 4

Cluster 2:
10, 11, 12
```

---

### Update

New centroids:

```text id="96ibxb"
C1 = (2+3+4)/3 = 3

C2 = (10+11+12)/3 = 11
```

Centroids moved.

Therefore:

> Continue.

---

## Iteration 2

Current centroids:

```text id="ncvpq0"
C1 = 3
C2 = 11
```

---

### Assignment Again

Check every point again.

Result:

```text id="8d9y40"
Cluster 1:
2, 3, 4

Cluster 2:
10, 11, 12
```

No points changed clusters.

---

### Update Again

New centroids:

```text id="z0q1ko"
C1 = 3

C2 = 11
```

Same as before.

---

# What Happened?

Previous centroids:

```text id="hq3x3n"
3
11
```

Current centroids:

```text id="jvbg2f"
3
11
```

No movement.

Therefore:

> **K-Means has converged.**

It stops.

---

# Visual Intuition

## Iteration 1

```text id="hqfx3k"
2   3   4         10   11   12
↑                            ↑
C1=2                       C2=12
```

Update:

```text id="6im6my"
2   3   4         10   11   12
    ↑                  ↑
   C1=3             C2=11
```

Centroids moved.

---

## Iteration 2

Assignment:

```text id="80aqkl"
Cluster 1:
2,3,4

Cluster 2:
10,11,12
```

Update:

```text id="l7ohvx"
C1=3
C2=11
```

No movement.

Stop.

---

# When Does K-Means Stop?

Usually, K-Means stops when one of these conditions is met.

---

## Condition 1 (Most Common)

Centroids stop changing.

```text id="qarfcc"
Old centroids
=
New centroids
```

---

## Condition 2

Cluster assignments stop changing.

Example:

```text id="dt91kj"
Iteration 5:
2 → Cluster 1
3 → Cluster 1
...

Iteration 6:
Exactly the same assignments.
```

Nothing changes.

Stop.

---

## Condition 3

Maximum iterations reached.

Example:

```text id="t5hf4k"
max_iter = 300
```

If convergence hasn't occurred by then:

> Stop anyway.

This prevents infinite loops.

Scikit-learn uses this approach.

---

# Does K-Means Always Converge?

> **Yes, K-Means eventually converges.**

Why?

Because:

```text id="wejlwm"
Assignment Step
↓
Never increases the objective

Update Step
↓
Never increases the objective
```

The total within-cluster distance keeps decreasing (or stays the same).

Eventually:

> It cannot improve further.

So it stops.

---

# Important Clarification

Many beginners think:

> "Convergence means K-Means found the perfect clustering."

Not necessarily.

Convergence only means:

> **The algorithm cannot improve further from its current starting point.**

It may have found:

```text id="vx7m9u"
A very good solution

OR

A local minimum
```

depending on initialization.

---

# Example of Local Minimum

Suppose:

True best clustering:

```text id="egdf0u"
Cluster A

Cluster B
```

Because of poor initialization:

K-Means converges to:

```text id="6ojw0h"
Cluster A'

Cluster B'
```

which is acceptable but not optimal.

This is why initialization matters.

---

# Important Points to Remember

* K-Means repeatedly performs Assignment and Update.
* Convergence means centroids stop changing.
* Cluster assignments may also stop changing.
* Scikit-learn additionally uses a maximum iteration limit.
* Convergence does not guarantee the globally best solution.

---

# Mini Summary

```text id="m8n6mr"
Step 5: Repeat Until Convergence

Assignment
↓
Update
↓
Assignment
↓
Update
↓
...

Stop when:
• Centroids stop moving, or
• Assignments stop changing, or
• Maximum iterations reached.

Convergence:
The algorithm has stabilized.
```

# Most Important Takeaway

```text id="t6p2k9"
K-Means is simply:

Guess centroids
↓
Assign points
↓
Move centroids
↓
Repeat
↓
Until nothing changes
```

If you understand this loop, you understand the heart of K-Means.

---

# Mini Summary

```text id="n4j7m2"
Full K-Means Algorithm:

1. Choose K
2. Initialize K centroids
3. Assign each point to nearest centroid
4. Update centroids using the mean
5. Repeat steps 3 and 4
6. Stop when convergence occurs

Core Idea:
Assign → Update → Repeat.
```
