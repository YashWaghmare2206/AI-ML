# **Topic 3: Entropy**

## **Subtopic 3.1: What is Entropy?**

**Theory**

Entropy measures:
*     How much uncertainty exists in a system before we know the outcome.

In simple words:

* Entropy = Amount of uncertainty
* Higher uncertainty → Higher entropy
* Lower uncertainty → Lower entropy

**Important Point**

Entropy does not look at what actually happened.

It looks at:
*     "How uncertain am I before the event occurs?"

**Key Takeaway**

* Entropy measures uncertainty.
* More uncertainty → Higher entropy.
* Less uncertainty → Lower entropy.
* Zero uncertainty → Zero entropy.

## **Subtopic 3.2: Entropy as Average Uncertainty**

In the previous topic:
*     Entropy = Uncertainty

But there is one refinement:
*     Entropy = Average uncertainty across all possible outcomes.

**Fair Coin**

*     50% Head
*     50% Tail

Hard to predict. | High entropy.

Biased Coin

*     99% Head
*     1% Tail

Easy to predict | Low entropy.

### **Important Observation**

Entropy depends on:
*     Number of outcomes
*     Probability distribution of outcomes

Not just the number of outcomes.

### Key Takeaway

Entropy is not simply:
*     "How many outcomes exist?"

It is:
*     "On average, how uncertain are we before observing the outcome?"

1. A balanced distribution → Higher entropy.
2. A highly predictable distribution → Lower entropy.

## **Subtopic 3.3: Entropy Formula Intuition**

What Should a Good Entropy Formula Do?

Since entropy measures uncertainty:
*     A good formula should satisfy these rules.

**Rule 1: More Uncertainty → Higher Entropy**

Example:

Fair Coin

| Outcome | Probability |
| ------- | ----------- |
| Head    | 50%         |
| Tail    | 50%         |

Hard to predict.

Entropy should be high.

| Outcome | Probability |
| ------- | ----------- |
| Head    | 99%         |
| Tail    | 1%          |

Easy to predict.

Entropy should be lower.

**Rule 2: No Uncertainty → Zero Entropy**

| Outcome | Probability |
| ------- | ----------- |
| Head    | 100%        |

Before tossing: Head

You already know the answer.
1. Uncertainty = 0
2. Entropy should be 0.

**Rule 3: Rare Events Should Contribute More**

Recall self-information:

| Event Probability | Information |
| ----------------- | ----------- |
| 99%               | Small       |
| 1%                | Large       |

1. Rare events are more informative.
2. The entropy formula must take this into account.

### Key Takeaway

The entropy formula is designed so that:

* Balanced outcomes → High entropy
* Predictable outcomes → Low entropy
* Certain outcomes → Zero entropy
* Rare events contribute more information than common events

That's the intuition.
