# **# Topic 1: Why Mutual Information?**

## **Subtopic 1.1: Why Correlation Is Not Enough?**

### **What Correlation Tries To Do**

Correlation tries to answer:

"When X changes, does Y also change in a predictable way?"

**Example**:

Study Hours	Marks

| Study Hours | Marks |
| ----------- | ----- |
| 1           | 40    |
| 2           | 50    |
| 3           | 60    |
| 4           | 70    |
| 5           | 80    |


As study hours increase, marks increase.

Correlation will be high.

So correlation says:

"Study Hours is a useful feature for predicting Marks."

### **When Correlation Works Well**

Correlation works best when the relationship is approximately a straight line.

**Visualize:**



This is called a linear relationship.

Correlation likes this type of pattern.

### **The Problem**

Now consider:

| X  | Y |
| -- | - |
| -3 | 9 |
| -2 | 4 |
| -1 | 1 |
| 0  | 0 |
| 1  | 1 |
| 2  | 4 |
| 3  | 9 |

Notice:



Clearly, Y depends completely on X.

If I know X = 3, I immediately know Y = 9.

If I know X = 2, I immediately know Y = 4.

There is a strong relationship.

### **But Correlation Gets Confused**

Look at the graph:



This is a curve, not a straight line.

Correlation mainly measures straight-line relationships.

So it may give a value close to 0.

### **Why This Is A Problem In ML**

Imagine:

Target = House Price

Feature = Distance from City Center

**The relationship might look like:**



Not a straight line.

But the feature is still useful.

If we use only correlation:

We may wrongly remove this feature.

### What We Need

**We need a method that asks:**

"Does X contain information about Y?"

**instead of**

"Is the relationship a straight line?"

This idea leads to **Mutual Information**.

## **## Subtopic 1.2: Linear vs Non-Linear Relationships**

Before understanding Mutual Information, you must clearly understand the difference between linear and non-linear relationships.

**What is a Relationship?**

When one variable changes and another variable changes because of it, we say there is a relationship.

| Study Hours | Marks |
| ----------- | ----- |
| 1           | 40    |
| 2           | 50    |
| 3           | 60    |
| 4           | 70    |

As study hours increase, marks increase.
There is a relationship.


### **### Linear Relationship**

A relationship is called linear when the change follows roughly a straight line.

**Example:**

| X | Y  |
| - | -- |
| 1 | 10 |
| 2 | 20 |
| 3 | 30 |
| 4 | 40 |

Notice:

**X increases by 1
Y increases by 10**

The rate of change remains constant

**Graph:**



A straight line.

**Important Property**

For a linear relationship:

Equal changes in X produce equal changes in Y.

Example:

X +1 → Y +10
X +1 → Y +10
X +1 → Y +10

Constant rate.

### **Non-Linear Relationship**

| X | Y  |
| - | -- |
| 1 | 1  |
| 2 | 4  |
| 3 | 9  |
| 4 | 16 |

Relationship:


Let's see the changes:

1 → 4   (+3)

4 → 9   (+5)

9 → 16  (+7)

The change is not constant.

**Graph:**



Curved.

This is a non-linear relationship.


**Another Example**

| Age | Learning Speed |
| --- | -------------- |
| 5   | Low            |
| 15  | High           |
| 25  | Very High      |
| 70  | Low            |

The relationship goes up and then down.

Graph:



Definitely not a straight line.

Still a relationship exists.

Real-world data is rarely perfectly linear.

### **The Problem with Correlation**

Correlation is designed mainly to detect straight-line patterns.

So:

✅ Linear relationship → Correlation works well

❌ Non-linear relationship → Correlation may miss important information

### **Why Mutual Information Is Better**

**Mutual Information doesn't ask:**

"Is the relationship a straight line?"

**It asks:**

"If I know X, do I learn something about Y?"

If yes, MI becomes high.

It doesn't matter whether the relationship is:

Straight | Curved | Complex

## **Subtopic 1.3: Dependency vs Correlation**

This is one of the most important ideas behind Mutual Information.

**Many people think:**

"If correlation is low, then the variables are unrelated."

This is wrong.

### **What is Dependency?**

Dependency simply means:

Knowing one variable helps us know something about another variable.

If X changes and Y is affected in some way, then Y depends on X.

**Example 1**

| Temperature | Ice Cream Sales |
| ----------- | --------------- |
| Low         | Low             |
| Medium      | Medium          |
| High        | High            |

If I know the temperature is high, I can guess ice cream sales will be high.

Temperature gives information about sales.

**Temperature and Sales are dependent.**

### **What is Correlation?**

Correlation is much more specific.

**It asks:**

"Do X and Y move together in a straight-line pattern?"

**Notice:**

Dependency = Any kind of relationship
Correlation = Linear relationship

**Correlation is only one type of dependency.**

### **Think of Dependency as a Big Circle**


* Dependency
* │
* ├── Linear Dependency
* │     └── Correlation measures this
* │
* ├── Curved Dependency
* │
* ├── Complex Dependency
* │
* └── Other Dependencies

**Correlation covers only one small part.**

**Mutual Information tries to capture all of them.**

### **Example 2: Strong Dependency, Weak Correlation**

Consider:


| X  | Y |
| -- | - |
| -3 | 9 |
| -2 | 4 |
| -1 | 1 |
| 0  | 0 |
| 1  | 1 |
| 2  | 4 |
| 3  | 9 |

If I tell you:

X = 2

You instantly know:

Y = 4

If I tell you:

X = 3

You instantly know:

Y = 9

So Y completely depends on X.

Dependency is very strong.

**But the graph is curved:**



Correlation may be close to zero.

### **Why This Happens**

**Correlation asks:**

"Do points form a straight line?"

The graph above does not.

So correlation becomes small.

**Mutual Information asks:**

"If I know X, do I know more about Y?"

**Answer:**

Yes, absolutely.

Therefore MI will be high.

### **Why Feature Selection Needs Dependency**

When selecting features, we don't care whether the relationship is straight.

**We care:**

"Does this feature contain useful information about the target?"

That's exactly what Mutual Information measures.
