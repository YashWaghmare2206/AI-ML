## **Subtopic 8.3: Reading and Interpreting MI Scores**



### **Example Output**

* Income      0.82
* Age         0.45
* Gender      0.12
* Color       0.01


What Does 0.82 Mean?

Wrong thinking:
* `0.82 = 82% important`

❌ Not true.

MI is not a percentage

**Correct Thinking**

MI scores are mainly used for:
* `Comparing features with each other.`

Example:

* Feature	MI
* Income	0.82
* Age	0.45
* Color	0.01

Interpretation:
* `**Income > Age > Color**`
