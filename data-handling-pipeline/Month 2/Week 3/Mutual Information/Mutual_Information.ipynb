# Topic 5: Mutual Information

## **Subtopic 5.1: What is Mutual Information?**

You already know:

* Entropy = Uncertainty
* Conditional Entropy = Remaining uncertainty after knowing X
* Reduction in uncertainty = Entropy − Conditional Entropy

Now we give this reduction a name:
* `Mutual Information (MI)`

**Definition**

Mutual Information measures:
* `How much knowing X reduces uncertainty about Y.`

### Example 1: Useful Feature

**Target:**
* Will customer buy?

**Feature:**
* Income

**Data:**

| Income | Buy |
| ------ | --- |
| High   | Yes |
| High   | Yes |
| Low    | No  |
| Low    | No  |

If I tell you: **Income =**  **High**

You immediately know:  **Buy = Yes**

**Income removed a lot of uncertainty**.

Therefore:
* `Mutual Information is high.`

### Example 2: Useless Feature

| Color | Buy |
| ----- | --- |
| Red   | Yes |
| Blue  | No  |
| Red   | No  |
| Blue  | Yes |

If I tell you: **Color = Red**

**Can you predict Buy? No**.

Still uncertain.

**Color removed almost no uncertainty**.

Therefore:
* `Mutual Information is low.`

### Simple Interpretation

**Mutual Information asks:**

* `"How useful is this feature for predicting the target?"`

* **More useful feature**
* ↓
* More uncertainty removed
* ↓
* Higher MI


* **Less useful feature**
* ↓
* Less uncertainty removed
* ↓
* Lower MI

**Key Takeaway**
* `Mutual Information = Amount of uncertainty removed by knowing a feature.`

## **Subtopic 5.2: Mutual Information Formula**

**Mutual Information is:**

* `MI(X,Y)=H(Y)−H(Y∣X)`

1. H(Y) : Uncertainty about Y before knowing X.
2. H(Y|X) : Uncertainty about Y after knowing X.

**Mutual Information = Original Uncertainty - Remaining Uncertainty**

## **Subtopic 5.3: Understanding MI Scores**

When you calculate Mutual Information, you'll get a number.

The question is:
* `What does that number mean?`

### **MI = 0**

This means:
* `Knowing X gives no information about Y.`

**Example**

| Color | Buy |
| ----- | --- |
| Red   | Yes |
| Blue  | No  |
| Red   | No  |
| Blue  | Yes |

Knowing color doesn't help predict buying. **MI = 0**

### **Low MI**

This means:
* `X provides a little information about Y.`

| Income | Buy |
| ------ | --- |
| High   | Yes |
| High   | No  |
| Low    | No  |
| Low    | No  |

Income helps somewhat.

But prediction is not perfect.

Therefore: **Low MI**

### **High MI**

This means:
* **X provides a lot of information about Y.**

| Income | Buy |
| ------ | --- |
| High   | Yes |
| High   | Yes |
| Low    | No  |
| Low    | No  |

Knowing income almost tells you the answer.

Therefore: **High MI**

### Important Point

Unlike correlation:

**Correlation:**
* `-1 to +1`

**Mutual Information:**
* `0 to positive values`
* `MI is never negative.`

## **Subtopic 5.4: Why Mutual Information Can Never Be Negative**

**Formula**
* `MI(X,Y)=H(Y)−H(Y∣X)`

MI = Original Uncertainty - Remaining Uncertainty

### **Case 1**

Reduce uncertainty

100 ->  40

* Reduction = 60
* MI > 0

### Case 2

Change nothing

100 -> 100

* Reduction = 0
* MI = 0

**Can Knowing a Feature Increase Uncertainty?**

Can this happen?

100 ->  130

No.
* A feature cannot make you know less than before.
* At worst, the feature is useless and changes nothing

**Remaining uncertainty can never be larger than original uncertainty.**

* Conditional Entropy ≤ Entropy
* Entropy - Conditional Entropy ≥ 0

Therefore:
MI ≥ 0


### Simple Interpretation

MI = 0
* `Feature gives no information.`

MI > 0
* `Feature gives some information.`

MI < 0
* `Impossible`
