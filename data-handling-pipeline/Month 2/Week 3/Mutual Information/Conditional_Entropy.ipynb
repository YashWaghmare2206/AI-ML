# **Topic 4: Conditional Entropy**

## **Subtopic 4.1: What is Conditional Entropy?**

We already know:
* Entropy = Uncertainty

Now we ask:
* What happens to uncertainty after we learn some new information?

That's Conditional Entropy.

Conditional Entropy measures:
* `How much uncertainty remains after knowing another variable.`

### Example

Target:
* Will customer buy?

**Before seeing any feature:**

| Customer | Buy |
| -------- | --- |
| A        | Yes |
| B        | No  |
| C        | Yes |
| D        | No  |

Prediction:
* `Yes?
* No?`

* Uncertainty exists.
* Entropy is high.


**Now suppose we know:**

| Income | Buy |
| ------ | --- |
| High   | Yes |
| High   | Yes |
| Low    | No  |
| Low    | No  |

If someone tells you:
* Income = High

You immediately know:
* Buy = Yes

Almost no uncertainty remains

**After knowing Income:**
* Know answer

Low uncertainty.

The remaining uncertainty is called:
*  `Conditional Entropy`

**Conditional Entropy = Uncertainty remaining after knowing X**

## **Subtopic 4.2: Entropy Before Knowing X vs After Knowing X**

**Step 1: Before Knowing X**

| Customer | Buy |
| -------- | --- |
| A        | Yes |
| B        | No  |
| C        | Yes |
| D        | No  |

There is uncertainty.

This uncertainty is called:
* Entropy of Y


**Step 2: After Knowing X**

| Income | Buy |
| ------ | --- |
| High   | Yes |
| High   | Yes |
| Low    | No  |
| Low    | No  |

I tell you:
* Income = High

Now what happens?

You immediately predict:
* Buy = Yes

**Much of the uncertainty disappears.**


**Before Knowing Income**
* Entropy(Y)

**After Knowing Income**
* Conditional Entropy H( Y | Income)
* Entropy of Y given Income.

**The Core Idea**

* Before knowing X
* ↓
* Entropy
* ↓
* Know X
* ↓
* Some uncertainty removed
* ↓
* Remaining uncertainty
* =
* Conditional Entropy

### Good Feature vs Bad Feature

**ML Interpretation**

A useful feature:

* High Entropy
* ↓
* Feature known
* ↓
* Low Conditional Entropy


A useless feature:

* High Entropy
* ↓
* Feature known
* ↓
* Still High Conditional Entropy


**Mutual Information = Entropy - Conditional Entropy**

In words:
* How much uncertainty was removed after knowing the feature?
