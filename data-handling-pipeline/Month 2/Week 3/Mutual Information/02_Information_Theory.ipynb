# Topic 2: Information Theory Basics

## **Subtopic 2.1: What is Information?**

This may sound like a simple question, but Mutual Information is built on a very specific meaning of the word information.

### 1. Information in Daily Life

Suppose I tell you:

**"The sun will rise tomorrow."**

Did you learn anything useful?

Not really.

You already expected it.

There was almost no surprise.

Therefore:

**Very little information was gained.**


**Now suppose I tell you:**

"Tomorrow all colleges in India are closed unexpectedly."

This is unexpected.

You learned something new.

Therefore:

**A lot of information was gained.**


### **2. Information and Prediction**

Think about ML.

A model is constantly trying to predict outcomes.

**If something is easy to predict:**

**Little information**

**If something is hard to predict:**

**More information**

**Weather Example**

| Weather | Probability |
| ------- | ----------- |
| Sunny   | 95%         |
| Rainy   | 5%          |

If I tell you:

"Today is sunny."

Not surprising.

Little information.

If I tell you:

"Today is rainy."

Very surprising.

More information.

### **Key Observation**

Rare events provide more information.

Common events provide less information.


### **ML Connection**

Suppose a feature tells us something rare but useful about the target.

That feature provides information.

**Mutual Information tries to measure:**

"How much information about the target is contained inside a feature?"

### **Simple Feature Example**

**Target:**  Will Customer Buy?

**Feature:**  Income

| Income | Buy |
| ------ | --- |
| High   | Yes |
| High   | Yes |
| High   | Yes |
| Low    | No  |

If I know income, I can predict buying behavior much better.

Income provides information about the target.

Therefore:

**Mutual Information between Income and Buy will be high.**

### **Important Distinction**

**Information ≠ Data**

Data = "Today is sunny"

Whether it contains much information depends on how expected it was.

If sunny happens every day:

Little information.

If sunny almost never happens:

High information.


### **Key Takeaways**

Information measures surprise.

**More surprise → More information.**

**Less surprise → Less information.**


Rare events carry more information.

Common events carry less information.

Mutual Information will eventually measure:
*     How much information a feature provides about the target.

## **Subtopic 2.2: What is Uncertainty?**

Uncertainty means:

*     "I don't know what will happen."

The more possible outcomes there are, the more uncertain we are.

### **Example 1: Coin Toss**

Before tossing a fair coin:

Possible outcomes:

Head | Tail

Can you predict exactly what will happen?

No.

There is uncertainty.

After the coin lands:

Suppose it lands Head.

Now you know the result.

Uncertainty becomes zero

### **Example 2: Dice Roll**

Before rolling:

Possible outcomes:

1
2
3
4
5
6

Can you predict the exact number?

No.

There is uncertainty.

Compared to a coin:

Coin → 2 possibilities
Dice → 6 possibilities

Which has more uncertainty?

The dice.

Because more possible outcomes exist.


### **Uncertainty and Prediction**

Think like a machine learning model.

Suppose we want to predict:

Will customer buy?

#### **Case 1**

| Customer | Buy |
| -------- | --- |
| A        | Yes |
| B        | Yes |
| C        | Yes |
| D        | Yes |

Before seeing a new customer:

What do we predict?

"Yes"

Almost certainly.

Very little uncertainty.


#### **Case 2**

| Customer | Buy |
| -------- | --- |
| A        | Yes |
| B        | No  |
| C        | Yes |
| D        | No  |

Now prediction becomes harder.

Uncertainty increases.


**Important Observation**

When prediction is difficult:

➡️ Uncertainty is high

When prediction is easy:

➡️ Uncertainty is low


### **Why ML Cares About Uncertainty**

Machine Learning is basically:

*     Reducing uncertainty about the target.

Example:

Target = Will customer buy?

**Before looking at features:**

Maybe yes
Maybe no

High uncertainty.

**After looking at Income:**

High income → likely Yes
Low income → likely No

Uncertainty decreases.

This reduction in uncertainty is exactly what Mutual Information measures later.

### **Key Takeaways**

Uncertainty

Means:
*     Not knowing which outcome will occur.

* More possible outcomes  → More uncertainty
* Easier prediction  → Less uncertainty
* Harder prediction  → More uncertainty


Information reduces uncertainty

This sentence is extremely important because later we'll define:

* Entropy = Amount of uncertainty
* Mutual Information = Reduction in uncertainty

## **Subtopic 2.4: Why Rare Events Carry More Information Than Common Events**

### **Core Idea**

The less likely an event is,

**➡️ the more information it carries when it happens.**

The more likely an event is,

**➡️ the less information it carries when it happens.**

Important Observation

Information and probability move in opposite directions.

High Probability
1. Expected
2. Less surprise
3. Less information

Low Probability
1. Unexpected
2. More surprise
3. More information

**Information and probability are inversely related**

1. Higher probability → Lower information
2. Lower probability → Higher information


## **Subtopic 2.5: Self-Information (How Information is Measured)**

Until now we've said:

*    Rare events carry more information.

But how much more?

Can we assign a number to information?

That's where **Self-Information** comes in.

### **What is Self-Information?**

Self-information measures:

How much information is gained when a specific event occurs.

Think of it as:

Event occurs

↓

How surprised am I?

↓

Information gained

### We Need a Measurement

We want a measure that follows these rules:

Rule 1

Less probability -> More information

Rule 2

More probability -> Less information

Rule 3

Probability = 1 -> Information = 0

Because if something is guaranteed:
No surprise


The Formula

Information Theory uses:

**I(x)=−log(P(x))**

where:

**I(x) = Information of event x**

**P(x) = Probability of event x**



### **Key Takeaways**

**Self-Information**

Measures information gained from a single event.

**Formula**

I(x)=−log(P(x))

High Probability → Low information

Low Probability → High information

Guaranteed Event

**P(x)=1**

→ Information = 0

## **Subtopic 3.4: The Entropy Formula**

The entropy formula is:

**H(X)=−∑P(x)logP(x)**

H(X)
* H stands for Entropy.
* X is the variable we're measuring

P(x)
* This is the probability of an outcome

| Outcome | Probability |
| ------- | ----------- |
| Head    | 0.5         |
| Tail    | 0.5         |


log(P(x))

You already learned from **self-information**:
* I(x)=−log(P(x))

This measures:

Information carried by one outcome.

* Rare outcome → larger value
* Common outcome → smaller value

**P(x) × Information**

This gives the outcome's contribution to entropy.

**Σ (Sigma)**

Sigma means:
* Add all outcomes together.

**Example: Coin**

**Head contribution + Tail contribution**

Entropy is simply:
* `Average self-information of all possible outcomes.`

## **Subtopic 3.5: High Entropy vs Low Entropy**

### **High Entropy**

High entropy means:
* `High uncertainty.`

| Outcome | Probability |
| ------- | ----------- |
| Head    | 50%         |
| Tail    | 50%         |

Before tossing:
* `Hard to predict.`
* `High uncertainty.`


### **Low Entropy**

Low entropy means:
* `Low uncertainty`.

| Outcome | Probability |
| ------- | ----------- |
| Head    | 99%         |
| Tail    | 1%          |

Before tossing:
*     `Head (Mostly to be predicted)`
*      `Uncertainty is small.`

Therefore:
*` Low entropy`


## **Subtopic 3.6: Zero Entropy and Maximum Entropy**

### **Zero Entropy**

Zero entropy means:
* No uncertainty at all.

You already know the outcome before it happens.

**Example:**
| Outcome | Probability |
| ------- | ----------- |
| Head    | 100%        |

**No uncertainty.**

Therefore:
**Entropy = 0**

### **Maximum Entropy**

Maximum entropy means:
* Highest possible uncertainty.

All outcomes are equally likely.

**Example:**
| Outcome | Probability |
| ------- | ----------- |
| Head    | 50%         |
| Tail    | 50%         |

You genuinely don't know.

This is the highest uncertainty possible for a coin.

Therefore:
* Maximum entropy for a coin
