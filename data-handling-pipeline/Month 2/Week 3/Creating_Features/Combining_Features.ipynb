## **Subtopic 3.1 — What Are Feature Interactions?**

### **Theory**

**Feature Interaction means:**

* `Combining two or more existing features to create a new feature.`

Instead of using features separately:

* `Feature A`
* `Feature B`

we create:

* Feature A + Feature B
* Feature A × Feature B
* Feature A / Feature B
* Feature A - Feature B

### **Why Do We Do This?**

* Sometimes the relationship is not inside a single feature.
* The useful information comes from how features relate to each other.

#### Example 1 — House Price

| Area | Bedrooms |
| ---- | -------- |
| 1000 | 2        |
| 1500 | 3        |
| 2000 | 4        |

Individually:

* Area
* Bedrooms

are useful.

But create: **Area_per_Bedroom = Area / Bedrooms**

| Area | Bedrooms | Area_per_Bedroom |
| ---- | -------- | ---------------- |
| 1000 | 2        | 500              |
| 1500 | 3        | 500              |
| 2000 | 4        | 500              |

This tells us something new:
* `How much space each room gets.`

### **Important Warning**

Not every combination is useful.

**Bad example:**

* `Age × ZipCode`

Usually meaningless.

**Good feature interactions should have:**
* `Business Meaning`
* `Domain Meaning`

## **Subtopic 3.2 — Addition Features**

We combine two or more features using:

* `New Feature = Feature A + Feature B`

The goal is to create a feature representing a total, combined score, or overall amount

### Example — Monthly Spending

| Food | Transport |
| ---- | --------- |
| 5000 | 2000      |
| 7000 | 3000      |

Create:
* `Total_Expense = Food + Transport`

Result:

| Food | Transport | Total_Expense |
| ---- | --------- | ------------- |
| 5000 | 2000      | 7000          |
| 7000 | 3000      | 10000         |

### **Key Principle**

Before creating:

* `**A + B**`

ask:

* **`Does the sum represent something meaningful?`**

* If yes, it may become a useful feature.
* If no, don't create it.

## **Subtopic 3.3 — Subtraction Features**

We create a new feature using:

* `New Feature = Feature A - Feature B`

Subtraction is useful when the difference between two values is more important than the values themselves.

### **Example 1 — Income and Expenses**

| Income | Expense |
| ------ | ------- |
| 50000  | 30000   |
| 70000  | 45000   |

Create:
* `Savings = Income - Expense`

| Income | Expense | Savings |
| ------ | ------- | ------- |
| 50000  | 30000   | 20000   |
| 70000  | 45000   | 25000   |


### **When Subtraction Features Are Useful**

**Useful**

* **`When the difference itself matters.`**

Examples:

* Income - Expense
* Current Marks - Previous Marks
* Actual Sales - Target Sales
* Actual Weight - Ideal Weight

## **Subtopic 3.4 — Multiplication Features**

What is a Multiplication Feature?

We create:

* **`New Feature = Feature A × Feature B`**

The idea is:

`Sometimes the effect of one feature depends on another feature.`

### **Example 1 — Revenue**

| Price | Quantity |
| ----- | -------- |
| 100   | 2        |
| 50    | 5        |

Create:
* `Revenue = Price × Quantity`

| Price | Quantity | Revenue |
| ----- | -------- | ------- |
| 100   | 2        | 200     |
| 50    | 5        | 250     |

Notice:

* Price alone ❌
* Quantity alone ❌
*
* Price × Quantity ✅

Revenue is what actually matters.

### **Why Multiplication Is Powerful**

Suppose we predict sales.

Features:

* `Advertisement Budget`
* `Festival Season`

Maybe:

* `High Budget + Festival`

causes much higher sales than expected.

The interaction:

* `Budget × Festival`

helps the model see that combined effect.

## **Subtopic 3.5 — Division Features**

What is a Division Feature?

We create:

* `New Feature = Feature A / Feature B`

The goal is usually to create:

* Ratios
* Rates
* Efficiency measures
* Per-unit metrics

### **Example 1 — Area Per Room**

| Area | Rooms |
| ---- | ----- |
| 1000 | 2     |
| 2000 | 4     |

Create:
* `Area_Per_Room = Area / Rooms`

| Area | Rooms | Area_Per_Room |
| ---- | ----- | ------------- |
| 1000 | 2     | 500           |
| 2000 | 4     | 500           |

## **Important Warning**

Division has the same issue as reciprocal transformation:

* `Denominator = 0`

causes:

* `Division by Zero ❌`

Example:

* Revenue = 1000
* Customers = 0

1000 / 0 ❌

**Always check for zeros.**

## **Subtopic 3.6 — How to Decide Which Interactions Are Worth Creating?**

### **Rule 1: Look for Business Meaning**

Good Interaction:

Price × Quantity

Result:

Revenue

Everyone understands what Revenue means.

### **Rule 2: Ask What Decision This Feature Helps Make**

Example:

Revenue / Customer

Helps answer:

Which customers are more valuable?

Useful.

### **Rule 3: Domain Knowledge Beats Random Combinations**

Suppose you're working on house prices.

Features:

Area
Rooms
Bathrooms

A domain expert may suggest:

Area / Rooms

because buyers care about room size.

That is far more valuable than generating 500 random combinations.

### **Rule 4: More Features ≠ Better Model**

* A + B
* A - B
* A * B
* A / B
* B / A
* A ** 2
* B ** 2

Result:

Thousands of features

**Most are noise.**

### **Rule 5: Validate with Model Performance**

Suppose you create:

* `df["Revenue"] = (
*     df["Price"] * df["Quantity"]
* )`

**How do you know it's useful?**

You test it.

* Without Revenue
* ↓
* Train Model
* ↓
* Accuracy = 82%
*
* With Revenue
* ↓
* Train Model
* ↓
* Accuracy = 86%

Then:

**Revenue is useful**

# Subtopic 3.7 — Common Mistakes in Combining Features

This is where many beginners lose marks in competitions and make models worse instead of better.

---

# Mistake 1 — Combining Everything with Everything

Suppose you have:

```text
Age
Salary
Experience
City_Code
Customer_ID
```

A beginner creates:

```text
Age × Salary
Age × Experience
Age × City_Code
Salary × Experience
Salary × Customer_ID
...
```

Soon:

```text
5 features
↓
50 features
↓
500 features
```

Most of them are useless.

---

## Why Is This Bad?

The model now sees:

```text
Useful Features
+
Noise
```

Too much noise can hurt learning.

---

# Mistake 2 — Ignoring Business Meaning

Bad feature:

```text
Customer_ID × Salary
```

Example:

| Customer_ID | Salary |
| ----------- | ------ |
| 1001        | 50000  |

Result:

```text
50,050,000
```

What does this mean?

```text
Nothing
```

The model gets a large number with no meaning.

---

# Mistake 3 — Division by Zero

Example:

```python
df["Revenue_Per_Customer"] = (
    df["Revenue"] / df["Customers"]
)
```

What if:

```text
Customers = 0
```

Then:

```text
Revenue / 0
```

causes problems.

Always check denominators.

---

# Mistake 4 — Data Leakage

This is a BIG one.

Suppose we predict:

```text
Will Customer Churn?
```

Target:

```text
Churn
```

Then you create:

```text
Refund_After_Churn
```

But refunds happen after the customer has already churned.

Now the model is secretly seeing future information.

This is called:

```text
Data Leakage
```

The model looks amazing during training and terrible in production.

---

# Mistake 5 — Duplicating Information

Suppose:

```text
Revenue = Price × Quantity
```

Then you create:

```text
Revenue × 2
Revenue × 3
Revenue × 4
```

These features don't add much new information.

They're mostly duplicates.

---

# Mistake 6 — Creating Features Without Testing

Many beginners do:

```text
Create Feature
↓
Keep Feature Forever
```

Wrong.

Correct workflow:

```text
Create Feature
↓
Train Model
↓
Evaluate
↓
Keep or Remove
```

---

# Real Kaggle Example

Imagine:

| Feature            | Useful? |
| ------------------ | ------- |
| Price × Quantity   | ✅       |
| Revenue / Customer | ✅       |
| Area / Rooms       | ✅       |
| Customer_ID × Age  | ❌       |
| ZipCode / Salary   | ❌       |

Why?

Because the first group represents meaningful concepts.

The second group represents random math.

# The 5-Second Test

Before creating a feature, ask:

### Question 1

```text
Can I explain what this feature means?
```

### Question 2

```text
Could it influence the target?
```

### Question 3

```text
Would a business person understand why I created it?
```

If all answers are:

```text
Yes
```

then the feature is worth trying.
