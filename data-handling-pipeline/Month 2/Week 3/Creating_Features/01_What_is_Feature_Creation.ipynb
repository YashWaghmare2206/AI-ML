# **Topic 1: What is Feature Creation?**

## **Subtopic 1.1 — What is a Feature?**

**A feature is simply an input variable used by a machine learning model.**

| Age | Salary | Experience | Purchased |
| --- | ------ | ---------- | --------- |
| 25  | 30000  | 2          | No        |
| 40  | 80000  | 15         | Yes       |

Here:

* Age → Feature
* Salary → Feature
* Experience → Feature

The target variable is:
* Purchased

So:
* Features (X) → Inputs
* Target (y) → Output



## **Subtopic 1.2 — What is Feature Creation?**

Feature Creation means:
* `Creating new useful features from existing features.`

Instead of giving raw data directly to the model, we create additional information that may help the model learn better patterns.

Example

Original:

| Height (m) | Weight (kg) |
| ---------- | ----------- |
| 1.8        | 80          |

Create:
**BMI = Weight / Height²**

New Dataset

| Height | Weight | BMI   |
| ------ | ------ | ----- |
| 1.8    | 80     | 24.69 |

* BMI did not exist before.
* We created it.
* This is Feature Creation.


Often:
* `Feature Engineering > Changing Algorithms`

A well-engineered feature can improve a model more than switching from one algorithm to another.

## **Subtopic 1.3 — How Feature Creation Helps ML Models Learn Better**

**A machine learning model can only learn from the information you give it.**

* `Model ≠ Intelligent Human`
* `Model = Pattern Finder`

The model cannot magically invent information.

If important information is hidden inside the data, we often create a feature to expose it.

### Example

Predict employee salary.

**Original**

| Age | Experience |
| --- | ---------- |
| 25  | 3          |
| 40  | 15         |

The model sees two separate numbers.

But maybe what matters is:
* `Experience Rate = Experience / Age`

| Age | Experience | Experience_Rate |
| --- | ---------- | --------------- |
| 25  | 3          | 0.12            |
| 40  | 15         | 0.375           |

This new feature may reveal a stronger pattern.

### **Why Not Let the Model Figure It Out?**

Some advanced models can.

For example:

* Neural Networks
* XGBoost
* Random Forests

can often discover complex relationships automatically.

But:
* `Explicit Feature + Good Model = Best Results`

### **Real ML Example**

Suppose we have:

| Date       |
| ---------- |
| 2026-06-10 |
| 2026-12-25 |

A model only sees:
* `2026-06-10`
* `2026-12-25`
Not very useful.

Instead we create:

| Day | Month | Weekend |
| --- | ----- | ------- |
| 10  | 6     | False   |
| 25  | 12    | True    |

Now the model can learn:
* Sales increase in December
* Sales increase on weekends
