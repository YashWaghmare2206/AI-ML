# **Topic 4 — Group-Based Features**

## **Subtopic 4.1 — What Are Group-Based Features?**

**Theory**

Instead of looking at a row individually, we look at:

* `Groups of rows`

and create features from the group.

Think:

* Customer Group
* Department Group
* City Group
* Product Group

**Example**

| Employee | Department | Salary |
| -------- | ---------- | ------ |
| A        | IT         | 50000  |
| B        | IT         | 60000  |
| C        | HR         | 40000  |
| D        | HR         | 45000  |

Normally a model only sees:

* Salary
* Department

But we can create:

* Average Salary of Department

**Step 1**

Group by Department:

IT
* 50000
* 60000

Average:
* 55000

| Employee | Dept | Salary | Dept_Avg_Salary |
| -------- | ---- | ------ | --------------- |
| A        | IT   | 50000  | 55000           |
| B        | IT   | 60000  | 55000           |
| C        | HR   | 40000  | 42500           |
| D        | HR   | 45000  | 42500           |

### **Why Is This Useful?**

Now the model knows:

* Not only my salary
* But also
* What is normal for my department

**That extra context is often extremely valuable.**

### **Important Intuition**

Feature interactions answered:

* `How do columns relate?`

Group-based features answer:

* `How does this row compare to its group?`

This distinction is very important.

# Group-Based Features — Complete Overview

The idea is:

```text
Group Rows
↓
Calculate Some Statistic
↓
Assign It Back To Each Row
↓
Create New Feature
```

Example dataset:

| Employee | Department | Salary |
| -------- | ---------- | ------ |
| A        | IT         | 50000  |
| B        | IT         | 60000  |
| C        | HR         | 40000  |
| D        | HR         | 45000  |

---

# 1. Group Mean (Average)

## Theory

Calculate average value within a group.

Formula:

```text
Department Average Salary
```

---

## Example

IT:

```text
(50000 + 60000)/2
=
55000
```

HR:

```text
(40000 + 45000)/2
=
42500
```

New feature:

| Employee | Salary | Dept_Avg_Salary |
| -------- | ------ | --------------- |
| A        | 50000  | 55000           |
| B        | 60000  | 55000           |
| C        | 40000  | 42500           |
| D        | 45000  | 42500           |

---

## Why Useful?

Gives context.

```text
My Salary
vs
Typical Department Salary
```

---

# 2. Group Count

## Theory

Count how many rows belong to a group.

---

## Example

Dataset:

| Customer | City   |
| -------- | ------ |
| A        | Mumbai |
| B        | Mumbai |
| C        | Mumbai |
| D        | Pune   |

Count customers per city.

Mumbai:

```text
3
```

Pune:

```text
1
```

New feature:

| Customer | City   | City_Count |
| -------- | ------ | ---------- |
| A        | Mumbai | 3          |
| B        | Mumbai | 3          |
| C        | Mumbai | 3          |
| D        | Pune   | 1          |

---

## Why Useful?

Tells popularity/frequency.

Examples:

```text
Customers per City

Orders per Customer

Transactions per Product
```

---

# 3. Group Max

## Theory

Find highest value in each group.

---

## Example

IT Salaries:

```text
50000
60000
```

Maximum:

```text
60000
```

HR Salaries:

```text
40000
45000
```

Maximum:

```text
45000
```

Feature:

```text
Dept_Max_Salary
```

---

## Why Useful?

Helps compare with the best performer in the group.

---

# 4. Group Min

## Theory

Find lowest value in each group.

---

## Example

IT:

```text
50000
60000
```

Minimum:

```text
50000
```

HR:

```text
40000
45000
```

Minimum:

```text
40000
```

Feature:

```text
Dept_Min_Salary
```

---

## Why Useful?

Helps measure lower bounds.

Examples:

```text
Lowest Price in Category

Lowest Salary in Department
```

---

# 5. Difference From Group Mean ⭐

This is one of the most useful Kaggle features.

## Theory

Instead of just storing the average:

```text
My Value - Group Average
```

---

## Example

Employee A:

```text
Salary = 50000

Dept Average = 55000
```

Feature:

```text
50000 - 55000
=
-5000
```

Employee B:

```text
60000 - 55000
=
5000
```

Result:

| Employee | Diff_From_Avg |
| -------- | ------------- |
| A        | -5000         |
| B        | 5000          |
| C        | -2500         |
| D        | 2500          |

---

## Why Useful?

Now the model knows:

```text
Above Average?
Below Average?
```

instead of only seeing the raw salary.
