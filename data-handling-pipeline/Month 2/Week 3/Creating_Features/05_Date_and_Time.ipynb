
# Topic 5 — Date & Time Features

## Why?

Models do not understand:

```text
2026-08-15
```

They only see a value.

We extract meaningful information from dates.

---

# 1. Year Feature

## Create

```text
2026-08-15
↓
2026
```

Python:

```python
df["Year"] = df["Date"].dt.year


# `The .dt accessor means:`

"This column contains datetime objects. Let me access date-related properties."
```

## Use

* Long-term trends
* Growth over years

---

# 2. Month Feature

## Create

```text
2026-08-15
↓
8
```

Python:

```python
df["Month"] = df["Date"].dt.month
```

## Use

* Seasonality
* Monthly sales trends

Example:

```text
December → Higher sales
```

---

# 3. Day Feature

## Create

```text
2026-08-15
↓
15
```

Python:

```python
df["Day"] = df["Date"].dt.day
```

## Use

* Monthly billing cycles
* Salary/payment behavior

---

# 4. Day of Week ⭐⭐⭐⭐⭐

## Create

```text
Monday
Tuesday
...
Sunday
```

Python:

```python
df["DayOfWeek"] = df["Date"].dt.dayofweek
```

Result:

```text
Monday = 0
Sunday = 6
```

## Use

Example:

```text
Restaurant sales

Weekend > Weekday
```

---

# 5. Weekend Feature ⭐⭐⭐⭐⭐

## Create

```text
Is_Weekend

0 = No
1 = Yes
```

Python:

```python
df["IsWeekend"] = (
    df["Date"].dt.dayofweek >= 5
).astype(int)
```

## Use

Examples:

```text
Shopping
Food Delivery
Movie Tickets
```

---

# 6. Quarter Feature

## Create

```text
Q1 = Jan-Mar
Q2 = Apr-Jun
Q3 = Jul-Sep
Q4 = Oct-Dec
```

Python:

```python
df["Quarter"] = df["Date"].dt.quarter
```

## Use

Business and finance datasets.

# 7. Holiday Feature

---

# 8. Elapsed Time Features ⭐⭐⭐⭐⭐

Most important date feature.

## Example

Signup Date:

```text
2024-01-01
```

Today:

```text
2026-01-01
```

Create:

```text
Days_Since_Signup = 730
```

Python:

```python
df["DaysSinceSignup"] = (
    today - df["SignupDate"]
).dt.days
```

## Use

```text
Customer Churn
Customer Lifetime Value
Retention
```

---

# 9. Age Feature ⭐⭐⭐⭐

Instead of:

```text
Birth Date
```

Create:

```text
Age
```

Example:

```text
1999-06-01
↓
26 years
```

Python:

```python
df["Age"] = (
    today.year -
    df["BirthDate"].dt.year
)
```

# 10. Cyclical Features ⭐⭐⭐⭐

Problem:

```text
December = 12
January = 1
```

They are close in reality but far numerically.

Use:

```python
month_sin = np.sin(
    2*np.pi*month/12
)

month_cos = np.cos(
    2*np.pi*month/12
)
```

Used in:

```text
Time Series
Forecasting
Weather
Traffic
```

# Common Mistakes 🚨

## Mistake 1

Using raw dates directly.

Bad:

```text
2026-08-15
```

Better:

```text
Year
Month
Day
Weekend
```

---

## Mistake 2

Ignoring seasonality.

Example:

```text
December sales ≠ June sales
```

---

## Mistake 3

Ignoring weekends.

Example:

```text
Restaurant traffic
```

---

## Mistake 4

Data Leakage.

Using future dates to create features.

---

## Mistake 5

Treating cyclical values as normal numbers.

Example:

```text
Month 12 and Month 1
```
