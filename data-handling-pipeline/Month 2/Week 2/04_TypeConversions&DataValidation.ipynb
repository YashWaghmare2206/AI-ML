# Type Conversions & Data Validation

## Topic 1 — df.dtypes + df.info()

Before fixing datatype problems: first we must detect them

1️⃣ Why This Matters

Example:

Salary = "50000"
looks numeric…

but actually it may be:
string (object)

Then: df["Salary"].mean() may fail or behave wrongly.

2️⃣ df.dtypes

df.dtypes

This shows: datatype of every column

```python
# Example

import pandas as pd
from unicodedata import category

df = pd.DataFrame({
    "Age": [22, 25, 30],
    "Salary": [40000, 50000, 60000],
    "Name": ["A", "B", "C"]
})

print(df.dtypes)
```

Meaning

| Type         | Meaning                                 |
| ------------ | --------------------------------------- |
| `int64`      | integer                                 |
| `float64`    | decimal values / NaN often causes float |
| `object`     | usually string/text                     |
| `bool`       | True / False                            |
| `datetime64` | date/time                               |
| `category`   | repeated categories                     |


3️⃣ Why Age Sometimes Becomes Float

Example:

Age = [22, None, 30]

Output type:  float64

because: NaN is treated as float

This confuses many beginners.

4️⃣ df.info()
Concept
df.info()

This gives:

column names | datatype | non-null count | memory usage

Much more detailed than dtypes.

```python
# Example Output
#
# print(df.info())
#
# <class 'pandas.core.frame.DataFrame'>
# RangeIndex: 3 entries
# Data columns:
# Column   Non-Null Count   Dtype
# 0   Age      3 non-null       int64
# 1   Salary   3 non-null       int64
# 2   Name     3 non-null       object
```

5️⃣ Interview-Level Use

First thing many data scientists do:

df.head() | df.info() | df.describe()

before anything else.

This is standard workflow.

6️⃣ Real Problem Example

Suppose:

Salary → object

instead of:

int64

Then you know:

conversion is needed

This is why this topic matters.

## Practice

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, None, 30],
    "Salary": ["40000", "50000", "60000"],
    "Joined": ["2024-01-01", "2024-02-10", "2024-03-15"]
})

print(df.dtypes)
print(df.info())  # No need of print statement here

# Because pandas represents missing values as NaN,
# and NaN is stored as float,
# so the whole column becomes float64.

# None itself is not “float64”
# Pandas converts missing numeric values into:
# NaN → float
# That’s the exact reason.
```

## Topic 2 — .astype() (Basic Type Conversion)

Now we move from:

detect datatype problems to actually fixing them. This is one of the most used pandas operation

1️⃣ Concept

df["column"].astype(new_type)

Meaning: Convert a column from one datatype to another

2️⃣ Common Types

| Type         | Example         |
| ------------ | --------------- |
| `int`        | whole numbers   |
| `float`      | decimal values  |
| `str`        | text            |
| `"category"` | repeated labels |
| `bool`       | True / False    |


category is not a Python variable here

It must be written as a string.


3️⃣ Example — String to Integer

df = pd.DataFrame({
    "Salary": ["40000", "50000", "60000"]
})

Current type: object

Convert:

df["Salary"] = df["Salary"].astype(int)

Now: int64

4️⃣ Example — Integer to String

df["Age"] = df["Age"].astype(str)

Used when:

ID values should be treated as labels

Example:

Pincode
Employee ID
Roll Number

Even if numeric, they are not for math.

5️⃣ Example — Category Type
df["Department"] = df["Department"].astype("category")

Used for:

IT | HR | Finance | Marketing

Benefits:

memory efficient | better grouping | cleaner ML preprocessing

Very useful.

6️⃣ Important Danger

This can fail:

df["Salary"] = df["Salary"].astype(int)

if data contains:

"abc"
"?"
"missing"

Then:

Error occurs

That is why later we learn:

pd.to_numeric()

which is safer.

7️⃣ Example of Failure

df = pd.DataFrame({
    "Salary": ["40000", "abc", "60000"]
})

This fails:

df["Salary"].astype(int)

because:

"abc" is not numeric

8️⃣ Interview-Level Rule

Use:  astype()

when:  data is already clean

Use:  pd.to_numeric()

when:  data may be dirty

Very important interview answer.

### Practice Question

```python
import pandas as pd

df = pd.DataFrame({
    "Age": ["22", "25", "30"],
    "Salary": ["40000", "50000", "60000"],
    "Department": ["IT", "HR", "IT"]
})

df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].astype(float)
df["Department"] = df["Department"].astype("category")




```

## Topic 3 — pd.to_numeric() + pd.to_datetime()

This is one of the most important real-world preprocessing topics.

Because: real datasets are messy

and: astype()

often fails.

### Part A — pd.to_numeric()

1️⃣ Concept

pd.to_numeric(column)

Meaning:

Convert values to numeric safely

Used for:

int | float | mixed numeric strings

```python
# 2️⃣ Example

import pandas as pd

df = pd.DataFrame({
    "Salary": ["40000", "50000", "60000"]
})

# Convert:

df["Salary"] = pd.to_numeric(df["Salary"])
print(df["Salary"])


# Now:

# numeric type
```

3️⃣ Why Better Than astype()?

Because it supports:

errors="coerce"

This is VERY important.

```python
# 4️⃣ Example with Dirty Data
import pandas as pd

df = pd.DataFrame({
    "Salary": ["40000", "abc", "60000"]
})

# This fails:
# astype(int)

# But this works:

df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

# 5️⃣ What errors="coerce" Does
#
# It converts invalid values into:
# NaN

# Example:
# "abc" → NaN
#
# This is powerful because:
# then we can handle missing values normally
#
# Very important ML workflow.
```

### Part B — pd.to_datetime()

6️⃣ Concept

pd.to_datetime(column)

Meaning:

Convert strings into proper datetime format

Used for:

joining dates | DOB | transaction dates | timestamps

Very common.


```python
# 7️⃣ Example
import  pandas as pd

df = pd.DataFrame({
    "Joined": [
        "2024-01-01",
        "2024-02-10",
        "2024-03-15"
    ]
})

# Convert:

df["Joined"] = pd.to_datetime(df["Joined"])

# Now:

# datetime64

## Imp Note

# Before converting in datetime64 the date should be in any of the following format ..these are commonly accepted

"2025-04-26"
"26-04-2025"
"04/26/2025"
"26/04/2025"
"2025/04/26"
"April 26, 2025"
"26 Apr 2025"
"2025-04-26 14:30:00"
```

8️⃣ Invalid Dates

Example:

"2024-13-50"

invalid date.

Use:

errors="coerce"

Again:

invalid date → NaT

(NaT = missing datetime value)

Very important.

9️⃣ Interview-Level Rule

to_numeric → dirty numbers

to_datetime → dirty dates

astype → clean data only

This is a strong interview answer.

## Practice

```python
import pandas as pd

df = pd.DataFrame({
    "Age": ["22", "25", "abc"],
    "Salary": ["40000", "50000", "?"],
    "Joined": ["2024-01-01", "wrong_date", "2024-03-15"]
})

df["Age"] = pd.to_numeric(df["Age"] , errors="coerce")
df["Salary"] = pd.to_numeric(df["Salary"] , errors="coerce")
df["Joined"] = pd.to_datetime(df["Joined"] , errors="coerce")

```

## Topic 4 — Impossible Value Checks (Data Validation)

```python
# This is where preprocessing becomes:
#
# 🔥 Professional-level
#
# Because not all bad data is missing.
#
# Sometimes values exist…
#
# but they are logically impossible.
#
# Example:
#
# Age = -5
# Salary = 0
# Experience = -2
# DOB = future date
#
# These must be detected and handled.
#
# Very important in real ML projects.
```

1️⃣ What is Data Validation?

It means:

checking whether values make sense

Not just:

is value present?

but:

is value valid?

This is a huge difference.

2️⃣ Common Impossible Values

Age

Age < 0

Age > 100 or 120


Salary

Salary <= 0

Usually invalid.
(Depends on business logic)

Experience

Experience < 0 |
Experience > Age

Very important check.

Example:

Age = 20 |
Experience = 25

Impossible.

Dates
DOB in future | Joining date before birth

```python
# 3️⃣ Example Dataset
import pandas as pd

df = pd.DataFrame({
    "Age": [25, -3, 30, 150],
    "Salary": [40000, 0, 50000, -1000],
    "Experience": [2, 3, -1, 200]
})
# many values are invalid
```

4️⃣ Detect Negative Age

print(df[df["Age"] < 0])

This filters: bad rows only
Very useful.

5️⃣ Detect Salary ≤ 0

print(df[df["Salary"] <= 0])
Common real-world check.

6️⃣ Detect Invalid Experience

print(df[df["Experience"] < 0])

and stronger: print(df[df["Experience"] > df["Age"]])

Very important.

```python
# 7️⃣ What To Do After Detection?
#
# Options:
# Replace with NaN
#
# Best common practice:
# df.loc[df["Age"] < 0, "Age"] = pd.NA
#
# Then handle using:
# fillna()
#
# This is often best.
```

```python
# Remove rows
#
# Used when:
# row is clearly unusable
#
# Cap values
#
# Used when:
# value is extreme but still fixable
#
# Example:
# Age = 120
```

8️⃣ Interview-Level Rule

Best practice:

Convert invalid values → NaN

then handle like missing values

This is a very strong interview answer.

### Practice

```python
import  pandas as pd

df = pd.DataFrame({
    "Age": [22, -5, 30, 120],
    "Salary": [30000, 0, 50000, -2000],
    "Experience": [1, 2, -1, 40]
})

print(df[df["Age"] < 0])
print(df[df["Salary"] <= 0])
df.loc[df["Experience"] < 0 , "Experience"] = pd.NA
print(df)

print(df[df["Experience"] > df["Age"]])

# Why is converting invalid values to NaN often better than direct deletion?
# Converting invalid values to NaN allows us to handle them systematically
# using missing-value methods like fillna(), interpolation, or model preprocessing,
# instead of silently keeping wrong values.
```

## Topic 5 — df.describe() + Sanity Checks

This is one of the fastest ways to understand a dataset.

Before modeling, many data scientists immediately run:

df.describe()

because it quickly reveals: suspicious values | without writing many filters.

Very important.

1️⃣ Concept

df.describe()

This gives summary statistics for: numerical columns

```python
# 2️⃣ Example

import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, 30, 120],
    "Salary": [30000, 35000, 50000, 5000000],
    "Experience": [1, 2, 3, 40]
})

print(df.describe())
```

3️⃣ Output Meaning

| Metric | Meaning            |
| ------ | ------------------ |
| count  | non-null values    |
| mean   | average            |
| std    | standard deviation |
| min    | minimum            |
| 25%    | Q1                 |
| 50%    | median             |
| 75%    | Q3                 |
| max    | maximum            |


4️⃣ How We Use It

Suppose: Age max = 120

Immediately suspicious.

Suppose: Salary max = 5000000

Likely outlier.

Suppose: Salary min = -5000

Impossible value.

This is: sanity checking

5️⃣ Quick Validation Example

print(df.describe())

and ask:

1. Does min make sense?
2. Does max make sense?
3. Does mean look realistic?

This is exactly what real analysts do.

6️⃣ For Categorical Columns

describe() default focuses on numeric.

For all columns:

df.describe(include="all")

Now it also shows:

1. unique
2. top
3. freq

for text/category columns.

Very useful.

7️⃣ value_counts() (Important Partner)

For checking category balance:

df["Department"].value_counts()

Used for:

* How many IT?
* How many HR?

Very common.

8️⃣ Interview-Level Workflow

First steps:

1. df.head()
2. df.info()
3. df.describe()

This is standard professional workflow.

Interview favorite.

### Practice

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, 30, 120],
    "Salary": [30000, 0, 50000, 5000000],
    "Department": ["IT", "HR", "IT", "Finance"]
})

print(df.describe())
df.describe(include="all")  # This include the category values also df.describes() show jst numeric values
print(df["Department"].value_counts())

# ⚠ Q5 — Why is min useful? it comes in df.describe()
# min helps detect impossible or suspicious values like
# negative salary, age = -5, or salary = 0.
# It helps validate whether the lowest value is logically valid.
```
