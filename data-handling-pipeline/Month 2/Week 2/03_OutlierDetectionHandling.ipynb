```python
# 1️⃣ What is an Outlier?
# An outlier is:
# A value that is unusually far from the rest of the data
# Example:
# [10, 12, 11, 13, 15, 200]
# Here:
# 200
# is likely an outlier. Because all others are near: 10–15
```

2️⃣ Why Outliers Matter

They can badly affect:

Mean
Standard deviation
Linear Regression
K-Means Clustering
Distance-based models

Example:

Salary average becomes unrealistic

because of one huge value.

3️⃣ IQR Concept

IQR means:

Interquartile Range

Formula:

IQR=Q3−Q1

Where:

Q1 = 25th percentile
Q3 = 75th percentile

Meaning:

Middle 50% of the data

4️⃣ Outlier Boundaries
Formula:

Lower Bound
Lower Bound=Q1−1.5×IQR

Upper Bound
Upper Bound=Q3+1.5×IQR

Outside these bounds = Outlier

5️⃣ Example Dataset
salary = [10, 12, 14, 15, 18, 20, 22, 100]
Clearly: 100 looks suspicious

6️⃣ Step-by-Step

Step A — Find Q1 and Q3
Suppose:
Q1 = 12.5
Q3 = 21

Step B — Calculate IQR
IQR=21−12.5=8.5

Step C — Lower Bound
12.5−1.5×8.5=−0.25

Step D — Upper Bound
21+1.5×8.5=33.75

Step E — Check Values
Since:
100 > 33.75
it is an: 🔥 Outlier

```python
# 7️⃣ Pandas Code
import pandas as pd

df = pd.DataFrame({
    "Salary":[10,12,14,15,18,20,22,100]
})

Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Salary"] < lower_bound) |
    (df["Salary"] > upper_bound)
    ]

print(outliers)
```

Why IQR is Popular

Because:
It works well even when data is not normally distributed
That makes it better than Z-score in many practical cases.

## Topic 2.2.1 — Removing Outliers using IQR

1️⃣ Concept

After finding:
Lower Bound
Upper Bound

we remove rows outside that range.

Rule:
Keep only values inside the bounds

```python
import pandas as pd
# 2️⃣ Example Dataset
df = pd.DataFrame({
    "Salary":[10,12,14,15,18,20,22,100]
})
#
# Clearly:
# 100
# is suspicious.

# 3️⃣ Step-by-Step Code

# Step A — Calculate Q1 and Q3
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

# Step B — Calculate IQR
IQR = Q3 - Q1

# Step C — Find Bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Step D — Remove Outliers
clean_df = df[
    (df["Salary"] >= lower_bound) &
    (df["Salary"] <= upper_bound)
    ]
print(clean_df)
```

## 2.2.2 — Capping / Winsorization

Instead of deleting outliers:
remove the row completely

we do:
🔥 Keep the row, but limit the extreme value

This is called:
Capping (Winsorization)

Very important in real ML pipelines.

1️⃣ Why Capping?

Suppose:

Salary = 5000000

If we remove it:

we lose an entire row

Maybe that row contains useful:

Age
Department
Experience
Target variable

So instead:

we keep the row
but reduce the extreme impact

2️⃣ Concept

Using IQR bounds:

If value > upper bound
replace with upper bound

If value < lower bound
replace with lower bound

This is:

Winsorization

3️⃣ Example Dataset
df = pd.DataFrame({
    "Salary":[10,12,14,15,18,20,22,100]
})

Here: 100 is an outlier.

Suppose:

Upper Bound = 33.75

Instead of removing: 100

we replace it with:
33.75

```python
# 4️⃣ Code

import pandas as pd

df = pd.DataFrame({
    "Salary":[10,12,14,15,18,20,22,100]
})

Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df["Salary"] = df["Salary"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print(df)
```

```python
df["Salary"] = df["Salary"].clip(
    lower=lower_bound,
    upper=upper_bound
)
#
# What clip() Does
# clip(lower, upper)
#
# means:
# force every value to stay inside this range
#
# Rule
# If value < lower
# replace with:
# lower
#
# If value > upper
# replace with:
# upper

# If value is normal keep it unchanged
```

5️⃣ Result

Before: 100

After: 33.75

Row stays.
Only extreme effect is reduced.

6️⃣ Why This Is Better

| Remove           | Cap            |
| ---------------- | -------------- |
| Lose rows        | Keep rows      |
| Risk losing info | Safer          |
| Simple           | More practical |

Usually:

🔥 Capping > Removing for business datasets.

8️⃣ Interview Question

Very common interview question.

## 2.2.3 — Log Transform

This is one of the most powerful techniques for handling:

🔥 Highly Skewed Data

Highly skewed data means:
the data is not balanced evenly
Most values are concentrated on one side, while a few extreme values stretch the distribution toward the other side.

1️⃣ What is Skewed Data?

Example:

Salary = [30000, 35000, 40000, 45000, 5000000]

Most values are small.

One value is extremely large.

This creates:
Right Skew (Positive Skew)

Graph looks like:
■■■■■■■■■────────────►

Long tail on the right.

2️⃣ Why This Is a Problem

Skewed data affects:

Mean |
Standardization |
Linear Regression |
Distance-based models |
Clustering

and makes models unstable.

3️⃣ Solution — Log Transform

We apply:

logarithm to compress large values more than small values.

This reduces skewness.

4️⃣ Formula

y = log(x)  OR  y = log(1 + x)

Mostly used is y = log(1 + x) as log(0) is invalid

used is np.log1p() which means log(1 + x)

```python
# 5️⃣ Example

# Before: 1000 , 2000 , 3000 , 5000000

import numpy as np
np.log1p([1000,2000,3000,5000000])

# After : 6.9 , 7.6 , 8.0 , 15.4

# Notice:
# Huge gap becomes smaller
# That helps ML models a lot.
```

```python
# 6️⃣ Code Example

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Salary":[30000,35000,40000,45000,5000000]
})

df["Log_Salary"] = np.log1p(df["Salary"])

print(df)
```

7️⃣ Important Rule

Use log transform for:

✔ Salary
✔ Income
✔ House price
✔ Followers
✔ Sales
✔ Population

Avoid for:

❌ Negative values
❌ Categorical data
❌ Small bounded features
