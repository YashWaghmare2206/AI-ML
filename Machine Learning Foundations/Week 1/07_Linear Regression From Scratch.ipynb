# Entire Training Pipeline

```text
Dataset
   ↓
Initialize w,b
   ↓
Predict
   ↓
Calculate Cost
   ↓
Calculate Gradients
   ↓
Update w,b
   ↓
Repeat
   ↓
Converged
   ↓
Trained Model
```

This is the complete learning process of Linear Regression.

---

# What Does "Training" Actually Mean?

A lot of beginners think:

```text
Training = Learning Rules
```

More precisely:

```text
Training
=
Finding optimal parameters
```

For Linear Regression:

```text
Training
=
Finding best w and b
```

That's all.
