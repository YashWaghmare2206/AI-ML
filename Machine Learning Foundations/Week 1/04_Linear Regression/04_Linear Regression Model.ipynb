## Definition

Linear Regression assumes a **linear relationship** between input features and the target variable.

For a single feature:

## Why "Linear"?

Because the prediction is a linear combination of the inputs.



---

# Understanding the Components

## 1. Input Feature ((x))

The independent variable.

Example:

| Area (sq ft) |
| ------------ |
| 1000         |
| 1200         |
| 1500         |

Here:

**`x = Area`**

---

## 2. Weight ((w))

Determines how strongly the feature affects the prediction.



---

### Interpretation

If:

**`w > 0`**

Positive relationship.

Increase (x) → Increase (y)

---

If:

**`w < 0`**

Negative relationship.

Increase (x) → Decrease (y)

---

If:

**`w = 0`**

Feature has no effect.

---

## 3. Bias ((b))

The intercept.

Represents the prediction when:

**`x = 0`**

# Goal of Linear Regression

Find values of:

**`w`**

and

**`b`**

such that prediction error is minimized.

In other words:



through the data.

---

# How Are w and b Found?

We do not choose them manually.

The algorithm learns them from data.

Process:

```text
Initialize w and b
        ↓
Make Predictions
        ↓
Calculate Error
        ↓
Adjust w and b
        ↓
Repeat
```

This adjustment process is called:

### Gradient Descent

which you'll study in Topics 6 and 7.

### Why is it called "fitting a line"?

Suppose your data is:

| Area (x) | Price (y) |
| -------- | --------- |
| 1        | 5         |
| 2        | 7         |
| 3        | 9         |
| 4        | 11        |

If we plot these points on a graph, they look like they lie on a straight line.

which corresponds to:



The question becomes:

> Which values of (w) and (b) make this line pass as close as possible to the data points?

That process is called **fitting the line to the data**.

---

### What does "best fitting" mean?

Real datasets are noisy.

Example:

| Area | Price |
| ---- | ----- |
| 1    | 5     |
| 2    | 8     |
| 3    | 8     |
| 4    | 12    |

No single straight line passes through every point.

So we search for the line whose predictions have the **smallest overall error**.

Example:



Both are possible lines.

We compute errors for each line and choose the one with the lower cost (MSE).

---

### What is the model actually learning?

When training Linear Regression, we are not learning the line itself.

We are learning:

**`w`**

and

**`b`**

Because once (w) and (b) are known, the line is completely defined.



This is the learned model.

---

### Why a line specifically?

Because Linear Regression assumes:

> The relationship between input and output is approximately linear.

For example:

* Area ↑ → Price ↑
* Experience ↑ → Salary ↑

A straight line is often a reasonable first approximation.

Later you'll learn:

* Polynomial Regression → curve fitting
* Decision Trees → rule fitting
* Neural Networks → highly complex function fitting

But all of them are still doing the same thing:

> Find a mathematical function that best matches the data.

For Linear Regression, that function happens to be a **straight line**.

So whenever you hear:

> "Linear Regression fits a line to the data"

read it as:

> "Linear Regression finds the values of (w) and (b) that make the line (y=wx+b) have the smallest prediction error on the dataset."
