# Week 1 → Topic 6: Gradient Descent

Now we know:



The next question is:

> How do we actually find the values of (w) and (b) that minimize the cost?

The answer is **Gradient Descent**.

---

## Core Idea

Think of training as:

```text
Current Line
     ↓
Calculate Cost
     ↓
Move Line Slightly
     ↓
Cost Decreases
     ↓
Repeat
```

Gradient Descent is the algorithm that decides:

```text
Which direction to move?
How much to move?
```

---

## Cost Surface

Every possible pair of:

**`(w,b)`**

produces a cost:

**`J(w,b)`**

Imagine:

```text
w,b values
      ↓
Cost Function
      ↓
Cost
```

This creates a surface.

For Linear Regression, that surface looks like a bowl.

```text
      .
    .   .
  .       .
 .         .
.     ↓     .
 .         .
  .       .
    .   .
      .
```

The bottom of the bowl is:

```text
Minimum Cost
```

That is where we want to reach.

---

## What Gradient Descent Does

Start somewhere on the bowl.

```text
Random w,b
```

Then repeatedly move downhill.

```text
High Cost
     ↓
Lower Cost
     ↓
Lower Cost
     ↓
Minimum Cost
```

Until the bottom is reached.

---

## What Is a Gradient?

From your March Calculus month:

A derivative tells us:

> How rapidly something changes.

The gradient tells us:

> Which direction causes the largest increase.

For cost:

**`J(w,b)`**



These tell us:

* How cost changes when (w) changes
* How cost changes when (b) changes

---

## Why Move in the Negative Direction?

Gradient points toward:

```text
Maximum Increase
```

But we want:

```text
Maximum Decrease
```

Therefore:



(the negative gradient)

points downhill.

## Learning Rate

The learning rate is critical.



determines how large each step is.

---

### Too Small

```text
Tiny Steps
```

```text
Slow Learning
```

May require thousands of iterations.

---

### Too Large

```text
Huge Steps
```

Can jump over the minimum.

```text
Left
Right
Left
Right
```

and never converge.

---

### Good Learning Rate

Moves efficiently toward the minimum.

## Training Process

Initialize:



randomly.

Then:

```text
Predict
   ↓
Compute Cost
   ↓
Compute Gradients
   ↓
Update Parameters
   ↓
Repeat
```

This loop is called an **iteration**.

---

## Convergence



Updates become very small.

Cost stops decreasing significantly.

We say:

```text
Model Converged
```

## Key Takeaway

Gradient Descent is an optimization algorithm that minimizes the cost function by repeatedly updating:



It does **not** measure error.

The **Cost Function** measures error.

Gradient Descent uses that information to move toward the values of (w) and (b) that produce the lowest possible error.
