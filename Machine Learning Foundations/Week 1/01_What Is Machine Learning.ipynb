### Week 1 → Topic 1: What Is Machine Learning

#### Formal Definition

Machine Learning (ML) is a field of AI that develops algorithms capable of learning patterns from data and improving performance on a task without being explicitly programmed with task-specific rules.

A classic definition by Tom M. Mitchell:

> A computer program is said to learn from experience **E** with respect to some task **T** and performance measure **P** if its performance at **T**, as measured by **P**, improves with experience **E**.

---

## Why Machine Learning Exists

Traditional programming works like:

```
Rules + Data
      ↓
   Output
```

Example:

```python
if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

The programmer explicitly writes the rules.

---

For many real-world problems, writing rules is impossible.

Examples:

* Detect spam emails
* Predict house prices
* Recognize faces
* Translate languages
* Recommend videos

There may be millions of possible patterns.

Instead of manually writing rules, we let the algorithm learn them from data.

```
Data + Correct Answers
          ↓
   Learning Algorithm
          ↓
        Model
```

The learned model contains the rules.

## What Is Actually Learned?

A model learns **parameters**.

Example:

**`y = wx + b`**

The model learns:

* (w) (weight)
* (b) (bias)

from data.

The entire training process is basically:

```
Find parameters
that minimize prediction error.
```

This idea will appear again in:

* Linear Regression
* Logistic Regression
* Neural Networks
* Deep Learning

## Training vs Inference

### Training

Learning parameters from data.

```
Data
 ↓
Training Algorithm
 ↓
Learned Parameters
```

Usually computationally expensive.

---

### Inference

Using trained parameters to make predictions.

Example:

```python
price = model.predict(new_house)
```

Fast operation.
