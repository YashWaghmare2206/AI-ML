```python
import numpy as np
import pandas as pd
```

```python
# Dataset

x = np.array([1,2,3,4,5])
y = np.array([5,7,9,11,13])
```

```python
# Initialization

w = 0
b = 0
learning_rate = 0.01
itr = 1000
m = len(x) # Number of rows or data points in total
```

```python
def predict(w, x, b):
    return w * x + b
```

```python
def J_func(y_hat, y, m):
    return np.sum((y_hat - y) ** 2) / (2 * m)
```

```python
def gradient_w(y_hat, y, x, m):
    return np.sum((y_hat - y) * x) / m

def gradient_b(y_hat, y, m):
    return np.sum(y_hat - y) / m
```

```python
cost_history = []

for i in range(itr):

    y_hat = predict(w , x , b)
    J = J_func(y_hat , y , m)

    if i <= 100:    # Storing starting 100 iterations cost
        cost_history.append(J)

    d_w = gradient_w(y_hat , y , x , m)
    d_b = gradient_b(y_hat , y , m)

    w = w - learning_rate * d_w
    b = b - learning_rate * d_b


print(f"w : {w} ")
print(f"b : {b} ")
```

```python
import matplotlib.pyplot as plt

plt.plot(range(0 , len(cost_history)) , cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Cost vs Iterations")
plt.show()
```

## **THe above is the way to check gradient descent Convergence**
