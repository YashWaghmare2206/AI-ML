# Week 1 → Topic 2: Supervised vs Unsupervised Learning

This is one of the most important distinctions in Machine Learning.

The difference is based on whether the dataset contains the **correct answer (target variable)**.

---

## Supervised Learning

In supervised learning, the dataset contains:

**`(X, Y)`**

where:

* (X) = features/input
* (Y) = target/output (ground truth)

The model learns a mapping:



---

### Example

House Price Dataset

| Area | Bedrooms | Price |
| ---- | -------- | ----- |
| 1000 | 2        | 50L   |
| 1200 | 3        | 60L   |
| 1500 | 3        | 75L   |

Features:

**`X = [Area, Bedrooms]`**

Target:

**`Y = Price`**

The model learns:

**`Price = f(Area, Bedrooms)`**

Because the correct answer (Price) is known, this is supervised learning.

---

## Goal of Supervised Learning

Learn a function that generalizes well to unseen data.

Not:

```text
Memorize training data
```

But:

```text
Learn patterns from training data
```

so future predictions are accurate.

---

# Types of Supervised Learning

There are two major categories.

---

## 1. Regression

Target variable is continuous.

Examples:

| Input          | Output           |
| -------------- | ---------------- |
| House Features | ₹75,00,000       |
| Weather Data   | 34.6°C           |
| Engine Data    | Fuel Consumption |

Output can be any real number.

Examples:

```text
21.5
150.8
4500000
```

---

### Mathematical View



The target belongs to the set of real numbers.

---

## 2. Classification

Target variable is categorical.

Examples:

| Input          | Output               |
| -------------- | -------------------- |
| Email          | Spam                 |
| Medical Report | Disease / No Disease |
| Image          | Cat / Dog            |

Possible outputs:

```text
Spam
Not Spam
```

or

```text
0
1
```

or

```text
Cat
Dog
Bird
```

---

### Mathematical View



Finite categories.

---

# Unsupervised Learning

In unsupervised learning:

Only inputs are available.

X

No target variable.

No correct answers.

Dataset:

| Age | Income |
| --- | ------ |
| 20  | 30000  |
| 21  | 32000  |
| 55  | 100000 |
| 57  | 110000 |

No label column.

---

## Goal

Discover hidden structure in data.

Instead of:

```text
Predict Y
```

we try to:

```text
Find patterns
```

---

# Main Types of Unsupervised Learning

---

## 1. Clustering

Group similar observations together.

Example:

Customer Data

| Age | Income |
| --- | ------ |
| 22  | 30k    |
| 25  | 35k    |
| 55  | 120k   |
| 58  | 130k   |

Algorithm may discover:

```text
Cluster 1 → Young Customers

Cluster 2 → High Income Customers
```

without being told these groups exist.

---

### Common Algorithms

* K-Means
* Hierarchical Clustering
* DBSCAN

You already studied K-Means during Feature Engineering.

---

## 2. Dimensionality Reduction

Reduce number of features while preserving information.

Example:

100 features

↓

10 features

Common algorithms:

* PCA
* t-SNE
* UMAP

Useful for:

* Visualization
* Compression
* Noise reduction

---

# Key Difference

### Supervised

Dataset:

**`(X,Y)`**

Goal:

**`Predict Y`**

Examples:

* House price prediction
* Spam detection
* Disease prediction

---

### Unsupervised

Dataset:

**`X`**

Goal:

**`Find hidden structure`**

Examples:

* Customer segmentation
* Pattern discovery
* Data compression

---

# Why Supervised Learning Usually Performs Better

Because it has access to the correct answers.

Think mathematically:

Supervised learning optimizes:



The algorithm knows when it is wrong.

Unsupervised learning has no ground truth labels, making the task harder.
