# 3. Mathematics

## Dataset



---

For binary classification we actually want:

**`P(y=1|x)`**

This is a conditional probability.

Meaning:

Given feature vector (x),

what is probability that class is 1?

---

Example



---

This shift is important:

Regression predicts values.

Classification predicts probabilities.

---

# Why Probability?



---

Therefore classification gives:

# 4. Python Implementation

Binary labels:

```python
import numpy as np

X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

y = np.array([0, 0, 0, 1, 1])

print(X.shape)
print(y.shape)
```

Output:

```python
(5,1)
(5,)
```

---

Check classes:

```python
np.unique(y)
```

Output:

```python
array([0,1])
```

This is a binary classification dataset.

---

Count each class:

```python
unique, counts = np.unique(y, return_counts=True)

print(dict(zip(unique, counts)))
```

Output:

```python
{0:3, 1:2}
```
