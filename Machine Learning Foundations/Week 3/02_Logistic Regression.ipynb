# 1. Theory

Logistic Regression is a classification algorithm used mainly for:



Binary classification.

Despite its name, **it is not a regression algorithm**.

It is a classification algorithm that predicts probabilities.

---

Input:

**`x`**

Output:

**`P(y=1|x)`**

a probability between 0 and 1.

---

Example:

Student Exam Dataset

| Hours Studied | Pass |
| ------------- | ---- |
| 1             | 0    |
| 2             | 0    |
| 3             | 0    |
| 4             | 1    |
| 5             | 1    |



meaning:

92% probability of passing.

---

# 2. Intuition



---



---

So the idea behind Logistic Regression is:

### Step 1

Compute the linear score:



Exactly the same as Linear Regression.

---

### Step 2

Pass that score through a function that squeezes everything into:

**`(0,1)`**

This function is called the Sigmoid Function.

---



---

So Logistic Regression is basically:

```text
Linear Regression
        ↓
    Score z
        ↓
 Sigmoid Function
        ↓
 Probability
```

---

# Geometric Interpretation

You learned in Linear Regression:



defines a hyperplane.

---



---

For (n) features:

Hyperplane.

---

Logistic Regression uses this hyperplane as the separator between classes.

So the real goal is:

Find the hyperplane that best separates class 0 and class 1.

# Statistical Interpretation

This is where Logistic Regression becomes beautiful.

Instead of predicting:

**`y`**

we model:

**`P(y=1|x)`**

---

Meaning:

For every point in feature space,

we assign a probability.

---

Example:

| x | Probability |
| - | ----------- |
| 1 | 0.02        |
| 2 | 0.10        |
| 3 | 0.45        |
| 4 | 0.88        |
| 5 | 0.98        |

---

probability smoothly changes from:



This smooth transition is exactly what Sigmoid provides.

# 4. Python Implementation

Suppose:



---

Compute score:

```python
import numpy as np

x = 4

w = 2
b = -5

z = w*x + b

print(z)
```

Output:

```python
3
```

---

Currently:

[
z=3
]

Not a probability.

---

Later we'll convert it using Sigmoid.

---

Vectorized version:

```python
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

w = np.array([2])
b = -5

z = X @ w + b

print(z)
```

Output:

```python
[-3 -1  1  3  5]
```

These are scores.

Not probabilities yet.
