# 4. Python Implementation

### Single Value

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print(sigmoid(0))
```

Output:

```python
0.5
```

---

### Multiple Values

```python
z = np.array([-5,-2,0,2,5])

print(sigmoid(z))
```

Output:

```python
[0.0067 0.1192 0.5 0.8808 0.9933]
```

---

### Derivative

```python
def sigmoid_derivative(z):
    s = sigmoid(z)
    return s*(1-s)
```
