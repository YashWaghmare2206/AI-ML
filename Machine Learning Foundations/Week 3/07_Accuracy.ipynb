> **Accuracy is the proportion of predictions that are correct.**

# 4. Python Implementation

Without sklearn

```python
actual = [1,0,1,1,0]

predicted = [1,0,1,0,0]

correct = 0

for a,p in zip(actual,predicted):
    if a==p:
        correct += 1

accuracy = correct/len(actual)

print(accuracy)
```

Output

```python
0.8
```

---

Using NumPy

```python
import numpy as np

actual = np.array([1,0,1,1,0])

predicted = np.array([1,0,1,0,0])

accuracy = np.mean(actual==predicted)

print(accuracy)
```

Output

```python
0.8
```

---

Using sklearn

```python
from sklearn.metrics import accuracy_score

accuracy_score(actual,predicted)
```

Output

```python
0.8
```

# 5. Is Accuracy Always Good?

No.

This is one of the biggest misconceptions in Machine Learning.

Let's see why.

---

## Example

Suppose you're building a model to detect a rare disease.

Dataset

1000

patients.

Only

990

---

**Everyone is healthy.**

```text
Healthy
Healthy
Healthy
...
Healthy
```

Accuracy



Amazing?

No.

The model completely failed to detect **every sick patient**.

---

This shows a major limitation.

A model can have **99% accuracy** and still be useless.

---

# When Accuracy Works Well

Accuracy is useful when the dataset is **balanced**.

Example

Dogs

50%

Cats

50%

Now

95%

accuracy actually means the model is genuinely performing well.

---

# When Accuracy Fails

Accuracy becomes misleading when classes are **imbalanced**.

Example

Fraud Detection

99.9%

transactions are genuine.

Only

0.1%

are fraud.

A model predicting

"Genuine"

for every transaction gets almost

100%

accuracy.

Yet it catches

zero frauds.

---

# Why Do We Need Other Metrics?

Accuracy tells us

> "How often is the model correct?"

It does **not** tell us

* How many positive cases it found.
* How many false alarms it made.
* How many positives it missed.

To answer these questions,

we need

* Precision
* Recall
* F1 Score

All of these are built using the **Confusion Matrix**.
