# 06. Overfitting

This is one of the most important concepts in ML.

---

# Theory

**Overfitting** occurs when a model learns:

* Actual patterns ✅
* Noise and random fluctuations ❌

from the training data.

As a result:

* Training performance becomes very good.
* Validation/Test performance becomes poor.

---

# Intuition

Imagine a student.

Training questions:

```text
2+2 = 4
3+5 = 8
7+1 = 8
```

Instead of learning addition, the student memorizes every answer.

---

Exam:

```text
4+6 = ?
```

Student fails.

The student memorized rather than learned.

That is overfitting.

# How to Detect Overfitting

Look at training and validation performance.

### Example

Training Accuracy:

```text
99%
```

Validation Accuracy:

```text
75%
```

Huge gap.

Likely overfitting.

---

For regression:

Training MSE:

```text
1
```

Validation MSE:

```text
50
```

Again, overfitting.

---

# Train vs Validation Error

Typical overfitting pattern:

```text
Training Error    Very Low
Validation Error  High
```

The gap is the important clue.

---

# Why It Happens

Common reasons:

### 1. Model Too Complex

Example:

Degree 20 polynomial for 30 samples.

---

### 2. Too Many Features

1000 features.

Only 100 training examples.

---

### 3. Too Little Data

Model doesn't have enough examples to learn general patterns.

---

### 4. Noise in Dataset

Random fluctuations get memorized.

---

# Learning Curve View

As training continues:

Training error:

```text
Keeps decreasing
```

Validation error:

```text
Decreases
then
Starts increasing
```

When validation error starts rising, overfitting is beginning.

---

# How to Reduce Overfitting

### 1. More Training Data

More examples make memorization harder.

---

### 2. Fewer Features

Remove unnecessary features.

---

### 3. Lower Polynomial Degree

Degree 20 → Degree 3

---

### 4. Regularization

Penalize large weights.

We'll cover this soon.

---

### 5. Early Stopping

Stop training when validation performance stops improving.

Mostly used in deep learning.

---

---

# Important Insight

A model can have:

```text
99% training accuracy
```

and still be a bad model.

What matters is:

```text
Validation/Test Performance
```

because those represent unseen data.

---

# What You Should Remember

1. Overfitting = learning noise along with patterns.
2. Training performance becomes very good.
3. Validation/Test performance becomes worse.
4. Usually caused by excessive model complexity.
5. Polynomial degree that's too high often causes overfitting.
6. More data, simpler models, and regularization help reduce overfitting.
7. Low training error alone does NOT mean the model is good.

## **Important**

```text id="0k0k2i"
Underfitting
------------
Model complexity < Data complexity

Cause:
- Degree too low

Result:
- Cannot learn the true pattern
- High training error
- High validation error

Example:
True pattern: y = x²
Model: y = wx + b


Good Fit
--------
Model complexity ≈ Data complexity

Result:
- Learns the actual pattern
- Low training error
- Low validation error
- Generalizes well to new data

Example:
Learns y ≈ x²


Overfitting
-----------
Model complexity > Data complexity

Cause:
- Degree too high
- Too little training data
- Too much noise

Result:
- Learns the pattern
- Also memorizes noise and training examples
- Very low training error
- High validation error

Example:
Learns y ≈ x² + training data quirks


Quick Memory Trick
------------------

Underfit:
"I can't learn the pattern."

Good Fit:
"I learned the pattern."

Overfit:
"I learned the pattern and memorized the noise."
```
