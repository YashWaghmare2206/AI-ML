# 05. Feature Engineering

This is where ML starts becoming less about algorithms and more about understanding data.

A simple model with good features often beats a complex model with poor features.

---

# Theory

**Feature Engineering** is the process of creating new features from existing data to help the model learn better patterns.

You are not changing the algorithm.

You are improving the inputs given to the algorithm.

---

# Intuition

Suppose you're predicting house price.

Available features:

```text
Length = 20
Width = 30
```



So create a new feature:

```text
Area = 600
```

Now the model can learn more useful relationships.

---

# Why Feature Engineering Helps

The model only sees the features you provide.

If important information is hidden inside the data, the model cannot automatically discover it (especially simple models like linear regression).

Feature engineering exposes useful information.

---

# Common Types of Feature Engineering

## 1. Combining Features

Example:

```text
Length = 20
Width = 30
```

Create:

```text
Area = Length × Width
```

---

Salary example:

```text
Salary = 50000
Experience = 5
```

Create:

```text
Salary_Per_Year = Salary / Experience
```

---

## 2. Extracting Information

Date:

```text
2026-07-05
```

Create:

```text
Year = 2026
Month = 7
Day = 5
Weekday = Sunday
```

Sometimes month or weekday contains more predictive information than the full date.

---

## 3. Domain Knowledge Features

Predicting student performance:

Original:

```text
Attendance
Study Hours
```

Create:

```text
Attendance × Study Hours
```

if domain knowledge suggests both together matter.

---

## 4. Categorical Transformations

Original:

```text
Mumbai
Delhi
Pune
```

Convert into model-friendly features using encoding.

Example:

```text
Mumbai -> [1,0,0]
Delhi  -> [0,1,0]
Pune   -> [0,0,1]
```

You already encountered this during preprocessing.

# House Price Example

Original features:

| Size | Bedrooms |
| ---- | -------- |
| 1000 | 2        |
| 1500 | 3        |

Create:

| Size | Bedrooms | Size per Bedroom |
| ---- | -------- | ---------------- |
| 1000 | 2        | 500              |
| 1500 | 3        | 500              |

New feature:



Sometimes this captures information better than either feature alone.

---

# Feature Engineering vs Feature Selection

Many beginners confuse these.

### Feature Engineering

Create new features.

Example:

```text
Area = Length × Width
```

---

### Feature Selection

Choose which features to keep.

Example:

Keep:

```text
Area
Bedrooms
```

Remove:

```text
Length
Width
```

---

# Python Example

Dataset:

```python
import pandas as pd

df = pd.DataFrame({
    "length":[20,30,40],
    "width":[10,15,20]
})
```

Create area:

```python
df["area"] = df["length"] * df["width"]
```

Result:

```text
length width area
20     10    200
30     15    450
40     20    800
```

---

# Another Example

Age:

```python
df["age_squared"] = df["age"] ** 2
```

Log transformation:

```python
import numpy as np

df["salary_log"] = np.log(df["salary"])
```

---

# Why Linear Regression Needs It

Linear regression can only learn linear relationships.



This idea directly leads to the next topic: **Polynomial Regression**.

---

# Good Feature Engineering Characteristics

A good engineered feature:

* Contains useful information.
* Reduces noise.
* Makes relationships easier to learn.
* Is based on logic or domain knowledge.

---

# Bad Feature Engineering

Creating random features:

```python
random_number = np.random.randint(0,100)
```

usually adds noise.

More features ≠ better model.

Useful features matter.

---

# What You Should Remember

1. Feature engineering = creating new features from existing features.
2. The goal is to expose useful information to the model.
3. Common operations:

   * Multiplication
   * Division
   * Log
   * Square
   * Date extraction
   * Encoding
4. Better features often improve performance more than changing algorithms.
5. Feature engineering is especially important for linear models.
6. Polynomial Regression is essentially feature engineering using powers like (x^2, x^3), etc. and is the next topic.
