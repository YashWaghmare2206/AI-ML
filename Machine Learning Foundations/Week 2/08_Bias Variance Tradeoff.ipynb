# 8. Bias–Variance Tradeoff

This topic explains **why underfitting and overfitting happen**.

Everything we've learned so far leads here.

---

# Theory

Prediction error can be thought of as coming from:

1. **Bias**
2. **Variance**
3. Irreducible Noise

In ML we mainly focus on:

```text
Bias
Variance
```

---

# Bias

Bias means:

> How wrong are the model's assumptions?

A high-bias model is too simple and cannot capture the real pattern.

---

### Example



A straight line assumes:

> "Data should look linear."

That assumption is wrong.

Result:

```text
Underfitting
```

---

# Variance

Variance means:

> How sensitive is the model to changes in training data?

A high-variance model changes a lot if the training data changes slightly.

---

### Example

Degree 20 polynomial.

Dataset A:

```text
slightly noisy
```

Produces one curve.

Dataset B:

```text
same pattern, different noise
```

Produces a very different curve.

The model is unstable.

Result:

```text
Overfitting
```

---

# Intuition

Imagine drawing a curve through points.

### High Bias

Model is too rigid.

```text
Straight line
```

Cannot follow the pattern.

---

### High Variance

Model is too flexible.

```text
Wiggly curve
```

Follows every tiny fluctuation.

---

### Balanced

Captures trend without chasing noise.

---

# Relationship

| Bias     | Variance | Result              |
| -------- | -------- | ------------------- |
| High     | Low      | Underfitting        |
| Low      | High     | Overfitting         |
| Moderate | Moderate | Good Generalization |

---

# Why It's Called a Tradeoff

Usually:

When Bias ↓

Variance ↑

---

When Variance ↓

Bias ↑

---

Example:

Degree 1 polynomial:

```text
High Bias
Low Variance
```

---

Degree 20 polynomial:

```text
Low Bias
High Variance
```

---

Middle degree:

```text
Balanced
```

---

# Mathematical View



You usually don't calculate this in practice.

But understanding the idea is important.

---

# Polynomial Example

Degree 1:

```text
High Bias
```

Misses curvature.

---

Degree 2:

```text
Balanced
```

Captures pattern.

---

Degree 20:

```text
High Variance
```

Fits noise.

---

This is why choosing model complexity matters.

---

# Train and Validation Errors

### Underfitting

```text
Training Error    High
Validation Error  High
```

High Bias.

---

### Good Fit

```text
Training Error    Low
Validation Error  Low
```

Balanced.

---

### Overfitting

```text
Training Error    Very Low
Validation Error  High
```

High Variance.

---

# Bias and Features

Suppose predicting house price.

Features:

```text
Bedrooms
```

only.

Likely high bias because important information is missing.

---

Add:

```text
Area
Age
Location
Bathrooms
```

Bias decreases.

Model can learn richer patterns.

---

---

---

# How to Reduce High Bias

If underfitting:

* Add features
* Increase polynomial degree
* Use a more powerful model
* Reduce regularization

---

# How to Reduce High Variance

If overfitting:

* More training data
* Simpler model
* Lower polynomial degree
* Regularization
* Feature selection

---

# The Goal

We do not want:

```text
Minimum Bias
```

or

```text
Minimum Variance
```

individually.

We want:

```text
Best balance between them
```

because that gives the lowest validation/test error.

---

# Connection to Next Topic

Suppose a model is overfitting.

One way to reduce variance is:

```text
Regularization
```

Regularization discourages extremely large weights and helps the model generalize better.

That is exactly the next topic.

---

# What You Should Remember

1. Bias = error due to overly simple assumptions.
2. Variance = sensitivity to training data.
3. High Bias → Underfitting.
4. High Variance → Overfitting.
5. Increasing complexity usually decreases bias but increases variance.
6. Good ML models balance bias and variance.
7. Regularization is one of the main tools for controlling variance.
