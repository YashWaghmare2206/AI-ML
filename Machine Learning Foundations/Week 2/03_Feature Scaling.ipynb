# 03. Feature Scaling

This topic exists because not all features have the same numerical range.

---

# The Problem

Suppose we're predicting house prices.

Features:

| Size (sq ft) | Bedrooms |
| ------------ | -------- |
| 1000         | 2        |
| 1500         | 3        |
| 2000         | 4        |

Notice:

* Size → values around 1000–2000
* Bedrooms → values around 2–4

The ranges are very different.

---

# Why Is This Bad?



The "bedrooms" feature contributes almost nothing compared to size.

During gradient descent, the algorithm has difficulty finding optimal weights efficiently.

---

# Intuition

Imagine finding the minimum of a cost function.

### Well-scaled features

Ranges are similar:

* Age: 0–10
* Bedrooms: 0–10

Cost surface looks roughly circular.

Gradient descent moves directly toward the minimum.

---

### Poorly-scaled features

Ranges are very different:

* Area: 0–5000
* Bedrooms: 0–5

Cost surface becomes long and narrow.

Gradient descent zig-zags.

Needs many more iterations.

---

Andrew Ng often shows:

* Unscaled → stretched ellipse
* Scaled → almost circle

Gradient descent works much faster on the circular version.

---

# Mathematical View

Suppose:



A tiny change in (w_1) affects the cost much more than the same change in (w_2).

Therefore gradients have very different magnitudes.

Gradient descent struggles because one direction is much steeper than another.

---

# What Feature Scaling Does

It transforms features so they have comparable ranges.

Example:

Before:

| Area | Bedrooms |
| ---- | -------- |
| 1000 | 2        |
| 1500 | 3        |
| 2000 | 4        |

After scaling:

| Area | Bedrooms |
| ---- | -------- |
| 0.2  | 0.4      |
| 0.3  | 0.6      |
| 0.4  | 0.8      |

Now both features are on similar scales.

---

# When Should We Scale?

Usually:

✅ Linear Regression (Gradient Descent)

✅ Logistic Regression

✅ Neural Networks

✅ KNN

✅ K-Means

✅ SVM

---

Often not necessary for:

✅ Decision Trees

✅ Random Forests

✅ XGBoost

because they split on thresholds rather than distances or gradients.

---

# Python Example

Dataset:

```python id="jot7rv"
import numpy as np

X = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 4]
])
```

Manual scaling:

```python id="ykwx5i"
X_scaled = X / np.max(X, axis=0)

print(X_scaled)
```

Output:

```python id="3e98mb"
[
 [0.5 0.5]
 [0.75 0.75]
 [1.0 1.0]
]
```

---

# Scikit-Learn

Feature scaling is usually done with preprocessing tools.

For example:

```python id="d8ff18"
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)
```

# Important Distinction

Many beginners think:

> Scaling changes the information.

It doesn't.

If:

* House A is larger than House B before scaling,

it remains larger after scaling.

Scaling changes the numerical representation, not the underlying relationships.

---

# What you should remember

1. Feature scaling makes features have similar numerical ranges.
2. It helps gradient descent converge faster.
3. Unscaled features create elongated cost surfaces and slow learning.
4. Scaling does not remove information.
5. Common methods:

   * Divide by maximum
   * Min-Max scaling
   * Z-Score normalization
6. Z-Score normalization is the most commonly used method in ML and is next.
