# Problem Regularization Solves

Suppose your model is overfitting.

Training error:

```text
Very Low
```

Validation error:

```text
High
```

The model is learning noise.

We need a way to make the model simpler **without removing features**.

That's what regularization does.

---

# Core Idea

Regularization says:

> "Large weights are suspicious."

If a model needs extremely large weights to fit the data, it may be memorizing noise.

So we add a penalty for large weights.

# What is λ (Lambda)?



controls how strong the penalty is.

---

### λ = 0



No regularization.

---

### Small λ

```text
Little regularization
```

---

### Large λ

```text
Strong regularization
```

Weights are pushed toward zero.

---

# Intuition



Huge weights.

Regularization says:

> "Are these really necessary?"

and penalizes them.

---

After regularization:




Model becomes smoother and usually generalizes better.

---

# Geometric Intuition

Without regularization:

The algorithm only cares about minimizing training error.

---

With regularization:

The algorithm tries to:

1. Fit the data.
2. Keep weights small.

Both goals must be satisfied simultaneously.

---

# Why Small Weights Help

Consider polynomial regression:



If weights become huge:

```text
Very wiggly curve
```

Often overfits.

---

If weights remain moderate:

```text
Smooth curve
```

Usually generalizes better.

---

# Effect on Bias and Variance

Regularization:

```text
Bias ↑ slightly
Variance ↓ significantly
```

This is often a good tradeoff.

---

Remember:

Overfitting = High Variance.

Regularization reduces variance.
