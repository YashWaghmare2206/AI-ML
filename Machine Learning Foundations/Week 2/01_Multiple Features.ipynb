# 01. Multiple Features

## Theory

In simple linear regression, we predict using **one feature**:



Example:

* (x) = house size
* (y) = house price

---

In **multiple linear regression**, we use **multiple features** to predict the target:



---

## Intuition

Suppose you're predicting a house price.

Using only:

* House size

may not be enough.

Price can also depend on:

* Number of bedrooms
* Age of house
* Distance from city
* Number of bathrooms

So instead of learning one weight, the model learns a weight for each feature.



The model automatically learns how much each feature contributes.

## Gradient Descent

Weights update individually:



for every feature (j).

If there are 100 features:

* 100 weights
* 100 gradients

plus bias.

---

## Python Implementation (Without Vectorization)

```python
def predict(x, w, b):
    prediction = 0

    for j in range(len(w)):
        prediction += w[j] * x[j]

    prediction += b

    return prediction
```

Example:

```python
x = [1000, 3, 5]

w = [200, 10000, -500]

b = 50000

y_hat = predict(x, w, b)

print(y_hat)
```

---

## NumPy Implementation

```python
import numpy as np

x = np.array([1000, 3, 5])

w = np.array([200, 10000, -500])

b = 50000

y_hat = np.dot(w, x) + b

print(y_hat)
```

This is the standard approach used in ML.

---

## Important Insight

Single Feature:



* Input = one number
* Weight = one number

Multiple Features:



* Input = vector
* Weight = vector

The model is still a **linear regression model**.

The only difference is that the input has multiple dimensions.
